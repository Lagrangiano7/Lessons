# 1. Grupos

## 1.1 Definición

**Def.:** un "grupo" es un conjunto $C$ sobre el que hay definida una operación $\odot$ tal que:

$$\phi: G \times G \to G$$
$$(g, g') \mapsto g\odot g'$$

que cumple las siguientes 4 propiedades definitorias:

1. El producto es **cerrado**: $gg'\in G; \ \forall g, g'\in G$
2. El producto es **asociativo**: $(ab)c=a(bc)$
3. Existe elemento **neutro**: $eg=ge=g; \ \ \forall g\in G$
4. Existe elemento **inverso**: $\forall g\in G \ \exists g^{-1}\in G \text{ tq. } gg^{-1}=g^{-1}g=e$



**Def.:** el "orden" de un grupo es el cardinal de su conjunto

**Def.:** dado un elemento del grupo $g\in G$, su "orden" es el número de veces que hay que aplicar $a$ a sí mismo para obtener el elemento neutro: $a^r=e$

**Def.:** un grupo se dice "cíclico" si existe un elemento del mismo que lo genere completamente: $G=\{e, a, a^2, \ldots, a^{r-1}\}$ donde $a^r=e$

**Def.:** un grupo se dice "abeliano" o "conmutativo" si el producto interno, además de ser asociativo, es conmutativo (para todos los elementos del grupo)

## 1.2 Subgrupos

**Def.:** un "subgrupo" $H$ de un grupo $G$ es un subconjunto de $G$ con la misma operación interna tal que tiene estructura de grupo en sí mismo. Se denota $H< G$. Al igual que sucede con los espacios vectoriales, todos los grupos tienen como mínimo dos subgrupos, llamados "impropios", que son $\{e\}$ y él mismo

**Criterio de subgrupo:** $H$ subgrupo de $G$ si y solo si dados cualesquiera dos elementos de $H$, entonces $g(g')^{-1}$ también está en $H$:

$$\boxed{H< G \Longleftrightarrow g(g')^{-1}\in H; \ \ \forall g, g'\in H}$$

**Dem.:**

$\Longrightarrow$:

Como $H$ subgrupo de $G$, sabemos que para todo $g, g'\in H$ sus inversos también están incluidos en $H$. Entonces, como tanto $g$ como $(g')^{-1}$ son elementos de $H$ y este es cerrado por ser grupo, $g(g')^{-1}\in H.$

$\Longleftarrow$:

Ahora tenemos que demostrar que un subconjunto de $G$ con el producto interno heredado de este grupo cumple las 4 propiedades de grupo:

1. Asociatividad: trivial, el producto es heredado y como $G$ es grupo y todos los elementos de $H$ lo son también de $G$, todos los elementos de $H$ obedecen asociatividad en el producto
2. Existencia de elemento neutro: escogiendo $g'=g$, tenemos que $gg^{-1}=e\in H$
3. Existencia de elemento inverso: escogiendo $g=e$, tenemos que $(g')^{-1}\in H$
4. Clausura: sean $h_1, h_2\in H$. Tomando $g=h_1$ y $(g')^{-1}=h_2^{-1}$, tenemos que $h(h_"^{-1})^{-1}\in H$ y por ende $h_1h_2\in H_\blacksquare$

**Proposición:** sean $H_1, H_2\subset G$, entonces $H_1\cap H_2$ es también subgrupo de $G$

## 1.3 Grupos de permutaciones

**Def.:** se define el grupo de permutaciones de $n$ elementos $S_n$ como el conjunto de todas las operaciones de permutación de $n$ elementos. El producto de grupo es la composición de las permutaciones. Por ser permutaciones, tiene orden $n!$ (todas las maneras en las que podemos ordenar $n$ elementos)

**Notación de permutaciones:** dados $n$ elementos, podemos escribir una permutación sobre ellos de varias formas. Una de ellas es con una matriz de 2 filas: en la primera, ponemos los $n$ elementos en la configuración original. En la segunda fila, los ponemos ya permutados. Así, sabemos que el elemento i-ésimo ha ido a parar al j-ésimo. Alternativamente, podemos escribir la permutación en notación de **ciclos**: si el elemento i va al j, el j al k y el k al i, escribiríamos $(ijk)$

**Def.:** el "orden" $m$ de un ciclo es el número de elementos que contiene

**Def.:** la "paridad" de un ciclo es $\pi_c = (-1)^{m-1}$. Como era de esperar, viene de descomponer el ciclo en trasposiciones y ver que para realizar una permutación cíclica de $m$ elementos hacen falta $m-1$ trasposiciones

**Descomposición de una permutación:** dada una permutación de $n$ elementos, podemos descomponerla en ciclos disjuntos que conmutan entre sí, y la descomposición es **única**. Alternativamente, podemos descomponerla en trasposiciones, pero hay infinitas formas de hacerlo. Sin embargo, el número de trasposiciones que empleemos en las infinitas descomposiciones de una misma permutación siempre tendrá un número de elementos par ó impar.



**Proposición:** todo grupo de permutaciones $S_n$ contiene un "subgrupo alternado" $A_n$ compuesto por todas las permutaciones de orden par de $S_n$, cuyo orden es $|S_n|/2=n!/2$. Por ende, tiene índice 2 y el subgrupo alternado es normal

### 1.3.1 Grupo de permutaciones y teoría de grafos (mío)

- Podemos relacionar los elementos de un grupo de permutaciones de $n$ elementos con las transformaciones sobre los vértices de un polígono regular de $n$ vértices tales que lo dejan invariante (reflexiones, rotaciones, composiciones de estas, etc.)

**Proposición:** sea el grafo completo de $n$ nodos. El número de nexos nos da el número de trasposiciones que podemos hacer.

**Proposición:** como toda permutación se puede descomponer en trasposiciones, podemos generar el grupo $S_n$ solamente a partir del conjunto de todos los 2-ciclos (trasposiciones) de los $n$ elementos del grupo (+ la identidad), componiéndolos. Para hallar estas trasposiciones, basta con dibujar el grafo completo correspondiente e identificar cada nexo entre A y B como la trasposición $A\leftrightarrow B$

### 1.3.2 Ejemplo: grupo $S_3$

- Estudiemos el grupo de permutaciones de 3 elementos (polígono regular asociado: triángulo equilátero)
- El grupo es: $S_3=\{e, (12), (13), (23), (123), (132)\}$
- Tiene un subgrupo llamado "subgrupo alternado": $A_3=\{e, (123), (132)\}$, conformado por ciclos de paridad +1

### 1.3.3 Grupo de isotropía

**Def.:** el "grupo de isotropía" de un grupo $G$ para un punto dado es el subgrupo de $G$ conformado por todas aquellas transformaciones que dejan invariante al punto considerado

$$\boxed{Iso_x(G):=\{g \in G \text{ tq. } g\cdot x=x\}}$$

**Ejemplo:** dado un punto en $\mathbb R^3$ y $G=SO(3)$, el grupo de isotropía de dicho punto para el grupo considerado será el subgrupo de todas las rotaciones en el espacio cuyo eje pase por el punto (infinitas porque el parámetro es real) $\Longrightarrow$ el grupo de isotropía de $SO(3)$ para cualquier punto (que no sea el origen) es $SO(2)$:

$$Iso_x(SO(3))=SO(2); \ \forall x\in \mathbb R^3, x\neq 0$$

## 1.4 Grupos de transformaciones

**Def.:** un "grupo de transformaciones" $G$ sobre el conjunto $X$ es un grupo que además tiene una acción definida sobre el conjunto $X$. Esto es, existe una aplicación $g\cdot x=\phi(g, x)$ que define la acción del grupo sobr el conjunto. Además, la aplicación debe cumplir:

1. Ser homomorfismo: $(gg')\cdot x=g\cdot(g'\cdot x); \forall x \in X; \forall g, g'\in G$
2. Ser compatible con la identidad: $e\cdot x =x; \ \forall x \in X$ (viene de primera propiedad, lo pongo por ser explícito)

- Por ejemplo, un grupo de transformaciones es el de las traslaciones en el de rotaciones en el plano: $SO(2)$. La aplicación $\phi$ es aquella a la que le damos el elemento correspondiente del grupo y el punto que deseamos rotar, y nos da el punto rotado

### 1.4.1 Órbita

**Def.:** dado un punto $x\in M$, la "órbita" es la región de la variedad a la que el grupo mapea a $x$:

$$\boxed{Orb_G(x):=\{g\cdot x; \ \forall g \in G\} \subseteq M}$$

**Ejemplo:** si $G=SO(2)$ y $x\in \mathbb R^2$ con $x\neq 0$, entonces la órbita de $x$ será una circunferencia de radio la separación de $x$ del origen, porque habremos construido la órbita como el conjunto resultante de aplicar todas las rotaciones posibles en el plano al punto $x$

### 1.3.2 Trasitividad

**Def.:** la acción del grupo, $\phi$, se dirá "transitiva" si dada cualquier pareja de puntos sobre la variedad, puedo pasar de un punto al otro con una transformación del grupo:

$$\boxed{\phi \text{ transitiva } \Longleftrightarrow \exists g\in G \text{ tq. } \phi(g, x_1)=x_2; \ \forall x_1, x_2\in M}$$

**Ejemplo:** el grupo de traslaciones es transitivo. Sin embargo, el grupo de rotaciones en el plano no lo es, porque si tengo dos puntos en $\mathbb R^2$ con diferente separación del origen, no puedo pasar de uno al otro con una rotación sobre el plano

**Proposición:** en un grupo transitivo, todos los puntos tienen la misma órbita, que es toda la variedad

### 1.3.3 Efectividad

**Def.:** la acción de un grupo de transformaciones se dice "efectiva" si para cada elemento de la variedad existe al menos una transformación que lo mueve

$$\boxed{\phi \text{ efectiva }\Longleftrightarrow \forall x\in M \ \exists g\in G \text{ tq. } g\cdot x \neq x}$$

**Ejemplo:** $SO(2)$ no es efectiva porque el origen no se mueve con ninguna rotación

### 1.3.4 Fidelidad

**Def.:** la acción de un grupo de transformaciones se dice "fiel" si para cada transformación de $G$ hay al menos un punto de la variedad que muevo

$$\boxed{\phi \text{ fiel } \Longleftrightarrow \forall g\in G-\{e\} \ \exists x\in M \text{ tq. } g\cdot x \neq x}$$

**Ejemplo:** $SO(2)$ sí es fiel porque toda rotación no trivial mueve al menos un punto (de hecho, mueve todos salvo el origen)

# 2. Aplicaciones entre grupos

- Queremos ahora una forma de establecer relaciones entre elementos de distintos grupos. Esto lo haremos con homomorfismos porque permiten conservar la estructura algebraica

**Def.:** un "homomorfismo" es una aplicación entre dos estructuras algebraics tal que conserva la estructura de la estructura de partida

- De este modo, un homomorfismo entre grupos $f: G\to G'$ mapeará los elementos de $G$ a elementos de $G'$ de tal forma que:

$$\boxed{f(g_1\odot g_2)=f(g_1)\odot'f(g_2)}$$

**Kernel:** el "kernel / núcleo" de un homomorfismo es el conjunto de todos los elementos del dominio tales que mapean a la identidad del grupo de llegada. **Es un subgrupo normal de $G$**

$$\boxed{Ker[f] := \{g\in G \text{ tq. } f(g)=e'\}}$$

**Imagen:** la "imagen" de un homomorfismo es el conjunto de todos los elementos del grupo de llegada a los que se mapea el grupo de partida. **Es un subgrupo de G'** (no necesariamente normal)

$$\boxed{Im[f]:=\{g'\in G' \text{ tq. } \exists g\in G \text{ tq. } f(g)=g'\}\equiv f(G)}$$

**Propiedad:** un elemento inverso mapea al elemento inverso en el espacio de llegada. En particular, la identidad se mapea a la identidad:

$$\boxed{f(g^{-1})=\left(f(g)\right)^{-1}}$$

$$\boxed{f(e)=e'}$$

**OJO!:** esta notación en RHS no significa "inversa del homomorfismo". De hecho, en general un homomorfismo no tiene por qué tener inversa definida. Esta notación significa que mapea al elemento de $G'$ tal que es el inverso de $f(g)$, también elemento de $G'$

## 2.1 Análogo a la fórmula de las dimensiones

- En álgebra lineal, para un homomorfismo $f: V\to V'$, tenemos que se cumple $dim(V)=dim(Ker[f])+dim(Im[f])$
- Para los homomorfismos sobre grupos, tenemos una relación análoga:

$$\boxed{f: G\to G' \Longrightarrow |G|=|Ker[f]|\times |Im[f]|}$$

donde $|A|$ significa "orden del grupo A"

## 2.1 Ejemplo: homomorfismo entre cuaterniones y grupo de Klein

- Grupo de cuaterniones: $Q=\{1,- 1, i, -i, j, -j, k, k\}$ (omito resultados de la op. de grupo)
- Grupo de Klein: $V_4=\{e, a, b, c\}$ tq. todos los elementos son de orden 2

- Podemos definir un homomorfismo $f:Q\to V_4$ tal que:

$$f(\pm 1)=e$$
$$f(\pm i)=a$$
$$f(\pm j)=b$$
$$f(\pm k)=ab$$

- De aquí leemos directamente que $Ker[f]=\{-1, 1\}$ y que $Im[f]=V_4$, de modo que el homomorfismo no es inyectivo pero sí suprayectivo
- Comprobamos inmediatamente que se cumple la fórmula de los órdenes

## 2.2 Primer toerema de isomorfía de Noether

- Dado un homomorfismo entre grupos: $f: G\to G'$, entonces el grupo cociente del kernel de la aplicación es isomorfo a la imagen de la aplicación:

$$\boxed{f \text{ homomorfismo entre G y G'} \Longrightarrow G/Ker[f]\cong Im[f]}$$

**Intuición:** dado un homomorfismo entre dos grupos, es totalmente equivalente estudiar su imagen que estudiar el grupo cociente de su kernel

## Automorfismos

**Def.:** un "automorfismo" es un isomorfismo de un grupo sobre sí mismo (endomorfismo + isomorfismo):

$$\varphi: G\to G$$

mapeando todo el grupo de partida a todo el grupo de llegada de forma invertible respetando la operación de grupo

**Proposición:** el conjunto de todos los automorfismos de un grupo con la operación de composición conforma a su vez un grupo: $\text{Aut}_G$

**Ejemplo:** podemos definir un homomorfismo que nos dé el conjugado de los elementos de $G$ con respecto a algún elemento de $G$ previamente fijado:

$$\varphi_1: G\to G$$
$$g\mapsto \varphi_1(g)=(g_1)^{-1}gg_1$$

### Automorfismos internos

**Proposición:** el conjunto de todos los automorfismos de conjugación conforma un grupo llamado "grupo de automorfismos internos de $G$", y es subgrupo normal del grupo de automorfismos de $G$

$$\text{Int}_G :=\{\varphi_i\in \text{Aut}_G \ \text{ tq. } \varphi_i(g)=(g_i)^{-1}gg_i\} \subset_N \text{Aut}_G$$

### Automorfismos externos

**Def.:** los "automorfismos externos" son todos aquellos que no son internos:

$$\text{Ext}_G := \text{Aut}_G-\text{Int}_G$$

**Proposición:** el grupo de permutaciones de 3 elementos $S_3$ no tiene automorfismos externos. Todo automorfismo que podamos definir sobre este grupo es una conjugación

**OJO!:** si tengo por ejemplo el subgrupo $A_3=\{e, (123), (132)\}$, entonces la aplicación $\varphi(g)=(12)^{-1}g(12)$ es un automorfismo de $A_3$ porque coge elementos de este subgrupo y devuelve elementos del mismo, pero no es interno porque el elemento $(12)$ no pertenece a $A_3$. Por tanto, aunque tenga forma de conjugación, realmente es un automorfismo externo

# 4. Cosets

## Relaciones de equivalencia

**Def.:** sea un conjunto no vacío $K$ y $R$ una relación binaria definida sobre $K$ tal que si dos elementos están relacionados, entonces $R(a, b)=1$ o bien $a\sim b$. Diremos que $R$ es relación de equivalencia si cumple 3 propiedades definitorias:

1. Reflexividad: todo elemento es equivalente a sí mismo: $a\sim a; \ \forall a\in K$
2. Simetría: $a\sim b\Longleftrightarrow b\sim a; \ \forall a, b\in K$
3. Transitividad: $a\sim b$ y $b \sim c$, entonces $a \sim c; \ \forall a, b, c\in K$

## Clases de equivalencia

**Def.:** dado un elemento $a\in K$, su "clase de equivalencia" es el conjunto de todos los elementos de $K$ equivalentes a $a$:

$$[a]:=\{b \in K \text{ tq. } a\sim b\}$$

**Def.:** decimos entonces que $a$ es un "representante" de esta clase de equivalencia. Intuitivamente, como todos los elementos de la clase de equivalencia son por definición equivalentes entre sí, cualquiera de ellos nos sirve como representante de dicha clase

**Def.:** el "orden" de una relación de equivalencia es el número de clases de equivalencia que esta genera

**Particiones:** una relación de equivalencia sobre un conjunto induce una "partición" del mismo. Esto es, el conjunto sobre el que se ha definido la relación se puede separar en subconjuntos disjuntos (las clases de equivalencia) tales que la unión de todos recupera el conjunto original:

$$K=\displaystyle\bigcup_{s}^N[a_s]; \ \ [a_i]\cap [a_j]=\varnothing, \ i\neq j$$

donde $N$ es el orden de la relación de equivalencia definida sobre $K$

## Conjunto cociente

**Def.:** el "conjunto cociente" $K/\sim$ de una relación de equivalencia es el conjunto de todas las clases de equivalencia a las que esta relación da lugar. Si se quiere, el orden de la relación de equivalencia es el cardinal del conjunto cociente

$$\boxed{K/\sim := \{[a] : \ a\in K\}}$$

## Cosets

**Def.:** sea $H< G$. Diremos que dos elementos de $G$ son equivalentes si cumplen el criterio de subgrupo de $H$ (sin estar en él necesariamente):

$$g\sim g' \Longleftrightarrow g^{-1}(g')^{-1}\in H; \ \ g, g'\in G$$

**Def.:** un "coset" de $H$ no es más que cada clase de equivalencia de esta relación

**Importante:** en otros recursos se define directamente el coset de $H$ del que $g$ es un representante como $gH$ ("coset por la izda.") ó $Hg$ ("coset por la dcha."). Nuestra definición de clase de equivalencia da lo mismo que esta, pero es más rica porque permite entender que **los cosets particionan el grupo grande $G$ en subconjuntos disjuntos**

**Proposición:** de entre todos los cosets de un subgrupo, solo hay uno de ellos que tenga asimismo estructura de grupo, y es el coset trivial $eH=He$. Esto se debe a que el resto no contienen la identidad

**Proposición (# de elementos de un coset):** dado $H< G$, los cosets de $H$ tienen (1) todos el mismo número de elementos y (2) este coincide con el orden de $H$

Concretamente:

$$|G|=[G:H]|H|$$

Pues hay $\frac{|G|}{|H|}$, todos ellos de orden $|H|$

## Subgrupos normales

**Def.:** $H< G$ se dirá "subgrupo **normal** de $G$" si todo coset por la izda. coincide con su correspondiente coset por la dcha.:

$$\boxed{H\triangleleft G \Longleftrightarrow gH=Hg; \ \ \forall g\in G}$$

**Nota:** podemos pasar el elemento del coset de la dcha. a la izda. como su inverso y reescribir la condición como "un subgrupo es normal si es invariante bajo conjugación por todo elemento de $G$":

$$gHg^{-1}=H; \forall g\in G$$

**OJO!:** esta notación no significa que la relación se cumpla para cada elemento de $H$ sustituyendo el elemento donde pone el nombre del subgrupo, sino que "cualquier elemento de $H$, al someterlo a conjugación por cualquier elemento de $G$, vuelve a dar un elemento de $H$"

**Proposición:** la conjugación de un subgrupo por un elemento dado de $G$ siempre dará otro subgrupo de $G$ (no se pierde la estructura). Si en particular el subgrupo es normal, dará él mismo

**Def.:** $G$ se dice "simple" si no contiene subgrupos normales

### ¿Qué pasa si el grupo es Abeliano?

- Todos los cosets por la dcha. coinciden con sus correspondientes cosets por la izda. y por ende todos los subgrupos de $G$ son normales.

### Teorema (orden del grupo y cosets)

- Dado un grupo $G$ y un subgrupo suyo $H\subset G$, entonces el orden de $G$ se puede calcular como el producto del orden de $H$ por el número de cosets que este induce

$$|G|=[G:H]|H|$$

donde $[G:H]$ se llama "índice de $H$ en $G$"

**Dem.:**

- Es bien fácil: como los cosets de cualquier subgrupo, por ejemplo $H$ particionan el grupo $G$, está claro que si contamos todos los elementos de los cosets tenemos el orden de $G$. Sin embargo, no hace falta ir contando uno a uno porque sabemos que todos los cosets tienen el mismo orden, y coincide con el orden del subgrupo que los induce. Por ende, si tenemos $q=[G:H]$ cosets, cada uno de orden $|H|$, particionando el grupo $G$, entonces el orden de $G$ será precisamente el producto de ambos números$_\blacksquare$

### Teorema (grupo cociente)

- Sea el conjunto cociente de la relación de equivalencia de los cosets (el conjunto de todos los cosets que particionan $G$):

$$G/\sim \equiv G/H = \{g_1H, \ldots, Hg_1, \}$$

- $H$ es subgrupo normal de $G \Longleftrightarrow$  entonces el conjunto cociente es un grupo bajo la operación (ahora entre cosets):

$$(gH)(g'H)=(gg')H$$

**Fíjate:** si $H$ no fuera normal, esto no tendría sentido porque la operación no estaría bien definida: no sabríamos si coger el coset por izq. o por dcha.

### Teorema ($q=2$ normal)

- Todo subgrupo de índice 2 es normal

**Nota:** equivalentemente, todo subgrupo de dimensión $|H|=|G|/2$ es normal. Este criterio obviamente solo sirve para cuando el orden del grupo grande sea par

**Dem.:**

- Si el índice del subgrupo en $G$ es 2, significa que 2 cosets particionan el grupo $G$ entero. Sin embargo, como solo uno de ellos puede ser grupo (el trivial, que coincide con el subgrupo) y como la conjugación de un subgrupo siempre da otro subgrupo, pero solo hay 1 grupo que pueda dar la conjugación de $H$, que es él mismo (la otra mitad de la partición no es un subgrupo de $G$) entonces el subgrupo $H$ es normal$_\blacksquare$

### Teorema de Lagrange

$$\boxed{\text{En un grupo finito, el orden de todo subgrupo es divisor del orden del grupo}}$$

**OJO!:** no quiere decir que si un grupo tiene $|G|=20$ deba tener subgrupos de órdenes $1, 2, 4, 5, 10, 20$, sino que dado cualquier subgrupo de $G$, es seguro que tendrá un orden de entre esta lista. Puede que haya más de un subgrupo con el mismo orden o puede que para un orden dado no haya ningún subgrupo con ese orden

# 5. Clases de conjugación

**Def.:** la "conjugación" es una relación de equivalencia tal que:

$$g\sim g'\Longleftrightarrow \exists \tilde g\in G \text{ tq. } g'=(\tilde g)^{-1}g\tilde g$$

**Def.:** las "clases de conjugación" son las clases de equivalencia de esta relación

### ¿Si el grupo es Abeliano?

- Entonces cada elemento de $G$ solo es equivalente a sí mismo y por ende el grupo $G$ queda particionado en $|G|$ clases de equivalencia de un solo elemento ("singletones")

**MUCHÍSIMO OJO!:** esto es solo si el grupo entero es abeliano. Si tengo un subgrupo que es abeliano, entonces no dará clases de conjugación en singletons necesariamente, porque el elemento de $G$ que uso para construir la clase de conjugación puede que no esté en ese subgrupo y por ende puede que los elementos del subgrupo y los del grupo no conmuten, de modo que el argumento no se sostiene (ver ej. 13b)

### Teorema ($S_n$ y clases de conjugación)

- Las clases de conjugación de $S_n$ separan por tipo de ciclos

### Centralizador de un subconjunto y centro de grupo

**Def.:** dado un subconjunto $S$ de $G$, el "centralizador" $Z_S$ de dicho subconjunto es el conjunto de todos los elementos de $G$  que conmutan con todos los elementos de $S$:

$$Z_S:=\{g \in G \text{ tq. } gg'=g'g; \ \ \forall g'\in S\}$$

**Proposición:** el centralizador de cualquier subconjunto de $G$ es subgrupo de $G$

**Def.:** el "centro del grupo $G$" es el centralizador del propio grupo, es decir, $Z_G$ (el conjunto de los elementos de $G$ que conmutan con todos los elementos de $G$)

**Proposición:** el centro del grupo es un subgrupo normal. Esto es trivial porque, como todos los elementos del centro conmutan con todos los elementos del grupo, ellos "ven" el grupo como abeliano. El subgrupo es por ende normal, porque las conjugaciones son triviales. Además, es un subgrupo abeliano

# 6. Producto Directo

**Def.:** sean dos grupos $G_1, G_2$, se define su producto directo como el producto cartesiano de todos sus elementos y donde la operación de grupo es:

$$(g_1, g_2)(g_1', g_2')=(g_1g_1',g_2g_2')$$

**Proposición:** el producto directo de dos grupos forma asimismo un grupo, con el conjunto y operación de su definición

**Propiedad:** el elemento inverso es el elemento de los inversos, cada uno de su grupo correspondiente: $(g_1, g_2)^{-1}=(g_1^{-1}, g_2^{-1})$

**Propiedad**: si los dos grupos son abelianos, entonces su producto directo también lo es

**Ejemplo:** el grupo de Klein $V_4$ es el producto directo del grupo cíclico de dos elementos: $C_2=\{e, \sigma\}$:

$$V_4\cong C_2\times C_2=\{(e, e), (e, \sigma), (\sigma, e), (\sigma, \sigma)\}$$

donde efectivamente cada elemento es de orden 2



## Proposición (subgrupos normales)

- Si $G=G_1\times G_2$, entonces $\{e_1\}\times G_2$ y $G_1\times \{e_2\}$ son subgrupos normales de $G$

**Dem.:**

Lo hacemos para el primero, y el segundo es igual. Que es subgrupo de $G$ es evidente porque es subconjunto de $G$ y además $G_2$ es grupo. La parte de la identidad es trivial. Entonces, basta con demostrar que $\{e_1\}\times G_2$ es invariante bajo conjugaciones de $G$:

$$(g_1, g_2)^{-1}(e, g_2')(g_1, g_2)=(e, g_2^{-1}g_2'g_2)\in \{e\}\times G_2$$

porque naturalmente el grupo $G_2$ es invariante bajo conjugaciones de elementos de sí mismo (si se quiere, por propiedad 1 de producto de grupo: clausura)$_\blacksquare$

## Teorema (factorización de un grupo)

- Sea $G$ grupo y sean $H, K$ subgrupos de $G$ tal que:

1. Son normales
2. No comparten más elementos de lo estrictamente necesario: $H\cap K=\{e\}$
3. Todos los elementos de $G$ se pueden escribir como el producto de un elemento de $H$ por otro de $K$

- Entonces, podemos factorizar el grupo grande como: $G\cong H\times K$  y esta descomposición es **única**

**Dem.:**

- **Proposición:** los elementos de $H$ y de $K$ conmutan
    - Por ser $H$ subgrupo normal de $G$, en particular cumplirá que $k^{-1}hk\in H$ para todo $h\in H, k\in K$. Lo mismo va para $K$
    - En particular, el elemento $k^{-1}hkh^{-1}=(k^{-1}hk)h^{-1}=k^{-1}(hkh^{-1})$ estará contenido tanto en $H$ como en $K$, pero como el único elemento que comparten es la identidad, entonces $k^{-1}hkh^{-1}=e \Longrightarrow hk=kh_\square$

- Ahora definimos el homomorfismo:

$$\varphi: H\times K\to G$$
$$(h,k)\mapsto hk$$

- Este es un isomorfismo, y por ende $G\cong H\times K_\blacksquare$

# 7. Producto Semidirecto

**Idea:** lo único que cambia con el producto directo es que el producto de grupo ya no es el mismo en la primera posición de la tupla

**Def.:** sean dos grupos $H, K$. Sea el homomorfismo que a cada elemento de $K$ le asigna un automorfismo de $H$: 

$$\phi: K\to \text{Aut}_H$$
$$k\mapsto \varphi_k:H\to H$$

Entonces, el producto semidirecto de $H$ y $K$, denotado $H\rtimes_\phi K$, es el conjunto de todos los pares $(h, k)$ con la operación de grupo:

$$\boxed{(h_1, k_1)(h_2, k_2)=(h_1\varphi_{k_1}(h_2), k_1k_2)}$$

**Propiedades del producto semdirecto de dos grupos:**

1. Es grupo
2. No es único, porque depende de la elección del homomorfismo $\phi$
3. Si el homomorfismo mapea todo elemento de $K$ al automorfismo identidad: $\varphi(h)=h$, entonces recuperamos el producto directo como caso particular del producto semidirecto
4. $|H\rtimes_\phi K|=|H|\cdot |K|$
5. $\bar H=H\rtimes_\phi\{e\}$ es subgrupo **normal** de $G$ e isomorfo a $H$
6. $\bar K= \{e\}\rtimes_\phi K$ es subgrupo de $G$ e isomorfo a $K$
7. La intersección de estos dos subgrupos es trivial: $\bar H\cap \bar K=\{(e, e)\}$

**Ejemplo:**

- ¿Es $S_3\cong C_3\times C_2$?: No, porque $C_2, C_3$ son ambos abelianos, de modo que su producto directo también. Sin embargo, $S_3$ no es abeliano, de modo que los dos grupos no pueden ser isomorfos
- ¿Puedo construirlo con algún producto semidirecto de estos dos grupos cíclicos?: Sí!
    - Sea $H=A_3$ subgrupo normal de $S_3$, isomorfo a $C_3$
    - Sea $K=\{e, (12)\}$ subgrupo de $S_3$, isomorfo a $C_2$
    - Sea el homomorfismo $\phi$ tal que:

$$\phi: K\to \text{Aut}_{A_3}$$
$$e\mapsto \varphi_e$$
$$(12)\mapsto \varphi_{(12)}$$

tal que: $\varphi_{(12)}(e)=e$, $\varphi_{(12)}(123)=(132)$, $\varphi_{(12)}(132)=(123)$ (automorfismo externo de conjugación por $(12)$)

Entonces, tengo que la operación de grupo será ahora:

$$(h_1, k_1)(h_2, k_2)=(h_1\varphi_{k_1}(h_2), k_1k_2)$$

por definición de producto semidirecto. Por ejemplo:

$$((123), (12))((132), (12))=((123)\varphi_{(12)}(132), (12)(12))=((132), e)\in G=H\rtimes_\phi K$$

Se comprueba que la tabla de Cayley de $C_3\rtimes_\phi C_2$ es la misma que $S_3$ y por ende $S_3\cong C_3\rtimes_\phi C_2 \ _\blacksquare$

**Nota:** fíjate que hemos empezado hablando de los subgrupos de $S_3$: $A_3$ y $\{e, (12)\}$  y hemos acabado hablando de $C_3$ y $C_2$. No pasa nada porque son isomorfos. Por tanto, también podíamos haber dicho que $S_3=A_3\rtimes_\phi \{e, (12)\}$
