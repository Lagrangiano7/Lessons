# 1. Grupos de Lie

**Def.:** un "grupo de Lie" es aquel conjunto que es una variedad diferenciable y que además tiene estructura algebraica de grupo, de tal forma que la operación binaria interna $\odot$ es diferenciable, y la inversa de cada elemento también

Por ende, cada elemento del grupo vendrá representado en una carta concreta por una tupla de $n$ parámetros, y todo es igual que en variedades diferenciables, solo que aquí además tenemos la operación interna de grupo

## Traslaciones y campos invariantes

**Def.:** se define la operación "traslación a izquierdas" de un punto de la variedad como "aplicarle a ese punto otro punto de la variedad actuando por la izquierda":

$$L_g:G\to G$$
$$g'\mapsto gg'$$

Por ser variedad diferenciable, tenemos además los correspondientes campos vectoriales. Entonces, podemos servirnos de esta operación para definir la correspondiente sobre los campos vectoriales:

**Def.:** se construye la "diferencial" de la traslación a izquierdas como sigue. Lo vemos con un ejemplo:

Sea un grupo de Lie de dimensión 2 donde la operación de grupo es:

$$gg'=(\alpha, \lambda)\odot(b, \mu)=(\alpha+be^\lambda, \lambda+\mu)$$

Entonces, la operación de traslación a izquierdas por $g_0$ será:

$$L_{g_0}g=(\alpha_0, \lambda_0)\odot (\alpha, \lambda)=(\alpha_0+\alpha e^{\lambda_0}, \lambda_0+\lambda)$$

Entonces, se define el **diferencial** de la traslación a izquiedas como aquella aplicación que traslada un campo a izquiedas, y no es más que la matriz Jacobiana de esta aplicación. Por eso a menudo se denota con $DL_{g_0}$

$$DL_g:\mathfrak X^1_0\to \mathfrak X^1_0$$

$$X\mapsto DL_g X$$

$$DL=\begin{pmatrix}\frac{\partial L^1}{\partial \alpha} & \frac{\partial L^1}{\partial \lambda} \\ \frac{\partial L^2}{\partial \alpha} & \frac{\partial L^2}{\partial \lambda}\end{pmatrix}$$

Donde la aplicación está expresada en la base natural de los campos vectoriales

En este caso, sería:

$$DL=\begin{pmatrix}e^{\lambda_0} & 0 \\ 0 & 1\end{pmatrix}$$

**Def.:** se define un "campo invariante a izquierdas" como aquel que, al trasladarlo a izquiedas, coincide consigo mismo en la nueva posición:

$$\boxed{DL_g X(g') = X(gg')}$$

**Ejemplo:** hallemos los campos invariantes a izquiedas del grupo del ejemplo anterior

La forma más general de un campo vectorial en un punto de la variedad es $X(\alpha, \lambda)=\xi(\alpha, \lambda)\partial_\alpha + \eta(\alpha, \lambda)\partial_\lambda$. Escribamos entonces la condición de invariancia a izquiedas y veamos qué condición han de cumplir estos campos. Esto nos dará el subconjunto de todos los campos vectoriales que son invariantes a izquiedas, que de hecho es subálgebra de Lie del espacio de campos vectoriales $\mathfrak X_{inv, L}< \mathfrak X$

$$\begin{pmatrix}e^{\lambda_0}\xi(\alpha, \lambda) \\ \eta(\alpha, \lambda)\end{pmatrix} = \begin{pmatrix}\xi(\alpha_0+\alpha e^{\lambda_0}, \lambda_0+\lambda) \\ \eta(\alpha_0+\alpha e^{\lambda_0}, \lambda_0+\lambda)\end{pmatrix}$$

De estas dos ecuaciones funcionales deducimos que:

$$\xi(\alpha, \lambda)=C_1 e^\lambda; \ \ \eta(\alpha, \lambda)=C_2$$

Por ende, el campo invariante a izquierdas más general será:

$$X_{inv, L}=C_1e^\lambda \partial_\alpha + C_2\partial_\lambda$$

**Forma mucho más fácil de hallar campos inv. a izdas (no hay que resolver nada):**

- Cojo un vector tangente a la identidad arbitrario: $X_e=\begin{pmatrix}A \\ B\end{pmatrix}$, donde las coordenadas son para la base natural
- Sé que por definición un campo inv. a izquierdas cumple: $DL_gX(g')=X(gg')$. Si en particular $g'=e$, entonces $X(g)$ obtenido como $X(g)=DL_gX_e$ es por definición inv. a izquierdas

En el ejemplo anterior, tendríamos que el campo más general inv. a izquierdas es:

$$X(\alpha, \lambda)=\begin{pmatrix}e^\lambda & 0 \\ 0 & 1\end{pmatrix}\begin{pmatrix}A \\ B\end{pmatrix}=Ae^\lambda \partial_\alpha +B\partial_\lambda$$

De aquí deducimos que hay dos campos inv. a izquierdas independientes:

$$X_1(\alpha, \lambda)=e^\lambda\partial_\alpha; \ \ X_2(\alpha, \lambda)=\partial_\lambda$$

**Def.:** el "álgebra de Lie" asociada al grupo es precisamente el subálgebra $\mathfrak X_{inv, L}\equiv \mathfrak g$ (también podíamos haber hecho un desarrollo equivalente con traslaciones a derechas)

**Observación:** un campo invariante a izquierdas está completamente determinado por su valor en un solo punto de la variedad (usualmente se coge el elemento neutro), porque de ahí ya podemos recuperar el campo en el resto de la variedad aplicándole la traslación a izquierdas

## Teorema (campos inv. y álgebra de Lie)

El álgebra de Lie es isomorfa al espacio tangente al elemento neutro del grupo:

$$\mathfrak g\cong T_e G$$

# 2. Álgebras de Lie

## Definiciones

**Def.:** un "álgebra" es un espacio vectorial sobre el cuerpo $\mathbb K$ que además tiene definida una aplicación bilineal (y obviamente es cerrado bajo esta). Es decir, $\mathcal A =(V, (\cdot, \cdot))$

**Def.:** un "álgebra asociativa" es aquella que tiene una aplicación asociativa

**Def.:** un "álgebra de Lie" es aquella cuya aplicación bilineal cumple lo siguiente:

1. Antisimétrica
2. Satisface identidad de Jacobi

**Nota:** obviamente, podemos construir un álgebra de Lie a partir de un álgebra asociativa, definiendo una nueva aplicación bilineal sirviéndonos de la anterior, tal que $[v_1, v_2]=(v_1, v_2)-(v_2, v_1)$. Esto es lo que haremos en el caso de que los elementos del álgebra sean matrices y queramos construir un álgebra de Lie, porque normalmente el producto bilineal se toma como el producto de matrices, y este es asociativo

**Def.:** diremos que un álgebra de Lie es **compacta** si es el álgebra de Lie de algún grupo de Lie compacto

**Def.:** como el álgebra es por definición cerrada bajo su aplicación bilineal, es de suponer que $(v_1, v_2)$ da otro elemento del álgebra, y como tal se puede descomponer en la base del álgebra. En general, tendremos entonces que:

$$(v_1, v_2)=\alpha_1 e_1 + \ldots + \alpha_n e_n$$

Pues a los coeficientes que dan la descomposición de la aplicación bilineal para elementos de la base se les llama **constantes de estructura** del álgebra:

$$\boxed{(e_i, e_j)=\displaystyle\sum_k^n C^k_{ij}e_k}$$

**OJO!:** dependen de la elección de base que hagamos!

**Nota:** naturalmente, hay $n^3$ constantes de estructura, pero dependiendo de la aplicación bilineal definida no todas son independientes. Por ejemplo, en álgebras de Lie, tenemos que cumplen: $C^k_{ij}=-C^j_{ji}$

**Def.:** un "subálgebra" es un subespacio vectorial $W<V$ que además es cerrado bajo la aplicación bilineal

## Teoremas relacionados con grupos

### As above so below

**Def.:** un subálgebra de Lie se dice "ideal" o "invariante" ($\mathfrak J$) si su conmutador con cualquier elemento del álgebra está dentro de $\mathfrak J$. Se denota $\mathfrak J \triangleleft \mathcal A$

$$\mathfrak J \triangleleft \mathcal A \Longleftrightarrow [v, \mathfrak J]\subseteq\mathfrak J; \forall v\in \mathcal A$$

**Fíjate:** todo álgebra de Lie $\mathcal A$ tendrá siempre al menos dos ideales, que son $\{0\}$ y $\mathcal A$. Estos se llaman ideales "impropios". Los no triviales se llaman "propios"

**Teorema:** sea $G$ grupo de Lie y sea $\mathfrak g$ su álgebra de Lie. Entonces:

- $\mathfrak g'<\mathfrak g \Longrightarrow \exists G'<G \text{ conexo tq. } Lie(G')=\mathfrak g'$
- $\mathfrak J \triangleleft \mathfrak g\Longrightarrow \exists H\triangleleft G \text{ conexo tq. } Lie(H)=\mathfrak  J$

Si además estos subgrupos son simplemente conexos, entonces $G'=exp(\mathfrak g')$ y $H=exp(\mathfrak J)$

### Correspondencia con grupos

- Sea $\mathfrak g$ álgebra de Lie de dimensión finita. Existe un **único** grupo de Lie **simplemente conexo** tal que su álgebra de Lie es precisamente $\mathfrak g$

## Homomorfismos de álgebras y Teorema del Kernel

**Def.:** un "homomorfismo" es una aplicación entre dos estructuras algebraicas de igual tipo (y por ende preserva sus propiedades)

**Def.:** un "homomorfismo de álgebras" será entonces aquella aplicación $f:\mathfrak g_1\to \mathfrak g_2$ tal que preserva la estructura de álgebra (espacio vectorial + aplicación bilineal)

1. $f(\alpha v_1+\beta v_2)=\alpha f(v_1)+\beta f(v_2)$
2. $f([v_1, v_2])=[f(v_1), f(v_2)]$

**Teorema:** el Kernel de la aplicación es un ideal del álgebra de partida:

$$\boxed{Ker [f] \triangleleft \mathfrak g_1}$$

**Teorema:** la imagen de la aplicación es un subálgebra (en general no será ideal) del álgebra de llegada:

$$\boxed{Im[ f] < \mathfrak g_2}$$


**Dem:**

(1)

$v_1\in Ker f \Longleftrightarrow f(v_1)=0$. Sea ahora $v_2\in \mathfrak g_1$. Entonces, el teorema será cierto si $[v_1, v_2]\in Ker f \ \forall v_1\in Ker f$. Precisamente, este conmutador es otro vector del álgebra, que estará en $Ker f$ si y solo si su aplicación da 0. Es decir: $f([v_1, v_2])=[f(v_1), f(v_2)]=[0, f(v_2)]=0\Longrightarrow [v_1, v_2]\in Ker f$ y por ende $Ker f $ ideal de $\mathfrak g_{1\blacksquare}$

(2)

Sea $B=\{v_1, \ldots, v_n\}$ base del álgebra de partida. Nos preguntamos si es subespacio vectorial, es decir, si la combinación lineal de cualquier elemento de la imagen es asimismo elemento de la imagen. Vemos que por las propiedades de los homomorfismos esto es cierto: $\alpha_1f(v_1)+\ldots + \alpha_n f(v_n)\in Im[f]$ porque $f(\alpha_1v_1+\ldots + \alpha_n v_n)\in Im[f]$. Ahora nos preguntamos si es subálgebra, es decir, si es cerrado bajo la aplicación bilineal del álgebra de llegada. $[f(v_i), f(v_j)]\in Im[f]$ porque $f([v_i,v_j])\in Im[f]$. Por ende, la imagen de un homomorfismo es subálgebra del álgebra de llegada$_\blacksquare$

## Tipos de álgebras

**Def.:** un álgebra de Lie <span style="color: red">no abeliana</span> se dice **simple** si no tiene ideales (propios)

**Def:** un álgebra de Lie <span style="color: red">no abeliana</span> se dice **semisimple** si no tiene ideales abelianos (pero en general puede tener ideales)

**Proposición:** un álgebra semisimple siempre se puede descomponer como suma directa de álgebras simples, que son todos los ideales del álgebra original

**Teorema:** todo álgebra semisimple (sobre los complejos) tiene un subálgebra de Cartan $h$, que cumple:

1. Es abeliana maximal (ningún otro conmutador da 0)
2. La representación adjunta de todo elemento de $h$ es diagonalizable simultaneamente (y por ende existe base común de autovectores en $\mathfrak g$)

**Proposición:** las distintas subálgebras de Cartan de un álgebra de Lie son conjugadas, y por ende isomorfas. Basta con hablar del "álgebra de Cartan" aunque haya varias

**Def.:** dada un álgebra simple (compleja), se llama **rango** a la dimensión del su subálgebra de Cartan

**Propisición:** $Rg[\mathfrak{sl}(n, \mathbb C)]=n-1$

### Clasificación de álgebras semisimples (complejas) (Cartan)

La clasificación de las álgebras semisimples se divide en 2 bloques: las "usuales" A-D y las "excepcionales" E-G. Dentro de estos dos grandes bloques, se enumeran por su rango

- $A_n \to \mathfrak {sl}(n+1, \mathbb C)$
- $B_n \to \mathfrak{so}(2n+1)$
- $C_n \to \mathfrak{sp}(2n, \mathbb C)$
- $D_n \to \mathfrak{so}(2n)$
- $E_6, E_7, E_8$
- $F_4$
- $G_2$

**OJO!:** $D_n$ empieza a partir de $n=2$ porque $D_1$ sería abeliana

**OJO!:** un álgebra compleja puede tener muchas formas "reales" si deseamos expresarla como tal. Por ejemplo, la forma real compacta de $\mathfrak{sl}(2, \mathbb C)$ es $\mathfrak{su}(2, \mathbb R)$

### Propiedades de las primeras álgebras

Veamos algunas relaciones de isomorfía entre elementos concretos de esta clasificación. Son de gran utilidad porque nos dicen que las álgebras que relacionemos son esencialmente la misma, y si además son álgebras de Lie de unos correspondientes grupos de Lie, ya sabemos que sendos grupos tienen los mismos generadores

1. $A_1\cong B_1 \cong C_1 \text{(complejas)} \Longrightarrow \mathfrak{su}(2)\cong \mathfrak{so}(3)\cong \mathfrak {sp}(2)$ (reales)
2. $A_3\cong D_3 \text{(complejas)} \Longrightarrow \mathfrak{su}(4)\cong \mathfrak{so}(6)$ (reales)
3. $B_2\cong C_2 \text{(complejas)} \Longrightarrow \mathfrak{so}(5)\cong \mathfrak{sp}(4)$ (reales)
4. $D_2\cong A_1\oplus A_2 \text{(complejas)} \Longrightarrow \mathfrak{so}(4)\cong \mathfrak{sl}(2)\oplus \mathfrak{sl}(2)$ (reales)

## Casimires

**Def.:** un "Casimir" es un elemento del álgebra universal de un álgebra que conmuta con todos los generadores de esta

**Def.:** el "álgebra universal" de un álgebra es el álgebra asociativa con unidad más pequeña que contiene a $\mathfrak g$ y cuya estructura de conmutador reproduce la estructura de $\mathfrak g$

**Teorema:** en álgebras semisimples, el número de Casimires independientes es igual al rango de dicho álgebra

**Proposición:** si el álgebra es semisimple, una forma de construir un Casimir es empleando la inversa de la forma de Killing, que denotaremos $(K^{-1})^{\alpha\beta}=K^{\alpha\beta}$. Si los generadores son $J_a$, tenemos entonces que:

$$C=K^{\alpha\beta}J_\alpha J_\beta$$

### Teorema irreps. y Casimires

$$\boxed{\text{irrep.} \Longrightarrow \text{los Casimires y el centro del grupo son proporcionales a la identidad}}$$

**Nota:** el inverso de esta implicación es muy útil para identificar representaciones que no sean irreducibles: basta con que uno de los Casimires no sea proporcional a $I$ para que la representación sea reducible

**Dem:**

Consideremos la representación irreducible de un grupo. Trivialmente, esta es idéntica a sí misma, de modo que estamos en el tercer caso del lema de Schur, y se cumple que "todo aquel operador que conmute con todos los elementos de la representación es proporcional a la identidad". Esto quiere decir que (1) los elementos del centro del grupo y (2) todos los Casimires (con esto agotamos el listado de elementos que conmutan con los elementos del grupo) han de ser proporcionales a la identidad, y hemos acabado$_\blacksquare$

## Representación de álgebras

**Def.:** se define la "representación" sobre un espacio vectorial $V$ de una estructura algebraica como aquel homomorfismo que asocia a cada elemento de dicha estructura un endomorfismo sobre $V$. Véase que, como es homomorfismo, debe preservar la estructura del espacio de partida. En el caso de la representación de grupos, esto quiere decir que los endomorfismos que se asignen deben ser regulares, porque tiene que existir inversa, pero en el caso de las álgebras no hay tal restricción porque no hay noción de elemento inverso, y los endomorfismos asignados bien podrían ser degenerados

**Nota:** también podemos definir representaciones de estructuras algebraicas sobre álgebras, pero por ser homomorfismo ahora la representación debe preservar asimismo la aplicación bilineal

### Representación adjunta de un álgebra de Lie

**Def.:** la representación adjunta de un álgebra de Lie $\mathfrak g$ es aquella representación del álgebra sobre sí misma que a cada elemento del álgebra le asocia el endomorfismo construido contrayendo el conmutador con dicho elemento. Es decir:

$$\boxed{\begin{gather}\rho: \mathfrak g\to End(\mathfrak g) \\  X\mapsto ad_x\equiv [X, \cdot]\end{gather}}$$

**Propiedades:** 

1. $[ad_X, ad_Y]=ad_{[X, Y]}$
2. $\rho_1\otimes \rho_2 = \rho_1\otimes I_2 + I_1\otimes \rho_2$
3. Los coeficientes de la matriz del endomorfismo serán precisamente las constantes de estructura del álgebra (en la base en la que estemos expresando todo)

**Def.:** se define la "forma de Killing" ($K$) como una forma bilineal y simétrica que a cada par de vectores les asocia la traza de sus representaciones adjuntas

$$\boxed{\begin{gather}K:\mathfrak g\times \mathfrak g \to \mathbb C \\ (X, Y) \mapsto Tr[ad_X\circ ad_Y]\end{gather}}$$

**Nota:** lógicamente, podemos escribir la forma de Killing en forma matricial si calculamos sus entradas en cada par de elementos de la base

**Proposición:** la forma de Killing en álgebras semisimples es un producto escalar, por ser una forma bilineal simétrica y no degenerada

## Teoremas de álgebras semisimples

### Teorema de Weyl (reducibilidad)

- Las representaciones de un álgebra semisimple (de dim. finita) son completamente reducibles

### Teoremas de clasificación (criterio de Cartan)

1. $\mathfrak g$ álgebra semisimple $\Longleftrightarrow$ la forma de Killing es no-degenerada ($\det K\neq 0$)
2. $\mathfrak g$ es álgebra (1) semisimple, (2) real y (3) compacta $\Longleftrightarrow$ la forma de Killing tiene signo definido

## Pesos de una representación

- Vamos a ver ahora una forma de traducir el estudio de las álgebras semisimples a estudiar un conjunto finito de vectores en el espacio euclideo. Introducimos el concepto paulatinamente

- Sabemos que todo álgebra de Lie semisimple de rango $r$ tiene un subálgebra de Cartan $\mathfrak h$ de dimensión $r$
- Sabemos también que todos los elementos del álgebra de Cartan conmutan: $[h', h'']=0; \ \ \forall h', h''\in \mathfrak h$
- Como las representaciones son homomorfismos, toda representación debe reflejar esta conmutatividad en el subálgebra de Cartan, de modo que debe existir un subespacio de los endomorfismos $End(V)$ en el que dichos endomorfismos también sean conmutativos

$$\mathfrak h \text{ abeliano } \mapsto End_{\mathfrak h}\text{ (abeliano) }<End(V) $$

- Si la representación es fiel (i.e: inyectiva), entonces habrá $r$ endomorfismos independientes que conmuten entre sí (si se quiere, son base de $End_{\mathfrak h}$). Entonces, como admitimos que estamos trabajando sobre los complejos, es un resultado conocido que $r$ matrices que conmutan entre sí se pueden diagonalizar simultaneamente
- Por ende, existe base de $V$ de autovectores de estos endomorfismos, $B=\{v_{1}, \ldots, v_n\}$. Si llamamos a los endomorfismos asociados a los elementos del subálgebra de Cartan como los nombres de sus elementos en mayúsculas: $H_1=\rho(h_1)$, tenemos:

$$H_iv_j=\lambda_{ji}v_j$$

**Def.:** los $\lambda_{ji}$ son los **pesos** de la representación

**Def.:** se define el "vector de pesos" de un cierto vector $v_j\in V$ como la r-tupla de los autovalores que tiene un autovector DETERMINADO para todos los endomorfismos

$$\lambda_j=(\lambda_{j1}, \ldots, \lambda_{jr})$$

**Fíjate:** como vamos cambiando de endomorfismo y hay $r$ endomorfismos independientes (si la repr. es fiel), entonces el las tuplas tendrán exactamente $r$ elementos

**Fíjate:** como hay $n=dim(V)$ vectores independientes en la base, tendremos en total $n$ vectores de pesos

<div style="border:1px solid black; padding:10px; width:60%; margin:auto; text-align:center;">
Una representación (fiel) de un álgebra semisimple de rango $r$ sobre un espacio vectorial de dimensión $n$ tiene $n$ vectores de peso (en general no todos son linealmente independientes) con $r$ componentes 
</div>


**Def.:** fijada una irrep., se dice que un vector de pesos es "de peso máximo" si sus componentes son positivas ó 0

**IMPORTANTE:** el peso máximo de una irrep. determina todos los demás pesos de la irrep.

**Algoritmo:**

1. Empieza con el peso máximo y fíjate en una componente positiva
2. Resta la raíz correspondiente a esa componente hasta que obtengas el número opuesto en esa posición
3. Repite este procedimiento con todos los nodos que haya en el diagrama

**OJO!:** es fácil determinar qué pesos salen en cada representación. Lo que no es tan fácil es determinar la multiplicidad geométrica de cada uno, y para ello existen fórmulas como la de Freudenthal

## Raíces

**Def.:** si en concreto escogemos la representación adjunta de un álgebra semisimple, entonces los pesos reciben el nombre especial de "raíces". Este nombre no es arbitrario, sino que viene de que los vectores de peso en esta representación adjunta forman un sistema de raíces

### Sistemas de raíces

**Def.:** sea $E$ un espacio vectorial euclideo, y como tal dotado del producto escalar usual. Se define un "sistema de raíces" $\Phi$ en $E$ como un conjunto finito de vectores **no nulos** de $E$ (llamados "raíces") que satisfacen:

1. Generan $E$ ($E=lin\{\Phi\}$)
2. Fijada una raíz, el sistema es cerrado bajo reflexiones con un plano perpendicular a dicha raíz

Diremos que el sistema es "reducido" si se cumple:

3. Los únicos múltiplos de la raíz $\alpha$ son ella misma y su opuesta, $-\alpha$

Diremos además que el sistema es "cristalográfico" si se cumple la "condición de integralidad":

4. Si $\alpha, \beta$ son raíces, entonces la proyección de $\beta$ sobre la recta que define $\alpha$ es un múltiplo semientero ó entero de $\alpha$ $\langle\beta, \alpha \rangle:=\frac{2(\alpha,\beta)}{(\alpha, \alpha)}\in \mathbb Z^+\cup \{0\}$

**OJO!:** estos brackets $\langle\cdot, \cdot\rangle$ no son un producto escalar, primero porque solo son lineales en el primer argumento, y segundo porque no es simétrico. Tampoco hay garantía de que sea definido positivo

**Proposición:** si $\alpha\in \Phi$, entonces $-\alpha\in \Phi$ (por cond. 2 de la definición, es un copia y pega)

**Teorema:** si el sistema es cristalográfico, fijada una raíz, las demás raíces solo pueden formar un ángulo con respecto a ella de:

1. $90^\circ$ sin restricción sobre el módulo
2. $60^\circ, 120^\circ$ de igual módulo
3. $45^\circ, 135^\circ$ de ratio $\sqrt 2$
4. $30^\circ, 150^\circ$ de ratio $\sqrt 3$

**Dem:**

Sean dos raíces cualesquiera $\alpha, \beta\in \Phi$. Sabemos que la proyección de una sobre la otra ha de ser un número entero, por la condición de integralidad. La estrategia para esta demostración será tratar de recuperar el ángulo que forman por medio del producto escalar usual, que está codificado en los brackets de proyección. Concretamente:

$$\begin{gather}\langle \beta, \alpha\rangle\langle\alpha, \beta\rangle=2\frac{(\alpha, \beta)}{(\alpha, \alpha)}2\frac{(\beta, \alpha)}{(\beta, \beta)}
\\
=4\frac{(\alpha, \beta)^2}{|\alpha|^2|\beta|^2} = 4\cos^2\theta\in \mathbb Z^+\cup \{0\}\end{gather}$$

Los únicos valores del ángulo formado entre sendas raíces que cumplen esta condición son (además de los triviales, que son 0 y 180, son los citados)$_\blacksquare$

**Def.:** se definen las "raíces positivas" $\Phi^+\subset \Phi$ como aquellas que cumple:

1. Por cada raíz $-\alpha\in \Phi$, su opuesta está contenida en $\Phi^+$
2. Si $\alpha, \beta\in \Phi^+$, entonces $\alpha+\beta\in \Phi^+$

De estas propiedades se deduce que las raíces positivas son la mitad de las raíces, y que se puede construir como "todas aquellas raíces que quedan a un lado de un plano que pasa por el origen y que no contiene ninguna raíz"

**Def.:** se define una **raíz simple** como aquella raíz positiva que no se puede expresar como suma de cualesquiera dos raíces positivas. El conjunto de raíces simples se llama "base de $\Phi$" y se denota $\Delta$

**Propiedades:**

1. $\Delta$ es base de $E$
2. Todas las raíces se pueden expresar como combinaciones lineales de la base con coeficientes enteros, y fijada una raíz, estos han de ser todos no negativos o todos no positivos
3. Fijada una raíz simple, si existe otra, estará formando un ángulo de $120^\circ, 135^\circ, 150^\circ$ con ella, no busques ángulos pequeños

### ¿Qué tiene que ver la representación adjunta con sistemas de raíces?

**Teorema:** los vectores de pesos de la representación adjunta forman un sistema de raíces reducido y cristalográfico

**Nota:** si encuentras un número de raíces simples igual al rango del álgebra, tienes una base del sistema de raíces y ya has acabado y puedes construir los demás geométricamente sin matarte

**Teorema:** podemos escribir el álgebra semisimple como suma directa de su subálgebra de Cartan y los subespacios propios de cada autovalor de la base de $ad(\mathfrak h)$:

$$\boxed{\mathfrak g = \mathfrak h\oplus \bigoplus_{\alpha \in \Phi}\mathfrak g_\alpha}$$

donde $\mathfrak g_\alpha :=\{x\in \mathfrak g \ | \ ad(\mathfrak h)x=\alpha(h)x \ \forall h\in \mathfrak h\}$

**Lema:** todos los subespacios propios son de dimensión 1: $\boxed{dim(\mathfrak g_\alpha)=1, \forall \alpha\in \Phi}$, y por ende $\boxed{dim(\mathfrak g)=r+\#\Phi}$



## Diagramas de Dynkin

- Los diagramas de Dynkin son una forma gráfica de representar el sistema de raíces del álgebra, y como tal de clasificarlo

**Construcción a partir de raíces simples positivas:**

1. Pon un nodo por cada raíz simple positiva
2. Conecta cada par de nodos dependiendo del ángulo que formen:
    1. Ortogonales: no conectar
    2. $120^\circ$: 1 nexo
    3. $135^\circ$: 2 nexos
    4. $150^\circ$: 3 nexos
3. En el nexo que une cada par de nodos, lo hacemos direccional, apuntando al nodo de menor módulo. Si los módulos son iguales, entonces ese nexo es adireccional

**Fíjate:** como las raíces de un álgebra forman un sistema de raíces cristalográfico, los únicos ángulos contemplados en la construcción de los diagramas de Dynkin son precisamente los que pueden formar dos raíces entre sí

### Ejemplo: raíces de $A_1$ ($\mathfrak{sl}(2, \mathbb C)$)

**Nota:** como $\mathfrak{su(2)}$ es la forma real compacta de $A_1$, esto que hagamos también servirá para estudiar este otro álgebra

Como $A_1$ es subálgebra del tipo especial lineal, sabemos que la dimensión de su subálgebra de Cartan es $r=n-1$, de modo que sabemos que estamos tratando con $\mathfrak{sl}(2, \mathbb C)$

**OJO!:** este 2 no significa ni que tenga 2 parámetros ni que tenga dimensión 2 como espacio vectorial (de hecho, tiene dimensión 3 sobre los complejos). Este 2 solo hace alusión a que $\mathfrak{sl}(2, \mathbb C)$ es el álgebra de Lie asociada al grupo de matrices de dimensión 2: $SL(2, \mathbb C)$.

Como se trata de un álgebra sobre los complejos de dimensión 3, tengo que dar una base de 3 elementos, y unas relaciones de conmutación entre ellos. Escogiendo la base como $B:=\{h, e, f\}$ donde $h\in \mathfrak h$, tenemos las relaciones de conmutación:

$$\begin{gather}[h, e]=2e \\ [h, f]=-2f \\ [e, f]=2h\end{gather}$$

Ahora queremos construir las raíces. Naturalmente, hemos hecho una buena elección de la base para este propósito, porque aquí ya se ve directamente la aplicación de $ad(\mathfrak h)$ sobre los demás vectores de la base. De aquí sacamos entonces las raíces, fijando un vector de la base cada vez y escribiendo en una tupla sus autovalores correspondientes a cada $ad(h)$. Como solo hay uno, entonces obtendremos números. Concretamente:

$$\begin{gather}\alpha_e=2 \\ \alpha_f=-2\end{gather}$$

El sistema de raíces serán sencillamente un vector de módulo $2$ y otro de módulo $-2$ en una recta. Véase que este es el único sistema de raíces que se puede construir para rango 1

Como solo hay una raíz simple, el diagrama de Dynkin será sencillamente un nodo

```python
import matplotlib.pyplot as plt

# --- Crear figura ---
fig, ax = plt.subplots(figsize=(3,3))

# Coordenadas del nodo
x, y = 0, 0

# Dibujar nodo
ax.plot(x, y, 'ko', markersize=12)


# Ajustes visuales
ax.set_xlim(-0.5, 0.5)
ax.set_ylim(-0.5, 0.5)
ax.set_aspect('equal', adjustable='box')
ax.axis('off')  # quitar ejes
ax.set_title("Diagrama de Dynkin de $A_1$")

plt.show()
```

### Ejemplo: raíces de $A_2$ ($\mathfrak{sl}(3, \mathbb C)$)

**Nota:** como la forma real compacta de $A_2$ es $\mathfrak{su}(3)$, su sistema de raíces será el mismo que el que vamos a calcular ahora

$\mathfrak{sl}(3, \mathbb C)$ tiene dimensión $8$ sobre los complejos, de modo que necesitaremos 8 elementos de la base. De nuevo, vemos que facilita mucho el trabajo de sacar raíces el coger una base del subálgebra de Cartan como parte de la base del álgebra, de modo que escogemos la base: $B:=\{h_1, h_2, e_\alpha, e_\beta, e_{-\alpha}, e_{-\beta}, e_{\alpha+\beta}, e_{-(\alpha+\beta)}\}$ donde $B_{\mathfrak h}:=\{h_1, h_2\}$ y donde lógicamente la notación de la base de $A_2$ ya denota cómo van a ser los vectores de raíces. Las relaciones de conmutación que nos atañen son

$$\begin{gather}[h_1, e_\alpha]=2e_\alpha \\ [h_2, e_\alpha]=-e_\alpha \\ [h_1, e_\beta]=-e_\beta \\ [h_2, e_\beta]=2e_\beta \\ \vdots\end{gather}$$

Ya tenemos todo lo que necesitamos para construir las raíces. El primer vector de raíces es $\alpha=(2, -1)$ y el segundo es $\beta=(-1, 2)$. Podríamos continuar extrayendo vectores de raíces, pero se comprueba que estas dos ya son simples, y como el número de raíces simples es igual a la dimensión del subálgebra de Cartan, el resto de raíces las podemos construir geométricamente (sumas y reflexiones). Finalmente, tenemos que el sistema de raíces de $A_2$ es un hexágono regular:

```python
import matplotlib.pyplot as plt
import numpy as np

# --- Datos de las raíces de A2 ---
angles = np.linspace(0, 2*np.pi, 6, endpoint=False)
x = np.cos(angles)
y = np.sin(angles)

# --- Crear figura con dos subplots ---
fig, axes = plt.subplots(1, 2, figsize=(10,5))

# --- Primer subplot: raíces como vectores ---
ax1 = axes[0]
for xi, yi in zip(x, y):
    ax1.arrow(0, 0, xi, yi, head_width=0.1, head_length=0.15, fc='b', ec='b')

# Etiquetas α y β
ax1.text(1.2, 0, r'$\alpha$', fontsize=14, color='r', ha='center', va='center')
ax1.text(-0.6, np.sqrt(3)/2 + 0.1, r'$\beta$', fontsize=14, color='r', ha='center', va='bottom')

ax1.axhline(0, color='k', linestyle='--', linewidth=0.5)
ax1.axvline(0, color='k', linestyle='--', linewidth=0.5)
ax1.set_aspect('equal', adjustable='box')
ax1.set_xlim(-1.5, 1.5)
ax1.set_ylim(-1.5, 1.5)
ax1.set_title("Raíces de A2")

# --- Segundo subplot: diagrama de Dynkin de A2 ---
ax2 = axes[1]
# Coordenadas de los nodos
nodes_x = [0, 1]
nodes_y = [0, 0]

# Dibujar nodos
for xi, yi in zip(nodes_x, nodes_y):
    ax2.plot(xi, yi, 'ko', markersize=12)

# Dibujar línea que une los nodos
ax2.plot(nodes_x, nodes_y, 'k-', linewidth=2)



ax2.set_xlim(-0.5, 1.5)
ax2.set_ylim(-0.5, 0.5)
ax2.set_aspect('equal', adjustable='box')
ax2.axis('off')  # quitar ejes
ax2.set_title("Diagrama de Dynkin de A2")

plt.tight_layout()
plt.show()
```

### Ejemplo: otras representaciones de $\mathfrak{su}(3)$

- La representación "quark" o sencillamente "3" de este álgebra de Lie viene dada por el peso máximo $(1,0)$. Calculemos los demás:

$$(1,0)\xrightarrow[]{-\alpha_1} (-1,1) \xrightarrow[]{-\alpha_2}(0,-1)$$

**Nota:** véase que claramente esto ya no es un sistema de raíces. Obviamente, no tenía por qué serlo, porque es una representación, pero no la adjunta

Si continuamos aplicando operadores de bajada, llegaremos a que uno de los dos números de la tupla siempre excede $1$ en valor absoluto, pero eso no puede ser porque sabemos que ese es el autovalor máximo. Por ende, hemos acabado. Si dibujamos el "diagrama de pesos" (en analogía al sistema de raíces) veremos que forman un triángulo equilátero, llamado "triángulo de Gelfand-Tsetlin", que es precisamente el que representa los 3 quarks ligeros


- Otra representación es "$3^*$", que tiene como peso máximo $(0,1)$:

$$(0,1)\xrightarrow[]{-\alpha_1} (1,-1) \xrightarrow[]{-\alpha_2}(-1,0)$$

**Nota:** véase que esta representación es independiente de la anterior

- Finalmente, tenemos el "eightfold way", una representación de $\mathfrak{su}(3)$ que tiene como peso máximo $(1,1)$. Tenemos que los pesos son:

```python
import matplotlib.pyplot as plt
import networkx as nx

# Pesos de la representación (1,1)
weights = {
    "1,1": (0, 3),
    "-1,2": (-1, 2),
    "2,-1": (1, 2),
    "0,0": (0, 1),
    "-2,1": (-1, 0),
    "1,-2": (1, 0),
    "-1,-1": (0, -1),
}

# Conexiones al restar raíces
edges = [
    ("1,1", "-1,2"),
    ("1,1", "2,-1"),
    ("-1,2", "0,0"),
    ("2,-1", "0,0"),
    ("0,0", "-2,1"),
    ("0,0", "1,-2"),
    ("-2,1", "-1,-1"),
    ("1,-2", "-1,-1")
]

# Crear grafo
G = nx.DiGraph()
for w in weights:
    G.add_node(w, pos=weights[w])
G.add_edges_from(edges)

pos = weights

plt.figure(figsize=(4,10))   # <--- más alto

# Dibujar nodos
nx.draw_networkx_nodes(G, pos, node_size=200, node_color="lightblue")

# Dibujar flechas
nx.draw_networkx_edges(G, pos, arrowstyle="->", arrowsize=20)

# Dibujar etiquetas desplazadas
label_pos = {k:(v[0], v[1]+0.25) for k,v in pos.items()}  # <--- desplazamiento arriba
nx.draw_networkx_labels(G, label_pos, font_size=12)

plt.axis("off")
plt.show()
```

**Nota:** si representamos el diagrama de pesos, veremos que tenemos un hexágono con 2 puntos en el origen (multiplicidad geométrica del autovalor $0$ es 2, la del resto 1). Este es precisamente la organización de los mesones / bariones más ligeros cuando los dibujamos en ejes de carga y extrañeza. Esto sirvió para revelar que los hadrones tienen estructura, y fue lo que revivió la propuesta de los quarks.

# 3. Grupos de matrices

- De entre todos los grupos, los que son pertinentes para la Física son los de matrices. Estudiemos estos:

```python
import networkx as nx
import matplotlib.pyplot as plt

# Crear grafo no dirigido (sin flechas)
G = nx.Graph()

# Nodos
nodes = [
    "GL(n)", "SL(n)", "Sp",
    "O(p,q)", "O(n)", "SO(p,q)", "SO(n)",
    "U(p,q)", "U(n)", "SU(p,q)", "SU(n)"
]

G.add_nodes_from(nodes)

# Conexiones jerárquicas (sin flechas)
edges = [
    # Nivel 1 -> Nivel 2
    ("GL(n)", "SL(n)"),
    ("GL(n)", "O(p,q)"),
    ("GL(n)", "U(p,q)"),
    
    # Simpléctico
    ("SL(n)", "Sp"),
    
    # Pseudo-ortogonal
    ("O(p,q)", "O(n)"),
    ("O(p,q)", "SO(p,q)"),
    ("O(n)", "SO(n)"),
    ("SO(n)", "SO(p,q)"),
    ("SO(p,q)", "SL(n)"),
    
    # Pseudo-unitario
    ("U(p,q)", "U(n)"),
    ("U(p,q)", "SU(p,q)"),
    ("U(n)", "SU(n)"),
    ("SU(n)", "SU(p,q)"),
    ("SU(p,q)", "SL(n)")
]

G.add_edges_from(edges)

# Posiciones fijas para layout vertical y simetría horizontal
pos = {
    "GL(n)": (0, 8),
    "SL(n)": (0, 7),
    "Sp": (0, 6),
    
    "O(p,q)": (-3, 7),
    "O(n)": (-4, 5),
    "SO(p,q)": (-2, 5),
    "SO(n)": (-3, 4),
    
    "U(p,q)": (3, 7),
    "SU(p,q)": (2, 5),
    "U(n)": (4, 5),
    "SU(n)": (3, 4)
}

# Dibujar el grafo
plt.figure(figsize=(12,9))
nx.draw(G, pos, with_labels=True, node_color='lightblue', node_size=2500, font_size=10, edge_color='gray')
plt.title("Jerarquía de grupos de matrices")
plt.axis('off')
plt.show()
```

**Proposición:** un conjunto de matrices con parámetros reales o complejos forma una variedad sobre el cuerpo correspondiente. Además, se puede definir una norma como:

$$||A||=\sqrt{\displaystyle\sum_{i,j}^n|a_{ij}|^2}$$

De modo que es un espacio de Banach. Si además las matrices son invertibles, entonces tenemos un grupo de Lie

## $GL(n, \mathbb K)$

**Def.:** el "grupo general lineal" es aquel que contiene todas las matrices invertibles de dimensión $n$. Si está definido sobre los reales, tiene $n^2$ parámetros, y si lo está sobre los complejos, $2n^2$. En cualquier caso, siempre trabajaremos con parámetros reales para no lidiar con variedades complejas, aunque esto implique duplicar el número de parámetros

**Condición:** $det M_n\neq 0$

**Álgebra de Lie:** $\mathfrak{gl}(n, \mathbb K)$, álgebra de todas las matrices de dimensión $n$ (invertibles o no)


## $SL(n, \mathbb K)$

**Def.:** el "grupo especial lineal" es aquel que contiene todas las matrices de dimensión $n$ invertibles que además tienen determinante +1

**Condición:** $detM_n=1$

**Álgebra de Lie:** $\mathfrak{sl}(n, \mathbb K)$, álgebra de todas las matrices de dimensión $n$ de traza nula

**Proposición:** $SL(n,\mathbb R)$ es cerrado en $GL(n,\mathbb R)$

## $O(n)$

**Def.:** el "grupo ortogonal" es aquel que contiene todas las matrices **reales** de dimensión $n$ tales ortogonales

**Condición:** $M M^t=I$

**Álgebra de Lie:** $\mathfrak{o}(n)$, álgebra de todas las matrices **reales y antisimétricas** de dimensión $n$ ($X^t=-X$)

## $SO(n)=O(n)\cap SL(n,\mathbb R)$

**Def.:** el "grupo especial ortogonal" es aquel grupo de matrices ortogonales que además tienen determinante 1

**Condición:** $MM^t=I; \ \ detM=1$

**Álgebra de Lie:** $\mathfrak{so(n)}$, álgebra de todas las matrices reales y antisimétricas de **traza nula** de dimensión $n$

## $O(p,q)$

**Def.:** el "grupo pseudo-ortogonal" de signatura $(p,q)$ es el grupo de todas las matrices reales que dejan invariante una matriz de la forma: $\eta=\begin{pmatrix}I_p & 0 \\ 0 & -I_q\end{pmatrix}$. El grupo ortogonal es el caso particular en el que $p=n, q=0$

**Condición:** $M^t\eta M=\eta$

**Álgebra de Lie:** $\mathfrak{o}(p,q)$, matrices $X$ tales que: $X^tK+KX=0$


## $SO(p,q)=O(p,q)\cap SL(n,\mathbb R)$

**Def.:** el "grupo especial pseudo-ortogonal" es aquel grupo de matrices ortogonales que además tienen determinante 1

**Condición:** $M^t\eta M=\eta; \ \ detM=1$

**Álgebra de Lie:** $\mathfrak{so(p,q)}$, matrices $X$ tales que $X^t K +KX=0$ y de traza nula

**Nota:** el grupo de Lorentz propio ortocrono es un subgrupo de $SO(1,3)$ donde la primera componente es mayor o igual que 1

## $Sp(2n)$

**Def.:** el "grupo simpléctico" es aquel grupo de matrices que dejan invariante la matriz simpléctica $\Omega=\begin{pmatrix}0 & -I_n \\ I_n & 0\end{pmatrix}$

**Condición:** $M^t\Omega M=\Omega$

**Álgebra de Lie:** $\mathfrak{so(p,q)}$, matrices $X$ tales que $X^t \Omega +\Omega X=0$ y de traza nula

**Nota:** el grupo de Lorentz propio ortocrono es un subgrupo de $SO(1,3)$ donde la primera componente es mayor o igual que 1

## Diferencial de traslación a izquierdas es fácil en grupos de matrices

**Lema:** la diferencial de traslación a izquierdas de un vector tangente al elemento neutro por una matriz es precisamente la aplicación de la matriz a dicho vector:

$$\boxed{DL_UX_e=UX_e; \ X_e\in T_e(G)}$$

## Campos fundamentales

**Def.:** un "grupo de transformaciones" es aquel que tiene definida una operación que, a cada elemento del grupo, le asocia un difeomorfismo sobre una variedad $M$:

$$\phi: G\times M\to M$$
$$(g,p) \mapsto \sigma_g(p)$$

que además cumpla $\sigma_g\odot \sigma_{g'}=\sigma_{gg'}$ (esto no lo convierte en un homomorfismo, porque $\sigma_g:M\to M$, pero $M$ no tiene ninguna operación interna que haya que respetar, es solo un conjunto de puntos o, a lo sumo, una variedad. La composición en $g$ es algo que pedimos adicionalmente)

En grupos de matrices, las transformaciones son especialmente sencillas, porque **la acción natural de un grupo de matrices sobre $M=\mathbb R^n$ es aplicar la propia matriz** :

$$\sigma_g(x)=Ax$$

**OJO!:** así definido, es imperativo que **la dimensión de las matrices sea igual a la dimensión de la variedad** (porque estamos considerando la acción natural)

**Def.:** sea un subgrupo uniparamétrico en un grupo de transformaciones $G$ tal que $\gamma(t)=exp(tX)$. Naturalmente, $\sigma_{\gamma(t)}(p_0)$ es una curva en $M$ que pasa por el punto $p_0\in M$. Se define el "campo fundamental" en $p_0$, denotado, $X^*(p_0)$, como el vector tangente sobre $p_0$ a la curva así construida. Lógicamente, si repetimos este proceso para cada $p_0$, construimos un campo vectorial:

$$\boxed{X^*(p):=\frac{d}{dt}\Big |_{t=0}\phi(exp(tX), p)}$$

**Interpretación:**

1. Hago una curva en el grupo, $\gamma_G(t)=exp(tX)$ con $X$ algún elemento del álgebra de Lie
2. Como a cada elemento del grupo le corresponde un difeomorfismo sobre $M$, construyo esa familia de difeomorfismos y se la aplico a un punto $p_0\in M$, generando así la curva correspondiente en $M$, parametrizada con el mismo parámetro $\gamma_M(t)=\phi(exp(tX), p_0)$. Véase que lo que hemos conseguido es poner en correspondencia puntos de $G$ con puntos de $M$ mediante la acción del grupo
3. Defino el campo fundamental sobre $p_0$ como el vector tangente de esta curva en $p_0$. La gracia es que, en este mismo punto de la parametrización ($t=0$), el vector tangente a nuestra curva en el grupo sería el vector del álgebra $X$, de modo que hemos conseguido poner en correspondencia vectores del álgebra del grupo con vectores tangentes a la variedad

**Teorema:** si la acción es efectiva, entonces el álgebra de Lie del grupo es isomorfa al álgebra de Lie generada por los campos fundamentales

**Recuerda:** una acción se dice "fiel / efectiva" si para todo punto existe al menos un elemento del grupo que lo transforma de manera no trivial

$$\forall p \in M, \ \exists g \in G \ tq. \ g\cdot p\neq p$$

**Forma práctica de hallar campos fundamentales (grupos generales):** literalmente es aplicar la definición: si tu grupo tiene $m$ parámetros, tendrás que generar las $m$ curvas correspondientes en $\mathbb R^n$ de acuerdo a la definición y luego sacar sus vectores tangentes en el origen de la parametrización. Para afianzar el concepto, es mejor poner el subíndice $0$ en el punto transformado a la hora de construir la curva en $\mathbb R^n$, pero luego recuerda sistituir por las variables de la variedad porque el punto $p_0$ es arbitrario. Así has pasado de tener vectores a tener campos vectoriales: los campos fundamentales

### Campos fundamentales en grupos de matrices

- En la práctica, en grupos de matrices no hace falta hacer toda esta correspondencia entre curvas. Podemos obtener los campos fundamentales sencillamente aplicando el vector del álgebra de Lie correspondiente directamente sobre los puntos de la variedad:

$$\boxed{X^*=\displaystyle\sum_i \xi_i(x)\partial_i; \ \ \xi_i(x)=(Xx)_i}$$

**Propiedad:** sean $V^*, W^*$ dos campos fundamentales, generados respectivamente por $X, Y\in \mathfrak g$. Entonces, el conmutador de los campos fundamentales viene dado por el de los generadores:


$$\boxed{[V, W]=-[X, Y]_{kl}x_l\partial_k}$$

Si se quiere, podemos definir un mapa que a cada vector del álgebra de Lie del grupo le asocia un campo fundamental sobre la variedad:

$$\phi: \mathfrak g\to \mathfrak X^*(M)$$
$$X\mapsto X^*=X_{kl}x_l\partial_k$$

Entonces, esta propiedad se puede expresar como: "si $X\mapsto V$ y $Y\mapsto W$, entonces $[X, Y]\mapsto -[V, W]$ 

### Ejemplo: campos fundamentales de $SO(3)$ sobre $\mathbb R^3$

**Forma general:**

- Tracemos por ejemplo la curva en $SO(3)$ correspondiente al generador de rotaciones sobre el eje Z, que es: $M(\alpha)=exp(\alpha J_z)=\begin{pmatrix}1 & 0 & 0 \\ 0 &\cos\alpha & -\sin\alpha \\ 0 & \sin\alpha & \cos\alpha\end{pmatrix}$

Ahora pongamos esta curva en correspondencia con la curva que genera para un punto $x_0\in \mathbb R^3$ (i.e: estudiamos la órbita de dicho punto bajo este grupo uniparamétrico de transformaciones que hemos construido)

$\gamma(\alpha)=Mx_0=\begin{pmatrix}1 & 0 & 0 \\ 0 &\cos\alpha & -\sin\alpha \\ 0 & \sin\alpha & \cos\alpha\end{pmatrix} \begin{pmatrix}x_0 \\ y_0 \\ y_0\end{pmatrix} \longrightarrow \begin{cases}x'=x_0 \\ y'=y_0\cos\alpha -z_0\sin\alpha \\ z'=y_0\sin\alpha+z_0\cos\alpha\end{cases}$

Esto nos define una curva parametrizada sobre la variedad, y según la definición, debemos hallar el vector tangente en el origen de la parametrización. Esto se hace derivando con respecto al parámetro y evaluándolo en 0 (o en el valor del parámetro que nos dé el elemento identidad en el grupo):

$$X^*_1=-z_0\partial_y + y_0\partial_z$$

Como podemos repetir este procedimiento en cualquier punto de la variedad, concluimos que el primer campo fundamental es:

$$\boxed{X^*_1=-z\partial_y+y\partial_z}$$

Con los otros dos se procede exactamente igual

**Sirviéndonos de que $SO(3)$ es grupo de matrices:**

Hallamos el vector del álgebra de Lie del grupo que genera esta curva: $X=\frac{d}{d\alpha}\Big |_{\alpha=0}M(\alpha)=\begin{pmatrix}0 & 0 & 0 \\ 0 &0 & -1 \\ 0 & 1 & 0\end{pmatrix}$. Entonces, el campo fundamental asociado a este generador será:

$$X^*_1=X_1 x=\begin{pmatrix}0 \\ -z \\ y\end{pmatrix}\equiv -z\partial_y + y\partial_z$$

## Invariantes de un grupo de transformaciones

**Def.:** sea $G$ grupo de transformaciones sobre una variedad $M$, y sea $\mathbf T$ un campo tensorial definido sobre $M$ (tensorial porque se transforma adecuadamente bajo la acción de $G$) si su derivada de Lie con respecto a todos los campos fundamentales asociados al grupo es 0:

$$\mathbf T \text{ invariante bajo G } \Longleftrightarrow\mathfrak L_{X^*_i}\mathbf T=0$$

## Medida de Haar

**Motivación:** sabemos diferenciar los elementos del grupo, pero no sabemos integrarlos. Además, en grupos finitos, sabemos que en el álgebra del grupo (y en el grupo como tal si la función está definida en el grupo) tienen sentido expresiones como $\displaystyle\sum_{g\in G}f(g)$, y que, como el grupo es finito, estas cumplen la propiedad de que $\displaystyle\sum_{g\in G}f(g)=\displaystyle\sum_{g'\in G}f(gg')$, es decir, son invariantes a izquierdas / derechas. Es por eso que parece pertinente definir una suma (integral) sobre grupos de Lie que sea asimismo invariante a izquierdas / derechas

**Def.:** sea $f(g):G\to\mathbb R$ una función definida sobre el grupo (si se quiere, sobre la variedad subyacente). Se define la "medida de Haar" a izquierda como aquella medida $d\mu(g)$ definida sobre la variedad tal que la integral de una función trasladada a izquierdas sea igual a la integral de la función original

$$\int_G f(g'g)d\mu(g)=\int_G f(g)d\mu(g)$$

**Proposición:** si $G$  es grupo compacto, entonces la medida de Haar es finita: $\int_G d\mu(g)<\infty$

**Teorema de Haar:** todo grupo de Lie localmente compacto tiene una medida de Haar y esta es única salvo factor multiplicativo (solo hay una forma de integrar en el grupo)

**Ejemplo:** en $U(1)$, la medida de Haar es $d\mu(\varphi)=\frac{1}{2\pi}d\varphi$. Se comprueba que es invariante a izquierdas y derechas (que en este caso son lo mismo porque el grupo es abeliano). Tenemos que:

$$\int_0^{2\pi} f(e^{i\varphi})\frac{d\varphi}{2\pi}=\int_0^{2\pi}f(e^{i(\varphi'+\varphi)})\frac{d\varphi}{2\pi}$$

Porque, haciendo un cambio de variable en la segunda integral tal que $\tilde \varphi=\varphi'+\varphi$, tenemos que es $\int_{\varphi'}^{2\pi+\varphi'}f(\tilde \varphi)\frac{d\tilde \varphi}{2\pi}$, pero como las funciones definidas sobre $U(1)$ (en el fondo, las funciones sobre la variedad $S^1$) son periódicas, tenemos que las dos integrales de la ecuación valen lo mismo, porque lo único que cambia es que estamos empezando y acabando de integrar en puntos diferentes de la circunferencia, pero siempre recorremos exactamente una vez la misma, de modo que la integral, por la propiedad conmutativa de la suma, da lo mismo

# 4. $SO(3)$ y $SU(2)$

## $SO(3)$

### Parametrizaciones

**Proposición:** $SO(3)$ tiene dimensión 3

**Parametrización 1:** versor de rotación + cantidad rotada en torno a dicho eje

**Parametrización 2:** ángulos de Euler

$$R(\alpha, \beta, \gamma)=R_z(\alpha)R_y(\beta)R_z(\gamma); \ \ \alpha, \gamma\in [0, 2\pi); \ \ \beta\in [0,\pi]$$

### Variedad subyacente

**Proposición:** cuando hacemos una rotación en el espacio euclideo, $1$ siempre es un autovalor, de modo que existe un espacio de dimensión 1 que permanece invariante. Su autovector correspondiente será el eje de rotación

**Variedad subyacente:**

- Por ende, la variedad que subyace a $SO(3)$ es una esfera maciza de radio $\pi$, de modo que cualquier rotación es un punto dentro de dicha variedad: colatitud y acimut fijan la dirección ($\hat n$), y el radio fija la cantidad rotada en torno a dicho eje
- **Fíjate:** es lo mismo girar $\pi$ en torno a un eje que girar $-\pi$ en torno al eje opuesto. Por ende, hay una correspondencia entre las antípodas de la esfera
- **Consecuencia:** la variedad de $SO(3)$ es conexa por arcos, pero NO es simplemente conexa, porque no todas las curvas son homótopas a un punto: un punto sobre la superficie de la esfera es el mismo que su antípoda, y una curva que una a los dos no es homótopa a ningún punto

**Proposición:** $SO(3)$ es compacto y conexo, pero no simplemente conexo

### Medida de Haar

La medida de Haar en $SO(3)$ viene dada en las respectivas parametrizaciones por:

$$d\mu(\alpha, \beta, \gamma)=\frac{1}{8\pi^2}\sin(\beta) d\alpha d\beta d\gamma$$

$$d\mu(\hat n, \theta)=\frac{1}{2\pi^2}\sin^2\left(\frac{\theta}{2}\right)d\theta d\Omega(\hat n)$$

### $SO(3)$ como grupo de transformaciones

- No es transitivo
- No es efectivo (el origen no se ve transformado por ninguna rotación)
- Las órbitas de cualquier punto que no sea el origen son esferas

### Álgebra de Lie

- El álgebra de Lie de $SO(3)$ es $\mathfrak{so}(3)\cong \mathfrak{su}(2)$. Podemos tomar una base $B:=\{J_1, J_2, J_3\}$ con las relaciones de conmutación:

$$[J_i, J_j]=\varepsilon_{ijk}J_k$$

Y las constantes de estructura son entonces 1's, 0's y -1's en esta elección de base

Es muy importante ver que tiene el mismo álgebra de Lie que $SU(2)$, de modo que ambos grupos son localmente isomorfos en un entorno de la identidad. Como $SU(2)$ es la forma real compacta de $A_1$, pero el álgebra es la misma que la de $SO(3)$, tenemos entonces que $\mathfrak{so}(3)$ es semisimple (y por ende sus representaciones finitas reducibles son completamente reducibles), tiene como diagrama de Dynkin un único nodo (una raíz simple), y que posee un subálgebra de Cartan de dimensión 1. De aquí vienen los operadores escalera. Estos vienen de coger la base del álgebra $B:=\{J_3, J_+, J_-\}$. Con esta elección de base:

$$\begin{gather}[J_3, J_+]=2J_+ \\ [J_3, J_-]=-2J_- \\ [J_+, J_-]=J_3\end{gather}$$

Por ende, $h=lin\{J_3\}$, y vemos que las raíces de la representación adjunta serán precisamente $2$ y $-2$, como ya nos salía para $A_1$

## $SU(2)$

### Parametrizaciones

**Proposición:** visto como un grupo de parámetros reales, $SU(2)$ tiene dimensión 3

**Parametrización 1:** $M=\begin{pmatrix}\alpha & \beta \\ -\beta^* & \alpha^*\end{pmatrix}$

**Parametrización 2:** como exponencial de generadores: $M=e^{-\frac{i}{2}\vec \sigma\cdot \hat n}$

**Parametrización 3:** ángulos de Euler:

$$M=e^{\frac{i}{2}\alpha \sigma_z}e^{\frac{i}{2}\beta \sigma_y}e^{\frac{i}{2}\gamma \sigma_z}$$

### Álgebra de Lie

El álgebra de Lie $\mathfrak{su}(2)$ es el mismo (isomorfo) que $\mathfrak{so}(3)$. Sin embargo, los generadores se toman ahora como las matrices de Pauli (porque el grupo es de matrices de dimensión 2). Las relaciones de conmutación en esta base son entonces:

$$\left[\frac{\sigma_i}{2}, \frac{\sigma_j}{2}\right]=i\varepsilon_{ijk}\frac{\sigma_k}{2}$$

Véase que el factor imaginario es un mero convenio

**Nota:** en sí, entiendo que las matrices que tomemos dan igual si estudiamos el álgebra como algo abstracto, es decir, basta con hacer referencia a sus generadores. Sin embargo, si trabajamos con álgebras / grupos de matrices, puede ser bueno coger matrices que tengan las dimensiones especificadas, aunque con esto yo creo que ya estamos escogiendo una representación (aunque esta sea la fundamental)

### Doble recubridor de $SO(3)$

**Teorema:** $SU(2)$ es el doble recubridor de $SO(3)$, es decir, a cada matriz de $SU(2)$ le corresponde una matriz de $SO(3)$, pero a cada matriz de $SO(3)$ le corresponden dos matrices de $SU(2)$, que son una y su opuesta.

Entonces, es de esperar que podamos construir un homomorfismo sobreyectivo tal que a cada elemento de $SU(2)$ le asigne su elemento correspondiente de $SO(3)$. En concreto, vemos que tanto $I$ como $-I$ en $SU(2)$ mapean a $I$ en $SO(3)$. Este es el Kernel de la aplicación. Por el primer teorema de isomorfía: $Im[f]=Dom/Ker[f]$, tenemos:

$$\boxed{SO(3)=SU(2)/Z_2}$$

### Representaciones de $SU(2)$

**Teorema:** solo hay 1 representación irreducible de $SU(2)$ por cada dimensión, y cada dimensión tiene exactamente 1 representación irreducible de $SU(2)$.

**Notación:** la dimensión se suele expresar como $2j+1$, donde $j=0,1/2,1,\ldots$ En el contexto de la Física, $j$ suele recibir el nombre de "spin" de la irrep. de $SU(2)$

**Teorema:** una representación que es producto tensorial de dos irreps. se descompone en suma directa de irreps. de dimensiones desde $|j_1-j_2|$ hasta $j_1+j_2$ en pasos de uno, y cada una aparece solo una vez:

$$j_1\otimes j_2=|j_1-j_2|\oplus \ldots \oplus j_1+j_2$$

Si se quiere expresar vectores en la base de $j_1\otimes j_2$ en términos de vectores de la base adaptada de la representación reducida, se emplean los coeficientes de Clebsch-Gordan

**Proposición:** las representaciones de $SU(2)$ con $j$ entero son también representaciones de $SO(3)$

### Matrices de Wigner y símbolos "$d$"

- En cualquier irrep. de $SU(2)$ podemos parametrizar las matrices de dicho grupo mediante los ángulos de Euler. Las matrices reciben entonces el nombre de "matrices de Wigner":

$$D^j(R)=\langle l|e^{i\alpha J_3}e^{i\beta J_2}e^{i\gamma J_3}|l'\rangle$$

Sin embargo, como la tercera componente de momento angular se toma autovalor de $J_3$, podemos tomar elementos de matriz y darnos cuenta de que dos de las 3 exponenciales solo sacan un factor de fase:

$$\boxed{D^j_{mm'}(R)=\langle jm'|D^j(R)|jm\rangle = e^{i(\alpha m'+\gamma m)}d^j_{mm'}(\beta)}$$

donde $d^j_{mm'}(\beta)=\langle jm'|e^{i\beta J_2}|jm\rangle$

## Casimires

- Como $\mathfrak{su(2)}$ tiene rango 1, existe un único operador de Casimir:

$$\boxed{J^2=J_1^2+J_2^2+J_3^2}$$

- Al aplicarse sobre vectores de una irrep., tenemos $J^2|lm\rangle=l(l+1)|lm\rangle$
