# 1. Representaciones: introducción y definiciones

## Definición

**Def.:** sea un grupo $G$ y un espacio vectorial $V$. Entonces, una "representación de $G$ en $V$" es un **homomorfismo** que a cada elemento del grupo le asocia un **endomorfismo regular** sobre $V$:

$$\rho: G\to \text{End}_r(V)$$
$$g\mapsto \varphi_g: V\to V$$

**Def.:** si la representación $\rho$ es inyectiva (a distintos elementos de grupo le corresponden distintos endomorfismos), entonces diremos que es **fiel**

- Las aplicaciones que actúan sobre los elementos de un espacio vectorial de dimensión finita (en particular endomorfismos) se pueden representar con matrices, previa elección de la base

## Equivalencia de dos representaciones

**Def.:** sean $\rho_1, \rho_2$ dos representaciones de un grupo $G$. Diremos que son equivalentes si existe un isomorfismo $T:V_1\to V_2$ entre sus espacios vectoriales:

$$\rho_1\sim\rho_2 \Longleftrightarrow V_1\cong V_2$$

que además sea un morfismo de representaciones: $\rho_1(g)=T^{-1}\rho_2(g)T$

**Más fácil:** dos representaciones son equivalentes si sus matrices solo difieren en un cambio de base (el mismo siempre, lógicamente)

### Def. en lenguaje de matrices

- Dos representaciones son equivalentes si, para cada elemento del grupo, las matrices de ambas representaciones son semejantes, siempre a través de la misma matriz $T$:

$$\rho_1(g)=T^{-1}\rho_2(g)T$$

## Representación identidad

**Def.:** la "representación identidad" de $G$ sobre el espacio vectorial $\mathbb C$ (sobre el cuerpo $\mathbb C$) es aquella que a todo elemento del grupo le asocia el endomorfismo identidad:

$$\rho: G\to \text{End}_G(\mathbb C)$$
$$g\mapsto 1$$

### Proposición (no-equivalencia de la identidad)

- La representación identidad de un grupo no es equivalente a ninguna otra representación de dicho grupo

**Dem.:** supongamos que esta afirmación es falsa. Entonces, existe al menos una representación de $G$ tal que $\rho(g)=P\rho_I(g)P^{-1}; \ \forall g\in G$, pero como la representación identidad asigna precisamente el operador identidad a todos los elementos del grupo: $\rho(g)=PP^{-1}=I \Longrightarrow \rho(g)=\rho_I(g)_\blacksquare$

## Representación unitaria

**Def.:** sea $G$ grupo y $(\rho, V)$ representación de $G$ con producto escalar definido en $V$. Diremos que esta representación es "unitaria" si los endomorfismos son matrices unitarias:

$$\langle \rho(g)v, \rho(g)w\rangle = \langle v, w\rangle$$

### Teorema (equivalencia a representaciones unitarias)

- En un grupo finito, todas las representaciones son equivalentes a representaciones unitarias

## Representación fundamental

**Def.:** sea $G$ un grupo de matrices. Entonces, la "representación fundamental" de $G$ es aquella que a cada elemento de grupo (matriz) le asigna un endomorfismo tal que su matriz asociada coincide con la matriz del grupo que estamos representando

$$\rho_{fund}:G(M_n)\to V_n$$
$$M_n \mapsto \varphi_g \equiv M_n$$

**OJO!:** esto ya fija la elección de una base en $V$ de tal forma que las matrices del grupo coincidan con su representación. También fija la dimensión del espacio vectorial, que debe ser de la misma dimensión que las matrices del grupo

# 2. Operaciones sobre representaciones

- Definamos dos operaciones sobre las representaciones que nos permitan combinar dos para dar una nueva

## 2.1 Suma directa

**Def.:** sea $G$ grupo y $(\rho_1, V_1), (\rho_2, V_2)$ dos representaciones de $G$. La "suma directa" de las dos representaciones es una nueva representación, cuyo espacio vectorial es $V_T=V\oplus V'$ y cuyo homomorfismo es tal que:

$$\boxed{\rho_T(g)(v_1, v_2)=(\rho_1(g)v_1, \rho_2(g)v_2)}$$

**Fíjate:** puedo escoger una base de $V_1\oplus V_2$ tal que los $n_1$ primeros elementos de la base sean los de la base de $V$ y los últimos $n_2$ elementos de la base sean los de la base de $V_2$. Entonces, las matrices de la representación $\rho_T$ quedan diagonales por bloques

**Nota:** $dim(V_1\oplus V_2)=dim(V_1)+dim(V_2)$

## 2.2 Producto tensorial

**Def.:** sea $G$ grupo y $(\rho_1, V_1), (\rho_2, V_2)$ dos representaciones de $G$. El "producto tensorial" de las dos representaciones es una nueva representación, cuyo especio vectorial es $V_T=V_1\otimes V_2$ y cuyo homomorfismo es tal que:

$$\boxed{\rho_T(g)(v_1\otimes v_2) = (\rho_1(g)v_1)\otimes (\rho_2(g)v_2)}$$

**Fíjate:** ahora la acción de los endomorfismos se mezcla, mientras que antes permanecían separados.

**Matrices de los endomorfismos:** si tomamos la base (ordenada como sigue) $B_T=B_1\otimes B_2=\{u_1\otimes u_1', \ldots, u_1\otimes u_{n_2}', \ldots, u_{n_1}\otimes u_{n_2}\}$ tenemos que, en dicha base, la matriz del endomorfismo asociado a $g$ en la representación $\rho_1\otimes \rho_2$ es sencillamente el producto de Kronecker de las dos matrices correspondientes:

$$(\rho_1\otimes \rho_2)[g]=\rho_1(g)\otimes \rho_2(g)$$

# 3. Representaciones derivadas de otras

**Nota:** las representaciones que se deriven de otra representación pueden o no ser equivalentes a esta

## Representación Dual

**Idea:** una representación no es más que una regla que me coge un elemento de un grupo y le asocia un endomorfismo regular que actúa sobre elementos $V$ espacio vectorial. Sin embargo, también podría pedir que me asocie un endomorfismo regular que actúe sobre el espacio dual de $V$:

**Def.:** la "representación dual" de $G$ es un homomorfismo que a cada elemento de $G$ le asocia un endomorfismo regular sobre el espacio dual del espacio vectorial $V$:

$$\rho^t: G\to \text{End}_r(V^*)$$
$$g\mapsto \varphi_g: V^*\to V^*$$

La acción de estos endomorfismos se puede relacionar con la de los endomorfismos de la representación "ordinaria" como:

$$(\rho^t(g)\omega)v=\omega (\rho(g^{-1})v); \ \ v\in V, \omega \in V^*$$

## Representación Adjunta

**Def.:** sea $G$ grupo y $V$ espacio vectorial con producto escalar definido. Sea representación de $G$ $(\rho, V)$. Entonces, se define la "representación adjunta de $\rho$" como:

$$\langle \rho^\dagger (g)v, w\rangle:=\langle v, \rho(g)w\rangle; \ \ \forall v, w\in V$$

## Representación conjugada

**Def.:** sea $(\rho, V)$ representación de $G$, entonces su "representación conjugada" se define como:

$$\bar \rho(g):=[\rho(g)]^*; \ \ \forall g\in G$$

# 4. Representaciones irreducibles

## Representaciones reducibles

**Def.:** sea $G$ grupo y $(\rho, V)$ representación de $G$. Diremos que esta representación es "reducible" si existe algún subespacio de $V$ (no trivial) que sea invariante bajo todos los endomorfismos de la representación:

$$\boxed{\rho \text{ reducible } \Longleftrightarrow \exists W<V \text{ tq. } \rho(g)W = W; \ \ \forall g \in G}$$

- Decimos en este caso que el subespacio vectorial $W$ de $V$ es "G-invariante"

**Def.:** una representación "irreducible" es aquella que no es reducible (el espacio vectorial de la representación no tiene espacios invariantes bajo todos los endomorfismos de la repr.)

- Si tenemos una repr. reducible, podemos encontrar su subespacio vectorial invariante $W$, tomar la base de dicho subespacio como los $r=dim(W)$ primeros elementos de la base de $V$ y después completar el resto de la base (Teorema de ampliación de la base me permite hacerlo)
- En esta base, las matrices asociadas a los endomorfismos sobre $V$ tendrán la forma:

$$\rho(g)=\left(
\begin{array}{c|c}
A_r & B \\ \hline
0 & C_{n-r}
\end{array}
\right)$$

- Comprobamos que $\rho(g)W=W$

## Representaciones completamente reducibles

**Def.:** diremos que una representación es "completamente reducible" si, dado un subespacio vectorial invariante $W$, existe un subespacio complementario $W'$ que también sea invariante. En la base adaptada de estos dos subespacios, cuya unión recupera el espacio total por definición de complementariedad, las matrices de los endomorfismos quedan diagonales por bloques

**Recuerda:** $W, W'$ se dicen "complementarios" si $W\oplus W'=V$

**OJO!:** para que la representación sea irreducible, no basta con que las matrices de los endomorfismos sean diagonales por bloques, sino que los subespacios invariantes a los que hacen alusión no deben contener ningún subespacio invariante. De lo contrario, procedemos de forma recursiva, lo que sencillamente se traducirá en partir los bloques en otros más pequeños

**Fíjate:** si tienes producto escalar definido sobre $V$, entonces puedes coger el complementario como el complemento ortogonal de $W$ y por tanto siempre puedes dejar la matriz diagonal por bloques

# 5. Teoría de Caracteres

## Definición

**Def.:** sea $G$ grupo y sea $(\rho, V)$ representación finita de $G$. Se define el "carácter" del elemento $g$ en esta representación como la traza de su matriz asociada:

$$\boxed{\chi^{(\rho)}(g)=Tr[D_\rho(g)]}$$

**Fíjate:** el carácter de un elemento de $G$ depende de la representación que elijamos (dependiendo de la repr. tendremos asociadas unas matrices u otras, y por ende sus trazas serán diferentes en general)

## Teorema (caracteres son funciones de clase)

- Dos elementos pertenecen a la misma clase de conjugación $\Longrightarrow$ tienen el mismo carácter

**Interpretación:** los caracteres son funciones de clase: podemos pasarles como argumento la clase de conjugación, porque cualquier representante tiene el mismo carácter


**OJO!:** no se cumple necesariamente en el otro sentido, puede que dos clases de conjugación distintas tengan igual carácter

**Dem.:**

- Sea un elemento del grupo obtenido a partir de conjugar otro: $g'=(\tilde g)^{-1}g\tilde g$. Escogida una representación, su carácter es:

$$\chi^{(\rho)}(g')=Tr[D_\rho(g')]=Tr[D^{-1}_\rho(\tilde g)D_\rho(g) D_\rho(\tilde g)]$$

- Pero como la traza es cíclica, tenemos que:

$$\chi^{(\rho)}(g')=Tr[D_\rho(g)]=\chi^{(\rho)}(g)_\blacksquare$$

## Teorema (representación suma directa)

- Si una representación es suma directa de otras $r$ representaciones, entonces los caracteres de esta representación son suma de los caracteres de las representaciones involucradas:

$$\boxed{\rho=\bigoplus_{i=1}^r \rho_i \Longrightarrow \chi^{(\rho)}(g)=\displaystyle\sum_{i=1}^r\chi^{(\rho_i)}(g)}$$

**Dem.:**

- Si el grupo es finito, $\rho$ es completamente reducible y por ende todas las matrices asociadas a los endomorfismos se podrán expresar diagonales por bloques escogiendo la base adaptada a la suma directa. Entonces, la traza de estos endomorfismos será la suma de las trazas de los bloques

**Fíjate:** como en un grupo finito toda representación es completamente reducible (i.e: se puede expresar como suma directa de representaciones irreducibles), basta con estudiar los caracteres de las representaciones irreducibles, y así podremos estudiar los de cualquier representación tan solo sumándolos

## Teorema (representación producto tensorial)

- Si una representación es producto tensorial de otras $r$ representaciones, entonces los caracteres de esta representación son producto de los caracteres de las representaciones involucradas:

$$\boxed{\rho = \bigotimes_{i=1}^r \rho_i \Longrightarrow \chi^{(\rho)}(g)=\displaystyle\prod_{i=1}^r \chi^{(\rho_i)}(g)}$$

**Dem.:**

- Sea una representación de algún grupo $G$ construida como el producto tensorial de otras dos representaciones de dicho grupo. Entonces, la matriz asociada a un endomorfismo en esta representación será el producto de Kronecker de las matrices del mismo endomorfismo en las representaciones 1 y 2 con las que construimos esta. Es decir:

$$\rho = \rho_1\otimes \rho_2 \Longrightarrow D^{(\rho)}(g)=D_{\rho_1}(g)\otimes D_{\rho_2}(g)=\begin{pmatrix}a_{11}D_{\rho_2}(g) & \ldots \\ \ldots & a_{nn}D_{\rho_2}(g) \end{pmatrix}$$

- Tomando ahora la traza:

$$\chi^{(\rho)}=Tr[D_\rho(g)]=a_{11}Tr[D_{\rho_2}(g)]+\ldots + a_{nn}Tr[D_{\rho_2}(g)] = Tr[D_{\rho_2}(g)](a_{11}+\ldots + a_{nn})=Tr[D_{\rho_2}(g)]Tr[D_{\rho_1}(g)]=\chi^{(\rho_2)}\chi^{(\rho_1)}_\blacksquare$$

## Álgebra de un grupo

**Def.:** un "álgebra" sobre cierto cuerpo $\mathbb K$ es una estructura algebraica. Concretamente, es un espacio vectorial $V$ sobre $\mathbb K$ equipado con una forma bilineal $\cdot : V\times V\to V$

**Def.:** dado un grupo $G$, el "álgebra del grupo" es un álgebra asociativa donde: (1) el espacio vectorial es el generado por todas las combinaciones lineales de los elementos del grupo y (2) la forma bilineal definida sobre este espacio vectorial es el producto de grupo:

$$\mathcal A_G = (V, \cdot) \text{ tq. } V=\left\{\displaystyle\sum_i\lambda_i g_i; \  \lambda_i\in \mathbb K, g\in G\right\}; \ \  \cdot =\odot$$

**Nota:** lógicamente, una base del álgebra será precisamente el conjunto de $G$: $\{g_i\}$

**Aclaración:** se dice que un álgebra es "asociativa" si su producto bilineal es asociativo. Esto es, para cualesquiera tres elementos del álgebra $a, b, c\in \mathcal A$, tenemos que: $(a\cdot b)\cdot c=a\cdot(b\cdot c)$. Hay otras álgebras que no son asociativas (por ejemplo las álgebras de Lie, en las que el producto bilineal es un conmutador)

**Aclaración:** el producto de dos elementos del álgebra: $a_1 = \displaystyle\sum_i \lambda_i g_i; \ a_2 = \displaystyle\sum_j \mu_j g_j$ será entonces:

$$a_1\cdot a_2 = \left(\displaystyle\sum_i \lambda_i g_i\right)\cdot \left(\displaystyle\sum_j \mu_j g_j\right)=\displaystyle\sum_{i,j}\lambda_i\mu_jg_i\odot g_j$$

**Aclaración sobre endomorfismos en el álgebra:** si entendemos el álgebra como un espacio vectorial equipado con un producto bilineal, entonces los endomorfismos sobre este espacio vectorial no tienen por qué respetar la estructura del álgebra, en el sentido de que: $\varphi(a_1a_2)\neq \varphi(a_1)\varphi(a_2)$ necesariamente, porque estamos tratando solo el espacio vectorial a la hora de definir los endomorfismos. Por otro lado, podemos considerar el álgebra como una estructura algebraica en sí misma, y entonces definimos los "endomorfismos sobre el álgebra" precisamente pidiendo que sean homomorfismos en dicha estructura algebraica (de modo que sí se cumple esta igualdad)

### Caracteres del álgebra

- Ahora que conocemos el carácter de un elemento de un grupo en cierta representación de dicho grupo, podemos extender el concepto al álgebra del grupo para definir el carácter de un elemento de este álgebra:

$$\boxed{\chi^{(\rho)}(a)=\displaystyle\sum_i\lambda_i\chi^{(\rho)}(g)}$$

### Representación regular

**Def.:** la "representación regular" de un grupo es una representación **concreta** sobre su álgebra de tal forma que a cada elemento del grupo le asocia un endomorfismo regular del álgebra cuya acción es hacer actuar el elemento del grupo por la izquierda, ahora entendido como elemento del álgebra:

$$\rho_R: G\to \text{End}_r(\mathcal A_G)$$
$$g\mapsto \varphi_g: \mathcal A_G\to \mathcal A_G$$

y tal que $\varphi_g(a) = ga$

**Comparar con Tema 3:** si denotamos la aplicación bilineal del álgebra como $(a_1, a_2)$, entonces tenemos que la representación regular es:

$$\rho_R:G\to End_r(\mathcal A_G)$$
$$g\mapsto (g, \cdot)$$

**Ejemplo:** hallamos representación regular del grupo $G=A_3=\{e, (123), (132)\}$

- Primero construimos el álgebra:

$$\mathcal A_{A_3}=lin\{\lambda_1 e + \lambda_2 (123) + \lambda_3(132)\}$$

La base natural es $B=\{e, (123), (132)\}$. El producto del álgebra es el producto de grupo, que en este caso no es más que la composición de permutaciones. Hallemos ahora las matrices de los endomorfismos a los que mapea la representación regular a los elementos de $g$. Para ello, procedemos como siempre, evaluando la acción del elemento en la base concreta:

$$e\{e, (123), (132)\}=\{e, (123), (132)\} \Longrightarrow \rho(e)=I_3$$
$$(123)\{e, (123), (132)\} = \{(123), (132), e\} \Longrightarrow \rho(123)=\begin{pmatrix}0 & 1 & 0 \\ 1 & 0 & 0 \\ 0 & 1 & 0\end{pmatrix}$$
$$(132)\{e, (123), (132)\} = \{(132), e, (123)\}\Longrightarrow \rho(132)=\begin{pmatrix}0 & 1 & 0 \\ 0 & 0 & 1 \\ 1 & 0 & 0\end{pmatrix}$$

### Propiedades de la representación regular

1. Las matrices de sus endomorfismos (en la base natural) solo tienen 1s y 0s, y solo hay un 1 por fila y por columna (esto es lógico, porque si los elementos de la base son los elementos del grupo y como el producto de un elemento del grupo por otro va a dar un tercer elemento del grupo, que a su vez será elemento de la base del álgebra del grupo, es lógico que la acción de los endomorfismos de la representación regular sobre un elemento cualquier de la base sea dar otro elemento de la base y que por tanto solo haya 0s y 1s y solo uno por fila columna. Es como una relación entre elementos de la base)
2. Es reducible
3. Concretamente, contiene todas las representaciones irreducibles del grupo, cada una tantas veces como su dimensión:
   $$\boxed{\rho_R=\bigoplus_{i=1}^r (dim V_i)\rho_i}$$ donde $r$ es el número de irreps. del grupo (coincide con el número de clases de conjugación)
4. El carácter del elemento neutro coincide con el orden del grupo: $\chi^R(e)=|G|$
5. El carácter de cualquier otro elemento del grupo es nulo: $\chi^R(g)=0; \ g\neq r$

**Dem. 4:**

- Tenemos que la matriz del elemento neutro en la representación regular se descompone como suma directa de matrices de las irreps. de la forma:

$$\rho^R(e)=\bigoplus_{i=1}^r(dim V_i)\rho^{(i)}(e)=\bigoplus_{i=1}^rn_iI_{n_i}$$

- Tomando caracteres:

$$\chi^R(e)=\displaystyle\sum n_i^2=|G|_\blacksquare$$


**Dem. 5:**

- Las matrices que representan a un elemento distinto del neutro en la representación regular no tienen ningún 1 en la diagonal, porque eso querría decir que dejan invariantes todos los elementos del grupo, pero esto es imposible a menos que el elemento coincida con el neutro, que hace esto por definición. Por ende, al tomar la traza, se me va a 0$_\blacksquare$

**Ejemplo:** ahora que conocemos las matrices (en la base natural del álgebra del grupo) de la representación regular de $A_3$ y que sabemos que es reducible, pues vamos a reducirla (completamente, porque es grupo finito). Tenemos que hallar subespacio invariante y otro complementario a él. Como $A_3$ es abeliano, sus matrices en cualquier repr. conmutan y por ende se pueden diagonalizar a la vez. Encontramos que las matrices diagonalizadas son:

$$\rho(e) \to \begin{pmatrix}1 & 0 & 0 \\ 0 & 1 & 0 \\ 0 & 0 & 1\end{pmatrix}$$
$$\rho(123)\to \begin{pmatrix}1 & 0 & 0 \\ 0 & \omega & 0 \\ 0 & 0 & \omega^2\end{pmatrix}$$
$$\rho(132)\to \begin{pmatrix}1 & 0 & 0 \\ 0 & \omega^2 & 0 \\ 0 & 0 & \omega\end{pmatrix}$$

donde los subespacios propios son todos de dimensión 1 porque no hay autovalores repetidos (salvo en la identidad, lógicamente). Como ya hemos reducido los subespacios invariantes a dim. 1, hemos acabado seguro (porque no puede haber más subespacios invariantes dentro de ellos por dimensión), de modo que podemos construir la representación regular como suma directa de tres representaciones irreducibles:

$$\rho_R = \rho_1\oplus\rho_2\oplus \rho_3$$

donde:

$$\rho_1: G\to \mathbb C$$
$$e\mapsto 1$$
$$(123)\mapsto 1$$
$$(132)\mapsto 1$$

$$\rho_2: G\to \mathbb C$$
$$e\mapsto 1$$
$$(123)\mapsto \omega$$
$$(132)\mapsto \omega^2$$

$$\rho_3: G\to \mathbb C$$
$$e\mapsto 1$$
$$(123)\mapsto \omega^2$$
$$(132)\mapsto \omega$$

donde $\omega=e^{i\frac{\pi}{3}}$ es la raíz cúbica de la unidad

**Recuerda:** la matriz de una repr. que es suma directa de reprs. irreducibles es la matriz que lleva las otras matrices en bloques en la diagonal. De este modo, hay 3 representaciones irreducibles de $A_3$, que son $\rho_1$, $\rho_2$ y $\rho_3$, y toda otra representación se puede construir como suma directa de estas (en general, cada una repetida varias veces)

**Aclaración:** si queremos por ejemplo hallar la matriz asociada a $(123)$ en la representación regular de $A_3$, tenemos que hacer:

$$\rho_R(123)=\rho_1(123)\oplus\rho_2(123)\oplus\rho_3(123)=1\oplus \omega\oplus\omega^2=\begin{pmatrix}1 & 0 & 0 \\ 0 & \omega & 0 \\ 0 & 0 & \omega^2\end{pmatrix}$$

**Fíjate:** el carácter del elemento neutro en una representación nos da la dimensión de dicha representación:

$$\boxed{dim \ \rho = \chi^{(\rho)}(e)}$$

## Lema de Schur

- Sea $G$ grupo (finito o infinito) y sean $(\rho, V), (\rho', V')$ dos representaciones **irreducibles** de $G$, tales que existe una aplicación que las relaciona: $f\circ \rho(g)=\rho'(g)\circ f$, entonces se cumple una (y solo una) de las siguientes posibilidades:

1. $dim(V)\neq dim(V')$: $f=0$
2. $dim(V)=dim(V')$ pero representaciones no son idénticas, entonces se cumple una de dos:
    - $f=0$ si $\rho\cancel \sim \rho'$
    - Si no, $\rho\sim \rho'$
3. Si $(\rho, V)=(\rho', V')$ (irreducibles) entonces $f$ múltiplo de la identidad


**Dem.:**

- Supongamos que existe una aplicación que nos lleva de $V$ a $V'$, $f: V\to V'$
- Entonces, $Ker[f]\leq V$ y $Im[f]\leq V'$
- Vemos que el Kernel de la aplicación es invariante bajo $\rho(g)$: $f(v)=0 \Longrightarrow f(\rho(g)v)=0$ porque $f(\rho (g)v)=\rho'(g)f(v)=0$
- Asimismo, vemos que la imagen de $f$ también es invariante bajo $\rho(g)$ porque si $v'=f(v)\in Im[f]$, entonces $\rho'(g)v'=\rho'(g)f(v)=f(\rho(g)v)$ esta también en $Im[f]$ sencillamente porque estamos aplicando $f$
- Tenemos entonces un subespacio invariante para cada representación, de modo que ambas son reducibles, a menos que los subespacios invariantes sean impropios (triviales)
- Pero como hemos dicho que han de ser irreducibles, este debe ser el caso, de modo que:

Para $Ker[f]$ se debe cumplir una de dos:

1. $Ker[f]=\{0\} \Longrightarrow f$ inyectiva
2. $Ker[f]=V \Longrightarrow f=0$

Para $Im[f]$ se debe cumplir el mismo razonamiento, ahora en $V'$:

1. $Im[f]=\{0\} \Longrightarrow f=0$
2. $Im[f]=V'\Longrightarrow f$ suprayectiva

Por ende, $f=0$  ó $f$ isomorfismo entre $V, V'$, que solo es posible si los espacios tienen la misma dimensión.

La tercera tesis del lema se demuestra fácilmente: si las representaciones son idénticas, entonces podemos escribir:

$$\rho(g)=f^{-1}\circ \rho(g)\circ f$$

Si la dimensión de las representaciones es finita, entonces podemos escribirlo todo en forma matricial porque las $\rho(g)$ son endomorfismos en el espacio vectorial de la representación y las $f$ son ahora automorfismos en dicho espacio vectorial (porque son isomorfismos entre $V$ y $V'$ pero ambos son ahora idénticos). Entonces, tenemos que:


$$D_\rho(g)=P^{-1}D_\rho(g)P \Longrightarrow P=\lambda I_\blacksquare$$

## Teorema (reducibilidad completa)

$$\boxed{\text{Toda representación de un grupo finito es completamente reducible}}$$

# 6. Teoremas en términos de caracteres

- Hasta ahora, hemos definido los caracteres extensivamente y hemos visto varios conceptos aparentemente desrelacionados. Veremos ahora la potencia y la elegancia de los caracteres, reformulando algunos teoremas de la teoría de representaciones de grupos

## Ortogonalidad de dos representaciones

- Sea $G$ finito de orden $N$ y sean dos representaciones irreducibles de este, $(\rho^{(1)}, V_1)$ y $(\rho^{(2)}, V_2)$. Entonces, tenemos la siguiente relación entre las matrices de estas representaciones:

$$\frac{1}{N}\displaystyle\sum_{g\in G}D_{ik}^{(1)}(g^{-1})D_{lm}^{(2)}(g)=\frac{1}{n_1}M_{im}(M^{-1})_{lk}\delta_{\rho^{(1)}\rho^{(2)}}$$

donde $\delta_{\rho^{(1)}\rho^{(2)}} = \begin{cases}1, \rho^{(1)}\sim \rho^{(2)} \\ 0, \text{ si no}\end{cases}$

**Fíjate:** en caso de que ambas representaciones sean equivalentes, entonces $M$ es la matriz de cambio de una a otra

**Fíjate:** esta relación vale para todas las componentes de las matrices de ambas representaciones, y no es ni un producto de matrices ni tampoco un producto de Kronecker de las matrices

### Corolario

- Si las dos representaciones irreducibles son idénticas, entonces la expresión se reduce a:

$$\frac{n_1}{N}\displaystyle\sum_{g\in G}D_{ij}^{(1)}(g^{-1})D^{(1)}_{kl}(g)=\delta_{il}\delta_{kj}$$

## Ortogonalidad de dos representaciones (reformulación en caracteres)

- Sea $G$ finito de orden $N$ y sean dos representaciones irreducibles de este, $(\rho^{(1)}, V_1)$ y $(\rho^{(2)}, V_2)$. Entonces, tenemos la siguiente relación entre los caracteres de las matrices de estas representaciones:

$$\boxed{\frac{1}{N}\displaystyle\sum_{g\in G}\bar \chi^{(1)}(g)\chi^{(2)}(g)=\delta_{\rho^{(1)}\rho^{(2)}}}$$

**Fíjate:** 

Al tomar trazas sobre el teorema de ortogonalidad, nos ha salido un conjugado complejo. Esto se debe a que, como el grupo es finito, toda representación de este es equivalente a una representación unitaria (se puede llegar a ella con un cambio de base), y como las trazas son invariantes bajo cambios de base, no hay pérdida de generalidad al admitir que trabajamos siempre con representaciones unitarias. Entonces, tenemos que:


$D^{(1)}(g^{-1})=[D_1(g)]^{-1}=[D_1(g)]^\dagger\Longrightarrow Tr[D^{(1)}(g^{-1})]=Tr([D_1(g)]^\dagger)\Longrightarrow \chi^{(1)}(g^{-1})=\bar \chi^{(1)}(g)$ en grupos finitos

**Dem.:**

Partimos del teorema de ortogonalidad. Escogemos concretamente el caso en el que $k=i, l=m$, de modo que la expresión queda como:

$$\frac{1}{N}\displaystyle\sum_{g\in G}D_{ii}^{(1)}(g^{-1})D_{mm}^{(2)}(g)=\frac{1}{n_1}M_{im}(M^{-1})_{mi}\delta_{\rho^{(1)}\rho^{(2)}}$$

Ahora sumamos sobre todos los índices $i, m$ para obtener las trazas en LHS:

$$\displaystyle\sum_{i=1}^{n_1}\displaystyle\sum_{m=1}^{n_2}\left(\frac{1}{N}\displaystyle\sum_{g\in G}D_{ii}^{(1)}(g^{-1})D_{mm}^{(2)}(g)\right)=\frac{1}{n_1}\delta_{\rho^{(1)}\rho^{(2)}}\displaystyle\sum_{i=1}^{n_1}\displaystyle\sum_{m=1}^{n_2} \frac{1}{n_1}M_{im}(M^{-1})_{mi}$$

Donde en RHS reconocemos un producto de matrices, ahora sí, y además sumado en sus índices de modo que es la traza de la identidad. Si las dos representaciones no son equivalentes, RHS es 0. Si en efecto lo son, entonces la traza de la identidad da la dimensión de la matriz, $n_1$, y recuperamos la expresión del teorema$_\blacksquare$

### Corolario

- Como los caracteres son funciones de clase y el teorema corre por todos los elementos del grupo, los elementos del grupo pertenecientes a la misma clase de conjugación darán el mismo carácter, de modo que podemos agrupar por clases de conjugación de la forma:

$$\boxed{\frac{1}{N}\displaystyle\sum_{i=1}^r c_i\bar\chi^{(1)}_i\chi^{(2)}_i=\delta_{\rho^{(1)}\rho^{(2)}}}$$

donde $c_i$ es el orden de la i-ésima clase de conjugación del grupo y $r$ es el número de clases de conjugación del grupo.

## Reformulación del teorema de ortogonalidad (vectorial)

- Sea una representación de grupo $(\rho, V)$ con caracteres $\chi^{(\rho)}_i$, donde el subíndice $i$ hace referencia al carácter de la clase de conjugación i-ésima. Construyamos el vector que en la componente i-ésima tiene el carácter de la clase de conjugación i-ésima (con un factor de normalización):


$$\boxed{\vec \chi^{(\rho)}=\left(\sqrt{\frac{c_1}{N}}\chi^{(\rho)}_1, \ldots, \sqrt{\frac{c_r}{N}}\chi^{(\rho)}_r\right)}$$

Entonces, el teorema de ortogonalidad se puede expresar como:

$$\boxed{\vec \chi^{(1)}\cdot \vec \chi^{(2)}=\delta_{\rho_1\rho_2}}$$

**OJO!:** fíjate en que, como los caracteres pueden en general ser complejos, el producto escalar conjuga el primer vector.

### Vectores de clase

- Sabemos que, dada una representación, podemos construir un vector que en cada componente tenga el carácter de una clase de conjugación, pesado por la raíz del orden de dicha clase y normalizado con $1/\sqrt N$
- También podemos definir "vectores de clase" como, fijada una clase, el vector que en cada componente tiene el carácter de dicha clase en una irrep. dada:

$$\vec \xi_k = \sqrt{\frac{ c_k}{N}}(\chi^{(1)}(c_k), \ldots, \chi^{(r)}(c_k))$$

**Propiedades:**

- $\langle \vec \xi_i, \vec \xi_j \rangle = \delta_{ij}$

Es decir, cumple ortogonalidad y normalización, igual que los vectores de caracteres. La ventaja es que podemos usarlos más fácilmente, porque no tenemos que andar introduciendo factores de peso para construirlos. Si quiero calcular $\langle \vec \xi_i, \vec \xi_j \rangle=0$ para hallar algún carácter que no sepa, puedo hacerlo directamente leyendo las columnas de la tabla de caracteres e ignorando los factores de normalización porque se van al 0

## Teorema

$$\boxed{\text{Un grupo tiene tantas representaciones irreducibles como clases de conjugación}}$$

## ¿Cómo hallo la descomposición de una repr. en irreps.?

- Sea un grupo $G$ y pongamos que tenemos cierta representación $(\rho, V)$ que sabemos que no es irreducible (como admitimos que $G$ es finito es por ende completamente reducible). Queremos descomponerla en las representaciones irreducibles del grupo:

$$\rho = \alpha_1\rho_1\oplus \ldots \oplus\alpha_r \rho_r$$

donde $\alpha_i\in \mathbb N$ es el número de veces que se repite una representación irreducible en la nuestra. Con los caracteres, hallar esta descomposición es fácil! Como hay tantas representaciones irreducibles del grupo como clases de conjugación tenga este, podemos construir los caracteres de todas las clases de conjugación y recordando la propiedad:

$$\chi^{(\rho)}(c_k)=\displaystyle\sum_{i=1}^r \alpha_i\chi^{(\rho_i)}(c_k)$$

Tenemos un sistema de $r\times r$ ecuaciones, donde las incógnitas son precisamente los coeficientes de la suma directa de representaciones. Por teorema de Rouché-Frobenius es sistema compatible determinado de modo que tiene solución única $\longrightarrow$ saco $\{\alpha_i\}\longrightarrow$ ya conozco la descomposición de mi representación $\rho$ en representaciones irreducibles

**Más fácil: por ortogonalidad**

- Sea el vector de caracteres $\vec\chi^{(\rho)}$ de cierta representación $\rho$ que quiero descomponer en irreps. de cierto grupo:

$$\rho = \alpha_1\rho^{(1)}\oplus\ldots\oplus\alpha_r\rho^{(r)}$$

- Entonces, puedo hallar el i-ésimo coeficiente de la forma:

$$\boxed{\alpha_i=\langle \vec \chi^{(i)}, \chi^{(\rho)}\rangle}$$

**Nota:** como $\alpha_i\in \mathbb N$, da igual en qué orden tomes el producto escalar, así que si un vector es complejo y el otro real, pon el complejo en el segundo argumento para no tener que conjugarlo

**Dem.:**

- Consideremos la descomposición de $\rho$ en irreps. (no-equivalentes) y tomemos caracteres:

$$\chi^{(\rho)}(c_k)=\displaystyle\sum_i\alpha_i\chi^{(i)}(c_k)$$

Podemos multiplicar a izq. y dcha. por $\sqrt{c_k/N}$ y se sigue cumpliendo:

$$\sqrt{\frac{c_k}{N}}\chi^{(\rho)}(c_k)=\sqrt{\frac{c_k}{N}}\displaystyle\sum_i\alpha_i\chi^{(i)}(c_k)$$

- Esto se cumple para cada clase de conjugación $C_k$ por separado, así que podemos considerar la relación en las $r$ clases a la vez. Es decir, es cierto que:

$$\frac{1}{\sqrt N}(\sqrt{c_1}\chi^{(\rho)}(c_1), \ldots, \sqrt{c_r}\chi^{(\rho)}(c_r))=\frac{1}{\sqrt N}\alpha_1(\sqrt c_1\chi^{(1)}(c_1), \ldots, \sqrt c_r\chi^{(1)}(c_r))+\ldots + \frac{1}{\sqrt N}\alpha_r(\sqrt c_1\chi^{(r)}(c_1), \ldots,\sqrt c_r\chi^{(r)}(c_r))$$

En términos de vectores de caracteres:

$$\vec\chi^{(\rho)} = \displaystyle\sum_{i=1}^r\alpha_i\vec \chi^{(i)}$$

Y hemos logrado expresar el vector de caracteres de una representación en la base **ortnonormal** de los vectores de caracteres de las irreps. independientes del grupo: es precisamente el que tiene componentes $(\alpha_1, \ldots, \alpha_r)$. Podemos ahora servirnos de la ortogoanlidad para aislar una componente concreta:

$$\alpha_l = \langle \vec \chi^{(l)}, \vec \chi^{(\rho)}\rangle_\blacksquare$$

**Nota:** recuerda que los elementos de esta base son de módulo unidad

## Equivalencia de representaciones: reformulación con caracteres

$$\boxed{\text{Dos representaciones son equivalentes $\Longleftrightarrow$ tienen los mismos caracteres}}$$

## Irreducibilidad: reformulación con caracteres

$$\boxed{\text{$\rho$ irreducible $\Longleftrightarrow \langle \vec \chi^{(\rho)}, \vec \chi^{(\rho)}\rangle$  }=1}$$

## Descomposición de la representación regular

**Proposición 1:** sea $(\rho, V_{\mathcal A_G})$ la representación regular del grupo $G$, entonces sus caracteres son nulos para todos los elementos del grupo salvo la identidad:

$$\chi^{(\rho_R)}(c_k)=N\delta_{ke}$$

y en consecuencia los vectores de caracteres son:

$$\vec \chi^{(\rho_R)}=\left(\sqrt{\frac{1}{N}}N, 0, \ldots, 0\right)$$

porque la clase de conjugación del elemento identidad es $\{e\}$


**Proposición 2:** la representación regular no es irreducible porque:

$$\langle \vec \chi^{(\rho_R)}, \vec \chi^{(\rho_R)}\rangle =N\neq 1$$

**Teorema:** la representación regular se puede expresar como suma directa de todas las representaciones irreducibles del grupo, tantas veces como dimensión tengan:

$$\boxed{\rho^R=\bigoplus n_i \rho^{(i)}}$$

**Corolario:** como $dim(\rho^R)=N$, tenemos entonces tomando dimensiones a ambos lados del teorema que:

$$|G|=\displaystyle\sum_i^rn_i^2$$

**Ejemplo:**

- Sea $G=C_3=\{e, a, a^2\}$, entonces tenemos que $|G|=3=n_1^2+n_2^2+n_3^2$. La única solución es $n_1=n_2=n_3$, de modo que las 3 representaciones irreducibles (que sabemos que son 3 porque el grupo, al ser abeliano, tiene 3 clases de conjugación, todas ellas singletones) tienen dimensión 1

# 7. Representaciones Restringidas

**Def.:** sea $G$ grupo y $H\subset G$. Sea $(\rho, V)$ una representación de $G$. Entonces, se define la "representación de $G$ restringida a $H$" como aquella representación de $H$ tal que coincida con la representación de los elementos de $H$ en la representación global (la de $G$):

$$\rho_G|_H: H\to \text{End}_r(V)$$
$$h\mapsto \varphi_h: V \to V$$

tal que $\rho_G(h)=\rho_G|_H(h); \ \ \forall h\in H$

- Por eso se llama restringida, porque en el fondo lo único que estamos haciendo es servirnos de una representación de todo el grupo para construir una representación de un subgrupo tan solo copiando las matrices de la representación que ya sabemos que funcionan
- Esto sirve por ejemplo cuando se rompe la simetría de un sistema pero no completamente. Por ejemplo, sea un sistema con simetría rotacional en el espacio. Ahora, pongamos que introducimos un agente (usualmente un campo magnético uniforme), que rompe esta simetría definiendo una dirección privilegiada del espacio para los momentos angulares. Entonces, el sistema aún presenta simetría de rotación en el espacio, pero solo en torno al eje del campo magnético. Decimos que el grupo de simetría se ha reducido, y las matrices de rotación, que antes eran las matrices en $SO(3)$, ahora se han reducido al subgrupo de rotaciones en el plano $SO(2)$. Entonces, podemos usar las matrices de $SO(3)$ que representen las rotaciones en el plano.

**OJO!:** incluso si una representación de $G$ es irreducible, su representación restringida en $H$ no tiene por qué serlo, y en general podrá reducirse en representaciones irreducibles de $H$. La matriz que nos da la descomposición de las representaciones irreducibles del grupo grande en términos de las irreducibles del pequeño es la **matriz de Frobenius**:

$$\rho^{(i)}_G|_H=\bigoplus_{j=1}^{r_H}\alpha_{ij}\rho^{(j)}_H$$

# 8. Representaciones inducidas

- Con las representaciones restringidas, hemos aprendido a construir la representación de un subgrupo de $G$ a partir de una representación de $G$
- Podemos ahora preguntarnos si podemos proceder al revés: si conocemos la representación de un subgrupo $H\subset G$, ¿podemos construir una representación de $G$ a partir de él? $\longrightarrow $ sí!

**Procedimiento:**

1. Construyo todos los cosets de $H$ en $G$ (si se quiere, el conjunto cociente $G/H$)
2. Para cada elemento de $G$, construyo su matriz en la representación inducida como:

$$D_G(g)=D_H(g_p^{-1}gg_q)\Delta_{pq}$$

donde:

- $g_p\in \text{coset}_p$
- $g_q\in \text{coset}_q$
- $\Delta_{pq}=\begin{cases}1, \  \ g_p^{-1}gg_q\in H \\ 0, \text{ si no}\end{cases}$

**Fíjate:** como el grupo no es un conjunto ordenado, la representación de $G$  inducida por $H$ que me dé esta fórmula va a depender del orden en el que coja yo los elementos de los cosets. Sin embargo, esto solo me va a cambiar el orden de filas y columnas, de modo que si construyo dos representaciones tomando los elementos de los cosets en distinto orden, puedo pasar de una matriz a otra con un cambio de base compuesto por transformaciones elementales y son por ende equivalentes.

**Def.:** la matriz de coeficientes que nos dice cómo se descompone la representación inducida en las representaciones irreducibles de $G$ es $(\beta_{ij})$. Por ejemplo, si usamos la representación irreducible 1 de $H$ para inducir una representación de $G$, entonces:

$$\rho^{(1)}_H\to \rho_{H\to G}=\beta_{11}\rho^{(1)}_G\oplus \beta_{12}\rho^{(2)}_G\oplus\beta_{13}\rho^{(3)}_G$$

## Teorema de reciprocidad de Frobenius

- La matriz de Frobenius de la representación restringida de $G$ en $H$ y de la inducida de $G$ por $H$ se relacionan por una trasposición:

$$\boxed{(\alpha_{ij})=(\beta_{ji})}$$
