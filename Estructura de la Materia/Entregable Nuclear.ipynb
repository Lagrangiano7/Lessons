## Sr-90

**Modos de desintegración:** $\beta^-$

**Esquema de desintegración:** se desintegra siempre por beta- en Itrio-90 en su estado fundamental (véase que es uno de esos casos en los que el padre es mucho más estable que el hijo, habida cuenta de que el estroncio-90 tiene vida media de décadas y el Itrio-90 de unos pocos días)

**Cálculo de $Q$:**

- Tenemos la desintegración:

$$_{38}^{90}Sr\to \ _{39}^{90}Y+e^-+\bar \nu_e$$

- Calculamos su factor $Q$ de la forma:

$$Q=(m(Sr)-m(Y)-m_e)c^2 \approx (M(Sr)-38m_e-M(Y)+39m_e-m_e)c^2=(M(Sr)-M(Y))c^2$$

- Empleando los valores de masa atómica referidos en la propia tabla nuclear, obtenemos:

$$Q=545.965(keV)$$

que está en muy buen acuerdo con el factor Q tabulado: $Q=545.9keV$


- **Relación de Q con energía de partículas emitidas:** como se trata de una desintegración beta, se generan 3 productos de desintegración, de modo que la repartición de momentos lineales (y por ende de energías cinéticas) no será siempre la misma para todos los productos como sucede, por ejemplo, en las desintegraciones alpha. Sin embargo, sí que podemos sacar del factor Q un dato relevante, que es la energía cinética máxima de los electrones beta- emitidos:

$$T_{e,max}\approx Q=545.965$$


Examinando la gráfica grande y los datos numéricos que proporciona la IAEA, vemos que para $T_{e,max}=545.9keV=Q$ no hay ninguna instancia, y que ya para $T_{e,max}=545.0keV$ tenemos $dN/dE=	0.0000000808348$, lo que pone de manifiesto la calidad de nuestra estimación de la energía cinética máxima del electrón. Naturalmente, el motivo por el que no llega nunca a coincidir con $Q$ es que el antineutrino siempre se llevará algo de energía. De hecho, si creemos a pies juntillas en el modelo estándar, el neutrino es una partícula sin masa, de modo que nunca puede quedar en reposo como resultado de la relatividad especial, y en consecuencia siempre estaremos introduciendo cierto error en nuestros cálculos cuando despreciemos el término de energía cinética del neutrino en el límite de $T_e>>T_\nu$, que es precisamente el que hace al caso para calcular la energía cinética máxima del electrón en una desintegración beta-

## Cs-137

**Modos de desintegración:** $\beta^-$

**Esquema de desintegración:** se desintegra siempre por beta- en Ba-137, pero no siempre en su estado fundamental. De hecho, la mayoría de las veces ($94.7\%$) el hijo queda en su segundo excitado, y luego ya emite gammas hasta alcanzar el fundamental

**Cálculo de $Q$:**

- Tenemos la desintegración:

$$_{55}^{137}Cs\to \ _{56}^{137}Ba+e^-+\bar \nu_e$$

- Calculamos su factor $Q$ de la forma:

$$Q=(m(Cs)-m(Ba)-m_e)c^2 \approx (M(Cs)-55m_e-M(Ba)+56m_e-m_e)c^2=(M(Cs)-M(Ba))c^2$$

- Empleando los valores de masa atómica referidos en la propia tabla nuclear, obtenemos:

$$Q=1175.623(keV)$$

que vuelve a estar en muy buen acuerdo con el factor Q tabulado: $Q=1175.63keV$


- **Relación de Q con energía de partículas emitidas:** por la discusión de antes, volvemos a juzgar oportuno determinar la energía cinética máxima del electrón beta- emitido:

$$T_{e,max}\approx Q=1175.623$$


Examinando la gráfica grande y los datos numéricos que proporciona la IAEA, vemos que, si bien a $E=Q$ no hay ningún registro de electrones con esa energía, de modo que técnicamente hemos vuelto a acertar, observamos que en la práctica $dN/dE$ se hace aproximadamente 0 a $E\sim 500keV$, esto es, a mitad de la gráfica. Para energías superiores seguimos registrando instancias de electrones, pero la frecuencia es muy baja. Es interesante notar que esto no sucede en el caso del Sr-90. Tal vez este decaimiento prematuro de los registros de electrones en función de la energía tenga que ver con que la contribución del antineutrino es en este caso apreciable siempre, de modo que el electrón se ve imposibilitado para pasar de los $500keV$. Sería interesante estudiar qué dicta si la contribución energética del antineutrino depende del isótopo que se desintegre en beta-

## Pu-239

**Modos de desintegración:** $\alpha$

**Esquema de desintegración:** se desintegra siempre por alpha en U-235, principalmente en los primeros estados excitados de este

**Cálculo de $Q$:**

- Tenemos la desintegración:

$$_{94}^{239}Pu\to \ _{92}^{235}U+_2^4He$$

- Calculamos su factor $Q$ de la forma:

$$Q=(m(Pu)-m(U)-m(He))c^2 \approx (M(Pu)-94m_e-M(U)+92m_e-M(He)+2m_e)c^2=(M(Pu)-M(U)-M(He))c^2$$

- Empleando los valores de masa atómica referidos en la propia tabla nuclear, obtenemos:

$$Q=5244.498(keV)$$

que vuelve a estar en muy buen acuerdo con el factor Q tabulado: $Q=5244.50keV$


- **Relación de Q con energía de partículas emitidas:** como la desintegración considerada es ahora alpha, podemos calcular las energías cinéticas con que saldrá la partícula alpha. Deduzcamos la expresión de la energía cinética de la partícula alpha en función de lo que conocemos, que es el factor $Q$:

Si el Uranio acabara en el estado fundamental, tendríamos entonces en el sistema propio del núcleo de $Pu$ (donde $T_{Pu}\equiv 0$):

$$Q=T_\alpha+T_U$$

por conservación de energía. No obstante, como la desintegración alpha es no-relativista, podemos hacer $T_\alpha=\frac{\vec p_\alpha^2}{2m_\alpha}$ y lo propio para el núcleo de Uranio, de modo que existe una sencilla relación entre ambas energías cinéticas no relativistas: $T_U=\frac{m_\alpha}{m_U}T_\alpha$. Introduciendo esto en el factor $Q$ conseguimos que la energía cinética de la partícula alpha quede solo dependiente del factor $Q$:

$$Q=T_\alpha\left(1+\frac{m_\alpha}{m_U}\right)$$

Hacemos ahora $m_\alpha/m_U\approx M_\alpha/M_U$, y escribimos:

$$Q\approx T_\alpha\frac{M_\alpha+M_U}{M_U}$$

pero como la desintegración es no-relativista, podemos admitir en buena aproximación que la masa se conserva. Esto es, el balance de energías de la reacción, que se cumple de forma exacta, es:

$$m_{Pu}c^2=m_Uc^2 + T_U+m_\alpha c^22+T_\alpha$$

pero admitimos que las energías cinéticas son despreciables frente a las energías en reposo de las partículas involucradas por ser una desintegración alpha, de modo que, de la expresión anterior:

$$m_{Pu}\approx m_U+m_\alpha$$

que es el mismo principio de conservación de masa que subyace a las reacciones químicas. En definitiva, tenemos $M_U+M_\alpha\approx M_{Pu}$:

$$Q\approx T_\alpha\frac{M_{Pu}}{M_U}$$

y en buena aproximación, como $M_X=Au+\delta$:


$$Q\approx T_\alpha\frac{A_{Pu}}{A_U}\Longrightarrow T_\alpha\approx \frac{235}{239}Q$$

Si ahora consideramos que el Uranio puede quedar en un estado excitado, parte de la energía disponible se la quedará este núcleo y no podrá emplearse en la energía cinética de los productos de desintegración, de modo que:

$$T_\alpha\approx \frac{235}{239}(Q-E^*)$$

Por ejemplo, en el nivel fundamental del Uranio (donde se toma el origen de energías), calculamos $T_\alpha=5156.73keV$, y el resultado tabulado es $T_\alpha=5156.7keV$. En el segundo excitado, encontramos $T_\alpha=5143.90keV$ y el valor tabulado es $T_\alpha=5144.3keV$. Notamos que la fórmula que se ha derivado funciona bien a pesar de que las aproximaciones sean un poco burdas.

**Nota:** aunque la fórmula que hemos obtenido para la energía cinética de la partícula alpha es rudimentaria, de ella se desprende un hecho importante, y es que $T_\alpha<Q$ siempre. Esto es, no podemos considerar, como hacíamos en la desintegración beta, que todo queda en reposo salvo la partícula alpha. Esto se debe, naturalmente, a la conservación de momento lineal en un sistema de 2 partículas. En el inercial propio del padre, este se encuentra, por construcción, en reposo, de modo que los productos de desintegración han de salir no solo colineales, sino con momentos lineales opuestos. Entonces, si la partícula alpha lleva cierta energía cinética asociada, es necesario que el hijo (U) también tenga energía cinética no nula en el inercial del padre. Como $Q=T_U+T_\alpha$, es por esto que $T_\alpha<Q$. La única forma de que una de las dos partículas quede en reposo sería que la otra quede asimismo en reposo. Esto no violaría la conservación de momento lineal (0 antes, 0 después), pero sí que violaría la conservación de energía, en tanto que el Plutonio-239 tiene energía en reposo de sobra para dar lugar a los dos productos de desintegración asimismo en reposo, de modo que el exceso de energía debe ir a algún sitio, y es a la energía cinética de dichos productos.

- La cadena de desintegración del Sr-90 es como sigue:

$$^{90}Sr \to \ ^{90}Y \to ^{90}Zr$$

donde ambas desintegraciones son por beta-. El circonio-90 es estable. Por tanto, podemos modelar esta cadena de forma exacta con el modelo de Bateman, si consideramos que partimos de una muestra aislada y carente de cualquier elemento que pueda decaer en alguna de las 3 especies involucradas en la cadena que estamos estudiando. Escribimos entonces:

$$\begin{cases}\dot N_{Sr}=-\lambda_{Sr}N_{Sr} \\ \dot N_Y=-\lambda_YN_Y+\lambda_{Sr}N_{Sr} \\ \dot N_{Zr}=\lambda_YN_Y\end{cases}$$

donde el Sr-90 sigue la ley de decaimiento exponencial porque nada le contribuye y el Zr-90 únicamente recibe aquellas contribuciones provenientes de la desintegración del Y-90.

Como vimos en clase, esta cadena sí es capaz de alcanzar un equilibrio secular, que gráficamente (en unidades arbitrarias) queda de la siguiente forma:

![Serie](Serie_Radiactiva.png)

- donde $N_1=N_{Sr}$, $N_2=N_{Y}$, $N_3=N_{Zr}$.

- **Implicaciones en la actividad en equilibrio secular:** la actividad de una muestra con varias especies desintegrándose es, naturalmente, una magnitud extensiva (i.e: las actividades son aditivas). Sin embargo, en equilibrio secular, la forma de la actividad total es especialmente sencilla, ya que se igualan la actividad de todos los nodos de la cadena de desintegración se iguala: $\lambda_iN_i=\lambda_jN_j$ (salvo, naturalmente, para el último nodo, que es por construcción estable). Por tanto, en una cadena con $p$ nodos en equilibrio secular, tendremos:

$$A=\displaystyle\sum_i^{p-1}A_i=(p-1)A_1(0)$$

donde $A_1(0)=\lambda_1N_1(0)$ es la actividad inicial del primer nodo de la cadena. En nuestro caso, como $p=3$, tenedremos que **en equilibrio secular, la actividad de la muestra se duplicará con respecto a lo esperado si considerásemos que el Sr-90 se desintegra directamente siguiendo la ley exponencial a un núclido estable**
