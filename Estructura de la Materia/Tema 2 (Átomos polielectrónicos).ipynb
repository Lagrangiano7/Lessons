**Outline:** no sabemos resolver la ec. de Schrödinger para un átomo polielectrónico, por lo que vamos a hacer aproximaciones al Hamiltoniano completo:

$$H=\displaystyle\sum_{i=1}^N\left(\frac{\vec p_i^2}{2m}-\frac{Z\alpha\hbar c}{r_i}\right) + \displaystyle\sum_{i< j}\frac{\alpha\hbar c}{||\vec r_i-\vec r_j||}$$

- Molesta el término de interacción de los electrones

1. Aprox. de electrones libres: admito que los electrones no se ven (mala, mejora con correcciones) --> formalismo de # de ocupación
2. Aprox. de apantallamiento cte.: me quito el término de interacción de electrones dando cuenta de él mediante una carga nuclear efectiva --> formalismo de # de ocupación
3. Aprox. de campo central (Hartree): sustituyo términos de interacción e-e y e-p por un potencial efectivo central

**Correcciones:**

- Aprox. de Hartree deja término suelto que será muy pequeño si hemos modelado bien nuestro potencial --> podemos introducirlo como una perturbación
- Metemos estructura fina (+reglas de Hund)

# 1. Aproximaciones del átomo polielectrónico

## 1.1 Electrones libres

- **Clave:** consideramos despreciable el término de interacción entre electrones, así que queda Hamiltoniano ideal para N fermiones --> invita a usar formalismo de números de ocupación

- Si la config. electrónica es: $1s^{\nu_1}2s^{\nu_2}\ldots$, la energía del sistema será $E=\nu_1E_1+\nu_2E_2+\ldots$

donde las energías son las del átomo hidrogenoide

- Se comprueba que ya para el He esta aproximación es muy mala. Mejora si introducimos el término de interacción como perturbación, pero es costoso porqe hay que evaluar integrales complicadas

## 1.2 Apantallamiento cte.

- Nos centramos en el átomo de He
- **Clave:** si uno de los dos electrones está muy cerca del núcleo, solo verá la carga de los protones: $q=Ze$
- Si está muy alejado, verá la carga del núcleo pero también la del otro electrón, que lo apantalla: $q=(Z-1)e$
- Para distancias intermedias, tendremos $q=(Z-\sigma), \sigma\in(0,1)$. Ahora tenemos:

$$H=\displaystyle\sum_{i=1}^N\left(\frac{\vec p_i^2}{2m}-\frac{(Z-\sigma)\alpha\hbar c}{r_i}\right)$$

donde las energías son las del átomo hidrogenoide con esta nueva carga nuclear efectiva:

$$E_n=-\frac{(Z-\sigma)^2}{n^2}E_0$$

- **Problema:** mi teoría es efectiva y no me dice cómo calcular $\sigma$
- **Solución:** la determino experimentalmente --> da resultados con precisión del 2%, no está mal pero el apantallamiento no es intrínseco a mi teoría (podría modelarlo tal vez, pero es mejor usar el siguiente método)

## 1.3 Potencial central (Hartree)

- **Clave:** las interacciones del Hamiltoniano completo me dan dolores de cabeza porque tienen términos no-centrales
- Me los quito a cambio de introducir un potencial central que modele estas interacciones, de modo que podré hacer separación de variables. Ahora tengo:

$$H=\displaystyle\sum_{i=1}^N\left(\frac{\vec p_i^2}{2m}+V_c(r_i)\right)-\displaystyle\sum_{i=1}^N\left(\frac{Z\alpha\hbar c}{r_i}+V_c(r_i)\right)+\displaystyle\sum_{i< j}\frac{\alpha\hbar c}{||\vec r_i-\vec r_j||}$$

- donde el primer sumatorio es nuestro nuevo Hamiltoniano sin pertubar y el resto de términos son parte de la "interacción electrostática residual", que incluiremos como una perturbación más adelante, cuando toque hacer correcciones

**Propiedades de V_c(r_i) (iguales que las de apantallamiento cte.):**

1. Cerca del núcleo, el electrón i-ésimo solo ve la carga de los protones: $V_c(r_i)\to-\frac{Z\alpha\hbar c}{r_i}$
2. Lejos del núcleo, el electrón se ve apantallado por los otros $N-1$ electrones: $V_c(r_i)\to-\frac{(Z-(N-1))\alpha\hbar c}{r_i}$

**Algoritmo de Hartree**

1. Escojo N funciones iniciales $\{\varphi_i\}$
2. Calculo $V_c(r)$ como:

$$V_c(r)=-\frac{Z\alpha\hbar c}{r}+\frac{1}{N_{pares}}\int f(\vec r)d\Omega$$

con $f(\vec r)=\int d^3\vec r'\displaystyle\sum_{i<j}\frac{e|\varphi_i(\vec r')|^2 e|\varphi_j(\vec r')|^2}{4\pi\varepsilon_0||\vec r'-\vec r||^2}$

es decir, $V_c(r)$ es el témino e-p (ese no se toca) + el valor esperado del término de interacción e-e con respecto a las funciones de prueba de turno solo que promediado en ángulo sólido para dar algo dependiente de $r$ y poder hacer separación de variables después

3. Resolvemos la ec. de Schrödinger para este Hamiltoniano
4. Si las energías se estabilizan, el algoritmo ha convergido y tenemos nuestros cálculos del espectro de $H$ y las autofunciones de $H$ (que no es el completo sino el de Hartree, pero se acercan mucho porque el método converge. Además, luego corregiremos con término electrostático).

El algoritmo se repite hasta que la diferencia de energías entre una iteración y la anterior sea menor que nuestra tolerancia, introduciendo las nuevas funciones de onda que hemos hallado en la iteración anterior

**Nota:** el Hamiltoniano de la aproximación de campo central también es ideal, de modo que seguimos pudiendo aplicar el formalismo de números de ocupación! Solo que ahora la energía de cada nivel será la que nos dé el algoritmo de Hartree

# 2. Configuraciones electrónicas

**Def.:** una "configuración electrónica" es aquella configuración de los electrones en un átomo, en general polielectrónico. Esto incluye parte espacial y de spin, y debe respetar el principio de exclusión de Pauli

## Función de onda del sistema en una config. electrónica

- Como los electrones son fermiones y por tanto obedecen Pauli, debemos construir la función de onda del sistema de tal forma que sea antisimétrica bajo permutaciones de $p$ electrones. Esto se logra mediante el determinante de Slater, de modo que obtendremos una parte espacial y una parte espinorial de simetría opuesta bajo permutaciones

## Deg. de una config. electrónica

**Problema:** si en una config. electrónica tenemos 2 partículas y 2 estados disponibles (i.e: $1s^2$), entonces solo hay un determinante de Slater y ya conocemos la función de onda de esa config. electrónica. Si por el contrario tenemos demasiados estados disponibles para las partículas que tenemos, no nos caben en el determinante de Slater y hay que hacer varios

- Decimos entonces que la config. electrónica está degenerada, y su deg. coincide con el número de determinantes de Slater independientes que se pueden construir, que es:

$$g=\begin{pmatrix}g_1\\ \nu_1\end{pmatrix}\begin{pmatrix}g_2\\ \nu_2\end{pmatrix}\ldots \begin{pmatrix}g_n\\ \nu_n\end{pmatrix}$$

donde $n$ es el último nivel ocupado.

**Nota:** como todos los determinantes de Slater de una configuración son igualmente válidos, conforman una base de los posibles vectores de estado (o funciones de onda) que representan dicha configuración electrónica. Por tanto, un sistema con una config. electrónica determinada (con degeneración >1) puede encontrarse en una superposición de sus dos o más determinantes de Slater

## Representación reducida de una config. electrónica

- Los determinantes de Slater son largos de escribir, de modo que deseamos encontrar una forma abreviada (y biunívoca) de referirnos a ellos. Es como sigue:

$$[X]: \ \left(m_{l_1}^{m_{s_1}}m_{l_2}^{m_{s_2}}, \ldots\right)$$

**Reglas para escribir la forma abreviada:**

1. Debe acompañarse de la forma alfanumérica para saber a qué niveles y orbitales nos referimos
2. Dentro de un mismo nivel, ponemos $m_l$ en orden decreciente, y $m_s$ positivos antes que negativos

- Con estas reglas, lograremos no dar determinantes de Slater redundantes

## Paridad de una config. electrónica

- Los determinantes de Slater de una config. electrónica tienen paridad bien definida, y es:

$$\Pi=(-1)^{\displaystyle\sum_i l_i}$$

# 3. Interacción electrostática residual

- Introducimos ahora el término electrostático como una perturbación (cosa que es lícita porque nuestro método ha convergido, de modo que nuestro último cálculo del potencial de Hatree ha de ser muy bueno)
- Buscamos ahora una base en la que la perturbación sea diagoal para poder aplicar como caso no-deg. de teoría de perturbaciones
- Veo que el término residual no conserva $l_i$ por separado, pero sí que conserva momento angular total --> acoplo momentos angulares
- Como del spin no se dice nada, acoplo spines también
- Aplico (computacionalmente) corrección y encuentro que:

**Resultado:** cada config. electrónica se desdobla en varios niveles de energía, llamados "términos electrostáticos" o solo "términos", que denotaremos según:

$$^{2S+1}L$$

**Nota:** sigue habiendo degeneración porque los términos no dependen de tercer número cuántico de L ni de S. Por tanto, cada término electrostático $^{2S+1}L$ tiene degeneración:

$$g_{term}=(2L+1)(2S+1)$$

**Notación:** en analogía con el átomo hidrogenoide, denotaremos $L=0 \to S$, $L=1 \to P$, etc. 

**Notación:** véase que la ventaja de poner $2S+1$ es que vemos a ojo si se trata de un singlete ($S=0\to 2S+1=1$), doblete ($S=1\to 2S+1=1/2$) o triplete ($S=1\to 2S+1=3$) de spin

**¿Cómo calculo términos en la práctica?**

1. Acopolo momentos angulares orbitales y de spin de la config. electrónica (niveles llenos no contribuyen)
2. Hago tabla con las posibles combinaciones de $L$ y $S$
3. Quito aquellas que violen Pauli / "conservación" de degeneración original

**Trucos:**

1. Si solo hay 2 electrones equivalentes (mismo n, mismo l) --> solo me quedo con combinaciones de $L+S$ par
2. Si solo hay 2 huecos "equivalentes", lo mismo
3. Si todos los electrones son no-equivalentes, todas las configuraciones son válidas

**Fíjate:** los electrones de niveles llenos se acoplan a $L=0, S=0$, por eso no contribuyen. Por tanto, todos los gases nobles se encuentran e $^1S$

## Los términos en triplete de spin están más ligados que los de singlete de spin

- Cuando dos electrones están en triplete, tienen spines paralelos
- Por Pauli, se deben repeler más que si los tuvieran antiparalelos (fuerza de canje por parte espacial antisimétrica bajo permutaciones)
- Al estar más separados, se apantallan menos y notan más el tirón dell núcleo ==> están más ligados

# 4. Corrección spin-órbita

- En el átomo hidrogenoide, tanto la corrección relativista como la S-O entraban dentro de la corrección de estructura fina. Ahora, solo consideraremos la S-O por ser la más relevante (tras la corrección de interacción electrostática residual)

- Para cada electrón, podemos despreciar el campo $\vec B$ de los otros electrones y quedarnos solo con el suyo, de modo que es como aplicar corrección S-O a cada electrón: $\Delta H_{SO}\sim \displaystyle\sum_{i=1}^N\vec l_i\cdot \vec s_i$


**Consecuencia:** la energía ahora depende de la orientación relativa entre spines y momentos angulares orbitales (ya no se conservan L y S por separado, sino que ahora se conserva $\vec J = \vec L +\vec S$)

**Aprox. de Russel-Saunders:** para átomos con $Z$ pequeño, se cumple:

$$\langle \displaystyle\sum_{i}\vec l_i\cdot \vec s_i\rangle \approx \langle \vec L\cdot \vec S\rangle$$

Por tanto, bajo esta aproximación, la corrección S-O para átomos polielectrónicos es:

$$\boxed{\Delta E_{SO}=\frac{A}{\hbar^2}\langle \vec L\cdot \vec S\rangle=\frac{A}{\hbar^2}(J(J+1)-L(L+1)-S(S+1))}$$

- **Notación:** cada nivel de estructura fina se denotará por $^{2S+1}L_J$
- La deg. de cada nivel de estructura fina será $(2J+1)$

## Reglas de Hund (ordenación de niveles de EF)

- Veamos primero cómo se comportan estos niveles de estructura fina

**Regla de intervalos de Landé:** la diferencia de energías entre dos niveles consecutivos de estructura fina es (1) proporcional a $A$ y (2) crece linealmente con $J$

**Útil:** la diferencia de energía entre dos niveles de EF consecutivos de un mismo término es:

$$\boxed{^{2S+1}L_{J+1}- ^{2S+1}L_{J}=A(J+1)}$$

- Llamaremos "multiplete normal" a aquel con $A>0$. Este empieza desde abajo en $J=0$ y va subiendo, cada vez le cuesta más subir
- Llamaremos "multiplete invertido" a aquel con $A<0$. Este empieza desde arriba en $J=0$ y va bajando, cada vez baja más trecho

**Def.:** llamaremos "configuración tipo Hund" a 2 tipos de configuraciones:

1. Todos los electrones de valencia en un mismo nivel
2. 1 electrón en orbital $s$ y el resto juntos en otro orbital

Las reglas de Hund, aunque tienen excepciones, casi siempre sirven para determinar, al menos, el término fundamental de los átomos con configuraciones tipo Hund. Las regla son como siguen:

**Regla 1:** los términos electrostáticos de mayor $S$ suelen ser los de menor energía

**Regla 2:** de entre ellos, si hubiera empate, el de mayor $L$ es el término fundamental de la configuración

**Regla 3 (estructura fina):** el signo de $A$ (y por tanto el tipo de multiplete) se puede deducir a partir de cómo de llenos están los niveles $nl^\nu$ (solo sirve para el tipo 1 de config. de Hund)

- Nivel menos que se semilleno ($\nu < 2l+1$) --> $A>0$ (multiplete normal)
- Nivel más que semilleno ($\nu > 2l+1$) --> $A<0$ (multiplete invertido)
- Nivel semilleno ($\nu=2l+1$) --> $A=0$, sin desdoblamiento a primer orden

**Fíjate:** los átomos con 2 electrones / huecos equivalentes tienen los mismos términos (y por tanto con la misma ordenación). Lo único en lo que van a distinguirse es que el primero, como es menos que semilleno, tendrá multiplete normal en estructura fina, y el segundo tendrá multiplete invertido

# 5. Reglas de selección

**Nota:** en aprox. dipolar eléctrica, la interacción fotón-electrón es 1 a 1, por lo que admitimos que solo 1 electrón cambia en la transición, y el resto de la config. electrónica se mantiene inalterada

**Para config. electrónica:**

1. $\Delta l=\pm 1$

**Para términos:**

1. $\Delta S=0$
2. $\Delta L=0,\pm 1$ (pero no se permite $L=0\to L=0$)

**Para estructura fina:**

1. $\Delta J=0,\pm 1$ (pero no se permite $J=0\to J=0$)
2. $\Delta M_J=0,\pm 1$

**Observación:** en el pasado, los científicos pensaban que había 2 tipos distintos de Helio, que llamaron "ortohelio" y "parahelio". Ahora sabemos que son el mismo átomo de $He$ solo que los spines de los dos electrones se acoplan, respectivamente, a $S=0$ o $S=1$. Véase que, de acuerdo a estas reglas de selección, no podemos pasar de un "tipo" de He a otro, al menos no en aprox. dipolar eléctrica

**Importante:** las reglas de selección de los **términos** asumen que el átomo no es demasiado pesado, de modo que $L$ y $S$ están bien definidos. Si el átomo fuera más pesado, podrían no estarlo y los efectos relativistas serían importantes (nosotros no los hemos considerado, por eso nuestra teoría tildaría de prohibidas transiciones que no lo son). Esto es, los átomos pesados pueden violar las reglas de selección de los términos. Las de CE y las de estructura fina **no las violan**
