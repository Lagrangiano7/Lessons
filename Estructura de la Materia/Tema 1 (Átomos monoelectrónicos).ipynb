# 1. Solución del átomo de hidrogenoide

## Modelo de Bohr (semiclásico)

**Postulados de Bohr:** (átomos hidrogenoides)

1. El momento angular orbital del electrón orbitando el protón está cuantizado: $L=n\hbar$
2. Las órbitas del electrón son circunferencias
3. Cuando el electrón pasa de una órbita a otra, emite / absorbe un fotón cuya energía es la diferencia entre niveles

$$r=\frac{n^2}{Z}a_0; \ \ a_0=\frac{\hbar}{\alpha cm_e}$$
$$v=\frac{Z}{n}\alpha c$$
$$E_n=-E_0\frac{Z^2}{n^2}; \ \ E_0=\frac{1}{2}m\alpha^2c^2=13.61eV$$

**Nota:** estrictamente sería con la masa reducida del sistema. Esto es importante para átomos exóticos como el muonio, donde la masa de las partículas ligadas deja de ser despreciable

## Solución cuántica

$$|nlm_l\rangle\equiv \Psi_{nlm}(r,\theta\phi)=R_{nl}(r)Y_{l}^{m_l}(\theta,\phi)$$

donde:

$$R_{nl}(r)=N\left(\frac{Zr}{a_\mu}\right)^lQ_{nl}(Zr/a_\mu)e^{-Zr/na_\mu}; \ \ a_\mu = \frac{m_e}{\mu}a_0$$

**Observaciones importantes:**

1. Los estados estacionarios del Hidrógeno ($|nlm\rangle$) tienen paridad bien definida. La parte radial es simétrica y los armónicos esféricos tienen paridad $PY_{l}^{m_l}=(-1)^lY_{l}^{m_l}$, de modo que **la paridad de un estado estacionario coincide con la paridad de su $l$**

2. Cuando $r\to 0$, $R(r)\sim r^l$

3. La parte radial tiene $n-l-1$ nodos (sin contar origen)

4. Las partes radial y angular se normalizan por separado

5. La degeneración del nivel n-ésimo es $n^2$ (o $2n^2$ si contamos spin)

**OJO!:** se definen varias densidades de probabilidad a partir de la total. La total se define incluyendo el jacobiano de la parte radial dentro de la densidad original para que así solo tengamos la densidad de probabilidad en una variable. Las densidades radial y angular se obtienen tomando la marginal respecto de las variables contrarias:

$$\rho_T(r,\theta\varphi)=|\Psi(r,\theta,\varphi)|^2r^2=|R(r)|^2|Y_l^{m_l}|^2r^2$$

- Si ahora tomamos la marginal con respecto a la parte angular, tenemos la densidad de probabilidad radial:

$$\rho_r(r):=\int  \rho_T d\Omega= r^2R^2(r)$$

# 2. Correcciones del átomo hidrogenoide

## 2.1 Estructura fina

1. Corrección relativista
2. Interacción $S_e-L_e$

### 2.1.1 Corrección relativista

- De la fórmula para la velocidad de Bohr vemos que el electrón, cuando está en el estado fundamnental, viaja con $v=c/137$. Esto es aproximadamente el $1\%$ de la velocidad de la luz, que ya empieza a ser relativista, por lo que hay que usar la relación E-p de Einstein, pero poco, de modo que podemos desarrollarla en Taylor a orden siguiente

$$T = E-mc^2=\sqrt{m^2c^4+p^2c^2}-mc^2 \to T\approx \frac{p^2}{2m}-\frac{p^4}{8m^3c^2}$$

**Nota:** fíjate que el primer término, por ser clásico, no lleva $c$. El siguiente ya es relativista así que sí la lleva (como cuando en cuántica recuperamos resultados clásicos y se va la $\hbar$)

- El Hamiltoniano queda de la forma:

$$H=\frac{p^2}{2m}+V(r)-\frac{p^4}{8m^3c^2}= H^0+\Delta H$$

**Corrección a primer orden**

$$\boxed{\Delta E_{rel}=E_n\alpha^2\frac{Z^2}{n^2}\left(\frac{n}{l+1/2}-\frac{3}{4}\right)}$$

**Efectos sobre los niveles de energía:**

1. Baja todos (ya sabíamos porque tenía un "-" delante de $p^4$)
2. Muestra que la deg. de $l$ en el átomo de Bohr era accidental y que realmente los niveles de energía dependen también de $l$
3. La corrección es más significativa para los niveles más bajos (claro, órbitas Keplerianas. Cuanto más cerca del núcleo, más rápido va el electrón)

### 2.1.1 Interacción $S_e-L_e$

- El electrón está orbitando alrededor del protón, y al ser una carga rotando, da lugar a un campo magnético
- Asimsmo, tiene spin, asociado al momento dipolar de la partícula, de modo que el campo magnético que genera el electrón al orbitar interacciona con el spin del mismo electrón. Esto hay que tenerlo en cuenta porque ahora tenemos un electrón sometido a un potencial con un término adicional de campo magnético

- La energía potencial de un momento dipolar $\vec M$ sometido a un campo magnético $\vec B$ es:

$$V=-\vec M\cdot \vec B$$

- El momento dipolar de una partícula está estrechamente relacionado con su spin:

$$\vec M = -g_s\frac{\mu_B}{\hbar}\vec S$$

donde $g_s$ es el factor giromagnético (aprox. 2 para el electrón)

- Queda solo calcular el campo magnético al que está sometido el electrón, generado por él mismo al orbitar alrededor de un punto. Como el electrón se mueve en círculos, habría que ir cambiando de sistema de referencia (transformada de Lorentz) a cada tramo diferencial que recorre, y esto introduce un factor 1/2 llamado **factor de Thomas**
- En la práctica, es más fácil ponerse en el sistema de referencia no inercial del electrón, de modo que ahora es el protón el que orbita el electrón, con la misma velocidad, y este está estacionario en el origen
- Este sistema es como una espira, que está resuelto de EM clásico. En el centro de una espira, el campo magnético es:

$$B=\frac{\mu_0 I}{2r}$$

- Haciendo $I=e/T$, $T=2\pi r/v$, multiplicando y dividiendo por $m_e$ y recogiendo el producto $m_evr=L_e$ llegamos finalmente a que:

$$\vec B=\frac{1}{2}\frac{\mu_0}{4\pi}\frac{Ze}{m_e}\frac{\vec L_e}{r^3}$$

donde el primer factor 1/2 es el factor de Thomas. Multiplicamos ambas expresiones para obtener el potencial de esta interacción y tenemos:

$$V=\frac{\mu_0}{4\pi}\frac{Ze^2}{2m_e^2}\frac{\vec S\cdot \vec L}{r^3}=\frac{Z\alpha\hbar}{2m_e^2c}\frac{\vec L\cdot \vec S}{r^3}$$


**Corrección a primer orden:**

$$\boxed{\Delta E_{L-S}=-E_nZ^2\alpha^2\frac{j(j+1)-l(l+1)-3/4}{2nl(l+1/2)(l+1)}}$$

**Corrección fina:**

- Sumando ambas correcciones, obtenemos finalmente:

$$\boxed{\Delta E_{fina}=E_n\alpha^2\frac{Z^2}{n^2}\left(\frac{n}{j+1/2}-\frac{3}{4}\right)}$$

**Observación:** el nivel n-ésimo se desdobla en $n$ niveles (pero como la degeneración de cada nivel va como $2n^2$, la corrección fina nunca logrará romper la degeneración de ningún nivel al completo)

**Notación:** los niveles pasan a denotarse con notación espectroscópica pero con subíndice $j$ del momento angular total al que se acoplan L y S: $\boxed{nl_j}$

## 2.2 Lamb shift

- En el vacío tiene lugar producción de pares virtuales electrón-positrón que apantallan la carga del núcleo y hacen que el electrón esté menos ligado a él --> los niveles de energía más cercanos al núcleo suben porque están menos ligados


- Para orbitales s (l=0), que son a los que más les afecta, la corrección es:

$$\boxed{\Delta E_{lamb}=\frac{Z^4\alpha^5m_ec^2}{4n^3}k(n)=-E_n\alpha^3\frac{Z^2}{2n}k(n)}$$

donde $k(n)$ es una integral que depende del número principal pero $k(n)\approx 13$

**Efecto:** sube los orbitales s con respecto de los p, desbloqueando la transición $2s_{1/2}\to 2p_{1/2}$ (es la más notoria porque para niveles superiores la diferencia de energías se atenúa. Con todo, pasa en todos los niveles)

**Efecto:** en un mismo nivel de energía $n$, los orbitales de igual $j$ pero distinto $l$ tendrán distintas energías

## 2.3 Estructura hiperfina

- La estructura hiperfina tiene en cuenta las interacciones de spin del núcleo:

1. Spin núcleo - Spin electrón
2. Spin núcleo - L electrón

- Momento magnético del electrón por estar girando:

$$\vec M_L = \frac{e}{2m_e}\vec L_e$$

- Momento magnético del electrón por tener spin:

$$\vec M_S=g_s\frac{e}{2m_e}\vec S$$

- Momento magnético total del electrón:

$$\vec M_e = -\frac{\mu_B}{\hbar}(\vec L+2\vec S)$$

- Momento magnético del núcleo por tener spin:

$$\vec M_p=g_I\frac{m_e}{m_p}\frac{\mu_B}{\hbar}\vec I$$

**Corrección hiperfina:**

$$\boxed{\Delta E_{hiperf}=-E_n\alpha^2\frac{Z}{n}g_I\frac{m_e}{m_p}\frac{f(f+1)-I(I+1)-j(j+1)}{j(j+1)(2l+1)}}$$

**Nota:** se llama "hiperfina" porque, aunque va con $\alpha^2$ como la fina, tiene un factor adicional $m_e/m_p\sim 10^{-4}$. En estados ligados donde ambas partículas tengan igual masa (como en positrón-electrón), la estructura hiperfina tendría entonces la misma relevancia que la fina

**Efecto:** todos los niveles se desdoblan en función de la proyección del momento angular total del electrón sobre el spin nuclear ($\vec F=\vec J+\vec I$)

**En la práctica:** los núcleos que estudiemos entiendo que tendrán $I=1/2$. Es por esto que la estructura hiperfina desdoblará cada nivel en 2 ($F\in\{J-1/2,J+1/2\}$)

**OJO!:** si el núcleo tiene más de una partícula, tienes que calcular el spin nuclear! Por ejemplo, el deuterio tiene spin nuclear $I=1$

# 3. Reglas de selección

- Se observa que los electrones de un átomo pueden ver su estado alterado y emitir / absorber la diferencia de energía (así se estudia experimentalmente el espectro del sistema)
- ¿Por qué sucede esto?

    1. Tu sistema está preparado en superposición de varios autoestados (prob. oscila)
    2. El átomo está sometido a una perturbación dependiente del tiempo
    
- Estudiaremos el segundo caso porque del primero ya está todo dicho

- El **ritmo de transición** es $R=10^8s^{-1}$
- El tiempo de vida en un cierto estado para un electrón es entonces $\tau=1/R=10^{-8}s$

- Las interacciones posibles con la radiación son 3:

    1. Absorción: me llega un fotón de frecuencia propicia, lo absorbo y subo de nivel
    2. Emisión estimulada: me llega un fotón de cierta frecuencia, me pone a mí a emitir otro fotón y así me desexcito
    3. Emisión espontánea
    
- **Nota:** habrá una regla de selección por cada número cuántico, y dependiendo del CCOC en el que estemos trabajando, las reglas cambiarán

## 3.1 Reglas de selección en átomo de Bohr

**OJO!:** estas reglas de selección son las más generales (i.e: todas las posibles). Sin embargo, para ver si una transición es posible o no, hay que calcular su elemento de matriz (si por ejemplo el campo eléctrico va en dirección z, el término perturbativo irá como $z$ y este solo permite transiciones que no cambien $m_l$). Esto se ve trivialmente con el teorema de Wigner-Eckart si estamos en aproximación dipolar

1. $\Delta n=0,1,2\ldots$ (nada)
2. $\Delta l=\pm 1$ (cambio de paridad por cons. de L, porque fotón tiene $\mp 1\hbar$ de l)
3. $\Delta m_l=0,\pm 1$

## 3.2 Reglas de selección en estructura fina

1. $\Delta n=0,1,2,\ldots$ (nada)
2. $\Delta l=\pm 1$
3. $\Delta j=0,\pm 1$
4. $\Delta m_j=0,\pm 1$

**Recuerda:** estamos trabajando en aproximación dipolar (en el fondo, que el campo eléctrico que causa la pertubración depende linealmente del espacio). Es por esto que la probabilidad de transición (y el ritmo de transición) dependen del momento dipolar de la partícula $||\langle d\rangle_{i,f}||^2$. Asimismo, las transiciones prohibidas son aquellas que tienen prob. 0 a primer orden, pero podrían tener prob. no nula a órdenes superiores de teoría de perturbaciones depenidentes del tiempo

# 4. Efecto Zeeman

- Se trata ahora de estudiar qué le pasa al átomo de Hidrógeno **en corrección fina** cuando se le imbuye en un campo B uniforme
- El Hamiltoniano de Bohr tendrá entonces dos perturbaciones:

$$H=H_{Bohr}+\Delta H_{fina}+\Delta H_B$$

- **Estrategia:** estudiamos casos límite, en los que una de las perturbaciones es más significativa que la otra, y tomamos entonces la más significativa como parte del Hamiltoniano resuelto. El otro término será la perturbación.

## 4.1 Débil ($\Delta H_B<<\Delta H_{fina}$)

- Esto sucede cuando la intensidad de los campos magnéticos del interior del átomo es mayor que la intensidad del campo externo

$$B<<1(T)$$

- $H^{(0)}=H_{Bohr}+\Delta H_{fina} \Longrightarrow$ adoptamos CCOC de estructura fina
- $\Delta H = \Delta H_{B}$
- El campo magnético externo interactuará con el electrón, ya que este tiene $\vec M_e=-\frac{\mu_B}{\hbar}(\vec L+2\vec S)$. Si tomamos por simplicidad el campo magnético en el eje Z, tenemos:

$$\Delta H_B=-\vec M_e\cdot \vec B=\frac{\mu_B}{\hbar}B(L_z+2S_z)$$

**Problema:** para calcular la corrección a primer orden, hay que calcular el valor esperado de un observable de un CCOC que no es el nuestro ($L_z$ y $S_z$)

**Solución: teorema de la proyección:**

$$\langle \vec L\rangle_{\text{base J}}=\frac{\langle \vec J\cdot \vec L\rangle}{\hbar^2j(j+1)}\langle\vec J\rangle$$

- Donde todos los valores esperados están tomados con respecto de la base del CCOC de estructura fina
- El resultado de esta corrección es:

$$\boxed{\Delta E_B=\mu_BBg_Lm_j; \ \ g_L=\frac{3}{2}+\frac{\frac{3}{4}-l(l+1)}{2j(j+1)}}$$

donde $g_L$ es el "factor de Landé"

**Nota:** se rompe completamente la degeneración!

- Esta corrección se le suma a la energía del átomo teniendo en cuenta estructura fina y hemos acabado

## 4.2 Fuerte (Paschen-Back)

- $H^{(0)}=H_{Bohr}+\Delta H_B \Longrightarrow$ adoptamos CCOC del Hamiltoniano de Bohr
- $\Delta H = \Delta H_{fina}$

- Antes conocíamos el espectro del átomo en estructura fina así que podíamos lanzarnos a calcular directamente la corrección que introduce el campo B. Hallemos primero el nuevo espectro del átomo en presencia de B y luego ya haremos la corrección fina en nuestra base

$$\Delta E_B=\langle nlsm_lm_s|\frac{\mu_B}{\hbar}B(L_z+2S_z)| nlsm_lm_s\rangle$$

- Ahora estos valores esperados son directos, y obtenemos:

$$\boxed{\Delta E_B = \mu_BB(m_l+2m_s)}$$

- Ahora que conocems el espectro del Hamiltoniano $H^{(0)}$, deberíamos hacer la corrección de estructura fina. No obstante, esta va como $\Delta E_{Am_lm_s}$ y es muy pequeña. También rompería deg. totalmente como en el caso débil, pero no la tendremos en cuenta, así que **ahora puede haber degeneración accidental entre niveles**

## Relación entre separación de líneas y separación de energías

- **Las transiciones desde los dos niveles más juntos de energía dan las longitudes de onda más cercanas**
- Si conocemos la separación en energías de dos niveles, tenemos una fórmula aproximada para la separación de sus longitudes de onda en la transición a un mismo nivel

$$\boxed{\delta\lambda\approx \frac{\lambda_0}{E_0}\delta E}$$
