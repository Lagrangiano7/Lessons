# 1. Ley exponencial de desintegración

**Nota:** la desintegración de un núcleo es un proceso aleatorio, y no sabemos con certeza cuándo se desintegrará un único núcleo. Sin embargo, si la muestra es suficientemente grande, podemos hacer predicciones con certeza porque el espacio muestral es suficientemente grande

**Regla de oro de la desintegración:** el ritmo de desintegración de un elemento radiactivo es proporcional al tamaño de la muestra

$$\frac{dN}{dt}=-\lambda N$$

donde:

- $N$: número de partículas en la muestra
- $\lambda$: "constante de desintegración", es probabilidad de desintegración por unidad de tiempo (lo que en teoría de perturbaciones dependientes del tiempo llamaríamos "ritmo de transición")

**Def.:** se define la "actividad" de una muestra como su ritmo de desintegración:

$$\boxed{A=-\frac{dN}{dt}\ \ [\text{Bq ó Ci ó dps}]}$$

- Las unidades de la actividad se suelen dar en (1) Becquerelios, (2) Curios ó (3) desintegraciones por segundo
- **Nota:** $1 Bq=1dps$; $1Ci=3.7\cdot 10^{10}Bq$. El Curio se define como la actividad de 1g de Radio226

- Integrando la regla de oro con respecto del tiempo se obtiene que las muestras suficientemente grandes de material radiactivo se desintegran siguiendo una ley exponencial:

$$N(t)=N_0e^{-t/\tau}$$

donde $\tau=1/\lambda$ es la "vida media" del núcleo estudiado (el tiempo de vida promedio de ese núcleo)

**Def.:** la "semivida" de una muestra ($t_{1/2}$) es el tiempo que tarda en llegar a la mitad de su tamaño, y es intrínseco del núcleo, no depende del tamaño original de la muestra. Se relaciona con $\tau$ según:

$$\boxed{t_{1/2}=Ln(2)\tau}$$

## Constante de desintegración ($\lambda$)

- **Nota:** un núcleo se puede desintegrar de varios modos, cada uno de ellos con cierta probabilidad. Es de esperar que el modo primario sea el causante de la mayoría de desintegraciones. De hecho, a cada modo le corresponde una constante de desintegración $\lambda_i$, y la total que se emplea luego en la ley exponencial es la suma de todas ellas:

$$\lambda=\displaystyle\sum_{modos}\lambda_i$$

**Def.:** se define la "constante de ramificación / branching ratio (BR)" de un modo de desintegración como la probabilidad de que un núcleo se desintegre por un modo dado:

$$\boxed{BR_i=\frac{\lambda_i}{\lambda}}$$

# 2. Cadenas radiactivas

- **Observación:** en general, los núcleos no llegan a la línea de estabilidad con 1 sola desintegración, sino que se necesita que el núcleo padre se desintegre en uno o más hijos, que a su vez se convierten en padres que dan más hijos y así hasta llegar a la línea de estabilidad. A todos los nodos de esta cadena, comenzando por el padre inestable y acabando en los hijos estables, se le llama "cadena de desintegración"

## Ecuaciones de Bateman

- Para entender las cadenas de desintegración, construyamos un modelo de juguete. Estudiemos la cadena:

$$1\to 2\to 3$$

con el núcleo 3 ya estable. Entre medias, naturalmente, habrá productos de desintegración residuales como electrones, positrones,  partículas alpha, etc.

- Inicialmente, tenemos una muestra pura de núcleos de tipo 1. Estos solo pueden decaer a hijos de tipo 2, que a su vez solo pueden decaer a hijos de segunda generación de tipo 3 y estos ya son estables. Las poblaciones de cada tipo de núcleo seguirán:

$$\begin{cases}\dot N_1=-\lambda_1N_1 \\ \dot N_2=-\dot N_1-\lambda_2N_2 \\ \dot N_3=-\dot N_2\end{cases}$$

**Explicación:** el núcleo (1) decae en 2. Entonces, 2 crece porque 1 está aumentando su población pero a su vez decrece porque es inestable y decae en 3. Como 3 es estable, solo puede aumentar su población con las aportaciones de 2

Aplicando la regla de oro a cada núcleo, tenemos:

$$\begin{cases}\dot N_1=-\lambda_1N_1 \\ \dot N_2=+\lambda_1N_1-\lambda_2N_2 \\ \dot N_3=+\lambda_2N_2\end{cases}$$

- Nos queda un sistema de EDOs de orden 1 acopladas. Podemos diagonalizar la matriz de coeficientes y resolverlo de forma escalonada como resolveríamos un sistema de ecuaciones lineales algebraicas, y obtenemos:

$$\begin{cases}N_1(t)=N_1(0)e^{-t/\tau_1} \\ N_2(t)=\frac{\lambda_1}{\lambda_2-\lambda_1}N_1(0)\left(e^{-t/\tau_1}-e^{-t/\tau_2}\right)\\ N_3(t)=N_1(0)-N_1(t)-N_2(t)\end{cases}$$

- **Nota:** véase que $N_3(t)$ se puede sacar de que la magnitud $N_1+N_2+N_3$ es constante en el tiempo (sale de desacoplar la matriz de coeficientes, pero también del sentido común, porque 1 solo se desintegra en 2 y 2 solo en 3)

- **IMPORTANTE:** muy pocos núcleos se desintegran realmente siguiendo la ley exponencial, porque la gran mayoría son productos intermedios de una cadena radiactiva

- **Observación:** los términos intermedios de las cadenas siempre tienen un máximo en el tiempo. En el caso de nuestro modelo de juguete, $N_2(t)$ es el único término intermedio, y su valor máximo es:

$$N_{2}(t_{max})=\frac{\lambda_1}{\lambda_2}N_1(t_{max})$$

**OJO!:** como $N_2(t)$ involucra exponenciales con distintos coeficientes, no se puede determinar $t_{max}$ de forma analítica, hay que hacerlo de forma computacional

## Equilibrio secular

- Estudiemos el ritmo de desintegración de $N_2(t)$ bajo las siguientes hipótesis simplificadoras:

1. $t_{1/2}$ del padre (1) >> $t_{1/2}$ del hijo (2)
    - Por ejemplo, $^{226}Ra\to \ ^{222}Rn + ^4He$ con $t_{1/2}(Ra)=1600$ años, $t_{1/2}(Rn)=3.82$ días
2. El tiempo de observación es $t<<t_{1/2}(1)$
3. El tiempo de observación es $t>>t_{1/2}(2)$ (por lo menos 9 veces superior)



- Aplicando estas 3 hipótesis a la expresión exacta de $N_2(t)$ llegamos a que:

$$N_2(t)\approx \frac{\lambda_1}{\lambda_2}N_1(0)=cte.$$

- Pasando $\lambda_2$ multiplicando, tenemos:

$$\boxed{\lambda_2N_2=\lambda_1N_1}$$

- **Conclusión:** para tiempos **ni demasiado cortos ni demasiado largos** con respecto a las escalas de tiempo padre/hijo, el ritmo de desintegración del hijo se iguala al del padre --> "equilibrio secular"

- **¿Qué sucede en la práctica?:** el padre se desintegra muy lentamente y el hijo lo hace muy rápidamente, así que se adapta casi inmediatamente a las desintegraciones del padre. Siempre que el padre se desintegre, el hijo se desintegra (pero el hijo solo se puede desintegrar cuando se desintegra en él el padre), de modo que los ritmos de desintegración de padre e hijo se igualan

- **Resultado:** en equilibrio secular, las actividades son idénticas. Como la actividad es a tenemos una cadena de $p$ núcleos, entonces la actividad total es de:

$$\boxed{A_{T, eq. \ sec}=(p-1)A1(0)}$$

donde la actividad total va con $p-1$ porque el último nodo es estable. Entiendo que este resultado solo sirve para cadenas como la de Bateman, en la que $i$ solo recibe aportaciones de $i-1$ y solo se desintegra en $i+1$

- **Nota:** el equilibrio secular explica por qué el mineral de Uranio es 12 veces más radiactivo de lo que debería ser si decayera directamente a una especie estable, es que se desintegra a través de una cadena, y cada núcleo contribuye a la actividad de la muestra! Aplicando el razonamiento de las actividades aditivas, debe haber en torno a 13 núcleos distintos en la cadena

**OJO!:** es aquí donde se muestra la importancia de definir la actividad como $\lambda_iN_i$ en lugar de $\dot N_i$. Si en las ecuaciones de Bateman sumamos $\dot N_1+\dot N_2+\dot N_3$, nos da 0 (porque el número total de partículas de la muestra se conserva). Esto querría decir que la actividad de la muestra es siempre 0. No obstante, sabemos que esto no puede ser así, porque *hay* desintegraciones. Por otro lado, como 3 es estable, podríamos pensar que a la actividad solo contribuyen 1 y 2 y hacer $A=-\dot N_1-\dot N_2$, pero esto solo dejará un término $\lambda_2N_2$, y sabemos que 1 también contribuye a la actividad del sistema en su conjunto. Por tanto, debemos recurrir a la definición de actividad como $\lambda_iN_i$:


$$A_T=\lambda_1N_1+\lambda_2N_2+\cancel\lambda_3N_3$$

Por tanto, tenemos que la actividad de una muestra es siempre aditiva, y en equilibrio secular, como se igualan los ritmos de desintegración de padres e hijos:

$$A_T=\lambda_1 N_1+\lambda_2N_2=2\lambda_1N_1$$

que coincide exactamente con la expresión que hemos deducido anteriormente para la actividad de una cadena de $p$ nodos en equilibrio secular
