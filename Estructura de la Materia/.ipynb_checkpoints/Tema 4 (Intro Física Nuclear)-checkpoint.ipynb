**Objetivos:**

1. Conocer términos habituales
2. Estudiar empíricamente la estabilidad del núcleo
3. Estudiar los modos de desintegración
4. Estudiar la fisionomía del núcleo

# 1. Definiciciones

## Miscelanea

- **Def.:** un "nucleón" es aquello que compone un núcleo (son protones y neutrones)
- **Def.:** los distintos nucleones se mantienen unidos sobreponiéndose a la repulsión electrostática de los protones por virtud de la "fuerza nuclear fuerte". Esta es una fuerza efectiva, que viene de la interacción fuerte entre los quarks que conforman cada nucleón. Parte de ella se "filtra" más allá de su nucleón correspondiente y se acopla a otros quarks

## Relaciones entre núcleos

- **Def.:** la identidad (el elemento) de un núcleo la da el número de protones que tenga ("número atómico", $Z$). Para un mismo elemento, el núcleo puede variar su número de neutrones, y a cada una de estas variantes del mismo elemento la llamaremos "isótopo" (dos isótopos de un mismo elemento pueden tener propiedades nucleares distintas)
    - "Isótopo estable" es aquel que es eterno (no cambia número de protones, neutrones ni estado energético jamás espontáneamente. Otra cosa será que no esté aislado). En la práctica, se dice estable aquel isótopo del que nunca se ha registrado una desintegración, pero podría ser que su vida media sea tan larga que aún no hayamos esperado tiempo suficiente
    - "Isótopo inestable" es aquel que cambia su naturaleza espontáneamente incluso estando aislado, emitiendo partículas

- **Def.:** elementos "isóbaros" son aquellos que tienen el mismo número másico $A=Z+N$. Es decir, están en la misma diagonal de la tabla nuclear
- **Def.:** elementos "isótonos" son aquellos con mismo número de neutrones $N$ pero distinto $Z$. Son la contraparte de los isótopos

## Sobre tiempos de desintegración

- **Def.:** la "semivida" ($t_{1/2}$) de una especie es el tiempo que debe transcurrir para que se desintegre la mitad de una muestra pura de dicho material suficientemente representativa
- **Def.:** la "vida media" ($\tau$) de una especie es el tiempo promedio que vive un núcleo de dicha especie hasta desintegrarse. Se relaciona con la semivida a través de una constante:

$$t_{1/2}=\ln(2)\tau$$

- **Def.:** el "ritmo de desintegración para el modo de desintegración i-ésimo" ($\Gamma_i$) de una especie que se desintegra de varias formas (i.e: beta-, etc.) es $\Gamma_i=\frac{1}{\tau}$

# 2. La línea de estabilidad

- **Def.:** la "tabla nuclear" es un compendio de todos los núcleos conocidos. En el eje X está el número de neutrones $N$ y en el eje Y el número atómico $Z$.

- **Def.:** un "núcleo exótico" es aquel que tiene demasiados / demasiados pocos neutrones para los protones que tiene (i.e: está a la dcha. o a la izda. de la línea de estabilidad, respectivamente)

- **Def.:** la "línea de estabilidad" es la región de la tabla de núcleos estables. Sigue aprox. una curva. Para $Z, N$ pequeños (abajo a la izquierda) es aprox. lineal: $Z\approx N$ pero después los núcleos van adquiriendo más neutrones que protones (se acerca al eje X, parece una sqrt). Los núcleos por encima de la curva de estabilidad tienen demasiados pocos neutrones y muchos protones --> desintegración beta+/captura electrónica. Los de debajo tienen demasiados neutrones para demasiados pocos protones --> beta-

## Propiedades empíricas de la estabilidad

- **Observación:** hay como 70 más núcleos con $A$ par que impar
- **Observación:** en núcleos de A par, puedo tener N/Z ambos pares (mucho de esto) ó N/Z ambos impares (muy pocos)
- **Observación:** en núcleos de A impar, tengo tantos de N par / Z impar como del inverso --> no distingue

**Conclusión:** debe haber algo que favorezca la estabilidad de los núcleos de A par, y dentro de ellos N/Z ambos pares (me sirve para desarrollar un modelo de potencial efectivo)


**Def.:** los "números mágicos" son aquellos N / Z que dan lugar a núcleos que tienen más estabilidad que sus aledaños en la tabla nuclear (no garantiza estabilidad, solo los hace más estables). En el modelo de capas, es porque estos números completan su capa de protones / neutrones. Si un núcleo tiene tanto N como Z mágicos, se dice que es "doblemente mágico"

# 3. Modos de desintegración

- Los núcleos que son inestables se desintegran, en general siguiendo una cadena de desintegración, hasta llegar finalmente a un núcleo de la línea de estabilidad. Veamos los modos de desintegración principales

**IMPORTANTE:** estrictamente, la desintegración debe ser un proceso espontáneo y aislado, como alpha o beta+/-. Veremos que hay otros "modos" como la captura electrónica que involucran interacciones del núcleo con otra partícula, pero esto ya no es de por sí una desintegración

![Tabla nuclear](tabla_nuclear.png)

## 3.1 Alpha

**Proceso:** un núcleo suelta un núcleo de He

$$_Z^A X \to \  _{Z-2}^{A-4} Y+_2^4 He$$

**¿Dónde ocurre?:** en átomos por encima de la línea de estabilidad, cerca del borde y con bastante N / Z

**Nota:** los núcleos de He son muy estables y por eso son los que más se sueltan. No obstante, hay veces en las que se eyectan otros núcleos ligeros como el Carbono (ya no se considera alpha). La línea entre este tipo de desintegración y la fisión espontánea es difusa, pero la clave está en que estos núcleos son ligeros y los de fisión espontánea siguen siendo bastante pesados

## 3.2 Beta -

**Proceso:** un neutrón del núcleo se convierte en protón, y para compensar la carga suelta un electrón (y su antineutrino)

$$_Z^A X \to \  _{Z+1}^{A} Y+e^-+\bar\nu_e$$

**A nivel de partículas:**

$$n\to p + e^-+\bar \nu_e$$

**¿Dónde ocurre?:** en núcleos por debajo de la línea de estabilidad, que quieren perder neutrones y ganar protones

## 3.3 Beta +

**Proceso:** igual que beta- pero ahora es protón $\to$ neutrón (emitiendo positrón)

$$_Z^A X \to \  _{Z-1}^{A} Y+e^++\nu_e$$

**Nota:** el proceso a nivel de partículas no tiene lugar para partículas libres porque está cinéticamente prohibido. El protón no tiene suficiente energía en reposo para dar lugar a todos los productos de desintegración asimismo en reposo. Lo que pasa es que como los protones del núcleo no son libres sino que están ligados, pueden desintegrarse de esta forma

**¿Dónde ocurre?:** en núcleos por encima de la línea de estabilidad, que quieren ganar neutrones y perder protones

## 3.4 Captura electrónica (CE)

**Proceso:** el núcleo captura un electrón, que se fusiona con uno de sus protones para dar un neutrón (como beta- pero invertido)

$$_Z^A X +e^-\to \  _{Z-1}^{A} Y+\nu_e$$

**OJO!:** estos electrones se toman de los del propio átomo. Para núcleos grandes, algunos electrones incursionan dentro del núcleo y son estos los que se absorben para beta-

**¿Dónde ocurre?:** igual que en beta+, quiere perder protones y ganar neutrones

## 3.5 Emisión espontánea de protones / neutrones

**Proceso:** el núcleo eyecta un protón / neutrón

Neutrón: $_Z^A X \to \  _{Z}^{A-1} X+n$

Protón: $_Z^A X \to \  _{Z-1}^{A-1} Y+p$


**¿Dónde ocurre?:** en núcleos con un exceso extremo de protones / neutrones. Se quieren librar de ellos ya y los eyectan. La región donde tiene lugar la emisión espontánea de protones / neutrones se llaman "dripping lines / líneas de goteo". Para protones, sucede en el extremo superior y para neutrones en el inferior. Se teoriza que toda la tabla está delimitada por líneas de goteo, y que aquellas regiones en las que ya hemos descubierto dichas líneas están acabadas de explorar

## 3.6 Fisión espontánea

**Proceso:** el núcleo fisiona (espontáneamente, estando aislado) en 2 o más hijos + productos de desintegración

$$_Z^A X \to \  _{Z_1}^{A_1}Y+_{Z_2}^{A_2}W+\text{productos}$$

**Nota:** no tienen por qué ser siempre los mismos núcleos hijos, un mismo núcleo puede tener distintos procesos de fisión espontánea

**¿Dónde ocurre?:** en núcleos muy pesados, con muchos protones y neutrones

## 3.7 Gamma

**Proceso:** un núcleo en un estado excitado se desexcita, emitiendo un fotón

$$_{Z}^{A}X^*\to _{Z}^{A}X+\gamma$$

# 4. Mecánica del núcleo

- Los núcleos son estados ligados de protones y neutrones (y estos a su vez de quarks), de modo que podemos estudiar su espectro ligado al igual que hacíamo con los átomos

- Cada nivel de energía del núcleo lleva asociado (1) un spin nuclear y (2) una paridad. Se toma origen de energías en nivel fundamental

- Como los núcleos se desexcitan, a cada nivel se le puede asociar una vida media, entendida como "el tiempo promedio que dura un núcleo en dicho nivel"

- Hay algunos niveles con vida media excepcionalmente larga --> los llamamos "isómeros" a los núcleos en dichos estados excitados, y los denotamos con una $m_i$ tras $A$, donde el subíndice i indica el número de isómero que es (en general puede haber más de uno, si solo hubiera uno se omite el subíndice)

$$^{99m}Tc$$

## Spin nuclear

- Se define el "spin nuclear" como el momento angular total del núcleo:

$$\vec J = \displaystyle\sum_i^A(\vec l_i+\vec s_i)$$

**Fíjate:** cada contribución $\vec j_i$ ha de ser semientera porque $s_i=1/2$. Entonces, si acoplamos $A$ par tenemos $J$ entero. Si acoplamos $A$ impar, $J$ semientero

**Observación:** el estado fundamental de todo núcleo con $A$ par, Z/N par-par se acopla a $J=0$. Esto se conoce como "pairing"

- En el estado fundamental, cuando hay pairing, sabemos que los nucleones que se acoplan dos a dos son del mismo tipo porque la fuerza nuclear fuerte acopla preferentemente nucleones de igual paridad, y la paridad del fundamental es +1

## Esquemas de desintegración

- **Def.:** un "esquema de desintegración" es una representación gráfica de un modo de desintegración de un núcleo, donde se presentan (1) los niveles de energía involucrados, (2) flechas para los modos de desintegración del padre al hijo, (3) la probabilidad con la que ocurren estos modos y (4) las subsecuentes desexcitaciones posibles del hijo. Se suele dar un esquema de desintegración por cada modo que presenta el padre

# 4. Fisionomía del núcleo

- Los núcleos tienen distintas geometrías (esféricos, elipsoidales, core orbitado por otros nuclones a gran distancia, etc.)
- Nosotros nos centramos únicamente en los átomos esféricos

## Radio del núcleo

- **Problema:** Como el núcleo en sí no es una esfera rígida sino que los nucleones tienen asociadas densidades de probabilidad de presencia, la frontera del mismo no está bien definida
- **Lo que se hace:** estudio la densidad de carga del núcleo normalizada con respecto a la densidad de carga en su centro en función del radio y asigno:
    - Radio del núcleo ($R$): $\rho/\rho_0=0.5$
    - Corteza superficial ($t$): empieza cuando $\rho/\rho_0=0.9$ y acaba cuando $\rho/\rho_0=0.1$ (da idea de cuán compacto es el núcleo)
- El orden de magnitud del radio son unos pocos fermis ($1fm=10^{-15}m$)

**Fórmula empírica para el radio:**

$$\boxed{R\approx R_0A^{1/3}}$$

- donde $R_0\sim 1.2fm$

**Relación entre distribución de masa y de carga:**

$$\rho_m(r)\approx\frac{A}{Ze}\rho_c(r)$$

## Estimación de la densidad media de un núcleo esférico

- Si tomamos el núcleo como una esfera rígida de radio $R$ entonces su densidad de masa media será:

$$\langle \rho_m\rangle = \frac{m}{V}$$

donde $m\approx Au$ ($u$ son unidades de masa atómica). Por su parte, $V=\frac{4\pi}{3}R^3$, por tanto:

$$\boxed{\rho_m\sim 10^{17}(kg/m^3)}$$
