**Objetivos:**

1. Construir modelo semiempírico de masa del núcleo y enteder la naturaleza de sus 5 contribuciones
2. Entender el balance energético de los procesos nucleares

# 1. Modelo semiempírico de masas / Fórmula de Bethe-Weizsäcker

## Masa nuclear ($m$) y masa atómica ($M$)

- **Observación:** un núcleo compuesto de $A$ nucleones tiene menor masa que los $A$ nucleones por separado
- **Resultado:** la energía que "falta" en masa en reposo se emplea en mantener los nucleones ligados

Por tanto, la masa de un núcleo vendrá dada por:

$$m=Zm_p+Nm_N-\frac{\Delta E}{c^2}$$

donde $\Delta E>0$ es la energía de ligadura o "Binding Energy" (BE)

**Nota:** en la práctica, con lo que se trabaja no es la masa nuclear sino la "masa atómica" o los "excesos de masa"

**Def.:** la "masa atómica" se define como $\boxed{M=Au+\Delta}$, donde $\Delta$ se llama "exceso de masa". Como la u.m.a se toma con respecto del Carbono 12, el exceso de masa para este es nulo: $\Delta(^{12}C)=0$

**En la práctica:** la masa atómica contempla asimismo la masa de los electrones y la masa correspondiente que sustraerían sus energías de ligadura, pero esta última podemos despreciarla y queda:

$$\boxed{M\approx m+Zm_e}$$

**Def.:** se define la "energía de ligadura" como la energía necesaria para disgregar un núcleo en sus $A$ nucleones constituyentes y separarlos unos de otros hasta el infinito en reposo y en el mismo nivel energético al que se encontraban

- Vemos que la masa y la BE están estrechamente ligadas, por lo que el estudio de ambas es equivalente

## 1.1 Energía de ligadura

- **Nota.:** habitualmente, se habla de la "energía de ligadura por nucleón, $BE/A$
- **IMPORTANTE:** la energía de ligadura por nucleón, aunque lo diga Wikipedia, **no es** una medida fidedigna de la estabilidad de un núcleo. Esto se debe a que esta magnitud representa la energía (por nucleón) necesaria para partir un núcleo en todos sus nucleones constituyentes, pero los procesos de desintegración ni parten el núcleo en todos sus nucleones constituyentes, ni a veces eliminan ningún nucleón (véase beta+/- y CE). Por tanto, hay que llevar cuidado cuando se relacione BE/A y estabilidad, no siempre funciona que el núcleo con más BE/A es el más estable
- **Ejemplo:**
    - BE/A($^51$Fe)$=8.46MeV/nuc$; $\tau=305ms$ (inestable)
    - BE/A($^4$He)$=7.07MeV/nuc$; $\tau=\infty$ (estable)

- Lo que es indicador de estabilidad es el BE/A con respecto a sus vecinos en la tabla nuclear

## 1.2 Las 5 contribuciones de la masa nuclear

- El objetivo ahora es encontrar una expresión empírica de la BE en función de las observaciones que hemos hecho en este tema y el anterior sobre el núcleo y su estabilidad

### 1. Volumen

- Cuanto más volumen ocupen los nucleones, más energía va a costar mantenerlos unidos

**Contribución:** $+a_VA$

**Nota sobre las contribuciones:** se dan en términos de $A$, pero es más fácil entenderlas en términos de $R$. Por ejemplo, el término de volumen va con $A$, osea que va con $R^3$

### 2. Superficie

- Los protones en la superficie del núcleo no van a notar el tirón de todos los otros nucleones como los que están en el centro, notarán menos porque tendrán menos primeros vecinos y estarán más alejados de algunos y estarán por ende menos ligados ==> cuanta mayor superficie, a más protones les pasará y menor será la BE

**Contribución:** $-a_SA^{2/3}$

### 3. Repulsión de protones

- Los protones presentan repulsión Coulombiana entre ellos, y la energía potencial de esta decae con el inverso del módulo de la distancia:

**Contribución:** $-a_cZ^2A^{-1/3}$

**Observación:** como esta contribución va con $1/R$, podríamos pensar que los núcleos más pesados tienen más BE al tener mayor radio (se les quita menos por repulsión) pero fíjate que va también con $Z^2$. Si lo graficas, sale que cuanto más grande el núcleo, más repulsión Coulombiana y por tanto menos BE (como es lógico. Esto va en concordancia con que el átomo de He sea de los más estables, sin meternos en números mágicos. La fuerza nuclear fuerte solo tiene que vencer a la repulsión de 1 único par de protones)

### 4. Asimetría

- La línea de estabilidad va aproximadamente como $Z=N$, si bien esta tendencia es más marcada en núcleos ligeros. Por tanto, favorezco esta tendencia pero lo peso con $1/A$ para que afecte más a los núcleos ligeros

**Contribución:** $-a_A\frac{(A-2Z)^2}{A}$

**Observación:** contribución es 0 para los núcleos con $N=Z$ y si no solo resta

### 5. Pairing

- En el Tema anterior vimos que hay como 70 núcleos más con $A$ par que impar. Dentro de estos, los de Z/N ambos pares estaban favorecidos y los de Z/N ambos impares estaban suprimidos. Para $A$ impar, no importaba si era Z par / N impar o al revés. Damos cuenta de ello así:

**Contribución:** $\delta A^{-1/2}$

donde:

$$\delta=\begin{cases}>0, \ \ Z,N \ \text{ambos pares} \\ <0, \ \ Z,N \ \text{ambos impares} \\ =0, \ \ \text{$A$ impar}\end{cases}$$

**Nota:** que $A$ par con Z/N ambos pares esté favorecido tiene que ver con el pairing en que, si Z/N ambos pares, se pueden formar $Z/2$ pares de protones y $N/2$ pares de neutrones, dando lugar a un estado fundamental con $J=0$, que parece que está favorecido

## Poniendo todo junto

- Nos queda entonces una fórmula semiempírica del BE como sigue:

$$\boxed{BE(Z,A) = a_VA-a_SA^{2/3}-a_cZ^2A^{-1/3}-a_A\frac{(A-2Z)^2}{A}+\delta A^{-1/2}}$$

## Comentarios sobre el modelo

1. La BE/A tiene una fuerte componente que es constante,
2. Funciona mejor para núcleos de masa intermedia y elevada que para núcleos ligeros, porque estos tienen unos picos que el modelo no predice. Sin embargo, sí que predice la tendencia general de la curva de BE/A vs. A
3. La gráfica de BE/A vs. A tiene una fuerte tendencia constante para núcleos de masa intermedia con un máximo en el hierro (Z=55) y luego baja. El modelo predice bien esta componente constante (en torno a $7,8MeV/nuc$), el máximo y el subsiguiente decaimiento de la curva

## ¿Predice bien la línea de estabilidad?

- Ahora que tenemos un modelo, queremos usarlo. Vamos a ver si, para un $A$ determinado, podemos predecir qué $Z$ es el estable

- Hagamos $A=cte.$ (nos movemos en isóbaras, que en la tabla nuclear son rectas de pendiente -1)
- **Clave:** una vez fijado $A$, la EB/A sí que se convierte en un indicador fidedigno de estabilidad, y el elemento estable será aquel que minimice $BE$ con respecto a $Z$, que es el único parámetro libre. Geométricamente, trazando una isóbara en la tabla nuclear para un $A$ dado, se trata de ver con qué $Z$ interseca en la línea de estabilidad

- Haciendo $\frac{\partial BE}{\partial Z}=0$, encontramos que la masa atómica (equivalente a hablar sobre -BE) traza parábolas en $Z$ tales que:

$$M(Z, A)=A(m_N-\frac{1}{c^2}(a_V+a_SA^{-1/3}+a_A))+\boldsymbol{Z}(m_p+m_e-m_N-4a_A/c^2)+\boldsymbol{Z}^2\frac{1}{c^2}(a_CA^{-1/3}+4a_C/A)+\delta(A)$$

**IMPORTANTE:** si A impar, entonces $\delta=0$ y obtenemos una parábola. Si A par, puede ser que Z/N ambos pares o ambos impares, de modo que obtenemos **dos** parábolas coincidentes pero la impar-impar está elevada --> esto replica bien el comportamiento de la naturaleza, donde impar-impar está muy suprimido y par-par muy favorecido

![Desdoblamiento](Desdoblamiento_masa_atomica.png)

**Nota:** fíjate que, aunque nuestro modelo prediga un solo mínimo, puede ser que dos núcleos estén suficientemente cerca de este para ser ambos estables. Esto explica por qué la línea de estabilidad no es exactamente una línea sino que para una misma isóbara puede haber dos elementos estables

**Nota:** fíjate que haciendo esto de las isóbaras podemos reconstruir la línea de estabilidad tomando un valor de $A$ cada vez, y que no hemos hablado de desintegraciones. Finalmente, entiendo que podría hacerse lo mismo tomando $Z=cte.$ (líneas horizontales y ver dónde cortan) o bien $N=cte.$ (líneas verticales), aunque no sé si para estas dos condiciones se mantendría el criterio de mínima $EB$ ==> mayor estabilidad

**Nota:** fíjate que, técnicamente, nuestro método de maximización de EB no garantiza estabilidad, solo garantiza que el núcleo al que hemos llegado es el más estable de su isóbara. Sin embargo, siempre conseguimos llegar a un elemento estable. ¿Cómo puede ser esto? Es porque, salvo para núcleos muy pesados, la línea de estabilidad tiene siempre al menos un elemento estable por cada $Z$. Véase en la tabla nuclear que, pasado $Z\approx 82$, no hay más núcleos estables conocidos, y por tanto nuestro método para encontrar núcleos estables fallaría, porque para esas Z's no los hay. Como mucho, tal vez nos daría el elemento más estable de la isóbara, pero a masas tan altas entiendo que el modelo fallaría porque es casi de juguete (y no tiene en cuenta números mágicos, etc.)

# 2. Balance energético de procesos nucleares

- Consideremos la reacción general:

$$A+B\to C+D$$

- Deseamos saber si esta reacción se puede dar de forma esponténea o si requiere de intervención por parte del investigador (i.e: hay que aportarle energía). Para ello, el criterio será que los dos núcleos en reposo puedan originar los otros dos núcleos-producto asimismo en reposo. Es decir:

$$(m_A+m_B)c^2\geq (m_C+m_D)c^2$$

O bien, si llamamos factor $Q$ a la resta de estas magnitudes:

$$\boxed{Q=-\Delta mc^2}$$

donde $\Delta m=\displaystyle\sum_{prod}m_i-\displaystyle\sum_{react}m_i$

**Criterio de espontaneidad:**

- $Q \begin{cases}>0, \ \ \text{proceso podría suceder espontáneamente (``exotérmico")} \\ < 0, \ \ \text{proceso no sucede espontáneamente (``endotérmico") }\end{cases}$

**Nota:** incluso si $Q>0$, podría ser que el proceso no se dé espontáneamente porque su probabilidad, si bien no nula, podría ser arbitrariamente cercana a 0

**Nota:** los términos "exotérmico / endotérmico" no significan que se libere o se absorba calor, son más bien en analogía con las reacciones químicas. La energía se conserva en las reacciones nucleares, de modo que toda ella va a la energía de reposo + cinética + de ligadura + excitaciones de los productos

## Relación de factor $Q$ y energía cinética de los productos

- Si imponemos la conservación de energía en la reacción general de arriba, obtenemos:

$$-\Delta mc^2=\Delta T$$

donde $\Delta T$ es la diferencia de las energías cinéticas de los productos y de los reactivos. Entonces, el factor $Q$ también se puede calcular de la forma:

$$Q=-\Delta mc^2=\Delta T$$

**Nota:** véase que hemos admitido que ninguno de los hijos acaba en estado excitado. De lo contrario, habría que considerar esto también al hacer el balance de energías. Los valores tabulados de $Q$ se suelen dar admitiendo que todo está en estado fundamental, y después ya tú sumas o restas las energías correspondientes a la excitación de reactivos y productos

## Factor $Q$ en desintegraciones

**Nota:** en una desintegración solo participa un padre, de modo solo hay 1 reactivo --> si $Q<0$, no podremos conseguir dar lugar a la reacción ni aunque pongamos el núcleo padre a velocidad arbitrariamente cercana a la de la luz, porque siempre podremos ponernos en su inercial, donde vuelve a estar en reposo. Diremos entonces que la desintegración por ese modo para ese núcleo está "cinéticamente prohibida"

**Nota:** en general, querremos dejar $Q$ en términos de la masa atómica $M$, no de la nuclear $m$, porque es la primera la que está tabulada

- Estudiemos ahora el factor $Q$ de algunos de los modos de desintegración que hemos estudiado en el Tema anterior

### 1. Alpha

- Este lo haremos paso por paso y en los siguientes pondremos directamente el resultado porque todos se hacen igual

$$_Z^AX \to \  _{Z-2}^{A-4}Y+_2^4He$$

$$Q=-\Delta mc^2=(m_X-m_Y-m_{He})c^2$$

Ahora pasamos a masa atómica. Recuerda que: $M\approx m+Zm_e$

$$Q=(M_X+Zm_e-M_Y-(Z-2)m_e-M_{He}-2m_e)$$

Simplificando:

$$\boxed{Q=(M_X-M_y-M_{He})c^2}$$

### 2. Beta+

$$_Z^AX \to \  _{Z-1}^{A}Y+e^++\nu_e$$

$$\boxed{Q=(M_X-M_Y-2m_e)c^2}$$

### 3. Beta-

$$_Z^AX \to \  _{Z+1}^{A}Y+e^-+\bar\nu_e$$

$$\boxed{Q=(M_X-M_Y)c^2}$$

### 4. CE

$$_Z^AX +e^-\to \  _{Z-1}^{A}Y+\nu_e$$

$$\boxed{Q=(M_X-M_Y)c^2}$$

**Observación:** $Q_{\beta^+} <  Q_{CE}, Q_{\beta^-}$. Es posible que haya $\beta^+\Longrightarrow $ es posible que haya $\beta^-$ y CE

**Nota:** podría parecer un sinsentido que un núcleo se desintegre con probabilidad no nula tanto en beta- como en beta+, porque uno indica un exceso de neutrones y el otro un déficit de estos. Sin embargo, hay núcleos como el $^{64}Cu$ que decaen en ambos modos (con probs. 38% para beta- y 62% para beta+/CE) porque por ambas vías pueden llegar a la línea de estabilidad eventualmente

## Energía cinética de los productos

- Nos interesa ahora estudiar con qué energía cinética salen despedidos los productos de desintegración
- **Nota:** en desintegraciones con 2 productos como la alpha, la conservación de energía y de momento lineal determina unívocamente la energía cinética con la que saldrán los productos. Sin embargo, si hay 3 o más productos como en las beta, el momento lineal se puede distribuir de distintas formas entre los 3 productos

### Alpha

$$_Z^AX\to \ _{Z-2}^{A-4}Y+_2^4He$$

1. El momento lineal se conserva: $\vec p_\alpha=-\vec p_Y$
2. La energía cinética de la partícula alpha es pequeña (no relativista): $T_\alpha=\frac{\vec p_\alpha^2}{2m_\alpha}$

Por tanto, podemos escribir la energía cinética de Y en términos de la de la partícula alpha y queda:

$$Q=T_\alpha\left(1+\frac{m_\alpha}{m_Y}\right)$$

**Nota:** hemos admitido que tanto el padre X como el hijo Y están en estado fundamental. La partícula alpha siempre sale enn estado fundamental

**Aproximación:** $m_\alpha/m_Y\approx M_\alpha/M_Y$

$$Q=T_\alpha\left(1+\frac{M_\alpha}{M_Y}\right)=T_\alpha\frac{M_\alpha+M_Y}{M_Y}$$

- **Fíjate:** como las energías cinéticas en desintegraciones alpha son pequeñas, en buena aproximación la masa se conserva (como en las reacciones químicas)

$$m_Xc^2+T_x=(m_Y+m_\alpha)c^2+T_Y+T_\alpha$$

pero si $T's$ mucho más pequeñas que las energías en reposo:

$$m_X\approx m_Y+m_\alpha$$

- Entonces queda:

$$Q=T_\alpha\frac{M_X}{M_Y}$$

- Recordemos que en las desintegraciones alpha las masas atómicas cumplen esta misma igualdad porque los términos de electrones se cancelan. Finalmente, **despreciando los excesos de masa**, $M_X\approx A_Xu$. Por tanto:

$$\boxed{Q=T_\alpha\frac{A_X}{A_Y}}$$

de donde se puede determinar la energía cinética de la partícula alpha.

**Nota:** si el hijo Y acaba en un estado excitado, no toda la energía disponible de la desintegración se habrá empleado en dar energía cinética a los productos de desintegración sino que parte de ella se la habrá quedado Y "intrínsecamente" si se quiere, de modo que los productos tendrán menos energía disponible:

$$\boxed{T_\alpha=\frac{A_Y}{A_X}(Q-E^*)}$$

donde $E^*$ es la energía del nivel excitado del hijo (recuerda que se toma su nivel fundamental como 0)

**Resultado:** la partícula alpha en una desintegración de este tipo podrá salir con tantos valores distintos de energía cinética como niveles excitados tenga el núcleo hijo (y que den valores de energía cinética no nulos)

### Beta-

$$_Z^AX\to \ _{Z+1}^AY+e^-+\bar \nu_e$$

(la disertación también vale para beta+)

- Inicialmente, se creía que la desintegración beta daba 2 productos al igual que la alpha: un núcleo hijo y un electrón / positrón. Sin embargo, se observó que el electrón no salía con valores de energía cinética cuantizados, como corresponde a una desintegración de 2 productos como hemos visto arriba, sino que se observaba que cada electrón salía con la energía cinética que quería
- Se hizo una gráfica Energía de salida - Número de electrones con esa energía de salida y salió un continuo
- **Propiedad:** el continuo no seguía hasta E infinita sino que se paraba en un valor (la energía cinética del electrón estaba acotada superiormente, claro, porque una reacción solo tiene una cantidad limitada de energía)
- **Pauli:** postula que no es una desintegración de 2 sino de 3 productos, y que el tercero ((anti)neutrino) se está llevando parte del momento. Por eso los electrones salen en un continuo

**¿Cuál es la energía cinética máxima del electrón/positrón?:**

- Hagamos el balance de energía de la deisntegración beta-. Si me pongo en el inercial propio de $X$ (o si lo tengo en reposo en laboratorio) entonces $T_X=0$, de modo que:

$$Q=\Delta T=T_Y+T_e+T_\nu$$

pero el electrón / positrón y el neutrino son mucho más ligeros que el núcleo hijo, de modo que saldrán con mucha más energía cinética. Es por esto que puedo despreciar $T_Y$ frente a los otros términos y tengo:

$$Q\approx T_e+T_\nu$$

Por tanto, el electrón / positrón alcanzará su máxia energía cinética cuando el (anti)neutrino quede en reposo (no sucede pero es máximo), de modo que:

$$\boxed{T_{e,max}=Q}$$

Lo que explica por qué el continuo de Número de electrones con energía E vs. E tenía un límite de energía (coincidente aprox. con Q)
