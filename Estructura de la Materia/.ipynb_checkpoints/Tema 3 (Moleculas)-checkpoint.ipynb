# 1. Función de onda molecular

**Problema:** si en átomos no sabíamos resolver la ec. de Schrödinger, ahora menos. Además, ya no hay un solo núcleo, de modo que la dificultad es considerablemente mayor porque no podemos situarnos sobre él y tratarlo solo como una fuente de potencial, sino que hay que considerar los núcleos como partes del sistema. Concretamente, el Hamiltoniano es:

$$H=\displaystyle\sum_{a}\frac{\vec P_a^2}{2M}+\displaystyle\sum_{a<b}\frac{\hbar\alpha cZ_aZ_b}{R_{ab}}+\displaystyle\sum_{i}\frac{\vec p_i^2}{2m}+\displaystyle\sum_{i<j}\frac{\hbar\alpha c}{r_{ij}}-\displaystyle\sum_{i,a}\frac{Z_a\hbar\alpha c}{r_{ia}}$$

**Solución:** hago aprox. de Born-Oppenheimer y así puedo estudiar la dinámica de electrones y núcleos por separado. Entonces, la función de onda del sistema, que en general sería $\Psi(\{\vec r_i\}, \{\vec R_a\})$ ahora se puede factorizar en parte de electrones y parte de núcleos: $\Psi(\{\vec r_i\}, \{\vec R_a\})=\phi(\{\vec r_i\})F(\{\vec R_a\})$

- De acuerdo a esta aproximación, tenemos entonces ahora 2 problemas:

**Parte electrónica:**

$$H_{e^-}\phi=E_{n,R_a}\phi$$

donde la energía depende de $R_a$, que es la distancia entre los núcleos

**Parte nuclear:**

- Una vez he resuelto la parte electrónica, conozco los autovalores de su Hamiltoniano: $E_{n,R_a}$. Ahora escribo la ec. entera:

$$H\Psi(\{\vec r_i\}, \{\vec R_a\})=(H_{e^-}+H_N)\phi(\{\vec r_i\})F(\{\vec R_a\})=(H_N+E_{n,R_a})\phi(\{\vec r_i\})F(\{\vec R_a\})$$

Por tanto, podemos definir un nuevo Hamiltoniano para la parte nuclear: $H'=(H_N+E_{n,R_a})$ y resolver la parte nuclear con él para completar el problema. Véase que la parte nuclear solo se puede resolver una vez hemos resuelto la parte electrónica, porque necesitamos su espectro

## Parte electrónica

- Seguimos sin saber resolver analíticamente el problema de autovalores de la parte electrónica. En la práctica, lo que se hace es admitir la aprox. de enlace fuerte, de modo que las soluciones de la parte electrónica son combinaciones lineales de las funciones de onda de los átomos constituyentes de la molécula. Así, en $H_2^+$, tendríamos que el estado fundamental es la combinación de los estados fundamentales, que los conocemos porque serían en este caso los del Hidrógeno (en general, los de Hartree)

$$\phi_{1s}(r)=N\left(\Psi_{1s}(r+R/2)\pm \Psi_{1s}(r-R/2)\right)$$

**OJO!:** el origen ahora se toma entre los dos núcleos, de modo que las funciones de onda quedarán desplazadas. Concretamente:

$$\phi_{1s}(r)=\frac{N}{\sqrt{\pi a_0^3}}\left(e^{-|r+R/2|/2}\pm e^{-|r-R/2|/2}\right)$$

**Nota:** recuerda que estamos con el $H_2^+$, de modo que solo hay un electrón, por eso hacemos $r_1\equiv r$.

Como podemos elegir el desfase entre ambas funciones de onda, damos lugar a dos tipos de "orbital molecular": el "enlazante" (+) y el "antienlazante" (-). En ambos, la probabilidad de presencia tiene 2 máximos, uno sobre cada núcleo. Sin embargo, el antienlazante tiene probabilidad 0 de que el electrón se encuentre a medio camino entre los 2 núcleos, de modo que no forma molécula (veremos que en general podemos tener moléculas con orbitales antienlazantes, pero esto es porque hay más enlazantes que antienlazantes)

**Notación:** en moléculas homonucleares, la paridad está bien definida, de modo que denotaremos con subíndice $g$ (gerade; par) a las funciones de onda pares de OMs y con $u$ (ungerade; impar) a aquellas con simetría impar. Asimismo, para decir que cierto orbital molecular es antienlazante, le pondremos un asterisco. Así por ejemplo, $\phi_{1s}(r)$ con el signo "+" sería $\sigma_g$, y con el signo "-" sería $\sigma_u^*$

**Importante:** como tenemos dos orbitales moleculares, la energía del nivel original en los átomos ahora se parte en 2 (una para el enlazante, que debe ser más pequeña que en los átomos originales de los que se compone la molécula para que estos prefieran formar molécula que estar separados, y otra para el antienlazante, que habrá de ser mayor que la de los átomos por separado). Concretamente:

$$E_{\pm}=E_{1s}-\alpha\hbar c\left(\frac{J\pm K}{1\pm S}\right)$$

donde $J$ "integral Coulombiana", $S$ "integral de solapamiento" y $K$ "integral de canje"

### Orbitales moleculares en moléculas diatómicas

- Hasta ahora hemos estudiado el $H_2^+$ por ser la molécula diatómica más sencilla. Veamos qué pasa en general

**Orbitales $\sigma$:** (eje alineado con el de la molécula)

1. s+s
2. $p_z+p_z$
3. $s+p_z$
4. (Atómicos sin superponer)

**Orbitales $\pi$:**

1. $p_i+p_j$ con $i,j\neq z$

**OJO!:** para que dos orbitales atómicos se puedan superponer para formar un orbital molecular, han de tener energías similares

- Podemos mostrar los orbitales moleculares que se forman mediante **diagramas de correlación**

- **Regla de ordenación de OMs:** si $Z<8$, en una superposición que da lugar tanto a orbitales $\sigma$ como $\pi$, (1) van intercambiados empezando por los $\pi$. Si no, $\sigma$ van por fuera y $\pi$ por dentro

**Nota sobre paridad:** en OMs de moléculas homonucleares, la paridad está bien definida porque en átomos lo está, y como estamos en enlace fuerte, estamos sumando orbitales atómicos --> homonuclear --> sumo 2 veces el mismo tipo de orbital atómico --> paridad sigue estando bien definida. En heteronucleares, al superoner orbitales atómicos de distinta paridad, su paridad deja de estar bien definida

**Nota sobre moléculas homonucleares:** toda molécula homonuclear tiene (1) momento dipolar eléctrico 0 y (2) enlace covalente

## Parte nuclear

- Una vez hemos resuelto la parte electrónica del problema, podemos resolver la nuclear
- Si modelamos los núcleos como 2 partículas clásicas unidas por un muelle, vemos que pueden (1) vibrar y (2) rotar

### Vibración de la molécula

- En moléculas diatómicas, los 2 núcleos solo pueden vibrar en el eje de la molécula --> aproximo a oscilador armónico simple unidimensional (reduzco problema de 2 cuerpos a 1)

- Obtengo espectro:


$$E_n=\hbar\omega_0(n+1/2)$$

donde $\omega_0$ sale, como siempre, de hacer la aprox. armónica al potencial que sienten los 2 núcleos.

**Def.:** se le llama "constante armónica" a $\hbar\omega_0$ y toma valores típicos de $0.1-0.01eV$

**Separación entre niveles vibracionales:** como los niveles de energía son equidistantes, los gammas que emita la molécula debido a transiciones entre niveles rotacionales van a ser siempre de la misma frecuencia! (Términos superiores en el desarrollo en serie dan separaciones ligeramente distintas)

**Regla de selección entre niveles vibracionales:** $\boxed{\Delta n=\pm 1}$

**IMPORTANTE:** como el nivel fundamental de un oscilador no es 0, **los núcleos jamás se encontrarán en la separación óptima, sino oscilando en torno a ella**. Por tanto, el sistema está menos ligado de lo que parecía, y la profundidad del potencial de los núcleos se corrige como:

$$D_0=D_E-\frac{1}{2}\hbar\omega_0$$

donde $D_0$ es la profundidad del potencial considerando esta corrección y $D_E=(V(R_0))$ es antes de considerarla

**Nota sobre la energía de disociación de un enlace iónico:** si deseamos romper una molécula con enlace iónico, debemos darle una energía tal que llevemos los átomos constituyentes al infinito. Si permitimos que sigan estando en forma iónica, entonces la energía de disociación del enlace habrá sido $D_0$. Si además queremos que estén neutros, necesitamos quitarle el electrón extra al anión (tengo que aportar energía para quitárselo) y darle ese electrón al catión (dependiendo del catión, gano o pierdo energía al hacerlo):

$$\boxed{D_{0}^{neutr}=D_0^{iones}-(E_I+E_A)}$$

donde:

- $E_I>0$: energía de ionización (energía necesaria para arrancar el $e^-$ menos ligado de un átomo neutro)
- $E_A$: afinidad electrónica (energía liberada por un átomo neutro al capturar un $e^-$)

**IMPORTANTE:** $D_0^{neutr}<D_0^{iones}$. Cuesta menos separarlos como átomos neutros que como iones

### Rotación de la molécula

- Un sólido rígido en rotación (libre) con momento de inercia $I$ tiene un Hamiltoniano:

$$H_{rot}=\frac{N^2}{2I}$$

donde $N$ es el momento angular orbital del sistema de 2 partículas que conforman los núcleos

**Nota:** $I=\mu R_0^2$, donde $\mu$ es la masa reducida del sistema de 2 núcleos

- Este Hamiltoniano tiene espectro:

$$E_N=\frac{\hbar^2}{2I}N(N+1), \ \ N=0,1,2,\ldots$$

**Def.:** la "constante rotacional" se define como $B=\frac{\hbar^2}{2I}$ y toma valores típicos de $10^{-4}-10^{-3}eV$ (**niveles rotacionales son más pequeños que vibracionales**)

**Separación entre niveles rotacionales:** (ahora sí tengo varias líneas)

$$\boxed{E_{N+1}-E_N=2B(N+1)}$$

**Gammas de dos niveles rotacionales contiguos están equiespaciadas:** sabemos calcular la energía de la gamma que sale de la transición de un nivel a otro, por ejemplo (3)$\to$(2). La siguiente gamma que registre el espectroscopio será (2)$\to$(1). Vemos que la distancia (en energías) entre estas dos gammas es de $\boxed{2B}$:

$$E_{\gamma,2}-E_{\gamma,1}=E_{N+1}-E_N-(E_N-E_{N-1})$$

$$=2B(N+1)-2BN=2B$$

**Regla de selección de niveles rotacionales:** $\boxed{\Delta N=\pm 1}$

**Útil:** $\boxed{I=\mu R_0^2}$

## Clasificación de espectros de una molécula

- Hay 3 tipos, que clasificamos en orden de energía **creciente**

1. Rotacional puro: dentro de un mismo nivel electrónico y dentro de un mismo nivel vibracional, salto de un nivel rotacional a otro ($10^{-4}-10^{-3}eV$)
2. Vibro-rotacional: dentro de un mismo nivel electrónico, salto de un nivel vibracional a otro (y puede que rotacional también) ($10^{-2}-10^{-1}eV$)
3. Bandas moleculares: salto de un nivel vib./rot. de un nivel electrónico a los de otro nivel electrónico ($1-10eV$)

**Nota:** recuerda que la parte nuclear es dependiente de la electrónica, concretamente del nivel electrónico en el que nos encontremos, porque la energía de la parte electrónica va metida en el Hamiltoniano de la nuclear. Cambios en el nivel electrónico conllevan cambios en los niveles nucleares. Asimismo, cambios en la distancia relativa de los dos núcleos influyen sobre el potencial que ven los electrones y por tanto sobre la parte electrónica
