# 1. Difusión del calor

- **Nota:** no consideraremos el calor de transporte, solamente el calor aparente (el relacionado puramente a la difusión del calor por un medio, sin transporte de materia)

- Ecuación de conservación de $u$ se puede escribir como:

$$\frac{\partial}{\partial t}(\rho cT)=\vec\nabla\cdot(\kappa\vec\nabla T)$$

- Si $\rho, c, \kappa=$cte., entonces recuperamos la ec. del calor:

$$\boxed{\frac{\partial T}{\partial t}=\chi_T\nabla^2T}; \ \ \chi_T=\frac{\kappa}{\rho c}$$

- donde $c$ es el calor específico del material

**En estado estacionario:**

1. La temperatura será una función armónica

$$\boxed{\nabla^2T=0}$$

2. La producción de entropía será constante

$$\boxed{P=cte.}$$

- Esto último se debe a que $\sigma=\vec J_q\cdot \vec\nabla\left(\frac{1}{T}\right)$, pero como en estado estacionario ninguna de estas magnitudes es función del tiempo, $P=\int\sigma d^3\vec r$ tampoco lo será, así que será un número 

## Cumplimiento aprox. del teorema de Prigogine

- Sabemos que un sistema térmico jamás puede alcanzar coeficiente fenomenológico constante en general (a menos que $\kappa\sim T^{-2}$), de modo que no puede alcanzar la mínima producción de entropía en un estado estacionario
- Sin embargo, si la temperatura varía poco en el espacio:

$$\frac{|T(x)-\langle T\rangle|}{\langle T\rangle}<<1 \ \forall x \in I$$

- Entonces $T(x)\approx \langle T\rangle=cte.$ y sí se cumple esta condición del teorema de Prigogine, ya que ahora

$$L_{qq}=\kappa\langle T\rangle^2=cte.$$

- Esto es, naturalmente, si $\kappa=cte.$, que será en la mayoría de casos

# 2. Conducción eléctrica

## Difusión de la carga

- Sabemos de la ley de Ohm que $\vec J_e=-\kappa_e\vec\nabla\phi$
- Sabemos también que la ecuación de continuidad para la carga establece

$$\frac{\partial \rho_e}{\partial t}=-\vec\nabla\cdot\vec J_e$$

- Introduciendo la ley de Ohm en esta ecuación, vemos que la carga se difunde de la forma

$$\frac{\partial \rho_e}{\partial t}=\kappa_e\nabla^2\phi$$

- De la ley de Gauss sabemos que $\nabla^2\phi=-\frac{\rho}{\varepsilon_0}$ en el vacío (o materiales no dieléctricos, admito conductor)

- Entonces, tenemos la ecuación diferencial para la densidad de carga:

$$\frac{\partial\rho_e}{\partial t}+\frac{\kappa_e}{\varepsilon_0}\rho=0$$

- Que tiene por solución:

$$\boxed{\rho_e(\vec r, t)=\rho_0(\vec r)e^{-t/\tau}}; \ \ \tau=\frac{\varepsilon_0}{\kappa_e}$$

- De modo que el exceso de carga en el interior de un conductor se va a la superficie, resultado conocido de electromagnetismo

## La conductividad es siempre positiva

- Sabemos que $L_{ee}=\kappa_e T$
- Si introducimos esto en la producción local de entropía:

$$\sigma=L_{ee}X_e^2=\kappa_e T(\vec\nabla\phi)^2>0$$

- Por tanto:

$$\boxed{\kappa_e>0}$$

## Producción de entropía en proceso conductivo

- La producción local de entropía también se puede escribir en función del flujo:

$$\sigma=\frac{J_e^2 r_e}{T}$$

- donde $r_e=1/\kappa_e$ es la "resistividad" del material. Reagrupando términos e integrando en el volumen, queda:

$$T\int_V\sigma d^3\vec r=\int_VJ_e^2r_ed^3\vec r$$

- donde la integral de la izquierda es la producción de entropía y el de la derecha la potencia desarrollada por el conductor, que disipa en forma de calor de Joule (T=cte. porque estamos estudiando procesos directos y ahora toca conducción)

$$\boxed{TP=I^2R}$$

**Conclusión:** el paso de corriente a través de un conductor genera calor que es disipado al entorno

**Nota:** recuerda que $\int_VT\sigma d^3\vec r$ es el trabajo mínimo por unidad de tiempo que se le ha de dar al sistema para mantener el estado estacionario! Así que el sistema disipa en forma de calor de Joule este trabajo mínimo, y así se mantiene el estado estacionario

# 3. Difusión de materia

## Primera ley de Fick

- La primera ley de Fick es una ley empírica que relaciona el flujo de materia del componente i-ésimo con su concentración:

$$\boxed{\vec J_i=-D\vec \nabla c_i}$$

- donde $D$ se llama "coeficiente de difusión"

## Segunda ley de Fick

- La segunda ley de Fick describe la difusión en el sistema

$$\boxed{\frac{\partial c}{\partial t}=\vec\nabla\cdot(D\vec\nabla c_i)}$$

- Si $D=cte.$, vemos que **la concentración de cada componente obedece una ecuación de difusión, como la temperatura**

- La segunda ley de Fick sale de introducir la primera en la ecuación de balance de concentraciones **en ausencia** de reacciones químicas (producción de componentes). Recordemos que estamos con procesos directos

## Procesos desproporcionados

- Sea un proceso difusivo de dos componentes. Entonces:

$$\sigma=\vec J_1\cdot \left(\frac{-\vec\nabla\mu_1}{T}\right)+\vec J_2\cdot \left(\frac{-\vec\nabla\mu_2}{T}\right)$$

- Por ser flujos difusivos de materia, se cumple:

$$\boxed{M_1\vec J_1+M_2\vec J_2=0}$$

donde $M_i$ es la masa molar del i-ésimo flujo difusivo de materia

- Además, según la ecuación de Gibss-Duhem:

$$\boxed{\displaystyle\sum_{i=1}^N n_id\mu_i=-SdT+VdP}$$

- Como consideramos proceso isotérmico e isobárico, podemos dividir por el volumen el número de moles y obtener las concentraciones correspondientes, de modo que:

$$\displaystyle\sum_{i=1}^Nc_i\vec\nabla\mu_i=0$$

- De este modo, **en un sistema difusivo de materia de 2 componentes, ni los flujos ni las concentraciones son independientes**

$$\vec J_2=-\frac{M_2}{M_1}\vec J_1; \ \ \vec\nabla\mu_2=-\frac{c_1}{c_2}\vec \nabla\mu_1$$

- La producción local de entropía queda entonces:

$$\sigma=\vec J_1\cdot\left(1+\frac{M_1c_1}{M_2c_2}\right)\left(\frac{-\vec\nabla\mu_1}{T}\right)$$

**Si la densidad del componente 1 es mucho menor que la del componente 2 (ya sea por menor masa molar o menor concentración), entonces el proceso difusivo depende exclusivamente (aprox.) del componente de menor concentración**

$$\sigma\approx \vec J_1\cdot \left(\frac{-\vec\nabla\mu_1}{T}\right)$$

## Procesos ideales

- Debemos ahora relacionar el coeficiente fenomenológico $L_{mm}$ con el coeficiente de difusión (admitimos que estamos en proceso desproporcionado)

- Comparando la primera ley de Fick con la ecuación constitutiva correspondiente, tenemos:

$$D=\frac{L_{mm}}{T}\left(\frac{\partial \mu_1}{\partial c_1}\right)$$

- donde se ha admitido que el potencial químico depende solo de la concentración, así que: $\vec\nabla\mu_1=\left(\frac{\partial \mu_1}{\partial c_1}\right)\vec\nabla c_1$

- **Para gases ideales con N componentes** se cumple

$$\boxed{\mu_i(T,P,c_i)=\mu_i^0(T,P)+RT\log v_i+RT\log c_i=\mu_i^0(T,P)+RT\log X_i}$$

- donde:

$$\mu_i^0(T,P)=\mu_i^{\ \ \bullet}(T)+RT\log P$$

y $\mu_i^{\ \ \bullet}(T)$ es el "potencial químico estándar" (tabulado)

- Por tanto, tenemos que $\frac{\partial \mu_i}{\partial c_i}=\frac{RT}{c_i}$

- De aquí directamente:

$$\boxed{D=L_{mm}\frac{R}{c_i}}$$

**Nota:** llamaremos "proceso ideal" a todo aquel en el que las fuerzas intermoleculares sean razonablemente débiles, de modo que podemos aproximar su comportamiento difusivo de materia al de un gas ideal

# 4. Procesos químicos

## Consideraciones previas

- La ley de balance de concentraciones establece que la concentración varía localmente por flujos de materia y por producción (reacciones químicas generando más cantidad del componente estudiado)

$$\frac{\partial c_i}{\partial t}=-\vec\nabla\cdot\vec J_i+\displaystyle\sum_{j=1}^\alpha\nu_{ik}\mu_k$$

- Como estamos con procesos directos, nos interesa estudiar las reacciones químicas sin difusión de materia, por lo que nos quedamos solo con el segundo término de la derecha
- **Nota:** admitiremos que las concentraciones no dependen del espacio, solo del tiempo. En la práctica, esto se puede conseguir agitando el reactor para homogeneizar la mezcla y por ende quitar la dependencia espacial de las concentraciones

- **Nota:** las reacciones químicas tienen lugar en el rango de temperaturas de $10-10^4(K)$

## Convenios de las reacciones químicas

- Las reacciones químicas se producen por colisiones, y no es usual que haya más de 2 partículas colisionando (a veces hay 3, pero no más)
- La reacción más general se puede escribir de la forma:

$$\nu_1 B_1+\nu_2 B_2\rightleftharpoons \nu_3B_3+\nu_4B_4$$

**Convenio 1:** a lo que está en la izquierda se le llama "reactivos" y sus coeficientes estequiométricos $\nu_1, \nu_2$ serán negativos en nuestros cálculos. Lo que está a la derecha se llama "productos" y sus coeficientes estequiométricos $\nu_3, \nu_4$ serán positivos en nuestros cálculos (todo esto incluso si la reacción se produce en ambas direcciones)

## Grado de avance de una reacción ($\xi$)

**Ley de las proporciones constantes:**

- La variación de la cantidad de moles (i.e: de la concentración) de cada componente pesado por su coeficiente estequiométrico es igual para todos los componentes, y viene dada por la variación del grado de avance

$$\boxed{\frac{dn_{1}}{\nu_1}=\frac{dn_{2}}{\nu_2}=\frac{dn_{3}}{\nu_3}=\frac{dn_{4}}{\nu_4}=d\xi}$$

**Def.:** el grado de avance $\xi(t)$ es una función únicamente del tiempo cuya variación indica en qué dirección sucede la reacción en promedio (se puede dar en ambos sentidos pero en uno más que en otro hasta alcanzar el equilibrio)

$$d\xi\begin{cases}>0\to\text{"Reacción directa" (izda. a dcha.)}
\\ < 0\to\text{"Reacción inversa" (dcha. a izda.)}
\\
=0 \to \text{Equilibrio}\end{cases}$$

**Nota:** de la ley de proporciones constantes se deriva:

$$\boxed{dn_k=\nu_kd\xi}$$

## Afinidad de una reacción ($A$)

- Podemos entender una reacción química como un proceso fuera del equilibrio en un sistema abierto multicomponente
- Entonces, la el potencial de Gibbs queda de la forma:

$$dG=-SdT+VdP+\displaystyle\sum_{k=1}^N\mu_k dn_k$$

- Como consideramos proceso isotermo e isobárico (estamos con procesos directos así que no hay ninguna otra fuerza termodinámica más que la que induce las reacciones químicas), tenemos:

$$dG=\displaystyle\sum_{k=1}^N\mu_k dn_k=\left(\displaystyle\sum_{k=1}^N\mu_k \nu_k\right)d\xi < 0$$

- Porque se debe alcanzar un mínimo de $G$

**Def.:** se define la "afinidad de una reacción química" $A$ como:

$$\boxed{A:=-\displaystyle\sum_{k=1}^N\mu_k \nu_k=-\Delta G}$$

- **Nota:** Decíamos que $d\xi>0$ quiere decir que la reacción es directa. Como $dG<0$, entonces en una reacción directa se cumple $A>0$. Razonando así, tenemos otro criterio para el sentido de avance de una reacción

**Criterio 2 para el sentido de avance de una reacción:**

$$A\begin{cases}>0\to\text{"Reacción directa" (izda. a dcha.)}
\\ < 0\to\text{"Reacción inversa" (dcha. a izda.)}
\\
=0 \to \text{Equilibrio}\end{cases}$$

**Nota:** ahora se entiende por qué decíamos que cuando $\Delta G<0$ la reacción puede suceder de manera espontánea, porque $A>0$

## Constante de equilibrio

**Def.:** se define la constante de equilibrio $K$ de una reacción química de la forma:

$$K=-\frac{\displaystyle\sum_k\nu_k\mu_k^0(T,P)}{RT}$$

- Esto viene de escribir la afinidad en términos de potenciales químicos y fracciones molares, de modo que podemos reescribirla de la forma:

$$A=RT\log\left(\frac{K}{\displaystyle\prod_{k=1}^N(X_k)^{\nu_k}}\right)$$

### Ley de acción de masas

- Como en el equilibrio es $A=0$, se debe cumplir:

$$\boxed{K_{eq}=\displaystyle\prod_{k=1}^N(X_k)^{\nu_k}}$$

- O bien haciendo $X_k=c_k v$ con $v=V/N$, tenemos la ley de acción de masas en función de las concentraciones:

$$\boxed{K_{eq}=\displaystyle\prod_{k=1}^N(c_k)^{\nu_k}}$$

**Nota:** no aparece el término de volumen molar porque por estequiometría los exponentes siempre deben sumar 0 (para que la reacción esté bien ajustada)

### Expresión tabulada de la afinidad

- La afinidad se puede escribir entonces separando el logaritmo:

$$\boxed{A=A^0-RT\log\left(\displaystyle\prod_{k=1}^N(c_k)^{\nu_k}\right)}$$

donde $A^0=RT\log K$ está tabulado

## Velocidad de reacción

- Las reacciones suceden, en general, en ambos sentidos, pero en cada sentido con una velocidad

**Notación:**

- $r_f$ (forward) será la velocidad de reacción de izquierda a derecha (**directa**)
- $r_r$ (reverse) será la velocidad de reacción de derecha a izquierda (**inversa**)
- $r_n=r_f-r_r$ (net) será la velocidad neta de la reacción (hacia qué lado avanza en promedio) (**neta**)

### Modelo cinético de las reacciones químicas

- Las reacciones químicas (directas) suceden por colisiones entre dos o tres partículas (no más, es muy improbable)
- El modelo cinético admite que la velocidad de reacción de un proceso directo viene dada por la **ecuación cinética**:

$$r:=k(c_A)^{m}(c_B)^{n}$$

donde:

- $k$ es la "constante de velocidad" de la reacción en ese sentido
- $c_i$ es la concentración de los **reactivos** (para $r_f$ serán los reactivos y para $r_r$ los productos)

**Def.:** la "molecularidad" de una reacción es el número de moléculas que participan en un choque reactivo (en general, esto no será representativo de la reacción real porque las reacciones no tienen por qué ser directas, pueden tener mecanismos de reacción (reacciones en cadena))

**Def.:** el "orden de reacción" de un componente es el exponente que lleva en la ecuación cinética. Para reacciones directas, coincide con el coeficiente estequiométrico (**siempre tomándolo positivo**)

- Entonces, para la reacción química general (directa):

$$\displaystyle\sum_{i}\nu_i R_i\rightleftharpoons \displaystyle\sum_j\nu_jP_j$$

Tendremos que las velocidades serán:

$$r_f=k_f\displaystyle\prod_{i}([R]_i)^{\nu_i}$$
$$r_r=k_r\displaystyle\prod_j([P]_j)^{\nu_j}$$
$$r_n=k_f\displaystyle\prod_{i}([R]_i)^{\nu_i}-k_r\displaystyle\prod_j([P]_j)^{\nu_j}$$

**MUY IMPORTANTE:** en esta definición de las velocidades de reacción, se toman todos los coeficientes estequiométricos como positivos

- Como en el equilibrio $r_n=0$, esto permite hacer la asociación entre constantes de velocidad y la constante de equilibrio:

$$\boxed{\left(\frac{k_f}{k_r}\right)_{eq}=\frac{\displaystyle\prod_{j}([P]_j)^{\nu_j}}{\displaystyle\prod_{i}([R]_i)^{\nu_i}}\equiv K_{eq}}$$

**Explicación:** tal y como se define la constante de equilibrio, los coeficientes estequiométricos de la derecha se toman positivos y los de la izquierda negativos. Como cuando trabajamos con velocidades de reacción se toman todos positivos, en equilibrio queda precisamente el cociente de productos y reactivos. En la definición de $K_{eq}$ quedará lo mismo, porque los productos tienen coeficientes estequiométricos positivos y los reactivos, negativos

## Relación afinidad-velocidad de reacción

- Afinidad y velocidad de reacción inducen el mismo criterio del sentido de avance de la reacción, así que cabe pensar que están relacionadas

$$\boxed{A=RT\log\left(\frac{r_f}{r_r}\right)}$$

- Entonces, la velocidad neta de la reacción se puede escribir en términos de la afinidad según:

$$\boxed{r_n=r_f\left(1-e^{-A/RT}\right)}$$

## Aplicando el formalismo TPI a las reacciones químicas

- Sea una única reacción química, entonces la producción local de entropía vendrá dada por:

$$\boxed{\sigma=r_n\frac{A}{T}}$$

donde $X=A/T$ y $J=r_n$

- Si se cumple que $A/RT << 1$, podemos hacer Taylor a primer orden en $r_n$ y obtenemos una relación lineal entre $r_n$ y $A$ --> esta es la condición de validez de la hipótesis lineal en reacciones químicas!

$$r_n\approx \frac{r_f}{RT}A$$

- Comparando con la ecuación fenomenológica:

$$J=LX\to r_n=L\frac{A}{T}$$

- identificamos el coeficiente fenomenológico:

$$\boxed{L=\frac{r_f}{R}}$$

- Y la producción local de entropía se puede escribir en función de las velocidades de reacción o del grado de avance

$$\sigma=R(r_f-r_r)\log\left(\frac{r_f}{r_r}\right)=\frac{A}{VT}\frac{d\xi}{dt}$$

## Relación afinidad-grado de avance

- Podemos encontrar una expresión para el grado de avance en función de la afinidad de la reacción desarrollando $A$ a primer orden en $\xi$ y tomando que el grado de avance inicialmente es 0. Entonces, se llega a la expresión:

$$\boxed{\xi(t)=\xi_{eq}\left(1-e^{-t/\tau}\right)}$$

donde $-\frac{1}{\tau}=\frac{V}{RT}r_{f,eq}\left(\frac{\partial A}{\partial \xi}\right)_{eq}$

- Vemos entonces que las reacciones químicas se pueden entender como procesos de relajación, lo que encaja perfectamente con el uso de nuestro formalismo (el tiempo de relajación microscópico es mucho menor que el tiempo de relajación macroscópico, porque las colisiones entre moléculas toman un tiempo del orden de $10^{-10}s$)
