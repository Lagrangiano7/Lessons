**Objetivo:** hasta ahora hemos estudiado únicamente sistemas en equilibrio o en estado estacionario. Queremos ahora un criterio para analizar su estabilidad.

# Sistemas homogéneos

- **Def.:** llamaremos "sistema homogéneo" a aquel cuyas variables solo dependen del tiempo, no de la posición (un ejemplo típico son las reacciones químicas donde se ha homogeneizado la mezcla agitando el recipiente)

- Tenemos en general $N$ variables $\{x_k(t)\}_{k=1}^N$, y sabemos cómo evolucionan en el tiempo porque podemos plantear sus ecuaciones con el formalismo desarrollado. Por tanto, tenemos en general el sistema de EDOs de primer orden:


$$\begin{cases}\frac{dx_1}{dt}=f_1(x_1, \ldots, x_N) \\ \quad \quad \ \vdots \\ \frac{dx_N}{dt}=f_N(x_1, \ldots, x_N)\end{cases}$$

**Def.:** se define un "punto estacionario" del sistema como aquel que no evoluciona en el tiempo. Es decir, dados los valores $(x_1, \ldots, x_N)=(x_1^0, \ldots, x_N^0)$, esta tupla permanecerá inmutable en el tiempo.

**Observación:** por construcción de estado de equilibrio y de estado estacionario, ambos son puntos fijos (independientemente de si el sistema es homogéneo o no, porque no evolucionan en el tiempo). Por tanto, la ecuación para encontrar puntos fijos es:

$$f_k(x_1, \ldots, x_N)=0$$

Esto es, los puntos fijos son el kernel de la matriz de coeficientes.

## Atractores

**Def.:** en el espacio de fases, un "atractor" es una región ($A\subset \Gamma$) hacia la que convergen las trayectorias. Hay 3 tipos de atractores:

1. Puntos fijos estables (trayectorias convergen a un punto)
2. Atractores de ciclo límite (trayectorias convergen pero se quedan orbitando un punto. Las magnitudes del sistema se quedan oscilando en el tiempo)
3. Atractores extraños: las trayectorias son irregulares y sin periodicidad

**Propiedades:** 

1. Si una trayectoria llega a un atractor, ya no puede salir nunca
2. No existen otros atractores dentro de un atractor

## Estabilidad lineal

- En general, las relaciones $f_k(x_1, \ldots, x_N)$ no son lineales. Sin embargo, si consideramos una perturbación pequeña a un punto fijo, esta sí obedecerá una dinámica lineal. Lo vemos:

- Sea $\vec x^s$ un punto fijo del sistema, y sea $\vec \varepsilon (t)$ una perturbación del sistema. El nuevo estado del sistema si lo perturbamos cuando está en un punto fijo es:

$$\vec x(t)=\vec x^s+\vec\varepsilon(t)$$

- Sabemos cómo evoluciona $\vec x(t)$, y podemos desarrollar las funciones $f_k$ a primer orden porque tenemos una perturbación pequeña en torno a un punto fijo, donde este último anula, por definición, las funciones. De este modo, queda que la perturbación tiene un comportamiento lineal en el tiempo:

$$ \frac{d}{dt}\vec\varepsilon(t)=M\vec\varepsilon(t)$$

donde $$M=\left.\begin{pmatrix}\frac{\partial f_1}{\partial x_1} & \ldots & \frac{\partial f_1}{\partial x_N} \\ \vdots & \vdots & \vdots \\ \frac{\partial f_N}{\partial x_1} & \ldots & \frac{\partial f_N}{\partial x_N}\end{pmatrix}\right |_{(x_1,\ldots,x_N)=(x_1^s, \ldots, x_N^s)}$$

es la matriz Jacobiana del sistema evaluada en el punto fijo.

La solución de esta ecuación es inmediata, y tenemos cómo se comporta la perturbación:


$$\vec\varepsilon(t)=e^{Mt}\vec\varepsilon(0)$$

**Nota:** podemos descomponer $\vec\varepsilon(0)$ en la base de autovectores de $M$, que llamaremos $B_M:=\{\vec v_1, \ldots, \vec v_N\}$, donde la acción de la exponencial de la matriz es trivial:

$$\boxed{\vec\varepsilon(t)=\displaystyle\sum_{i=1}^N c_ie^{\lambda_i t}\vec v_i}$$

**Nota:** Si algún autovalor fuera imaginario puro, tendríamos un comportamiento oscilante en la dirección de su correspondiente autovalor, con frecuencia $\omega=\lambda_k$

**Criterio de estabilidad:**

1. Todos los autovalores negativos $\to$ punto fijo es estable
2. Algún autovalor tiene parte real positiva $\to$ punto fijo es inestable en la dir. del autovector correspondiente a dicho autovalor
3. Algún autovalor tiene parte real 0 $\to$ criterio no decide, habría que considerar términos de orden superior. Podemos estar ante una bifurcación para la selección de parámetros que tenemos

**Def.:** en un sistema de 2 variables, si al menos uno de los autovalores es 0 ($\Delta=0$) o si los autovalores son imaginarios puros conjugados ($\Delta>0, \tau=0$) diremos que el punto fijo presenta "estabilidad marginal / neutra". No sabemos si es estable o inestable (corresponde al caso 3 del criterio general)

**Def.:** si para que el punto fijo presente estabilidad marginal se debe satisfacer una relación entre los parámetros del sistema, $h(\beta_1, \ldots, \beta_N)=0$, entonces decimos que esta relación es una "curva de estabilidad marginal" (si hay 2 parámetros, de lo contrario sería una hipersuperficie de estabilidad marginal en el espacio de parámetros)

**Nota:** si la perturbación crece con el tiempo en un entorno del punto fijo, diremos que este "no tiene relevancia física". Si se relaja en torno a dicho punto, el punto fijo sí tendrá relevancia física

**Práctico:** en sistemas de 2 variables, podemos calcular los autovalores de $M$ a partir de su determinante ($\Delta$) y traza ($\tau$) de la forma:

$$\boxed{\lambda_{1,2}=\frac{\tau\pm \sqrt{\tau^2-4\Delta}}{2}}$$

**Observaciones:**

- Cuando los autovalores tienen parte imaginaria no nula, además de crecimiento / decrecimiento exponencial tenemos espirales en el espacio de fases
- Cuando los autovalores son imaginarios puros, tendremos entonces atractores de ciclo límite, en los que hay espirales que ni crecen ni decrecen --> elipses

### Teorema de Hartman-Grobman

- "Si todos los autovalores de $M$ tienen $Re(\lambda_k)\neq 0$, el sistema dinámico es equivalente a su aprox. lineal en un entorno del punto fjo"

**Interpretación:** si se cumple la condición del teorema, el sistema se comporta como su perturbación (en un entorno del punto fijo).

## Clasificación de puntos fijos

1. Dos autovalores reales<br/>
   a) Dos negativos $\to$ Nodo. Asintóticamente estable
   
   b) Dos positivos $\to$ Nodo. Asintóticamente inestable

   c) Signos opuestos $\to$ Punto de silla. Una parte de la sol. tiende al punto fijo a lo largo del "espacio estable" (recta) y la otra se aleja de él por el espacio inestable
   
2. Dos autovalores complejos (forman par complejo conjugado)<br/>

   a) Ambos parte real negativa $\to$ "Foco / Punto espiral". Estable. Las órbitas tienden hacia el punto fijo formando espirales

   b) Ambos parte real positiva $\to$ Sin nombre. Inestable. Las órbitas se alejan del punto fijo formando espirales

   c) Ambos imaginarios puros $\to$ "Vórtice / Centro". Estable, pero no asintóticamente (las órbitas nunca llegan al punto fijo sino que se quedan orbitándolo)

3. Dos autovalores iguales<br/>

    a) $|a_{12}|+|a_{21}|=0\to$ Nodo propio / en estrella. Estable si $\tau<0$, inestable si $\tau>0$

    b) $|a_{12}|+|a_{21}|\neq 0\to$ Nodo degenerado. Estable si $\tau<0$, inestable si $\tau>0$

## Bifurcaciones

- **Def.:** se dice que un sistema dinámico presenta una "bifurcación" en un cierto valor $\beta_0$ de uno de sus parámetros cuando una modificación suave y pequeña del parámetro desde $\beta<\beta_0$ hasta $\beta'>\beta_0$ da lugar a un cambio cualitativo en el comportamiento del sistema, que puede ser:

1. Aparición / Desaparición de puntos fijos
2. Cambio en la estabilidad de algún punto fijo
3. Aparición / Desaparición de algunas estructuras (i.e: ciclos límite)

**Def.:** al parámetro que tiene la potestad de dar lugar a bifurcaciones se le llama "parámetro de control"

**En la práctica, ¿cómo encuentro bifurcaciones?:** Busca el valor de $\beta$ tal que al menos uno de los autovalores de $M$ tiene parte real nula (donde ya no aplica el teorema de Hartman-Grobman) y estudiar qué pasa para valores del parámetro menores y mayores que $\beta_0$. Si cambia número de puntos fijos, su estabilidad o el tipo de atractor que son, para ese valor del parámetro $\beta$ tenemos bifurcación. Si hubiera $p$ parámetros, entonces habría que encontrar la combinación $(\beta_1, \ldots \beta_p)$ que origine el mismo comportamiento que he descrito

### Bifurcación de Hopf

- **Clave:** punto fijo se desestabiliza y en su lugar aparece ciclo límite
- Supercrítica: pasamos de estado estacionario estable a ciclo límite estable
- Subcrítica: pasamos de estado estacionario estable a ciclo límite inestable

### Bifurcación de Pitchfork

- **Clave:** surgen nuevos puntos fijos (habitualmente 2) y cambia la estabilidad de todos ellos
- Suelen surgir porque soluciones de $f_k$ que antes eran complejas ahora pasa a ser reales con el nuevo valor del parámetro
- Normalmente, el nuevo punto fijo es estable mientras que el antiguo pasa a ser inestable

# Sistemas heterogéneos

- Consideremos ahora que las variables del sistema no dependen únicamente del tiempo sino también del espacio (para simplificar, lo haremos en 1 dimensión espacial, pero la generalización es trivial)
- Para 2 variables $c_1(x, t), c_2(x,t)$ tenemos ahora las ecuaciones:

$$\begin{cases}\frac{\partial c_1}{\partial t}=f_1(c_1, c_2)+D_1\frac{\partial^2c_1}{\partial x^2} \\ \frac{\partial c_2}{\partial t}=f_2(c_1, c_2)+D_2\frac{\partial^2c_2}{\partial x^2}\end{cases}$$

como podría ser la segunda ley de Fick para un sistema de dos componentes, con concentraciones $c_1$ y $c_2$. 

- Los estados estacionarios serán la mayoría de las veces homogeneos (podría haberlos también heterogeneos), de modo que no dependen del espacio $\Longrightarrow$ son los mismos que para el problema homogeneo.

- Perturbemos ahora estos puntos fijos y veamos qué sucede con la perturbación:

$$c_1(x,t)+c_1^*+\eta_1(x, t); \ \ \ c_2(x, t)=c_2^*+\eta_2(x, t)$$

- Haciendo aprox. lineal queda, como antes, que las funciones se pueden desarollar en Taylor a primer orden de la perturbación y tenemos:

$$\begin{cases}\frac{\partial \eta_1}{\partial t}=a_{11}\eta_1+a_{12}\eta_2+D_1\frac{\partial^2\eta_1}{\partial x^2} \\ \frac{\partial \eta_2}{\partial t}=a_{21}\eta_1+a_{22}\eta_2+D_2\frac{\partial^2\eta_2}{\partial x^2}\end{cases}$$

donde $a_{ij}=\left.\left(\frac{\partial f_i}{\partial c_j}\right)\right |_{(c_1^*, c_2^*)}$ son los elementos de matriz del Jacobiano evaluados en el punto fijo

- **Ansnatz:** hacemos la hipótesis de que las perturbaciones tienen forma exponencial, como las soluciones de la difusión (solo que, mientras que la difusión siempre se estabiliza, estas soluciones podrían no hacerlo porque dependen de los autovalores de $M$ exponencialmente):

$$\begin{cases}\eta_1(x,t)=a_ne^{\lambda_n t}\cos(q_nx) \\ \eta_2(x,t)=b_ne^{\lambda_n t}\cos(q_nx)\end{cases}$$

donde $q_n=\frac{n\pi}{L}$

**Nota:** el subíndice $n$ indica que solo estamos teniendo en cuenta uno de los infinitos modos que componen la perturbación general:

$$\tilde \eta_1=\displaystyle\sum_{n=0}^\infty a_n e^{\lambda_n t}\cos(q_nx)$$

$$\tilde \eta_2=\displaystyle\sum_{n=0}^\infty b_n e^{\lambda_n t}\cos(q_nx)$$

- La ventaja de que todo nuestro análisis sea lineal es que, si sabemos qué le pasa al modo n-ésimo, luego podemos sumar y saber qué le pasa a la perturbación general

- Introducir este ansnatz en las ecuaciones hace que pasen de ser EDPs a un sistema algebraico (lineal), que solo tiene solución no trivial si su determinante se anula. Por tanto, podemos determinar los autovalores a partir del polinomio característico:

$$\begin{vmatrix}m_{11}-D_1q_n^2-\lambda_n & m_{12} \\ m_{21} & m_{22}-D_2q_n^2-\lambda_n\end{vmatrix}=0$$

que es el mismo problema de autovalores de siempre solo que ahora introducimos el término $D_k-q_n^2$, que puede alterar la estabilidad del punto fijo

- El análisis de estabilidad es el mismo que en el caso homogeneo

**Nota:** en general, algunos modos serán estables y otros inestables. Como cada modo es independiente de todos los demás, se comportarán por separado. Entonces, los estables se relajarán hacia el punto fijo y los inestables se alejarán de él, de modo que en general las perturbaciones, cuando hay difusión, se deforman (algunos de sus modos se atenúan y otros se refuerzan)
