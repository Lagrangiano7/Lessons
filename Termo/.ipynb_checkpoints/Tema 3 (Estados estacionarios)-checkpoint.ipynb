# 1. Definición y condiciones

**Def. 1:** sea un sistema sobre el que actúan $n$ fuerzas termodinámicas independientes y en el que, en consecuencia, aparecen $n$ flujos termodinámicos. Manténganse constantes en el tiempo (pero no nulas) $k$ de estas $n$ fuerzas. Experimentalmente, se observa que los flujos correspondientes a estas $k$ fuerzas serán alcanzarán un valor constante pero no nulo, mientras que el resto de flujos se anularán (han satisfecho las fuerzas que estaban libres). Entonces, decimos que el sistema se encuentran en un "estado estacionario de orden k"

**Def. 2:** un "estado estacionario" es aquel estado termodinámico en el que las variables termodinámicas no varían en el tiempo pero sí en el espacio

**Nota:** las variables extensivas dependen del sistema en su conjunto y por tanto no pueden depender del espacio, solo del tiempo. Para cualesquiera variables $A$ extensiva y $b$ intensiva, se cumple en un estado estacionario:

$$\boxed{\frac{dA(t)}{dt}=0; \ \ \frac{\partial b(\vec r, t)}{\partial t}=0; \ \ \vec\nabla b(\vec r, t)\neq 0}$$

## Los sistemas aislados no alcanzan estados estacionarios

- Por definición de estado estacionario, la entropía total del sistema debe mantenerse constante en el tiempo

$$\frac{dS}{dt}=0\Longleftrightarrow \frac{dS_e}{dt}=-\frac{dS_i}{dt}$$

- Como la producción de entropía $P=\frac{dS_i}{dt}>0$ siempre, entonces el sistema debe intercambiar cierta entropía con el entorno por unidad de tiempo
- Recordando que

$$\frac{dS_e}{dt}=-\int_\Sigma \vec J_S\cdot d\vec \Sigma=-\int_\Sigma \left(\frac{\vec J_Q}{T}-\displaystyle\sum_{i=1}^N\frac{\mu_i}{T}\vec J_i\right)\cdot d\vec\Sigma$$

- Vemos que el sistema debe ser capaz de intercambiar materia y / o calor con el entorno para alcanzar un estado estacionario

**Conclusión:** sistema puede alcanzar estado estacionario $\Longrightarrow$ sistema no es aislado (es cerrado o abierto)

**Nota:** en retrospectiva, esta disertación me parece innecesaria. Es evidente que un sistema aislado no puede tener estados estacionarios, porque estos precisan de interacción externa para sostener las fuerzas termodinámicas constantes (i.e: mantener en el tiempo las condiciones de contorno). Además, por Física Estadística sabemos que un sistema aislado busca maximizar su entropía con respecto de sus ligaduras internas (i.e: avanzará hacia el macroestado con el mayor número de microestados asociados, siendo todos los que induzcan dicho macroestado equiprobables)

## Sustentación de un estado estacionario

- En la práctica, para conseguir un estado estacionario, debemos alimentar al sistema desde el exterior
- La potencia que hay que darle al sistema debe ser, como mínimo:

$$\boxed{\dot W=\int_VT\sigma d^3\vec r}$$

# 2. Teoremas sobre la producción de entropía del sistema

## 2.1 Teorema de mínima prod. de entropía (Prigogine)

- La producción de entropía en un estado estacionario es mínima cuando se satisfacen las siguientes condiciones:

1. Régimen lineal
2. Se satisface la relación de reciprocidad de Onsager
3. $L_{ij}=$cte.
4. Condiciones de contorno son independientes del tiempo (se ha fijado alguna fuerza termodinámica)

## Intento de demostración (mío):

- Por definición, $(L_{\alpha\beta})$ es una forma bilineal y definida positiva. Si además se satisface Onsager, es simétrica y por tanto puede representar la métrica de una variedad Riemanniana
- Los flujos vienen dados en régimen lineal por:

$$J_\alpha=L_{\alpha\beta}X^\beta$$

- La producción local de entropía vendrá entonces dada por:

$$\sigma=J_\alpha X^\alpha=L_{\alpha\beta}X^\alpha X^\beta$$

- La producción de entropía debe minimizarse, de modo que debe cumplirse que:

$$\int \sigma dt=\int L_{\alpha\beta}X^\alpha X^\beta dt$$

sea extremal. Si hacemos la identificación con el Lagrangiano de geometría diferencial:

$$g_{\alpha\beta} \to L_{\alpha\beta}$$
$$\dot u^\alpha\to X^\alpha$$

- Entonces podemos formular un Lagrangiano equivalente para el sistema termodinámico tal que:

$$\mathcal L = \frac{1}{2}L_{\alpha\beta}X^\alpha X^\beta = \frac{1}{2}(L_{11}(X^1)^2+2L_{12}X^1X^2+L_{22}(X_{22})^2)$$

- Introduciendo este Lagrangiano en las ecuaciones de Euler-Lagrange para, por ejemplo, la primera fuerza termodinámica:

$$\frac{\partial \mathcal L}{\partial u^1}-\frac{d}{dt}\left(\frac{\partial \mathcal L}{\partial \dot u^1}\right)=0$$

- El primer término es nulo si admitimos que los coeficientes de $(L_{\alpha\beta})$ son constantes como indica el teorema. Entonces, queda:

$$L_{11}X^1+L_{12}X^2=cte.$$

pero esto es precisamente el primer flujo termodinámico, $J_1$. Entonces:

$$J_1=cte.$$

- Esto aplica para todos los flujos. El estado estacionario se define como aquel en el que todos los flujos son constantes, de modo que hemos encontrado que la condición de que la entropía sea extremal en un estado estacionario (con las condiciones adicionales impuestas) es consistente con la condición de estado estacionario y por ende queda probado el teorema.

## 2.2 Proposición (mía) sobre la imposibilidad de alcanzar estado estacionario en algunos procesos

- Los procesos directos como los de carácter térmico o eléctrico siempre romperán el Teorema de Prigogine (y por ende no alcanzarán estado estacionario) a menos que sus coeficientes vayan como:

$$\kappa\propto T^{-2}; \ \ \kappa_e\propto T^{-1}$$

- Esto se debe a que, al comparar las ecuaciones constitutivas de estos procesos con sus respectivas leyes empíricas, se obtiene:

$$L_{qq}=\kappa T^2; \ \ L_{ee}=\kappa_e T$$

## 2.2 Evolución de la producción de entropía

- La producción de entropía $P$ en un sistema disminuye con el tiempo si se satisfacen las siguientes condiciones:

1. Régimen lineal
2. Se satisface la relación de reciprocidad de Onsager
3. $L_{ij}=$cte.
4. Todas las fuerzas termodinámicas deben poder escribirse como gradientes de magnitudes intensivas
5. Todas las ecuaciones de balance del sistema carecen de término de producción

- Además, si se cumplen todas estas condiciones, la variación en el tiempo de la producción de entropía de un sistema se puede calcular de la forma:

$$\boxed{\frac{dP}{dt}=2\int \left(\frac{\partial a_\alpha}{\partial M^\beta}\right)\left(\frac{\partial M_\beta}{\partial t}\right)\left(\frac{\partial M^\alpha}{\partial t}\right)d^3\vec r=2\int\left(\frac{\partial a_\alpha}{\partial t}\right)\left(\frac{\partial M^\alpha}{\partial t}\right)d^3\vec r<0}$$

donde:

- $a_i$ es la densidad de la magnitud extensiva $A$ i-ésima del sistema, asociada a la intensiva $M_i$ (i.e: densidad de energía interna para temperatura)
- $M_j$ es la magnitud intensiva en función de la cual se escriben las fuerzas termodinámicas: $X_j=-\vec\nabla M_j$
