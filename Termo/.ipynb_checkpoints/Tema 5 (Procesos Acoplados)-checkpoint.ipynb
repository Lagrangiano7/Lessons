**Def.:** un "proceso acoplado" es aquel que involucra dos procesos de distinta naturaleza (i.e: difusión de calor y conducción eléctrica) y en el que un proceso de una naturaleza dada encuentra una causa o un efecto en un proceso de la otra naturaleza (i.e: un gradiente de potencial eléctrico puede dar lugar a un flujo de calor (efecto Peltier))

- En la práctica, un proceso acoplado es aquel en el que tienen lugar dos procesos de naturaleza distinta y los elementos de matriz fuera de la diagonal principal de $(L_{\alpha\beta})$ son no-nulos (si no, serían dos procesos simples ocurriendo a la vez, pero los dos fenómenos no estarían acoplados, no se verían uno a otro)

**Nota:** recuerda que por el principio de simetría de Curie solo se pueden acoplar (en medios isótropos, que asumiremos siempre) procesos de igual orden tensorial. Por tanto, podemos hacer 4 combinaciones:

1. Difusión de calor + conducción eléctrica: Termoeléctricos
2. Difusión de calor + difusión de materia: Termodifusivos
3. Difusión de materia + conducción eléctrica: Electrodifusivos
4. Reacción química + Reacción química: Reacciones químicas acopladas

**Ecuaciones constitutivas de 2 fenómenos acoplados:**

$$\begin{cases}J_1 = L_{11}X_1+L_{12}X_2 \\ J_2 = L_{21}X_1+L_{22}X_2\end{cases}$$

- Los términos de la diagonal corresponden al proceso directo y los de fuera diagonal a los procesos acoplados

# 1. Efectos termoeléctricos

## Definiciones de las magnitudes relevantes

**Def.:** la "conductividad térmica" se define en procesos termoeléctricos como:

$$\kappa:=-\left(\frac{J_q}{\nabla T}\right)_{J_e=0}=\frac{1}{T^2}\left(L_{qq}-\frac{L_{eq}^2}{L_{ee}}\right) \geq 0$$

**Def.:** la "resistividad eléctrica" se define como:

$$r_e:=-\left(\frac{\nabla\phi}{J_e}\right)_{\nabla T=0}=\frac{T}{L_{ee}} \geq 0$$

**Def.:** el "coeficiente de Peltier" se define como:

$$\Pi := \left(\frac{J_q}{J_e}\right)_{\nabla T=0}=\frac{L_{qe}}{L_{ee}} \ [V]$$

**Def.:** el "coeficiente Seebeck" se define como:

$$\varepsilon := -\left(\frac{\nabla\phi}{\nabla T}\right)_{J_e=0}=\frac{L_{eq}}{TL_{ee}} = \frac{\Pi}{T} \ [V/K]$$

- Donde la última igualdad (que solo se satisface si se cumple Onsager) se llama **segunda ley de Kelvin**

**Def.:** el "coeficiente Thomson" se define como:

$$\tau := \varepsilon-\frac{\partial \Pi}{\partial T}=-T\frac{\partial \varepsilon}{\partial T}$$

- Donde la segunda igualdad recibe el nombre de **primera ley de Kelvin**

## Ecuaciones constitutivas

$$\begin{cases}\vec J_q = -\kappa\vec\nabla T+\Pi\vec J_e\\\vec E = \varepsilon\vec\nabla T+r_e\vec J_e\end{cases}$$

## 1.1 Efecto Peltier

**Clave:** una diferencia de potencial en un conductor da lugar a un flujo de calor en condiciones isotermas ($\nabla T=0$)

**IMPORTANTE:** nosotros llamaremos "efecto Peltier" a esto, es decir, al segundo término de la primera ecuación constitutiva. No obstante, en todos los recursos online que he consultado, el efecto Peltier se entiende como la refrigeración Peltier

### Aplicación: refrigeración Peltier

- Este efecto de refrigeración solo ocurre cuando se juntan dos conductores con distinto coeficiente de Peltier
- Lo que se busca es enfriar el medio, no el propio cable

**Fundamento:**

- En la práctica, se toman dos placas a través de las cuales se puede hacer pasar una corriente
- Estas placas tienen distinto coeficiente de Peltier, asociado a la energía de los electrones en un medio, de modo que los electrones tendrán diferente energía en A que en B
- Pongamos que los electrones de A tienen más energía térmica que en B, pero los requerimientos de energía de B son superiores a los de A
- Entonces, al pasar de A a B, tendrán menor energía térmica --> le habremos dado a B electrones fríos --> B se enfriará
- Podría usarse también el razonamiento inverso (invirtiendo la corriente) para que A se caliente, pero hay formas mejores de generar calor
- En el fondo, los módulos Peltier son máquinas frigoríficas (les damos trabajo y absorben calor)

- En un punto de soldadura, el ritmo de absorción de calor Peltier es:

$$\boxed{\dot Q_\Pi=(\Pi_A-\Pi_B)I}$$

- donde la corriente fluye del conductor $A$ al $B$

## 1.2 Efecto Seebeck

**Clave:** una diferencia de temperaturas puede originar una diferencia de potencial en ausencia de flujo de corriente($\vec J_e=0$)

### Aplicación: termómetros de termopar

- Véase que si tenemos un conductor y le ponemos un foco caliente en un extremo y un foco frío en el otro, estaremos induciendo una diferencia de potencial (los electrones cerca del foco caliente vibrarán más rápido y la densidad electrónica en esa parte se reducirá, sucediendo lo contrario a la derecha)
- Diferentes metales tienen diferentes propiedades de conducción de la electricidad, así que si ponemos dos conductores distintos unidos por los extremos e inducimos esta diferencia de temperaturas, ambos tendrán una diferencia de potencial pero para el de arriba (por ejemplo) será mayor que para el de abajo
- Entonces, el empuje del de arriba gana al de abajo y como el circuito está cerrado, fluye corriente
- Hemos logrado originar una diferencia de potencial sin introducir corriente de forma externa


- Para encontrar la relación entre el gradiente de potencial y el de temperatura, veamos el "empuje" (la diferencia de potencial) en ambos conductores. La diferencia entre estos nos dará la temperatura

![Termopar](Termopar.png)

- La ventaja de nuesto cálculo es que podemos hallar la diferencia de potencial que origina la diferencia de temperaturas en cada cable por separado, sin cerrar el circuito (i.e: no pasa corriente), y luego compararlos

**Camino A:**

$$-\int_1^2d\phi=\int_{T_1}^{T_2}\varepsilon dT$$

- Si admitimos $\varepsilon=cte.$, cosa que en general no es cierta, entonces tenemos:

$$\Delta\phi_{12}=-\varepsilon\Delta T$$

- Si ahora vamos por el camino B, tenemos que hacer el recorrido 1-i-d-2. Al final, restamos ambas diferencias de potencial y tenemos:

$$\boxed{\Delta\phi=\phi_i-\phi_d=(\varepsilon_A-\varepsilon_B)(T_2-T_1)}$$

- Lo que se desea medir son temperaturas, no diferencias de temperatura, así que se suelen fabricar los termopares de forma que uno de los extremos esté a T cte. Si se toma esta temperatura como referencia (se toma el 0 de potencial en ella), entonces es:

$$\boxed{\Delta\phi=\varepsilon_{AB}T}$$

## 1.3 Efecto Thomson

- El efecto Joule es que un conductor eléctrico disipa energía en forma de calor en condiciones isotermas $\nabla T=0$, y siempre es positivo (siempre calienta el sistema)
- El efecto Thomson es lo mismo pero en presencia de un gradiente de temperatura ($\nabla T\neq 0$), y puede ser positivo o negativo
- El efecto Thomson es el acoplo de los efectos Peltier y Seebeck, ya que es necesario que (1) el coeficiente de Peltier sea variable (con la temperatura) y (2) que exista un gradiente de temperatura que origine un flujo de corriente

- Introduzcamos las ecuaciones constitutivas del efecto termoeléctrico en la ecuación de conservación de la energía interna y veamos qué contribuciones hay

$$\boxed{\frac{\partial u}{\partial t}=\kappa\nabla^2T-(\vec\nabla\Pi)_T\cdot \vec J_e+r_eJ_e^2+\varepsilon\vec\nabla T\cdot\vec J_e-\frac{\partial\Pi}{\partial T}\vec\nabla T\cdot\vec J_e}$$

1. La densidad de energía interna puede cambiar si la temperatura se redistribuye (se difunde) por el sistema
2. Calor Peltier (un coeficiente de Peltier variable da lugar al efecto Peltier)
3. Efecto Joule
4. Efecto Seebeck (un gradiente de temperatura origina un flujo de corriente en condiciones de potencial nulo)
5. Efecto Thomson (si el coeficiente de Peltier es variable Y ADEMÁS hay un flujo de corriente originado por un gradiente de temperatura, se intercambia un calor adicional)

# 2. Efectos termodifusivos

## Definiciones de las magnitudes relevantes

**Def.:** la conductividad térmica en procesos termodifusivos se define como:

$$\kappa:=-\left(\frac{J_q}{\nabla T}\right)_{J_m=0}=\frac{1}{T^2}\left(L_{qq}-\frac{L_{qm}^2}{L_{mm}}\right) > 0$$

**Def.:** el coeficiente de difusión se define como:

$$D:=-\left(\frac{J_m}{\nabla c}\right)_{\nabla T=0}=\frac{R}{c}L_{mm} > 0$$

**Def.:** el "calor de transferencia" se define como:

$$q^*:=\left(\frac{J_q}{J_m}\right)_{\nabla T=0}=\frac{L_{qm}}{L_{mm}}$$

y es la energía térmica transportada por un mol de materia que se difunde en un proceso isotermo (es equivalente al calor de transporte, despeja la fórmula y lo verás)

**Def.:** el "coeficiente de difusión térmica" se define como:

$$D_T:=-\frac{J_m}{c(\nabla T)_{\nabla c=0}}=\frac{L_{mq}}{cT^2}$$

y cuantifica el flujo de materia que se genera en respuesta a un gradiente de temperatura en condiciones de concentración cte.

**Def.:** el "coeficiente de Soret" se define como:

$$s_T:=\frac{D_T}{D} \ [K^{-1}]$$

y es el cociente entre la inclinación del sistema a generar un flujo de materia cuando este es causado por un gradiente de temperatura y cuando este es causado por un gradiente de concentración

**Def.:** el "coeficiente Dufour" se define como:

$$D_F:=-\frac{J_q}{(\nabla c)_{\nabla T=0}}=\frac{R}{c}L_{qm}$$

**Nota:** igual que en el caso termoeléctrico, en el difusivo hay dos ecuaciones análogas a la primera y segunda ley de Kelvin (que salen de admitir que se cumple Onsager):

$$\boxed{D_F=q^*D=RT^2D_T}$$

## Ecuaciones constitutivas

$$\begin{cases}\vec J_q = -\kappa\vec \nabla T+q^*\vec J_m \\ \vec J_m = -cD_T\vec\nabla T-D\vec \nabla c\end{cases}$$

## 1.1 Efecto Dufour (como Peltier)

**Clave:** un gradiente de concentración puede dar lugar a un flujo de calor en condiciones isotérmicas

## 1.2 Efecto Soret (como Seebeck)

**Clave:** un gradiente de temperatura puede dar lugar a un flujo de materia en condiciones de concentración constante

## Consideración en efecto estacionario

- Sea un sistema termodifusivo
- Sabemos que si fijamos el gradiente de temperatura y dejamos el de concentración libre, al final el sistema alcanzará un estado estacionario con $J_q=cte.\neq 0$ y con $J_m=0$. Entonces, de acuerdo a nuestras ecuaciones constitutivas, existirá un gradiente de concentración que no inducirá ningún flujo de materia:

$$\vec\nabla c=-c\frac{D_T}{D}\vec\nabla T=-cs_T\vec \nabla T$$

**Nota:** por analogía con las ecuaciones constitutivas de los procesos termoeléctricos, seguramente haya un análogo de la refrigeración por efecto Peltier o de los termopares por efeco Seebeck, pero no los estudiamos

# 3. Reacciones químicas acopladas

## 3.1 Acoplo de reacciones químicas

- **Def.:** se dice que un conjunto de reacciones químicas está acoplado si sus afinidades están relacionadas (admitimos que linealmente). Entonces, podemos expresar una en términos de todas las otras:

$$A_m=\displaystyle\sum_{k=1}^{m-1}b_{mk}A_k$$

- La producción local de entropía será entonces:

$$\sigma=\displaystyle\sum_{k=1}^{m-1}\tilde r_k\frac{A_k}{T}$$

con $\tilde r_k=r_k+r_mb_{mk}$

- Hacemos esto porque así nos quitamos 1 grado de libertad. Podemos entonces definir unos nuevos coeificientes fenomenológicos:

$$\tilde L_{ki}=L_{ki}+L_{km}b_{mk}$$

y sigue cumpliendo Onsager

**En la práctica:** diremos que dos reacciones químicas están acopladas si en ambas aparece la misma especie química (porque pelean por ella)

**Resultado:** como la producción local de entropía de cada reacción por separado se suma a un todo, podemos tener reacciones que ocurran en sentido contrario al habitual ($\sigma_i<0$) si el resto dan cuenta de ello para que $\sigma$ siga siendo positivo

## 3.2 Máquinas moleculares

- **Def.:** una "máquina molecular" es una máquina termodinámica que opera a nivel microscópico. A diferencia de la Termodinámica Clásica, donde la máquina intercambia calor, las moleculares funcionan típicamente con energía química

**OJO!:** las máquinas moleculares cogen energía de reacciones químicas y la usan para transportar materia, pero por el principio de Curie pensaríamos que esto no es posible, porque estamos acoplando un proceso químico (escalar) con uno de difusión de materia (vectorial). El truco está en que este principio solo aplica a medios isótropos, y el medio donde operan las máquinas moleculares (el cuerpo humano a nivel microscópico) dista mucho de ser isótropo, de modo que podemos acoplar lo que queramos

### Transporte de moléculas a lo largo de filamentos

- Sea desea transportar una macromolécula a lo largo de un filamento
- Saco la energía para transportar la macromolécula de sus reacciones químicas, pero como se mueve en un medio acuoso, me aparece una fuerza de arrastre (proporcional a la velocidad porque es partícula pequeña)

$$\boxed{\sigma=\frac{\mu A}{T}+vF\frac{1}{T}>0}$$

con $F$ fuerza de arrastre

### Bombas a través de membranas

- Los cationes pueden desafiar la osmosis, yendo de zonas de menor concentración a zonas donde esta es mayor (normalmente se bombean los cationes $Na^+, K^+, Ca^{2+}, H^+$)
- La energía para transportar estos cationes se obtiene una vez más de sus reacciones químicas

$$\boxed{\sigma=\frac{\mu A}{T}+J\nabla \mu\frac{1}{T}>0}$$

- Un ejemplo de este proceso es el intercambio de $Na^+$ de una célula con su medio exterior a través de su membrana. En el interior de la célula, tenemos $[Na^+]=145(mol/m^3)$, y en el exterior es $[Na^+]=12(mol/m^3)$. Sin embargo, se observa que el flujo es hacia dentro de la célula. Por sí solo, este proceso no podría suceder, porque sería $\sigma=J\nabla \mu\frac{1}{T}<0$, porque el flujo natural sería hacia el otro lado para equilibrar las concentraciones. No obstante, gracias al acoplo de este proceso con la reacción química "hidrólisis ATP", las células son capaces de retener sus cationes sodio y absorber nuevos del medio

**Nota:** nuevamente, vemos que el medio no es isótropo porque puedo distinguir la dirección dentro/fuera de la célula

## 3.4 Procesos electrocinéticos

- **Def.:** un "proceso electrocinético" es aquel en el que se acoplan fenómenos de (1) transporte de carga y (2) transporte de materia. El tratamiento es un poco diferente al del resto de procesos en tanto que los electrocinéticos se dan en medios heterogeneos y por tanto más que estudiar su producción local de entropía conviene estudiar su producción de entropía (integramos en todo el espacio para quitarnos el problema de la heterogeneidad)

- **Nota:** estudiaremos los procesos electrocinéticos admitiendo que hay un tanque (1) y un tanque (2) llenos de fluido polarizado. Entre ellos, un tanque mucho más pequeño (3). Las paredes son porosas para permitir el intercambio de partículas

- **Electro-ósmosis:** en un fluido cuyas partículas tienen momento dipolar permanente, como agua con sal disuelta, una diferencia de potencial arrastrará las cargas ==> flujo de materia. Esto a su vez genera una diferencia de presiones hidrostáticas ==> flujo de materia en sentido contrario. En el estacionario, ambos flujos de materia se igualarán y a la diferencia de presiones entre los sistemas 1 y 2 se le llamará "presión electro-osmótica"

- **Efecto Rouss:** efecto inverso. Una diferencia de presión hidrostática origina un flujo de corriente

### Ecuaciones fenomenológicas

$$\begin{cases}I=L_{11}\frac{\Delta\phi}{T} + L_{12}\frac{\Delta P}{T} \\ J_V=L_{21}\frac{\Delta\phi}{T} + L_{22}\frac{\Delta P}{T}\end{cases}$$

**Def.:** se define el "potencial de flujo" como: $\left(\frac{\Delta\phi}{\Delta P}\right)_{I=0}=-\frac{L_{12}}{L_{11}}$

**Def.:** se define la "permeabilidad electro-osmótica" como: $\left(\frac{J_V}{I}\right)_{\Delta P=0}=\frac{L_{21}}{L_{11}}$

**Def.:** se define la "presión electro-osmótica" como: $\left(\frac{\Delta P}{\Delta \phi}\right)_{J_V=0}=-\frac{L_{21}}{L_{22}}$

**Def.:** se define la "corriente de flujo" como: $\left(\frac{I}{J_V}\right)_{\Delta \phi=0}=\frac{L_{12}}{L_{22}}$

- Si se cumple Onsager, entonces se satisfacen las relaciones de Saxen, del todo análogas a las dos leyes de Kelvin:

$$\left(\frac{J_V}{I}\right)_{\Delta P=0}=-\left(\frac{\Delta\phi}{\Delta P}\right)_{I=0}$$
$$\left(\frac{I}{J_V}\right)_{\Delta \phi=0}=-\left(\frac{\Delta P}{\Delta \phi}\right)_{J_V=0}$$

### Producción de entropía


- La producción de entropía para estos procesos es (si flujos y fuerzas no dependen de la posición, solo del tiempo):


$$\boxed{\frac{dS_i}{dt}=J_V\frac{\Delta P}{T}+I\frac{\Delta\phi}{T}}$$

donde $J_V=A\displaystyle\sum J_iv_i$ se llama "flujo de volumen" y $A$ es la sección de los tanques

# 4. Eficiencia de un proceso acoplado

- Al igual que a los procesos clásicos podemos asociarles una eficiencia (habitualmente, a las máquinas termodinámicas), a los procesos acoplados también se les puede asignar
- Sea el proceso acoplado de dos pares conjugados:

$$\sigma=J_1X_1+J_2X_2$$

donde $\sigma_1=J_1X_1>0$ y ambos positivos, y $\sigma_2=J_2X_2<0$ con $J_2>0, X_2<0$. Naturalmente, el segundo proceso no podría suceder por sí solo

**Def.:** la "eficiencia termodinámica" ($\eta$) de un proceso acoplado es el cociente de las producciones de entropía individuales de ambos procesos aislados:

$$\eta=-\frac{\sigma_2}{\sigma_1}$$

- Podemos desarrollar esta expresión hasta llegar a:

$$\eta=\frac{Z\chi(q+Z\chi)}{1+qZ\chi}$$

donde:

- $\xi = \frac{X_2}{X_1}$
- $z=\sqrt{\frac{L_{22}}{L_{11}}}$
- $q=\frac{L_{12}}{\sqrt{L_{11}L_{22}}}$ llamado "grado de acoplamiento"

**Propiedad:** $q\in(-1,1)$ porque, por ser $(L_{\alpha\beta})$ definida positiva, es $det(L)>0\Longrightarrow L_{11}L_{22}-L_{12}^2>0$

- Buscamos ahora maximizar la eficiencia del proceso. Como lo que controlamos son las fuerzas termodinámicas, buscamos aquel $\chi$ que haga $\frac{d\eta}{d\chi}=0$ y llegamos a que:

$$\boxed{\chi_{opt}=\frac{\sqrt{1-q^2}-1}{qZ}}$$

$$\boxed{\eta_{max}=\left(\frac{q}{\sqrt{1-q^2}+1}\right)^2}$$

**Observaciones:**

1. $\eta\in(0,1)$
2. La eficiencia máxima de un proceso acoplado solo depende de su grado de acoplamiento
3. Si los procesos no están acoplados $q=0$, entonces $\eta_{max}=0$

**Problema que veo yo:** si $\eta_{max}=0$ y $\eta\in(0,1)$ entonces $\eta$ debería ser 0 también, pero no da 0
