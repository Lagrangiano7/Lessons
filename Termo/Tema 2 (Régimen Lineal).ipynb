# Producción local de entropía

- En general, la producción local de entropía se puede escribir como la contribución de $m$ pares conjugados de flujos termodinámicos $J$ y fuerzas termodinámicas $X$
- La fuerza (i.e: diferencia de temperaturas) es la causa, y el flujo (i.e: flujo de calor) es el efecto
- Por tanto, podemos escribir $\sigma$ de la forma:

$$\boxed{\sigma=J_\alpha X^\alpha}$$

## Identificación de flujos y fuerzas en nuestros sistemas

1. Proceso térmico
$$J=\vec J_q; \ \ X=\vec\nabla\left(\frac{1}{T}\right)$$

2. Proceso eléctrico
$$J=\vec J_e; \ \ X=-\frac{\vec\nabla\phi}{T}$$

3. Proceso de transporte de materia
$$J_i=\vec J_i; \ \ X_i=-\frac{\vec\nabla(\mu_i)_T}{T}$$

4. Proceso químico:
$$J_k= r_k; \ \ X_k=\frac{A_k}{T}$$

## Criterio de equilibrio termodinámico

- Se dice que un sistema está en equilibrio termodinámico cuando no existen en él fuerzas ni flujos termodinámicos:

$$\boxed{\sigma=0\Longleftrightarrow \text{sistema en equilibrio termodinámico}}$$

- Particularización: entiendo que se dirá que un sistema se encuentra en equilibrio térmico, por ejemplo, cuando su fuerza térmica sea 0

# Matriz fenomenológica

- Desarrollemos los flujos en función de las fuerzas termodinámicas en torno al equilibrio:

$$J_\alpha=J_\alpha(X^\beta=0)+\left(\frac{\partial J_\alpha}{\partial X^\beta}\right)_{eq}X^\beta+\ldots$$

- Pero como en el equilibrio no hay fuerzas, tampoco hay flujos (sin causa no hay efecto), de modo que el término a orden 0 es nulo. Por tanto:

$$\boxed{J_\alpha=L_{\alpha\beta}X^\beta}$$

en régimen lineal, donde $L_{\alpha\beta}=\left(\frac{\partial J_\alpha}{\partial X^\beta}\right)_{eq}$ se llama "matriz de coeficientes fenomenológicos"

**Nota:** a cada ecuación de un flujo en función de las fuerzas presentes en el sistema se le llama "ecuación constitutiva"

**Nota:** como admitimos que el régimen es lineal, los coeficientes no pueden depender de fuerzas ni flujos, pero sí de cualquier otra variable termodinámica, así como del espacio y del tiempo

## Ecuaciones empíricas

- Para cada tipo de proceso ya existen ecuaciones empíricas que relacionan flujo y fuerza de forma lineal

1. Procesos térmicos (ley de Fourier)

$$\vec J_Q=-\kappa\vec \nabla T$$

2. Procesos eléctricos (ley de Ohm)

$$\vec J_e=-\kappa_e\vec\nabla \phi$$

3. Procesos de difusión de materia (ley de Fick)

$$\vec J_i=-D\vec\nabla c_i$$

4. Procesos químicos

$$r_k=r_{k,f}\frac{A}{RT}$$

## Tipos de procesos según relación flujo-fuerza

**Def.:** un proceso se dice "directo" si aparece una fuerza que da lugar a su flujo conjugado y ninguno más

**Def.:** un proceso se dice "acoplado / efecto cruzado" si aparece una fuerza que da lugar a un flujo que no es su flujo conjugado (i.e: proceso termoiónico, termoeléctrico, piezoeléctrico, etc.)

## Propiedades de $(L_{\alpha\beta})$

### Forma bilineal definida positiva

- Veamos qué sucede si introducimos ahora esta expresión de los flujos en la producción local de entropía:

$$\boxed{\sigma=L_{\alpha\beta}X^\alpha X^\beta}$$

- Vemos que la matriz de coeficientes fenomenológicos es una **forma bilineal definida positiva** (para que $\sigma>0$)
- Por ende, el criterio de Sylvester establece que todos los menores principales son estrictamente positivos

$$L_{11}>0; \Delta_{22}>0; \ldots; det(L_{\alpha\beta}>0)$$

### Relación de reciprocidad de Onsager

- "La fuerza termodinámica de una naturaleza A da lugar a un flujo de naturaleza B de la misma forma que la fuerza termodinámica de naturaleza B da lugar a un flujo de tipo A"

$$\boxed{L_{\alpha\beta}=L_{\beta\alpha}}$$

- Por ejemplo, si tenemos un proeso acoplado en el que actúan fuerzas y flujos de dos naturalezas (por ejemplo, térmica y eléctrica) tenemos que los flujos serán:

$$J_q=L_{qq}X_q+L_{qe}X_e$$
$$J_e=L_{eq}X_q+L_{ee}X_e$$

- Pero como $L_{qe}=L_{eq}$, la diferencia de temperatura inducirá un flujo de corriente de la misma forma que la diferencia de potencial inducirá un flujo de calor

**Nota:** se puede demostrar que la relación de Onsager se cumple en la mayoría de casos por medio de argumentos de Física Estadística

### Principio de simetría de Curie

- "Los elementos de simetría de las causas han de estar presentes en las de los efectos y viceversa"

**En la práctica:** en un medio **isótropo** los flujos y fuerzas de distinta naturaleza solo se pueden acoplar si tienen el mismo orden tensorial (por ejemplo, una diferencia de temperatura, cuyo par conjugado es vectorial, no puede inducir un proceso químico, cuya par conjugado es escalar)

### Matriz inversa

- Así como los flujos se relacionan con las fuerzas mediante la matriz de coeficientes fenomenológicos, podemos encontrar la relación inversa inviertiendo $(L_{\alpha\beta})$
- A esta matriz se le llama "matriz de resistividad"

$$X^\beta=L^{\alpha\beta}J_\alpha$$

- El interés de esto es que como las fuerzas vienen dadas por gradientes en la mayoría de casos, si conozco los flujos puedo plantear ecuaciones diferenciales que me den las fuerzas
- Sin embargo, en la práctica siempre admitiremos que lo que se conoce son las fuerzas (que es lo que controlamos sobre el sistema, los estresores) y desearemos calcular los flujos. Las ecuaciones son entonces lineales algebraicas y no lineales diferenciales parciales
