# Formulación de Termo del no-equilibrio como teoría de campos

**Def.:** un sistema se dice "en equilibrio" si sus coordenadas termodinámicas son independientes del tiempo y el espacio

- Los procesos reales toman tiempo finito en suceder (no son cuasiestáticos) y producen entropía (no son irreversibles)
- Debemos encontrar un formalismo que nos permita modelarlos --> Termodinámica de los Procesos Ireversibles (TPI)

## Hipótesis del equilibrio termodinámico local

- Se admite que el sistema es una malla compuesta por muchos cubos pequeños que están en equilibrio termodinámico entre sí (salvo una variación diferencial entre cubos aledaños) --> esto implica que la variación de las variables termodinámicas es suave --> teoría de campos

- Las celdas se admiten suficientemente grandes como para poder hacer promedios significativos

**Validez:** esta hipótesis (en el fondo, el tratamiento de la termo como teoría de campos) es válida si el tiempo de relajación de cada celda es mucho menor que el tiempo que tarda el proceso en suceder a escala global en el sistema

**Nota:** cualquier proceso basado en colisiones atómicas / moleculares tiene tiempo de relajación interno de $\sim 10^{-10} (s)$, así que reacciones químicas, procesos difusivos, etc. se pueden modelar bien con este formalismo

- La hipótesis no se cumplirá en procesos que involucren polímeros grandes (cadenas de proteínas), porque tienen gran tiempo de relajación interno. Tampoco en procesos con tiempo de relajación global muy corto (ultrasonidos, ondas de choque, colisiones nucleares, etc.)

# Ecuaciones de balance y conservación

- En Termo del equilibrio tenemos variables termodinámicas que no dependen del espacio ni del tiempo. Tenemos que transformarlas en campos --> las magnitudes intensivas se transforman en campos trivialmente admitiendo dependencia espacio-temporal
- En el caso de las extensivas, definiremos densidades de estas y luego definiremos la magnitud extensiva correspondiente integrando en todo el espacio

$$T\to T(\vec r, t)$$
$$P\to P(\vec r, t)$$
$$U\to \int_V u(\vec r, t)d^3\vec r$$

**Def.:** una "ecuación de balance" es aquella que da cuenta de cómo varía en el tiempo la densidad de una magnitud extensiva. Tiene en cuenta los flujos en los que se intercambia dicha magnitud y la producción local de esta

$$\frac{\partial A}{\partial t}=F_{in}-F_{out}+P_A$$

**Def.:** una "ecuación de conservación / de continuidad" es una ecuación de balance en la que no hay término de producción local $P_A$. Esto es, la magnitud asociada se conserva en el sistema en su conjunto (i.e: masa, carga)

**Nota:** como estamos en una teoría de campos, ahora los principios de la termodinámica del equilibrio se deben formular localmente --> serán ecuaciones de balance para $u$ y $s$

## Conservación de carga

$$\boxed{\frac{\partial{\rho_e}}{\partial t}+\vec \nabla\cdot \vec J_e=0}$$

## Conservación de masa

$$\boxed{\frac{\partial \rho_m}{\partial t}+\vec \nabla \cdot (\rho\vec v)=0}$$

- donde $\vec v=\frac{\displaystyle\sum_{i=1}^NM_ic_i\vec v_i}{\displaystyle\sum_{i=1}^NM_ic_i}$ es la velocidad del centro de masas de las componentes del sistema, y $\vec v_i=\frac{1}{c_i}\vec J_i$ es la "velocidad local" del componente i-ésimo (no es la velocidad real, es como la velocidad promedio de las cargas en el modelo de Drude)


## Ec. de balance de la concentración

$$\boxed{\frac{\partial c_i}{\partial t}=-\vec \nabla\cdot \vec J_i+\displaystyle\sum_{k=1}^\alpha \nu_{ik}\mu_k}$$

- donde $\alpha$ es el número de reacciones químicas teniendo lugar en el sistema que involucran a la componente i-ésima

## Primer principio (ec. de conservación para $u$)

$$\boxed{\frac{\partial u}{\partial t}=-\vec \nabla \cdot \vec J_Q-\vec \nabla\cdot ( \phi\cdot \vec J_e)}$$

**Nota:** fíjate que no hay término de producción, porque el primer principio solo te dice que un sistema intercambia energía en forma de calor y trabajo, nunca se habla de producción interna (no tendría sentido, a menos que englobes en tu sistema los elementos que antes daban calor, pero siempre puedes formular el sistema de tal forma que esta expresión se mantenga)

- El flujo de calor $\vec J_Q$ se puede expresar como la suma de dos contribuciones:

$$\boxed{\vec J_Q=\vec J_q+\displaystyle\sum_{i=1}^nh_i\vec J_i}$$

donde:

1. $\vec J_q$ es el "flujo de calor aparente" (asociado a la difusión de calor)
2. $\displaystyle\sum_{i=1}^nh_i\vec J_i$ es el "calor de transporte" (energía térmica que porta la materia al desplazarse por el espacio)
3. $h_i=\mu_i+Ts_i$ es la entalpía molar parcial
4. $\vec J_i$ es el flujo de materia de la componente i-ésima del sistema

- Entonces, en general el calor se transporta de dos formas por un sistema: por difusión (colisiones entre partículas sin transporte neto de materia) y convección (la materia se desplaza y lleva consigo energía térmica). La radiación no la damos y la conducción se ve con el flujo de carga

## Segundo principio (ec. de balance para $s$)

$$\boxed{\frac{\partial s}{\partial t}=-\vec \nabla\cdot \vec J_S+\sigma}$$

- donde $\vec J_S = \frac{1}{T}\vec J_Q-\displaystyle\sum_{i=1}^N\frac{\mu_i}{T}\vec J_i$

**Nota:** $\sigma > 0$ en todos los puntos (e instantes)

- Por la desigualdad de Clausius: $dS>\frac{\delta Q}{T}$ (no incluyo igualdad porque solo estudio procesos irreversibles). Para conseguir la igualdad, hay que añadir algo más:

$$dS=\frac{\delta Q}{T}+\text{dS_i}$$

- Y ese "algo" ha de provenir del interior del sistema porque la interacción con el entorno ya la da el primer término. La Termodinámica del equilibrio no nos puede decir qué es ese término interno de entropía porque no sabe lo que sucede en el interior del sistema! Además, debe cumplirse:

$$\boxed{\frac{dS}{dt}=\frac{dS_e}{dt}+\frac{dS_i}{dt}}$$

donde:

- $\frac{dS_e}{dt}=-\int_\Sigma\vec J_s\cdot d\vec \Sigma$
- $\frac{dS_i}{dt}=\int_V\sigma d^3\vec r$

**Def.:** la "producción de entropía" se define como:

$$\boxed{P\equiv \frac{dS_i}{dt}=\int_V\sigma d^3\vec r}$$

**Intuitivamente:** la variación de la entropía del sistema con el tiempo tiene dos contribuciones: una por las interacciones del sistema con el entorno ($dS_e/dt$) y otra por la producción interna de entropía ($dS_i/dt$). La primera va con una integral de superficie porque el sistema solo interactúa con el entorno a través de esta. La segunda va con una integral de volumen porque la entropía se produce en el interior del sistema

- La producción local de entropía tiene varias contribuciones:

1. Difusión de calor
2. Trabajo eléctrico
3. Transporte de materia
4. Procesos químicos

- Es decir, todos los procesos que pueden tener lugar en nuestro sistema generan entropía. La producción local de entropía queda entonces de la forma:

$$\boxed{\sigma=\vec J_Q\cdot \vec \nabla\left(\frac{1}{T}\right)+\vec J_e\cdot \left(\frac{-\nabla\phi}{T}\right)+\displaystyle\sum_{i=1}^N\vec J_i\cdot \vec \nabla\left(\frac{-\mu_i}{T}\right)+\displaystyle\sum_{k=1}^\alpha r_k\frac{A_k}{T}}$$

- Pero vemos que no está escrito de forma "bonita" porque las contribuciones no son independientes! Esto es, el término $\vec J_Q$ está influenciado también por el transporte de materia. Podemos solucionarlo haciendo:

$$d\mu_i=(d\mu_i)_T-s_idT$$

- Y entonces queda:

$$\boxed{\sigma=\vec J_q\cdot \vec \nabla\left(\frac{1}{T}\right)+\vec J_e\cdot \left(\frac{-\nabla\phi}{T}\right)+\displaystyle\sum_{i=1}^N\vec J_i\cdot\left(\frac{-\vec\nabla(\mu_i)_T}{T}\right)+\displaystyle\sum_{k=1}^\alpha r_k\frac{A_k}{T}}$$

- Ahora las cuatro contribuciones de la producción local de entropía están separadas en sus respectivas naturalezas

# Transformadas galileanas

- Los flujos en los que estén involucradas cargas / partículas con masa serán sujetos a transformadas Galileanas que los alteren (el flujo de calor aparente, por ejemplo, no se ve afectado)

$$\vec J_{i,ref}=\vec J_i-c_i\vec V$$
$$\vec J_{Q,ref}=\vec J_Q-\vec V\displaystyle\sum_{i=1}^Nh_ic_i$$
$$\vec J_{S, ref}=\vec J_s-\vec V\displaystyle\sum_{i=1}^Nc_is_i$$
