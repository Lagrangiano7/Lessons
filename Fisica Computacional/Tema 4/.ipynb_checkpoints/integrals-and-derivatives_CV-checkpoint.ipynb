# 4.1 Métodos para evaluar integrales de una variable

* En primer lugar nos vamos a centrar en evaluar la integral definida de una función real 
    en el sentido de Riemann:

    $$I(a,b)=\int\limits_a^b\,f(x)\,\text{d}x.$$
    
* Es equivalente a calcular el área bajo la curva de f(x) a través de las sumas de Riemann:

    $$I(a,b)=\lim_{N\to\infty}\sum_{k=0}^{N-1}\,h\,f(x_k),\quad\text{con}\quad h=\frac{b-a}{N}\quad\text{y}\quad x_k=a+ k\cdot h, \ \ \ k=0,\ldots,N-1$$

    es decir, dividir la superficie en rectángulos de altura $f(x_i)$ y anchura $h\to 0$. 
    
    
* En una ordenador simplemente estimamos la superficie haciendo que $N$ tome un valor grande:

```python
# Nota: si modificas h un factor 10, estás modificando tu error un factor 10

# Trapecio: es como interpolar con splines de orden 1 e integrar la figura resultante
```

```python
from IPython.display import Image,display
display(Image(filename ="int_rect.png", width=600, height=600))
```

```python
def I_rect(f,a,b,N):
    h=(b-a)/N
    I=0
    for n in range(N):
        I+=h*f(a+h*n)   
    return I
```

```python
def f(x):
    return x

print(I_rect(f,0,1,10000))
print(1/2)
```

* Sin embargo, ya hemos visto en el ejemplo 2.15 que es una aproximación imprecisa.

   Se requiere valores grandes de $N$ para obtener una buena estimación.

### Ejemplo 4.2:  aplicando la regla del trapecio

* Escribir un programa que permita integrar una función dada en un intervalo aplicando la regla del trapecio.


* Aplicarlo a la función $x^4-2x+1$ en el intervalo (0,2) con N=10.


* Comparar con el resultado exacto.


* Calcular el valor de $N$ que permite obtener una precisión $\epsilon<0.01 \%$

```python
def trap_f(f, a, b, N):
    h = (b-a)/N
    I = h/2*(f(a)+f(b))
    for k in range(1,N):
        I+=h*f(a+k*h)
    return I

f = lambda x: x**4-2*x+1

def trap_data(a, b, f_a, f_b):
    base = b-a
    height_rect = f_a
    A = base*height_rect + 1/2*base*(f_b-f_a)
    return A

def trap_intervs(x, y):
    Is = np.zeros(x.size-1)
    for i in range(Is.size):
        Is[i]=trap_data(x[i], x[i+1], y[i], y[i+1])
    return (np.sum(Is), Is)
```

```python
def trap_puntos(x, y):
    # Ahora no puedo hacer el método adaptativo, trabajo con lo que hay
    N = len(x)
    I = 0
    for i in range(N-1):
        a = x[i]
        b = x[i+1]
        h = b-a
        I+=h/2*( y[i] + y[i+1] )
    return I
```

### 4.1.1.1 Error de truncación de la regla del trapecio

- Tal y como hemos visto, la integración numérica del área bajo una curva en un intervalo $I=b-a$ radica en:
    <ol>
   <li> Dividir $I$ en un conjunto $N$ de sub-intervalos $I_k=[x_{k-1},x_{k}]$ con $x_k =a + k\,h$ y $h=\left(b-a\right)/N$ su longitud.
       
    <li> Aproximar el valor de la curva en $I_k$ por funciones simples. 
        
    <li> El error de la aproximación se basa en estimar cuan bien estamos describiendo la curva en $I_k$.  
    </ol>       
    
       
       
- Cuando el intervalo $I_k$ es pequeño, podemos hacer un desarrollo de Taylor de f(x) en torno a $x_{k-1}$:

    $$f(x)=f(x_{k-1})+(x-x_{k-1})\,f'(x_{k-1})+\frac{1}{2}(x-x_{k-1})^2\,f''(x_{k-1})+\dots,$$  

  de forma que su integral es:

    $$\int\limits_{x_{k-1}}^{x_k}\,f(x)\,\text{d}x=h\,f(x_{k-1})+ \frac{1}{2} h^2\,f'(x_{k-1})+\frac{1}{6}h^3\,f''(x_{k-1})+{\cal O}(h^4)$$


- Sin embargo, es igual de correcto hacer el desarrollo en torno a $x_k$, que al integrar da un área:

    $$\int\limits_{x_{k-1}}^{x_k}\,f(x)\,\text{d}x=h\,f(x_{k})- \frac{1}{2} h^2\,f'(x_{k})+\frac{1}{6}h^3\,f''(x_{k})+{\cal O}(h^4),$$

   por lo que, tomando su media:

    $$\int\limits_{x_{k-1}}^{x_k}\,f(x)\,\text{d}x=\frac{h}{2}\,(f(x_{k-1})+f(x_k))+\frac{h^2}{4}\,(f'(x_{k-1})-f'(x_k))+\frac{h^3}{12}\,\left(f''(x_{k-1})+f''(x_{k})\right)+{\cal O}(h^4),$$

   de forma que la integral total es: 
 
    $$I(a,b)=\int\limits_{a}^{b}\,f(x)\,\text{d}x=\frac{h}{2}\sum_{k=1}^N{(f(x_{k-1})+f(x_k))}+\frac{h^2}{4}\left(f'(a)-f'(b)\right)+\frac{h^3}{12}\sum_{k=1}^N{(f''(x_{k-1})+f''(x_k))}+{\cal O}(h^4).$$


- En la aproximación rectangular, el valor de la curva en $I_k$ se aproxima por $f(x)\sim f(x_{k-1})$  
(o en general por el valor de la función en un $x_0\in I_k$).


- Por tanto:

    $$I(a,b)_{\text{rect}}=h\sum_{k=1}^N{f(x_{k-1})}\, ,$$

  y

    $$\epsilon_\text{rect}=\vert I(a,b)-I(a,b)_\text{rect}\vert= \frac{1}{2} h^2\sum_{k=1}^N{f'(x_{k-1})}\sim {\cal O}(h).$$


- En la regla del trapecio la curva se aproxima por la recta definida por el punto inicial y final del intervalo:

    $$ I(a,b)_\text{trap}=\frac{h}{2}\sum_{k=1}^N{\left(f(x_{k-1})+f(x_k)\right)}$$

  por lo que el error es 

    $$\epsilon_\text{trap}=\big\vert I(a,b)-I(a,b)_\text{trap}\big\vert =  \frac{h^2}{4}\left(f'(a)-f'(b)\right)+\frac{h^3}{12}\sum_{k=1}^N{(f''(x_{k-1})+f''(x_k))}+{\cal O}(h^4),$$


* Sin embargo, el segundo término no es otra cosa que la regla del trapezoide para la integral:

    \begin{align}
    \frac{h^2}{6}\int\limits_a^b{\text{d}x\,f''(x)}=&\frac{h^3}{12}\sum_{k=1}^N{(f''(x_{k-1})+f''(x_k))}+{\cal O}(h^4),\\
    =&\frac{h^2}{6}\left(f'(b)-f'(a)\right),
    \end{align}

    de forma que

    $$\epsilon_\text{trap}=\frac{h^2}{12}\left(f'(a)-f'(b)\right)+{\cal O}(h^4),$$

    que es la regla de Euler-Maclaurin para el error en la aproximación trapezoidal.  
    
 \textbf{Nota:} también tenemos el error de redondeo:
 $$\frac{\epsilon_M(b-a)}{\sqrt{2N}}\sqrt{\bar{f(x)}}$$


* La regla del trapecio es por tanto una aproximación de primer orden:  
    es correcta en los términos proporcionales a $h$ y el error de truncación es ${\cal O}(h^2)$.


* Por supuesto, la regla del trapecio también sufre el error de redondeo que es de orden:

    $$\sigma\left(I(a,b)_\text{trap}\right) = \frac{h}{2}\, \epsilon_M\left(\sum_{k=1}^N{\left(f(x_{k-1})^2+f(x_k)^2\right)}\right)^{1/2}\sim \frac{\epsilon_M (b-a)}{\sqrt{2 N}}\,\sqrt{\overline{f(x)^2}}.$$


* La regla del trapecio nos permite aproximar mejor y mejor una integral a medida que aumentamos el valor de N hasta llegar al punto en que el error vendrá dominado por el error de redondeo.


* Que ocurre para un valor de N:

    $$N\sim\left(\frac{(b-a)}{12\,\epsilon_M\,}\frac{f'(a)-f'(b)}{\sqrt{\overline{f(x)^2}}}\right)^{2/3}\gtrsim10^8.$$


* A partir de ese momento el error disminuye muy lentamente al precio de que el programa se hace más y más lento.

## 4.1.2 Regla de Simpson.

- La regla del trapecio aproxima el área bajo una curva sustituyendo la curva en un intervalo por líneas rectas. 


- Esto da lugar a una aproximación de primer orden. 


- Por tanto, una mejor aproximación seria sustituir cada segmento por curvas cuadráticas:

    $$A\,x^2+B\,x+C$$


- Donde ahora, para aproximar cada intervalo, necesitamos tres puntos.


- Es lo que se conoce la regla de Simpson:

El número de intervalos debe ser par porque tenemos que coger cada $2h$

Mejor que el trapecio para funciones suaves, pero peor que la cuadratura gaussiana

```python
display(Image(filename ="int_simp.png", width=600, height=600))
```

* Tomando el intervalo definido entre los puntos $x=-h$ y $x=+h$, se tiene:

    $$f(-h)=A\,h^2 +B\,h +C,\quad f(0)=C,\quad f(h)=A\,h^2-B\,h+C$$

  que implica que:

    $$A=\frac{1}{2\,h^2}\left(f(-h)-2\,f(0)+f(h)\right),\quad B=\frac{1}{2h}\left(f(h)-f(-h)\right),\quad C= f(0),$$

   y su integral es:

    $$\int\limits_{-h}^h\left(A\,x^2+B\,x+C\right)\,\text{d}x = \frac{h}{3}\left(f(-h)+4f(0)+f(h)\right)$$
    
  que puede utilizarse para cualesquiera tres puntos igualmente espaciados.


* Para estimar una integral usando el método de Simpson sólo debemos:
<br>

    * Dividir el intervalo en un número par de subintervalos. 
    
    * Usar la regla de Simpson para estimar el área de todos los pares de subintervalos adyacentes. 

    * Sumar los resultados.
    
    
* En el fondo, un método de integración numérica no es más que una interpolación: una forma de aproximar la función que queremos integrar a partir de funciones simples en un conjunto finito de puntos.       

### Ejemplo  4.3: obteniendo la regla de Simpson 

Obtener la expresión final para la aproximación de una integral por la regla de Simpson. 

$I(a, b) = \int_a^b P_2(x)dx = \frac{h}{3}\left[f(a)+4f(a+h)+f(a+2h)+f(a+2h)+4f(a+3h)+f(a+4h)+f(b-2h)+4f(b-h)+f(b)\right]$

$=\frac{h}{3}\left[f(a)+f(b)+4\displaystyle\sum_{k=1,3,5\ldots}^{N-1}f(a+kh) + 2\displaystyle\sum_{k=2,4,6\ldots}^{N-2}f(a+kh)\right]$

$\boxed{I(a,b)=\frac{h}{3}\left[f(a)+f(b)+4\displaystyle\sum_{k=1}^{N/2}f(a+(2k-1)h)+2\displaystyle\sum_{k=1}^{N/2-1}f(a+2kh)\right]}$

### Ejemplo 4.4: aplicando la regla de Simpson

1- Definir un programa que permite integrar una función en un intervalo dado con el método de Simpson.  
2- Aplicarlo para volver a integrar la función $x^4-2x+1$ del Ejemplo 3.2 con $N=10$ en el intervalo [0,2].  
3- Calcular el error relativo al aumentar $N$ y compararlo con el error obtenido con la regla del trapecio.

```python
def simpson(f, a, b, N):
    if N%2!=0:
        print("N debe ser par")
        return
    h = (b-a)/N
    I = (f(a)+f(b))
    for k in range(1, N//2): # Llego solo hasta N/2-1 (me quedo corto en 1 término de la primera sumatoria) y luego se lo añado
        I+=4*f(a+(2*k-1)*h)+2*f(a+2*k*h)
    I+=4*f(b-h)
    I=h*I/3
    return I

f = lambda x: x**4-2*x+1
I = simpson(f,0,2,10)
print(I)
```

Ahora error va con $h^4$. Puedes ver el error relativo de un método calculando desv. porcentual de un método con un h dado, multiplicar h por un factor 10 y ver en qué orden de magnitud de cambia el error (en este caso, te salen 4 0's más ==> error va con $h^4$)

### Ejemplo 4.5: el límite difractivo de un telescopio

El principal factor que limita la resolución de un sistema óptico está asociado al fenómeno de la difracción y es conocido como límite difractivo. Cuando la luz proveniente de una estrella pasa por la abertura circular de un telescopio produce un patrón difractivo consistente en un punto central rodeado por anillos concéntricos, el llamado patrón de Airy.  

```python
display(Image(filename ="airy_rings.jpg", width=400, height=400))
```

(Vista de los anillos de Airy de un láser pasando por una lente circular)

La intensidad de la luz en el patrón difractivo (difracción de Fraunhofer) viene dada por:

$$I(r)=\left(\frac{J_1(k\,r)}{ k\,r}\right)^2,\quad k=2\pi/\lambda,$$

donde $r$ es al distancia en el plano focal al centro del patrón difractivo, $\lambda$ es la longitud de onda de la luz y  $J_m(x)$ es la función de Bessel, definida por:

$$J_m(x)=\frac{1}{\pi}\int\limits_0^\pi{\cos\left(m\,\theta-x\,\sin\,\theta\right)\,\text{d}\theta}.$$

donde $J_1(x)$ satisface: $\lim_{x\to0}J_1(x)/x=1/2.$ 

1- Escribe una función que calcule el valor de la función de Bessel usando la regla de Simpson con N=1000. 


2- Representa en una misma gráfica las funciones de Bessel $J_0$, $J_1$, $J_2$ y $J_3$ como función de x, en el intervalo $[0,50]$.  


3- Escribir un segundo programa que genere una gráfica de densidad en un grid 100x100 de la intensidad de un patrón circular de una fuente luminosa con $\lambda = 500$ nm en un grid que cubra valores de $r$ hasta $1 \mu$m.

```python
import numpy as np
import matplotlib.pyplot as plt

def J(m, x):
    f = lambda t: np.cos(m*t-x*np.sin(t))
    I = 1/np.pi*simpson(f, 0,np.pi,1000)
    return I

x = np.linspace(0,50,100)
for i in range(4):
    plt.plot(J(i,x))
plt.show()
```

```python
# Podríamos aprovechar la simetría de revolución de los anillos de Airy, sería lo óptimo, pero aquí vamos a hacerlo con un meshgrid

t = np.linspace(0,500,500)

X, Y = np.meshgrid(t,t)
r = np.sqrt((X-250)**2+(Y-250)**2)

k = 2*np.pi/(500)

I = (J(1,k*r)/(k*r))**2

plt.imshow(I, origin="lower", cmap="hot")
plt.colorbar()
plt.show()
```

## 4.1.3 Estimación práctica de errores

* Las fórmulas para estimar el error que hemos visto por ahora dependen de las derivadas de f.


* Pero hay ocasiones en que esto puede no ser fácil o posible (por ejemplo si la función es el resultado de ciertas medidas en un experimento).


* Sin embargo, teniendo en cuenta que el error del método del trapecio y Simpson es ${\cal O}(h^2)$ y ${\cal O}(h^4)$, respectivamente, podemos aplicar:  
<br>

    * Obtener nuestra integral $I_1$ con $h_1=(b-a)/N_1$.
    * Volverla a integrar usando ahora $h_2=(b-a)/N_2$ y $N_2=2 N_1$, lo que proporciona $I_2$.
    * De esta forma la integral exacta $I$ será:
    
    $$I=I_1 +\epsilon\,h_1^n = I_2+\epsilon\,h_2^n,$$
    
    con $n=2$ o $4$ para la regla del trapezoide o el método de Simpson. 
  <br>
  
    * Teniendo en cuenta que $h_1=2h_2$, se obtiene que:

    \begin{align}
        \epsilon_\text{trap}=&\,\vert\epsilon h_2^2\vert=\frac{1}{3}\vert I_2-I_1\vert,\\
        \epsilon_\text{simp}=&\,\vert\epsilon h_2^4\vert=\frac{1}{15}\vert I_2-I_1\vert.\\
     \end{align}    

Si sabemos el error que queremos cometer, podemos saber de antemano el número de intervalos necesarios. Así, optimizamos el rendimiento del programa, porque solo usamos lo que necesitamos

## 4.1.4 Métodos adaptativos

* Hasta ahora hemos estudiado métodos en los que el número de intervalos esta fijado por adelantado. 


* Sin embargo, lo óptimo sería fijar $N$ a partir de la precisión que uno quiere obtener. 


* Este objetivo es el que define los métodos integrales adaptativos.


* La idea básica de estos métodos se basa en la estimación práctica de errores que acabamos de ver:

    $$\epsilon\propto \vert I_2-I_1\vert,\quad\text{con}\quad h_2 \propto h_1. $$


* El procedimiento es el siguiente:

    <ol>
        <li> Escoger un valor pequeño de intervalos $N_1$ y obtener $I_1$ con el método correspondiente.</li>
        <li> Doblar el número de intervalos $N_2=2\,N_1$ y obtener $I_2$.</li>
        <li> Estimar el error de truncación a partir de:

    $$\epsilon_2=\alpha\vert I_2-I_1\vert,$$
    
    donde $\alpha=1/3$ para la regla del trapecio y $1/15$ para el método de Simpson.</li>
    <li> Sí $\epsilon_2$ es mayor que la precisión deseada volvemos a doblar el número de puntos $N_3=2\,N_2$, obtenemos $I_3$ y calculamos el valor de $\epsilon_3$.</li>
    <li> Repetimos el punto 4 tantas veces como sea necesario hasta que:
        
    $$\epsilon_i=\alpha\vert I_{i}-I_{i-1}\vert,$$
        
    sea menos que el error deseado; siendo $I_i$ la $i$-esima estimación de la integral.</li> 
    </ol>
    
    
*  Desde un punto de vista de eficacia computacional los métodos adaptativos tienen sentido porque en la evaluación de la integral $I_i$ podemos usar el resultado de $I_{i-1}$.    

### 4.1.4.1 Método del Trapecio con paso adaptado

* Sea $N_i$ el número de intervalos de la iteración $i$-esima y $h_i=(b-a)/N_i$ su anchura, $N_{i-1}=\frac{1}{2} N_i$ y $h_{i-1}=2h_i$ será el número de intervalos y anchura de la iteración $i-1$. 


* La regla del trapecio para la iteración $i$ viene dada por: 

\begin{align}
    I_i=&h_i\left(\frac{1}{2}f(a)+\frac{1}{2}f(b)+\sum_{k=1}^{N_i-1}f(a+k\,h_i)\right)\\
       =&h_i\left(\frac{1}{2}f(a)+\frac{1}{2}f(b)+\sum_{k=2,4,\cdots}^{N_i-2}f(a+k\,h_i)+\sum_{k=1,3,\cdots}^{N_i-1}f(a+k\,h_i)\right).
\end{align}


* Es decir, separamos la suma en una para los intervalos pares y otra para los impares.


* Sin embargo, haciendo j=k/2, la suma sobre los intervalos pares puede expresarse como:

$$\sum_{j=1}^{N_i\,/2-1}f(a+2j\,h_i)=\sum_{j=1}^{N_{(i-1)}-1}f(a+j\,h_{i-1}),$$

* En el primer sumatorio usar $2h_i$ y sumar hasta $N_i/2-1$ es como usar la mitad de intervalos, que es precisamente lo que hacíamos en la iteración anterior

* Y por tanto:

\begin{align}
I_i=&\frac{h_{i-1}}{2}\left(\frac{1}{2}f(a)+\frac{1}{2}f(b)+\sum_{k=1}^{N_{(i-1)}-1}f(a+k\,h_{i-1})\right) +  h_i\!\!\!\sum_{k=1,3,\cdots}^{N_i-1}f(a+k\,h_i)\\
=& \frac{1}{2}I_{i-1}+ h_i\!\!\!\sum_{k=1,3,\cdots}^{N_i-1}f(a+k\,h_i).
\end{align}

```python
display(Image(filename ="intervals_adap.png", width=400, height=400))
```

* Esta propiedad permite definir el siguiente algoritmo:

##### Algoritmo para el método del Trapecio con paso adaptado.

1. Decidir la precisión objetivo y empezar con un valor pequeño de intervalos $N_1$. Calcular $I_1$ con la formula estándar para la regla del trapecio.  


2. Doblar el número de intervalos y usar:

    $$I_i=\frac{1}{2}I_{i-1}+ h_i\!\!\!\sum_{k=1,3,\cdots}^{N_i-1}f(a+k\,h_i)$$

   para obtener el valor de la integral. 
   
* Si no quieres dar pasos impares, entonces puedes reescribir la sumatoria de la forma: $\displaystyle\sum_{j=0}^{N_i/2-1}f(a+(2j+1)h_k)$
   
   Estimar el error aproximado por medio de:

$$\epsilon_i=\frac{1}{3}(I_i-I_{i-1})$$

3. Si el valor absoluto de $\epsilon_i$ es menor que el valor de la precisión objetivo, parar. De otra forma repetir el paso 2. 

```python
# Por fuerza bruta, solo con lo que sé de cálculo empírico de errores de truncación
from numpy import abs

def trap_adapt(f, a, b, tol):
    # Empiez con un solo intervalo que abarque todo el [a,b]
    N = 1
    h = b-a
    I1 = h/2*(f(a)+f(b))
    err = 1.0 # Error arbitrario para asegurar que hagamos al menos una iteración
    cont=1
    
    while err > tol:
        cont+=1
        # Ahora calculamos la misma integral con el trapecio para el doble de cajas (lo llamo I2)
        N*=2
        h/=2
        I2 = 1/2*(f(a)+f(b))

        for k in range(1,N):
            I2+=f(a+k*h)
        I2*=h

        err = 1/3*abs(I2-I1)
        I1=I2
    
    return (I1, err, cont)

f = lambda x: x**4

I, err,cont = trap_adapt(f, 0,1,10**(-6))

print(I, err,cont)
```

```python
import numpy as np

def trap_adapt(f, a, b, tol):
    N = 1
    h = b-a
    I1 = h/2*(f(a)+f(b)) # Aplicamos trapecio con N = 1
    err = 1.0
    count = 1
    
    while np.any(err) > tol:
        I2 = I1/2
        N*=2
        h/=2
        for k in range(1, N,2):
            I2+=h*f(a+k*h)
        err = 1/3*np.abs(I2-I1)
        I1=I2
        count+=1
    return (I2, err, count)

f = lambda x: x**2

I,_, _ = trap_adapt(f, 0, 1, 10**(-2))
print(I)
```

### 4.1.4.2 Método de Simpson con paso adaptado

* En el caso de la regla de Simpson, la integral viene dada por:

\begin{align}
  I_i=&\frac{h_i}{3}\left(f(a)+f(b)+2\!\sum_{k=1}^{N_i\,/\,2-1}\!\!\!f(a+2k\,h_i)+4\sum_{k=1}^{N_i\,/\,2}f(a+(2k-1)\,h_i)\right)\\
     =&\frac{h_i}{3}\left(f(a)+f(b)+2\!\!\!\sum_{k=1}^{N_{(i-1)}-1}\!\!\!f(a+k\,h_{i-1})+4\!\sum_{k=1,3,\cdots}^{N_i-1}\!\!\!f(a+k\,h_i)\right)\\
     =&\frac{h_{i}}{3}\left(f(a)+f(b)+2\!\!\!\sum_{k=2,4,\cdots}^{N_{(i-1)}-2}\!\!\!f(a+k\,h_{i-1})+2\!\!\sum_{k=1,3,\cdots}^{N_{(i-1)}-1}\!\!\!f(a+k\,h_{i-1})+4\!\!\!\sum_{k=1,3,\cdots}^{N_i-1}\!\!\!f(a+k\,h_i)\right)
\end{align}

* Por consiguiente, llamando:
$$S_i=\frac{1}{3}\left(f(a)+f(b)+2\!\!\!\sum_{k=2,4,\cdots}^{N_{i}-2}\!\!\!f(a+k\,h_{i})\right)\quad\text{y}\quad T_i=\frac{2}{3}\!\sum_{k=1,3,\cdots}^{N_{i}-1}\!\!\!f(a+k\,h_{i}),$$

* Se tiene la identidad:

    $$S_i=S_{i-1}+T_{i-1},$$

    y

$$I_i=h_i\left(S_i+2\,T_i\right),$$

* Por tanto, en cada nueva iteración sólo tenemos que calcular el valor de $T_i$, lo que permite definir el siguiente algoritmo:

##### Algoritmo para el método del Simpson con paso adaptado

1- Decidir la precisión objetivo y empezar con un valor pequeño de intervalos $N_1$. Calcular $I_1$ con la formula estándar para la regla de Simpson.


2- Doblar el número de intervalos y usar:

$$S_i=S_{i-1}+T_{i-1},\quad\text{y}\quad I_i=h_i\left(S_i+2\,T_i\right)$$

   con 

$$S_i=\frac{1}{3}\left(f(a)+f(b)+2\!\!\!\sum_{k=2,4,\cdots}^{N_{i}-2}\!\!\!f(a+k\,h_{i})\right)\quad\text{y}\quad T_i=\frac{2}{3}\!\sum_{k=1,3,\cdots}^{N_{i}-1}\!\!\!f(a+k\,h_{i}).$$

para calcular el valor de la integral. Estimar el error aproximado por medio de:

$$\epsilon_i=\frac{1}{15}(I_i-I_{i-1}).$$


3- Si el valor absoluto de $\epsilon_i$ es menor que el valor de la precisión objetivo, parar. De otra forma repetir el paso 2. 

```python
# Fuerza bruta
import numpy as np

def simpson_adapt(f, a, b, tol):
    N=2
    h=(b-a)/2
    I1 = h/3*(f(a)+f(b) + 4*f(a+h))
    err = 1.0

    cont=1

    while err > tol:
        cont+=1
        N*=2
        h/=2

        # Calculo la misma integral por el mismo método (Simpson) pero con el doble de cajas
        I2 = (f(a)+f(b))
        for k in range(1,N//2):
            I2 += 2*f(a+2*k*h) + 4*f(a+(2*k-1)*h)

        I2+=2*f(b-2*h)
        I2*=h/3
        err = 1/15*np.abs(I2-I1)

        I1=I2
    
    return (I1, err, cont)


f = lambda x: np.sin(10*np.sqrt(x))**2
I, err, cont = simpson_adapt(f, 0,1,10**(-6))

print(I, err, cont)
```

## 4.1.5 Método de Romberg

En trapecio: $I=I_1+\epsilon h_1^2 + O(h^4) = I_2 + \epsilon h_2^2$ con $h_1=2h_2$

Por tanto: $\epsilon h_2^2=\frac{I_2-I_1}{3}$ y se lo puedo meter a la otra integral

* Partamos de la regla del trapecio para una integral $I$ en un intervalo $[a,b]$ con $N_i$ sub-intervalos de anchura $h_i=(b-a)/N_i$:

$$I=I_i+\epsilon\,h_i^2+{\cal O}(h_i^4).$$


* Y del hecho de que podemos calcular su error a primer orden a partir de:

    $$\epsilon\,h_i^2=\frac{1}{3}\left(I_i-I_{i-1}\right),$$
  
  donde $I_{i-1}$ es la estimación de la integral del trapecio con $h_{i-1}=\displaystyle 2h_i$.


* Esto nos permite obtener una mejor estimación de la integral:

    $$I=I_i+\frac{1}{3}\left(I_i-I_{i-1}\right)+\epsilon_2 h_i^4+ {\cal O}(h_i^6).$$

  donde $\epsilon_2$ es ahora desconocido y por tanto, el error es ${\cal O}(h_i^4)$.


* Con el fin de simplificar la notación, llamemos a esas dos estimaciones:

    $$R_{i,1}=I_i\quad\text{y}\quad R_{i,2}=I_i+\frac{1}{3}\left(I_i-I_{i-1}\right)=R_{i,1}+\frac{1}{3}\left(R_{i,1}-R_{i-1,1}\right),$$

    que implica que 

    $$I=R_{i,2}+\epsilon_2 h_i^4+{\cal O}(h_i^6).$$


* De la misma forma, usando $h_{i-1}$, tenemos:

    $$I=R_{i-1,2}+\epsilon_2 h_{i-1}^4+{\cal O}(h_{i-1}^6)=R_{i-1,2}+16\,\epsilon_2 h_i^4+ {\cal O}(h_i^6),$$

    que nos permite obtener que:

    $$\epsilon_2 h_i^4=\frac{1}{15}\left(R_{i,2}-R_{i-1,2}\right).$$

    y que por tanto:

    $$I=R_{i,2}+\frac{1}{15}\left(R_{i,2}-R_{i-1,2}\right)+{\cal O}(h_i^6)=R_{i,3}+{\cal O}(h_i^6).$$


* Podemos continuar este proceso cancelando sistemáticamente los errores de ordenes superiores y obteniendo resultados cada vez más precisos. 


* De forma general, si $R_{i,m}$ es la estimación de la integral al orden $i$-esimo con error de orden $h^{2m}$, tenemos:
<br>

    \begin{align}
    I=&R_{i,m}+\epsilon_m h_i^{2m}+{\cal O}(h_i^{2m+2}),\\
    \\
    I=&R_{i-1,m}+\epsilon_m h_{i-1}^{2m}+{\cal O}(h_{i-1}^{2m+2})=R_{i-1,m}+4^m\epsilon_m h_i^{2m}+{\cal O}(h_i^{2m+2}),
    \end{align}

    y por tanto:

    $$\epsilon_m h_i^{2m}=\frac{1}{4^m-1}(R_{i,m}-R_{i-1,m})+{\cal O}(h_i^{2m+2}).$$


* Por lo que la siguiente iteración $R_{i,m+1}$ es:

    \begin{equation}
    R_{i,m+1}=R_{i,m}+\frac{1}{4^m-1}\left(R_{i,m}-R_{i-1,m}\right),
    \end{equation}

    que proporciona una estimación:

    $$I=R_{i,m+1}+{\cal O}(h_i^{2m+2})\,.$$

### 4.1.5.1 Algoritmo para el método de Romberg

<ol>
  <li>Calcular las dos primeras estimaciones de la integral usando la regla del trapecio: $I_1 \equiv R_{1,1}$ y $I_2\equiv R_{2,1}$.</li>
  <li> Calcular a partir de ellas $R_{2,2}$. </li>
  <li> Usamos la regla del trapecio para obtener $I_3\equiv R_{3,1}$,  y de ahí, $R_{3,2}$ y $R_{3,3}$. </li>
  <li> Para cada subsiguiente etapa calculamos $I_i\equiv R_{i,1}$, y de ahí la secuencia $R_{i,2}\cdots R_{i,i}$.</li>
  <li> Para cada estimación el error es conocido, lo que nos permite parar la secuencia cuando hayamos obtenido la precisión objetivo.    
</ol>

|     |     |     |     |     |     |     | |
| --- | --- | --- | --- | --- | --- | --- | --- |
| $I_1\equiv R_{1,1}$ | | | | | | |
| $\qquad$ | $\searrow$| | | | | | |
| $I_2\equiv R_{2,1}$ |$\to$|$R_{2,2}$  | | | | | |
| $\qquad$ | $\searrow$| | $\searrow$ | | | | |
| $I_3\equiv R_{3,1}$ |$\to$|$R_{3,2}$  |$\to$ |$R_{3,3}$ | | | |
| $\qquad$ | $\searrow$| | $\searrow$ | |$\searrow$ | | |
| $I_4\equiv R_{4,1}$ |$\to$|$R_{4,2}$  |$\to$ |$R_{4,3}$ |$\to$ | $R_{4,4}$| |
| $\qquad$ | $\searrow$| | $\searrow$ | |$\searrow$ | |$\searrow$ | 


```python
# ESTA ES LA CORRECTA

import numpy as np

def romb(f, a, b, tol):
    err = 1.0
    N=1
    h = b-a
    R11 = h/2*(f(a)+f(b))
    R1 = np.array([R11], float)
    n=1
    
    while err > tol:
        n+=1
        h/=2
        N*=2
        tmp = 0
        for k in range(1,N, 2):
            tmp+=f(a+k*h)
        R21=R11/2 + h*tmp
        R2 = np.empty(n, float)
        R2[0]=R21
        
        for i in range(1,n):
            error = (R2[i-1]-R1[i-1])/(4**i-1)
            R2[i]=R2[i-1]+error
        err = np.abs(error)
        R11=R21
        R1=R2
    return (R1[-1], err, n)
    
    while err > tol:
        n+=1
        h/=2
        N*=2
        tmp = 0
        for k in range(1,N, 2):
            tmp+=f(a+k*h)
        R21=R11/2 + h*tmp
        R2 = np.empty(n, float)
        R2[0]=R21
        
        for i in range(1,n):
            error = (R2[i-1]-R1[i-1])/(4**i-1)
            R2[i]=R2[i-1]+error
        err = np.abs(error)
        R11=R21
        R1=R2
    return (R1[-1], err, n)

# f = lambda x: np.sin(10*np.sqrt(x))**2

# I, err, cont=romb(f,0,1,10**(-6))

# print(I, err, cont)
```

valor de la integral con precisión aproximada de $\epsilon =10^{-6}$.  
   Empezar con un sólo intervalo y aumentarlos en cada iteración por un factor 2.  
   En cada iteración imprimir el número de intervalos, la estimación de la integral y su error. 


2. Modificar el programa para evaluar la integral con el método de Romberg con la misma precisión.

```python
# Método de Romberg
# Partimos del trapecio con un solo paso
import numpy as np

def romberg(f, a, b, tol):
    N=1
    h = b-a
    I1 = h/2*(f(a)+f(b))
    R1 = np.array([I1], float)
    err = 1.0
    n = 1
    
    while err > tol:
        n+=1
        h/=2
        R21 = 0
        for k in range(1,N):
            R21+=f(a+(2*k+1)*h)
        R21=I1/2+h*R21 # Hasta ahora solo he aplicado trapecio adaptado, es decir, R21
        R2 = np.empty(n,float)
        R2[0]=R21
        for i in range(1,n):
            error = (R2[i-1]-R1[i-1])/(4**i-1)
            R2[i]=R2[i-1]+error
        err = np.abs(error)
        I1 = R21
        R1=R2
        N*=2
    return (R1[-1], err, n)
    
    
f = lambda x: np.sin(10*np.sqrt(x))**2
I, err, cont = romberg(f, 0,1,10**(-6))

print(I, err, cont)
```

## 4.1.6 Métodos de integración de orden superior: fórmulas de Newton-Cotes

- La regla del trapecio se basa en aproximar una curva por una recta.


- La de Simpson por una función cuadrática.


- Por tanto, podemos aumentar la precisión creando reglas de integración con polinomios de órdenes superiores.


- De forma general, estos métodos consisten en asignar pesos a los valores de la función a integrar en un número N de puntos:

    $$I=\int\limits_{a}^bf(x)\,\text{d}x \simeq \sum_{k=1}^N{w_k\,f(x_k)},$$
<br />

   - donde $x_k =a+k\,h$ son el conjunto de puntos, equiespaciados por una distancia $h$, en los que evaluamos la función. 
   - $w_k$ su peso correspondiente.
<br />
<br />

- Para la regla del trapecio: $w_1 =w_N=1/2$ y $w_i =1$ con $i=2,\cdots,N-1$.


- Para la regla de Simpson los pesos son: $w_1 =w_N=1/3$, $w_i =2/3$ para $i=2,4\cdots,N-2$ y $w_i =4/3$ con $i=3,5\cdots,N-1$.


- Para ordenes superiores la idea es la misma:
<br />

    - Interpolar la curva en un número $N-1$ de sub-intervalos por medio de un polinomio de orden $n$. 
    - Integrar el polinomio en el conjunto de subintervalos correspondientes. 
    - Sumar todas las integrales hasta obtener la estimación completa de la integral.

<br />  

- Hasta cuarto orden la distribución de pesos es la siguiente: 

| Grado                 | Polinomio    |  Pesos                                    |
| :-------------------: | :----------: | :--------------------------------------- |
| 1 (regla del trapecio)|   Recta      | $\frac{1}{2}$,1,1,1,$\cdots$,1,1,1,$\frac{1}{2}$| 
| 2 (regla de  Simpson) |   Cuadrático | $\frac{1}{3}$,$\frac{4}{3}$,$\frac{2}{3}$,$\frac{4}{3}$,$\cdots$,$\frac{4}{3}$,$\frac{1}{3}$|
| 3                     |   Cúbico     | $\frac{3}{8}$,$\frac{9}{8}$,$\frac{9}{8}$,$\frac{3}{4}$,$\frac{9}{8}$,$\frac{9}{8}$,$\frac{3}{4}$$\cdots$,$\frac{9}{8}$,$\frac{3}{8}$|
| 4                     |   Cuártico     | $\frac{14}{45}$,$\frac{64}{45}$,$\frac{8}{15}$,$\frac{64}{45}$,$\frac{28}{45}$,$\frac{64}{45}$,$\frac{8}{15}$,$\cdots$,$\frac{64}{45}$,$\frac{14}{45}$|

- Que se conocen como fórmulas de Newton-Cotes.


- En general, si conocemos la función en un conjunto de N puntos, podemos ajustarla con un polinomio de orden N-1.


- La distribución de pesos de la fórmula de Newton-Cotes para N puntos se basa en los interpoladores de Lagrange.

### 4.1.6.1 Determinando los pesos: la función interpoladora de Lagrange

   - Tal y como hemos visto en el tema 3, dada una función $f$, que conocemos en $N$ puntos $x_i$ con $i=1,\cdots,N$, su interpolador de Lagrange:
 
 $$L(x,f)=\sum_{i=1}^N f(x_i)\,l_i(x),$$ 
 
   - Donde $l_i(x)$ forman la base de polinomios de Lagrange de grado $N-1$, definidos como:
     
     $$l_i(x)=\prod_{\begin{array}{c}j=1\\[-4pt]j\neq i\end{array}}^N{\frac{x-x_j}{x_i-x_j}}=\frac{(x-x_1)}{(x_i-x_1)}\times\cdots\times \frac{(x-x_{i-1})}{(x_i-x_{i-1})}\times \frac{(x-x_{i+1})}{(x_i-x_{i+1})}\times\cdots\times \frac{(x-x_{N})}{(x_i-x_{N})}$$

     que cumplen la propiedad:     
    
     $$l_i(x_j)=\left\{\begin{array}{lr}1&\text{if}\quad j=i,\\0&\text{if}\quad j\neq i,\end{array}\right.$$
     
   - De forma que L(x,f) es el polinomio de orden $N-1$ que interpola (coincide con) la función $f$ en los puntos $x_i$ con $i=1,\cdots,N$. 
<br />

- Por tanto, los pesos de la fórmula de Newton-Cotes de orden N vienen dados por:

\begin{align}
    \int\limits_a^bf(x)\,\text{d}x \simeq& \sum_{i=1}^N w_i\cdot f(x_i),\\
    w_i=&\int\limits_a^b l_i(x)\,\text{d}x.
\end{align}

Desarrollo de clase

$$\int_a^bf(x)dx \sim \int_a^b P_n(x)dx = \int_a^b \displaystyle\sum_{i=0}^Nl_i(x)f(x_i)dx$$

$$=\displaystyle\sum_{i=0}^N w_if(x_i), \ \ w_i = \int_a^bl_i(x)dx$$

Para hacer una integral debo calcular N integrales, pero una vez las he calculado, me vale para cualquir función $f$. Solo hay que calcularlas una vez. Por ejemplo, para el trapecio tengo 1/2 para los extremos y 1 para el resto de puntos

Gracias a la cuadratura Gaussiana puedo usar un polinomio de grado 2N-1 sumando solo N términos

### 4.1.6.2 Cálculo de los pesos de Newton-Cotes

- Lamentablemente no existe una expresión analítica general para calcular el valor de los pesos $w_i$.


- Para obtenerlos tenemos que calcular otra integral, y a priori, parece que no hemos conseguido mucho. 


- Sin embargo, lo importante es darse cuenta que para un valor de $N$ y un intervalo $[a,b]$ dado, sólo es necesario calcular los pesos una vez. 


- Es más, conocidos los pesos para un valor de $N$ en un intervalo de integración, es posible usarlos para cualquier otro intervalo. 


- De forma habitual, se suelen calcular en el intervalo $x\in [-1,1]$ de forma que calculados los pesos:

    $$w_i =\int\limits_{-1}^1{l_i(x)\,\text{d}x}$$

  el cambio de variable:
  
  $$x_k'=\frac{1}{2} \left((b-a)x_k+(b+a)\right),$$
  
  nos permite obtener los pesos en el intervalo [a,b] a partir de:
 
  $$w_k'=\frac{1}{2}(b-a)w_k.$$
 
  de forma que la integral es:
 
     $$I=\int\limits_{a}^bf(x)\,\text{d}x\simeq \sum_{k=1}^N w'_k\,f(x'_k),$$
 
    y para un número $N$ de puntos sólo hace falta calcular para cualquier integrando los pesos $w_k$.


- La fórmula de Newton-Cotes nos permite integrar una función a partir de un polinomio de grado $N-1$ evaluándola en $N$ puntos equiespaciados,  
  garantizando que la integral es exacta cuando se integra un polinomio de ese grado.

## 4.1.7 Cuadratura Gaussiana

- Hasta ahora siempre hemos evaluado la integral en un número equiespaciado de puntos.


- Pero los interpoladores de Lagrange nos permiten evaluar la función en cualquier punto, de modo que uno pueda seleccionar los más convenientes. 


- De forma general, si uno puede seleccionar los $N$ puntos en los que evaluar la función, tenemos $N$ grados de libertad adicionales.   


- Lo cual nos permite obtener una regla de integración exacta para polinomios de grado $2\,N-1$.


- Esta regla es conocida como la cuadratura gaussiana. 

### 4.1.7.1 Obteniendo los puntos en la cuadratura gaussiana: los polinomios de Legendre

- La obtención de las posiciones $x_k$ en los que evaluar la función en la cuadratura gaussiana esta basada en los polinomios de Legendre. 


- Los polinomios de Legendre $P_N(x)$ son un conjunto de polinomios ortogonales que cumplen las propiedades (entre otras):

    \begin{align}
    \int\limits_{-1}^1{x^k\,P_N(x)\,\text{d}x}=0,\quad 0\le k< N,\\
    \int\limits_{-1}^1{P_N(x)\,P_M(x)\,\text{d}x}=\frac{2}{2N+1}\delta_{NM},\\
         P_N(x_i)=0,\quad\text{con}\quad -1\le x_i \le 1\quad \text{y}\quad i=1,\cdots,N.
   \end{align}

    donde la última propiedad indica que $P_N(x)$ tiene $N$ raíces reales en el intervalo $[-1,1]$.


- Supongamos ahora que $q(x)$ es un polinomio de grado $m<N$, entonces puede escribirse como $q(x)=\sum_{k=0}^{N-1}c_k\,x^k$, y por tanto, satisface:

$$\int\limits_{-1}^1\,q(x)\,P_N(x)\,\text{d}x=\sum_{k=0}^{N-1}c_k\int\limits_{-1}^1{x^k\,P_N(x)\,\text{d}x}=0.$$

- Por tanto, dado un polinomio $f(x)$ de grado $2N-1$, éste puede ser expresado como:

    $$f(x)=q(x)\,P_N(x)+r(x),$$

    con $q(x)$ y $r(x)$ polinomios de grado $N-1$ o menor, lo que implica.

    $$\int\limits_{-1}^1\,f(x)\,\text{d}x=\int\limits_{-1}^1\,q(x)\,P_N(x)\,\text{d}x+\int\limits_{-1}^1\,r(x)\,\text{d}x = \int\limits_{-1}^1\,r(x)\,\text{d}x=\sum_{i=1}^N w_k r(x_k),$$

    donde en el último paso estamos utilizando una igualdad porque $r(x)$ es un polinomio y

    $$w_k=\int\limits_{-1}^1\,l_k(x)\,\text{d}x.$$


- La fórmula anterior nos permite integrar de forma exacta cualquier polinomio de orden $2N-1$ o menor sobre el intervalo $[-1,1]$.


- Para ello, hay que  dividirlo por el polinomio de Legendre de grado $N$ e integrar su resto usando $N$ puntos arbitrarios.


- Sin embargo, el cociente de polinomios es en general un problema complejo. 


- Alternativamente, usamos que los polinomios de Legendre tiene $N$ raíces reales $x_k$ en el intervalo $[-1,1]$, lo que permite escribir:

    $$\sum_{k=1}^N\,w_k\,f(x_k)=\sum_{k=1}^N\,w_k\, q(x_k)\,P_N(x_k)+\sum_{k=1}^N\,w_k\,r(x_k)=\sum_{k=1}^N\,w_k\,r(x_k),$$


- por lo que la integral de $f$ (un polinomio de grado 2N-1) se puede expresar como:

    $$\int\limits_{-1}^1\,f(x)\,\text{d}x=\sum_{k=1}^N\,w_k\,f(x_k),$$

  con $x_k$ las raíces de los polinomios de Legendre y $w_k$

    $$w_k=\int_{-1}^1\,l_k(x)\,\text{d}x =\left[\frac{2}{1-x^2}\left(\frac{\text{d}P_N(x)}{\text{d}x}\right)^{-2}\right]_{x=x_k}.$$

- Una vez calculados los ceros de los polinomios de Legendre y sus pesos correspondientes, sólo es necesario re escalarlos al intervalo de integración deseado:

    $$x_k'=\frac{1}{2} \left((b-a)x_k+(b+a)\right),$$

    y 

    $$w_k'=\frac{1}{2}(b-a)w_k.$$

   para obtener:

     $$I=\int\limits_{a}^bf(x)\,\text{d}x\simeq \sum_{k=1}^N w'_k\,f(x'_k),$$
     
**<p style="color:red;">NOTA IMPORTANTE:</p>** Fíjate que este cambio de variable es para cuando $x$ y $\omega$ se obtienen en el intervalo canónico $(-1,1)$, es decir, con la función gaussx. Si tienes que usarla más de una vez, llama sollo una vez a gaussx y luego transforma. Esto se debe a que lo costoso del método de Gauss es calcular las raíces del polinomio de Legendre de grado 2N-1

Propiedades de los polinomios de Legendre a partir de los cuales construiremos la cuadratura Gaussiana:

1. $\int_{-1}^1 x^kP_N(x)dx = 0, \ k < N$
2. $\int_{-1}^1 P_NP_Mdx=\frac{2}{N+1}\delta_{NM}$
3. $P_N(x_i)=0, \ \ -1\leq x_i \leq 1$ (tiene 0's entre -1 y 1)

Puedo construir un polinomio de orden 2n-1 con un polinomio de Legendre

$$\int_{-1}^1 f(x)dx = \int_{-1}^1 P_{2N-1}(x)dx = \int_{-1}^1q_{N-1}(x)^N(x)dx + \int_{-1}^1 r_{N-1}(x)dx$$

La idea de la cuadratura Gaussiana es sacar el polinomio de grado 2N-1 que da la integral cuando evaluaos en los 0s del polinomio de Legendre correspondiente

El cambio de variable que hay que hacer para pasar de límites de integración -1 y 1 a "a", "b" hay que hacer el cambio de variable:

$$x'=\frac{1}{2}\left((b-a)x+(b+a)\right)$$

### Ejercicio 4.4: capacidad calorífica de un sólido 

- Según la teoría de Debye la capacidad calorífica de un sólido a una temperatura $T$ viene dada por:
<br/>

    $$C_V=9\,V\rho\,k_B\left(\frac{T}{\theta_D}\right)^3\int\limits_0^{\theta_D\,/\,T}{\frac{x^4\,\text{e}^x}{\left(\text{e}^x-1\right)^2}\,\text{d}x},$$

     - con $V$ el volumen del sólido, 
     
     - $\rho$ la densidad numérica del material, 
     
     - $k_B=1.38065\cdot 10^{-23}$ J/K la constante de Boltzmann,
     
     - $\theta_D$ la temperatura de Debye, que depende de la temperatura y la velocidad del sonido en el material. 
     
     
1. Escribir un programa que calcule $C_V$ en función de $T$, para una muestra de 1000 $\text{cm}^3$ de aluminio, cuya densidad numérica es $\rho=6.022\times 10^{28}$ $\text{m}^{-3}$ y $\theta_D=428$ K. Usar la cuadratura gaussiana con N=50. 


2. Usar la función para representar la capacidad calorífica del aluminio para temperaturas entre 5 y 500 K.

```python
from gaussxw import gaussxwab
import numpy as np
import matplotlib.pyplot as plt

# Cv = f(T)*I(T)

T = np.linspace(5,700,1000)
V=1/1000 # m^3
p = 6.022*10**28
kb = 1.38065*10**(-23)
theta_D = 480

def I_(T):
    N=50
    x, w = gaussxwab(N,0,theta_D/T)
    res = 0
    f = lambda x: x**4*np.exp(x)/(np.exp(x)-1)**2
    
    for k in range(N):
        res += w[k]*f(x[k])
    return res

I = np.vectorize(I_)

Cv = 9*V*p*kb*(T/theta_D)**3*I(T)

plt.grid()
plt.xlabel("$T$")
plt.ylabel("$C_V$")
plt.plot(T, Cv)
plt.show() # Observamos que se cumple la ley de Dulong-Petit como debería
```

### 4.1.7.2 Estimación de errores en la cuadratura Gaussiana

- La estimación del error de la cuadratura Gaussiana en función de la fórmula de Euler-Maclaurin es posible pero muy complicada. 


- Para hacernos una idea de cómo escala su error veamos un ejemplo.

#### Ejemplo 4.8: la función Beta

La función Beta se define como la integral

$$B(x,y)=\int\limits_0^1{t^{x-1}\,(1-t)^{y-1}\,\text{d}t}=\frac{\Gamma(x)\,\Gamma(y)}{\Gamma(x+y)},$$

donde para $n\in \mathbb N$, $\Gamma(n)=(n-1)!$.

Calcular la función Beta para $x=3$ e $y=5$ usando la cuadratura de Gauss yendo de N=1 a N=5.
Comparar con el resultado analítico.  
¿Cómo escala el error de la cuadratura de Gauss?


```python
from gaussxw import gaussxwab
from numpy import abs

def f(t):
    x = 3
    y = 5
    return t**(x-1)*(1-t)**(y-1)

analitico = 1/(7*5*3)
print(f"Valor exacto: {analitico}")

for N in range(1,6):
    x, w = gaussxwab(N,0,1)
    I=0
    for k in range(N):
        I+=w[k]*f(x[k])
    desv = abs(analitico-I)
    print(f"N={N}; I={I}; desv={analitico-I}") # Por qué me da valor negativo de desv en primera it.?
```

#### El error de la cuadratura guassiana

* Al pasar de N a N+1 el error de la cuadratura gaussiana mejora por un factor $c/N^2$.


* Al pasar de N=10 a N=11, la precisión mejora por un factor 100. 


* Por consiguiente, para valores de $N$ suficientemente grandes, la reducción total del error es del orden $N^{-2N}$.


* Si $I_N$ es la estimación gaussiana con N puntos al doblar el número de puntos se obtiene que:

    $$I=I_N+\epsilon_N=I_{2N}+\epsilon_{2N}\Rightarrow I_{2N}-I_N=\epsilon_N-\epsilon_{2N}\sim \epsilon_N,$$

    pues la mejora en cada paso es tan grande que uno puede asumir $\epsilon_{2N}\ll \epsilon_{N}$.


* Para valores de $N\sim 100$, la precisión alcanzada es generalmente mayor que la precisión de la máquina.

**¿Cual es son los pros y contras de definir una cuadratura gaussiana de paso adaptado?**

- **Contras**
<br/>

    - En cada nuevo paso tenemos que volver a calcular los puntos y pesos de la cuadratura.   
      El algoritmo adaptado será más lento que haber considerado desde el principio un número grande de puntos. 
    - El error que obtendremos siempre corresponderá al paso anterior. 
<br/>


- **Pros**
<br/>    

    - El número total de elementos que vamos a sumar en todos los pasos nunca será mayor que dos veces el número de puntos del último paso.
    - Casi nunca es necesario ir más allá de $N=100$ para obtener un error de truncación menor que el de redondeo.
    - Si empezamos con $N=10$, es razonable suponer que no tendremos que doblar el número de puntos más de tres o cuatro veces.
    
  
 - **Conclusiones**
<br/>      

    - A pesar del esfuerzo computacional extra, la cuadratura gaussiana es normalmente más eficiente que la regla del trapecio o de Simpson.    

## 4.1.8 Eligiendo un método de integración

**Regla general:** 

1. Si la función es suficientemente suave los métodos de orden superior Romberg, Newton-Cotes o la cuadratura Gaussiana permiten obtener gran precisión con pocos puntos. 


2. Si la función no es suave, los métodos más sencillos, como el trapezoidal, suelen ser el camino adecuado.
 

 - Método Trapezoidal:
 
    - sencillo de programar (respuesta rápida aunque menos precisa).
    - usa puntos igualmente espaciados (para utilizar en datos de laboratorio).
    - útil para funciones con problemas (singularidades, ruido,...).
    - en su forma de paso adaptado podemos garantizar una precisión a costa de más tiempo de computo.


 - Método de Simpson:
 
   - sencillo de programar (respuesta rápida aunque menos precisa).
   - usa puntos equiespaciados (para utilizar en datos de laboratorio).
   - mayor precisión que el trapezoidal con los mismos puntos.
   - mayor orden de la aproximación del integrando (problemas si la función tiene ruido,...).
   - en su forma de paso adaptado podemos garantizar una precisión, con posibles problemas si la función no es lo suficientemente suave.


 - Método de Romberg:
 
    - con puntos igualmente espaciados es el mejor con órdenes altos de aproximación.
    - aproximación excepcional con pocos puntos.
    - garantiza una precisión con estimación del error.
    - problemas para funciones que se comporten mal, con ruido, singularidades,...
    - muy útil para funciones suaves cuya forma puede ser determinada por pocos puntos.


 - Cuadratura Gaussiana:
 
    - puntos no equiespaciados.
    - similares ventajas e inconvenientes que el método de Romberg.
    - complejidad: cálculo de los puntos y los pesos.
    - mayor orden de aproximación.




# 4.2 Integrales sobre rangos infinitos

- Hasta ahora nos hemos limitado a estudiar técnicas para analizar integrales definidas en un intervalo $[a,b]$.


- Sin embargo, estas técnicas pueden emplearse de forma natural para estudiar integrales sobre rangos infinitos haciendo cambios de variable. 

## 4.2.1 Integrales en el rango $[0,\infty]$ y $[-\infty,0]$

- Una integral sobre el rango $[0,\infty]$ puede expresarse como otra en el rango $[0,1]$ a través de:

    $$z=\frac{x}{x+1}\quad\Rightarrow\quad x=\frac{z}{1-z},\quad dx=\frac{\text{d}z}{(1-z)^2}$$

   que nos permite hacer:

    $$\int\limits_0^\infty\,f(x)\,\text{d}x = \int\limits_{0}^1\frac{1}{(1-z)^2}\,f\left(\frac{z}{1-z}\right)\,\text{d}z,$$

    y por tanto, estudiarla por medio de las técnicas que hemos visto hasta ahora. 


- Por supuesto, $z=x/(x+1)$ no es el único cambio de variable posible. Por ejemplo, el cambio de variable

    $$z=\frac{x^\gamma}{x^\gamma+c}\quad \forall\,c,\gamma >0$$

    también puede ser usado.


- En general, dependiendo de la función un cambio de variable puede funcionar mejor que otro, y uno tiene que analizar cual es el óptimo en cada caso. 


- De la misma forma, una integral sobre el rango $[-\infty,0]$ puede expresarse como otra en el rango $[0,1]$ teniendo en cuenta que: 


$$\int\limits_{-\infty}^0\,f(x)\,\text{d}x = \int\limits_0^{\infty}\,f(-x)\,\text{d}x =  \int\limits_{0}^1\frac{1}{(1-z)^2}\,f\left(\frac{z}{z-1}\right)\,\text{d}z.$$

## 4.2.2 Integrales en los rango $[a,\infty]$ y $[-\infty,b]$

- Equivalentemente, una integral en el rango $[a,\infty]$ también puede escribir como otra en $[0,1]$ por medio del cambio de variable:

    $$z=\frac{x-a}{x-a+1}\quad\Rightarrow\quad x=\frac{z}{1-z}+a,\quad dx=\frac{\text{d}z}{(1-z)^2}$$

   de forma que

    $$\int\limits_a^\infty\,f(x)\,\text{d}x = \int\limits_{0}^1\frac{1}{(1-z)^2}\,f\left(\frac{z}{1-z}+a\right)\,\text{d}z,$$

-  De forma similar, una integral en $[-\infty,b]$ vendrá dada por:

$$\int\limits_{-\infty}^b\,f(x)\,\text{d}x = \int\limits_{-b}^{\infty}\,f(-x)\,\text{d}x =  \int\limits_{0}^1\frac{1}{(1-z)^2}\,f\left(\frac{z}{z-1}+b\right)\,\text{d}z.$$



## 4.2.3 Integrales en los rango $[-\infty,\infty]$


- Para resolver integrales en el rango $[-\infty,\infty]$ podemos dividir el intervalo en dos, de $[-\infty,a]$ y de $[a,\infty]$,  
  aplicando en cada uno de ellos los cambios de variable anteriores. 


- Otra opción es efectuar el cambio de variable   

    $$x=\frac{z}{1-z^2},\quad dx=\frac{1+z^2}{(1-z^2)^2}\text{d}z,$$

  que da lugar a la integral
  
     $$\int\limits_{-\infty}^\infty\,f(x)\,\text{d}x = \int\limits_{-1}^1\frac{1+z^2}{(1-z^2)^2}\,f\left(\frac{z}{1-z^2}\right)\,\text{d}z.$$
 
* Otra opción es:
 
  $$x=\tan\,z,\quad dx=\frac{\text{d}z}{\cos^2z},$$
 
  que permite expresar:

  $$\int\limits_{-\infty}^\infty\,f(x)\,\text{d}x = \int\limits_{-\pi/2}^{\pi/2}\frac{f\left(\tan z\right)}{ \cos^2z}\,\text{d}z.$$


### Ejemplo 4.9: integrando la  gaussiana 

Resolver la integral gaussiana, definida por:

 $$\int\limits_{-\infty}^\infty\,\text{e}^{-x^2}\,\text{d}x =\sqrt \pi$$

usando la cuadratura gaussiana con 50 puntos.

```python
# MEJOR: Método 1: separo en dos integrales de (-inf,0] y de [0,inf) y aprovecho paridad

from gaussxw import gaussxwab
import numpy as np

N=15
x, w = gaussxwab(N,0,1)

f = lambda x: np.exp(-x**2)

def g(z):
    J = 1/(1-z)**2
    x = z/(1-z)
    return J*f(x)

I=0
for k in range(N):
    I+=w[k]*g(x[k])

I*=2
print(I)
print(np.sqrt(np.pi))

# Es el mejor método de los dos, como se puede comprobar en que se acerca más al resultado exacto que el cambio de variable directo.
# Esto se debe a que estamos usando el mismo número de puntos para la mitad de la gráfica y luego multiplicando por 2 aprovechando
# la paridad de la gaussiana ==> lograremos más precisión
```

```python
# PEOR: Método 2: cambio de variable de (-inf,inf) a [-1,1]

from gaussxw import gaussxw
import numpy as np

N=15
x, w = gaussxw(N)

f = lambda x: np.exp(-x**2)

def g(z):
    J = (1+z**2)/(1-z**2)**2
    x = z/(1-z**2)
    return J*f(x)

I=0
for k in range(N):
    I+=w[k]*g(x[k])

print(I)
print(np.sqrt(np.pi))
```

# 4.3 Integrales múltiples

## 4.3.1 Integrales bidimensionales rectangulares

- Integrales en más de una variable son muy comunes en física. 


- Consideremos como punto de partida una integral doble sobre un rectángulo.

    $$I=\int\limits_{a_1}^{b_1}\int\limits_{a_2}^{b_2}\,f(x,y)\;\text{d}x\,\text{d}y$$


- Se puede resolver con la técnicas que hemos vista ahora teniendo en cuenta que:

   \begin{align}
       F(y)=&\int\limits_{a_2}^{b_2}\,f(x,y)\;\text{d}x,\\
       I=&\int\limits_{a_1}^{b_1}\,F(y)\;\text{d}y,\\
   \end{align}

    es decir, una forma de resolver integrales múltiples es es evaluar primero $F(y)$, y luego usar esos valores para integrar la función total. 


- Usando, por ejemplo, la cuadratura gaussiana, se tiene que:

    $$F(y)\simeq \sum_{i=1}^N\,w_i\,f(x_i,y),\quad I\simeq \sum_{j=1}^N\,w_j F(y_j)\quad\Rightarrow\quad I\simeq \sum_{i=1}^N\sum_{j=1}^N\,w_i\,w_j\,f(x_i,y_j),$$

    que es la llamada ecuación de Gauss-Legendre para el producto de integrales, la cuadratura gaussiana en dos dimensiones.

**Nota:** una forma optimizada de calcular una integral doble rectangular es con un meshgrid. Pasas de tener dos bucles for a tener una sola cuenta de matrices, y recuerda que Python opera con matrices casi a la misma velocidad que con números. Veamos un ejemplo de cómo podemos hacer esto.

Se desea calcular la integral doble:

$$I=\int_0^1\int_0^1 x^2ydxdy=\frac{1}{6}$$

```python
from gaussxw import gaussxwab
import numpy as np

N=50

x, w = gaussxwab(N, 0, 1)

I=0

def f(x, y):
    return x**2*y

X, Y = np.meshgrid(x, x)
WX, WY = np.meshgrid(w, w)

tmp = WX*WY*f(X, Y)

I = np.sum(tmp)
print(I)

# for i in range(N):
#     for j in range(N):
#         I+=w[i]*w[j]*f(x[i], x[j])
```

## 4.3.2 Integrales bidimensionales no rectangulares

* Otro tipo de integrales que aparecen de forma habitual son aquellas en los que el límite de integración de una variable depende de otra.

$$I=\int\limits_{a_1}^{b_1}\text{d} y\int\limits_{a_2}^y\,\text{d}x\,f(x,y),$$

- Podemos usar el mismo método que antes, es decir, definir:

    $$F(y)=\int\limits_{a_2}^y\,\text{d}x\,f(x,y),$$

    de forma que

    $$I=\int\limits_{a_1}^{b_1}\text{d} y\,F(y),$$

    que nos permite llegar a la misma cuadratura gaussiana, pero ahora con los pesos localizados en un triangulo:

```python
from numpy import empty
from matplotlib.pyplot import scatter,show,xlabel,ylabel
from gaussxw import gaussxw,gaussxwab

N=10
x,w=gaussxw(N)
z=[]
w=[]
for i in range(N):
    xx,ww=gaussxwab(N,-1,x[i])
    for j in range(N):        
        z.append(x[i])
        w.append(xx[j])

xlabel("x i")
ylabel("y j")
show(scatter(w,z))
```

* Aunque el método funcionará y proporcionará un resultado correcto, no es ideal.



* En el límite inferior izquierdo los puntos están completamente acumulados,

# 4.4 Derivadas

## 4.4.1 Derivadas de primer orden

### 4.4.1.1 Derivadas con paso hacía adelante y hacía atrás

- La derivada de una función $f$ con respecto a la variable $x$ viene dada por 

    $$\frac{\text{d}f(x)}{\text{d}x}=\lim_{h\to0}\frac{f(x+h)-f(x)}{h},$$


- La implementación de una derivada en un ordenador consiste en tomar un valor pequeño pero no nulo de $h$

    $$\frac{\text{d}f(x)}{\text{d}x}\simeq\frac{f(x+h)-f(x)}{h},$$

    que es conocida como la diferencia hacía adelante pues la derivada se evalúa desplazando hacía delante el argumento.


- Sin embargo, es igual de correcto evaluar la derivada hacía atrás,

    $$\frac{\text{d}f(x)}{\text{d}x}\simeq\frac{f(x)-f(x-h)}{h},$$

    que en la mayoría de los casos da aproximadamente el mismo resultado cuando $h$ es pequeño. 


- Excepciones son aquellas funciones con una discontinuidad en $x$ o cuando $x$ corresponde a un punto de frontera. 

```python
from IPython.display import Image,display
display(Image(filename="derivada_back_forth.png", width=750, height=750))
```

**Errores de las derivadas con paso hacía adelante/atrás**

- Como con cualquier cálculo computacional, las derivadas sufren tanto errores de truncación como errores de redondeo.


- Mientras que para las integrales el error de redondeo es típicamente pequeño, en las derivadas es igual de importante que el de truncación.


- Para entenderlo, expandamos en serie de Taylor la función adelantada:

    $$f(x+h)=f(x)+h\,f'(x)+\frac{h^2}{2}f''(x)+\cdots,$$

     por lo que la primera derivada se puede expresar como

    $$f'(x)=\frac{f(x+h)-f(x)}{h}+\frac{h}{2}f''(x)+\cdots,$$


- De la misma forma para la diferencia atrasada:

    $$f(x-h)=f(x)-h\,f'(x)+\frac{h^2}{2}f''(x)+\cdots,$$

    y por tanto:

    $$f'(x)=\frac{f(x)-f(x-h)}{h}+\frac{h}{2}f''(x)+\cdots,$$


- Las formulas anteriores nos permite analizar los dos tipos de errores con los que tenemos que lidiar. 
<br>

    1. El error de truncación es: $\frac{1}{2}h\vert f''(x)\vert$ y por tanto lineal en h. 

    2. El error de redondeo involucra restar dos números muy próximos y dividirlos por un valor muy pequeño

    $$\sigma(f'(x))\simeq\frac{\epsilon_M}{h}\sqrt{f(x+h)^2+f(x)^2+\frac{h^4}{4}f''(x)^2}\simeq \frac{\epsilon_M}{h}\sqrt{2}\vert f(x)\vert,$$


- Por tanto, el error total (la suma del de truncación y redondeo será) como función de h será:

    $$\epsilon(h)\simeq\frac{\epsilon_M}{h}\sqrt{2}\,\vert f(x)\vert+\frac{1}{2}h\,\vert f''(x)\vert,$$

    que derivando con respecto a $h$ nos permite obtener el valor que minimiza el error total, $h_\text{opt}$:

    $$h_\text{opt}=\left( 2\sqrt 2\,\epsilon_M\left\vert\frac{f(x)}{f''(x)}\right\vert\,\right)^{1/2}\sim 10^{-8},$$

    para el cual el error es:

    $$\epsilon_\text{min}=\epsilon\,(h_\text{opt})=h_\text{opt}\vert f''(x)\vert=\left(2\sqrt 2\,\epsilon_M\left\vert f(x)\,f''(x)\right\vert\,\right)^{1/2}\sim10^{-8},$$


- Por consiguiente, para las derivadas numéricas obtenemos típicamente la mitad de la precisión usual. 

### 4.4.1.2 Diferencias centrales/centradas

- Las derivadas adelantadas y atrasadas tienen el mismo error de truncación.


- Por consiguiente, expandiendo la serie de Taylor h/2 hacía adelante y h/2 hacía atrás hasta un orden superior y restándolas, se obtiene:

    $$f'(x)=\frac{f(x+h/2)-f(x-h/2)}{h}-\frac{1}{24}h^2\,f'''(x)+....,$$

    que permite reducir el error de truncación a $\frac{1}{24}h^2\,f'''(x)$, que corresponde a un $h_\text{opt}$
    
    $$h_\text{opt}=\left( 12\sqrt 2\,\epsilon_M\left\vert\frac{f(x)}{f''(x)}\right\vert\,\right)^{1/3}\sim 10^{-5},$$

    y un error mínimo

    $$\epsilon_\text{min}=\frac{1}{8}h_\text{opt}^2\vert\,f'''(x)\vert=\left(\frac{9}{16}\,\epsilon_M^2\left\vert f(x)^2\,f''(x)\right\vert\,\right)^{1/3}\sim10^{-10},$$


- Para el valor de $h$ óptimo, las diferencia centrada es un factor 100 más precisa que la adelantada o atrasada.


- Hay ocasiones, sin embargo, en los que sólo conocemos la función en un número dado de puntos equiespaciados.


- En ese caso, sólo podemos calcular la derivada tomando la diferencia entre dos puntos separados una distancia $2h$

    $$f'(x)=\frac{f(x+h)-f(x-h)}{2h}-\frac{1}{6}h^2\,f'''(x)+....,$$


- La pregunta es si en este caso, las diferencias centradas proporcionan una estimación más o menos precisa que las adelantadas o atrasada. 


- Comprobando los errores mínimos en ambos casos vemos que la diferencia centrada dará una estimación más precisa sólo si:

$$h<8\left\vert\frac{f''(x)}{f'''(x)}\right\vert.$$


- En general, para funciones suaves las diferencias centradas van a ser más precisas, pero no esta garantizado que vaya a ser el caso para funciones con mucho ruido. 

### Ejemplo 4.11: calculando derivadas

Dada la función $f(x)=1+\frac{1}{4}\,\tanh(4x)$:

1. Calcular la derivada de la función en el intervalo $-1\le x\le1$ usando diferencias hacía adelante, atrás y centradas. 
2. Calcular la derivada analítica y representar las 4 curvas en la misma figura. 

```python
import numpy as np
import matplotlib.pyplot as plt

f = lambda x: 1+1/4*np.tanh(4*x) # Es función suave, las centradas darán mejor precisión que las adelantadas / atrasadas

# Adelantadas: f'(x)=(f(x+h)-f(x))/h con h pequeño (h~10⁸)
h = 10**(-8)
x = np.linspace(-1,1,1000)

der_adelant = (f(x+h)-f(x))/h

# Atrasadas: f'(x)=(f(x)-f(x-h))/h

der_atras = (f(x)-f(x-h))/h

# Centradas: f'(x)=(f(x+h/2)-f(x-h/2))/h
h = 10**(-5)
der_cent = (f(x+h/2)-f(x-h/2))/h

plt.grid()
plt.plot(x, der_adelant, "r--")
plt.plot(x, der_atras, "b--")
plt.plot(x, der_cent, "g--")

sol = lambda x: 1/np.cosh(4*x)**2
plt.plot(x, sol(x), "k")
plt.show()
```

### 4.4.1.3 Aproximaciones de orden superior

- Al igual que con la integrales, las derivadas adelantadas y atrasadas consisten en estimar la curva por una recta y obtener su pendiente.


- La derivada centrada no es más que la pendiente del ajuste a una función cuadrática. 


- En general, dado un polinomio

    $$p(x,N)=\sum_{k=0}^N a_k\,x^k$$ 

     de orden $N$, éste describirá la función $f(x)$ en $N+1$ puntos en el intervalo $[x_0-\frac{N}{2}h,x_0+\frac{N}{2}h]$, si satisface el sistema de ecuaciones:

    $$f(x_j)-p\left(x_j,N\right)=0,\quad\text{con}\quad x_j=x_0+\left(j-\frac{N}{2}\right)\,h,\quad j=0,\cdots,N,$$


* Dado ese polinomio, $p(x,N)$ su derivada centrada en $x_0$ será:

    $$p'(x_0,N)=\sum_{k=1}^N k\,a_k\,x_0^{k-1},$$

    que coincidirá con la de $f(x)$ hasta ${\cal O}(h^{N})$.


* Sin embargo, siempre podemos hacer un cambio de variable $x\to \left(x-x_0\right)$ de forma que, en la nueva variable,  
  sólo tengamos que calcular la derivada de la función en 0, y por tanto

    $$f'(0)=a_1,$$

   por lo que sólo es necesario calcular el primer coeficiente. 


- Para N=2,

    \begin{align}
    f\left(-h\right)&=\,a_0-a_1 h+a_2\,h^2,\\
    &\\
    f\left(0\right)&=\,a_0,\\
    &\\
    f\left(h\right)&=\,a_0+a_1 h+a_2 h^2,
    \end{align}

     que implica que 

    $$a_1=\frac{f\left(h\right)-f\left(-h\right)}{2h},$$

     y volviendo a la variable original

    $$f'(x_0)=\frac{f\left(x_0+h\right)-f\left(x_0-h\right)}{2h}+{\cal O}(h^2)$$

     que nos permite recuperar la formula para las diferencias centradas para dos puntos separados una distancia $2h$.


* De la misma forma, por ejemplo para $N=4$,

    \begin{align}
    f\left(-2h\right)&=\,a_0-2\,a_1\,h+4\,a_2\, h^2- 8\,a_3\,h^3+ 16\,a_4\,h^4,\\
    &\\
    f\left(-h\right)&=\,a_0-a_1\,h+a_2\,h^2- a_3\,h^3+ a_4\,h^4,\\
    &\\
    f\left(0\right)&=\,a_0,\\
    &\\
    f\left(h\right)&=\,a_0+a_1\,h+a_2\,h^2+ a_3\,h^3+ a_4\,h^4,\\
    &\\
    f\left(2h\right)&=\,a_0+2\,a_1\,h+4\,a_2\, h^2+ 8\,a_3\,h^3+ 16\,a_4\,h^4,\\
    \end{align}

     que implica

    $$a_1=\frac{\frac{1}{12}f\left(-2h\right)-\frac{2}{3}f\left(-h\right)+\frac{2}{3}f\left(h\right)-\frac{1}{12}f\left(2h\right)}{h},$$

     y por tanto:

    $$f'(x_0)=\frac{\frac{1}{12}\,f\left(x_0-2h\right)-\frac{2}{3}f\left(x_0-h\right)+\frac{2}{3}f\left(x_0+h\right)-\frac{1}{12} f\left(x_0+2h\right)}{h}+{\cal O}(h^4),$$

    que nos permite reducir el error de truncación de manera efectiva.


* En general si queremos calcular la primera derivada con una precisión ${\cal O}(h^N)$ necesitaremos conocer la función en $N+1$ puntos. 


* Definiendo estos puntos como $x_i=x_0+(i-\frac{N}{2})h$, con $i=0,\cdots N$, y por tanto en el intervalo $[x_0-\frac{N}{2}h,x_0+\frac{N}{2}h]$,  
  estamos buscando expresar la derivada como:

    $$f'(x_0)=\displaystyle\frac{a_{-N/2}\,f\left(x_0-\frac{N}{2}h\right)+a_{-N/2+1}\,f\left(x_0-\left(\frac{N}{2}-1\right)h\right)+\cdots+a_{N/2}\,f\left(x_0+\frac{N}{2}h\right)}{h}+{\cal O}(h^N)$$

     y el problema consiste en extraer los coeficientes $a_i$ correspondientes. 


* Para ello tenemos en cuenta que la expansión en serie de Taylor hasta orden $N$ de cada uno de los términos es:

    \begin{align}
        a_{-N/2}\,f\left(x_0-\frac{N}{2}h\right)=&a_{-N/2}\,\left[ f(x_0)-\frac{N}{2}hf'(x_0)+\cdots+\frac{1}{N!}\left(-\frac{N}{2}h\right)^N\,f^{(N)}(x_0)\right],\\
    a_{-N/2+1}\,f\left(x_0-\left(\frac{N}{2}-1\right)h\right)=&a_{-N/2+1}\,\left[ f(x_0)-\left(\frac{N}{2}-1\right)hf'(x_0)+\cdots+\frac{1}{N!}\left(-\left(\frac{N}{2}-1\right)h\right)^N\,f^{(N)}(x_0)\right],\\
    \vdots\qquad\qquad&\qquad\qquad\qquad\qquad\vdots\qquad\qquad\qquad\vdots\\
    \vdots\qquad\qquad&\qquad\qquad\qquad\qquad\vdots\qquad\qquad\qquad\vdots\\
    \vdots\qquad\qquad&\qquad\qquad\qquad\qquad\vdots\qquad\qquad\qquad\vdots\\
    a_{N/2}\,f\left(x_0+\frac{N}{2}h\right)=&a_{N/2}\,\left[ f(x_0)+\frac{N}{2}hf'(x_0)+\cdots+\frac{1}{N!}\left(\frac{N}{2}h\right)^N\,f^{(N)}(x_0)\right],
    \end{align}


* Y de forma general:

    $$a_{\left(i-\frac{N}{2}\right)}\,f\left(x_0+\left(i-\frac{N}{2}\right)h\right)=a_{\left(i-\frac{N}{2}\right)}\left[f(x_0)+\,\left(i-\frac{N}{2}\right)h\,f'(x_0)+\cdots+\frac{1}{N!}\,\left(\left(i-\frac{N}{2}\right)h\right)^N\,f^{(N)}(x_0)\right].$$


* Ahora bien, queremos determinar los coeficientes $a_i$ que nos proporcionan la primera derivada,  
  pero que cancelan la contribución de la función en $x_0$ y de sus derivada de orden  $1< n\le  N$. 
  
 
* Por tanto, sumando columna a columna, los coeficientes deben cumplir: 

  \begin{align}        \sum_{i=0}^{N} a_{i-N/2} =&0,\\
       \sum_{i=0}^{N} a_{i-N/2} \left(i-N/2\right) =&1,\\
       \vdots\qquad\qquad&\vdots\\
       \sum_{i=0}^{N} a_{i-N/2} \left(i-N/2\right)^i =&0,
   \end{align}

   que en forma matricial se puede escribir como:
    <br>

  \begin{equation}
    \left(\begin{array}{ccccc}
    1&1&\cdots&1&1\\
    -\frac{N}{2}&-\frac{N}{2}+1&\cdots&\frac{N}{2}-1&\frac{N}{2}\\        \left(-\frac{N}{2}\right)^2&\left(-\frac{N}{2}+1\right)^2&\cdots&\left(\frac{N}{2}-1\right)^2&\left(\frac{N}{2}\right)^2\\
    \cdots& \cdots&\cdots&\cdots&\cdots\\
        \left(-\frac{N}{2}\right)^N&\left(-\frac{N}{2}+1\right)^N&\cdots&\left(\frac{N}{2}-1\right)^N&\left(\frac{N}{2}\right)^N
    \end{array}\right)\left(\begin{array}{c}
    a_{-N/2}\\[1.9pt]
    a_{-N/2+1}\\[1.9pt]
    a_{-N/2+2}\\[1.9pt]
    \cdots\\[1.9pt]
    a_{N/2}
    \end{array}\right)=
    \left(\begin{array}{c}
    0\\[1.8pt]
    1\\[1.8pt]
    0\\[1.8pt]
    \cdots\\[1.8pt]
    0
    \end{array}\right)
    \end{equation}
    <br>

* Que representa un sistema de ecuaciones lineales que aprenderemos a resolver en el Tema 5.

## 4.4.2 Derivadas de orden superior

El error total de la derivada de segundo orden será:

$$\epsilon_T = \frac{h^2}{12}|f^{(4)}(x)| + \frac{\sqrt{6}}{h^2}\epsilon_M|\bar{f(x)}|$$

### 4.4.2.1  Derivadas de segundo orden

- La derivada segunda es la derivada de la primera derivada. Tomando por ejemplo la diferencia centrada:

    \begin{align}
    f''(x)\simeq& \frac{f'\left(x+\frac{h}{2}\right)-f'\left(x-\frac{h}{2}\right)}{h}\\
    =&\frac{\left(f(x+h)-f(x)\right)/h-\left(f(x)-f(x-h)\right)/h}{h}\\
    =&\frac{f(x+h)-2f(x)+f(x-h)}{h^2},
    \end{align}

  que es la aproximación más sencilla para la segunda derivada.



* Para determinar el error de truncación, expandamos una vez más la función en cada intervalo en serie de Taylor

   \begin{align}
    f(x+h)=&f(x)+hf'(x)+\frac{1}{2}h^2f''(x)+\frac{1}{6}h^3f'''(x)+\frac{1}{24}h^4f^{(iv)}(x)+\cdots,\\[4pt]
    f(x-h)=&f(x)-hf'(x)+\frac{1}{2}h^2f''(x)-\frac{1}{6}h^3f'''(x)+\frac{1}{24}h^4f^{(iv)}(x)+\cdots,
   \end{align}

   y por tanto

   $$f''(x)=\frac{f(x+h)-2f(x)+f(x-h)}{h^2}-\frac{1}{12}h^2f^{(iv)}(x)+\cdots$$



* Una vez más, podemos calcular el error total y el valor de $h$ que minimiza el error. 

    $$\epsilon(h)\simeq\frac{\sqrt 6\,\epsilon_M}{h^2}\,\vert f(x)\vert+\frac{1}{12}h^2\,\vert f^{(iv)}(x)\vert,$$

     por lo que el valor de $h$ que minimiza el error es

    $$h_\text{opt}=\left(12\sqrt 6\,\epsilon_M\left\vert\frac{f(x)}{f^{(iv)}}\right\vert\,\right)^{1/4}\sim10^{-4},$$

    que da un error 

    $$\epsilon_\text{min}=\frac{1}{6}h_\text{opt}^2\left\vert\,f^{(iv)}\,\right\vert=\left(\frac{1}{2\sqrt 6}\,\epsilon_M\left\vert\, f(x)f^{(iv)}\,\right\vert\right)^{1/2}\sim 10^{-9}.$$


* Tomando diferencias centradas estamos obteniendo la misma precisión que con diferencias adelantadas/atrasadas para la primera derivada.



* La precisión de una derivada decrece a medida que aumenta su orden. 

### 4.4.2.2  Derivadas de orden superior

- Repitiendo el mismo procedimiento podemos calcular derivadas de orden superior. 


- De forma general, sí queremos calcular una derivada de orden $m$ de una función $f$ con una precisión ${\cal O}(h^{n})$ necesitaremos evaluarla en $N$ puntos con:

    $$N=2\left\lfloor\frac{m+1}{2}\right\rfloor -1+n$$

    donde $\left\lfloor x\right\rfloor$ representa el mayor entero $\le x$. 


- Repitiendo el procedimiento empleado para la derivada de primer orden, el problema consisten en encontrar los coeficientes de la expansión:

    $$f^{(m)}(x_0)=\displaystyle\frac{a_{-N/2}\,f\left(x_0-\frac{N}{2}h\right)+a_{-N/2+1}\,f\left(x_0-\left(\frac{N}{2}-1\right)h\right)+\cdots+a_{N/2}\,f\left(x_0+\frac{N}{2}h\right)}{h^m}+{\cal O}(h^n),$$


- Expandiendo cada uno de ellos en serie de Taylor e imponiendo que sólo den una contribución no nula para la derivada de orden $m$  
  se obtiene que su solución viene dada por el sistema de ecuaciones:
  <br>
  
     \begin{equation}
    \left(\begin{array}{ccccc}
    1&1&\cdots&1&1\\
    -\frac{N}{2}&-\frac{N}{2}+1&\cdots&\frac{N}{2}-1&\frac{N}{2}\\
    \left(-\frac{N}{2}\right)^2&\left(-\frac{N}{2}+1\right)^2&\cdots&\left(\frac{N}{2}-1\right)^2&\left(\frac{N}{2}\right)^2\\
    \cdots& \cdots&\cdots&\cdots&\cdots\\
    \cdots& \cdots&\cdots&\cdots&\cdots\\
    \cdots& \cdots&\cdots&\cdots&\cdots\\
    \left(-\frac{N}{2}\right)^N&\left(-\frac{N}{2}+1\right)^N&\cdots&\left(\frac{N}{2}-1\right)^N&\left(\frac{N}{2}\right)^N
    \end{array}\right)\left(\begin{array}{c}
    a_{-N/2}\\[1.9pt]
    a_{-N/2+1}\\[1.9pt]
    a_{-N/2+2}\\[1.9pt]
    \cdots\\[1.9pt]
    \cdots\\[1.9pt]
    \cdots\\[1.9pt]
    a_{N/2}
    \end{array}\right)=
    \left(\begin{array}{c}
    0\\[1.8pt]
    0\\[1.8pt]
    0\\[1.8pt]
    \cdots\\[1.8pt]
    m!\\
    \cdots\\[1.9pt]
    0
    \end{array}\right)
    \end{equation}
    <br>

   donde el único coeficiente no nulo en el vector de la derecha esta en la fila $m+1$.

## 4.4.3 Derivadas parciales

- La generalización del procedimiento que acabamos de estudiar a funciones de más de una variable es trivial.


- Dada una función que depende de $n$ variables $x_i$ con $i=1,\cdots,n$, su derivada parcial centrada usando diferencias finitas viene dada por:


$$\frac{\partial f\left(x_1,\cdots,x_n\right)}{\partial x_i} =\frac{f\left(x_1,\cdots,x_i+h/2,\cdots,x_n\right)-f\left(x_1,\cdots,x_i-h/2,\cdots,x_n\right)}{h}+{\cal O}(h^2),$$


- De forma similar se obtienen las expresiones para derivadas parciales con precisión ${\cal O}(h^n)$.


- De la misma forma, uno puede calcular derivadas cruzadas, calculando la derivada de una derivada parcial dada:
<br>

    \begin{align}
    \frac{\partial^2 f\left(x_1,\cdots,x_n\right)}{\partial x_i\partial x_j} =&\frac{f\left(x_1,\cdots,x_i+h/2,\cdots,x_j+h/2,\cdots,x_n\right)-f\left(x_1,\cdots,x_i+h/2,\cdots,x_j-h/2,\cdots,x_n\right)}{h^2}\\
    -&\frac{f\left(x_1,\cdots,x_i-h/2,\cdots,x_j+h/2,\cdots,x_n\right)-f\left(x_1,\cdots,x_i-h/2,\cdots,x_j-h/2,\cdots,x_n\right)}{h^2}+{\cal O}(h^2),
\end{align}

    y de forma equivalente para derivadas parciales superiores.

### Ejemplo 4.12: derivadas con ruido

Dada una función arbitraria $f$, el siguiente código nos permite añadirle un error gaussiano aleatorio.

```python
def corr_vars(start,stop,step,mu,sigma,f):

    from numpy import arange,random,zeros
    
    # Calcumlamos los valores de x
    x = arange(start, stop, step)   
    
    # Generamos ruido gausiano de media mu y desviación típica sigma.
    e = random.normal(mu, sigma, x.size)
    
    # Inicializamos el valor de nuestra función 
    y = zeros(x.size)

    # Le añadimos ruido a la función y = func(x) + e
    for k in range(x.size):
        y[k] = f(x[k]) + e[k]
    
    return x,y
```

* Usemos esta función para añadir ruido a una señal sinusoidal. 

```python
from math import pi
from numpy import sin,cos
from numpy import random
from matplotlib.pyplot import plot,title,show

# Creamos una semilla para la secuencia de ruido aleatorio
random.seed(2)

x,y = corr_vars(-2*pi,2*pi,0.05,0,1, f=lambda x: 2*pi*sin(x))  

plot(x,y)
plot(x,2*pi*sin(x),"r")
title('y = 2*π*sin(x) + ruido')
show()
```

* Si una variable $\bf x$ toma valores muy grandes (y puede dar un **overflow**), no es lo mismo calcular la división $\bf\displaystyle\frac{a}{x y}$ como **a/x/y** que como **a/(x\*y)**.

    Mientras que en el **primer caso** estamos dividiendo por un **overflow**, y el resultado va a ser **0**, en el **segundo** también estamos multiplicando, y el resultado va a ser **NaN** (not a number).

```python
# Prueba de Miller-Rabin para ver si un número es primo
# Python3 program Miller-Rabin primality test
import random 
 
# Utility function to do
# modular exponentiation.
# It returns (x^y) % p
def power(x, y, p):
     
    # Initialize result
    res = 1; 
     
    # Update x if it is more than or
    # equal to p
    x = x % p; 
    while (y > 0):
         
        # If y is odd, multiply
        # x with result
        if (y & 1):
            res = (res * x) % p;
 
        # y must be even now
        y = y>>1; # y = y/2
        x = (x * x) % p;print
     
    return res;
 
# This function is called
# for all k trials. It returns
# false if n is composite and 
# returns false if n is
# probably prime. d is an odd 
# number such that d*2<sup>r</sup> = n-1
# for some r >= 1
def miillerTest(d, n):
     
    # Pick a random number in [2..n-2]
    # Corner cases make sure that n > 4
    a = 2 + random.randint(1, n - 4);
 
    # Compute a^d % n
    x = power(a, d, n);
 
    if (x == 1 or x == n - 1):
        return True;
 
    # Keep squaring x while one 
    # of the following doesn't 
    # happen
    # (i) d does not reach n-1
    # (ii) (x^2) % n is not 1
    # (iii) (x^2) % n is not n-1
    while (d != n - 1):
        x = (x * x) % n;
        d *= 2;
 
        if (x == 1):
            return False;
        if (x == n - 1):
            return True;
 
    # Return composite
    return False;
 
# It returns false if n is 
# composite and returns true if n
# is probably prime. k is an 
# input parameter that determines
# accuracy level. Higher value of 
# k indicates more accuracy.
def isPrime( n, k):
     
    # Corner cases
    if (n <= 1 or n == 4):
        return False;
    if (n <= 3):
        return True;
 
    # Find r such that n = 
    # 2^d * r + 1 for some r >= 1
    d = n - 1;
    while (d % 2 == 0):
        d //= 2;
 
    # Iterate given number of 'k' times
    for i in range(k):
        if (miillerTest(d, n) == False):
            return False;
 
    return True;
```

```python
from gaussxw import gaussxwab
import numpy as np

x, w = gaussxwab(15, 0, 1)

def zetta(z):
    f = lambda x: np.sin(z*np.arctan(x)) / ( (1+x**2)**(z/2) * (np.exp(2*np.pi*x)-1) )
    def g(x_):
        gamma = 1
        c = 1
        x = (c*x_/(1-x_))**(1/gamma)
        J = c/gamma * x**(1-gamma) / (1-x_)**2
        return f(x)*J
    I = np.sum(w*g(x))
    I*= 2
    I += 1/(z-1)+1/2
    return I

print(zetta(2))
print(np.pi**2/6)
```

## Ejercicio 4.12: el periodo de un oscilador anharmónico

El oscilador harmónico aparece en múltiples problemas físicos y su comportamiento puede estudiarse por medio de métodos analíticos.  
Su principal propiedad es que su periodo de oscilación es constante e independiente de la amplitud del movimiento.  
Un oscilador clásico puede interpretarse como una partícula en un pozo cóncavo que, desplazado del punto de equilibrio, se balancea alrededor de él.

```python
from IPython.display import display,Image
display(Image(filename ="anharmonic_oscillator.png", width=450, height=450))
```

donde el potencial es $V(x)\propto x^2$.  
Cualquier otro potencial define un oscilador anharmónico.  
La forma más fácil de estudiar el movimiento de un oscilador es aplicar la conservación de la energía

$$E=\frac{1}{2}m\,\dot x^2+V(x),$$

que para $V(x)\neq x^2$ representa una ecuación no lineal.  
Asumamos que el potencial es simétrico respecto a x=0.  
Perturbemos entonces nuestra partícula desplazándola a una posición $x=a$ en $t=0$, que satisface que $\dot x=0$ y $E= V(a)$.

1. Cuando la partícula llega al origen ha recorrido un cuarto de su periodo. Demostrar que el periodo $(T)$ viene dado por:

$$T=\sqrt{8m}\,\int\limits_0^a\frac{\text{d}x}{\sqrt{V(a)-V(x)}}.$$

2. Asumir ahora que el potencial es $V(x)=x^4$ y la masa de la partícula es m=1.  
   Escribir un programa que calcule el periodo del oscilador para una amplitud inicial $a$ usando la cuadratura gaussiana con $N=20$ puntos. 

3. Representa el periodo para valores de la amplitud en el intervalo $[0,2]$.

4. El resultado muestra que la partícula se mueve más rápido (el periodo disminuye) al aumentar el valor de la amplitud.   
   ¿Cómo es posible si la partícula tiene que recorrer una mayor distancia? El periodo diverge cuando a=0. ¿Cual es la razón?

```python
# Apartado 1

import numpy as np
from gaussxw import gaussxwab

V = lambda x: x**4
m = 1

f = lambda x: 1/np.sqrt(V(a)-V(x))

def get_T(a):
    x, w = gaussxwab(20,0,a)
    T = np.sqrt(8*m)*np.sum(w*f(x))
    return T
```

```python
# Apartado 2

import matplotlib.pyplot as plt

for a in np.linspace(0,2,300):
    plt.plot(a, get_T(a), "ro", markersize=1.3)
    
plt.grid()
plt.xlabel("Desplazamiento inicial (a)")
plt.ylabel("Periodo (T)")
plt.show()
```

El periodo diverge cuando $a\to 0$ por la sencilla razón de que, si no hay desplazamiento inicial, entonces la partícula no se mueve (se queda en el punto de equilibrio del potencial, no hace falta ni que sea estable). Como el periodo se define como "el tiempo que tarda una partícula en regresar a un estado mecánico determinado", este tiempo será infinito (también tendría sentido que fuese 0?)

Observamos que, cuanto más desplazamos la partícula de la posición de equilibrio, menos tarda en recorrer un periodo. Esto se debe a que va a mayor velocidad por conservación de energía.

## Ejercicio 4.13: calculando la constante de Stefan-Boltzmann

La teoría de Planck nos dice que la densidad de energía emitida por unidad de área, tiempo y frecuencia ($\nu$) por un cuerpo negro (cuerpo no reflectante en equilibrio termodinámico con su entorno) es

$$I(\nu)=\displaystyle\frac{2\pi h}{c^2}\frac{\nu^3}{\text{e}^{\,h\nu\,/\,k_B\,T}-1}$$

donde $h$ es la constante de Planck, $c$ es la velocidad de la luz, $k_B$ la constante de Boltzmann, y $T$ la temperatura del cuerpo negro.

1. Mostrar que la energía total radiada por segundo y unidad de área viene dada por la integral:
    
    $$W=\frac{2\pi k_B^4 T^4}{c^2 h^3}\int\limits_0^\infty\frac{x^3}{\text{e}^x-1}\,\text{d}x.$$
    
2. Escribir un programa que calcule la integral de la expresión. Describe que método has usado y por qué, y cual es la precisión esperada. 

3. Incluso antes de la teoría cuántica, era bien sabido que la energía total emitida por unidad de área y tiempo por un cuerpo negro seguiría la ley de Stefan:

    $$W=\sigma\,T^4,$$

    donde $\sigma= 5.670374419 \times 10^{-8}$ W⋅m$^{-2}$⋅K$^{-4}$ es la constante de Stefan Boltzmann. Comprobar que se obtiene el resultado correcto.

La integral a computar se puede expresar como el producto de una gamma de Euler y una zeta de Riemann. Expandiendo dichas funciones, podemos pasar del cálculo de una integral al de una sumatoria, que computacionalmente es más sencillo:

$$W_{A,\nu} = \frac{12\pi k_B^4T^4}{c^2h^3}\displaystyle\sum_{n=1}^\infty \frac{1}{n^4}$$

```python
# La expresión sin el término cuártico de T debería darnos precisamente la constante de Stefan-Boltzmann

kb = 1.38*10**(-23)
c = 3*10**8
h = 6.63*10**(-34)

n = 1/np.arange(1, 1000)

sigma = 12*np.pi*kb**4*np.sum(n**(4))/(c**2*h**3)
print(sigma) # El error introducido viene dado en gran parte porque hemos truncado las constantes.
```

Podemos deriva e igualar a 0 para obtener la posición del máximo de la curva. (Un breve análisis del integrando revela que solo tiene un extremo y siempre va a ser máximo). Sin embargo, lo hacemos derivando numéricamente para practicar

```python
import numpy as np

def der_cent(f, x0):
    h = 10**(-5) # Minimizo error
    der = (f(x0+h/2)-f(x0-h/2)) / h
    return der

x = 5

f = lambda t: t**(x-1)*np.exp(-t)

t = np.linspace(0,10,100)

der = der_cent(f, t)

plt.plot(t, der, label="Derivada del integrando")
plt.plot(t, np.zeros_like(t), "k--")
plt.vlines(x-1, np.min(der), np.max(der), "r", label="Predicción analítica del 0")
plt.legend()
plt.show()
```

## Ejercicio 4.19: el procesamiento de imágenes y el microscopio de efecto túnel 


Cuando la luz incide sobre una superficie, la cantidad que penetra por unidad de superficie depende no sólo de la intensidad de la luz, sino también del ángulo de incidencia. Si la luz forma un ángulo θ con la normal, sólo "ve" cos θ de área por unidad de área real en la superficie:

```python
from IPython.display import display,Image
display(Image(filename ="lightstm.png", width=450, height=450))
```

Así, la intensidad de la iluminación es $a\,\cos\theta$, siendo $a$ la intensidad total de la luz. Esta sencilla ley física es un elemento central de las gráficas en 3D. Nos permite calcular cómo incide la luz sobre los objetos tridimensionales y, por lo tanto, cómo se verán cuando se les ilumine desde diferentes ángulos. 

Supongamos, por ejemplo, que estamos mirando la Tierra desde arriba y vemos montañas. Conocemos la altura de las montañas $w(x,y)$ en función de su posición en el plano, por lo que la ecuación de la superficie de la Tierra es simplemente $z = w(x,y)$, o, de forma equivalente, $w(x,y)-z = 0$, y el vector normal $\bf v$ a la superficie viene dado por el gradiente de $w(x,y)-z$ así:

$${\bf v}=\nabla[z-w(x,y)]=\left(\begin{array}{c}\partial/\partial x \\\partial/\partial y\\\partial/\partial z\end{array}\right)[z-w(x,y)]=\left(\begin{array}{c}-\partial w/\partial x \\-\partial w/\partial y\\1\end{array}\right).$$

Ahora supongamos que tenemos una luz que entra representada por un vector $\bf a$ con magnitud igual a la intensidad de la luz. Entonces el producto escalar de los vectores ${\bf a}$ y ${\bf v}$ es

$${\bf a}\cdot {\bf v}=\vert {\bf a}\vert \vert {\bf v}\vert \cos \theta,$$

donde θ es el ángulo entre los dos vectores. Por tanto, la intensidad de la iluminación de la superficie de las montañas es 

$$I=\vert {\bf a}\vert \cos \theta=\frac{{\bf a}\cdot {\bf v}}{\vert {\bf v}\vert}=\frac{-a_x(\partial  w/\partial x)-a_y(\partial  w/\partial y)+a_z}{\sqrt{(\partial w/\partial x)^2+(\partial w/\partial y)^2+1}}.$$


Tomemos un caso sencillo en el que la luz incide horizontalmente con intensidad unitaria, a lo largo de una línea un ángulo φ en sentido contrario al de las agujas del reloj desde el eje este-oeste, de modo que a = (cos φ, sin φ, 0). Entonces nuestra intensidad de iluminación se simplifica a

$$I=\frac{-\cos\phi(\partial  w/\partial x)+\sin\phi(\partial  w/\partial y)}{\sqrt{(\partial w/\partial x)^2+(\partial w/\partial y)^2+1}}.$$

Si podemos calcular las derivadas de la altura $w(x,y)$ y conocemos φ podemos calcular la intensidad en cualquier punto.
