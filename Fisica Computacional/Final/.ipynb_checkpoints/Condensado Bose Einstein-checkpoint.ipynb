```python
import numpy as np
import matplotlib.pyplot as plt
from gaussxw import gaussxwab
from math import gamma
from scipy.special import zeta
```

```python
N=700
u, w = gaussxwab(N,0,1)

def g(z, n):
    f = lambda x: x**(n-1)/(np.exp(x)/z-1) / gamma(n)
    h = lambda u: 1/(1-u)**2 * f(u/(1-u))
    I = np.sum(w*h(u))
    return I

g = np.vectorize(g)
```

```python
z = np.linspace(0,1,N)

plt.plot(z, g(z, 1), "r", label="$g_1(z)$")
plt.plot(z, g(z, 2), "b", label="$g_2(z)$")
plt.plot(z, zeta(2)*np.ones_like(z), "g--", label="$\\zeta(2)$")
plt.xlabel("$z$")
plt.legend()
plt.show()
```
