## Apartado 1

```python
M=1
m=0.1
l=0.2
g=9.81
mu=10
```

```python
import numpy as np
import matplotlib.pyplot as plt

def F(r):
    theta, x, w, v = r
    lamb = 1/l/(M+m*np.sin(theta)**2)
    u=0
    a_theta = lamb*( (m+M)*g*np.sin(theta) -m*g*l*w**2*np.sin(theta)*np.cos(theta) +mu*v*np.cos(theta) -u*np.cos(theta) )
    a_x = lamb*( m*g*l**2*w**2*np.sin(theta) -m*g*l*np.sin(theta)*np.cos(theta) -l*mu*v +u*l )
    
    return np.array([w, v, a_theta, a_x], float)

t=0
t_max = 10
N=1000
h = t_max/N
tp = np.linspace(0,t_max, N)

# Conds. iniciales
r = np.array([0,0,0,0], float)
xp = []
thetap = []

# Usamos Euler porque, total, obtendremos una solución estática
for t in tp:
    thetap.append(r[0])
    xp.append(r[1])
    
    r+=h*F(r)
    
plt.xlabel("$t[s]$")
plt.ylabel("$\\theta(t) [rad]$")
plt.plot(tp, thetap)
plt.show()

plt.xlabel("$t[s]$")
plt.ylabel("$x(t) [m]$")
plt.plot(tp, xp)
plt.show()
```

## Apartado 2

```python
import numpy as np
import matplotlib.pyplot as plt
from random import random

def F(r):
    theta, x, w, v = r
    lamb = 1/l/(M+m*np.sin(theta)**2)
    u=0.01*random()
    a_theta = lamb*( (m+M)*g*np.sin(theta) -m*g*l*w**2*np.sin(theta)*np.cos(theta) +mu*v*np.cos(theta) -u*np.cos(theta) )
    a_x = lamb*( m*g*l**2*w**2*np.sin(theta) -m*g*l*np.sin(theta)*np.cos(theta) -l*mu*v +u*l )
    
    return np.array([w, v, a_theta, a_x], float)

t=0
t_max = 10
N=1000
h = t_max/N
tp = np.linspace(0,t_max, N)

# Conds. iniciales
r = np.array([0,0,0,0], float)
xp = []
thetap = []
wp = []
vp = []
T = []
V = []


for t in tp:
    theta, x, w, v = r
    thetap.append(theta)
    xp.append(x)
    wp.append(w)
    vp.append(v)
    
    T.append(1/2*M*v**2+1/2*m*l**2*w**2)
    V.append(m*g*l*np.cos(theta))

    k1 = h*F(r)
    k2 = h*F(r+k1/2)
    k3 = h*F(r+k2/2)
    k4 = h*F(r+k3)
    
    r+=1/6*(k1+2*k2+2*k3+k4)
    
plt.xlabel("$t[s]$")
plt.ylabel("$\\theta(t) [rad]$")
plt.plot(tp, thetap)
plt.show()

plt.xlabel("$t[s]$")
plt.ylabel("$x(t) [m]$")
plt.plot(tp, xp)
plt.show()
```

## Apartado 3

```python
xp = np.array(xp)
thetap = np.array(thetap)

xpend = xp + l*np.sin(thetap)
ypend = l*np.cos(thetap)

plt.xlabel("$x_{pend}$")
plt.ylabel("$y_{pend}$")
plt.plot(xpend, ypend)
plt.show()

plt.xlabel("$\\theta$")
plt.ylabel("$\\dot \\theta$")
plt.plot(thetap, wp)
plt.show()

plt.xlabel("$x$")
plt.ylabel("$\\dot x$")
plt.plot(xp, vp)
plt.show()
```

## Apartado 4

```python
T=np.array(T)
V=np.array(V)
E=T+V

plt.xlabel("$t[s]$")
plt.ylabel("$T [J]$")
plt.plot(tp, T)
plt.show()

plt.xlabel("$t[s]$")
plt.ylabel("$V [J]$")
plt.plot(tp, V)
plt.show()

plt.xlabel("$t[s]$")
plt.ylabel("$E [J]$")
plt.plot(tp, E)
plt.show()
```
