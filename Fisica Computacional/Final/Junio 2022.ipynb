## Apartado 1

```python
m1=2
m2=3
L1=2
L2=1
g=9.81
```

```python
# Como el péndulo doble es un sistema periódico, empleamos el salto de la rana
import numpy as np
import matplotlib.pyplot as plt

def F(r):
    theta1, theta2, w1, w2 = r
    c12 = np.cos(theta1-theta2)
    s12 = np.sin(theta1-theta2)
    n=c12**2*m2-m1-m2
    
    a1 = 1/(L1*n) * ( L1*m2*c12*s12*w1**2 +L2*m2*s12*w2**2 +(m1+m2)*g*np.sin(theta1) -m2*g*c12*np.sin(theta2) )
    a2 = -1/(L2*n) * ( L1*(m1+m2)*s12*w1**2 +L2*m2*c12*s12*w2**2 +(m1+m2)*g*c12*np.sin(theta1) -(m1+m2)*g*np.sin(theta2) )

    return np.array([w1, w2, a1, a2], float)
    
t0=0
t_max=180 #s
N=10000
h = t_max/N
tp=np.linspace(t0,t_max, N)

# Cond. inicial
r = np.array([0,0,0,0.1])

# Resolución por método del salto de la rana
r_med = r+h/2*F(r)
theta1p = []
theta2p = []
w1p = []
w2p = []
T=[]
V=[]

for t in tp:
    theta1, theta2, w1, w2 = r
    theta1p.append(theta1)
    theta2p.append(theta2)
    w1p.append(w1)
    w2p.append(w2)
    T.append(1/2*m1*L1**2*w1**2 + 1/2*m2*L2**2*w2**2)
    V.append(-m1*g*L1*np.cos(theta1) -m2*g*(L1*np.cos(theta1)+L2*np.cos(theta2)))
    
    r+=h*F(r_med)
    r_med+=h*F(r)
```

```python
# Representamos theta1(t) y theta2(t) en los primeros 10s

ind = int(10000/18)+2 # Primeros 10s

plt.title("$\\theta_1(t)$")
plt.xlabel("t[s]")
plt.ylabel("$\\theta_1(t) [rad]$")
plt.plot(tp[:ind], theta1p[:ind])
plt.show()

plt.title("$\\theta_2(t)$")
plt.xlabel("t[s]")
plt.ylabel("$\\theta_2(t) [rad]$")
plt.plot(tp[:ind], theta2p[:ind])
plt.show()
```

## Apartado 2

```python
# Representamos coordenadas cartesianas durante los últimos 10s

theta1p = np.array(theta1p)
theta2p = np.array(theta2p)

xpend1 = L1*np.sin(theta1p)
ypend1 = -L1*np.cos(theta1p)
xpend2 = xpend1 + L2*np.sin(theta2p)
ypend2 = ypend1 - L2*np.cos(theta2p)

ind = int(17/18*1e4)
plt.title("Péndulo 1")
plt.xlabel("x")
plt.ylabel("y")
plt.plot(xpend1[ind:], ypend1[ind:])
plt.show()


plt.title("Péndulo 2")
plt.xlabel("x")
plt.ylabel("y")
plt.plot(xpend2[ind:], ypend2[ind:])
plt.show()
```

```python
# Trayectorias de los 2 péndulos por sus respectivos espacios de fases los últimos 10s

plt.title("Espacio de fases P1")
plt.xlabel("$\\theta_1$")
plt.ylabel("$\\dot\\theta_1$")
plt.plot(theta1p[ind:], w1p[ind:])
plt.show()

plt.title("Espacio de fases P2")
plt.xlabel("$\\theta_2$")
plt.ylabel("$\\dot\\theta_2$")
plt.plot(theta2p[ind:], w2p[ind:])
plt.show()
```

## Apartado 3

```python
T = np.array(T)
V = np.array(V)
E=T+V

plt.plot(tp[ind:], T[ind:], label="T")
plt.plot(tp[ind:], V[ind:], label="V")
plt.plot(tp[ind:], E[ind:], label="E")
plt.legend()
plt.show()
```
