# 6.1 Series de Fourier

- Sea f(x) una función periódica en el intervalo $[0,L]$:

$$f(x+L)=f(x),\;\forall x\in {\cal D}_f$$

- que cumple los las condiciones:    
<br/>

     - Es absolutamente integrable.
     - Esta acotada. 
     - Tiene un número finito de discontinuidades en su periodo, y estas son finitas. 
    
    
    
- Entonces dicha función puede ser escrita como una serie de Fourier.


- Si la función es par con respecto al punto medio de su intervalo periódico: 

    $$f(L/2-x)=f(L/2+x),$$

    entonces puede ser escrita como:

    $$f(x)=\sum_{k=0}^\infty{\alpha_k\,\cos\left(\frac{2\pi k x}{L}\right)}.$$


- Si la función es impar: 

    $$f(L/2-x)=-f(L/2+x),$$

    su serie de Fourier es:

    $$f(x)=\sum_{k=1}^\infty{\beta_k\,\sin\left(\frac{2\pi k x}{L}\right)}.$$


- Mientras que en el caso general, cualquier función periódica sin simetría definida puede ser escrita como:

   $$f(x)=\sum_{k=0}^\infty{\alpha_k\,\cos\left(\frac{2\pi k x}{L}\right)}+\sum_{k=1}^\infty{\beta_k\,\sin\left(\frac{2\pi k x}{L}\right)} =\sum_{k=-\infty}^\infty{\gamma_k\,\exp\left(i\frac{2\pi k x}{L}\right)},\quad$$

    donde los coeficientes $\gamma$ vienen dados por:

$$\gamma_k=\left\{\begin{array}{lr}\frac{1}{2}\left(\alpha_{-k}+i\beta_{-k}\right)&\text{if}\; k<0,\nonumber\\\alpha_0&\text{if}\; k=0,\nonumber\\\frac{1}{2}\left(\alpha_{k}-i\beta_{k}\right)&\text{if}\; k>0.\nonumber\\\end{array}\right.$$

- Una vez determinado el valor de los coeficientes $\gamma_k$ de una función periódica $f(x)$, la serie de Fourier permite obtener una representación de la función simple y compacta para su análisis numérico.


- La series de Fourier se limitan a funciones periódicas.   
  Sin embargo, si estamos interesados en describir una función no periódica $f(x)$ solamente en un intervalo $[0,L]$, podemos describirla en función de la serie de Fourier de una función periódica $g(x)$ que coincide con ella en el intervalo $[0,L]$.

## 6.1.1 Determinación de los coeficientes de Fourier

- Los coeficientes $\gamma_k$ de la serie de Fourier de una función $f(x)$ se pueden obtener trivialmente usando que:

    $$\int\limits_0^L{\exp\left(i\frac{2\pi\left(k'-k\right)x}{L}\right)\text{d}x}=\frac{-i L}{2\pi\left(k'-k\right)}\left[\text{e}^{i\,2\pi(k'-k)}-1\right]=L\,\delta_{k'\,k}\,,$$


- De esta forma se obtiene que:

    $$\int\limits_0^L{f(x)\exp\left(-i\frac{2\pi kx}{L}\right)}\text{d}x=\sum_{k'=-\infty}^{\infty}{\gamma_{k'}\int\limits_0^L{\exp\left(i\frac{2\pi\left(k'-k\right)x}{L}\right)\text{d}x}}=L\gamma_k,$$

    y los coeficientes vienen dados por la integral:

    $$\gamma_k=\frac{1}{L}\int\limits_0^L{f(x)\exp\left(-i\frac{2\pi kx}{L}\right)}\text{d}x.$$

# 6.2 Transformada de Fourier discreta

- Mientras que para algunas funciones podremos calcular las integrales que involucran $\gamma_k$ de forma analítica, en general esto no será posible, bien porque la integral tiene que calcularse numéricamente o porque no se tiene una expresión analítica para $f(x)$.


- En estos casos las integrales de los coeficientes tendrán que calcularse numéricamente.


- Por ejemplo aproximando la integral por la regla del trapecio se tiene con $N$ particiones en los puntos  $x_n=\displaystyle\frac{n\,L}{N}$

    $$\gamma_k=\frac{1}{L}\frac{L}{N}\left[\frac{1}{2}f(0)+\frac{1}{2}f(L)+\sum_{n=1}^{N-1}{f\left(x_n\right)\exp{\left(-i\frac{2\pi k x_n}{L}\right)}}\right].$$


- Sin embargo, como la función es periódica $f(0)=f(L)$ y se tiene

    $$\gamma_k=\frac{1}{N}\sum_{n=0}^{N-1}{f\left(x_n\right)\exp{\left(-i\frac{2\pi k x_n}{L}\right)}}.$$

    que se puede calcular perfectamente en un ordenador.


- Es por tanto la expresión a usar cuando sólo conocemos la función $f(x)$ en un conjunto equiespaciado de puntos $x_n$. 


- Por ejemplo si $f(x)$ representa una señal de audio (una onda de sonido muestreada a un ritmo constante).


- En ese caso llamando:

    $$x_n=\frac{n L}{N},\quad\text{y}\quad y_n=f(x_n),$$

    tenemos que:

    $$\gamma_k=\frac{1}{N}\sum_{n=0}^{N-1}{y_n\exp{\left(-i\frac{2\pi k\,n}{N}\right)}}.$$


- Sólo necesitamos conocer los valores de la muestra $y_n$ y el número total de puntos $N$.


- Por convención llamamos **Transformada de Fourier Discreta (DFT)** a la suma

    $$c_k=N \gamma_k=\sum_{n=0}^{N-1}{y_n\exp{\left(-i\frac{2\pi k\,n}{N}\right)}},$$

    donde de forma habitual tanto a los coeficientes $c_k$ como $\gamma_k$ se les suele llamar coeficientes de Fourier.

## 6.2.1 Transformada de Fourier inversa discreta

- Aunque para obtener los coeficientes de Fourier hemos realizado la integral aproximadamente (hemos usado la regla del trapecio), esta aproximación no involucra ninguna perdida de información.


- Teniendo en cuenta que la suma
    
    $$\sum_{k=0}^{N-1}\text{e}^{i2\pi k (m-n)/N}=N\delta_{m\,n N},\;\text{con}\; n\in{\mathbb N},$$

    partiendo de la transformada de Fourier discreta tenemos que:

    $$\sum_{k=0}^{N-1}c_k \exp\left(i\frac{2\pi k n}{N}\right)=\sum_{k=0}^{N-1}\sum_{m=0}^{N-1}y_m\exp\left(i\frac{2\pi k (n-m)}{N}\right)=Ny_n,$$

    y por tanto usando nuestros coeficientes de Fourier obtenemos exactamente los valores de la muestra

    $$y_n=\frac{1}{N}\sum_{k=0}^{N-1}c_k \exp\left(i\frac{2\pi k n}{N}\right),$$

    que es conocida como  **Transformada de Fourier inversa discreta (DFT inversa)**.


- Podemos por tanto ir hacia delante y hacia detrás del conjunto $\{c_k\}$ al $\{y_n\}$ y viceversa sin que haya ninguna perdida de información.


- Es importante notar que, dada la periodicidad de la función, sólo necesitamos calcular  los coeficientes $\{c_k\}$ hasta k = N − 1.  
  Por tanto nuestra  transformada de Fourier discreta difiere de la expresión general de la serie de Fourier:
  
    $$f(x)=\sum_{k=-\infty}^\infty{\gamma_k\,\exp\left(i\frac{2\pi k x}{L}\right)},\quad$$

    pues el sumatorio de la serie de Fourier va de $-\infty$ a $+\infty$.


- La transformada de Fourier discreta sólo nos da los valores de la muestra, $y_k$ y no nos dice nada de los valores de $f(x)$ en los puntos intermedios.  


- Dos funciones que tengan los mismo valores en los puntos de la muestra, tendrán la misma DFT, aunque sean funciones distintas.

```python
display(Image("DFT.png",width=800))
```

- Sin embargo, si la función es suficientemente suave, y lo que nos interesa es describirla a partir de un conjunto de puntos, la DFT será una aproximación muy buena de la función.


- La DFT funciona igual de bien para funciones reales como complejas.


- Sin embargo, en el caso de funciones reales los coeficientes de Fourier cumplen ciertas propiedades adicionales:
<br/>

    - Dado el coeficiente $c_r$ con $1\le r < \frac{1}{2} N$, el coeficiente $c_{N-r}$  cumple que:
        
    $$c_{N-r}=\sum_{n=0}^{N-1}y_n\exp\left(-i\frac{2\pi(N-r)n}{N}\right)=\sum_{n=0}^{N-1}y_n\exp\left(i\frac{2\pi r n}{N}\right)=\displaystyle c_r^*,$$


- Sólo es necesario calcular la mitad de los coeficientes 
<br/>

    - Si $N$ es par tendremos que calcular $\frac{N}{2}+1$, es decir $c_k$ con $0\le k \le \frac{N}{2}$.
<br/>

    - Si $N$ es impar necesitaremos  $\frac{N+1}{2}$, es decir $c_k$ con  $0\le k \le \frac{N-1}{2}$.
<br/>
        
    - Ambos casos se pueden representar en python con la operación: N//2+1.
    
    
    
- Si la función es compleja esta simplificación no se aplica que tendremos que calcular los $N$ coeficientes.     

### Ejemplo 6.1: la transformada de Fourier discreta

Escribir un programa que calcule la transformada de Fourier discreta de una muestra real de datos contenidos en una array.   
Cargar el archivo "pitch.txt" en la carpeta de datos del tema 6 y representar los coeficientes de Fourier de esa función.

```python
import numpy as np
import matplotlib.pyplot as plt

y = np.loadtxt("pitch.txt")
plt.plot(y)
plt.show()
```

```python
import numpy as np

def DFT_re(y_n): # calcula los coefs. c_k de Fourier
    N = len(y_n)
    n = np.array([i for i in range(0,N)])
    
    c_k = [np.sum(y_n*np.exp(-2*np.pi*1j*k*n/N)) for k in range(0,N//2+1)]
    return c_k

def DFT(y_n): # calcula los coefs. c_k de Fourier
    N = len(y_n)
    n = np.array([i for i in range(0,N)])
    
    c_k = [np.sum(y_n*np.exp(-2*np.pi*1j*k*n/N)) for k in range(0,N)]
    return c_k

plt.plot(np.abs(DFT(y)))
plt.show()
```

```python
def DFT(y): # me da los coefs
    N = len(y)
    c = np.zeros(N//2+1, complex)
    for k in range(N//2+1):
        for n in range(N):
            c[k]+=y[n]*np.exp(-2j*np.pi*k*n/N)
    return c


c = DFT(y)
plt.plot(np.abs(c))
plt.xlim(0,500)
plt.show()
```

## 6.2.2 Eligiendo los puntos de muestra

- Dada una muestra/partición inicial $\{x_n\}$, podemos desplazar los puntos una distancia arbitraria $\Delta$ obteniendo:

    $$x_n'=x_n+\Delta=\frac{n}{N}L+\Delta,$$

    cuyos coeficientes de Fourier serán:

    \begin{align}
    c_k=&\sum_{n=0}^{N-1}f\left(x_n+\Delta\right)\exp\left(-i\frac{2\pi k(x_n+\Delta)}{L}\right)=\exp\left(-i\frac{2\pi k \Delta}{L}\right)\sum_{n=0}^{N-1}f\left(x'_n\right)\exp\left(-i\frac{2\pi kx_n}{L}\right)\nonumber\\
=&\,\exp\left(-i\frac{2\pi k \Delta}{L}\right)\sum_{n=0}^{N-1}y'_n\exp\left(-i\frac{2\pi kx_n}{L}\right).\nonumber\\
    \end{align}


- Por tanto, llamando 

    $$c'_k=c_k \exp\left(i\frac{2\pi k \Delta}{L}\right)$$

    se obtiene que

    $$c'_k=\sum_{n=0}^{N-1}y'_n\exp\left(-i\frac{2\pi kx_n}{L}\right),$$

    que es expresión equivalente a la de los coeficientes de Fourier para la partición inicial $\{x_n\}$. 


- La transformada de Fourier discreta es independiente de los puntos de muestra que se escojan. Los pesos sólo cambiaran por un factor de fase. 


- Dado un intervalo $[0,L]$ en el que la función es periódica, dividido en $N$ subintervalos hay dos elecciones habituales para definir la transformada de Fourier discreta:
<br/>

    - La llamada DFT de tipo I, en el que los puntos de muestra se toman en el comienzo de cada subintervalo.
    - La DFT de tipo II en que los puntos de muestra se toman en la mitad de cada subintervalo.

```python
display(Image("DFTI.png",width=600))
display(Image("DFTII.png",width=600))
```

- Los coeficientes de Fourier para cada una de estas transformadas de Fourier discretas obviamente serán diferentes:
<br/>

    - En primer lugar, los valores de la función para las dos muestras son diferentes.
    - En segundo, porque los coeficientes para la DFT de tipo II llevan un factor de fase. 

## 6.2.3 Transformada de Fourier en 2-dimensiones

**IMPORTANTE:** la primera función es real --> solo hay que calcular los N//2+1 primeros coefs, pero la transformada es compleja, de modo que en la segunda transformada hay que tener en cuenta los N puntos!!!

- La transformada de Fourier discreta se puede generalizar de forma trivial a funciones de más de una variable.


- En el caso particular de una función de dos dimensiones, dada una muestra  con $M\times N$ puntos, si fijamos la coordenada $x$ a la fila $m$:

    $$c_{ml}=\sum_{n=0}^{N-1}y_{mn}\exp\left(-i\frac{2\pi l n}{N}\right).$$


- Variando ahora cada una de la filas:

    $$c_{kl}=\sum_{m=0}^{M-1}\sum_{n=0}^{N-1}y_{mn}\exp\left[-2i\pi\left(\frac{ k m}{M}+\frac{ l n}{N}\right)\right],$$

    que proporciona el valor de los coeficientes de Fourier para una función en dos dimensiones. 


- La correspondiente transformada de Fourier inversa será:

    $$y_{mn}=\frac{1}{M N}\sum_{k=0}^{M-1}\sum_{l=0}^{N-1}c_{kl}\exp\left[2i\pi\left(\frac{ k m}{M}+\frac{ l n}{N}\right)\right].$$


- Si la función es real, es importante notar que, mientras que al fijar una fila el valor de la transformada discreta corresponde a la de una función real, al variar las filas, estamos haciendo la transformada de la función:

    $$\sum_{n=0}^{N-1}y_{mn}\exp\left(-i\frac{2\pi l n}{N}\right),$$

    que es compleja.


- Por tanto, en este caso tendremos que calcular $M\times \left(\frac{1}{2}N+1\right)$ si $N$ es par y $M\times \frac{1}{2}\left(N+1\right)$ si $N$ es impar. 

    
- Eso implica que, dependiendo del valor de $M$ y $N$ podría ser más conveniente hacer transformada de Fourier en un orden o el otro. 

    Si, por ejemplo, $M\gg N$, será más conveniente hacer primero la transformada para las columnas. 
    

- Las transformadas de Fourier de dos dimensiones se utilizan de forma general en el tratamiento y procesamiento de imágenes, por ejemplo en:
<br/>

    - en astronomía para analizar señales de estrellas y galaxias.
    - en la microscopía electrónica.

## 6.2.4 Interpretación Física de la Transformada de Fourier

- La transformada de Fourier nos permite expresar/separar una función real o compleja en términos de ondas sinusoidales.


-  Que una función se pueda expresar como una transformada de Fourier es equivalente a decir que la función:   
   se puede expresar como una suma de ondas sinusoidales de diferentes frecuencias.
   
   
- Por tanto, los coeficientes de la transformación nos dicen cual es la contribución de cada frecuencia en la suma.
   

- Viendo la transformada de Fourier de una señal obtenemos una imagen de como se separa en frecuencias la señal.  
  Este es básicamente el funcionamiento del análisis de señales. 
 
 
- Volvamos a analizar la señal de ejercicio anterior. 

```python
show(plot(y))
```

- La señal parece consistir en una onda sinusoidal con una frecuencia bien definida. 


- Sin embargo, esta señal no es perfecta:
<br/>

    - Cada módulo tiene pequeñas variaciones.    
    
    - Sobre cada una de ellas hay oscilaciones que básicamente parecen aleatorias. 
    
    
    
- Volvamos a estudiar sus coeficientes de Fourier:

```python
show(plot(abs(c)))
```

- La gráfica pone en manifiesto una serie de picos dominantes.

```python
plot(abs(c))
xlim(0,125)
show()
```

- El primero de ellos es claramente el más grande, e indica la frecuencia dominante que produce ese comportamiento casi sinusoidal.


- El resto de picos, son armónicos del primero, y hacen que la señal no sea exactamente una onda sinusoidal.


- Finalmente entre cada pico hay un conjunto casi aleatorio de de frecuencias.  
  Representan un ruido de fondo de la señal, el llamado ruido blanco.

# 6.3 Transformadas de cosenos y senos discretas

Me interesa porque, al hacer periódica una función que no lo es, doy lugar a discontinuidad --> problemático porque el punto de discont. me converge a la media aritmética del valor por izq. y por dcha. Si hago solo extensión par / impar de la función, me ahorro la discont. ==> no tengo problemas

- Hasta ahora hemos usado la representación compleja de la transformada de Fourier. 


- Sin embargo si nuestra función tiene una simetría par o impar podemos usar sólo la representación en cosenos o senos.


- A primera vista, esto parece restringir de forma evidente su aplicación.


- No obstante, al igual que podemos aplicar la DFT a una función no periódica al repetirla continuamente,   
  podemos hacer algo parecido con las series de senos y cosenos.


- Por ejemplo para el serie del coseno, podemos simetrizar una función cualquiera definida en el intervalo $[0,\frac{L}{2}]$ simplemente añadiendo su imagen especular en el intervalo $[\frac{L}{2},L]$.  


- Eso implica que en su versión discreta, el numero de puntos de una muestra siempre va a ser par. 

```python
from IPython.display import Image,display
display(Image("cosine-transform.png"))
```

- Para la serie del seno podríamos hacer exactamente lo mismo. Sin embargo, la antisimetría del seno requeriría que la función se anulase en los extremos del intervalo $[0,\frac{L}{2}]$, lo que hace que sea mucho menos usada.


- Una vez esta función ha sido construida, su transformada de Fourier se puede calcular trivialmente. 


- Además, al ser simetría con respecto a $x=\frac{L}{2}$ tenemos que, dada una partición $x_n=\frac{n L}{N}$ la función satisface que:

    $$y_n=f(x_n)=f(x_{N-n})=y_{N-n},$$

    que implica que $N$ tiene que ser par. 


- De esta forma, la transformada coseno de la función así definida viene dada por:

    \begin{align}
    c_k=\sum_{n=0}^{N-1}{y_n\exp\left(-i\frac{2\pi k n}{N}\right)}=&\sum_{n=0}^{N/2}{y_n\exp\left(-i\frac{2\pi k n}{N}\right)}+\sum_{n=N/2+1}^{N-1}{y_n\exp\left(-i\frac{2\pi k n}{N}\right)}\nonumber\\
    =&\sum_{n=0}^{N/2}{y_n\exp\left(-i\frac{2\pi k n}{N}\right)}+\sum_{n=N/2+1}^{N-1}{y_{N-n}\exp\left(i\frac{2\pi k (N-n)}{N}\right)}\nonumber\\
    =&\sum_{n=0}^{N/2}{y_n\exp\left(-i\frac{2\pi k n}{N}\right)}+\sum_{n=1}^{N/2-1}{y_{n}\exp\left(i\frac{2\pi k n}{N}\right)}\nonumber\\
    =&y_0+y_{N/2}\cos\left(\pi k \right)+2\sum_{n=1}^{N/2-1}{y_{n}\cos\left(\frac{2\pi k n}{N}\right)}.\nonumber\\
    \end{align}

    que se denomina **transformada discreta coseno (DCT)** de tipo I (la función es evaluada en los puntos de la muestra).


- Cuando se aplica sobre funciones reales, la mayoría de los casos, los coeficientes $c_k$ son reales, y por tanto cumplen que $c_{N-n}=c_n*=c_n$.


- De esta forma, la transformada inversa coseno de una función real es:

    \begin{align}
    y_n=\frac{1}{N}\sum_{j=0}^{N-1}{c_k\exp\left(i\frac{2\pi k n}{N}\right)}=&\frac{1}{N}\sum_{k=0}^{N/2}{c_k\exp\left(i\frac{2\pi k n}{N}\right)}+\sum_{k=N/2+1}^{N-1}{c_k\exp\left(i\frac{2\pi k n}{N}\right)}\nonumber\\
    =&\frac{1}{N}\sum_{k=0}^{N/2}{c_k\exp\left(i\frac{2\pi k n}{N}\right)}+\sum_{k=N/2+1}^{N-1}{c_{N-k}\exp\left(-i\frac{2\pi (N-k)n}{N}\right)}\nonumber\\
    =&\frac{1}{N}\sum_{n=0}^{N/2}{c_k\exp\left(i\frac{2\pi k n}{N}\right)}+\sum_{k=1}^{N/2-1}{c_{k}\exp\left(-i\frac{2\pi k n}{N}\right)}\nonumber\\
    =&\frac{1}{N}\left[c_0+c_{N/2}\cos\left(\pi n\right)+2\sum_{k=1}^{N/2-1}{c_{k}\cos\left(\frac{2\pi k n}{N}\right)}\right],\nonumber\\
    \end{align}

    que se denomina  **transformada discreta coseno inversa (DCT inversa)** y es equivalente a la transformada directa salvo por el factor $1/N$. 

## 6.3.1 Transformadas de cosenos y senos discretas de tipo II

- De la misma forma que para las transformadas de Fourier, también es habitual usar las transformadas de senos y cosenos de tipo II, aquellas en las que la función se evalúa entre dos puntos de la muestra. 


- Para el caso de la series de cosenos de tipo II, la simetría con respecto al punto medio $x=\frac{L}{2}$ implica que  $y_n=y_{N-1-n}$ por lo que pueden escribirse como:

    \begin{align}
c_k=&\sum_{n=0}^{N/2-1}{y_n\exp\left(-i\frac{2\pi k n}{N}\right)}+\sum_{n=N/2}^{N-1}{y_n\exp\left(-i\frac{2\pi k n}{N}\right)}\nonumber\\
    =&\exp\left(i\frac{\pi k}{N}\right)\left[\,\sum_{n=0}^{N/2-1}{y_n\exp\left(-i\frac{2\pi k(n+1/2)}{N}\right)}+\sum_{n=N/2}^{N-1}{y_{N-n-1}\exp\left(i\frac{2\pi k (N-1/2-n)}{N}\right)}\,\right]\nonumber\\
    =&\exp\left(i\frac{\pi k}{N}\right)\left[\,\sum_{n=0}^{N/2-1}{y_n\exp\left(-i\frac{2\pi k (n+1/2)}{N}\right)}+\sum_{n=0}^{N/2-1}{y_{n}\exp\left(i\frac{2\pi k (n+1/2)}{N}\right)}\,\right]\nonumber\\
    =&2\exp\left(i\frac{\pi k}{N}\right)\sum_{n=0}^{N/2-1}{y_{n}\cos\left(\frac{2\pi k (n+1/2)}{N}\right)}.\nonumber\\
    \end{align}


- En general el factor $\exp\left(i\frac{\pi k}{N}\right)$ se suele reabsorber definiendo un nuevo parámetro

    $$a_k=\exp\left(-i\frac{\pi k}{N}\right)c_k=2\sum_{n=0}^{N/2-1}{y_{n}\cos\left(\frac{2\pi k (n+1/2)}{N}\right)},$$

    que es la forma habitual de escribir la **transformada de cosenos discreta de tipo II**.


- De la misma forma, la **transformada de cosenos discreta inversa de tipo II** viene dada por:

    $$y_n=\frac{1}{N}\left[ a_0+2\sum_{k=1}^{N/2-1}{a_{k}\cos\left(\frac{2\pi k (n+1/2)}{N}\right)}\,\right]$$


- Dada su simplicidad la transformada de cosenos discreta de tipo II es probablemente la versión  más usada.


- Transforma $N/2$ elementos en otros $N/2$. Por eso normalmente se suele sustituir $N/2\to N$ y $a_k\to 2 a_k$ de forma que su expresión se reduce a:

    $$a_k=\sum_{n=0}^{N-1}{y_{n}\cos\left(\frac{\pi k (n+1/2)}{N}\right)},\quad y_n=\frac{1}{N}\left[ a_0+2\sum_{k=1}^{N-1}{a_{k}\cos\left(\frac{\pi k (n+1/2)}{N}\right)}\,\right],$$

    donde ahora, $N$ es el número de puntos de nuestra función inicial, sin simetría par,  que se están transformando.


- La razón por la que se usa más frecuentemente la DCT que la DFT para funciones no periódicas es que mientras que en la DFT, al repetir periódicamente la función, estamos creando discontinuidades en los extremos del intervalo, la transformada coseno, al simetrizar la función, nos da una función continua.


- Esas discontinuidades en los extremos de la DFT para funciones no periódicas puede generar problemas, efecto  Gibbs, mientras que tal problema no aparece en la DCT.


- La transformación discreta de seno se obtiene de la misma forma, aunque como ya hemos comentado se aplica de forma menos frecuenta pues requiere que la función se anule en los extremos. 


- Sin embargo, hay problemas físicos en los que es interesante, por ejemplo para una cuerda vibrante, funciones de onda en un pozo de potencial infinito, etc.


## 6.3.2 Aplicaciones de la transformada del coseno

- La DCT tiene usos tecnológicos muy importantes. Por ejemplo es la base del funcionamiento de los archivos comprimidos JPEG, MPEG o MP3.


- Digitalmente una imagen no es más que un conjunto de píxeles (un cuadrado de la pantalla con un color asignado).  
  Un ordenador lo único que necesita es guardar el número que describe (array) el color de cada uno de ellos.   
  Cómo el número de píxeles es muy grande, los archivos que almacenan esa información son muy grandes.
  
  
- El formato JPEG aglutina primero conjuntos de píxeles en bloques y sobre cada uno de ellos aplica una DCT de tipo II.   
  Con ello obtiene todos los coeficientes $a_k$ asociados a la imagen, pero sólo almacena los más grandes.    
  Esto permite disminuir el tamaño del fichero de almacenamiento. 


- El formato MPEG hace lo mismo, pero sobre cada una de las imágenes que genera un vídeo. 


- El formato de audio MP3 funciona de forma parecida. Digitalmente una señal de sonido no es más que un conjunto de ondas espaciadas en el tiempo.  
  El formato MP3 aplica a cada una de esas muestras una DCT y sólo guarda aquellas frecuencias que son perceptibles para el hombre.

# 6.4 La transformada de Fourier rápida

- La transformada de Fourier discreta de una función real

    $$c_k=\sum_{n=0}^{N-1}y_n\exp{\left(-\frac{2\pi k n}{N}\right)},$$

    involucra $\#N\cdot(\frac{1}{2}N+1)\sim\frac{1}{2}N^2$ operaciones.


- En el Tema 2 ya vimos que el número de operaciones que se pueden hacer en un ordenador es de aproximadamente $10^9$.


- Eso implica que el tamaño máximo de una muestra a la que se le puede aplicar la DFT es de:

```python
from math import sqrt

n_oper=1e9
N=sqrt(2*n_oper)
print("%0.f" %N)
```

- Este es aproximadamente el número de puntos que se necesita para representar un segundo de una señal musical.


- Por tanto, a la hora de hacer práctico el análisis de Fourier es necesario encontrar otro algoritmo para calcular sus coeficientes.


- El algoritmo usado fue desarrollado C. J. Gauss en 1805 y redescubierto por Cooley y Tukey en 1965.   
  Es conocido como la transformada de Fourier rápida (FFT).
  

- Para explicarlo:

    1. Asumamos que $N=2^m$ (de forma general no es necesario este requisito pero simplifica su descripción).
    2. Dividamos nuestro muestra inicial de $N$ puntos en 2 submuestras con valores alternos de la función,  
       es decir, una con los valores pares y otra con los impares de la muestra.  
       Para la submuestra par tenemos:
       
    $$c^\text{par}_k=\sum_{r=0}^{N/2-1}{y_{2r}\exp\left(-i\frac{2\pi k (2r)}{N}\right)}=\sum_{r=0}^{N/2-1}{y_{2r}\exp\left(-i\frac{2\pi k r}{N/2}\right)},$$

        mientras que para la impar el resultado es:
       
    $$\hat c^\text{impar}_k=\sum_{r=0}^{N/2-1}{y_{2r+1}\exp\left(-i\frac{2\pi k (2r+1)}{N}\right)}=\exp\left(-i\frac{2\pi k}{N}\right)\sum_{r=0}^{N/2-1}{y_{2r+1}\exp\left(-i\frac{2\pi k r}{N/2}\right)}=\text{e}^{\left(-i\,2\pi k/N\right)}c^\text{impar}_k.$$

       Cada una de ellas no es más que la transformada de Fourier de una muestra con $N/2$ puntos, y la transformada de Fourier completa es
  
    $$c_k=c^\text{par}_k+\text{e}^{\left(-i\,2\pi k/N\right)}c^\text{impar}_k.$$
<br>

    3. Repetimos iterativa el proceso para cada uno de nuestras submuestras, dividiéndolas en otras dos.   
    De forma que después de $k$ pasos tendremos $2^k$ submuestras con $N/2^k$ elementos cada una.  
    Al llegar al paso $m$, con $m$ tal que $N=2^m$, tendremos $N$ submuestras de un elemento, cuya DFT es:

    $$c_0=\sum_{n=0}^0y_n\text{e}^0=y_0,$$

    es decir, coincide con los elementos de la muestra. 


- La transformada de Fourier rápida se calcula reconstruyendo la DFT a través de los elementos de la muestra.   
  Es decir, aplicando el proceso inverso al anterior. 

    1. Dados los elementos de la muestra, estos se combinan en pares siguiendo la ecuación de arriba. 
    <br>

    2. Estas combinaciones de pares se combinan en conjuntos sucesivos de 4 elementos.   
       Estos en conjuntos de 8 elementos y así sucesivamente hasta que hemos reconstruido el conjunto total.
        <br>
    
    3. Lo relevante es calcular el número de operaciones que hemos necesitado para recuperar la DFT completa.  
       En primer lugar tenemos $N$ muestras de 1 elemento.         
       A continuación, tenemos $N/2$ muestras cada una con 2 coeficientes, lo que requiere otra vez $N$ operaciones. 
       En la tercera ronda, tenemos $N/4$ muestras formadas con 4 elementos cada una, que implica de nuevo $N$ coeficientes. 
       Finalmente en la ronda final $m$, tenemos $N/2^m=1$ muestra con $2^m$ elementos.  
       Por tanto, en cada ronda siempre tenemos que realizar $N$ operaciones.  
       Falta calcular cuantos rodas necesitamos:
       
    $$m=\log_2 N,$$


- Que por tanto implica que el número total de operaciones que hemos necesitado para recuperar la DFT es de:
       
    $$\#(\text{operaciones})= N\cdot m=N\cdot\log_2N.$$


- Para $10^9$ operaciones el tamaño de la muestra es:

```python
from math import log
from scipy.optimize import fsolve

N_oper=[1e9]

def f(x):
    return x*log(x,2)-N_oper


print("%0.1e" %fsolve(f,1e6))
```

- Que implica que ahora podemos trabajar con muestras de hasta 40 millones de elementos.

## 6.4.1  Formulas para la transformada de Fourier rápida

- Como ya hemos visto, en el paso inicial tenemos la DFT completa que queremos recuperar, una muestra con $N$ elementos. 


- Para el paso siguiente, tendremos 2 muestras con $N/2$ elementos cada una.


- Para un paso $n$ tendremos $2^n$ conjuntos con $N/2^n$ elementos cada uno.
<br/>

    - El primero de estos conjuntos constará de los elementos de la muestra inicial: $\{0,2^n, 2\cdot 2^n,3\cdot 2^n,\cdots,N-2^n\}$.  
      Es decir por los elementos $2^n\cdot r$, con $r=0,\cdots,N/2^n-1$.
    <br>
    
    - De la misma forma, la segunda muestra contará con los elementos: $2^n\cdot r+1$, la tercera con $2^n\cdot r+2$, etc... 
        <br>
    
    - Por tanto, la DFT de la muestra $j+1$ del paso $n$ será:
        <br>
    
    $$\sum_{r=0}^{N/2^n-1}y_{2^n r+j}\exp{\left(-i\frac{2\pi k(2^n r +j)}{N}\right)}=\text{e}^{-i2\pi kj\,/N}\sum_{r=0}^{N/2^n-1}y_{2^n r+j}\exp{\left(-i\frac{2\pi k r}{N/2^n}\right)}=\text{e}^{-i2\pi kj\,/N}c_k^{\,(n,j)}.$$

     con $k,j=0,\cdots,2^n-1.$ 
 
 
- Pero ya hemos visto que esa DFT se puede descomponer en 2 submuestras con la mitad de elementos de forma que:
   
    $$c_k^{(n,j)}=c_k^{(n+1,\,j)}+\text{e}^{-i2\pi k 2^n\,/N}c_k^{(n+1,\,j+2^n)},$$


- Con esta formula el cálculo de la transformada de Fourier rápida (FFT) es fácil:


- Empezamos con el último paso $m$ con $N=2^m$, en ese caso, sólo hay un coeficiente de Fourier para nuestras $N$ muestras
   
    $$c_0^{(m,j)}=y_j, \quad\text{con}\quad j=0,\cdots, N-1,$$


-  Usamos estos valores para construir los coeficientes de Fourier de la fase anterior, hasta llegar a $m=0$ en donde:
    
    $$c_{k}^{0,0}=\sum_{r=0}^{N-1}y_n\exp{\left(-i\frac{2\pi k r}{N}\right)}=c_k,$$   

    tenemos los coeficientes que estábamos buscando.

## 6.4.2 La transformada de Fourier rápida en numpy

- Cómo no podía ser de otra forma ya hay un módulo en numpy en el que la transformada de Fourier rápida esta incorporada. Su implementación es además mucho más eficiente de la que hemos programado nosotros. 


- Es el módulo fft. 


- Este módulo incorpora las funciones rfft y irfft para la transformada de Fourier rápida de una muestra real y su inversa, respectivamente.


- También incorpora las funciones fft y ifft, en este caso para una muestra compleja. 


- Para muestras de dos dimensiones incorpora las funciones rfft2, irfft2 y fft2, ifft2. 


- Para funciones de más de dos dimensiones la función se llama rfftn, irfftn, y fftn, ifftn.

```python
from numpy import array, arange
from numpy.fft import rfft

y = arange(0,1,0.1)
c = rfft(y)
print(c)
```

```python
from numpy.fft import irfft
from numpy import set_printoptions

set_printoptions(precision=2)
set_printoptions(suppress=True)

z = irfft(c)
print(z)
```

$$f(t)=\left\{\begin{array}{rl}1&\text{if}\quad\lfloor 2t\rfloor\quad\text{es par}\\
                              -1&\text{if}\quad\lfloor 2t\rfloor\quad\text{es impar}\\\end{array}\right.$$
                              
donde $\lfloor x\rfloor$ representa el mayor valor entero que no excede de x. 

Vamos a intentar suavizar esta función por medio de la transformada de Fourier.

## 6.4.3 Transformada Coseno y Seno Rápidas

- La transformada del coseno y seno se puede obtener trivialmente usando la FFT pues no son  mas que una transformada de Fourier con una muestra simétrica o antisimétrica, respectivamente.


- Método para calcular una FCT y una FST:

   - Dada una muestra de N puntos, creamos una muestra con el doble de puntos por simetría especular par o impar dependiendo de si queremos una transformada seno o coseno.
   
   - Se realiza una FFT sobre la muestra obtenida.
   
   - Descartamos la parte imaginaria o real de los coeficientes para la transformada coseno y seno, respectivamente, que obviamente deberían ser 0.

## Ejercicio 6.2: la transformada de Fourier corta (SFT)

Cuando tenemos una señal con cierta periodicidad, la transformada de Fourier nos permite obtener la información de los datos, discernir señales del ruido, o incluso interpolar resultados. Sin embargo, en otros ámbitos, como es el de la industria musical, el cine, la radio, etc. las señales no se repiten indefinidamente en el tiempo (imaginar una canción, donde las notas van cambiando a lo largo del tiempo). ## Ejercicio 6.5: deconvolución de imágenes

Estamos habituados a ver en películas o series de televisión de temática policíaca escenas en las que un técnico transforma una imagen borrosa en una imagen nítida casi por arte de magia. 
Pero de hecho, no hay ninguna magia, no es más que análisis de Fourier.  

En una imagen borrosa cada punto queda difuminado por medio de una distribución que llamamos función de dispersión de punto.  
Para simplificar el problema asumamos que estamos trabajando con una imagen en blanco y negro, por lo que la imagen puede ser representada por una sóla función $a(x,y)$ que determina el brillo de un punto $(x,y)$ de la imagen.

Definamos la función de dispersión de punto como $f(x,y)$, que hace que dado un punto $(x,y)$ en la imagen inicial, su brillo (valor) se traslade al punto $f(x,y)$. Sí $f(x,y)$ es una función ancha, la imagen se emborronara mucho. Sin embargo, si la función esta descrita por un pico muy estrecho entonces la imagen puede ser aceptáblemente nítida.

En general, el brillo de la imagen borrosa $b(x,y)$ puede describirse por una función de convolución, que describe como se ha desplazado el  brillo de la imagen original:

$$b(x,y)=\int_{0}^K\int_0^L a(x',y')f(x-x',y-y')\,\text{d}x'\text{d}y',$$

donde $K$ y $L$ es la dimensión de la imagen en cada uno de sus ejes. Esta ecuación es la llamada convolución de la imagen a través de la función de dispersión de punto.

Para explicar como funciona la técnica de deconvolución (recuperación de una imagen borrosa) trabajemos primero en una dimensión (la generalización a más dimensiones es trivial). La convolución de una imagen en una dimensión viene dada por:

$$b(x)=\int_0^L a(x')f(x-x')\,\text{d}x'.$$

Esta función se puede representar por medio de una serie de Fourier de forma que:

$$b(x)=\sum_{k=-\infty}^\infty{\bar b_k\exp\left(i\frac{2\pi k x}{L}\right)},$$

donde los coeficientes de Fourier vendrán dado por:

\begin{align}
\bar b_k=&\frac{1}{L}\int_0^L{b(x)\exp{\left(-i\frac{2\pi k x}{L}\right)}\text{d}x}\nonumber\\
=&\frac{1}{L}\int_0^L\int_0^La(x')f(x-x')\,\exp{\left(-i\frac{2\pi k x}{L}\right)}\,\text{d}x\,\text{d}x'\nonumber\\
=& \frac{1}{L}\int_0^La(x')\,\exp{\left(-i\frac{2\pi k x'}{L}\right)}\int_{-x'}^{L-x'}f(x'')\,\exp{\left(-i\frac{2\pi k x''}{L}\right)}\,\text{d}x\,\text{d}x'',
\end{align}

donde en el último paso hemos llamado $x''=x-x'$.

Si ahora hacemos $f(x)$ una función periódica repitiéndola infinitamente hacía la izquierda y derecha de su intervalo, la segunda integral

\begin{align}
\int_{-x'}^{L-x'}f(x'')\,\exp{\left(-i\frac{2\pi k x''}{L}\right)}\,\text{d}x''=&\int_{-x'}^{0}f(x'')\,\exp{\left(-i\frac{2\pi k x''}{L}\right)}\,\text{d}x''+\int_{0}^{L-x'}f(x'')\,\exp{\left(-i\frac{2\pi k x''}{L}\right)}\,\text{d}x''\nonumber\\
=&\int_{L-x'}^{L}f(x'')\,\exp{\left(-i\frac{2\pi k x''}{L}\right)}\,\text{d}x''+\int_{0}^{L-x'}f(x'')\,\exp{\left(-i\frac{2\pi k x''}{L}\right)}\,\text{d}x''\nonumber\\
=&\int_{0}^{L}f(x'')\,\exp{\left(-i\frac{2\pi k x''}{L}\right)}\,\text{d}x''=L\,\bar f_k,
\end{align}

no es más que la transformada de Fourier $\bar f_k$ de $f(x)$ multiplicada por $L$.

Por tanto, 
$$\bar b_k=\bar f_k\int_0^La(x')\,\exp{\left(-i\frac{2\pi k x'}{L}\right)\,\text{d}x'}=L\,\bar f_k\bar a_k,$$

la transformada de Fourier de la imagen borrosa viene dada por el producto de la transformada de Fourier de la imagen original y de la función de dispersión de punto.

Para recuperar la imagen original por tanto sólo tenemos que coger la imagen borrosa y hacer su transformada de Fourier. Hacer lo mismo con la función de dispersión de punto y tomar su cociente dividido por el tamaño del intervalo, de forma que obtenemos

$$\bar a_k=\frac{\bar b_k}{L \bar f_k}.$$

Finalmente sólo tenemos que hacer la transformada de Fourier inversa para recuperar la imagen nítida. 
Este es llamado proceso de deconvolución.

El proceso para una imagen de dos dimensiones es completamente equivalente. Para una imagen de dimensiones $L\times K$ se obtiene

$$\bar b_{kl}=KL\bar a_{kl}\bar f_{kl}.$$


La única complicación asociada al proceso de deconvolución de una imagen es que a priori no sabemos cual es la función de dispersión del punto, por lo que uno tiene que probar con varias alternativas hasta que obtiene un resultado satisfactorio.

La mayoría de las cámaras, sin embargo, involucran una función de dispersión del punto que puede ser descrita de forma bastante satisfactoria como una función gaussiana:

$$f(x,y)=\exp\left(-\frac{x^2+y^2}{2\sigma^2}\right),$$

donde el valor de la desviación típica $\sigma$ es desconocido.
Este es por ejemplo el procedimiento que se usó para obtener la primera imagen nítida de un agujero negro.
