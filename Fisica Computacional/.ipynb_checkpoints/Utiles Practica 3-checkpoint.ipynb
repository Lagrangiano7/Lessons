# EDOs

## 1. Euler

$$x(t+h)=x(t) + hf(x,t)$$

```python
import numpy as np
import matplotlib.pyplot as plt
################
t0=0
t_max=10
N=1000
tp = np.linspace(t0, t_max, N)
h = t_max/N

x=0
f = lambda x, t: -x**3+np.sin(t)
################

xp=[]

for t in tp:
    xp.append(x)
    x+= h*f(x, t)

plt.plot(tp, xp)
plt.show()
```

## 2. RK2 / Punto medio

$$k_1=hf(x,t)$$
$$k_2=hf(x+k_1/2, t+h/2)$$
$$x(t+h)=x(t)+k_2$$

```python
import numpy as np
import matplotlib.pyplot as plt
################
t0=0
t_max=10
N=100
tpp = np.linspace(t0, t_max, N)
h = t_max/N

x=0
f = lambda x, t: -x**3+np.sin(t)
################

xpp = []

for t in tpp:
    xpp.append(x)
    x_med = x + h/2*f(x, t)
    x += h*f(x_med, t+h/2)
    
plt.plot(tpp, xpp)
plt.show()
```

## 3. RK4

$$k_1=hf(x,t)$$
$$k_2=hf(x+k_1/2,t+h/2)$$
$$k_3=hf(x+k_2/2,t+h/2)$$
$$k_4=hf(x+k_3,t+h)$$

$$x(t+h)=x(t)+\frac{1}{6}(k_1+2k_2+2k_3+k_4)$$

```python
import numpy as np
import matplotlib.pyplot as plt
################
t0=0
t_max=10
N=100
tp = np.linspace(t0, t_max, N)
h = t_max/N

x=0
f = lambda x, t: -x**3+np.sin(t)
################

xp = []

for t in tp:
    xp.append(x)
    k1 = h*f(x, t)
    k2 = h*f(x+k1/2, t+h/2)
    k3 = h*f(x+k2/2, t+h/2)
    k4 = h*f(x+k3, t+h)
    
    x += 1/6*(k1+2*k2+2*k3+k4)
    
plt.plot(tp, xp)
plt.show()
```

## 3.1 RK4 con paso adaptado

$$h_{opt}=h\rho^{1/4}=h\left(\frac{h\delta}{\epsilon}\right)^{1/4}$$

- Caso 1: $\rho < 1$: mal, regresamos al punto de partida con $h_{opt}$
- Caso 2: $\rho>1$: bien, pero estoy usando un paso demasiado pequeño $\longrightarrow$ el siguiente lo tomo con $h_{obj}$

**OJO!:** $\rho$ puede ser muy grande y dar overflow cuando el punto calculado con 1 paso de $2h$ y el calculado con dos pasos de $h$ queden demasiado cercanos entre sí $\longrightarrow$ tomo $h=min\{h_{opt}, 2h\}$

**Modificación: Extrapolación local:** $x(t+2h)=x_1+\frac{1}{15}(x_1-x_2)$
Mejora el error en un orden de magnitud, a $h^6$

```python
import numpy as np
import matplotlib.pyplot as plt
################
t=0
t_max=10
N=100
h = (t_max-t)/N

x=0
f = lambda x, t: -x**3+np.sin(t)
################
tp=[0]
xp=[0]

delta = 1e-6

while t < t_max:
    # Primero tomo paso con longitud 2h
    k1 = 2*h*f(x, t)
    k2 = 2*h*f(x+k1/2, t+h)
    k3 = 2*h*f(x+k2/2, t+h)
    k4 = 2*h*f(x+k3, t+2*h)
    
    x2h = x + 1/6*(k1+2*k2+2*k3+k4)
    
    # Ahora tomo 2 pasos con longitud h
    k1 = h*f(x, t)
    k2 = h*f(x+k1/2, t+h/2)
    k3 = h*f(x+k2/2, t+h/2)
    k4 = h*f(x+k3, t+h)
    
    x1h = x + 1/6*(k1+2*k2+2*k3+k4)
    
    k1 = h*f(x1h, t+h)
    k2 = h*f(x1h+k1/2, t+3*h/2)
    k3 = h*f(x1h+k2/2, t+3*h/2)
    k4 = h*f(x1h+k3, t+2*h)
    
    x11h = x1h + 1/6*(k1+2*k2+2*k3+k4)
    
    # Ahora calculo si h es o no válido
    p = 30*h*delta/np.abs(x11h-x2h)
    h_opt = h*p**(1/4)
    
    if p > 1: # Me vale pero tengo que tomar h_opt para la próxima
        x=x11h + 1/15*(x11h-x2h)
        xp.append(x)
        t+=2*h
        tp.append(t)
        if p==np.inf:
            h=2*h
        else:
            h = min(2*h, h_opt)
    else:
        h=h_opt

plt.plot(tp, xp)
plt.show()
```

## 4. Salto de la rana $O(h^2)$

$$x_{med}(t+h/2)=x(t)+\frac{h}{2}f(x,t)$$
$$x(t+h)=x(t)+hf(x_{med}, t+h/2)$$
$$x_{med}(t+3h/2)=x_{med}(t+h/2)+hf(x, t+h)$$

```python
import numpy as np
import matplotlib.pyplot as plt
################
t0=0
t_max=10
N = 1000
tp = np.linspace(t0,t_max, N)
h = t_max/N

x=0
f = lambda x, t: -x**3+np.sin(t)
################

xp=[]

x_med = x+h/2*f(x, t0)

for t in tp:
    xp.append(x)
    x += h*f(x_med, t+h/2)
    x_med+=h*f(x, t+h)



plt.plot(tp, xp)
plt.show()
```

## 6. Bulirsch-Stoer

```python
import numpy as np
import matplotlib.pyplot as plt


d = 1e-7
t0=0
t_max = 10
N=100
H = t_max / N # Particiono el intervalo [t0, t+h] en 100
tol = H*d # quiero que para cuando llegue al final la sol. se haya errado por un máx. de 1e-6


tp = np.arange(t0, t_max, H) # Construyo el intervalo

f = lambda x, t: -x**3+np.sin(t)
x=0
xp = []


for t in tp:
    xp.append(x)
    
    # 1. R11 con punto medio modificado, 1 solo paso
    n = 1
    x_med = x+H/2*f(x, t)
    x2 = x + H*f(x_med, t+H/2)

    R1 = np.empty(n, float)
    R1[0] = 1/2*(x_med+x2+H/2*f(x2, t+H))
    
    # 2. Ahora vamos incrementando el número de pasos en 1 cada vez hasta que lleguemos a tol.
    err = 2*H*d # cogemos este error para asegurarnos de que el while se ejecuta al menos 1 vez, porque tol es precisamente H*d
    while err > tol:
        n+=1
        h=H/n
        x_med = x+h/2*f(x, t)
        x2 = x + h*f(x_med, t+h/2)
        
        for i in range(n-1):
            x_med+=h*f(x2, t+h*(i+1))
            x2+=h*f(x_med, t+(i+3/2)*h)
            
        # 3. Extrapolación de Richardson hasta completar la fila
        R2 = R1
        
        R1 = np.empty(n, float)
        R1[0] = 1/2*(x_med+x2+h/2*f(x2, t+H))
        for m in range(1, n):
            eps = (R1[m-1]-R2[m-1])/((n/(n-1))**(2*m)-1)
            R1[m] = R1[m-1]+eps
        err = np.abs(eps)

    x = R1[n-1]
    
plt.plot(tp, xp)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

g=9.8
l=0.1

f_theta = lambda v: v
f_v = lambda theta: -g/l*np.sin(theta)


r = np.array([179*np.pi/180, 0], float)
thetap = []

d = 10**(-8)
t_0=0
t_max=10
N=110

H = t_max/N
tol = H*d
tp = np.arange(0,10, H)

def F(r):
    theta, v = r
    return np.array([f_theta(v), f_v(theta)], float)


for t in tp:
    thetap.append(r[0])
    
    # 1. R11 con punto medio modificado
    n = 1
    r_med = r+H/2*F(r)
    r2 = r + H*F(r_med)

    R1 = np.empty([1, 2], float)
    R1[0] = 1/2*(r_med+r2+H/2*F(r2))
    
    # 2. Ahora vamos incrementando el número de pasos en 1 cada vez hasta que lleguemos a tol.
    err = 2*H*d # cogemos este error para asegurarnos de que el while se ejecuta al menos 1 vez, porque tol es precisamente H*d
    while err > tol:
        n+=1
        h=H/n
        r_med = r+h/2*F(r)
        r2 = r + h*F(r_med)
        
        for i in range(n-1):
            r_med+=h*F(r2)
            r2+=h*F(r_med)
            
        # 3. Extrapolación de Richardson hasta completar la fila
        R2 = R1
        
        R1 = np.empty([n,2], float)
        R1[0] = 1/2*(r_med+r2+h/2*F(r2))
        for m in range(1, n):
            eps = (R1[m-1]-R2[m-1])/((n/(n-1))**(2*m)-1)
            R1[m] = R1[m-1]+eps
        err = np.abs(eps[0])

    r = R1[n-1]
    
plt.plot(tp, thetap)
plt.show()
```

## 1. Método del disparo

Busco aquella condición inicial que, al resolver la EDO, esta cumpla la cond. de contorno

```python
def biseccion(f, a, b, tol):
    err = 1.0
    
    while err > tol:
        x_med = (a+b)/2
        if f(a)*f(x_med) > 0: # Seguimos estando en la parte de a
            a = x_med
        else:
            b = x_med
        err = np.abs(b-a)
        
    return (a+b)/2
```

```python
import numpy as np

g = 9.8

v0=0 # Empezamos por el valor más bajo posible para asegurar que no nos pasamos y garantizar convergencia

x0=0
t1 = 10 # s
h = t1/100

tp = np.linspace(0,t1,100)
########### SECCIÓN: RESOL. DE LA EDO CON CONDS. INICIALES DADAS ###########

def F(r):
    x, v = r
    return np.array([v, -g])

###### COND: x(10s)=0

def x_t1(v0): # Resuelvo hasta t1. Por ejemplo, uso RK4
    r = np.array([x0, v0], float) # Cond. inicial
    
    for t in tp:
        k1 = h*F(r)
        k2 = h*F(r+k1/2)
        k3 = h*F(r+k2/2)
        k4 = h*F(r+k3)
        
        r += 1/6*(k1+2*k2+2*k3+k4)
    return r[0] # x(t_1) para cond. inicial v0

# Ahora quiero saber para qué cond. inicial v0 se cumple x(10s)=0 --> Anulo x_t1(v0)

v_a = 45
v_b = 55

v0 = biseccion(x_t1, v_a, v_b, 1e-6)

print(v0)
```

## 1.1 Problemas de autovalores

No podemos aplicar el método del disparo a la velocidad inicial porque la ec. es autónoma por definición de ec. de autovalores ==> ahora nos interesa hallar los autovalores que permitan el cumplimiento de las conds. de contorno. En este sentido, podemos ver las ecuaciones como una función de un parámetro, tal que si ese parámetro es autovalor, entonces se cumple la cond. de contorno que queremos

Un ejemplo es la ec. de Schrödinger independiente del tiempo:

$$\frac{-\hbar^2}{2m}\frac{d^2\Psi}{dx^2}+V(x)\Psi=E\Psi$$

Podemos ahora hacer un procedimiento del todo análogo al método del disparo, solo que en lugar de aplicarlo sobre la velocidad inicial, lo hacemos ahora sobre la energía, buscando que, cuando la energía que hayamos dado sea correcta (autovalor) entonces se cumpla la cond. de contorno

```python
# Pozo infinito de potencial, para adaptarlo a otro potencial basta con cambiar V(x) y ajustar conds. de contorno

import numpy as np
import matplotlib.pyplot as plt

L = 5.292e-11
hPlanck= 6.63e-34
hbar = hPlanck/(2*np.pi)
m = 9.1e-31
xp = np.linspace(0,L,1000)
h = L/1000

def V(x):
    return 0

def F(r, x, E):
    Psi, der = r
    return np.array([der, (V(x)-E)*2*m/hbar**2*Psi], float)

def Psi_L(E):
    r = np.array([0,1], float) # da igual qué valor le dé a la derivada porque ec. es autónoma
    for x in xp: # Resuelvo con RK4
        k1 = h*F(r, x, E)
        k2 = h*F(r+k1/2, x+h/2, E)
        k3 = h*F(r+k2/2, x+h/2, E)
        k4 = h*F(r+k3, x+h, E)
        
        r+= 1/6*(k1+2*k2+2*k3+k4)
    return r[0]

# Ahora encuentro E tal que me anule Psi_L

Psi_L = np.vectorize(Psi_L)
E = np.linspace(0,500*1.6e-19, 100)
plt.plot(E, Psi_L(E))


E0 = biseccion(Psi_L, 0, 500*1.6e-19, 1e-20)
plt.plot(E0, Psi_L(E0), "ro")
plt.grid()
plt.show()
```

## 2. Relajación (por diferencias finitas)

Ahora en lugar de resolver la EDO y ver si se cumplen o no las condiciones de contorno, hacemos lo contrario: imponemos que se cumplan las conds. de contorno desde el inicio y resolvemos la EDO en concordancia

$$\frac{d^2x}{dt^2}= f(x,t)$$

$$\frac{x_{i+1}-2x_i+x_{i-1}}{h^2}= f(x,t)$$

$$\boxed{x_i = \frac{1}{2}(x_{i+1}+x_{i-1}+h^2f(x_i, t_i))}$$

**Si EDO es de orden 1**

$$\frac{dx}{dt}=f(x,t)$$

- Por diferencias adelantadas:
$$\frac{x_{i+1}-x_i}{h}=f(x_i, t_i) \Longrightarrow x_i = x_{i+1}-hf(x_i, t_i)$$

- Por diferencias atrasadas:

$$\frac{x_i-x_{i-1}}{h}= f(x_i, t_i) \Longrightarrow x_i = x_{i-1}+hf(x_i, t_i)$$

- Tomando la media aritmética de ambas:

$$\boxed{x_i=\frac{1}{2}(x_{i+1}-x_{i-1})}$$

**MÉTODO DE JACOBI EN 1 DIMENSIÓN?**


```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, t):
    return -9.8

t0=0
t_max=10

N=100
tp = np.linspace(t0, t_max, N+1)
h = t_max/N
xp = np.zeros_like(tp) # Automáticamente tengo los extremos cumpliéndome las conds. de contorno, basta con no modificarlos

tol=1e-6
err = 1.0

while err > tol:
    x0=xp.copy()
    xp[1:N] = 1/2*(xp[2:N+1]+xp[0:N-1]-h**2*f(xp[1:N], tp[1:N]))
    err = np.max(np.abs(xp-x0))
    
plt.plot(tp, xp)
plt.show()
```

## 2.1 Diferencias finitas con SOR

$$x_i^{(n+1)}=\frac{1+\omega}{2}\left(x_{i+1}^{(n)}+x_{i-1}^{(n)}-h^2f(x_i^{(n)},t_i)\right)-\omega x_i^{(n)}; \ \ \omega\in (-1,1)$$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, t):
    return -9.8

t0=0
t_max=10

N=100
tp = np.linspace(t0, t_max, N+1)
h = t_max/N
xp = np.zeros_like(tp) # Automáticamente tengo los extremos cumpliéndome las conds. de contorno, basta con no modificarlos

tol=1e-6
err = 1.0
w = 0

while err > tol:
    x0=xp.copy()
    xp[1:N] = (1+w)/2*(xp[2:N+1]+xp[0:N-1]-h**2*f(xp[1:N], tp[1:N]))-w*x0[1:N]
    err = np.max(np.abs(xp-x0))
    
plt.plot(tp, xp)
plt.show()
```

## Problemas de contorno

### 1. Ec. de Laplace (Jacobi)

 $$\phi'(x,y)=\frac{1}{4}\left(\phi(x+a,y)+\phi(x-a,y)+\phi(x,y+a)+\phi(x,y-a)\right)$$

```python
import numpy as np
import matplotlib.pyplot as plt

L=1
V=1
a=1/100
N=int(L/a)+1

Psi = np.zeros((N,N), float)
Psi[0,:]=V # conds. de contorno: V arriba y 0 en los otros 3 lados

err = 1.0
tol = 1e-6

while err > tol:
    Psi0 = Psi.copy()
    Psi[1:N-1,1:N-1] = 1/4*(Psi[2:N,1:N-1] + Psi[0:N-2,1:N-1] + Psi[1:N-1,2:N] + Psi[1:N-1,0:N-2])
    err = np.max(np.abs(Psi-Psi0))

plt.imshow(Psi, cmap="hot")
plt.colorbar()
plt.show()
```

**Si es ec. de Poisson**

$$\nabla^2\phi=f$$

 $$\phi'(x,y)=\frac{1}{4}\left(\phi(x+a,y)+\phi(x-a,y)+\phi(x,y+a)+\phi(x,y-a)-a^2f(x,y)\right)$$

**Y si son 3 dimensiones?**

 $$\phi'(x,y)=\frac{1}{6}\left(\phi(x+a,y)+\phi(x-a,y)+\phi(x,y+a)+\phi(x,y-a)-a^2f(x,y)\right)$$

### 2. Ec. de Laplace (Gauss-Seidel SOR)

$$\phi_\omega(x,y)\leftarrow\frac{1+\omega}{4}\left[\phi(x+a,y)+\phi(x-a,y)+\phi(x,y+a)+\phi(x,y-a)\right]-\omega\,\phi(x,y)$$

```python
import numpy as np
import matplotlib.pyplot as plt

L = 1
a=1/100
V = 1
N = int(L/a)

Psi = np.zeros((N+1, N+1), float)

Psi[0,:]=V

tol = 1e-6
err = 1.0

w = 0.94

while err > tol:
    err = 0
    for i in range(1, N):
        for j in range(1, N):
            Psi_p = Psi[i,j]
            Psi[i,j]=(1+w)/4*(Psi[i+1,j]+Psi[i-1,j]+Psi[i,j+1]+Psi[i,j-1]) - w*Psi_p
            err = max(err, np.abs(Psi_p-Psi[i,j])) # Con esto me aseguro de que el error máximo de los elementos
                                                   # sea menor que la tolerancia
    
plt.imshow(Psi, cmap="hot")
plt.colorbar()

plt.show()
    
```

## Problemas de condiciones iniciales

### 1. FTCS (ec. del calor)

$$\phi(x, t+h)=\phi(x,t)+h\frac{D}{a^2}\left[\phi(x+a,t)-2\phi(x,t)+\phi(x-a,t)\right]$$

```python
import numpy as np
import matplotlib.pyplot as plt

L=1/100
T0=20
T_f=0
T_c=50
D=4.25*1e-6
N=100
a = L/N

xp = np.linspace(0,L,N+1)

# Existe un gradiente de temperaturas en el fondo del recipiente (50 on tp, 20 middle, 0 bottom) que hace que
# el sistema evolucione hacia un estado estacionaro
t0=0
t_max=10 # s
h = 1e-5
tp = np.arange(t0,t_max+2*h,h)

# cond. inicial
Psi = np.ones(N+1)*20
Psi[0]=50
Psi[-1]=0

times = [0.01,0.1,0.4,1,10]

for t in tp:
    Psi[1:N] += h*D/a**2 * (Psi[2:N+1] -2*Psi[1:N] + Psi[0:N-1])
    for i in times:
        if np.abs(t-i)<h/100:
            plt.plot(xp, Psi, label=f"$t={i}$")
            plt.xlabel("x (m)")
            plt.ylabel("T (C)")

plt.legend()
plt.show()
```

### 2. Euler implícito

Consiste en invertir el tiempo en FTCS y sumar +h a todas las dependencias temporales (tomar nuevo origen de tiempos) de modo que no tengas que trabajar con dependencias temporales negativas

### 3. Crank-Nicolson

Media entre FTCS y Euler implícito y llevas todo con t+h a la izq. y todo con t a la dcha.

$$A\phi(t+h)=B\phi(t)$$

```python
import numpy as np
import matplotlib.pyplot as plt
from banded import banded

L=2.0e-9
m = 9.109*1e-31
k = 5*1e10
hbar = 6.63*1e-34/(2*np.pi)
x0=L/2
sigma=1e-10
k=5*1e10

# Parte espacial
N=999
xp = np.linspace(0,L,N+1)
a = L/N

# Parte temporal
h=1e-18
tp = np.arange(0,100*h,h)


a1 = 1+1j*h*hbar/(2*m*a**2)
a2 = -1j*h*hbar/(4*m*a**2)
b1 = 1-1j*h*hbar/(2*m*a**2)
b2 = 1j*h*hbar/(4*m*a**2)

A = np.empty((N+1,N+1), complex)

A[0,:] = a2
A[1,:] = a1
A[2,:] = a2

def solveEvol(t_max):
    t=0
    # Cond. inicial
    Psi = np.exp(-(xp-x0)**2/(2*sigma**2))*np.exp(1j*k*xp)
    Psi[0]=Psi[-1]=0
    while t < t_max:
        # Resuelvo el sistema Av(t+h)=Bv(t) para obtener evol. temporal
        b = Psi[1:N]*b1 + b2*Psi[2:N+1] + b2*Psi[0:N-1]
        Psi[1:N] = banded(A, b, 1, 1)
        t+=h
    return Psi
        
    
# for i in range(100,500,100):
#     plt.plot(xp,np.real(solveEvol(i*h)))

plt.plot(xp, np.real(solveEvol(400*h)))
plt.show()
```
