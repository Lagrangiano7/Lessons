# 1. Foundations

## Definitions

**Def. Topological space:** let $X$ be a set. For each point $x\in X$ we ask that it has a nonempty collection of subsets of $X$, called "neighbourhoods of $x$, satisfying 4 defining properties:

1. $x$ lies in each of its neighbourhoods
2. The intersection of two neighbourhoods is a neighbourhood
3. If $N$ is a neighbourhood of $x$ and $U\subset X$ containing $N$, then $U$ is also a neighbourhood of $x$
4. If $N$ is a neighbourhood of $x$ and $M:=\{z\in N|N\text{is neighbourhood of $z$}\}$ then $M$ is a neighbourhood of $x$ (the "interior")

**Def. Topology:** a "topology on a set $X$" is the assignment of a collection of neighbourhoods to each point $x\in X$

## Substructure

**Def. Subspace Topology:** let $Y\subset X$ with $X$ a topology. Then, we can define a topology on $Y$ and this is called a "subspace topology"

**Proposition:** since any surface embedded in $\mathbb R^n$ is a subset of $\mathbb R^n$ and this is a topology, these surfaces are therefore also topologies (subset topologies of $\mathbb R^n$)

## Mappings

- Now that we have defined a type of space, we need to be concerned with its mappings

**Def. Continuity:** let $X, Y$ be topological spaces. A map $f:X\to Y$ is said to be "continuous" if, for each $x\in X$ and each neighbourhood $N\subset Y$ of $f(x)$ the set $f^{-1}(N)$ is a neighbourhood of $X$

**Def. Homeomorphism:** a map $f:X\to Y$ is called a "homeomorphism" if it is (1) injective, (2) surjective, (3) continuous and (4) $f^{-1}$ is also continuous (continuous bijection)
