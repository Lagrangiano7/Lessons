# Magnetic monopoles in SU(N) compactified to $\mathbb R^3\times S^1$

**Mechanism:**

- Choose the radius of the compactified dimension ($L_4$) such that $NL_4<<\Lambda^{-1}$, where $\Lambda$ is the dynamical scale of the theory (obtained through dimensional transmutation). The theory is now weakly coupled
- We can now build the Polyakov loop around the compactified dimension: $P_4[a](\vec x)=\mathcal P e^{i\oint_4 a_4dx_4}$
- **Observation:** the 1-loop effective potential of $P_4$ SSBs $Z_N$ to a centre element of the gauge group. If we are to relate this theory to 4D YM, this cannot be, because we need an unbroken centre group in the confined phase
- To stabilize this, we introduce a modification term to the effective potential called **double-trace deformation**: $V_{eff}(P_4)+=\displaystyle\sum_{m=1}^{N-1}J_m|\text{Tr}P_4^m|^2$
- Since the Polyakov loop carries information about the gauge fields, we may as well use the values of said loop for gauge fixing. To this end, we choose the gauge such that the VEV of the Polyakov loop is given by the clock matrix: $P_4\propto C=\text{diag}(1, \omega, \ldots, \omega^{N-1})$ (roots of unity)
- Equivalently, this gauge choice forces all the eigenvalues of $a_4$ to be different. The non-Abelian (non-Cartan) components of $a_4$ do not commute with the VEV of $a_4$ and hence acquire a mass term. In the low-energy regime, we will only notice the massless modes: the Cartan componentes. Therefore, we have $SU(N)\to U(1)^{N-1}$, i.e. an effective low energy theory would only observe the Cartan subgroup / the Abelian degrees of freedom

**Writing the low-energy EFT**

- We now wish to study the low-energy regime. For low energies, the modes related to $L_4$ are too energetic to be excited, so the EFT is 3-dimensional. What is its Lagrangian? It certainly only has a $U(1)^{N-1}$ symmetry
- **Note:** 3D gauge fields are dual to scalar fields, so we will be using this picture instead. $f_{\mu\nu}=\partial_\mu a_\nu-\partial_\nu a_\mu$, and its Hodge dual is $\star f = \frac{1}{2}\varepsilon_{\mu\nu\rho}f_{\nu\rho}$, and the Bianchi identity / Maxwell's equation claims $d\star f=0$, so it is exact. Therefore, it can be expressed as an exterior derivative: $d\star f=0\Longrightarrow \star f = d\sigma$, so $f=\star d\sigma$. If we now have $N-1$ copies of EM, we do this procedure for each of them and assemble them in a tuple, getting $\vec f= \star d\vec \sigma$ (up to a normalization factor). Each of these scalar $\sigma$ fields is called a **dual photon**
- The sigma fields are periodic up to a fundamental weight, so the instanton splits into N-1 BPS monopoles (carrying root charge) and 1 Kaluza-Klein monopole (carrying affine root charge), arising from the compactified dimension. Each has 1/N topological charge
- The contribution of a monopole to the action will be instanton-like: $e^{-S}e^{i\vec \alpha\cdot \vec \sigma}e^{i\theta/N}$. The first term must be the classical Boltzmann suppression to the path integral, $e^{-S_I/N}$. The second term remembers its magnetic charge and the third its topological charge. Considering monopoles and anti-monopoles in the action, the imaginary exponentials collapse to a cosine and we get:

$$\mathcal L_{3D}=\frac{g^2}{16\pi^2L_4}|d\vec \sigma|^2-\displaystyle\sum_{n=1}^N2\xi_n\cos\left(\vec \alpha_n\cdot\vec \sigma+\frac{\theta}{N}\right)$$

- This interaction term between the dual photons gives rise to a mass term: $m_\sigma^2 \sim \xi_m$
- A Wilson loop imposes a nontrivial boundary condition on $\vec\sigma$; its minimal configuration is a kink/domain-wall sheet whose action is proportional to the enclosed area. Hence we have an area law

**IMPORTANT:** we have found a confining mechanism in a theory obtained from compactifying ($\mathbb R^4\to \mathbb R^3\times S^1$) and modifying (addition of double-trace modification) YM 4D. However, in order to check whether this mechanism is also the one in YM 4D, we would need to find an adiabatic continuation of this theory ($R, \{J_m\})\neq 0$, to the original YM 4D theory in which ($R, \{J_m\})=0$

**Dumb mechanism:**

- Compactify and modify YM 4D (to stabilize $Z_N$)
- Choose gauge such that $P_4$ is clock matrix $\to$ off-diagonal modes of $a_\mu$ acquire mass and diagonal ones remain massless
- At low energies:
    - Relevant gauge group is $U(1)^{N-1}$ (only Cartan modes are massless and therefore exciteable)
    - We have 3 dimensions (the compactified dimension has such a small radius that its modes are too energetic to be excited)
- Instantons split in N-1 BPS monopoles and 1 KK monopole
- Formulate EFT for the dual fields $\to$ monopoles generate interaction $\to$ dual photons are massive $\to$ area law
