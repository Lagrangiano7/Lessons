# Conservation laws

Every particle decays into lighter particles unless a specific conservation law prevents it from doing so.

What are the conserved quantities?

1. Charge
    - Even if the particle entering the Feynman diagram has a different charge from the one exiting it, charge is conserved because the mediator particle carries away the excess / defect of charge
2. Color
    - EM and Weak interactions do not see color, this is just a property of the strong force
    - Just as in charge conservation, even if the particle entering the Feynman diagram for strong interaction has a different color from the one exiting it, color is conserved because the gluon carries away the difference
    - **Note:** remember all particles are colorless, and so the conservation law is always "0 color in, 0 color out"
    - Since there are 3 colors of quarks, the gluons mediating the interaction of 2 quarks changing colors must have 2 colors to even out everything
3. Baryon number
    - If a quark goes in, a quark has to come out. Nonetheless, since quarks can only be in mesons and baryons, and since mesons have 1+(-1)=0 "quark number", we can have as many mesons as we like, there is no such thing as a "conserved meson number"
    - Therefore, it is more practical to talk about "conserved baryon number" --> If a baryon comes in, another one has to come out
4. Electron / muon / tau number
    - Leptons are only subject to EM and Weak interactions
    - **Weak interactions only mix leptons of the same generation**
    - This is why instead of a leptonic number conservation, as was proposed in the beginning of the study of the Weak force, there are actually 3 separate conserved numbers, one for each generation of leptons
5. Approximate conservation of flavour
    - Because weak forces are so weak, we say flavours are "approximately conserved" in weak interactions (this is what brought Gell-Mann to introduce the concept of strangeness to account for the fact that strange particles are always produced in pairs). Nonetheless, this is incorrect
    - In practice, however, it is correct (it applies) because the energies needed to break this conservation are far beyond our reach in the lab
    - However, in decays this strangeness conservation is not fulfilled
    - **Conclusion:** flavour is generally conserved (in our attainable energy range) for interactions but not for decays
    - **FOR STRONG INTERACTIONS, TOTAL FLAVOUR CHARGE IS INDEED CONSERVED (COUNT QUARKS AT BOTH SIDES)**

# OZI rule

"If a strong process' Feynman diagram can be sliced in 2 parts just cutting gluons, then the process is supressed" (less likely to happen)

**Explanation:** if we can cut a Feynman diagram in 2 parts and just cut the gluons, this means there is a part of the process in which we would have exclusively gluons. Therefore, since these gluons must give place to the corresponding particles when the process finishes, they must have a high energy. Nonetheless, high energy gluons (or those which are closer, same thing) have a weaker binding than those whose energy is lower (this is called "asymptotic freedom"). Therefore, since it is harder for gluons to bind when they have high energies, the process is less likely to happen

## Why does the $\psi$ meson live 1000 times longer than other mesons?

The $\psi$ meson can decay (a priori) into either 3 pions (with 3 virtual gluons mediating the process, so the OZI rule applies) or into 2 D mesons (like Ks but with charm instead of strange). Nonetheless, this second process is kinematically forbidden (the product is more energetic than the $\psi$ meson) and therefore the only way to go for $\psi$ to disintegrate is into 3 pions, which is suppressed. Therefore, this makes the $\psi$ meson much more stable than other mesons

# Cabibbo mixing

We know from conservation law 4 that weak interactions only mix leptons of the same generation. Does the same apply for quarks? That is to say, is flavour conserved in weak interactions? **No**

We can have transitions between quarks of different generations, unlike what happens with leptons. Of course, it is most likely that a quark decays into its counterpart of the same generation, but it is not impossible that it decays to some quark from another generation. This is portrayed by the Cabibbo-Kobayashi-Maswaka 3x3 matrix

$$\begin{pmatrix}d'\\s'\\b'\end{pmatrix} = \begin{pmatrix}V_{ud}&V_{us}&V_{ub}\\ V_{cd}&V_{cs}&V_{cb}\\ V_{td}&V_{ts}&V_{tb}\end{pmatrix}\begin{pmatrix}d\\s\\b\end{pmatrix}$$

**Interpretation:**

1. $|V_{ij}|^2$ is the probability of j-th flavour quark transitioning to the i-th flavour quark
2. The meaning of this equation is the weak force does not couple the pairs in generations: $\begin{pmatrix}u\\d\end{pmatrix}, \begin{pmatrix}c\\s\end{pmatrix}, \begin{pmatrix}t\\b\end{pmatrix}$ (in the sense that one of each pair shall always transition to the other of the same pair and viceversa) but rather it binds the pairs: $\begin{pmatrix}u\\d'\end{pmatrix}, \begin{pmatrix}c\\s'\end{pmatrix}, \begin{pmatrix}t\\b'\end{pmatrix}$ (in the sense that up can transition to any of down-like quarks with some probability)
3. The matrix is unitary, which means the election of down-like quarks instead of up-like quarks in the equation is arbitrary (i.e: it is equally likely for an up quark to transition to a down quark as the opposite to happen)

**Caveat:** although the CKM matrix implies this last point, the truth is in nature the probabilities of $u\to d$ and $d\to u$ are not the same, due to the fact that quarks are not on their own but rather conforming 2 and 3 quark bound states, so their behaviour is more complex. $d\to u$ occurs naturally (as in beta disintegration of a neutron) but the inverse, $u\to d$ does not occur naturally. Otherwise, protons could spontaneously become neutrons but this is not observed because the proton is more stable than the neutron on their own. See how this has to do with bound states?
