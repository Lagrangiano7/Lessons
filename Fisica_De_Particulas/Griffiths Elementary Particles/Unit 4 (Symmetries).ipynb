# 1. Isospin (flavour symmetry)

- Isospin is an approximate symmetry for up and down quarks, although it predates them
- Since the proton and the neutron (regarded as elementary particles) are so close in mass, they could be considered different excitations of the same particle
- Therefore, one can account for "how proton-like" or "how neutron-like" a particle is by introducing the concept of isospin $\vec I$, which behaves just like an angular momentum. We evidently choose the basis elements to be protons and neutrons:

$$|p\rangle = \begin{pmatrix}1\\0\end{pmatrix}; \ \ |n\rangle = \begin{pmatrix}0\\1\end{pmatrix}$$

- Therefore, any nucleon (a **single** particle of the nucleus) is, in general, a superposition of proton and neutron:

$$N = \alpha|p\rangle+\beta|n\rangle = \begin{pmatrix}\alpha\\ \beta\end{pmatrix}$$

- Isospin can be treated with the formalism of angular momentum in QM. Therefore, we can identify an isospin state by giving its modulus $I$ and its third component $I_3$. We assign $I_3=+1/2$ for the proton and $I_3=-1/2$ for the neutron. In the uncoupled basis, we have:

$$|p\rangle = |\frac{1}{2}, \frac{1}{2}\rangle; \ \ |n\rangle = |\frac{1}{2}, -\frac{1}{2}\rangle$$

**Strong interactions are invariant under rotations in isospin space and therefore (Noether's theorem) isospin is conserved in every strong interaction**

## 1.1 Extending isospin to other particles

- Isospin can be associated to more particles, such as mesons (since it is essentially linked to quarks, it is sound to assume it can be associated to any baryon / meson)

- **Observation:** the horizontal rows of Eightfold Ways have very similar masses but different charges. We can expand the isospin treatment. To each charge we assign a different value of $I_3$, and therefore the number of particles in a row determines the $I$, since $I=2I_3+1 \Longrightarrow N=2I+1$ (with $N$ number of particles in the row)

**Pions:** $I=1$

$$|\pi^+\rangle = |11\rangle; \ \ |\pi^0\rangle = |10\rangle; \ \ |\pi^-\rangle = |1-1\rangle $$

**Lambda:** $I=0$

$$|\Lambda\rangle = |00\rangle$$

**Delta's:** $I=3/2$

$$|\Delta^{++}\rangle=|\frac{3}{2}\frac{3}{2}\rangle; \ \ |\Delta^{+}\rangle=|\frac{3}{2}\frac{1}{2}\rangle; \ \ |\Delta^{0}\rangle=|\frac{3}{2}\frac{-1}{2}\rangle; \ \ |\Delta^{-}\rangle=|\frac{3}{2}\frac{-3}{2}\rangle$$

**u/d quarks** $I=1/2$

$$|u\rangle=|\frac{1}{2}\frac{1}{2}\rangle; \ \ |d\rangle=|\frac{1}{2}\frac{-1}{2}\rangle$$

**Rest of quarks:** $I=0$

**Note:** we assign the highest $I_3$ value to the member with highest charge

# 1.2 Gell-Mann-Nishijima equation

- For hadrons composed only of u/d/s quarks, we have:

$$Q=I_3+\frac{1}{2}(A+S)$$

where:

- $Q$ charge
- $A$ baryon number
- $S$ strangeness

- All this isospin conservation does not work so well when heavier hadrons are involved

# 1.3 The deuteron

- Since isospin behaves like an angular momentum, we can ask ourselves what happens when we have a system of a 2 nucleons. Of course, we shall work in the coupled basis. According to their isospin assignations, we have 4 possibilities:

**Isospin triplet (I=1)**

$$\begin{cases}pp=|11\rangle \\ \frac{1}{\sqrt{2}}(|pn\rangle+|np\rangle)=|10\rangle \\ nn = |1-1\rangle\end{cases}$$

**Isospin singlet (I=0)**

$$\frac{1}{\sqrt{2}}(|pn\rangle-|np\rangle)=|00\rangle$$

- Experimentally, we only observe the singlet, and we call it "deuteron". We know it is the singlet because, if it were the triplet, all three possibilities should occur (after all, they only differ by a rotation in isospin space)

- Isospin is also closely related with scattering amplitudes in nucleon-nucleon scattering, allowing us to determine ratios between scattering amplitudes by comparing the factors of superposition. The square of these amplitudes yields the ratios between cross sections

# 2. Parity

**Def.:** parity is the inversion of some magnitude with respect to some origin point. It typically acts on (pseudo)scalars and (pseudo)vectors, and it does so as follows:

- Scalar: $P(s)=s$
- Pseudoscalar: $P(s)=-s$
- Vector / "polar" vector: $P(\vec v)=-\vec v$
- Pseudovector: $P(\vec v)=\vec v$ (this happens, for instance, when $\vec v=\vec a\times \vec b$ with $\vec a, \vec b$ polar vectors)

- Parity is an operator with a spectrum of just 2 eigenvalues: $\pm 1$
- Parity is also its own inverse: $P^2=I$ (inverting twice yields the same result)
- The parity of a composite system is the product of the parities of the individual subsystems

## 2.1 Experiment by C.S. Wu (weak interaction violates parity-symmetry)

- Radioactive Cobalt60 nuclei aligned with spin up
- This isotope is radioactive (beta decay) and therefore emits electrons
- Most electrons were observed to be emitted antiparallel to the direction of spin

**Mirror image (applying parity to everything)**

- Since spin is a pseudovector, it doesn't change
- The direction of emission of beta particles should shift, and they should be emitted parallel to the direction of spin
- THIS WAS NOT OBSERVED

**Classical hypothesis:** nature is parity-symmetric. If a process is allowed by the laws of physics, then its mirror image is also allowed

**Wu:** nature is not parity-symmetric because weak interaction violates this symmetry. If it were parity-symmetric, then an equal number of electrons should be emitted in the direction of spin and its opposite, but nature **CHOOSES** to emit preferentially in the direction opposite to nuclear spin

## 2.2 Neutrinos are exclusively left-handed

**Def.:** the value $\frac{m_s}{s}$ of a particle is called its "helicity"

- For particles with $s=1/2$, helicity can take on the values $+1$ (right handed) and $-1$ (left handed)

**Helicity is not Lorentz-invariant:** let us take the third component in the direction of the particle's velocity. If we travel faster than the particle, we see the spin just the same (pseudovector) but the particle now seems to be traveling backwards with respect to us. Therefore, its helicity has switched signs because their velocity (with respect to us) has, too. The third component of spin is, according to us, antiparallel to their direction of motion

- What if we found a particle that cannot be subject to this argument? For instance, neutrinos. According to the Standard Model, they are massless and therefore travel at the speed of causality (the speed of light). Hence, the argument is not valid since we cannot catch up with a neutrino, let alone travel faster than it does
- **We can then define helicity unambigously. In a Lorentz invariant way.**

- Physicists assumed then half of the neutrinos would be ("objectively") right-handed and half of them would be left-handed, but this is far from the truth, because **parity violation is maximal**

$$\boxed{\textbf{ALL NEUTRINOS ARE LEFT-HANDED, and all antineutrinos are right-handed}}$$

- Of course, this argument also holds for light. Since it has spin $s=1$, we would expect photons to have 3 helicities (-1,0,1) but they do not seem to have 0. This is due to the fact that light is a purely transversal wave and therefore lacks the longitudinal mode which would account for 0 helicity.

## 2.3 Parity of fermions and bosons

- The parity of a fermion must be opposite to that of its antiparticle
- The parity of a boson must be equal to that of its antiparticle

**Convention:** quarks have positive parity. Antiquarks have negative parity.

- We now understand why, if all neutrinos are left-handed, automatically all of the antineutrinos are right-handed. NONETHELESS, if neutrinos are discovered to have mass, small as it may be, their helicities are no longer Lorentz-invariant and therefore the boxed remark would make no sense, as parity in neutrinos would not be objectively defined

- The parity of an excited state with orbital angular momentum $l$ is its parity in the ground state multiplied by a factor $(-1)^{l+1}$
- The prefix **pseudo** tells you the parity is negative
- Photons are vector particles and as such have -1 parity

# 3. Charge conjugation

**Def.:** the "charge conjugation" operator $C$ converts a particle into its antiparticle

$$C|p\rangle = |\bar p\rangle$$

**Properties:**

1. $C^2\equiv I$(the antiparticles of antiparticles are ordinary particles)
2. Its spectrum is $\pm 1$
3. Its eigenstates are those particles which are their own antiparticles (Majorana particles). These are:
    - Photon
    - Mesons in the center of the Eightfold Way (all those composed of a quark and its corresponding antiquark, i.e: $u\bar u=\pi^0$)
    
- From these properties I deduce only neutral particles are eligible to be Majorana particles (otherwise their charge would change)

## 3.1 Charge conjugation number

- It is the eigenvalue of the $C$ operator and therefore we can only assign charge conjugation number to the eigenstates of $C$

- **A system consisting of a s=1/2 particle and its corresponding antiparticle in a configuration with orbital angular momentum $l$ and total spin $s$ is an eigenstate of $C$ with eigenvalue $(-1)^{l+s}$}}**

- For instance, $\pi^0=u\bar u$ fulfils this theorem and it has $C=(-1)^{0+0}=+1$
- In general: every meson composed of a quark and its antiquark fulfill this theorem. In particular, pseudoscalar mesons (l=0,s=0) have $C=+1$ and vector mesons (l=0,s=1) have $C=-1$

## 3.2 G operator (strong interactions only)

- Few particles are their own antiparticles, so charge conjugation alone is of limited interest
- Its use can be extended by focusing only on strong interactions
- If we then rotate it $180^\circ$ around the second axis of isospin space, we might get its antiparticle (because we change $I_3\to -I_3$). If we then apply charge conjugation, we get the original particle:

$$C(R_2\pi^+)=C(\pi^-)=\pi^+$$

- We define the operator:

$$G=CR_2; \ \ R_2=e^{i\pi I_2}$$

$$\boxed{\text{All mesons carrying only u/d quarks are eigenstates of $G$}}$$

- For an isospin multiplet, the eigenvalue of $G$ is $\boxed{G=(-1)^IC}$, where $C$ is the eigenvalue of $C$ of the neutral member
- For a system of $n$ pions, $G=(-1)^n$
- Combining these two, we can predict whether the number of pions emitted in a decay is even or odd (for instance $\rho$ has $I=1, C=-1$ and therefore $G=+1$, which implies we can only have even numbers of pions emitted. It emits 2)

# 4. CP violation

- We know parity symmetry is violated maximally (see neutrinos), but if we combine charge conjugation and parity: $CP$, then we seem to recover the behaviour of nature
- If we understand $CP$ as the mirror image of things instead of just $P$, then we do get coherent results: the mirror image of a (left-handed) neutrino is a (right-handed) antineutrino, and so on

## 4.1 Kaons violate CP symmetry

- It is experimentally observed neutral kaons (with strangeness +1) can turn into their antiparticles (with strangeness -1), just as $D^0$ and $B^0$ can, too
- Therefore, the particles we detect in the lab are not $K^0, \bar K^0$ but rather a linear combination of the two. Particularly, we can build eigenstates of the $CP$ operator as follows:

$$|K_1\rangle = \frac{1}{\sqrt 2}(|K^0\rangle-|\bar K^0\rangle); \ \ |K_2\rangle = \frac{1}{\sqrt 2}(|K^0\rangle+|\bar K^0\rangle)$$

such that:

$$CP|K_1\rangle=|K_1\rangle; \ \ CP|K_2\rangle=-|K_2\rangle$$

- $K_1$ decays into 2 pions (this makes sense, since it has CP=+1 and pions have C=+1 and P=-1 but there is 2 of them)
- $K_2$ decays into 3 pions. This decay is less energetic than the other and therefore slower

- If we start with a beam of $K^0$ particles, the $K_1$ component will soon die out (we expect many $2\pi$ events close to the source accounting for $K_1$ decays, and then exclusively $3\pi$ decays)

- Cronin and Fitch performed an experimnt in 1964 where they measured decay events of this type at the end of a 57ft beam and registered that of all 22,700 decays, 45 where $2\pi$ decays and so the long-lived neutral kaon is not exactly $|K_2\rangle$ as we thought, but rather $|K_L\rangle = \frac{1}{\sqrt{1+|\epsilon|^2}}(|K_2\rangle+\epsilon|K_1\rangle), \ \ \epsilon\simeq 2.3\cdot 10^{-3}$


- CP violation can be incorporated in the Standard Model by introducing a phase factor $\delta$ in the Kobayashi-Maskawa matrix, but it is currently an arbitrary input parameter

**My ideas:** perhaps Cronin and Fitch detected K1 decays because they travel faster than K2 particles and therefore their lifetime is elongated by special relativity. Or perhaps some of the K1 particles quantum-tunneled next to the measuring device, where they decayed. Perhaps a mixture of both

## 4.2 Convention-free definition of positive electric charge

- Subsequent studies of the long-lived Kaons $K_L$ have been made, and it has been found $39\%$ of the time $K_L$'s decay either in:

$$\pi^++e^-+\bar \nu_e$$

or in

$$\pi^-+e^++\nu_e$$

If we apply the $CP$ operator to the first decay we get the second one, and therefore if nature had CP symmetry (is $|K_L\rangle$ were an eigenstate of the $CP$ operator), both decays would have to be equally probable. Nonetheless, the second decay occurs more frequently than the first one by a factor of $3.3\cdot 10^{-3}$

- **The decay of $K_L$ particles makes an absolute distinction between matter and antimatter** (the fact that CP violation allows for the unequal treatment of particles and antiparticles may be the reason for the dominance of matter over antimatter in the Universe)
- **Convention-free way of defining positive electric charge:** it is the charge carried by the lepton produced preferentially in the decay of the long-lived neutral K meson

# 5. TCP theorem

- Newton's equations of motion and in fact all of Physics (except for Thermodynamics) has time-reversibility. This implies equations remain the same under the transformation $T: t\to -t$

- Nonetheless, there is a theorem which states "it is impossible to construct a quantum field theory in which the product TCP is violated". Since we know CP is indeed violated, then there must also be a slight violation of T (time reversal symmetry) in order to make up for it and make TCP an exact conservation

- This theorem is based only on the most general aspects of Physics (Lorentz invariance, QM and the idea that interactions are carried by fields), which makes it hard to debunk theoretically
- Experimentally, it has been checked to a good precision for strong and EM interactions. Weak ones (which are historically the most likely to break this symmetry) are hard to check because other interactions which are much stronger get in the way and the T violation is as slight as the CP one
- **If the TCP theorem is on the right, then all particles should have exactly the same mass and lifetime as their corresponding antiparticles**
