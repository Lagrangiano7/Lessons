# Quarks

- Quarks are one of the 3 elementary groups of particles (at least by now)

## Flavour

Quarks come in 3 flavours and their opposites (6 in total)

1. Up (u) / Down (d)
2. Charm (c) / Strange (s)
3. Top (t) / Bottom (b)

The only difference between up, charm and top / down, strange and bottom (apart from the fact that they are essentially different because the have different flavours) is their masses. Up, charm and top have the same charge, and are basically identical except for the fact that they have different masses. Since up is the lightest, it is also the most stable, so bulk baryons are made of it.

This threefold multiplicity leads us to state particles are ordered in 3 **generations**

## Color

**Def.:** a "meson" is a bound state of a quark and an antiquark: $q\bar q$

**Def.:** a "baryon" is a bound state of 3 quarks

**Question:** if quarks have spin 1/2 and if a baryon is uuu, how come this baryon can exist? According to Pauli's exclusion principle, it shouldn't, but we do observe it ($\Delta^{++}$)

**Answer:** quarks have yet another intrinsic property called "color" which differentiates them. Quarks come in 3 colors: Red, Blue and Green. Since Pauli's exclusion principle applies only to identical particles, the quarks composing this particle must have different colors.

**THEOREM:** all naturally occurring particles are colorless (as in their color charges cancel out or there is an equal representation of all 3 colors).

This is why we only have mesons (quark+antiquark= 0 color charge because antiparticles have opposite color charge of the same color) and baryons (3 quarks, all of them with a different color)

## Quark confinement

We cannot observe any quarks alone. They are always conforming mesons / baryons and they cannot be taken out of this bound state to study them isolated. This is due to the nature of the strong force acting on them

## Eightfold Way

The Eightfold Way is to Particle Physics as the periodic table is to Chemistry. It is a way of arranging particles (not quarks but mesons and baryons) according to their flavour charges (electric charge, strangeness, charm, etc.). We could also account for "upness" / "downness" but this is not necessary because the only quark which has charge 2/3 and 0 in every other charge (apart from upness) is the up quark, ídem for the down quark, so we identify them by the lack of the rest of charges.

Drawing diagonals corresponding to charge (Q) and strangeness (S) we get the supermultiplets. For instance, we can arrange the 8 lightest baryons in an hexagone (**the baryon octet**) placing one baryon on each vertex (corresponding to the intersection of a charge and a strangeness lines) and 2 baryons in the middle (which have Q=0, S=-1)

We can also draw the meson octet with the 8 lightest mesons in a completely analogous way.

**NOTE:** baryons can only have strangeness $S=0$ or $S<0$. Mesons, on the other hand, can have whichever strangeness (mind you, all these charges only take on integer values, except for charge, which takes on 1/3 or -2/3, where the unity is set by the charge of the electron)

We can also draw eightfold ways for quarks.

**Note:** we can take into account charm and other flavours in the eightfold way to obtain diagrams in higher dimensions (for instance, we can draw the 3D diagrams of mesons / baryons taking into account u/d, strangeness and charm).

# Leptons

Leptons are the other group of elementary particles which are not mediators.

**List of all leptons:** electron, muon, tau, electron neutrino, mu neutrino, tau neutrino, and their antiparticles (positron, electronic antineutrino, etc.)

There also fall in 3 generations, which are as described.

To every one of the generations of the electron corresponds a neutrino which is always with them when a particle decays into one of these leptons.

### Characteristics

Just as quarks have flavour, colour and charge, leptons have charge and leptonic numbers. Depending on the type of interaction, the characteristics associated with it must be conserved (for instance, in strong force interactions, charge and strangeness are to be conserved, but strangeness need not be conserved in weak interactions, such as the disintegration of the $\Omega^-$ baryon)

Electrons (and electronic neutrinos) have +1 electronic number $L_e$. Same goes for muons and taus with their respective numbers.

# Mediators

There are 4 fundamental forces (although at higher energies electromagnetic + weak + strong unify, and it is hypothesized that at high enough energies gravity unifies too). The meaning of unification is the interactions to which these forces give place become indistinguishable.

Each fundamental force has a carrier associated to it:

**Gravity:** graviton (?)
**Electromagnetic:** photon
**Weak:** intermediate vector bosons ($W^+, W^-, Z^0)$
**Strong:** gluons

**Note:** photons do not interact between them because they are "immune" to the electromagnetic interaction. Nonetheless, gluons are not immune to the strong interaction (even if they carry it!) and so they are much more condensed. It is hypothesized that gluons can conform a particle formed only by gluons called "glueball"
