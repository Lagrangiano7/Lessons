# 1. Analogy between QED and QCD

- When studying Hydrogen (the bound state of a proton and electron), we perform corrections which allow us to better understand the energy spectrum (fine / hyperfine structure)
- This bound state is ruled by QED exclusively, so how can we study bound states of quarks, which interact mainly through QCD? --> We need a potential, but we don't know it
- QCD and QED are similar in structure, except for the fact that the first one has some nonlinear terms which do not contribute much at short distances (remember, asymptotic freedom, the farther away they are, the more they attract eachother)

- Also, in QCD, short distance interactions are dominated by 1 gluon exchange, and in QED they are dominated by 1 photon exchange, and both bosons have spin 1 --> everything looks very similar --> we assume the short distance potential is identical in form to that of QED and the long distance potential must be confining

$$\boxed{V(r)=-\frac{4}{3}\frac{\alpha_s\hbar c}{r}+F_0r}; \ \ F_0\sim 16 \text{ tons}$$

- Sometimes the confining potential is modeled with a square term or a logarithm, but they all yield quite similar results because they do not differ that much in the short distances for which we have sensible probes

# 2. Quarkonium

**Def.:** quarkonium is the bound state conformed by a quark and its corresponding antiquark. It is chiefly used for heavy quarks, so in essence it only applies for **charmonium** ($c\bar c$) and **bottomonium** ($b\bar b$). Toponiu does not exist because, due to its large mass, it decays weakly before having time to form a bound state with its antiparticle

- **Why are we interested in heavy quark bound states insteaf of light ones?:** if the binding energy is comparable to the masses of the constituent particles, then the kinetic energy of said particles is very high and they are relativistic. This happens, for instance, for the light mesons. Therefore, we cannot apply the non-relativistic theory of quantum mechanics, which is the one we have been applying all along (quantum bound state theory) and we should use the relativistic bound state theory, which does not have exact solutions at this stage. However, for heavy quarks, the binding energy is less comparable to their masses and therefore they are non-relativistic --> we can use the bound state theory we know to perform calculations and corrections

- **NONETHELESS** even for heavy quarks the binding energy is still a significant fraction of the total energy and therefore we are compelled to regard different energy levels of the same bounf system as different particles

- **Why do hadrons have much higher masses than their constituent quarks?:** the total mass of the hydrogen atom comes from the rest masses of the constituent particles, the binding energy and the energy stored in the electrostatic field, which is about one part in a million of the total energy. Nonetheless, the energy stored in the gluon field of quarks makes up for 99% of the hadron's mass. Of course, the binding energy still subtracts mass, because the bound state must be les energetic than the free states separately to occur naturally

## 2.1 Comparison with positronium

- For quarkonium then, we have a quark and an antiquark, which have exactly the same rest mass (if TCP holds) and opposite charges. Therefore, we should expect bound states similar to those of positronium!
- Positronium (positron+electron) can be solved by altering the masses in the hydrogen atom. Of course, the hyperfine correction is now of the same order as the fine correction, since it goes with a factor of $m_e/m_p$ which is now equal to 1
- Since we said in short distances QCD behaves like QED, we essentially have the same system! Quark and antiquark where we had electron and antielectron before, so they should have the same bound states (of course with different energies, but same spacing) --> this is observed!
- **Note:** the Schrödinger solution for the full strong potential (Coulombic + linear term) is not known

# 3. Light mesons

- These are relativistic
- They can have total spin 0 (pseudoscalar nonet) or 1 (vector nonet)
- We shall only study their ground state (l=0)

**Note:** if SU(3) were a perfect symmetry, then all particles ina given supermultiplet would have the same mass, but they don't (since neutrons and protons have practically the same mass, we say isospin is an approximate symmetry which holds well for u and d quarks. This is due to the fact that their masses are close). Flavour symmetry is broken precisely because different quarks have different masses

- Why is $\pi^0$ lighter than $\rho^0$ if they have the same quark content? --> they only differ in the total spin, so the difference in masses must only be attributable to the spin-spin interactions of the quark and the antiquark composing the mesons (remember, $\pi$ has s=0 and $\rho$ has s=1)
- There is a "hyperfine correction" which accounts for the difference in masses (remember, since both the quark and the antiquark have the same mass, they hyperfine correction is as significant as fine corrections)
- For the Hydrogen atom, the hyperfine correction is:

$$\Delta E = \frac{8\pi \gamma_p e^2}{3m_em_pc^2}(\vec S_e\cdot \vec S_p)|\Psi_{100}(0)|^2$$

- Therefore, sticking to our analogy between QCD and QED at short distances, we propose a "spin-spin correction" for mesons:

$$\Delta E = k(\vec S_1\cdot \vec S_2)$$

which is plausible because we saw the only possibility to explain the mass difference is the different spin-spin coupling, so it makes sense to have said term. Therefore, the mass of a meson is to be determined with the semi-empirical law:

$$M=m_1+m_2+A\frac{\vec S_1\cdot \vec S_2}{m_1m_2}$$

- Pseudoscalar mesons (s=0): $\vec S_1\cdot \vec S_2=-\frac{3}{4}\hbar^2$
- Vector mesons (s=1): $\vec S_1\cdot \vec S_2=\frac{1}{4}\hbar^2$
- A is the same for all mesons and is determined fitting data

# 4. Baryons

- Baryons are bound states of 3 quarks, and therefore all baryons are fermions
- As fermions, they must abide the Pauli exclusion principle as well as the antisymmetry of their wavefunction / state vector
- We now have a 3 body problem, and as such it has 2 orbital angular momenta (the combined angular momentum of 2 particles around their centre of mass and the angular momentum of particle 3 with this combination with respect to the total centre of mass)

## 4.1 State vector of a baryon

- In order to completely specify the state of a baryon, we must give:

1. Spatial state / dynamic state
2. Flavour state
3. Spin state
4. Colour state

$$\boxed{|\text{baryon}\rangle\equiv |\text{space}\rangle|\text{flavour}\rangle|\text{spin}\rangle|\text{colour}\rangle}$$

- Spatial ground state (l=l'=0) so it must be symmetric
- Spin state can be j=3/2 (completely symmetric) or j=1/2 (mixed symmetry, meaning it is only symmetric under the permutation of 2 given particles but not the other)
- We shall only study light baryons (those formed by u/d/s), so there are 27 possible combinations. We can form linear combinations of flavour (isospin) to give place to flavour states with well-defined symmetry

### 4.1.1 Building flavour states with well-defined symmetry

- Say we have a proton, which is composed of uud quarks. The proton has $I=1/2, I_3=1/2$, so in the coupled isospin base we can write it as $|p\rangle \equiv |1/2,1/2\rangle$. We now want to obtain the symmetric state vector representing the proton in the uncoupled base (which is formed by the individual quark combinations). Since $|u\rangle \equiv |1/2,1/2\rangle; \ |d\rangle\equiv |1/2,-1/2\rangle$ in the uncoupled base, we have:

$$|p\rangle = |1/2,1/2\rangle_C\equiv \frac{1}{\sqrt 3}(|uud\rangle + |udu\rangle+|duu\rangle)$$

which are the three possible combinations which yield this coupled basis ket

### 4.1.2 EVERY HADRON IS IN A COLOUR SINGLET

- Just as the 3 lightest quark flavours generate flavour SU(3), the 3 colours generate colour SU(3). The former is an approximate symmetry, because quarks of different flavour have different mass. The latter is an exact symmetry (quarks of different flavour carry exactly the same mass)
- We can therefore form a colour decuplet, 2 colour octets and a colour singlet just as we did for flavour, but nature always chooses the singlet. For baryons, this is:

$$|\text{colour}\rangle = \frac{1}{\sqrt 6}(|\text{rgb}\rangle-|\text{rbg}\rangle+|\text{gbr}\rangle-|\text{grb}\rangle+|\text{brg}\rangle-|\text{bgr}\rangle)$$

**THIS IS COMMON FOR ALL BARYONS**
**THIS STATE VECTOR IS ANTISYMMETRIC**

- Since it is common for all baryons, we usually do not bother to include it in the ket
- We must keep this in mind when constructing the state ket of a baryon!

- For instance, $\Delta^+$ is a baryon with s=3/2, which means the spin part is completely symmetric. If we wish to study its state in the spatial ground state, then the spatial part is also symmetric. Therefore, the flavour part must be symmetric too, and we thus obtain an antisymmetric state ket (due to the colour part being antisymmetric)
- The state ket for $\Delta^+$ in the spatial ground state and spin $m_j=-1/2$ configuration, then we would say:

$$|\Delta^+\rangle=|\text{flavour}\rangle|\text{spin}\rangle=\frac{1}{\sqrt 3}(|\text{uud}\rangle+|\text{udu}\rangle+|\text{duu}\rangle)\otimes \frac{1}{\sqrt 3}(|\downarrow\downarrow\uparrow\rangle+|\downarrow\uparrow\downarrow\rangle+|\uparrow\downarrow\downarrow\rangle)$$

- **Note:** baryons with s=3/2 form the baryon decuplet and those with s=1/2 form the baryon octet
- The baryon octet is trickier because there are 4 spin 1/2 configurations in the coupled base depending on which of the two indexes are antisymmetric and which one is symmetric (this is what we call "mixed symmetry")
- We can make linear combinations of these spin states to get one with well-defined symmetry
- For the baryon octet (s=1/2), we can form symmetric states as follows:

$$|\text{octet}\rangle = \frac{\sqrt 2}{3}(\Psi_{12}(\text{spin})\Psi_{12}(\text{flavour})+\Psi_{23}(\text{spin})\Psi_{23}(\text{flavour})+\Psi_{13}(\text{spin})\Psi_{13}(\text{flavour}))$$

**Note:** when you build these state vectors, you will surely obtain states such as $u(\uparrow)u(\uparrow)d(\downarrow)$ where two quarks of the same flavour have the same spin state. Nonetheless, remember every particle is colorless and therefore these quarks are not violating the Pauli exclusion principle by being in the same spin state since they have different colors. This applies for mesons too, where the composing quarks have opposite color charges

## 4.2 Baryon magnetic moments (octet)

- Baryons from the spin decuplet and vector mesons do have a magnetic dipole moment but it has never been measured
- Pseudoscalar mesons have s=0 by definition so they do not have any magnetic dipole moment
- We have only measured the magnetic dipole moment of baryons from the octet, so we'll study these (therefore, only those composed of u/d/s)

- In the spatial ground state, l=l'=0, so the only contribution to the baryon's magnetic dipole moment is that of its 3 composing quarks:

$$\vec \mu=\vec \mu_1+\vec \mu_2+\vec \mu_3$$

- **Note:** different flavours of quarks carry different magnetic dipole moment, generations differ in more than just mass

- The magnetic dipole moment of a point-like particle of charge $q$ and mass $m$ is:

$$\vec \mu = \frac{q}{mc}\vec S$$

- We shall study their third component, denoted by $\mu$ and will refer to as "the magnetic moment" although strictly speaking this should be the whole vector $\vec \mu$:

$$\mu = \pm\frac{q\hbar}{2mc}$$

- Therefore, the magnetic momenta of the 3 quarks we will be working with will be (for spin up):

$$\mu_u=\frac{2}{3}\frac{e\hbar}{2m_uc}; \ \ \mu_d=-\frac{1}{3}\frac{e\hbar}{2m_dc}; \ \ \mu_s=-\frac{1}{3}\frac{e\hbar}{2m_sc};$$

- Therefore, the magnetic moment of a baryon B of the octet is determined as:

$$\mu = \langle B\uparrow |\mu_1+\mu_2+\mu_3|B\uparrow\rangle$$

- Note: we take the spin up case just as we could take the spin down, it doesn't really matter, since the sign depends on our choice of the Z axis
- Note: in order to account for the sign introduced when the individual quark's spin is up or down (with respect to the whole particle's third component) we can write the expectation value introducing $S_z$ and normalizing by a factor $2/\hbar$. Therefore, we will only be introducing a sign. Finally:

$$\mu = \frac{2}{\hbar}\displaystyle\sum_{i=1}^3\langle B\uparrow| \mu_iS_{i_z} |B\uparrow\rangle$$

**Example: calculating the magnetic dipole moment of the proton**

- The proton is a baryon from the octet and its state vector is given by:

$$|p\rangle=\frac{1}{3\sqrt 2}[2u(\uparrow)u(\uparrow)d(\downarrow)-u(\uparrow)u(\downarrow)d(\uparrow)-u(\downarrow)u(\uparrow)d(\uparrow)] \text{;  and permutations of $d$}$$

- Therefore, we only need to calculate the contribution of these 3 first terms and the rest will be the same (the algebraic sum is commutative), so we can just multiply what we get by 3. We have:

$$\mu_p=\frac{1}{3}(4\mu_u-\mu_d)$$

## 4.3 Baryon masses (octet and decuplet)

- If SU(3) were a perfect symmetry (which it isn't), all baryons in the octet would have the same mass, but they don't
- This could be because the $s$ quark is heavier than u/d, but this cannot be the whole truth because otherwise $\Delta(uud)$ would have the same mass as the proton $p(udd)$, and this is not the case
- **Explanation:** spin-spin coupling contributes to the mass of the baryon (again, hyperfine correction as was the case in mesons)
- We proceed as before, although now we have 3 coupled spins, so:

$$M(\text{baryon})=m_1+m_2+m_3+A'\left[\frac{\vec S_1\cdot \vec S_2}{m_1m_2}+\frac{\vec S_1\cdot \vec S_3}{m_1m_3}+\frac{\vec S_2\cdot \vec S_3}{m_2m_3}\right]$$

### 4.3.1 Case for baryons of equal mass quarks

- Note: we admit u/d have the same mass (approx.), so we can factor out the mass terms in the denominator and couple the spins altogether

$$\vec S_1\cdot\vec S_2+\vec S_1\cdot\vec S_3+\vec S_2\cdot\vec S_3=\frac{\hbar^2}{2}(j(j+1)-9/4)$$

therefore:

- **Proton/Neutron:** $M_{p/n}=3m_u-\frac{3}{4}\frac{\hbar^2}{m_u^2}A'$
- $\Delta$: $m_\Delta = 3m_u+\frac{3}{4}\frac{\hbar^2}{m_u^2}A'$
- $\Omega^-$: $m_{\Omega^-}=3m_s+\frac{3}{4}\frac{\hbar^2}{m_s^2}A'$

### 4.3.2 Case for baryons of the decuplet

- They have s=3/2, so every pair of spins must make s=1

$$\vec S_1\cdot \vec S_2=\frac{1}{2}(S_{12}^2-S_1^2-S_2^2)=\frac{\hbar^2}{4}$$

and so on for the other 2 pairs

- **Note:** notice this is in perfect accordance with what we would obtain coupling the 3 spins altogether for the decuplet baryons in the case of equal mass quarks (it adds up to $3\hbar^2/4$)

### 4.3.3 Case for other baryons

- For other baryons the calculation is not so simple, but it always boils down to calculating the separate pairs of $\vec S_i\cdot \vec S_j$. This can be done by studying the symmetry of the ket and we obtain similar formulae
