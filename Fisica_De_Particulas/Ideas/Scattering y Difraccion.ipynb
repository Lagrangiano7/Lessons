**Idea:** el scattering de una partícula cuántica no es sino la difracción de su función de onda, siempre estudiada en campo lejano (difracción de Fraunhofer) porque no tenemos la resolución para observar en campo cercano. Entonces, un haz coherente de partículas cuánticas debería formar el mismo patrón de difracción que una onda EM clásica en el mismo experimento. Como los fotones son relativistas, intuyo que la difracción de la función de onda se asemejará más a aquella de la luz cuando las partículas sean muy rápidas / energéticas, lo que me permite utilizar la aproximación de Born.

## Planteamiento del experimento

- Sea un haz coherente de partículas cuánticas de momento $\vec p$ sobre el eje Z que inciden sobre una "máscara" (un potencial) de rendija. Por el principio de Babinet, será lo mismo usar una apertura de rendija que una máscara con la forma de esa rendija, de modo que escogemos la segunda por sencillez. Consideramos el problema en 2 dimensiones, de modo que el potencial queda:

$$V=V_0\delta(x)\delta(z)\theta(a-|y|)$$

![Difraccion](Diffraction.png)

En aproximación de Born, la amplitud de probabilidad es la transformada de Fourier del potencial, al igual que en la difracción de campo lejano, la intensidad de la onda va como la transformada de Fourier de la máscara. Tenemos entonces una amplitud de scattering:

$$f^{(1)}(\vec p \ '-\vec p)=-\frac{m}{2\pi}\int d^3\vec x e^{-i\vec q\cdot \vec x}V(\vec x)$$

dode $\vec q=\vec p \ '-\vec p$ es el momento transferido. Integrando:

$$f^{(1)}(\vec p \ '-\vec p)=-\frac{mV_0}{\pi q_y}sin(q_ya)$$

pero $q_y=(\vec p \ '-\vec p)\cdot \hat y=p\sin\theta$, por lo que:

$$f^{(1)}(\vec p \ '-\vec p)=-\frac{mV_0a}{\pi}sinc(pa\sin\theta)$$

Asimismo, en campo lejano, $\sin\theta\approx \tan\theta = \frac{y}{L}$, por lo que finalmente:


$$f^{(1)}(\vec p \ '-\vec p)=-\frac{mV_0a}{\pi}sinc\left(pa\frac{y}{L}\right)$$

Y por tanto la sección eficaz diferencial será:

$$\boxed{\frac{d\sigma^{(1)}}{d\Omega}=\left(\frac{mV_0a}{\pi}\right)^2sinc^2\left(pa\frac{y}{L}\right)}$$

que es precisamente la forma que tiene un patrón de difracción de ondas EM por una sola rendija
