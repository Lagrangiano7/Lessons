- Vamos a aplicar el estudio de las invariancias de un Hamiltoniano a la teoría de scattering. El objetivo es ver cómo la presencia de una invariancia en el sistema restringe notablemente la forma de la matriz de scattering

**Nota:** restringiremos la discusión cuando sea preciso al problema de scattering de 1 sola partícula por un potencial fijo, porque el problema de 2 partículas es del todo equivalente

# 1. Traslación

- Se define el operador de traslación como:

$$T(\vec a)=e^{-i\vec P_{CM}\cdot \vec a}$$

apréciese que la traslación va con el momento lineal total del sistema porque la traslación se efectúa de igual manera para todas las partículas del sistema. Supongamos que el sistema es invariante bajo traslaciones, es decir

1. $[H_0, T(\vec a)]=0$ (se cumple siempre porque va con $\vec P_{CM}^2$)
2. $[H, T(\vec a)]=0$

si el sistema es invariante bajo traslaciones, es fácil ver que el operador de traslación conmuta asimismo con los operadores de onda de Moller:

$$[T(\vec a), \Omega_{\pm}]=0$$

y por ende conmuta también con el operador de scattering:

$$[S, T(\vec a)]=0$$

Esto significa que:

$$\langle \phi|S|\psi\rangle = \langle \phi_T|S|\psi_T\rangle$$

**Interpretación:** si el sistema es invariante bajo traslaciones, entonces el proceso de scattering es asimismo invariante bajo traslaciones (realizar el experimento en distintos puntos del espacio nos dará los mismos resultados)

## Corolario: conservación de momento lineal total

- Además, si el sistema es invariante bajo traslaciones, el momento lineal total del sistema se conserva (las interacciones entre dos partículas que hemos considerado hasta ahora siempre venían mediadas por un potencial central, de modo que el sistema era invariante bajo traslaciones implícitamente. Es por esto que nos salía el término $\delta^3(\vec P_{CM}'-\vec P_{CM})$: garantizaba que se debía conservar el momento lineal total del sistema porque este era invariante bajo traslaciones

# 2. Rotación (espacio y spines)

- El operador de rotación (en el espacio) viene dado por:

$$D(\alpha, \hat n)=e^{-i\vec J_{T}\cdot \hat n\alpha}$$

Si el sistema es invariante bajo rotaciones, es decir, $[H, D(\alpha, \hat n)]=0$ para todo ángulo y eje, entonces el momento angular total (que no es el del CM) se conserva

**Nota:** el momento angular orbital total es:

$$\vec J_T=\vec X_{CM}\times \vec P_{CM}+\vec J$$

donde $\vec J=\vec X_{rel}\times \vec p_{rel}+\vec S_1+\vec S_2$ es el momento angular total del sistema efectivo de 1 sola partícula

## Rotación en el espacio

- Supongamos que las partículas tienen ambas spin 0. Entonces, la única contribución al momento angular será el orbital, y tendremos que si el sistema es invariante bajo rotaciones en el espacio, este se conserva
- Esta invariancia bajo rotaciones se satisface si el potencial de interacción es esférico: $V=V(r)$

**Resultado 1:** si el sistema es invariante bajo rotaciones, entonces el operador de scattering también lo es y tenemos que la amplitud de scattering es la misma si, para un par $\vec p_{rel}, \vec p_{rel}'$ fijo, sometemos el sistema (o los momentos) a una rotación en el espacio. Esto es:


$$\langle \vec p_{rel}'|S|\vec p_{rel}\rangle = \langle \vec p_{rel, R}'|S|\vec p_{rel, R}\rangle$$

**Nota:** este resultado es inmediato, porque otra forma de expresar $[S, D]=0$ es $D^\dagger S D=S$

de él se deriva que:

$$f(\vec p_{rel'} \leftarrow \vec p_{rel})=f(\vec p_{rel', R} \leftarrow \vec p_{rel, R})$$

**OJO!:** véase que realmente estas dos amplitudes de scattering podrían ser iguales salvo una fase. En el caso particular que estamos estudiando, en el que ninguna de las dos partículas (o la partícula y el target fijo) no tienen spin, entonces la fase es nula y las amplitudes son idénticamente iguales. Sin embargo, en el caso más general en el que sí permitamos que tengan spin, tendremos que diferirán por una cierta fase

**Resultado 2:** si el sistema es invariante bajo rotaciones (seguimos sin spin), la amplitud de scattering no es ya función de 5 variables: $f(\vec p_{rel}'\leftarrow \vec p_{rel})$ (digo 5 porque estamos con colisiones elásticas, de modo que $|\vec p_{rel}'|=|\vec p_{rel}|$ ya fija una de las 6 componentes) sino que depende únicamente de (1) la energía del sistema y (2) de la colatitud, pero no del ángulo acimutal

$$\boxed{\text{Invariancia bajo rotaciones + sin spin} \Longrightarrow f = f(E_p, \theta)}$$

Esta reducción de grados de libertad del sistema tiene sentido: antes teníamos 5 y ahora tenemos 2, porque tenemos 3 magnitudes conservadas (las 3 componentes del momento angular orbital). Es de esperar que si el grupo de simetría es reducido (por ejemplo, si tenemos un campo magnético en el eje Z, donde solo se conservaría $L_z$), entonces tendremos 4 grados de libertad porque solo se conserva una de las tres componentes del momento angular orbital

## Expansión en ondas parciales para partículas sin spin

- Exploremos más en profundidad el caso del scattering de una partícula (o la partícula efectiva que surge de un problema de scattering de 2 partículas) sin spin cuando el sistema tiene simetría rotacional (i.e: el potencial de interacción es esférico)
- En este caso, conviene tomar el CCOC $\{H_0, L^2, L_3\}$. Véase que, como estamos admitiendo a priori que el sistema es invariante bajo rotaciones, se cumple $[S, L_i]=0$ y por ende **el operador de scattering es diagonal en esta base** porque se puede diagonalizar junto con los tres observables del CCOC (como la energía se conserva, también es $[S, H_0]=0$)
- Trabajamos entonces en la base:

$$|E, l, m\rangle$$

Estos kets se llaman "ondas esféricas", porque no son otra cosa que armónicos esféricos! (Con un ket añadido de energía, pero esencialmente fíjate en que esto es lo mismo que en el átomo de Hidrógeno, donde tenemos $|n,l,m\rangle$). El CCOC también es el mismo en el átomo de Hidrógeno, y es que conviene tomar el mismo CCOC porque la simetría del problema es la misma. Sin embargo, ahora como estamos en scattering y nos interesa estudiar los estados en el continuo, la energía no está cuantizada sino que es continua, de modo que mientras que la normalización del átomo de Hidrógeno es $\langle n'l'm'|nlm\rangle = \delta_{nn'}\delta_{ll'}\delta_{mm'}$, tenemos ahora $\langle E'l'm'|Elm\rangle=\delta(E-E')\delta_{ll'}\delta_{mm'}$

Las funciones de onda de la base esférica en representación de posiciones vienen dadas por una parte radial que son funciones de Ricatti-Bessel (que corresponden a una partícula libre) y una parte angular de armónicos esféricos:

$$\langle \vec x|Elm\rangle = i^l\left(\frac{2m}{\pi p}\right)^{1/2}\frac{1}{r}\hat j_l(pr)Y_l^m(\hat x)$$

**Nota:** véase que, como $S$ es diagonal en esta base (bajo las suposiciones de que la partícula no tiene spin y de que el sistema tiene simetría esférica), tenemos inmediatamente que los elementos de matriz de $S$ en esta base son:

$$\langle E'l'm'|S|Elm\rangle = s_l(E)\delta(E-E')\delta_{ll'}\delta_{mm'}$$

**Nota:** como $S$ es un operador escalar, por el teorema de Wigner-Eckart es incapaz de cambiar el número magnético (y desde luego no puede cambiar el número acimutal). Véase también que, como el problema tiene simetría esférica, todas las direcciones del espacio son equivalentes e indistinguibles y por ende los autovalores del operador de scattering no dependen del número magnético

**Def.:** como la matriz $S$ es unitaria, sus autovalores han de tener módulo unidad. Denotamos entonces sus autovalores por:

$$\boxed{s_l(E)=e^{2i\delta_l(E)}}$$

donde $\delta_l(E)$ recibe el nombre de "refasaje". Nótese que, así definido, el refasaje tiene una ambigüedad en tanto que podemos sumarle un múltiplo de $\pi$ y obtener el mismo resultado

**Comentario sobre la importancia de cada base:** hasta ahora, hemos trabajado en la base de momentos, y esto ha sido provechoso porque nos permitía relacionar la amplitud de scattering y las magnitudes directamente mensurables asociadas a dicha amplitud (sección eficaz diferencial, etc.) con los elementos de matriz del operador de scattering. Por otro lado, bajo ciertas condiciones de simetría (y ausencia de spin) elegimos ahor trabajar en la base de ondas esféricas. La gran ventaja de esta base frente a la de momentos es que diagonaliza el operador de scattering, de modo que podemos obtener más información sobre la amplitud de scattering, como veremos a continuación, al precio de que nos quede en forma de sumatorio. Sin embargo, en general este se podrá truncar a los primeros términos en buena aproximación

## Desarrollo en ondas parciales de la amplitud de scattering

Por un lado, en la base de momentos sabemos calcular exactamente la amplitud de scattering:

$$\langle \vec p'|S-1|\vec p\rangle = \frac{i}{2\pi m}\delta(E_{p'}-E_p)f(\vec p'\leftarrow \vec p)$$

Por otro lado, también sabemos expresar las ondas esféricas en representación de momentos:

$$\langle \vec p|Elm\rangle = \sqrt{\frac{1}{mp}}\delta(E_{p}-E)Y_l^m(\hat p)$$

Introduciendo la identidad de $|Elm\rangle$ entre el operador y el ket $|\vec p\rangle$, tenemos entonces:

$$\langle \vec p'|S-1|\vec p\rangle = \int dE\displaystyle\sum_{l,m}\langle \vec p'|S-1|Elm\rangle \langle Elm|\vec p\rangle=\frac{1}{mp}\delta(E_{p'}-E_p)\displaystyle\sum_{l,m}Y_l^m(\hat p')[s_l(E_p)-1]Y_l^m(\hat p)^*$$


- Comparando con la expresión de arriba, podemos despejar la amplitud de scattering inmediatamente, de modo que:

$$f(\vec p'\leftarrow \vec p)=\frac{2\pi}{ip}\displaystyle\sum_{l,m}Y_l^m(\hat p')[s_l(E_p)-1]Y_l^m(\hat p)^*$$

**Nota:** hasta aquí, el desarrollo es completamente general y no se precisa que el sistema tenga ningún tipo de simetría adicional. Es ahora cuando imponemos simetría esférica (**no:** porque ya has admitido que S es diagonal en la base de L)

Como se admite que el sistema tiene simetría esférica, podemos coger la trayectoria de entrada de la partícula en la dirección del eje $Z$, de modo que únicamente los armónicos esféricos $Y_l^0(\hat p)\neq 0$ (recuerda que los que tienen $m=0$ van con $p_z$ separado, mientras que los que van con $m\neq 0$ van con $p_x, p_y$ multiplicando a $p_z$, de modo que se anulan. De forma un poco más técnica, estamos escogiendo solo aquellos tensores esféricos que vayan con un otro operador, $p_z$, que a su vez ya es tensor esférico de $m=0$, de modo que es esperable que las únicas contribuciones que no se anulen sean las que tengan asimismo $m=0$). Entonces, tenemos que la amplitud de scattering queda de la forma:

$$f(E_p, \theta)=\frac{1}{2ip}\displaystyle\sum_{l=0}^\infty (2l+1)[s_l(E)-1]P_l(\cos\theta)$$

donde hemos aprovechado la definición de los armónicos esféricos con $m=0$ como polinomios de Legendre de argumento $\cos\theta$. Vemos entonces que la amplitud de scattering exacta queda en forma de sumatorio infinito. Para compactar la notación, podemos hacer:

$$\boxed{f(E_p, \theta)=\displaystyle\sum_{l=0}^\infty (2l+1)f_l(E)P_l(\cos\theta)}$$

donde hemos definido unas "amplitudes efectivas", si se quieren entender así, de modo que la amplitud exacta es un sumatorio de dichas amplitudes parciales. De ahí viene que este desarrollo reciba el nombre de "desarrollo en ondas parciales".

**Nota:** las amplitudes parciales quedan entonces de la forma:

$$\boxed{f_l(E)=\frac{s_l(E)-1}{2ip}=\frac{e^{i\delta_l(E)}\sin\delta_l(E)}{p}}$$

**Nota:** el desarrollo en ondas parciales es invertible. Esto es, si conocemos todas las amplitues parciales $\{f_l(E)\}$, podemos calcular la amplitud de scattering con el sumatorio. Si por otro lado lo que conociéramos, de alguna manera, fuera la amplitud de scattering completa, podríamos encontrar su onda parcial l-ésima haciendo:

$$f_l(E)=\frac{1}{2}\int_{-1}^1d(\cos\theta)f(E,\theta)P_l(\cos\theta)$$

## Desarrollo en ondas parciales de la sección eficaz total

- Si deseamos conocer la sección eficaz diferencial, no tenemos más que tomar $|f(E_p, \theta)|^2$. Sin embargo, esto nos deja con un doble suumatorio que ya de por sí es difícil de manejar:

$$\frac{d\sigma}{d\Omega}=\displaystyle\sum_{l,l'=0}^\infty (2l+1)(2l'+1)f_l(E)f_{l'}^*(E)P_l(\cos\theta)P_{l'}(\cos\theta')$$

Sin embargo, si integramos sobre el ángulo sólido para obtener la sección eficaz total, la ortogonalidad de los polinomios de Legendre cancela uno de los sumatorios, de modo que tenemos una expresión en serie para la sección eficaz total:

$$\boxed{\sigma(p)=\displaystyle\sum_{l=0}^\infty \sigma_l(p)}$$

donde $\boxed{\sigma_l(p)=4\pi (2l+1)|f_l(p)|^2=4\pi(2l+1)\frac{\sin^2\delta_l(p)}{p^2}}$

**Nota:** véase que las contribuciones a la sección eficaz total están acotadas superiormente. El término l-ésimo solo puede contribuir, como máximo, con:

$$\sigma_l(p)\leq 4\pi\frac{2l+1}{p^2}$$

Esta cota superior recibe el nombre de "unitarity bound" en inglés, porque viene de que los autovalores de $S$ tengan módulo unidad, y esto a su vez viene de que $S$ sea unitario.

**Nota:** en la práctica, la unitarity bound (el máximo de la contribución l-ésima) se alcanza cuando estamos cerca de alguna resonancia (entonces, se cumple que $\delta_l\approx (2k+1)\pi/2$)

## Ventajas del desarrollo en ondas parciales

1. Su mayor uso es a bajas energías, porque solo contribuyen los primeros términos (solo los primeros refasajes parciales, $\delta_l(p)$, no son 0), de modo que podemos conocer toda la amplitud de scattering en buena aproximación y en todas direcciones, mientras que en representación de momentos solo podíamos conocer el módulo de la amplitud de scattering (salvo hacia delante, por el teorema óptico)
2. En particular, a muy bajas energías se encuentra que la sección eficaz **diferencial** (en el inercial del CM) es isotrópica (no depende de $\theta$. De $\varphi$ ya sabíamos que no iba a depender por la simetría del sistema). Esto implica que solo el primer término del sumatorio de la amplitud de scattering contribuye (porque $P_0=1$, mientras que el resto ya dependen del ángulo)
3. En el caso de muy bajas energías, la amplitud de scattering queda idéntica a la de orden 0, de modo que:

$$f(E)=\frac{1}{p}e^{i\delta_0(E)}\sin\delta_0(E)$$

que, como vemos, efectivamente es isotrópico. Esto es ventajoso porque medir la sección eficaz diferencial (o total) nos da el refasaje (salvo signo). Estamos extrayendo más información que en la representación de momentos, a cambio de que la simetría de nuestro sistema es más exigente

# 3. Paridad

**Nota:** este desarrollo atañe solo a las partículas sin spin

**Def.:** se define el operador paridad $\Pi$ sobre kets como aqel que invierte la paridad de la magnitud asociada al ket. Esto es:

$$\Pi|\vec x\rangle = e^{i\eta}|-\vec x\rangle; \ \ \Pi|\vec p\rangle = e^{i\eta}|-\vec p\rangle; \ \ \Pi|\vec J\rangle = e^{i\eta}|\vec J\rangle $$

donde el momento angular no cambia porque es un vector "axial / par", mientras que posición y momento se dicen "polares / impares". Podemos tomar $\eta=0$, reabsorbiéndolo si se quiere dentro del propio operador.

**Nota:** en scattering relativista, esta reabsorción no es posible, porque el número de partículas del sistema puede variar, de modo que la inversión de paridad en general introducirá desfases con significado físico entre ellas. Por otro lado, como nosotros estamos considerando el caso no relativista, donde el número total de partículas se conserva, todas tienen el mismo desfase introducido por la paridad y por tanto se puede sacar como una fase global sin mayor significado físico.

**Resultado:** si $[\Pi, S]=0$, entonces $S$ no puede cambiar la paridad de los kets. Por tanto, el operador de scattering $S$ solo conecta estados de igual paridad:

$$\Pi|\phi_1\rangle = -|\phi_1\rangle; \ \ \Pi|\phi_2\rangle = +|\phi_2\rangle\Longrightarrow \langle \phi_1|S|\phi_2\rangle = 0$$

**Dem.:**

Suponemos que el sistema es invariante bajo inversión de paridad (esto lo cumplen todas las interacciones salvo la débil), de modo que $[S, \Pi]=0$. Suponemos también que tenemos dos estados de paridad bien definida tales que:

$$\Pi|\phi_p\rangle = |\phi_p\rangle; \ \ \Pi|\phi_i\rangle = -|\phi_i\rangle$$

Tomemos el elemento de matriz del conmutador:

$$\langle \phi_p|[\Pi, S]|\phi_i\rangle = 0 = \langle \phi_p|\Pi S|\phi_i\rangle - \langle\phi_p|S\Pi|\phi_i\rangle$$

Pero haciendo actuar el operador paridad del primer sumando por la izquierda y recordando que $\Pi=\Pi^\dagger$, tenemos:

$$2\langle \phi_p|S|\phi_i\rangle = 0 \Longrightarrow \langle \phi_p|S|\phi_i\rangle = 0_\blacksquare$$

**Resultado:** si el sistema es invariante bajo paridad, entonces en particular se cumple que $\langle \vec p'|S|\vec p\rangle = \langle -\vec p'|S|-\vec p\rangle $ y de aquí se sigue inmediatamente que:

$$\boxed{[S, \Pi]=0\Longrightarrow f(\vec p'\leftarrow \vec p)=f(-\vec p'\leftarrow -\vec p)}$$

**Observación:** véase que si el potencial tiene simetría esférica, entonces automáticamente encontramos que el sistema es invariante bajo paridad y se cumple todo esto. En particular, el resultado de las amplitudes de scattering al que acabamos de llegar se sostiene porque $|-\vec p\rangle$ equivale a aplicar una rotación de 180º al sistema, pero este es invariante bajo rotaciones en el espacio.

**Nota:** en general, veremos que la invariancia bajo rotaciones y la invariancia bajo inversión de paridad están bastante desconectadas. Por ejemplo, podríamos no tener simetría de rotación en los estados de spin pero sí simetría de paridad en estos

# 4. Inversión temporal del movimiento

**Teorema de Wigner:** toda operación de simetría viene representada o bien por un operador unitario o bien por uno antiunitario (de forma mutuamente exclusiva)

**Def.:** un operador unitario es aquel operador que cumple:

1. Es lineal
2. Mapea todo $\mathcal H\to \mathcal H$ (automorfismo)
3. Preserva la norma

**Def.:** un operador antiunitario es aquel que cumple 2 y 3 pero en lugar de ser lineal es antilineal, de modo que:

$$\theta (a|\psi_1\rangle + b|\psi_2\rangle) = a^*\theta|\psi_1\rangle + b^*|\psi_2\rangle$$

**Propiedad:** mientras que los operadores unitarios cumplen $\langle \psi_1|U|\psi_2\rangle=\langle \psi_2|U|\psi_1\rangle$, los antiunitarios conjugan este elemento de matriz: $\langle \psi_1|\theta|\psi_2\rangle = \langle \psi_2|\theta|\psi_1\rangle ^*$

## Actuación de $\Theta$ sobre funciones de onda

- Esta sección es más bien mía, voy a separarme un poco de la teoría de sattering para ahondar más en cómo se puede entender la acción de $\Theta$
- Como $\{\Theta, P\}=0$ y $[\Theta, X]=0$, tenemos entonces que:

$$\Theta|\vec p\rangle = |-\vec p\rangle; \ \ \Theta|\vec x\rangle = |\vec x\rangle$$

Ahora que sabemos cómo actúa sobre los vectores impropios que conforman las bases habituales para representar funciones de onda, podemos ver cómo actúa el equivalente funcional del operador $\Theta$ sobre funciones de onda en las dos distintas representaciones:


$$\Theta \Psi(\vec x)=\Theta\int d^3\vec x|\vec x\rangle \langle \vec x|\Psi\rangle = \int d^3\vec x|\vec x\rangle \langle \vec x|\Psi\rangle^* = \Psi^*(\vec x)$$

Por el mismo argumento, tenemos que:

$$\Theta\Psi(\vec p)=\Psi^*(-\vec p)$$

Probemos ahora a aplicar $\Theta$ a ambos lados de la ecuación de Schrödinger dependiente del tiempo:

$$\Theta(i\hbar\frac{\partial \Psi(\vec x, t)}{\partial t})=\Theta(H\Psi(\vec x, t))$$

Supongamos que el sistema es invariante bajo inversión temporal del movimiento, de modo que $[H, \Theta]=0$ (admitimos que ahora $\Theta$ es un operador funcional, no el operador que actúa sobre kets, aunque esencialmente son lo mismo). El resultado de este cómputo en base a lo que sabemos es inmediato (recuerda que la $i$ se conjuga por ser $\Theta$ antilineal):

$$-i\hbar\frac{\partial \Psi^*(\vec x, t)}{\partial t}=H\Psi^*(\vec x, t)$$

Pero esto nos da más bien poco significado físico. Supongamos que el Hamiltoniano solo tiene espectro ligado (discreto) por simplicidad. Entonces, podemos expandir las funciones de onda en la base de sus autoestados. La evolución temporal de estos es trivial, y su parte real se puede escoger real, de modo que:

$$\Psi(\vec x, t)=\displaystyle\sum_{n=1}^\infty \phi_n(\vec x)e^{-iE_nt} \Longleftrightarrow \Psi(\vec x, t)=\Psi(\vec x, -t)^*$$

de modo que tomar el complejo conjugado de la función de onda es lo mismo que hacer $t\to -t$. Definamos el operador funcional $W$ cuya actuación sea dejar el espacio invariante pero hacer $t\to-t$ y hagámoslo actuar a ambos lados de la ecuación de Schrödinger. Obtenemos entonces:

$$i\hbar \frac{\Psi(\vec x, -t)}{\partial (-t)}=H\Psi(\vec x, -t)$$

Donde he admitido que $H(t)=H(-t)$. Aplicando la igualdad que acabamos de obtener sobre conjugación compleja e inversión temporal, llegamos a la misma ecuación que obteníamos al aplicar $\Theta$ a ambos lados a la ecuación de Schrödinger en un sistema invariante bajo inversión temporal:

$$-i\hbar\frac{\partial \Psi^*(\vec x, t)}{\partial t}=H\Psi^*(\vec x, t)$$

de modo que debe cumplirse $\Theta\equiv W$ salvo una fase global.

**Resultado 1:** la acción del operador funcional de inversión temporal del movimiento es $t\to -t$

**Resultado 2:** un Hamiltoniano (sin spin) será invariante bajo inversión temporal del movimiento si y solo si el potencial es real

## Acción de $\Theta$ sobre $S$

- Nos preguntamos ahora cómo actúa el operador de inversión temporal sobre el operador de scattering:

- Sea un sistema invariante bajo inversión temporal del movimiento, de tal forma que $[H, \Theta]=0$. Entonces, tenemos:

$$\Theta e^{iHt}=e^{-iHt}\Theta$$

De este modo, la acción sobre los operadores de Moller será:

$$\Theta\Omega_{\pm}=\Theta\left[\displaystyle\lim_{t\to \mp\infty} e^{iHt}e^{-iH_0t}\right]=\left[\displaystyle\lim_{t\to \mp\infty} e^{-iHt}e^{iH_0t}\right]\Theta$$

donde hemos admitido que $[H_0, \Theta]=0$ también (esto siempre se cumplirá porque $H_0=\frac{\vec p^2}{2m}$). Analicemos ahora la acción de $\Theta$ sobre $S$:

$$\Theta S=\Theta\Omega_-^\dagger \Omega_+=\Omega_+^\dagger \Theta\Omega_+=\Omega_+^\dagger \Omega_-\Theta$$


Por tanto:



$$\boxed{\Theta S=S^\dagger \Theta}$$

Si existe una invariancia bajo inversión temporal del movimiento en el sistema, entonces se cumple:

$$\langle \chi, S\phi\rangle = \langle \chi, \Theta^\dagger S^\dagger \Theta \phi\rangle$$

$$=\langle \Theta\chi, S^\dagger \Theta \phi\rangle^* = \langle \chi_\Theta, S^\dagger \phi_\Theta\rangle ^*$$

Por tanto:

$$\boxed{\langle \chi | S |\phi\rangle = \langle \chi_\Theta |S|\phi_\Theta\rangle^*}$$

**Intepretación:** la probabilidad de que una partícula entrante salga con estado saliente es la misma que la probabilidad de que entre con el estado que antes era saliente y salga con el que antes era entrante (i.e: el proceso es indistinguible si sucede al revés en el tiempo, como es signatura de la invariancia bajo inversión temporal del movimiento)

En particular, para el scattering de 1 sola partícula, tenemos que:

$$f(\vec p'\leftarrow \vec p)=f(-\vec p\leftarrow -\vec p')$$

Si el sistema es en efecto invariante bajo inversión temporal del movimiento, ambas amplitudes de scattering serán iguales. Podíamos haber empleado la intuición física desde el pricipio para intuir este resultado

# Invariancias para procesos con spin

**Nota:** sea un proceso en representación de momento lineal que asimismo tiene cierto spin (que trataremos en la base acoplada). Los vectores de estado del sistema vendrán entonces determinados completamente por estos 5 números cuánticos:

$$|\vec p, s, m\rangle$$

donde $s$ es el módulo del spin total del sistema y $m$ es la tercera componente del spin total

Hasta ahora, las operaciones de simetría actuaban sobre el único ket que había, que era el de momento lineal. Sin embargo, ahora actuarán tanto sobre este como sobre el de spin. En algunos casos, como en la paridad, veremos que el resultado no se complica, mientras que en otros, como las rotaciones, en general sí

# 1. Inversión temporal del movimiento

- Recordemos que sobre kets de momento angular, la inversión temporal del movimiento invertía la tercera componente, de modo que:

$$\boxed{\Theta|\vec p, s, m\rangle = (-1)^{s-m}|-\vec p, s, -m\rangle}$$

donde la fase es global, y es solo una convención

# 2. Rotación

- Hasta ahora, hemos considerado únicamente las rotaciones en el espacio. Naturalmente, la rotación más general, $R=e^{-i\vec J_T\cdot \hat n\alpha}$, será una rotación de la parte espacial y otra de la parte de spin
- Consideramos ahora rotaciones completas, esto es, tanto en el espacio como en el espacio de spin
- Naturalmente, como el espacio y el espacio de spin son disjuntos (en tanto que podemos factorizar el espacio de Hilbert total como producto tensorial de ambos subespacios), es de suponer que podemos construir el operador de rotación total de la forma:

$$R=R_{esp}\otimes R_s = e^{-i\vec J\cdot \hat n\alpha}$$

Por tanto, la actuación de la rotación sobre un ket general será:

$$R|\vec p, \xi\rangle = |\vec p_R\rangle\otimes R_s|\xi\rangle$$

Debemos preguntarnos ahora cómo actúa la rotación de estados de spin. Introduciendo la identidad de la base del subespacio de spines, tenemos que:

$$R_s|\xi\rangle = \displaystyle\sum_{\xi'}|\xi'\rangle\langle \xi'|R_s|\xi\rangle$$

Ahora sabemos que una rotación en el subespacio de estados de spin se puede descomponer en términos de la base de dicho subespacio tomando elementos de matriz bastante sencillos. Tenemos entonces que la rotación total actúa de la forma:


$$R|\vec p, \xi\rangle = |\vec p_R\rangle\otimes \displaystyle\sum_{\xi'}|\xi'\rangle\langle \xi'|R_s|\xi\rangle$$

- Admitamos ahora que el sistema tiene invariancia bajo rotaciones (totales, tanto espaciales como de spin). Entonces, tenemos que:

$$\langle \vec p', \xi'|S|\vec p, \xi\rangle = \langle \vec p', \xi'|R^\dagger SR|\vec p, \xi\rangle$$

- Como siempre podremos encontrar una parte espacial de la rotación que nos alinee $\vec p$ con el eje Z, no perdemos generalidad si admitimos que el momento lineal es paralelo a dicho eje. Entonces, la matriz de amplitudes de scattering vendrá dada por:

$$F(\vec p'\leftarrow \vec p)=F(E_p, \theta,\varphi)$$

**Nota:** en el caso de 1 partícula, la amplitud de scattering no dependía de $\varphi$. Sin embargo, veremos ahora que sí lo hace, por la acción del spin. Como admitíamos antes que la partícula no tenía spin, lógicamente obteníamos una sola amplitud en lugar de una matriz de ellas, y esta no dependía del ángulo acimutal. 

**Observación:** si la rotación se realiza en torno al eje Z (con más generalidad, a lo largo del eje de la tercera componente de spin, que se toma por simplicidad como Z), entonces, la rotación sobre la parte de spines es trivial: $D_z(\alpha)|sm\rangle = e^{-im\alpha}|sm\rangle$. Por otro lado, el momento lineal sí que cambia en tanto que se ve rotado en el espacio

- Buscamos ahora una expresión para la matriz de amplitudes de scattering. Por un lado, como admitimos que el sistema es invariante bajo rotaciones, tenemos que $$\langle \vec p', \xi'|S|\vec p, \xi\rangle = \langle \vec p', \xi'|R^\dagger SR|\vec p, \xi\rangle$$

- Por otro lado, si tomamos $\vec p$ alineado con el eje Z, entonces el momento de salida vendrá completamente determinado por sus ángulos esféricos $(\theta, \varphi)$ (recordemos que estamos estudiando colisiones elásticas, de modo que el módulo del momento lineal se conserva). Entonces, podemos facilitarnos la vida empleando la igualdad de arriba:


$$\langle \vec p',s',m'|S|\vec p, s, m\rangle = \langle \vec p',s',m'|R_z^\dagger(-\varphi)S R_z(-\varphi)|\vec p, s, m\rangle $$

tomamos una rotación en torno el eje $Z$ de ángulo $-\varphi$ porque así le quitamos el momento lineal de salida su orientación acimutal. Tenemos que $R_z(-\varphi)|\vec p', s',m'\rangle = e^{im'\varphi}|\vec p'', s', m'\rangle$ con $\vec p''=(\theta, 0)$. En un momento veremos por qué nos interesa hacer esto. Llamando $f_{ss'mm'}(E,\theta, 0)=\langle \vec p''|S|\vec p\rangle$, tenemos entonces que:

$$\boxed{f_{ss'mm'}(E, \theta, \varphi)=e^{i(m-m')\varphi}f_{ss'mm'}(E, \theta, 0)}$$

**Resultado:** las amplitudes de scattering para un sistema de dos partículas con spin e invariancia rotacional son las mismas que para un sistema sin invariancia rotacional solo que su dependencia se simplifica de $f_{ss'mm'}(\vec p'\leftarrow \vec p)$ a la habitual de la invariancia rotacional $f_{ss'mm'}(E, \theta, \varphi)$, donde la dependencia con el ángulo acimutal una fase relativa

**Observación:** naturalmente, si admitimos que las partículas no tienen spin, solo tendremos una amplitud de scattering, y su desfase será nulo, de modo que como cabía esperar recuperamos la amplitud de scattering del caso sin spin y con invariancia rotacional. Sirva este desarrollo como justificación de por qué el caso sin spin no depende del ángulo acimutal

- Por ejemplo, en el caso de scattering de 2 partículas donde una de ellas tiene $s_1=1/2$ y la otra $s_2=0$, la matriz de amplitudes de scattering queda reducida de la forma más general cuando no hay simetría bajo rotaciones:

$$F(\vec p'\leftarrow \vec p)=\begin{pmatrix}f_{++}(\vec p'\leftarrow \vec p) & f_{+-}(\vec p'\leftarrow \vec p) \\ f_{-+}(\vec p'\leftarrow \vec p) & f_{--}(\vec p'\leftarrow \vec p)\end{pmatrix}$$

a la forma más simple:

$$F(E, \theta,\varphi)=\begin{pmatrix}a(E, \theta) & b(E, \theta)e^{-i\varphi} \\ c(E, \theta)e^{i\varphi} & d(E, \theta)\end{pmatrix}$$

# 3. Paridad +  Rotación

- Solo paridad: como el spin es un vector axial, no se invierte bajo paridad, es decir: $\Pi|s, m\rangle = |m, s\rangle$. Por ende:

$$\boxed{\Pi|\vec p, s, m\rangle = |-\vec p, s, m\rangle }$$

- Consideremos ahora el caso en el que el sistema es invariante bajo rotaciones y también bajo inversión de paridad (los sistemas son siempre invariantes bajo paridad salvo cuando las interacciones débiles son apreciables)
- Sigamos con el sistema de 2 partículas, donde la primera tiene spin 1/2 y la segunda spin 0
- Supongamos que el momento entrante es $\vec p$ y el saliente es $\vec p'$ (por simplicidad, tomemos estos en el plano $y=0$ con $\vec p$ alineado con el eje Z
- Si hacemos actuar paridad, los momentos se invierten (al spin no le pasa nada). Si ahora aplicamos una rotación de $\pi (rad)$ en torno al eje Y, logramos devolver los momentos a su configuración inicial. Sin embargo, la rotación también afecta al spin
- La operación que estamos considerando es: $M_y=R_y(\pi)\Pi$. Como admitimos que el sistema es invariante tanto bajo rotaciones como bajo inversión de paridad, sabemos que esta operación conmuta con el operador de scattering, de modo que:

$$S=M_y^\dagger SM_y$$

(naturalmente, la composición de dos operaciones unitarias es asimismo unitaria). De este modo, los elementos de matriz tomados a $S$ serán los mismos que tomados a $M_y^\dagger SM_y$

**Nota:** $R_y(\pi)|s,m\rangle = (-1)^{s-m}|s,-m\rangle$ (se comprueba fácilmente para el caso de spin 1/2, o de forma más general con la representación de las matrices de Wigner en ángulos de Euler)

- Como la operación $M_y$ relaciona los estados de iguales momentos entrante y saliente pero invierte los spines, encontramos inmediatamente que podemos obtener dos de los cuatro estados del sistema a partir de los otros dos. Efectivamente, tenemos que:

$$M_t|\vec p, m\rangle = (-1)^{s-m}|\vec p, -m\rangle$$

Como el sistema es invariante bajo rotaciones y bajo inversión de paridad, en particular será invariante bajo $M_y$, de modo que:


$$\langle \vec p', m'|S|\vec p', m'\rangle = (-1)^{m'-m}\langle \vec p', -m'|S|\vec p, -m\rangle$$

Donde del primer elemento de matriz se obtiene $f_{ss'mm'}(\vec p'\leftarrow \vec p)$ y del otro se obtiene la amplitud de scattering correspondiente al proceso en el que los spines se han invertido. Tenemos entonces que la matriz de amplitudes de scattering es:

$$F(E, \theta, \varphi)=\begin{pmatrix}a(E, \theta) & b(E, \theta)e^{-i\varphi} \\ -b(E, \theta)e^{i\varphi} & a(E, \theta)\end{pmatrix}$$

**Nota:** vemos entonces que la matriz de amplitudes de scattering se ha reducido de cuatro funciones a dos. Vemos que las simetrías de un sistema se pueden explotar para reducir el número de grados de libertad del mismo

# Desarrollo en ondas parciales en el caso con spin

- Anteriormente, hemos estudiado el problema de scattering de una (o dos) partículas sin spin cuando había invariancia rotacional
- Nos preguntamos ahora qué sucede si las partículas tienen spin

## Invariancia rotacional

- Para explotar la simetría de revolución, trabajaremos en la base de momento angular total (que es la magnitud conservada, porque admitimos que el sistema tiene invariancia rotacional en espacio y spin). Entonces, lo que hace al caso es trabajar en la base acoplada de $\vec J=\vec L+\vec S_1+\vec S_2$. Los kets de dicha base serán $|E, l, s, J, m_j\rangle$. Naturalmente, como el Hamiltoniano conmuta con $\vec J$, tendremos que $S$ vuelve a ser diagonal en esta base:

$$\boxed{\langle E', l', s', J', m_j' |S|E, l, s, J, m_j\rangle = \delta(E-E')\delta_{jj'}\delta_{mm'}s^j_{l's'ls}(E)}$$

donde notamos que, como antes, los autovalores de $S$ no dependen de la tercera componente del momento angular

**Def.:** podemos definir una matriz $S^j(E)$ cuyas columnas vengan dadas por una configuración de módulos de momento angular entrante ($, s$) y cuyas filas vengan determinadas por cierta configuración de módulos de momento angular saliente ($l',s'$). Naturalmente, su número de elementos de matriz será el número de combinaciones posibles de $l, s$ y de $l', s'$ tales que den un $J$ (que se mantiene constante porque conmuta con el Hamiltoniano)

![Entradas de la matriz S_J](Matriz_S_J.jpg)

**Nota:** si consideramos el caso particular en el que las partículas tienen spin 0, volvemos a que los autovalores de $S$ son, sencillamente, $e^{2i\delta_l(E)}$.

## Invariancia rotacional + de paridad

- La acción de la inversión de paridad sobre los estados de nuestro CCOC es $\Pi|E, l, s, J, m_J\rangle = (-1)^l |E, l, s, J, m_J\rangle$ (son autoestados de la paridad por construcción)
- **Resultado:** si el sistema es invariante bajo inversión de paridad, $S$ no puede conectar estados de entrada y salida de distinta paridad
- Si admitimos que el sistema tiene invariancia bajo rotación + paridad, entonces por rotación tenemos que los elementos de matriz de $S$ son los que hemos descrito arriba, y por paridad tenemos que $l-l'$ impar $\Longrightarrow $ elemento de matriz nulo. Combinando ambos resultados, tenemos entonces que:

$$\boxed{\text{Invariancia rot. + paridad} \Longrightarrow s^j_{l's'ls}=0 \text{  si $l,l'$ distinta paridad}}$$

- Por tanto, si el sistema tiene invariancia bajo rotaciones e inversión de paridad, aprox. la mitad de los elementos de matriz de $S^j$ son nulos

## Invariancia rotacional + inversión temporal del movimiento

- La acción de $\Theta$ sobre nuestros elementos de la base es: $\Theta|E, l, s, J, m_j\rangle = (-1)^{j-m_j}|E, l, s, J, -m_j\rangle$
- Por tanto, los elementos de matriz $(E', l', s', J', m_J' \leftarrow E, l, s, J, m_J)$ y $(E, l, s, J, -m_J \leftarrow E', l', s', J', -m_J')$ deben ser los mismos (salvo el factor de fase que saque $\Theta$. Estarán desfasados $(-1)^{j-j'-m_j+m_j'}$)

- Si además admitimos también invariancia bajo rotaciones, entonces los elementos de matriz vienen dados por $s^j_{l's'ls}(E)$ (y son indeppendientes de la tercera componente de momento angular), de modo que:

$$\boxed{\text{Invariancia rot. + $\Theta$}\Longrightarrow S^j \text{ simétrica}}$$

- Esto era esperable, porque da igual si el estado de entrada viene con $l, s$ y sale con $l', s'$ o si entra con $l', s'$ y sale con $l, s$, porque los procesos que suceden hacia delante en el tiempo son indisgintuibles de los que suceden hacia atrás

**Resumen:**

- Rot: elementos de matriz diagonales para salvo en dependencia de $l, s$. Son independientes de la tercera componente del momento angular
- Rot. + paridad: $S$ diagonal + elementos con $l, l'$ de distinta paridad son nulos
- Rot. + inv. temporal del movimiento: $S^j$ simétrica

**Ejemplo:**

- Sea nuestro problema habitual de scattering de 2 partículas donde una tiene $s_1=1/2$ y la otra tiene $s_2=0$
- Si admitimos que el sistema tiene invariancia bajo rotaciones, tendrá cierto momento angular total $J$ que permanecerá invariante en todo el proceso
- El momento angular de entrada será: $l=J\pm 1/2$ y el de salida, igual: $l'=J\pm 1/2$ (admitiendo que $J>0$ siempre). Entonces, las entradas de nuestra matriz $S^j$ serán 4, correspondientes a los dos posibles estados de entrada: $(J+1/2, 1/2)$ y $(J-1/2, 1/2)$, y los dos estados de salida, que serán los mismos. Tenemos entonces que la matriz $S^J$ tendrá en general la forma:

![](S_J_spin_medio.jpg)

- Si admitimos ahora que el sistema tiene asimsmo invariancia bajo inversión de paridad, es inmediato que los elementos fuera de la diagonal, que corresponden a un cambio en el momento angular orbital del sistema, son nulos
- Por tanto, si nuestro sistema de 2 partículas con spines 1/2 y 0 tiene invariancia bajo rot. + inv. de paridad, entonces su matriz $S^J$ vendrá dada por:

$$S^J=\begin{pmatrix}e^{2i\delta_+^J} & 0 \\ 0 & e^{2i\delta_-^J}\end{pmatrix}$$

**Observación:** notamos que la matriz $S^J$  es simétrica, de modo que en este caso la invariancia rot. + paridad implica invariancia bajo inversión temporal del movimiento. Sin embargo, este no será el caso en general (pero sí lo será si $S^J$ es bidimensional, como en nuestro caso)

- Esto permite realizar un desarrollo en ondas parciales de las amplitudes de scattering en la base de momentos en términos de los autovalores de $S^J$ como en el caso sin spin, aunque este cálculo dice Taylor que es complicado
- Recordemos que la forma más general de la matriz de amplitudes de scattering en el sistema de spines 1/2-0 cuando existe simetría máxima (de rotación + paridad) es (en representación de momentos):

$$F(E, \theta, \varphi)=\begin{pmatrix}a(E, \theta) & b(E, \theta)e^{-i\varphi} \\ -b(E, \theta)e^{i\varphi} & a(E, \theta)\end{pmatrix}$$

- Tenemos ahora un desarrollo en ondas parciales para cada una de las dos funciones:

$$a(E, \theta)=\displaystyle\sum_{l=0}^\infty [(l+1)f_l^+(E)+lf_l^-(E)]P_l(\cos\theta)$$
$$b(E, \theta)=\displaystyle\sum_{l=0}^\infty [f_l^+(E)-f_l^-(E)]\sin\theta P'_l(\cos\theta)$$

donde $f_l^j = \frac{e^{2i\delta_l^{\ j}}-1}{2ip}$

- Por ejemplo, en el estudio del scattering pion-nucleón por debajo de los $200MeV$, el los datos del scattering se pueden ajustar bien si nos quedamos solo con los términos hasta $l=1$ (inclusive). De este modo, el proceso entero se puede describir dando únicamente 3 parámetros reales: $\delta_0^+$, $\delta_1^+$, $\delta_1^-$ (donde el subíndice indica el $l$ y el superíndice indica si al acoplar $l$ y $s$, estos se suman o se restan, dando así distintos j's)

## Formalismo de helicidad

- Hemos estudiado qué sucede en un sistema de 2 partículas con ciertos spines cuando este presenta invariancia rotacional. No obstante, es bastante complicado y poco conveniente, de modo que desearíamos encontrar un acercamiento equivalente que nos facilite la vida
- **Def.:** la "helicidad" ($\lambda$) de una partícula es la proyección de su spin sobre la dirección del movimiento. Naturalmente, irá entonces desde $-s$ hasta $s$ en pasos de 1 (es sin más coger el eje Z en una dirección dada y proyectar el spin, como hacemos siempre, solo que ahora tomamos como eje Z la dirección del movimiento)
- **Propiedad:** como la helicidad es la proyección del spin sobre la dirección del movimiento, no se ve afectado por rotaciones
- **Def.:** el "operador de helicidad" es $\vec S\cdot \hat P$, pero puede escribirse como $\vec J\cdot \hat P$ porque $\vec L\cdot \hat P=0$ siempre (recuerda que $\vec L=\vec X\times \vec P$)
- Las partículas sin masa (i.e: fotón) solo pueden tener dos estados de spin: $-s, s$ en la dirección del movimiento (i.e: $\lambda=-s,s$) y puede que uno de dichos estados no se realice en la naturaleza (vemos que los neutrinos, que no tienen masa en el modelo estándar, son todos levógiros y no existen neutrinos dextrógiros). En estos casos, la helicidad es la forma natural de hablar del estado de spin, porque el estado de spin está íntimamente relacionado con el de movimiento (si la partícula no tiene masa)

**Nota:** al igual que podemos describir el estado de una partícula especificando su estado de movimiento y de spin $|\vec p, m\rangle$, también podemos hacerlo dando su estado de movimiento y su helicidad (es lo mismo pero cogiendo como tercer eje del operador de spin la dirección del movimiento, nada más)

**Construyendo la base de estados de helicidad:**

- Si consideramos que la partícula se mueve en el eje Z, tenemos entonces que el eje del movimiento coincide con el eje Z (absoluto) y entonces $m=\lambda$, nada cambia
- Si en general la partícula se mueve en la dirección $(\theta, \varphi)$, podemos construir el estado de movimiento en dicha dirección a partir de una rotación de un momento en el eje Z. Como la helicidad no se ve afectada por rotaciones (es operador escalar), entonces tenemos que:

$$|\vec p, \lambda\rangle = D(\varphi, \theta, -\varphi)|p\vec e_z, \lambda\rangle$$

donde la rotación es una matriz de Wigner dada en ángulos de Euler y generada por el momento angular total del sistema, como siempre:

$$D(\varphi, \theta, -\varphi)= e^{-iJ_3\varphi}e^{-iJ_2\theta}e^{iJ_3\varphi}$$

- La normalización de estos estados así construidos es entonces:

$$\langle \vec p',\lambda'|\vec p,\lambda\rangle = \delta^3(\vec p'-\vec p)\delta_{\lambda\lambda'}$$

**Extensión a sistema de 2 partículas:**

- Podríamos ponernos en el sistema de laboratorio y decir que el estado del sistema de 2 partículas no es más que:

$$|\vec p_1, \lambda_1\rangle \otimes |\vec p_2, \lambda_2\rangle$$

- Sin embargo, es más conveniente como siempre situarse en el inercial del CM, donde el problema queda reducido a aquel de una sola partícula efectiva. De este modo, tenemos que se puede tratar el estado como el de 1 sola partícula, donde ahora la rotación estará generada por el momento angular total del sistema: $\vec J=\vec J_1+\vec J_2$. Tendremos entonces:

$$|\vec p_{rel}, \lambda_1, \lambda_2\rangle = D(\varphi, \theta, -\varphi)|p_{rel}\vec e_z, \lambda_1, \lambda_2\rangle$$


- Como en la representación de tercera componente de spin, tenemos que los elementos de matriz de $S$ en esta base serán:

$$\langle \vec p_{rel}', \lambda_1', \lambda_2'|S|\vec p_{rel}, \lambda_1, \lambda_2\rangle=\delta^3(\vec p_{rel}'-\vec p_{rel})\delta_{\lambda_1\lambda_1'}\delta_{\lambda_2\lambda_2'}+\frac{i}{2\pi \mu}\delta(E_p'-E_p)f(\vec p_{rel}', \lambda_1', \lambda_2'\leftarrow \vec p_{rel}', \lambda_1', \lambda_2')$$

## Aprovechando invariancia bajo rotaciones

- Supongamos ahora que el sistema es invariante bajo rotaciones. Usaremos entonces el CCOC de momento angular
- **Nota:** como la helicidad es un operador escalar, a diferencia del spin, conmuta automáticamente con $\vec J$, de modo que podemos meterlo en el CCOC **sin necesidad de acoplar nada**. Esto es ventajoso porque ya no tenemos que preocuparnos de los Clebsch-Gordan. Tenemos entonces que el CCOC es:

$$CCOC:=\{H, J^2, J_3, \vec S_1\cdot \hat{P}_{rel}, \vec S_2\cdot \hat{P}_{rel}\}$$

De modo que los vectores de la base correspondiente serán:

$$|E, j, m, \lambda_1, \lambda_2\rangle$$

**Nota:** podemos hacer un desarrollo en ondas parciales como en el caso sin spin, solo que ahora estaremos desarrollando los distintos elementos de matriz de $S^J$. El razonamiento es del todo equivalente al del caso sin spin

- En representación de momentos, las ondas esféricas (ahora con estados de helicidad) son:

$$\langle \vec p\mu_1\mu_2|Ejm\lambda_1\lambda_2\rangle = \frac{1}{\sqrt{mp}}\delta(E_p-E)\delta_{\mu_1\lambda_1}\delta_{\mu_2\lambda_2}\left(\frac{2j+1}{4\pi}\right)^{1/2}R^j_{m\lambda}(\varphi, \theta, -\varphi)^*$$

con $\lambda=\lambda_1-\lambda_2$

- **Nota:** naturalmente, para el caso particular en el que las partículas no tienen spin, tenemos $\lambda=0$ y $J=l$, de modo que el factor $\left(\frac{2j+1}{4\pi}\right)^{1/2}R^j_{m\lambda}(\varphi, \theta, -\varphi)^*=Y_l^m(\theta, \varphi)$. Esto es, recuperamos las ondas esféricas habituales, a partir de las cuales sale el desarrollo en ondas parciales de la amplitud de scattering

- Ahora es un poco más complicado, porque tenemos varias amplitudes (recuerda la matriz $S^J$ que contiene cada posible income y outcome del experimento en polarizaciones bien determinadas). Al final, comparando los elementos de matriz de $S-1$ en la base $|\vec p_{rel}, \lambda_1 \lambda_2\rangle$ y en la base $|Ejm\lambda_1\lambda_2\rangle$, llegamos a que la amplitud de scattering es:

$$\boxed{f(\vec p'_{rel}, \lambda_1', \lambda_2'\leftarrow \vec p_{rel}, \lambda_1, \lambda_2) = f_{\lambda_1\lambda_2\lambda_1'\lambda_2'}(E, \theta, \varphi)=\displaystyle\sum_{j}(2j+1)f^j_{\lambda_1\lambda_2\lambda_1'\lambda_2'}(E)R^j_{\lambda\lambda'}(\varphi, \theta, -\varphi)^*}$$

**Nota:** para el caso de dos partículas sin spin, los elementos de matriz de la rotación quedan $R^l_{00}=P_l(\cos\theta)$ y recuperamos la amplitud de scattering para el scattering de 2 partículas sin spin
