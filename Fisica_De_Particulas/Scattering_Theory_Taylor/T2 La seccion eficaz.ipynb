# 1. $S$ conserva la energía

- Los estados asintóticos son autoestados del Hamiltoniano libre, $H_0$, de modo que como la energía se conserva en los procesos de scattering, la partícula ha de salir con la misma energía con la que entró. Una consecuencia inmediata de esta conservación de la energía es que el operador de scattering conmuta con el Hamiltoniano libre

$$H_0(S|\Psi_{in}\rangle)=H_0|\Psi_{out}\rangle = E_0|\Psi_{out}\rangle$$

$$S(H_0|\Psi_{in}\rangle)=E_0S|\Psi_{in}\rangle = E_0|\Psi_{out}\rangle$$

Y por tanto:

$$\boxed{[H_0, S]=0}$$

**Nota:** como el Hamiltoniano libre es $H_0=\frac{\vec p^2}{2m}$, sus autoestados serán los autoestados del momento lineal, denotados $|\vec p\rangle$. Esto podría llevarnos a pensar que los elementos de matriz de $S$ son nulos salvo cuando tomamos el mismo momento a cado lado: $\langle \vec p'|S|\vec p\rangle\propto \delta^3(\vec p'-\vec p)$, en analogía a cómo el operador de evolución temporal conmuta con el Hamiltoniano y sucede esto. Sin embargo, recordemos que el operador de evolución temporal se construye a partir del Hamiltoniano, mientras que el operador $S$ no. Además, no tenemos garantizado que se puedan diagonalizar a la vez porque ese teorema solo aplica a dos observables, y aquí tenemos un observable y un operador unitario. Por tanto, en general $\langle \vec p'|S|\vec p\rangle\neq 0$

**Nota:** habitualmente, a lo que se le llama "matriz $S$" es al elemento de matriz $\langle \vec p'|S|\vec p\rangle$

# 2. Descomposición de la matriz de scattering

- Queremos ahora saber qué forma tiene la matriz de scattering para poder empezar a calcular. En primer lugar, como conmuta con el Hamiltoniano libre, tenemos:

$$\langle\vec p'|[H_0, S]|\vec p\rangle = 0 = (E_{p'}-E_p)\langle\vec p'|S|\vec p\rangle$$

- De aquí se deduce que el elemento de matriz ha de ser nulo siempre que las energías sean distintas (es lógico, de lo contrario se violaría la conservación de energía) de modo que ya inferimos que la matriz de scattering lleva una delta en las energías (esta forma es típica de las conservaciones exactas)

$$\langle\vec p'|S|\vec p\rangle \sim \delta(E_{p'}-E_p)$$

- Si no hubiera potencial, la asíntota de entrada coincidiría con la de salida y el operador $S$ sería sencillamente la unidad. Puede ser que la partícula, pese a ver el potencial, no interactúe con el target y pase de largo. Podemos definir el operador $R$ que mide cuánto se separa $S$ de la identidad, haciendo $S=I+R$. Naturalmente, los elementos de matriz de $R$ también han de llevar una delta en las energías, de modo que:

$$\langle \vec p'|R|\vec p\rangle = -2\pi i\delta(E_{p'}-E_p)t(\vec p'\leftarrow\vec p)$$

Por tanto, los elementos de matriz del operador de scattering son:

$$\boxed{\langle \vec p'|S|\vec p\rangle=\delta^3(\vec p'-\vec p)-2\pi i\delta(E_{p'}-E_p)t(\vec p'\leftarrow\vec p)}$$

**Interpretación:** cuando disparamos una partícula a un target, pueden suceder dos cosas: (1) que pase de largo del potencial sin dispersarse (como sucedía, por ejemplo, con la mayoría de partículas alpha en el experimento de Rutherford). En este caso, es natural que se debe conservar el momento por componentes, porque no ha pasado nada. (2) Que la partícula sí se disperse. En este caso, lo que se debe conservar es la energía, de modo que la partícula salga de la región de potencial apreciable con la misma energía con la que entró. De este modo, lo que debe conservarse es el módulo del momento (porque la energía va con $\vec p^2$) pero no sus componentes, porque puede salir en una dirección distinta a aquella en la que entró. Como sabemos que solo se va a conservar el módulo pero no las componentes, vemos que ya tenemos todas las deltas que necesitamos en dicho factor (solo una, la de la energía), de modo que esperamos que $t(\vec p'\leftarrow \vec p)$ no contenga más deltas y por tanto debe ser una función suave en sus argumentos, que son los momentos de entrada y salida

**Def.:** véase que $t(\vec p'\leftarrow \vec p)$ va acompañado de una delta en el módulo de los momentos, de modo que decimos que solo está definida cuando el módulo del momento se conserva. Es decir, si es una función definida en el 6-espacio con ejes $\{p_x, p_y, p_z, p_x', p_y', p_z'\}$, la función solo está definida en las hipersuperficies donde $\vec p'^2-\vec p^2=0$. Se dice entonces que los elementos de matriz del operador $T$ (que son $t(\vec p'\leftarrow \vec p)$) están "en su capa de energías / on shell", porque solo están definidos sobre la shell $\vec p'^2=\vec p^2$ (imagina que dejas $\vec p^2$ fijo, equivalente a lanzar la partícula con un momento bien determinado, y que solo permites cambiar el momento de salida. Entonces, $t$ solo está definida sobre la esfera de radio $\vec p^2$)

- Veremos en el Tema 8 que se puede definir un operador $T$ tal que sus elementos de matriz $\langle p'|T|\vec p\rangle = t(\vec p'\leftarrow \vec p)$ **solo cuando** $\vec p'^2=\vec p^2$. Es decir, definiremos un operador que será una extensión de nuestro concepto, cuyos elementos de matriz nos den lo que queremos cuando la situación sea físicamente realizable. Sin embargo, no podríamos construir un operador solo a partir de aquellos elementos de matriz on shell porque para definir el operador hay que hacerlo en todos los kets de la base. A este operador extendido, más general, se le llama "operador $T$ off shell". Cuando queramos recuperar $t$, bastará tomar elementos de matriz del operador off shell e imponer la condición on shell (esto es, particularizar a una situación físicamente realizable)

**Def.:** se define la "amplitud de scattering" como:

$$\boxed{f(\vec p'\leftarrow \vec p)=-(2\pi)^2m \ t(\vec p'\leftarrow \vec p)}$$

- Esto se hace solo por conveniencia, porque será más fácil calcular la sección eficaz diferencial usando la amplitud de scattering. Sin embargo, $f$ y $t$ solo difieren en una constante, de modo que encierran la misma información. Tenemos entonces que el operador de scattering se puede descomponer de la forma (shell decomposition):

$$\langle \vec p'|S|\vec p\rangle = \delta^3(\vec p'-\vec p)+\frac{i}{2\pi m}\delta(E_{p'}-E_p)f(\vec p'\leftarrow \vec p)$$

# 3. Entendiendo el concepto de sección eficaz

- Sea un experimento de scattering clásico, en el que se tiene una partícula puntual que se lanza hacia un target rígido
- **Def.:** se define el "parámetro de impacto" $\vec \rho$ como el vector de posición de la asíntota de entrada desde el eje del centro de dispersión
- La partícula colisionará (o no) con el target y cambiará su dirección, pero poco podemos aprender de una sola colisión: tenemos que lanzar muchas partículas, y además tienen que estar uniformemente distribuidas en el espacio (en la práctica, en un plano perpendicular al momento incidente y de dimensiones superiores a las del target) para poder extraer resultados. Esto se consigue lanzando partículas con parámetro de impacto aleatorio (y uniformemente distribuido en dicho plano)
- **Def.:** tiene ahora sentido definir la densidad de partículas incidentes, $n_{inc}$, como el número de partículas por unidad de área contenidas en el plano perpendicualr al momento de entrada. Si la distribución de parámetros de impacto es uniforme, como hemos pedido, entonces esta densidad será además constante (independiente de la posición en el plano)
- El número total de partículas que sí han sido dispersadas (a diferencia de aquellas que (1) han pasado de largo o (2) han fallado) será entonces:

$$\boxed{N_{sc}=n_{inc}\sigma}$$

donde $\sigma$ es la "sección eficaz total" del target. Esta se entiende como el área que tiene "en la práctica" el target, suponiendo que lo hemos muestreado entero, repartiendo los parámetros de impacto uniformemente en el plano perpendicular al momento incidente

- El concepto de sección eficaz es mucho más potente que solo esto, porque podemos fijarnos en el número de partículas que han sido dispersadas en una cierta dirección (esto es, que caen dentro de cierta porción de ángulo sólido):

$$N_{sc}(\Delta\Omega)=n_{inc}\sigma(\Delta\Omega)$$

donde $\sigma(\Delta\Omega)$ es la sección eficaz del target que se ocupa de mandar las partículas a la porción de ángulo sólido considerada

## Sección eficaz diferencial

- Si queremos ver cuántas partículas se dispersan a un elemento infinitesimal de ángulo sólido, podemos tomar el límite $\Delta \Omega\to 0$, de modo que la expresión de arriba queda:

$$\frac{dN_{sc}}{d\Omega}=n_{inc}\frac{d\sigma}{d\Omega}$$

donde $\frac{d\sigma}{d\Omega}$ se llama "sección eficaz diferencial"

**Nota:** la sección eficaz diferencial **no está definida** hacia delante porque no podemos distinguir entre una partícula que ha sido dispersada hacia delante (i.e: ha dado dos vueltas y ha vuelto a salir por donde ha entrado) y una que sencillamente no se ha visto dispersada

- Estudiemos ahora el concepto de sección eficaz diferencial en dispersión de partículas cuánticas
- Planteamos un experimento completamente idéntico al clásico, solo que sustituimos las partículas clásicas puntuales por partículas cuánticas. En la práctica, estas estarán picadas en representación de momentos en torno a cierto a cierto valor de momento entrante $\vec p_0$. Ahora ponemos el target y disparamos las partículas cambiando el parámetro de impacto. Hacemos esto efectuando una traslación de los kets de las partículas de la forma:

$$\phi_\rho(\vec p)=e^{-i\vec p\cdot \vec \rho}\phi(\vec p)$$

ahora admitamos que repetimos el experimento muchas veces, con una partícula cada vez, con parámetros de impacto $\vec \rho$ uniformemente distribuidos en el plano perpendicular a $\vec p_0$, igual que en el caso clásico

- El número total de partículas dispersadas en un elemento de ángulo sólido será entonces la suma de la probabilidad de que la partícula acabe en dicho elemento de ángulo sólido en los $N$ experimentos llevados a cabo:

$$N_{sc}=\displaystyle\sum_{i=1}^NP(d\Omega\leftarrow \phi_{\rho_{i}})$$

Si tenemos suficientes eventos, podemos pasar al continuo:

$$N_{sc}=\int d^2\vec \rho \ n_{inc}P(d\Omega\leftarrow \phi_{\rho})$$

donde, recordemos, el parámetro de impacto $\vec \rho$ está contenido en el plano perpendicular a $\vec p_0$. Véase que da igual que integremos en todo el plano, porque la probabilidad se vuelve nula cuando el parámetro de impacto es demasiado grande, porque estamos tan alejados del centro de scattering que es muy improbable que la partícula se vea dispersada

Como la densidad de partículas incidentes se admite uniforme, podemos sacarla fuera de la integral y tenemos, al igual que en el caso clásico:


$$N_{sc}=n_{inc}\int d^2\vec \rho \ P(d\Omega \leftarrow \phi_{\rho}) = n_{inc}\sigma(d\Omega\leftarrow \phi)$$

**Interpretación de la sección eficaz diferencial:**

$$\sigma(d\Omega \leftarrow \phi)=\int d^2\vec \rho \ P(d\Omega \leftarrow \phi_\rho)$$

1. Como no conocemos exactamente el estado asintótico de entrada (porque el acelerador puede variar un poco al preparar uno), debemos promediar entre todos los experimentos realizados, y esto se consigue pesando la integral con la probabilidad de que una partícula con cierto parámetro de impacto acabe en el elemento infinitesimal de ángulo sólido que estamos considerando
2. Dado el elemento infinitesimal de ángulo sólio que estamos considerando, la integral se puede ver como una integral de superficie donde cada elemento de la superficie está pesado con una cierta probabilidad (la probabilidad de que dicho elemento de superficie mande la partícula a ese elemento infinitesimal de ángulo sólido), de modo que podemos imaginarnos el plano perpendicular a $\vec p_0$ como un relieve, más o menos picado en algunas regiones, correspondientes a aquellas que tengan mayor probabilidad de mandar la partícula al elemento infinitesimal de ángulo sólido considerado


En el caso clásico, que a diferencia del cuántico, es determinista, dada la asíntota de entrada cada elemento de área tendrá una probabilidad 0 ó 1 de mandar la partícula a un elemento infinitesimal de ángulo sólido dado. En el caso cuántico, las probabilidades se encuentran entre 0 y 1.

**Resultado:** se puede demostrar que, si la partícula se encuentra inicialmente suficientemente picada en torno a $\vec p_0$, entonces los resultados del scattering no dependen de la forma de la función de onda, de modo que tiene sentido escribir:


$$\sigma(d\Omega \leftarrow \vec p_0)$$

en lugar de $\sigma(d\Omega \leftarrow \phi)$

**Condiciones para que aplique nuestra definición idealizada de sección eficaz diferencial:**

1. Debemos usar un haz de muchas partículas y tener muchos centros dispersores puestos en tal disposición que las interacciones partícula-target se den 1 a 1 (por eso Rutherford usó una lámina delgada de oro, que tenía un grosor de aprox. 1 átomo)
2. Las partículas del haz han de estar uniformemente distribuidas o, alternativamente, los centros de dispersión en el target han de estarlo
3. El haz ha de ser suficientemente débil como para que las partículas no interaccionen apreciablemente unas con otras

# 4. Cálculo de la sección eficaz cuántica

- Hemos dicho que la sección eficaz diferencial (cuántica) viene dada por la integral de superficie del plano perpendicular al momento incidente donde cada elemento de superficie está pesado por la probabilidad de que este disperse la partícula a un elemento infinitesimal de ángulo sólido. Esta probabilidad será:

$$P(d\Omega \leftarrow \phi_\rho)=d\Omega\int_0^\infty p^2 |\Psi_{out}(\vec p)|^2dp$$

véase que, si integrásemos también sobre el ángulo sólido, obtendríamos la normalización del estado asintótico de salida, pero con toda seguridad divergería porque no es un estado propio (solo es normalizable en sentido de Dirac por ser autoestado del momento lineal). Entonces, tenemos:

$$\sigma(d\Omega\leftarrow \phi)=\int d^2\vec \rho \ P(d\Omega\leftarrow \phi_\rho(\vec p))$$

- Sabemos que $|\Psi_{out}\rangle = S|\Psi_{in}\rangle$, e introduciendo dos veces la identidad (en momentos) tenemos esta igualdad en funciones de onda en representación de momentos, a saber:

$$\Psi_{out}(\vec p)=\int d^3\vec p \ \langle \vec p|S|\vec p'\rangle \Psi_{in}(\vec p')$$

- También sabemos descomponer la matriz de scattering, de modo que tenemos que el estado asintótico de salida en función del de entrada viene dado por:

$$\Psi_{out}(\vec p)=\Psi_{in}(\vec p)+\frac{i}{2\pi m}\int d^3\vec p \ \delta(E_{p'}-E_p)f(\vec p'\leftarrow \vec p)\Psi_{in}(\vec p')$$

de modo que la función de onda de la partícula, lejos del centro de dispersión, es la función de onda entrante (correspondiente a la posibilidad de que la onda no se haya dispersado) más una amplitud correspondiente a la posibilidad de que sí se haya dispersado

- Centrémonos ahora en los valores del momento lineal que no se encuentran en un entorno del momento inicial $\vec p_0$. Hacemos esto porque lo que nos interesa es estudiar las partículas dispersadas, que serán aquellas para las que haya cambiado el momento (no en módulo pero sí en dirección y sentido). De este modo, $\Psi_{in}\approx 0$ porque está muy picada en torno a $\vec p_0$. Recordemos que preparamos el estado de entrada de la partícula de tal forma que está desplazado un parámetro de impacto del eje del centro de dispersión, de modo que tenemos $\Psi_{in}(\vec p')=e^{-i\vec \rho\cdot \vec p'}\phi(\vec p')$, donde $\phi(\vec p)$ son los paquetes de onda que hace nuestro acelerador. Entonces, tenemos que la función de onda lejos del target y en direcciones que no se encuentren en la vecindad del momento inicial, que en la práctica son las que observaremos, son:

$$\Psi_{out}(\vec p)=\frac{i}{2\pi m}\int d^3\vec p \ \delta(E_{p'}-E_p)f(\vec p'\leftarrow \vec p)e^{-i\vec \rho\cdot \vec p'}\phi(\vec p')$$

- Ahora que conocemos $\Psi_{out}(\vec p')$ podemos sustituirlo en la expresión de $P(d\Omega \leftarrow \phi_\rho)$ y a su vez esta en $\sigma(d\Omega \leftarrow \phi)$, de donde finalmente obtenemos la conocida expresión de la sección eficaz diferencial en términos de la amplitud de scattering, a saber:

$$\boxed{\frac{d\sigma}{d\Omega}(\vec p\leftarrow \vec p_0)=|f(\vec p\leftarrow \vec p_0)|^2}$$

**Nota:** véase que hasta ahora hemos escrito $\sigma(d\Omega\leftarrow \vec p_0)$. Sin embargo, como $\vec p$ apunta en la dirección de observación (define el eje del elemento infinitesimal de ángulo sólido en el que estamos observando), y como tiene el mismo módulo que el momento inicial: $|\vec p|=|\vec p_0|$, es lo mismo hablar de que una partícula ha sido dispersada en un elemento infinitesimal de ángulo sólido que decir que ha sido dispersada con un momento saliente $\vec p$. Como su módulo no cambia, no nos da más información que una dirección en el espacio con respecto al momento entrante, y por ende tiene sentido emplear la notación que se ha adoptado

# 5. Teorema Óptico

- La parte imaginaria de la amplitud de scattering hacia delante es directamente proporcional a la sección eficaz total del target:

$$\boxed{Im[f(\vec p\leftarrow \vec p)]=\frac{p}{4\pi}\sigma(\vec p)}$$

**Dem.:**

- Sea la descomposición habitual auxiliar de $S=I+R$. Como $S$ es unitario, tenemos que $S^\dagger S=I$, de modo que $R+R^\dagger = -R^\dagger R$
- Tomemos ahora elementos de matriz a ambos lados de la igualdad:

$$\langle \vec p'|R+R^\dagger|\vec p\rangle = -\langle \vec p'|R^\dagger R|\vec p\rangle$$

Recordemos que: $\langle \vec p'|R|\vec p\rangle=\frac{i}{2\pi m}f(\vec p'\leftarrow \vec p)\delta(E_{p'}-E_p)$

lo que nos permite calcular directamente el primer elemento del lado izquierdo. El segundo se hace tomando el complejo conjugado 2 veces. Para el lado derecho, insertamos la identidad en momentos entre los dos operadores, y tenemos:

$$\frac{i}{2\pi m}[f(\vec p'\leftarrow \vec p)-f^*(\vec p\leftarrow \vec p')]\delta(E_{p'}-E_p)=\frac{i}{2\pi m}\int d^3\vec p'' |f(\vec p''\leftarrow \vec p)|^2\delta(E_{p''}-E_{p'})\delta(E_{p''}-E_p)$$

Como en el lado derecho la primera delta condiciona la segunda, podemos sacar factor común de $\delta(E_{p'}-E_p)$. Si ahora hacemos $\vec p=\vec p'$, tenemos:

$$2iIm[f(\vec p\leftarrow \vec p)]=\frac{i}{2\pi m}\int d^3\vec p'' |f(\vec p''\leftarrow \vec p)|^2(2m)\delta(|\vec p''|^2-|\vec p|^2)$$

$$Im[f(\vec p\leftarrow \vec p)] = \frac{1}{2\pi}\int d^3\vec p'' |f(\vec p''\leftarrow \vec p)|^2\frac{1}{2p}\delta(|p''|-|p|)$$

$$Im[f(\vec p\leftarrow \vec p)] = \frac{1}{4\pi p}\int d\Omega\int_0^\infty p''^2dp'' \delta(p''-p)\frac{d\sigma}{d\Omega}$$

$$Im[f(\vec p\leftarrow \vec p)] = \frac{p}{4\pi}\sigma(\vec p)_\blacksquare$$

**Implicaciones:**

1. La amplitud de scattering no puede ser real en todo su dominio de definición (sabemos que al menos hacia delante tiene parte imaginaria. Asimismo, como admitimos que es suave, en un entorno del origen es asimismo compleja)
2. La amplitud de scattering hacia delante tiene parte imaginaria positiva
3. Por la definición de sección eficaz diferencial, podemos conocer el módulo de la amplitud de scattering de un proceso. Esto implica que podemos conocer la amplitud de scattering salvo una fase. Sin embargo, para la amplitud de scattering hacia delante, como el teorema óptico nos da la parte imaginaria y ya conocemos el módulo, podemos determinarla sin ambigüedad de fase (aunque la parte real podría ser positiva o negativa y no lo sabemos, pero yo entiendo que al ser una amplitud será positiva, aunque en general es compleja y esto no aplica)
5. Es un hecho experimental que la sección eficaz total en los procesos de scattering de partículas elementales tiende a ser constante a energías muy altas
6. El pico hacia delante se explica porque, por el teorema óptico, la sección eficaz diferencial vemos que ha de ir, como mínimo, como $p^2$
