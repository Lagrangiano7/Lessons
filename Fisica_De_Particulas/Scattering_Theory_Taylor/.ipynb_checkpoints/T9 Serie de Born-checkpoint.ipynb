# 1. La serie de Born

- Podemos coger la ecuación de Lippman-Schwinger e iterarla $\textit{ad infinitum}$, y esto se conoce como la serie de Born, a saber:

$$T(z)=\displaystyle\sum_{n=0}^\infty \left[VG_0(z)\right]^nV$$

- Entonces, ya tenemos una expresión en serie para la matriz T on shell sencillamente tomando los elementos de matriz en base de momentos y evaluando el límite on shell:

$$\boxed{t(\vec p'\leftarrow \vec p) = \displaystyle\sum_{n=0}^\infty\langle \vec p'|\left[VG_0(E_p+i0)\right]^nV|\vec p\rangle = \langle \vec p'|V|\vec p\rangle + \langle \vec p'|VG_0(E_p+i0)V|\vec p\rangle+\ldots}$$

**Teoremas de convergencia de la serie de Born:** (para potenciales centrales y que cumplan los 3 requisitos del Tema 1)

$$\boxed{\text{El potencial $V$ permite al menos un estado ligado $\Longrightarrow$ la serie de Born no converge para cualesquiera dos momentos}}$$

- Vemos que si el potencial es suficientemente débil como para no soportar estados ligados, la serie podría o no converger

**Criterio de convergencia a altas energías:**

- Para cualquier potencial $V$ existe una energía $\bar E$ para la que la serie de Born converge. Cuanto mayor sea la energía de la partícula (con $E>\bar E$, claro) tanto más rápido convergerá la serie, y a energías $E$ suficientemente altas, el primer término (aprox. de Born) domina}


**OJO!:** vemos que la serie de Born converge a altas energías, y que cuanto más alta sea la energía, mejor funciona. Sin embargo, hay que andar con cuidado, porque si la energía es suficientemente alta, empiezan a aparecer efectos relativistas (entre otros) que invalidan el uso de nuestra teoría y por ende de los resultados obtenidos con ella

# 2. La aproximación de Born

- Si estamos a energías suficientemente altas ($E>>\bar E$), entonces el término que domina en la serie de Born es el primero, y la amplitud de scattering es entonces la transformada de Fourier del potencial evaluada en la transferencia de momento $\vec q=\vec p'-\vec p$:

$$\boxed{f^{(1)}(\vec p'\leftarrow \vec p)=-\frac{m}{2\pi}\int d^3\vec x V(\vec x)e^{-i\vec q\cdot \vec x}}$$

**Nota:** como estamos en colisiones elásticas, $|\vec p'|=|\vec p|$, y por tanto:

$$|\vec q|=2|\vec p|\sin\frac{\theta}{2}$$

donde $\theta$ es el ángulo que forman $\vec p, \vec p'$

**Nota:** si el potencial es esférico, entonces la expresión se simplifica considerablemente:

$$\boxed{f^{(1)}(\vec p'\leftarrow \vec p)=-2m\int_0^\infty r^2\frac{\sin(qr)}{qr}dr}$$

# 3. Propiedades de la amplitud de Born

## 3.1 La amplitud hacia delante es independiente de la energía

- Ahora que sabemos calcular la amplitud de scattering como función de la transferencia de momento, podemos evaluar el caso particular en el que esta es nula: $f(\vec p\leftarrow \vec p)$. Esto da:

$$f(\vec p\leftarrow \vec p)=-\frac{m}{2\pi}\int V(\vec x)d^3\vec x$$

que es independiente de la energía (esto es una consecuencia natural de que la amplitud de Born dependa de la transferencia de momento en lugar de los momentos por separado, de modo que se cancela hacia delante y no hay mención a los momentos, y por ende tampoco a las energías)

**Nota:** este importante resultado implica que, si bien cada energía dará en general una sección eficaz diferencial en función de $\theta$ distinta, todas ellas tendrán el mismo valor en $\theta=0$ (por eso todas las curvas tienen un punto en común en el origen)

**Nota:** si en particular el potencial es esférico, recuperamos la regla de convergencia para $r\to \infty$ de la amplitud hacia delante para potenciales esféricos:

$$f(\vec p'\leftarrow \vec p)=-2m\int_0^\infty r^2V(r)dr$$

Y en general para que la integral de una función converja en infinito, dicha función ha de irse a 0 más rápido que $r^{-1}$ cuando $r\to \infty$. Por tanto, pedimos que $V(r)\sim r^{-3-\varepsilon}$ cuando $r\to \infty$

## 3.2 La amplitud de scattering a baja energía se vuelve isotrópica

- Si tomamos el límite en el que la partícula entra con energía nula, entonces obtenemos $f(\vec p'\leftarrow \vec p)=\int V(\vec x)d^3\vec x$ porque $\vec p=0$ y como estamos con scattering elástico, $\vec p'=0$ también. Esto implica que, a muy bajas energías, obtenemos el resultado de que la sección eficaz diferencial no depende del ángulo (las partículas se dispersan en todas direcciones por igual, o con más precisión, con igual probabilidad)

**OJO!:** esta propiedad es irrelevante porque la aprox. de Born no es válida a energías tan bajas

## 3.3 Cuanto más alta sea la energía, más picada estará $\frac{d\sigma}{d\Omega}(\theta)$

- Vemos que existe un término oscilante $e^{-i\vec q\cdot \vec x}$
- Cuanto mayor sea dicho término oscilante, tanto mayor será la supresión del integrando, y por ende más rápido se irá a 0

# 3.4 Aprox. de Born viola unicidad de $S$

- El teorema óptico no es sino una consecuencia directa de la unitariedad de $S$
- Vemos a continuación que la aprox. de Born viola el teorema óptico, de modo que viola asimismo la unitariedad de $S$

**Dem.:**

El teorema óptico nos da la parte imaginaria de la amplitud de scattering **hacia delante** de un proceso de scattering en términos (únicamente) de la sección eficaz total:

$$Im[f(\vec p\leftarrow \vec p)]=\frac{p}{4\pi}\sigma$$

Por otro lado, la aprox. de Born nos dice que la amplitud de scattering hacia delante ($\vec q=0$) viene dada por:

$$f(\vec p\leftarrow \vec p)=-\frac{m}{2\pi}\int V(\vec x)d^3\vec x \ \ \in \mathbb R; \ \ \forall \vec x\in \mathbb{R}^3$$

Vemos inmediatamente que la amplitud de scattering calculada con la aprox. de Born es una función real. Sin embargo, como siempre que haya scattering se podrá asociar al proceso una sección eficaz total no nula, vemos que la amplitud de scattering en efecto debería tener una parte imaginaria asimismo no nula. Por ende, la aprox. de Born viola el teorema óptico y con él la unitariedad de $S$. Sin embargo, en su rango de aplicabilidad, rompe la unitariedad de manera despreciable, de modo que podemos usarlo en buena aproximación$_\blacksquare$

# 3.5 Interpretación de la serie de Born con diagramas de Feynman

**Objetivo:** encontrar una expresión en serie de los elementos de matriz de $S$, esta vez en imagen de interacción, y encontrar un procedimiento gráfico que nos permita calcular el n-ésimo término de la serie

- En imagen de interacción, tanto los vectores de estado como los operadores dependen del tiempo, y lo hacen así:

$$|\Psi(t)\rangle_I = e^{iH_0t}|\Psi(t)\rangle_S$$
$$A_I(t)=e^{iH_0t}A_S(t)e^{-iH_0t}$$

donde el subíndice $I$ alude a la imagen de interacción (o de Dirac) y el subíndice $S$ a la de Schrödinger (recuerda que, en general, los operadores en imagen de Schrödinger pueden depender del tiempo, pero solo pueden hacerlo explícitamente. Los operadores canónicos no dependen del tiempo)

**Observación:** si el potencial es suficientemente débil, entonces el ket en imagen de interacción es constante. Esto quiere decir que, cuando el potencial no sea apreciable (cerca de los estados asintóticos) el ket será precisamente el estado asintótico de entrada o de salida, dependiendo de si la interacción ya ha tenido lugar o no. De este modo, tenemos que:


$$|\Psi_{-\infty}\rangle_I=|\Psi_{in}\rangle_S; \ \ |\Psi_{\infty}\rangle_I=|\Psi_{out}\rangle_S$$

Asimismo, el operador de Scattering, que en términos del operador de evolución temporal en imagen de Schrödinger toma la forma:

$$S=\lim_{\substack{t_1\to \infty \\ t_2\to-\infty}} U_0^\dagger(t_1)U(t_1)U^\dagger(t_2)U_0(t_2)$$

podemos expandirlo en exponenciales complejas para darnos cuenta de que queda de la forma:

$$\boxed{S=U_I(\infty, -\infty)}$$

donde $U_I(t,t_0)=e^{iH_0t}e^{-iH(t-t_0)}e^{-iH_0t_0}$

El problema se reduce entonces a encontrar una expresión del operador de evolución temporal en imagen de interacción, de modo que se ha simplificado considerablemente. Aplicamos el mismo razonamiento que la serie de Dyson (solo que ahora en lugar de estar en imagen de Schrödinger, estamos en imagen de interacción).

Sabemos que $U_I$ cumple la ecuación de operadores:

$$i\frac{dU_I}{dt}(t,t_0)=V_I(t)U_I(t,t_0)$$

De modo que ahora encontramos una solución formal integrando:

$$U_I(t,t_0)=I-i\int_{t_0}^tV_I(\tau)U_I(\tau,t_0)d\tau$$

Si ahora iteramos esta ecuación integral, tenemos:

$$U_I(t,t_0)=I-i\int V_I(\tau)d\tau + (-i)^2\int_{t_0}^t\int_{t_0}^{\tau'}V_I(\tau')V_I(\tau)d\tau+\ldots$$

Finalmente, para recuperar $U_I(\infty, -\infty)$, basta con tomar los límites correspondientes. De este modo, tomando elementos de matriz de estados de momento lineal bien definido: $\langle p'|S|\vec p\rangle$ (equivalentemente $\langle p'|U_I(\infty, -\infty)|\vec p\rangle$), tenemos que:

$$\langle\vec p'|S|\vec p\rangle = \delta^3(\vec p'-\vec p)-i\int_{-\infty}^\infty d\tau\langle \vec p'|e^{iE_{p'}\tau}Ve^{-iE_{p}\tau}|\vec p\rangle+(-i)^2\int_{-\infty}^\infty \int_{-\infty}^{\tau'}d\tau \langle \vec p'|e^{iE_{p'}\tau'}Ve^{-iH_0(\tau'-\tau)}Ve^{-iE_p\tau}|\vec p\rangle+\ldots$$

**Observación:** si simplificamos los términos, vemos que nos queda la serie de Born, como era de esperar (estamos calculando los elementos de matriz en momentos de $S$ y hemos dicho que la Física en ambas imágenes es la misma, esto es, los elementos de matriz de un mismo operador entre dos kets equivalentes entre imágenes dan el mismo resultado)

En cuanto al primer término, notamos que como $V$ (en imagen de Schrödinger) no depende del tiempo, podemos escribirlo de tal orma que nos queda una delta de Dirac en las energías:

$$-i\int_{-\infty}^\infty d\tau\langle \vec p'|e^{iE_{p'}\tau}Ve^{-iE_{p}\tau}|\vec p\rangle = -2\pi i\delta(E_{p'}-E_p)\langle \vec p'|V|\vec p\rangle$$

que es el primer término (no trivial) de la serie de Born

El segundo queda como sigue: primero integramos en los tiempos, donde introducimos el factor de damping (real) $-\varepsilon t$ en la exponencial compleja y admitimos teorema adiabático, de modo que la primitiva en $\tau\to-\infty$ se anula:

$$(-i)^2\int_{-\infty}^\infty \int_{-\infty}^{\tau'}d\tau \langle \vec p'|e^{iE_{p'}\tau'}Ve^{-iH_0(\tau'-\tau)}Ve^{-iE_p\tau}|\vec p\rangle=$$

$$(-i)^2\int d^3\vec k\langle \vec p'|V|\vec k\rangle \langle \vec k|V|\vec p\rangle \int_{-\infty}^\infty d\tau' \int_{-\infty}^{\tau'}e^{-i(E_p-E_k)\tau}e^{-i(E_k-E_{p'})\tau'}d\tau$$

Vemos que si integrásemos sobre $\tau$, no sabríamos qué daría la integral porque tendríamos exponenciales complejas evaluadas en límites infinitos, de modo que introducimos el factor de damping e invocamos el teorema adiabático. Tenemos entonces:

$$\displaystyle\lim_{\varepsilon\to 0^+}(-i)^2\int d^3\vec k\langle \vec p'|V|\vec k\rangle \langle \vec k|V|\vec p\rangle \int_{-\infty}^\infty d\tau' \int_{-\infty}^{\tau'}e^{-i(E_p-E_k)\tau}e^{-\varepsilon\tau}e^{-i(E_k-E_{p'})\tau'}d\tau$$

Integrando sobre $\tau$, queda entonces:

$$(-i)^2\int d^3\vec k\langle \vec p'|V|\vec k\rangle \langle \vec k|V|\vec p\rangle \int_{-\infty}^\infty d\tau' \frac{i}{E_p-E_k+i0}e^{i(E_{p'}-E_p)\tau'}d\tau'$$

Integramos ahora sobre $\tau'$ y notamos que queda una delta de Dirac en las energías:

$$(-i)^2\int d^3\vec k\langle \vec p'|V|\vec k\rangle \langle \vec k|V|\vec p\rangle  \frac{2\pi i}{E_p-E_k+i0}\delta (E_{p'}-E_p)$$

Como es natural, la delta solo comprende las energías "reales" de salida y de entrada, no las de los pasos intermedios. Ahora reorganizamos los términos de la integral restante, y ya nos podemos oler que son la expansión de los elementos de matriz del operador de Green libre on shell:

$$-2\pi i\delta (E_{p'}-E_p)\int d^3\vec k \frac{\langle \vec p'|V|\vec k\rangle \langle \vec k|V|\vec p\rangle }{E_p-E_k+i0}$$

De modo que finalmente queda:

$$-2\pi i\langle \vec p'|VG_0(E_p+i0)V|\vec p\rangle$$

Que es precisamente el elemento de matriz del segundo término no trivial de la serie de Born. Comprobamos nuevamente que el resultado al que hemos llegado en imagen de interacción es del todo equivalente al que se llega en imagen de Schrödinger.

**Interpretación del término a primer orden:**

El primer término no trivial, $-i\int_{-\infty}^\infty d\tau\langle p'|e^{iE_{p'}\tau}Ve^{-iE_{p}\tau}|\vec p\rangle$, se puede entender como sigue (leyendo de derecha a izquierda, en el orden que actúan los operadores sobre el primer ket): inicialmente, la partícula con momento bien definido $\vec p$ evoluciona libremente, hasta que en el instante $\tau$ se encuentra con cierto potencial. Este actúa instantáneamente sobre el estado de la partícula, provocando que cambie su momento a $\vec p'$, y finalmente la partícula, ahora con este nuevo momento, continúa evolucionando libremente). Para acabar, integramos sobre todos los instantes en los que sucede esta interacción con el potencial, porque "no sabemos cuándo sucede" (realmente sucede todo el tiempo))

Vemos entonces que, fijado el instante de la interacción, $\tau$, la interacción de la partícula con el potencial es instantánea y elástica, como el rebote de una pelota contra una pared.

**Interpretación del término a segundo orden y estados virtuales:**

Centrémonos ahora en el segundo término del desarrollo en serie de los elementos de matriz de $S$. Como no sabemos cómo actúa la exponencial compleja de $H_0$ sobre el potencial en imagen de Scrhödinger, introducimos un conjunto completo de partículas libres de momento $\vec k$, de tal forma que el segundo término queda:

$$(-i)^2\int_{-\infty}^\infty \int_{-\infty}^{\tau'}d\tau \int d^3\vec k e^{iE_{p'}\tau'} \langle \vec p'|V|\vec k\rangle e^{iE_k(\tau'-\tau)} \langle \vec k|V|\vec p\rangle e^{-iE_p\tau}$$

Tenemos entonces la siguiente interpretación de lo que sucede: la partícula entra inicialmente con momento $\vec p$, y evoluciona libremente hasta el instante $\tau$. Después, una primera interacción con el potencial $V$ la manda a un estado "intermedio" de momento $\vec k$, y continúa evolucionando libremente hasta el instante $\tau'$. En dicho instante, tiene lugar una segunda interacción con el potencial, de tal forma que este manda ahora la partícula a un estado con momento $\vec p'$. Finalmente, integramos sobre $\tau$ y sobre $\tau'$, de tal forma que consideramos todos los eventos donde las interacciones suceden en unos instantes determinados y los sumamos para obtener la amplitud de scattering a segundo orden

**Representación gráfica: diagramas de Feynman**

Podemos asociar diagramas a estas integrales. Hemos visto anteriormente que podemos entender los términos de orden superior como procesos en los que se da un número de interacciones entre la partícula y el potencial igual al orden del término que estamos calculando, y cada vez el potencial manda a la partícula a otro estado intermedio $|\vec k^{(r)}\rangle$ para finalmente acabar saliendo con el momento $\vec p'$ observado. **Podemos entender el término de orden r en el desarrollo en serie de la amplitud de scattering como aquel asociado al proceso en el que la partícula interacciona r veces con el potencial, pasando por r-1 estados intermedios no observables (virtuales).** Naturalmente, esto equivale a decir que introducimos la identidad en momentos r-1 veces porque en la serie de Born nos salen r-1 propagadores $G_0$ que sabemos cómo actúan en la base de momentos

Los diagramas de Feynman solo ponen esto de forma gráfica

**Reglas de los diagramas de Feynman (no relativistas):** (para sacar $t^{(r)}(\vec p'\leftarrow \vec p)$)

1. Dibújese el diagrama a orden requerido
2. Por cada vértice, póngase (multiplicando) un término $\langle \vec k_{n+1}|V|\vec k_{r}\rangle$
3. Por cada línea intermedia, escríbase el propagador (on shell) correspondiente $\frac{1}{E_p-E_{k_n}+i0}$
4. Intégrese para todos los momentos intermedios

**Ejemplo:** de este modo, el término $t^3(\vec p'\leftarrow \vec p)$ tendría 3 interacciones con el potencial, y por ende habría dos estados intermedios. Gráficamente:

![Diagrama de Feynman a tercer orden](Feynman_Diagram_Order3.jpg)

De acuerdo a las reglas que hemos derivado, tenemos entonces que el tercer término de la matriz T on shell para este proceso es:

$$t^{(3)}(\vec p'\leftarrow \vec p)=\int d^3\vec k_1\int d^3\vec k_2 \frac{\langle \vec p'|V|\vec k_2\rangle \langle \vec k_2|V|\vec k_1\rangle \langle \vec k_1|V|\vec p\rangle}{(E_p-E_{k_2}+i0)(E_p-E_{k_1}+i0)}$$

**OJO!:** los diagramas de Feynman no son sino una representación gráfica de la serie de Born, pero recuerda que esta no siempre tenía por qué converger. Por tanto, los diagramas de Feynman tampoco tienen por qué llevar siempre a un resultado convergente. Esto se llama "régimen no perturbativo", y hay que recurrir a otras herramientas que no sean la teoría de perturbaciones, como las relaciones de dispersión o el lattice QCD en el caso de dicha teoría
