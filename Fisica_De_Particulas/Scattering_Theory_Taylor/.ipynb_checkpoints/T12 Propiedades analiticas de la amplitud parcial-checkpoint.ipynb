# 1. Repaso: propiedades de las funciones de 1 variable compleja

**Def.:** diremos que una función de variable compleja es "analítica" en dicha variable en cierta región $R$ de su dominio si es diferenciable en todos los puntos de dicha región $R$

- Por ejemplo, $f(z)=1/z$ es analítica en todo el plano complejo salvo el origen ($\forall z \in \mathbb C\ \{0\}$)

## Teorema de Cauchy

- Sea una función analítica en una región $R$
- Entonces, cualquier integral de dicha función sobre un camino cerrado contenido en $R$ es 0:

$$\boxed{\text{$f(z)$ analítica en $R$} \Longrightarrow \oint_\gamma f(z)dz=0; \ \ \forall \gamma \subset R}$$

## Consecuencias del teorema

1. Una función analítica en $R$ es infinitamente diferenciable en dicha región (si es diferenciabl 1 vez, lo es infinitas veces)
2. La función analítica admite un desarrollo en serie (en general en serie de Laurent pero colapsa a Taylor) en torno a cualquier punto de $R$
3. Si la función es analítica salvo por un número finito de polos, entonces la integral cerrada por el camino $\gamma$ es $2\pi i$ por el residuo de la función en los polos contenidos por la curva

![Teorema de Cauchy, descripcion grafica](Cauchy.jpg)

4. Dos funciones analíticas que coinciden en cualquier segmento de linea han de coincidir en todas partes (en las que estén definidas). Si una tiene un dominio mayor que la otra, llamaremos a la grande la "extensión analítica única" de la primera. O bien por ejemplo, si tenemos que $f_1(z)$ es analítica en $R_1$ y $f_2(z)$ es analítica en $R_2$ y si $f_1(z)=f_2(z)$ en $R_1\cap R_2$, entonces una es la extensión analítica única de la otra en la región contraria y de forma efectiva lo que tenemos es una única función analítica $f(z)$ en todo $R=R_1\cup R_2$

5. Si $f(z)$ es analítica en una región $R$, en general su conjugado $f^*(z)$ no será analítico en dicha región. Sin embargo, la función $f^*(z^*)$ sí será analítica en la región $R^*$

## Principio de reflexión de Schwartz

- Sea una función analítica definida en el semiplano positivo. Si toma valores reales y no singulares sobre la recta real, entonces existe su continuación analítica al semiplano negativo y es:

$$\tilde f(z^*)=f^*(z); \ \ Im(z)\geq 0$$

Equivalentemente, si denotamos por $z$ un punto del semiplano negativo, entonces la extensión analítica a dicho plano queda de la forma:

$$\tilde f(z)=f^*(z^*); \ \ Im(z)\leq 0$$

**Nota:** en general se prescinde de la tilde para la continuación analítica de una función y toma su mismo nombre

**Comentario:** en el estudio de resonancias extendremos la variable momento lineal al plano complejo porque no deja de ser un parámetro en el cálculo de la función de onda reducida y de las subsecuente información que obtengamos de ella. La función de Jost es analítica en el semiplano superior ($Im(p)>0$) y real en el eje imaginario, no en el real. Por ende, no podemos aplicarle a esta función el principio de reflexión de Schwartz directamente, pero si hacemos el cambio de variable $p=ip'$, entonces ya cumple las condiciones del principio y tenemos:

$$\boxed{f(p)=f^*(-p^*)}$$

**Nota:** Muchas funciones de la teoría de scattering se dan como sumatorios o integrales de otras funciones, y nos interesa determinar la analiticidad de la función suma. He aquí un criterio de analiticidad para cada caso


## Serie de funciones analíticas 

- Sea el conjunto infinito de funciones $\{f^{(n)}(z)\}$, todas ellas analíticas en cierta región $R$
- Sea la serie

$$f(z)=\displaystyle\sum_{n=1}^\infty f^{(n)}(z)$$ uniformemente convergente en $R$


- Entonces la función $f(z)$ es analítica en $R$

**Nota:** entiendo que si la suma es finita, entonces $f(z)$ es analítica siempre

## Integral de una función analítica

- Sea la función $g(z, r): \ R\times [a,b]\to \mathbb C$ analítica en la variable $z$ e $R$ para cada valor fijo de $r$
- Sea la función de dos variables $g(z,r)$ continua en todo su dominio
- Entonces la función:

$$f(z)=\int_a^b g(z, r)dr$$

es analítica en todo $R$

**OJO!:** puede que los límites del intervalo en el que se encuentra $r$ sean infinitos

# 2. Propiedades analíticas de la solución regular

- Volvamos a la ecuación radial:

$$\left[\frac{d^2}{dr^2}-\frac{l(l+1)}{r^2}+2mV(r)+p^2\right]=0$$

donde ahora admitimos que $p$ es una variable compleja.

**Def.:** llamaremos "región física" al semieje positivo de la recta real de $p$ ($p\geq 0$), porque son los únicos valores con significado físico que puede tomar

**Teorema:** la ecuación radial tiene solución por iteración para cualquier valor complejo de $p$ y la solución $\phi_{l,p}(r)$ es una función entera de $p$

**Nota:** se dice de una función de variable compleja que es "entera" si es analítica en todo el plano complejo

## Propiedad de inversión de $p$

**Observación:** la ecuación radial presenta simetría bajo el cambio $p\to-p$, por lo que esperaríamos cierta relación entre las soluciones $\phi_{l,p}$ y $\phi_{l,-p}$. En efecto, la hay:

$$\boxed{\phi_{l,-p}(r)=(-1)^{l+1}\phi_{l,p}(r)}$$

# 3. Propiedades analíticas de la función de Jost y la matriz $S$

**Nota:** nos interesa estudiar las propiedades de la función de Jost porque los autovalores del operador de scattering $S$ en la base de momento angular (recuerda que siempre estamos bajo la suposición de que el sistema tiene simetría bajo rotaciones y que no tiene spin)

Partimos ahora de la definición integral de la función de Jost, donde hasta ahora admitíamos que el momento estaba en la región física. Sin embargo, podemos definir la continuación de $f_l(p)$ admitiendo que $p$ es complejo

1. $f_l(p)$ real en la región física ($p\geq 0$)
2. $f_l(p)$ continua en semiplano superior y recta real $Im [p]\geq 0$
3. $f_l(p)$ analítica en semiplano superior $Im[p]>0$

**Recuerda:** la analiticidad se define como diferenciabilidad, de modo que no podemos hablar de analiticidad en regiones cerradas, siempre tienen que ser abiertas

- En cuanto a $s_l(p)$, recordemos que se relacionaba con la función de Jost como:

$$s_l(p)=\frac{f_l^*(p)}{f_l(p)}$$

cuando $p$ estaba en la región física. Vemos que esta función así construida no es analítica, porque si bien $f_l(p)$ lo es, $f_l^*(p)$ no lo es. Entonces, aprovechando que $s_l(p)$ solo tiene sentido cuando el momento está en la región física, de modo que $p>0$ (real), podemos redefinir $s_l(p)$ de la forma:


$$\boxed{s_l(p)=\frac{f_l^*(p^*)}{f_l(p)}=\frac{f_l(-p)}{f_l(p)}}$$

**Nota:** ahora $s_l(p)$ es el cociente de dos funciones analíticas y esperaríamos que también lo fuera. Sin embargo, no lo es, porque el denominador hemos dicho que es analítico en $Im[p]>0$ y por ende el numerador es analítico en $Im[p]<0$. Esto es, las regiones de analiticidad de numerador y denominador no solapan, y por ende $s_l(p)$ así definido no es analítico en ninguna región del plano complejo de $p$

Si deseamos hacer el definir $s_l(p)$ analítico, está claro que debemos encontrar alguna forma de hacer que la función de Jost sea analítica también en el semiplano inferior, y esto se puede hacer, a cambio de imponer condiciones más restrictivas sobre el potencial en el infinito

## Potenciales truncados

**Resultado:** si el potencial se trunca idénticamente en algún $r=a$, entonces la integral de definición de la función de Jost no llega a $r\to\infty$, que es donde las funciones de Hankel dan problemas si extendemos $p$ al plano complejo, de modo que la integral es convergente y entera. Entonces, $s_l(p)$ será meromórfica (analítica en todo el plano complejo salvo en los ceros de su denominador, que darán polos). Sin embargo, los potenciales truncados rara vez son relevantes en casos reales

## Comportamiento asintótico exponencial

**Resultado:** si el potencial tiene comportamiento exponencial en $r\to\infty$, de la forma $O(e^{-\mu r})$ (como por ejemplo el de Yukawa), entonces $f_l(p)$ también será analítica en el semiplano inferior hasta $p=-\mu/2$. Por ende, $s_l(p)$ será meromórfica en $\mu/2 > Im[p]>-\mu/2$ (porque recuerda, es el cociente de una función de Jost con $-p$ y otra con $p$, asi que solo será analítica en la intersección de ambas regiones, que es la banda que decimos)

# 4. Amplitud parcial en el límite $p\to 0$

- Estudiemos ahora qué le sucede a la amplitud parcial de scattering $f_l(p)$ en el límite $p\to 0$

**Recuerda:** la amplitud parcial de scattering viene dada por:


$$f_l(p)=\frac{s_l(p)-1}{2ip}=\frac{1}{2ip}\frac{\mathbb f_l(-p)-\mathbb f_l(p)}{\mathbb f_l(p)}$$

- Naturalmente, solo podemos usar esta expresión en la región física ($p\geq 0$), pero como siempre podemos considerar su extensión al plano complejo
- Recuerda que si el potencial tiene tendencia asintótica exponencial, entonces existe una banda en la que la función de Jost es analítica y podemos por ende desarrollar en serie la amplitud de scattering en torno a $p=0$ para poder así estudiar su comportamiento a bajas energías

- Recordemos que la función de Jost se construía de forma integral como:

$$\mathcal f_l(p)=1-\frac{2m}{p}\int_0^\infty dr \ \hat{h_l}^+(pr)V(r)\phi_{l,p}(r)$$

- Introduzcamos ahora esta expresión para la función de Jost en la amplitud de scattering. Recordando que $\frac{\phi_{l,p}(r)}{\mathcal f_l(p)}=\Psi_{l,p}(r)$, tenemos que:

$$f_l(p)=\frac{2m}{p^2}\int_0^\infty dr \ \hat{j_l}(pr)V(r)\Psi_{l,p}(r)$$

**Acotación de la amplitud parcial:**

Como las funciones de Ricatti-Bessel y la función radial regular están acotadas por: $|\hat{j_l}(pr)|, |\phi_{l,p}(r)|\leq cte.\left(\frac{pr}{1+pr}\right)^{l+1}$, admitiendo que la función de Jost cumpla $\mathcal f_l(0)\neq 0$, tenemos entonces que la amplitud parcial de scattering en onda-l está acotada como sigue:


$$|f_l(p)|\leq \frac{cte.}{\mathcal f_l(0)}\frac{2m}{p^2}\int_0^\infty dr \ |V(r)|\left(\frac{pr}{1+pr}\right)^{2l+2}$$

en un entorno de $p=0$

## Recuperando el desarrollo en longitudes de scattering

### Potencial con tendencia asintótica exponencial

- Si el potencial tuviera en particular tendencia asintótica exponencial, entonces la acotación en un entorno del origen sería:

$$|f_l(p)|\leq \frac{cte.}{\mathcal f_l(0)}2mp^{2l}\int_0^\infty dr \ |V(r)|r^{2l+2}$$

- De este modo, deducimos que a bajas energías la amplitud parcial de scattering cumple $f_l(p)=O(p^{2l}) \ \ [p\to 0]$. Ah, pero como hemos dicho que en el caso de tener comportamiento asintótico exponencial es analítica en un entorno del origen (porque se abría la banda) entonces podemos hacer su desarrollo en serie en torno al origen, que claramente tendrá como primer término el de orden $p^{2l}$:

$$f_l(p)=-a_lp^{2l}+b_lp^{2l+1}+\ldots$$

- Este resultado ya era conocido hasta primer orden, pero ahora sabemos gracias a abrir $p$ al plano complejo que se trata solo del primer término de un desarrollo en serie

**Nota:** este resultado se cumple entonces si (1) el potencial tiene comportamiento asintótico exponencial y (2) si la función de Jost no se anula en el origen. Si sí lo hiciera, entonces el término a primer orden sería infinito (volaría porque hay una función de Jost en el denominador)

### Potencial con tendencia asintóticca $O(\frac{1}{r^\nu})$

- Tendríamos ahora que la tendencia de la amplitud parcial de scattering a baja energía sería:

$$f_l(p)=O(p^{\nu-3}); \ \ [p\to 0, l>(\nu-3)/2]$$

mientras que para los términos con $l\leq (\nu-3)/2$ se sigue cumpliendo la tendencia del anterior apartado

- Si deseamos hacer un desarrollo en serie, debemos primero concretar si $f_l(p)$ es analítica en un entorno del origen, ya que esto no está garantizado por el comportamiento del potencial. Por sus buenas propiedades, nos ceñiremos a los potenciales con tendencia asintótica exponencial

## Matriz $K$

- La expansión en potencias de $p$ de la amplitud de scattering resulta ser menos útil que la de la matriz $K$. Discutamos esta

**Def.:** se define la función $k_l(p)$ de la forma:

$$s_l(p)=\frac{1+ik_l(p)}{1-ik_l(p)}$$

Naturalmente, en la región física, por ser $S$ un operador unitario sus autovalores cumplen $|s_l(p)|=1$ y por ende $k_l(p)$ es real

- Despejando, tenemos que:

$$k_l(p)=i\frac{1-s_l(p)}{1+s_l(p)}=\tan\delta_l(p)$$

## Aproximación de rango efectivo

**Propiedad:** $k_l(p)$ es analítica en toda región donde $s_l(p)$ lo sea, salvo en aquellos puntos donde $s_l(p)=-1$, que dan polos. En partícular, es analítica en torno a $p=0$, de modo que podemos hacerle un desarrollo en serie en torno al origen

**Nota:** al hacer el cambio $p\to-p$, tenemos que: $s_l(-p)=1/s_l(p)$, porque se construye como $s_l(p)=\frac{\mathcal f_l(-p)}{\mathcal f_l(p)}$

**Propiedad:** $k_l(p)$ es una función impar en $p$ (se deduce a partir de la def. y de la propiedad de $s_l(p)$ que acabamos de ver

- Por ende, la expansión en potencias de $p$ en torno al origen de $k_l(p)$ tendrá únicamente potencias complejas. Veamos a partir de cuál empieza:
- Sabemos que $\delta_l\xrightarrow[p\to 0]{}-a_lp^{2l+1}$, que a su vez será muy pequeño. Por ende, $k_l(p)=\tan\delta_l(p)$ con $\delta_l(p)$ pequeño implica que $k_l(p)\xrightarrow[p\to0]{}\delta_l(p)$, cuya tendencia ya conocemos. Por tanto, tienen la misma tendencia en el origen. Tenemos entonces que:


$$k_l(p)\xrightarrow[p\to 0]{}-a_lp^{2l+1}$$

**Propiedad:** las funciones $k_l(p)/p^{2l+1}$ ó $p^{2l+1}/k_l(p)$ son (1) analíticas en torno al origen (salvo si $a_l$ no está acotado), (2) son funciones pares del momento y por ende pueden expandirse en potencias de $p^2$ y por ende de la energía

- Consideremos la función $p^{2l+1}/k_l(p)$ cerca del origen:

$$\boxed{p^{2l+1}/k_l(p)=-\frac{1}{a_l}+\frac{r_l}{2}p^2+O(p^4)}$$

- Esta expresión recibe el nombre de "expansión de rango efectivo" porque el parámetro $r_0$ ($r_l$ en onda-s) suele ser parecido al alcance del potencial

**Def.:** llamamos "aproximación de rango efectivo" a la expresión de arriba cuando solo retenemos los términos hasta primer orden de momento

- La aprox. de rango efectivo en onda-s ha sido uno de los métodos más importantes para parametrizar los datos de scattering a bajas energías, como el scattering n-pp por debajo de los $$

- Los potenciales cuyo comportamiento asintótico es de la forma $O(\frac{1}{r^\nu})$ propician que las funciones de scattering (Jost, amplitud parcial, $s_l(p)$, $k_l(p)$) suelen tener una singularidad en el origen y por ende en general no se pueden desarrollar en serie en torno a dicho punto
- Por ejemplo, la expansión en onda-s de la función $p^{2l+1}/k_l(p)$ para un potencial con cola $O(\frac{1}{r^4})$ es:

$$p/k_0(p)=p\text{ cotan } \delta_l(p)=-\frac{1}{a_0}+bp+cp^2\log p+O(p^2)$$

que, como se ve, no tiene desarrollo en serie de potencias de $p$
