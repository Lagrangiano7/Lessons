# 1. Operador densidad de un haz de partículas

- En un experimento, el haz de partículas saldrá en general parcialmente polarizado
- **Def.:** diremos que un haz de partículas con spin $s$ se encuentra totalmente polarizado si todas las partículas tienen el spin apuntando en la misma dirección. Diremos que está totalmente despolarizado si todos los posibles estados $|m\rangle$ están igualmente poblados. Diremos que está parcialmente polarizado si algunos estados se encuentran más poblados que otros
- En general, es buena aprox. usar la estadística de Boltzmann, de modo que siempre habrá estados ocupados más allá del fundamental, así que en general el haz estará parcialmente polarizadoOperador

- Si nos imaginamos un haz de electrones ($s=1/2$) en un haz despolarizado, nos imaginaríamos que sus spines apuntan en todas las direcciones del espacio, pero según nuestra definición, deberían apuntar la mitad de ellos en una dirección y la mitad de ellos en la dirección opuesta
- **Proposición:** estas dos concepciones de un haz despolarizado son equivalentes, de modo que un haz despolarizada en efecto se puede tratar conforme a nuestra definición

**Dem.:**

Si el haz se encuentra completamente despolarizado, no sabemos nada sobre él, de modo que su matriz densidad será $\rho = \frac{1}{2s+1}I$. Por otro lado, si nos imaginamos el haz como un conjunto de partículas cuyos spines apuntan cada uno en cierta dirección del espacio (distrbuidos de forma aleatoria y uniforme en las direcciones) entonces este sistema será invariante bajo rotaciones del spin: $R_s^\dagger \rho R_s=\rho$. Pero, por el lema de Schurr, las únicas matrices de dimensión $d=2s+1$ que conmutan con todas las matrices de rotación son aquellas proporcionales a la identidad, de modo que debe ser $\rho = cI$ y recuperamos la definición original. Por tanto, ambas concepciones de un haz despolarizado son equivalentes y podemos tratar dichos haces como un sistema donde todos los estados están poblados por igual$_\blacksquare$

**Recuerda:** para un sistema de $s=1/2$, como un haz de electrones, la matriz densidad tiene dimensión $2$ y puede por tanto expresarse en la base de la matriz identidad y las 3 matrices de Pauli, a saber:

$$\rho = \lambda(I+\vec r\cdot \vec \sigma)$$

donde se admite que el operador densidad podría estar no normalizado. Si deseamos que lo esté, entonces $\lambda=1/2$

- Si $|\vec r|=1$, el haz está completamente polarizado
- Si $|\vec r|=0$, el haz está completamente despolarizado
- Si $|\vec r|\in (0,1)$, el haz está parcialmente polarizado

**Nota:** podemos incorporar los grados de libertad correspondientes al espacio construyendo el operador densidad adecuado (si por ejemplo estamos en un autoestado del estado de movimiento, entonces $\rho = (|\phi\rangle\langle \phi|)\otimes \rho_s$), si bien en general el estado de movimiento y el de spin se encuentran mezclados en un experimento de scattering. No obstante, como ya hemos desarrollado un formalismo en el que las magnitudes mensurables no dependen del estado de movimiento de entrada si este está suficientemente picado en torno a cierto valor de momento lineal, no nos preocupa tanto la parte espacial, en tanto que podemos usar autoestados del momento lineal

- Por tanto, podemos hacer una descripción híbrida del sistema, en la que demos la parte espacial con kets de momento lineal y la parte de spin con un operador densidad

# 2. Operadores densidad de entrada y salida

- Hasta ahora, hemos tratado los estados de entrada y salida con kets, y hemos admitido que podíamos generar partículas con tercera componente de spin bien determinada. Sin embargo, en general el haz de partículas estará parcialmente polarizado, de modo que debemos hablar ahora de operadores densidad de entrada (el correspondiente al haz de entrada) y de salida, $\rho_{in}, \rho_{out}$

**Expresando $\rho_{out}$ en términos de $\rho_{in}$**

De la teoría anterior, sabemos que el spinor de salida se puede expresar en términos del spinor de entrada a través de la matriz de amplitudes de scattering de la forma:

$$\chi_{out}=F(\vec p'\leftarrow \vec p)\chi_{in}$$

**Recuerda:** si el spinor de entrada está normalizado, entonces el de salida no tendrá módulo unidad, sino que su módulo será la sección eficaz diferencial de que la partícula salga con momento $\vec p'$ (para todos los posibles estados de spin de salida) provisto que el estado de entrada sea $\chi_{in}$. Entonces, tenemos que el operador densidad de salida será:

$$\rho_{out}=\displaystyle\sum_ip_i|\chi_{out}^i\rangle \langle \chi_{out}^i|$$

$$\rho_{out}=\displaystyle\sum_ip_i F|\chi_{in}^i\rangle \langle \chi_{in}^i|F^\dagger$$

Podemos sacar $F$ de la sumatoria y tenemos entonces que:

$$\boxed{\rho_{out}=F(\vec p'\leftarrow \vec p)\rho_{in}F^\dagger(\vec p'\leftarrow \vec p)}$$

- Como $\chi_{out}$ no está normalizado, $\rho_{out}$ tampoco lo está, sino que será la sección eficaz diferencial de que el haz salga con momento $\vec p'$ (para todos los posibles estados de spin de salida) provisto que el operador densidad de entrada sea $\rho_{in}$:

$$\frac{d\sigma}{d\Omega}(\vec p'\leftarrow \vec p, \rho_{in})=Tr(\rho_{out})$$
