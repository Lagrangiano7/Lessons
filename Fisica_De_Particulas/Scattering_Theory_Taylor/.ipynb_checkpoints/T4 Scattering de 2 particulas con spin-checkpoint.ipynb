# 1. Factorización del espacio de Hilbert

**Clave:** el problema de scattering de 2 partículas con spin se extiende naturalmente del de 2 partículas sin spin, con la única diferencia de que ahora tendremos una amplitud por cada posible configuración final de spines

El espacio de Hilbert se descompone, igual que antes, como:

$$\mathcal H=\mathcal H_{CM}\otimes \mathcal H_{rel}$$

y el operador de scattering continúa siendo trivial en el subespacio del CM. Sin embargo, el subespacio $\mathcal H_{rel}$ tiene ahora una parte de spin, de modo que podemos ver el espacio de Hilbert completo como:

$$\mathcal H=\mathcal H_{CM}\otimes (\mathcal H_{rel,espacio}\otimes H_{S1,1}\otimes H_{S,2})$$

Los kets tendrán ahora la forma:

$$|\vec P_{CM}\rangle \otimes|\vec p_{rel}, \xi\rangle$$

con $|\xi\rangle = \displaystyle\sum_{m,m'}\chi_{m,m'}|m,m'\rangle$, que también se puede escribir con spinores (matrices columna de dimensión $(2s_1+1)(2s_2+1)$)

**Interpretación del problema:** el problema de scattering de 2 partículas con spin se reduce al problema de scattering de 1 partícula de spin $s_1$ por un target fijo de spin $s_2$

# 2. Elementos de matriz de $S$

Si ahora tomamos elementos de matriz en $S$, veremos que obtenemos lo mismo que en el caso de 2 partículas sin spin, con la única diferencia de que el posible cambio de spines (originado por el potencial de interacción) está tenido en cuenta en la matriz $T$ on shell, de modo que:

$$\langle \vec p_1', \vec p_2', \xi|S|\vec p_1, \vec p_2, \xi\rangle=\delta^3(\vec p_1'-\vec p_1)\delta^3(\vec p_2'-\vec p_2)\delta_{\xi\xi'}-2\pi i \delta(\sum E_i'-\sum E_i)\delta^3(\sum \vec p_i'-\sum\vec p_i) \ t(\vec p', \xi'\leftarrow\vec p, \xi)$$

**Interpretación:** el primer término, como siempre, hace referencia a la amplitud de que las partículas no se dispersen. En este caso, como no interaccionan, se conserva el momento total (y por tanto la energía), así como los estados de spin (es lógico que, si las partículas no interactúan entre sí, sus estados de spin, al igual que sucede con sus estados de movimiento, se mantengan inalterados). El segundo término es la amplitud de que las partículas sí se dispersen, y es el mismo que en el caso de 2 partículas sin spin, con la diferencia de que ahora el elemento de matriz de $T$ on shell depende de los estados de spin inicial y final de las partículas, que en general son infinitos porque pueden ser superposiciones de los estados de la base de spin

**Proposición:** se puede calcular la sección eficaz diferencial en el CM para cualesquiera estados iniciales y finales de spin, $|\chi\rangle$  y $|\chi'\rangle$ a partir de los elementos de la base de spin de la forma:

$$\frac{d\sigma}{d\Omega}(\vec p_{rel}', \chi'\leftarrow \vec p_{rel}, \chi)=\left|\displaystyle\sum_{\xi, \xi'}\chi_{\xi'}'^*f(\vec p_{rel}', \xi' \leftarrow \vec p_{rel}, \xi)\chi_{\xi}\right|^2$$

donde $|\chi\rangle = \displaystyle\sum_\xi\chi_\xi|\xi\rangle$ y $\{|\xi\rangle\}$ son los elementos de la base de spines (que puede ser base desacoplada o base acoplada, da igual)

Esta igualdad se puede escribir de manera mucho más compacta empleando spinores y definiendo la **matriz de amplitudes**

$$\boxed{F(\vec p_{rel}'\leftarrow \vec p_{rel})=(f_{\xi,\xi'}(\vec p_{rel}'\leftarrow \vec p_{rel}))}$$

donde los elementos de matriz son $f_{\xi,\xi'}(\vec p_{rel}'\leftarrow \vec p_{rel})=f(\vec p_{rel}', \xi' \leftarrow \vec p_{rel}, \xi)$

de modo que la sección eficaz diferencial para cualesquiera estados inicial y final de spin de las dos partículas queda:

$$\boxed{\frac{d\sigma}{d\Omega}(\vec p_{rel}', \chi'\leftarrow \vec p_{rel}, \chi)\left|\chi'^\dagger F(\vec p_{rel}'\leftarrow \vec p_{rel})\chi\right|^2}$$

**Nota:** $dim(F)=[(2s_1+1)(2s_2+1)]^2$ (porque el número de elementos de la base de spines es $(2s_1+1)(2s_2+1)$, así que el número total de combinaciones que podemos hacer con ellas es precisamente ese número al cuadrado. Hablo de combinaciones porque es lo que estamos haciendo al calcular la sección eficaz diferencial de estados de spin arbitrarios, sumar sobre todas las combinaciones posibles de los estados de la base de spines)

**Ejemplo:** estudiemos el caso en el que la primera partícula tiene $s_1=1/2$ y la segunda $s_2=0$. Entonces, los posibles resultados son:

$$(1/2, 0)\to (1/2, 0)$$
$$(1/2, 0)\to (-1/2, 0)$$
$$(-1/2, 0)\to (1/2, 0)$$
$$(-1/2, 0)\to (-1/2, 0)$$

de modo que la matriz de amplitudes quedará de la forma:

$$F=\begin{pmatrix}f_{++}(\vec p_{rel}'\leftarrow \vec p_{rel}) & f_{+-}(\vec p_{rel}'\leftarrow \vec p_{rel}) \\ f_{-+}(\vec p_{rel}'\leftarrow \vec p_{rel}) & f_{--}(\vec p_{rel}'\leftarrow \vec p_{rel})\end{pmatrix}$$

donde los elementos de la diagonal principal se llaman "amplitudes spin-nonflip" y los que están fuera de esta se llaman "amplitudes spin-flip"

# 3. Suma y promedio sobre spines

- En la práctica, los haces de partículas las emiten despolarizadas (cada estado de spin de las partículas en los haces tiene la misma población) y los detectores que se usan no suelen ser sensibles al spin de las partículas. Por tanto, nos interesa ahora saber calcular la sección eficaz diferencial que nos medirán dichos detectores

- En general, como el haz incidente está despolarizado, deberíamos usar formalismo de matrices densidad. Sin embargo, en casos sencillos, es más fácil proceder como se va a explicar ahora

**Promedio sobre spines de entrada:**

- Sea un experimento de scattering de 2 partículas con spin
- Consideremos que tanto el haz de partículas de tipo 1 (con spin $s_1$) como las de tipo 2 (con spin $s_2$) están despolarizados
- El número de cuentas que registre nuestro detector para el evento de momento relativo de entrada $\vec p_{rel}$ y de salida $\vec p_{rel}'$ en los que el estado de entrada de spin sea $\xi=(m_1, m_2)$ será:

$$N_{sc}(\xi, \xi')=n_{inc}\frac{1}{(2s_1+1)(2s_2+1)}\sigma(\vec p_{rel}',\xi'\leftarrow \vec p_{rel}, \xi)$$

- Esto es esencialmente la probabilidad condicionada de que una de las partículas de tipo 1 con tercera componente de spin $m_1$ interaccione con otra partícula de tipo 2 con tercera componente de spin $m_2$. Entonces, el número de cuentas totales con cierto estado final de spin $\xi'$ será la suma de todos los eventos con cierto estado de spin de entrada:

$$N_{sc}(\xi')=n_{inc}\frac{1}{(2s_1+1)(2s_2+1)}\displaystyle\sum_{\xi}\sigma(\vec p_{rel}', \xi'\leftarrow \vec p_{rel}, \xi)$$

y vemos que, efectivamente, queda como un promedio sobre los estados iniciales. Podemos hacer la identificación $\sigma(d\Omega)=\frac{d\sigma}{d\Omega}d\Omega$ si el momento de entrada está muy picado en torno a algún valor, como se presupone que lo está, y tenemos entonces que:

$$\frac{d\sigma}{d\Omega}(\vec p_{rel}', \xi'\leftarrow \vec p_{rel})=\frac{1}{(2s_1+1)(2s_2+1)}\displaystyle\sum_{\xi}\frac{d\sigma}{d\Omega}(\vec p_{rel}', \xi'\leftarrow \vec p_{rel}, \xi')$$

- Esta es la contribución a la sección eficaz diferencial de todos los eventos que acaban con cierto estado de spin $\xi'$. Como nuestros detectores no distinguen spin, contarán todos los eventos que les lleguen, indistintamente de su estado de spin, de modo que debemos sumar sobre todos los estados de spin de salida, y tenemos entonces que la sección eficaz diferencial que medirán nuestros detectores será:

$$\boxed{\frac{d\sigma}{d\Omega}(\vec p_{rel}', \leftarrow \vec p_{rel})=\frac{1}{(2s_1+1)(2s_1+1)}\displaystyle\sum_{\xi, \xi'}\frac{d\sigma}{d\Omega}(\vec p_{rel}', \xi'\leftarrow \vec p_{rel}, \xi)}$$

donde, recuerda, $\frac{d\sigma}{d\Omega}(\vec p_{rel}', \xi'\leftarrow \vec p_{rel}, \xi)=|f_{\xi, \xi'}(\vec p_{rel}', \vec p_{rel})|^2$, que son los elementos de matriz de la matriz de scattering

**Recuerda:** esta sección eficaz diferencial, que será la que midan nuestros aparatos, es la que se obtiene cuando los haces están despolarizados y los detectores no discriminan estados de spin. Es por esto que recibe el nombre de "sección eficaz diferencial despolarizada"

**Ejemplo:** para el caso de scattering de la partícula de $s_1=1/2$ con otra de $s_2=0$, tendremos entonces que su sección eficaz diferencial despolarizada será:

$$\frac{d\sigma}{d\Omega}(\vec p_{rel}'\leftarrow \vec p_{rel})=\frac{1}{2}(|f_{++}|^2+|f_{+-}|^2+|f_{-+}|^2+|f_{--}|^2)$$

# 4. Reinterpretación de la matriz de scattering

- La matriz de scattering se puede entender como la restricción del operador de scattering al subespacio de los estados de spin. Por tanto, cabría esperar que su acción sobre cierto estado de spin de entrada nos dé el de salida (para cierta combinación de momentos de entrada y de salida), y así es:

$$\chi_{out}=F(\vec p_{rel}'\leftarrow \vec p_{rel})\chi_{in}$$

**OJO!:** nos lo da sin normalizar. De hecho, su normalización es precisamente la sección eficaz diferencial para un cierto estado de spin de entrada cuando los de salida no se monitorizan (suma)

$$||\chi_{out}||^2=\displaystyle\sum_{\xi'}|\chi_{\xi'}^{out}|^2 = \displaystyle\sum_{\xi'}\left|\displaystyle\sum_{\xi}f_{\xi\xi'}(\vec p_{rel}'\leftarrow \vec p_{rel})\chi_{\xi}^{in}\right|^2$$

donde identificamos el término dentro del módulo al cuadrado como la sección eficaz diferencial para cierto estado de spin de entrada arbitrario y cierto estado de spin de salida (de la base), de modo que:

$$||\chi_{out}||^2=\displaystyle\sum_{\xi'}\frac{d\sigma}{d\Omega}(\vec p_{rel}', \xi'\leftarrow \vec p_{rel}, \chi_{in})=\frac{d\sigma}{d\Omega}(\vec p_{rel}'\leftarrow \vec p_{rel}, \chi_{in})$$

**Dem.:**

- Admitamos que es cierto que el spinor de salida se puede escribir como la acción de la matriz de scattering sobre el spinor de entrada. Entonces, la sección eficaz diferencial queda de la forma:

$$\frac{d\sigma}{d\Omega}(\vec p_{rel}', \chi_{out}\leftarrow \vec p_{rel}, \chi_{in})=\left|\chi_{out}^\dagger F(\vec p_{rel}', \vec p_{rel})\chi_{in}\right|^2=||\chi_{out}||^2$$

que nos da la probabilidad de que el sistema salga con dicho estado de spin$_\blacksquare$
