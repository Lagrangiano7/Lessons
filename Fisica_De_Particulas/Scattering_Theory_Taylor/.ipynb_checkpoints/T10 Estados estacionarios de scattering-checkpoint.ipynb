# 1. Definición y propiedades

- Cuando estudiamos el estado del sistema en el instante $t=0$, podemos recuperarlo a partir de los estados asintóticos con el operador de Moller correspondiente:

$$|\Psi(t=0)\rangle = \Omega_+|\phi_{in}\rangle$$

- Expandamos el estado asintótico de entrada en la base de momentos:

$$|\phi_{in}\rangle = \int d^3\vec p\phi_{in}(\vec p)|\vec p\rangle$$

- Ahora expandamos el estado en $t=0$ en la base de momentos:

$$|\Psi(t=0)\rangle = \Omega_+|\phi_{in}\rangle = \int d^3\vec p \phi_{in}(\vec p)|\vec p_+\rangle$$

**Observación:** parece que la base natural para expandir el vector de estado "actual" del sistema es el de $|\vec p_\pm\rangle$. Por ende, cabe definir:

$$\boxed{|\vec p_\pm \rangle = \Omega_\pm |\vec p\rangle}$$

Llamaremos a estos estados impropios los **estados estacionarios de scattering.** A continuación vemos por qué reciben este nombre

## 1.1 Son autoestados del Hamiltoniano completo

**Proposición:** los estados estacionarios de scattering son autoestados del Hamiltoniano completo con autovalor $E_p$ ($=E_{p'}$)

**Dem.:**

Recordemos de temas anteriores la importante relación de conmutación entre el Hamiltoniano libre, el completo y los operadores de Moller:

$$H\Omega_\pm = \Omega_\pm H_0$$

Hagamos actuar ahora el Hamiltoniano completo $H$ sobre los estados estacionarios de scattering para ver si efectivamente son autoestados de dicho Hamiltoniano:

$$H|\vec p_\pm \rangle = H\Omega_\pm |\vec p\rangle = \Omega_\pm H_0|\vec p\rangle = E_p|\vec p_\pm \rangle$$

$$\therefore H|\vec p_\pm \rangle = E_p|\vec p_\pm \rangle_\blacksquare$$

**Conclusión:** los estados estacionarios de scattering se llaman así porque son autoestados del Hamiltoniano completo. Se puede demostrar que tanto $|\vec p_+\rangle$ como $|\vec p_-\rangle$ conforman bases ortogonales de autoestados del subespacio de scattering (recuerda, $\mathcal H = \mathcal R\oplus \mathcal B$) y tienen como autovalor la energía de la partícula

Podemos expandir la identidad (que toma validez solo cuando se considera todo el espacio de Hilbert, porque es la suma de todos los proyetores) de la forma:

$$I=\int d^3 \vec p|\vec p\rangle \langle \vec p|$$

o bien

$$I=\int d^3 \vec p|\vec p_+\rangle \langle \vec p_+|+\displaystyle\sum_n|n\rangle \langle n|$$

o bien

$$I=\int d^3 \vec p|\vec p_-\rangle \langle \vec p_-|+\displaystyle\sum_n|n\rangle \langle n|$$


**Nota sobre la condición asintótica:** podemos expandir el vector de estado de una partícula en el instante $t=0$ en la base de estados estacionarios de scattering como hemos visto. Ademas, se cumple la condición asintótica para cualquier estado normalizable: $U(t)|\phi_+\rangle\xrightarrow[t\to-\infty]{}U_0(t)|\phi\rangle$. Sin embargo, en el caso de los autoestados de scattering, como estos son autoestados tanto del Hamiltoniano complto, esta tendencia a la asíntota de entrada no se sostiene:

$$U(t)|\vec p_\pm\rangle = e^{-iE_pt}|p_\pm \rangle$$

$$U_0(t)|\vec p\rangle = e^{-iE_pt}|p \rangle$$

$$\therefore U(t)|\vec p_+\rangle !\xrightarrow[t\to-\infty]{}U_0(t)|\vec p\rangle$$

Con todo, esto no nos debería alarmar, porque estos estados son impropios y por ende no representan el estado de ningún sistema físico. Encontramos que cuando construimos paquetes de ondas, la tendencia sí se sostiene, de modo que el significado físico sigue sosteniéndose. Sabemos esto por el lema de Riemann-Lebesgue, que establece que si una función es tal que $f\in L(\mathbb R^n)$, entonces su transformada de Fourier tiende a 0 cuando su nueva variable tiende a +/- infinito: $\hat f(\vec \xi)=\int_{\mathbb R^n}f(\vec x)e^{-i\vec \xi\cdot \vec x}d^n\vec x \xrightarrow[||\vec \xi||\to\infty]{}0$. De este modo, los paquetes de onda cumplen la condición asintótica:

$$\int d^3\vec p\phi_{in}(\vec p)[U(t)|\vec p_\pm \rangle]\to \int d^3\vec p\phi_{in}(\vec p)[U_0(t)|\vec p_\pm \rangle]$$

# 2. Ecuación de los estados estacionarios de scattering

Deseamos ahora encontrar una ecuación que nos permita calcular los estados estacionarios de scattering. En el Tema 8 ya hemos demostrado que estos se pueden expresar en términos de los estados de momento lineal de la forma:

$$|\vec \phi_\pm \rangle = |\phi\rangle + \int d^3\vec pG(E_p\pm i0)V|\vec p\rangle \langle \vec p|$$

Si ahora expandimos como hemos aprendido el ket $|\phi_\pm \rangle$ (y $|\vec \phi\rangle$) en la base de momentos, tenemos que:

$$\int d^3\vec p\phi(\vec p)|\vec p_\pm \rangle = \int d^3\vec p\phi(\vec p)[|\vec p\rangle + G(E_p\pm i0)V|\vec p\rangle]$$

De donde encontramos la igualdad:

$$\boxed{|\vec p_\pm \rangle = |\vec p\rangle + G(E_p\pm i0)V|\vec p\rangle}$$

**OJO!:** esta igualdad solo se sostiene en el sentido en el que se integra sobre todos los momentos pesada por la función de onda $\phi(\vec p)$. Sin embargo, se puede demostrar que esta igualdad se sostiene cuando trabajamos con funciones de onda, de modo que es todo lo que necesitamos para empezar a trabajar

## 2.1 Expresión para la matriz T on shell

Sabemos de la ecuación de Lippman-Schwinger para el operador $T$ expresar este en términos del propagador (completo). De este modo, haciéndolo actuar sobre los autoestados de momento, tendremos que:

$$T(E_p\pm i0)|\vec p \rangle = [V+VG(E_p\pm i0)V]|\vec p\rangle = V[I+G(E_p\pm i0)V]|\vec p\rangle$$

pero esta es precisamente la ecuación de los autoestados de scattering (por un potencial), de modo que la acción del operador T sobre los autoestados de momento cuando está on shell es dar la acción del potencial sobre los autoestados de scattering:

$$\boxed{T(E_p\pm i0)|\vec p\rangle = V|\vec p_\pm \rangle}$$

Haciendo ahora actuar por la izquierda el bra $\langle \vec p'|$, conseguimos encerrar el operador $T$ on shell entre dos kets de momento lineal y tenemos la matriz T on shell (recuerda que debemos acercarnos por encima de la recta real, así que tomamos $+i0$ y por ende el ket $|\vec p_+\rangle$):

$$\boxed{t(\vec p'\leftarrow \vec p)=\langle \vec p'|V|\vec p_+\rangle}$$

Si deseamos tener la matriz T on shell en términos de $|\vec p_-\rangle$, basta con recordar la propiedad del operador $T$: $T^\dagger(z)=T(z^*)$, de modo que transponemos la primera expresión que hemos hallado y tenemos:

$$T(E_p-i0)|\vec p\rangle = V|\vec p_-\rangle \Longrightarrow \langle \vec p|T(E_p+i0) = \langle \vec p_-|V$$

Actuando este resultado sobre el ket $|\vec p'\rangle$ tenemos que:

$$\langle \vec p|T(E_p+i0) = \langle \vec p_-|V$$

Si finalmente intercambiamos $|\vec p\rangle, |\vec p'\rangle$ para dejar el resultado como siempre (solo es un cambio de etiqueta) tenemos entonces que:

$$\boxed{t(\vec p'\leftarrow \vec p)=\langle \vec p_-|V|\vec p\rangle}$$

## 2.2 Expresión de autoestados de scattering con propagador libre

- En la expresión que hemos dado al inicio, tenemos los autoestados de scattering expresados en términos de los autoestados de momento lineal a partir del propagador completo. Sin embargo, desearíamos expresarlos en términos del propagador libre, porque ese sí sabemos cómo actúa. Podemos hacer esto, pero pagando el precio de que nos queda una ecuación recursiva: sustituimos $GV$ por $G_0T$ y aplicamos la acción de $T$ sobre los autoestados de momento lineal, ahora conocida, de modo que nos queda:

$$\boxed{|\vec p_\pm \rangle = |\vec p\rangle + G_0(E_p\pm i0)V|\vec p_\pm \rangle}$$

- Esta ecuación recibe el nombre de **ecuación de Lippman-Schwinger para los autoestados de scattering**

## 2.3 Recuperando la serie de Born con autoestados de scattering

- Ahora que tenemos una ecuación (recursiva) para los autoestados de scattering, podemos resolverla iterando para hallar un desarrollo en serie de la solución y después usarla para calcular la matriz T on shell:

$$|\vec p_\pm \rangle = |\vec p\rangle +G_0(E_p\pm i0)V|\vec p\rangle + \ldots $$

Introduciendo esta solución en la expresión de $t(\vec p'\leftarrow \vec p)$ tenemos que:

$$t(\vec p'\leftarrow \vec p)= \langle \vec p'|V|\vec p\rangle + \langle \vec p'|VG_0(E_p+i0)V|\vec p\rangle +\ldots$$


y recuperamos la serie de Born, como era de esperar

**Reinterpretación de la aprox. de Born:** vemos de la ecuación de Lippman-Schwinger para los autooestados de scattering que estoss se construyen como un estado de momento lineal bien definido $|\vec p\rangle$ más una "distorsión" $G_0(E_p\pm i0)V|\vec p_\pm \rangle$. De este modo, la aprox. de Born consiste en despreciar la distorsión y tomar $|\vec p_+ \rangle \simeq |\vec p\rangle$

# 3. Funciones de onda de los estados estacionarios de scattering

- Se puede demostrar que, para mediciones a suficiente distancia del centro de scattering, la función de onda de los estados estacionarios de scattering adopta la forma de una partícula libre en superposición con una onda esférica cuyo origen se encuentra precisamente en el centro de scattering:

$$\langle \vec x|\vec p_\pm \rangle \xrightarrow[r\to \infty]{} (2\pi)^{-3/2}\left[e^{i\vec p\cdot \vec x}+f(p\hat x\leftarrow \vec p)\frac{e^{ipr}}{r}\right]$$

**Nota:** comprobamos que obtenemos el mismo resultado que cuando aplicábamos la teoría dependiente del tiempo y calculábamos la función de onda a gran distancia del centro de dispersión

**Nota:** fíjate que la parte esférica de la función de onda del estado estacionario de scattering $|\vec p_\pm\rangle$ sale hacia fuera. En el caso de $|\vec p_-\rangle$, obtendríamos una onda colapsando sobre sí misma, pero esto no tiene relevancia física
