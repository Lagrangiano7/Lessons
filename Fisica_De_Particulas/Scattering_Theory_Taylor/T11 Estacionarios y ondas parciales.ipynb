**Nota:** en este capítulo continuamos aprendiendo sobre los estados estacionarios de scattering, pero particularizamos el potencial a uno con simetría esférica (recuerda, si hubiera spin, simetría bajo rotaciones del tipo más general, que engloba espacial y de spin pero con el mismo ángulo). Hablamos del desfasaje

# 1. Función de onda libre

- Sabemos que una partícula libre tiene función de onda en representación de posiciones $e^{i\vec p\cdot \vec x}$
- Cuando el sistema exibe simetría esférica, es conveniente trabajar en la base de momento angular (y energía). Las funciones de onda libres deben entonces ir por fuerza como armónicos esféricos en la parte angular y como cierta función radial. Introduciendo este ansnatz en la ecuación independiente del tiempo de Schrödinger, queda que la parte radial obedece la ecuación:

$$\left[\frac{d^2}{dr^2}-\frac{l(l+1)}{r^2}+p^2\right]y(r)=0$$

Por ser de orden 2 y lineal, esta EDO tiene 2 familias independientes de soluciones: las funciones de Ricatti-Bessel y las funciones de Neumann. La función de onda libre en esta base vendrá dada por la superposición de estas ondas con momento angular bien definido:

$$\boxed{\langle \vec x|\vec p\rangle = (2\pi)^{-3/2}\frac{1}{pr}\displaystyle\sum_l(2l+1)i^l \  \hat {j_l}(pr)P_l(\hat x\cdot \hat p)}$$

donde $\hat {j_l}(pr)$ son las funciones de Ricatti-Bessel. Se comportan como $r^{l+1}$ cerca del origen y como $\sin(pr-l\pi/2)$ lejos de él. Si no consideramos el origen en nuestro problema, entonces tendríamos que meter también las funciones de Neumann (que van como $r^{-l}$ cerca del origen), o en general la superposición de ambos tipos de funciones, que son las funciones de Hankel

# 2. Función de onda con potencial

**Def.:** de la misma forma que definíamos los estados estacionarios de scattering $|\vec p_\pm \rangle$, definimos su análogo en la base de momento angular:

$$|Elm+\rangle = \Omega_+|Elm\rangle$$

le damos esta etiqueta a los estados estacionarios del Hamiltoniano completo, $H$

**Observación:** como (por hipótesis) estos estados son autoestados de $H$, tienen energía bien definida $E$. Asimismo, como $[\Omega_+, L_i]=0$, tenemos que también son autoestados del momento angular y por ende lo son de $L^2$ y de $L_3$. Es por esto que la parte angular de sus funciones de onda seguirán siendo armónicos esféricos (porque tienen momento angular bien definido).

Introduciendo nuevamente el ansnatz en la ecuación de Schrödinger, ahora con potencial, tenemos que la parte radial de esta es ahora:

$$\left[\frac{d^2}{dr^2}-\frac{l(l+1)}{r^2}-2mV(r)+p^2\right]y(r)=0$$

**Observación:** se trata de la misma ecuación que para la partícula libre, solo que ahora aparece además el potencial. Por ende, las funciones de onda serán como las libres, pero la parte radial será distinta:


$$\langle \vec x|Elm+\rangle = i^l\sqrt{\frac{2m}{\pi p}}\frac{1}{r}\Psi_{l,p}(r)Y_l^m(\hat x)$$

**Nota:** la condición de contorno es que $\Psi_{l,p}(r=0)=0$

## Forma asintótica de $\Psi_{l,p}(r)$

- Deseamos ahora conocer el comportamiento asintótico de la parte radial de las ondas dispersadas en el infinito, porque este está estrechamente ligado al desfasaje de dichas ondas

- Si comparamos la función de onda en representación de posiciones de $|\vec p_+\rangle$ que ya sabemos (en el infinito) con el desarrollo en funciones radiales, vemos que la onda esférica se acaba atenuando completamente y que solo queda la onda plana incidente, que va como una partícula libre y por ende con una función de Ricatti-Bessel, que en el infinito a su vez va como un seno. No obstante, tiene un desfase introducido por el potencial. Por ende, la parte radial de la función de onda de la partícula que está siendo dispersada tiene el comportamiento asintótico:

$$\boxed{\Psi_{l,p}(r)\xrightarrow[r\to\infty]{} e^{i\delta_l(p)} \sin[pr-l\pi/2+\delta_l(p)] }$$

**Observación:** se puede encontrar el desfasaje en onda $l$ sencillamente comparando la función de onda de una partículla libre con la de nuestra partícula dispersada

**Observación:** $\delta_l>0$ si el potencial es atractivo y $\delta_l<0$ si es repulsivo. Véase que la barrera centrífuga siempre hace las veces de potencial repulsivo

**Nota:** el hecho de que la única acción tangible del potencial sea la de desfasar la función de onda en lugar, por ejemplo, de cambiar su módulo, es que estamos en scattering de 1 solo canal. Esto es, no permitimos que se creen ni se destruyan partículas

# 3. Ecuación de Lippman-Schwinger para la parte radial

- Sabemos que, cuando el potencial tiene simetría esférica, la amplitud de scattering se puede desarrollar en ondas parciales de la forma:

$$f(\vec p'\leftarrow \vec p)=\displaystyle\sum_l(2l+1)f_l(p)P_l(\hat p'\cdot \hat p)$$

- Ahora deseamos obtener una expresión para el término l-ésimo $f_l(p)$
- Partimos de la ecuación de Lippman-Schwinger para los estados estacionarios de scattering:

$$|Elm+\rangle = |Elm\rangle + G_0(E+i0)V|Elm+\rangle$$

Ahora tomamos el bra $\langle \vec x|$ por la izquierda para obtener las funciones de onda en representación de posiciones. Como tanto $|Elm\rangle$ como $|Elm+\rangle$ tienen como parte radial los mismos armónicos esféricos, entonces nos queda una ecuación de la parte radial:


$$\Psi_{l,p}(r)=\hat{j_l}(r)-2m\int_0^\infty dr' \ G_0^{l,p}(r,r')V(r')\Psi_{l,p}(r')$$

donde $G_0^{l,p}(r,r')=-\frac{1}{p}\hat{j_l}(pr_<)\hat{h}_l^+(pr_>)$ y donde $r_<, r_>$ son, respectivamente, el valor más pequeño entre $r$ y $r'$ y el más grande

- A partir de la descomposición de $|\vec p_+\rangle$ en la base $|Elm+\rangle$ podemos desarrollar la expresión habitual en el acercamiento independiente del tiempo:

$$f(\vec p'\leftarrow \vec p)=-(2\pi)^2m\langle \vec p'|V|\vec p\rangle$$

- Y obtenemos finalmente una expresión analítica para los términos de la serie:

$$\boxed{f_l(p)=\frac{2m}{p^2}\int_0^\infty dr \hat{j_l}(pr)V(r)\Psi_{l,p}(r)}$$

**Observación:** esta expresión depende de la parte radial de la función de onda dispersada, que no conocemos de forma exacta, de modo que debemos iterarla. Para cada término $f_l(p)$ podemos iterar esta integral hasta el término que convenga y luego sumar para obtener una expresión aproximada de la amlitud de scattering

## Validez de la aproximación de Born

- En Temas anteriores ya hemos demostrado que la aproximación de Born da una amplitud de scattering real. Como el término l-ésimo es complejo: $f_l(p)=\frac{e^{i\delta_l}\sin(\delta_l)}{p}$, es de suponer que la aprox. de Born solo será buena si el desfasaje es suficientemente pequeño como para que la exponencial sea la unidad, o si la energía es suficientemente alta como para atenuar cualquier contribución
- En este caso, la aprox. de Born (para el término l-ésimo) vendría dada, como es evidente a partir de la expresión de arriba, cogiendo el primer término de la ec. de Lippman-Schwinger para la onda dispersada, que es la onda libre):

$$\boxed{f_l(p)\approx \frac{2m}{p^2}\int_0^\infty dr \ \hat{j_l}(pr)V(r)\hat{j_l}(pr)}$$

## Comportamiento extremo de $f_l(p)$ y $\delta_l(p)$

- Cuando el potencial es pequeño comparado con la energía cinética de la partícula, la ec. radial se torna la de la partícula libre y por ende $f_l(p)\to 0$. Asimismo, como apenas hay scattering, $\delta_l(p)\to n\pi$ porque deja inalterado el seno (concretamente, $\delta_l\to 0$ si requerimos que el desfasaje sea una función continua en su variable $p$ que se anule cuando $p\to\infty$, deshaciéndonos de la ambigüedad de módulo $\pi$)
- Cuando $l\to\infty$, como el término general de un sumatorio convergente ha de tender a 0 conforme $n\to\infty$, necesariamente se debe cumplir $f_l\to 0$. Asimismo, el desfasaje en onda $l$, por el mismo argumento de antes (apenas hay scattering en onda $l$) tiende a un múltiplo entero de $\pi$

## Criterio para ver hasta qué término es apreciable el desarrollo en ondas parciales

- Si $f_l\xrightarrow[l\to\infty]{}0$, es de suponer que existe un término a partir del cual las contribuciones ya no son apreciables en buena aprox. Encontrémoslo.
- Sea $a$ el tamaño (radial) aproximado del potencial dispersor
- La barrera centrífuga al borde del potencial será entonces: $l^2/2ma^2$ (hacemos $l(l+1)\approx l^2$ porque esta discusión es para $l$ grandes)
- Si la barrera centrífuga es mayor que la energía de la partícula, entonces el desfasaje debería ser pequeño: $l^2/2ma^2>>E$. La condición es entonces:

$$\boxed{l>>pa\Longrightarrow \text{$f_l$ apenas contribuye}}$$

- Alternativamente, podemos aplicar un argumento clásico: en teoría clásica de scattering, el parámetro de impacto viene dado por: $b=\frac{l}{p}$. De este modo, si $\rho>a$, la partícula fallará el disparo y pasará de largo, no habrá colisión. Por ende, la contribución en dicha onda será nula

**IMPORTANTE:** incluso si estamos a una energía tan baja que la aprox. de Born no es útil, sí lo será para términos suficientemente grandes de $l$. El procedimiento es entonces calcular $f(\vec p'\leftarrow \vec p)$ con la aprox. de Born y después restarle las contribuciones en ondas parciales hasta el término en el que empieza a ser válida. Finalmente, se recalculan las amplitudes parciales retiradas con otros métodos más precisos o ajustando datos experimentales y se suman de nuevo, dando así una amplitud de scattering final

## Bajas energías: longitud de scattering

- Sabemos que el término l-ésimo del desarrollo en ondas parciales de la amplitud de scattering toma la forma exacta:

$$f_l(p)=\frac{2m}{p^2}\int_0^\infty dr \hat {j_l}(pr)V(r)\Psi_{l,p}(r)$$

Si ahora admitimos que estamos a bajas energías, de modo que $p\to 0$, y recordamos el comportamiento de las funciones de Ricatti-Bessel cerca del origen del argumento, entonces llegamos a que:

$$\boxed{f_l(p)\xrightarrow[p \to 0]{}-a_l p^{2l}}$$

siempre y cuando el potencial se anule a cierta distancia finita $a$ (como sucede por ejemplo en un pozo cuadrado)

**Def.:** la magnitud $a_l$ se llama "longitud de scattering", aunque solo tiene unidades de longitud para onda $s$ ($l=0$)

**Observación:** a bajas energías, solo contribuyen los primeros términos del desarrollo en ondas parciales de la amplitud de scattering. A energías muy bajas, solo contribuye la onda $s$ y la amplitud es isotrópica (porque es cte.)

**¿Qué pasa si el potencial tiene alcance infinito?:**

1. Si $V(r)$ decae más rápido que cualquier potencia $1/r$ (i.e: exponencial) entonces el desarrollo en longitudes de scattering sigue valiendo
2. Si $V(r)$ decae como $1/r^\nu$, entonces hay que modificar los términos con $l\geq \frac{\nu-3}{2}$
3. En algunos casos, $a_l$ da infinito. En estos casos, Taylor dice que la "el comportamiento en potencias se ve reducido"

## Teorema de Levinson

Comparando la expresión exacta $f_l(p)=\frac{e^{i\delta_l(p)}\sin(\delta_l(p))}{p}$ con la aproximada para bajas energías: $f_l(p)\xrightarrow[p \to 0]{}-a_l p^{2l}$, y sabiendo que a bajas energías el desfasaje es pequeño (lo que nos permite linearizar el seno), vemos que, a bajas energías, $\delta_l(p)\sim p^{2l+1}$. Teniendo en cuenta que hemos pedido que, a bajas energías, el desfasaje sea un múltiplo entero de $\pi$, tenemos entonces que:

$$\boxed{\delta_l(p)\xrightarrow[p\to 0]{}n\pi-a_lp^{2l+1}}$$

El **teorema de Levinson** establece que "el número de estados ligados con momento angular $l$ que alberga el potencial viene dado por la diferencia entre el desfasaje a energía infinita y a energía nula":


$$\boxed{\delta_l(0)-\delta_l(\infty)=n_l\pi}$$

- Naturalmente, si tomamos $\delta_l(\infty)=0$ porque a energía infinita no hay desfasaje, entonces:

  $$\delta_l(0)=n_l\pi$$

**OJO!:** el teorema de Levinson es válido bajo nuestras suposiciones habituales del potencial, pero cuando $a_0\to \infty$, entonces todo sigue igual salvo que $\delta_0(0)=(n_0+1/2)\pi$

# 4. La solución regular

- Hasta ahora, hemos expresado la solución de la parte radial de la ec. de Scrhödinger como $\Psi_{l,p}(r)$, que se llama "función de onda radial normalizada". Su normalización es:

$$\int_0^\infty \Psi_{l,p'}(r)\Psi_{l,p}^*(r) = \frac{\pi}{2}\delta(p'-p)$$

- Esta función es solución de la ec. radial cuando las 2 condiciones de contorno impuestas son: $\Psi_{l,p}(0)=0$ y $\Psi_{l,p}(r)\xrightarrow[r\to\infty]{}\hat{j_l}(pr)+pf_l\hat h_l^+(pr)$

- No obstante, es más conveniente para el desarrollo que está por venir el trabajar con la función de onda regular $\phi_{l,p}(r)$, que también es solución de la parte radial pero cuando las condiciones de contorno son $\phi_{l,p}(r)\xrightarrow[r\to 0]{}\hat {j_l}(pr)$. Véase que $\phi$ es proporcional a $\Psi$

# 5. El método de desfasaje variable (ec. de Calogero)

- Hasta ahora, solo hemos aprendido a calcular la ecuación del desfase estudiando el comportamiento asintótico de la función de onda del estado estacionario de scattering y comparándolo con el de la onda libre. El desfase entre ambas ondas sinusoidales (que lo son porque, a suficiente distancia del potencial y cumpliendo este con nuestras propiedades de decaimiento habituales, el potencial tiene un efecto despreciable y ambas partículas son libres) será precisamente el desfasaje en onda $l$ que buscamos (cambiando el momento angular de la partícula entrante cambiamos $l$, y este se conserva porque admitimos que el sistema tiene simetría bajo rotaciones y que no hay spines)
- Sin embargo, deseamos ahora tener una forma más analítica de calcular el desfasaje. Vemos que el método de Calogero nos da el desfasaje en onda $s$ (así que solo podemos obtener a partir de este método el primer término del desarrollo en ondas parciales de la amplitud de scattering, $f_0(p)$)

## Derivación

- Sea el potencial (de alcance infinito) $V(r)$
- Sea otro potencial obtenido a partir de truncar el primer a cierta distancia $r=\rho$
- Denotamos la función de onda regular (en onda $s$) del potencial completo como $\phi(r)$ y la del potencial truncado como $\phi_\rho(r)$
- Denotamos el desfasaje en onda $s$ originado por el potencial completo como $\delta$ y al del potencial truncado como $\delta(\rho)$
- Imponemos como condición de contorno que $\delta(0)=0$. Esto es una definición, pero tiene sentido: si truncamos todo el potencial, nos quedamos con que ya no existe potencial en ninguna región del espacio, de modo que la onda no se dispersará y será idéntica a la onda libre. Por la ambigüedad módulo $\pi$ del desfasaje, podríamos pedir en general $\delta(0)=n\pi$ con $n$ entero, pero imponemos en concreto 0 para quitarnos esta ambigüedad. Si se quiere, es un convenio

**OJO!:** no te confundas con que antes impusiéramos $\delta(\infty)=0$, porque este infinito era para $p$, no para $\rho$! La idea detrás de esta dependencia en el espacio es que el desfasaje del potencial truncado depende de dónde trunquemos el potencial, en sí también tiene una dependencia en el momento y se siguen cumpliendo las condiciones de contorno para momento

- Por otro lado, cuando más lejos trunquemos el potencial, más se parecerán los desfasajes entre sí, de modo que:

$$\delta(\rho)\xrightarrow[r\to\infty]{}\delta$$

- En el intervalo en el que existe potencial: $r\in [0,\rho]$, ambos potenciales son idénticos, y las condiciones de contorno de la solución regular de uno y otro también lo son (teorema de existencia y unicidad). Por ende:

$$\phi(r) \equiv \phi_\rho(r); \ \ \forall r \in [0,\rho]$$

**Nota:** este no sería el caso si empleáramos la función radial normalizada, porque su normalización depende del potencial en todo el espacio, y estos no son iguales

- Pasado el punto donde truncamos, la partícula que ve el potencial truncado será ahora libre, y por ende tendrá la forma:

$$\phi_\rho(r)=\alpha(\rho)\sin[pr+\delta(\rho)]; \ \ r>\rho$$

- Naturalmente, debemos pedir regularidad de la función de onda $\phi_\rho(r)$ en la frontera donde truncamos el potencial: $r=\rho$. Sin embargo, como dentro de esta esfera de radio $\rho$ las dos funciones de onda coinciden, entonces esto nos da información sobre la función de onda del potencial no truncado. Imponiendo entonces continuidad y derivabilidad:

$$\phi(\rho)=\phi_\rho(\rho)=\alpha(\rho)\sin[p\rho+\delta(\rho)]$$

$$\phi'(\rho)=\phi_\rho'(\rho)=p\alpha(\rho)\cos[p\rho+\delta(\rho)]$$

Como estas dos condiciones han de cumplirse para cualquier $\rho$, admitimos que esta puede tomar cualquier valor y sustituimos la variable $\rho$ por $r$. Introduciendo ambas identidades en la ec. de Scrhödinger radial y deshaciéndonos del factor común $\alpha(\rho)$, acabamos finalmente con la ecuación para el desfasaje en onda $s$:

$$\boxed{\delta'(r)=\frac{2m}{p}V(r)\sin^2[pr+\delta(r)]}$$

- Finalmente, el desfasaje en onda $s$ será $\delta_0=\delta(r\to\infty)$, ya que queremos el caso particular en el que el potencial no se trunca (i.e: se trunca en infinito porque decae a 0 de todos modos). Naturalmente, si el potencial sí se trunca en algún punto, este razonamiento sigue siendo válido

**Nota:** entiendo que esta ecuación es trivialmente generalizable a onda $l$ poniendo un desfase adicional de $-l\pi/2$

## Propiedades de $\delta$ a partir de la ec. de Calogero

1. Las regiones donde el potencial es atractivo ($<0$) adelantan a la onda (estoy integrando algo positivo porque va con un "-" delante) y las regiones donde el potencial es repulsivo ($>0$) retrasan la onda, como ya habíamos visto
2. Dados dos potenciales $V_1(r)\geq V_2(r); \ \ \forall r$, entonces sus desfasajes en onda $s$ cumplirán $\delta_1(r)\leq \delta_2(r); \ \ \forall r$ y en particular $\delta_1\leq \delta_2$

3. Podemos expresar el desfasaje en onda $s$ en forma integral integrando a ambos lados de la ec. de Calogero entre 0 e $\infty$ y aplicando $\delta(r=0)=0$:

$$\boxed{\delta_0=\frac{2m}{p}\int_0^\infty dr \ V(r)\sin^2[pr+\delta(r)]}$$

4. Por el teorema de acotación de integrales, podemos  acotar el desfasaje en onda $s$:

$$|\delta_0| \leq \frac{2m}{p}\int_0^\infty |V(r)|dr $$

5. $d_0=0$ a muy altas energías ($p\to\infty$) y también cuando el potencial es muy débil ($V\to 0$)

# 6. Solución iterada de la función regular

- Recuerda que la función regular obedecía la ecuación integral:

$$\phi_{l,p}(r)=\hat{j_l}(r)-2m\lambda\int_0^r dr' \ g_{l,p}^0(r, r')V(r')\phi_{l,p}(r')$$

donde $\lambda \in \mathbb R$ es un parámetro que sirve para cuantificar cómo de débil (o fuerte) es el potencial

- El propagador $g_{l,p}^0(r,r')$ es 0 cuando $r'>r$ (ecuaciones así reciben el nombre de "ecuaciones de Volterra"). Esto implica que podemos resolver por iteración independientemente de cómo de fuerte sea el potencial. Siempre va a converger (no es así para la función radial normalizada). Lo formulamos como un teorema:

**Teorema:**

La solución iterada de la ecuación integral de la función regular converge siempre, independientemente de la intensidad del potencial ($\forall \lambda \in \mathbb C$), de modo que:

$$\boxed{\phi_{l,p}(r)=\displaystyle\sum_{n=0}^\infty \lambda^n\phi^{(n)}(r)}$$

con $\phi^{(n)}(r)=-2m\int_0^rdr' \ g_{l,p}^0(r, r')V(r')\phi^{(n-1)}(r')$

# 7. La función de Jost

- Ahora que tenemos una solución iterada de la función regular, deseamos obtener una expresión para la amplitud de scattering a partir de ella
- Lejos del centro de dispersión, la función regular habrá de ser una combinación lineal de las dos funciones de Haenkel que dan la onda libre:

$$\phi_{l,p}(r)\xrightarrow[r\to\infty]{}\frac{i}{2}[f_l(p)\hat{h_l}^-(pr)-f_l^*(p)\hat{h_l}^+(pr)]$$

donde $f_l(p)$ es la "función de Jost"

Por otro lado, la función normalizada lejos del centro de dispersión se comporta como:

$$\Psi{l,p}(r)\xrightarrow[r\to\infty]{}\frac{i}{2}[\hat{h_l}^-(pr)-s_l(p)\hat{h_l}^+(pr)]$$


- De este modo, recordando que ambas funciones son proporcionales entre sí y comparando ambas expresiones, llegamos a la conclusión de que los autovalores del operador $S$ son precisamente:

$$\boxed{s_l(p)=\frac{f_l^*(p)}{f_l(p)}}$$

Y la constante de proporcionalidad (cte. en tanto que es otra variable fija de la que no depende la función radial) es:

$$\boxed{\phi_{l,p}(r)=f_l(p)\Psi_{l,p}(r)}$$

Despejando la primera expresión, multiplicando por la función de Jost y tomando raíz cuadrada encontramos la relación de la función de Jost con el desfasaje:


$$\boxed{f_l(p)=|f_l(p)|e^{-i\delta_l(p)}}$$

**Resultado:** el desfasaje es "-" la fase de la función de Jost

- Finalmente, nos interesa calcular de forma explícita la función de Jost. Volviendo a la ecuación integral para la función regular y recordando que $g_{l,p}^0(r, r')=\frac{i}{2p}[\hat{h_l}^-(pr)\hat{h_l}^+(pr')-\hat{h_l}^+(pr)\hat{h_l}^-(pr')]$ obtenemos que:

  $$\boxed{f_l(p)=1-2m\frac{\lambda}{p}\int_0^\infty dr \ \hat{h_l}^+(pr)V(r)\phi_{l,p}(r)}$$

De modo que si conocemos (aun si es de forma aproximada) la función regular, podemos inmediatamente calcular la función de Jost

# 8. $\lambda$ complejo y convergencia de la serie de Born

- En todo nuestro desarrollo y en nuestro teorema, si bien hemos considerado $\lambda \in \mathbb R$ para que el problema tenga significado físico (de lo contrario, $H=H_0+\lambda V$ no sería autoadjunto y carecería de significado físico), todos nuestros resultados de convergencia se siguen manteniendo si consideramos $\lambda\in \mathcal C$
- De este modo, como cierto desarrollo en serie de una función de variable compleja solo converge si la función es analítica, y como sabemos que los desarrollos en serie de la función regular $\phi_{l,p}(r)$ y de $f_l(p)$ convergen para cualquier $\lambda$ incluso si es complejo, entonces tenemos que:

**Resultado:** la función radial regular y la función de Jost son analíticas de $\lambda$

Además, recordando que:

$$\Psi_{l,p}(r)=\frac{\phi_{l,p}(r)}{f_l(p)}$$

$$s_l(p)=\frac{f_l^*(p)}{f_l(p)}$$

son también analíticas en $\lambda$ por ser cociente de funciones con esta propiedad (siempre que la función de Jost no se anule)

**Resultado:** como $f_l=1$ en $\lambda=0$, es seguro que existe cierta región del plano complejo en torno a $\lambda=0$ en la que la función de onda normalizada y $s_l(p)$ no tienen polos, de modo que podemos expandirlas en serie de forma convergente en torno a $\lambda=0$.  Es decir, la serie de Born converge para $\lambda\to 0$, cuando el potencial es débil. Queda así demostrado y entendido uno de los dos casos en los que la serie de Born converge

- Para cualquier $\lambda\in \mathbb C$, en general la función de Jost se anulará. Entonces, la función de onda normalizada y $s_l(p)$ tendrán un polo en el $\lambda$ donde la función de Jost se anule. Por el teorema de convergencia de variable compleja, las respectivas series de estas dos funciones convergerán $\forall \lambda \in \mathbb C \ \ tq. \ \  |\lambda|<|\lambda_0|$, siendo $\lambda_0$ aquel valor de $\lambda$ de menor módulo que da un polo, y divergerán para cualquier lambda de mayor módulo (se puede ver como dentro o fuera de una circunferencia de radio $|\lambda_0|$ en el plano complejo). Entonces, la convergencia de la serie de Born dependerá de si $|\lambda_0|>1$ o no, porque $\lambda=1$ corresponde a nuestro problema
