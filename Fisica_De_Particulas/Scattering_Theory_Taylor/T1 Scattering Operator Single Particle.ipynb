**Objetivos:** entender el concepto de estados asintóticos y la condición asintótica. Entender motivación de la matriz de scattering, $S$

# 1. Scattering clásico

- El problema de scattering es el siguiente: sea una partícula que viene desde el infinito (partícula libre) y penetra en cierta región donde existe un potencial. La partícula interactúa con este y después sale en otra dirección. Se desea conocer la relación entre la "recta" de entrada y la de salida. Lo que le suceda a la partícula en la región con potencial estará codificado en las direcciones de salida y sus frecuencias. No podemos mirar directamente qué sucede dentro de la región con potencial porque las interacciones son muy cortas y además la resolución necesaria excede las capacidades de nuestros instrumentos

- **Nota:** en la práctica, se pueden formar también estados ligados. Esto me chirría un poco porque la partícula, si parte del reposo en el infinito, tendrá energía mecánica $E=0$, de modo que no debería poder formar un estado ligado. Sin embargo, parece que para potenciales suficientemente atractivos la partícula puede quedarse orbitando el centro de potencial

# 2. Scattering cuántico

- En analogía a la teoría clásica de scattering, se buscan ahora los "estados asintóticos" $|\Psi_{in}\rangle$ y $|\Psi_{out}\rangle$ de la partícula dispersada. Lo que suceda en la región con potencial no nos preocupa

## Condición de existencia de estados asintóticos para potenciales esféricos

- No todos los potenciales permiten la existencia de estados asintóticos (por ejemplo, un potencial que decaiga demasiado lentamente en el infinito seguirá ejerciendo una acción apreciable sobre las partículas incluso lejos de la posición del centro de scattering, y como tal la partícula jamás se comportará en suficiente buena aproximación como una partícula libre cuando se encuentre lejos del centro de scattering, de modo que no tendrá sentido hablar de estados asintóticos)
- No hay un conjunto de condiciones generales que garanticen la existencia de estados asintóticos para un potencial dado, pero los más aceptados son los siguientes:

1. $V(r) \sim r^{-3-\epsilon}$ cuando $r\to \infty$
2. $V(r) \sim r^{-3/2+\epsilon}$ cuando $r\to 0$
3. $V(r)$ continuo salvo en puntos aislados en $r\in (0,\infty)$

- Es decir, el potencial debe (1) decaer más rápido que $1/r^3$ cuando $r\to\infty$, (2) debe no ser demasiado singular en el origen y (3) debe ser razonablemente bien comportado en todos los puntos intermedios

# 3. La condición asintótica

- "Si el potencial $V$ satisface las 3 condiciones de existencia de estados asintóticos, entonces por cada vector $|\Psi_{in}\rangle$ del espacio de Hilbert existe otro vector $|\Psi\rangle$ tal que el primer estado sea el estado asintótico del segundo"

- **Nota:** esto se cumple para todos los vectores del espacio de Hilbert, y a cada vector solo le corresponde un estado asintótico de entrada (y otro de salida). Esto es como en el caso clásico, es determinista (para la solución de los kets) y la solución es única.

- **Nota:** se puede demostrar que el espacio de Hilbert del sistema tratado se puede descomponer en suma directa de un subespacio de estados ligados y otro de estados de scattering:


$$\mathcal H=\mathcal B\oplus \mathcal R$$

- La condición asintótica establece claramente que todo ket del espacio de Hilbert es el estad asintótico de alguna "trayectoria", y esto incluye a los estados ligados (que son los autoestados del Hamiltoniano completo, es decir, partícula libre + potencial: $H=H_0+V$). Entiendo que esto se debe a que nada impide que la partícula empiece cerca del potencial, ya que de hecho no se puede hablar de posiciones porque nuestras partículas serán autoestados del momento. Es por esto que la condición:

$$\displaystyle\lim_{t\to-\infty}U(t)|\Psi\rangle = \displaystyle\lim_{t\to-\infty}U_0(t)|\Psi_{in}\rangle$$

(que la partícula provenga de un estado asintótico) no se ve violada si se tiene que el estado asintótico es un estado ligado

- **Def.:** se dice que una teoría de scattering es "asintóticamente completa" si (1) todo estado con asíntota de entrada tiene asimismo una asíntota de salida y (2) si todo estado de scattering es ortogonal a todo estado ligado

# 4. La matriz de scattering ($S$)

- Sean $|\Psi_{in}\rangle, |\Psi_{in}\rangle$ los estados asintóticos de entrada y salida de una partícula, respectivamente. Sea $|\Psi\rangle$ el vector de estado de la partícula en $t=0$. Podemos recuperar el estado actual yendo al principio del estado asintótico y permitiendo a la partícula evolucionar con el Hamiltoniano completo, de modo que:

$$|\Psi\rangle = \displaystyle\lim_{t\to-\infty}U^\dagger(t)U_0(t)|\Psi_{in}\rangle$$

donde la evolución hacia atrás en el tiempo de $U^\dagger(t)$ equivale a la evolución hacia delante en el tiempo de $U(t)$. Para acortar la notación, llamaremos al operador que nos lleva desde el estado de entrada hasta el de $t=0$ $\Omega_+$, de forma que:

$$|\Psi\rangle = \Omega_+|\Psi_{in}\rangle$$

Se puede hacer un razonamiento del todo equivalente con el estado de salida, $|\Psi_{out}\rangle$. Podemos llegar al estado en $t=0$ si nos vamos al final del estado de salida y después retrocedemos en el tiempo hasta $t=0$, de modo que:

$$|\Psi\rangle = \displaystyle\lim_{t\to\infty}U^\dagger(t)U_0(t)|\Psi_{out}\rangle$$

A este operador le llamaremos $\Omega_-$, tal que:

$$|\Psi\rangle =\Omega_-|\Psi_{out}\rangle$$

**Def.:** los operadores que nos dan el estado en $t=0$ de la partícula a partir de sus estados de entrada o de salida se llaman "operadores de onda de Møller". Estos dos operadores son isométricos pero no unitarios.

**Nota:** se dice que un operador es "isométrico" si es un operador (1) lineal, (2) definido en todo el espacio de Hilbert y (3) que preserve la norma de los vectores de dicho espacio. Se diferencian de los operadores unitarios en que no pedimos que mapeen $\mathcal H\to \mathcal H$, sino que solo pedimos que mapeen a un subespacio: $\mathcal H\to \mathcal R$ con $R\subseteq \mathcal H$, donde si se cumple la igualdad entonces son unitarios

- **Nota:** en espacios de dimensión finita, isométrico y unitario son sinónimos. Sin embargo, en espacios de dimensión infinita, un operador puede ser isométrico pero no unitario (unitarios $\subset$ isométricos). Imagina por ejemplo que tomas la base infinita pero numerable: $B:=\{|1\rangle, \ldots, |n\rangle, \ldots\}$. Sea un operador que mapea $|1\rangle\to|2\rangle, |2\rangle\to|3\rangle$ y en general $|n\rangle\to|n+1\rangle$. Vemos entonces que es lineal, está definido en todo el espacio de Hilbert y preserva la norma, de modo que es isométrico. Sin embargo, no es unitario porque al tratar de invertirlo vemos que mapea bien todos los estados de vuelta a sus originales salvo para $|1\rangle$, ya que ninguna flecha le llega (hemos empezado a mapear el 1 al 2). Por ende, el recorrido no es el mismo que el dominio y el operador es isométrico pero no unitario

- Los operadores isométricos cumplen entonces:

$$\Omega^\dagger = \begin{cases}\Omega^{-1} \text{ en } Rec(\Omega) \\ 0 \text{ fuera } \end{cases}$$

- Nuestro interés último en la teoría de scattering es, dado un estado de entrada, determinar el estado de salida que le corresponde. Es decir, deseamos encontrar un operador tal que si le damos el estado de entrada, nos devuelva el de salida. Aún no sabemos hacer eso, pero sí que sabemos cómo ir del estado de entrada al de $t=0$ y del de salida al de $t=0$, de modo que podemos ir del de entrada al de $t=0$ e invertir el segundo para ir del de $t=0$ al de salida y entonces ya sabríamos hacerlo. Procedemos con los operadores de Møller:

$$|\Psi_{out}=\Omega_-^\dagger|\Psi\rangle = \Omega_-^\dagger\Omega_+|\Psi_{in}\rangle$$

- Por tanto, el "operador de scattering", que es el operador que nos lleva del estado de entrada al de salida, se puede construir a partir de los operadores de onda de Møller de la forma:

$$\boxed{S=\Omega_-^\dagger\Omega_+}$$

tal que:

$$\boxed{|\Psi_{out}\rangle = S|\Psi_{in}\rangle}$$

- En un experimento de scattering, la partícula sale preparada del acelerador en un estado de entrada $|\Psi_{in}\rangle$, y lo que deseamos estudiar es la probabilidad de que salga dispersada al estado asintótico de salida $\Psi_{out}$. El estado al que evolucionará la partícula a partir del estado de entrada será $|\Psi\rangle = \Omega_+|\Psi_{in}\rangle$, mientras que el estado que evolucionaría al estado de salida es $|\Psi'\rangle = \Omega_-|\Psi_{out}\rangle$. Entonces, la probabilidad de que la partícula, que entró con el estado de entrada $|\Psi_{in}\rangle$, salga con el estado de salida $|\Psi_{out}\rangle$, viene dado por:

$$P=|\langle \Psi'|\Psi\rangle|^2 = |\langle \Psi_{out}|\Omega_-^\dagger\Omega_+|\Psi_{in}\rangle|^2$$

donde reconocemos la matriz de scattering en términos de los operadores de onda de Møller. Por tanto, encontramos que la probabilidad de que una partícula salga a cierto estado de salida habiendo entrado con cierto estado de entrada viene dada por el elemento de matriz de la matriz de scattering con respecto de los estados de entrada y de salida de interés. Esto es:

$$\boxed{P(|\Psi_{out}\rangle \leftarrow |\Psi_{in}\rangle)=|\langle\Psi_{out}|S|\Psi_{in}\rangle|^2}$$

**Interpretación:** la matriz de scattering me dice con qué estado sale la partícula a partir del estado con el que entra. La probabilidad que hemos calculado también se puede ver sencillamente como la probabilidad de encontrar que el estado de salida es uno dado. Esto es, si entro con cierto estado $|\Psi_{in}\rangle$, sé que salgo con el estado $|\Psi_{out}\rangle=S|\Psi_{in}\rangle$, de modo que quiero conocer la probabilidad de medir que el estado de salida es $|\chi\rangle$. Entonces, es:

$$P=|\langle \chi|\Psi_{out}\rangle|^2=|\langle \chi|S|\Psi_{in}\rangle|^2$$

- **Propiedad:** la matriz de scattering es **unitaria** y lineal (me pregunto si por el teorema de Wigner está entonces asociada a una operación de simetría, aunque el teorema va en el sentido contrario: op. de simetría ==> unitaria+lineal ó antiunitaria+antilineal). Esto se debe a que el primer operador de Møller va $\mathcal H\to \mathcal R$ y el segundo, por ser el adjunto, va de $\mathcal R\to \mathcal H$

- **Unicidad de los estados asintóticos:** el hecho de que $S$ sea unitaria implica que es invertible (en todo el espacio de Hilbert, ya que para ser unitaria ha de estar definida en todo ello y lo está por lo que acabamos de ver). Por tanto, para cada estado asintótico de entrada existe un único estado asintótico de salida y viceversa. Además, por ser $S$ lineal, se preserva la superposición:

$$|\Psi_{in}\rangle = a|\phi_{in}\rangle + b|\chi_{in}\rangle \Longrightarrow |\Psi_{out}\rangle = a|\phi_{out}\rangle + b|\chi_{out}\rangle$$
