# 1. Reducción a scattering de 1 partícula

**Clave:** el problema de scattering de 2 partículas cuya interacción viene dada por un potencial central se puede reducir al de scattering de 1 sola partícula por un potencial fijo (temas anteriores)

La demostración de este importante resultado recuerda a que un sistema de dos partículas que interactúan a través de un potencial central se puede reducir al de una sola partícula sometida asimismo a un potencial central, con masa igual a la masa reducida del sistema (esto solo es válido en Mecánica no relativista, sea clásica o cuántica)

**Dem.:**

Sea el espacio de Hilbert de los vectores de estado del sistema compuesto: $\mathcal H=\mathcal H_1\otimes \mathcal H_2$. Se puede demostrar que dicho espacio se puede factorizar asimismo alternativamente como: $\mathcal H=\mathcal H_{CM}\otimes \mathcal H_{rel}$, donde el primer espacio contiene los estados del centro de masas y el segundo los de la partícula virtual.

Entonces, el Hamiltoniano, que es:

$$H=\frac{\vec p_1^2}{2m_1}+\frac{\vec p_2^2}{2m_2}+V(\vec x_1-\vec x_2)$$

se puede reescribir de la forma:

$$H=\frac{\vec P_{CM}^2}{2M}+\left(\frac{\vec p_{rel}^2}{2\mu}+V(\vec X)\right)=H_{CM}+H_{rel}$$

donde:

- $\vec P_{CM}=\vec p_1+\vec p_2$
- $\vec p_{rel}=\frac{m_2\vec p_1-m_1\vec p_2}{m_1+m_2}$
- $\vec X=\vec x_1-\vec x_2$

Naturalmente, $\vec P_{CM}$ está definido sobre $\mathcal H_{CM}$ y $\vec X, \vec p_{rel}$ están definidos sobre $\mathcal H_{rel}$, de modo que los que aparecen en el Hamiltoniano son los extendidos al espacio de Hilbert del sistema compuesto.

El operador de evolución temporal vendrá dado por:

$$U(t)=e^{-iH_{CM}t}\otimes e^{-iH_{rel}t}$$

Recuerda que $exp(A+B)\neq exp(A)\otimes exp(B)$ a menos que $[A,B]=0$, pero como los dos Hamiltonianos conmutan entre sí por actuar sobre espacios de Hilbert distintos, se puede factorizar el operador de evolución temporal total. Entonces, vemos que en un sistema de dos partículs que interctúan a través de un potencial central, podemos estudiar su dinámica por (1) el movimiento del centro de masas, que resulta ser una partícula libre (porque el sistema no está sometido a más fuerzas) y (2) el movimiento de una partícula efectiva. La clave está en ver que el Hamiltoniano de la partícula efectiva es idéntico al de la partícula que se dispersa por un potencial fijo, de modo que hemos demostrado que el scattering de un sistema de dos partículas se puede estudiar como el scattering de una sola partícula sometida a un potencial asimismo efectivo pero fijo$_\blacksquare$

# 2. Operador $S$ en un sistema de 2 partículas

- Se puede demostrar que la condición asintótica y la completitud asintótica se siguen cumpliendo en el caso de que el sistema esté compuesto por 2 partículas, y que el operador de scattering es sencillamente:

$$S=I_{CM}\otimes S_{rel}$$

donde la identidad se aplica a los kets del CM y la $S$ se aplica a los del subespacio de kets de la dinámica relativa entre las dos partículas. Esto va en perfecta consonancia con la afirmación del apartado anterior. Asimismo, la identidad significa que el CM no se dispersa. Esto tiene mucho sentido, porque hemos visto que su Hamiltoniano es el de una partícula libre, de modo que al no estar sometido a ningún potencial, efectivamente no se dispersa. La dinámica del CM y la de la partícula efectiva están desconectadas, y la del CM es trivial, de modo que una vez más vemos que el scattering de dos partículas es del todo equivalente al de una sola partícula con dispersada por un potencial fijo.

Encontramos asimismo que el operador de scattering de dos partículas se puede expresar, como en el caso de una sola partícula, como la composición de dos operadores de Møller, ahora extendidos, con la parte del CM trivial:

$$S=\Omega_-^\dagger\Omega_+$$

donde ahora:

$$\Omega_{\pm}=I_{CM}\otimes \Omega_{\pm, rel}$$

**Nota:** el operador $S$ sigue conmutando con el Hamiltoniano libre, que ahora es:

$$H_0=\frac{p_1^2}{2m_1}+\frac{p_2^2}{2m_2}=\frac{\vec P_{CM}^2}{2M}+\frac{\vec p_{rel}^2}{2\mu}$$

de modo que ahora tenemos:

$$\boxed{[S, H_0]=0}$$

como en el scattering de una sola partícula.

**Importante:** el momento total del sistema se conserva. Este no era el caso en el scattering de una sola partícula, donde veíamos que solo se conservaba el módulo del momento de la partícula (que era el total del sistema), pero no su dirección. Ahora, como el sistema no está sometido a ninguna fuerza externa, el momento lineal total (el del CM) se conserva. Lo sabemos porque:

$$[\vec P_{CM}, S]=[\vec P_{CM}\otimes I, I\otimes S_{rel}]=0$$

Como el scattering ahora conserva energía y momento, sus elementos de matriz han de ir como:

$$\langle \vec p_1'\vec p_2'|S|\vec p_1\vec p_2\rangle \sim \delta(E_1'+E_2'-E_1-E_2)\delta^3(\vec p_1'+\vec p_2'-\vec p_1-\vec p_2)$$

**Nota:** comparando con los elementos de matriz del problema de scattering de una sola partícula, vemos que la delta no va ya solo con la energía, sino que va acompañada de otra delta que conserva el momento lineal total del sistema. Sin embargo, hemos dicho antes que el problema de scattering de 2 partículas es idéntico al de 1 sola partícula con potencial efectivo fijo, de modo que cabe preguntarse por qué los elementos de matriz no tienen la misma forma. Entiendo a que esto se debe a que el problema solo queda equivalente al de scattering de 1 sola partícula si se estudia en representación CM-relativo, no 1-2. Es por esto que los elementos de matriz no quedan como $\delta^3(\vec p_{rel}'-\vec p_{rel})-2\pi i \ t(\vec p_{rel}'-\vec p_{rel})\delta(E_{rel}'-E_{rel})$ (ahora lo vemos)

**Nota:** los kets $|\vec p_1, \vec p_2\rangle$ no solo son autovectores de $\vec p_1, \vec p_2$ sino también de $\vec P_{CM}$ y de $\vec p_{rel}$ por como hemos definido antes dichos operadores. Por tanto, podemos hablar indistintamente del estado $\boxed{|\vec p_1, \vec p_2\rangle\equiv |\vec P_{CM}, \vec p_{rel}\rangle}$

Es mucho más conveniente trabajar en esta representación de los vectores de estado del sistema completo, porque ahora el operador de scattering es trivial en uno de los subespacios (en el del CM). Comprobamos entonces que, efectivamente, como dicta la intuición, los elementos de matriz (en la base adecuada) quedan formalmente idénticos a los del caso de 1 sola partícula (solo que imponiendo además conservación de momento lineal total):


$$\langle \vec p_1', \vec p_2'|S|\vec p_1, \vec p_2\rangle = \langle\vec P_{CM}', \vec p_{rel}'|I\otimes S_{rel}|\vec P_{CM}, \vec p_{rel}\rangle$$

$$\langle \vec p_1', \vec p_2'|S|\vec p_1, \vec p_2\rangle = \delta^3(\vec P_{CM}'-\vec P_{CM})\langle \vec p_{rel}'|S_{rel}| \vec p_{rel}\rangle$$

pero como hemos visto que el problema de scattering de 2 partículas se reduce al de 1 sola partícula, tenemos entonces:

$$\langle \vec p_{rel}'|S_{rel}|\vec p_{rel}\rangle = \delta^3(\vec p_{rel}'-\vec p_{rel})-2\pi i \ t(\vec p_{rel}'\leftarrow \vec p_{rel})\delta(E_{p_{rel}'}-E_{p_{rel}})$$

Combinando estos dos resultados, volvemos a la base de momentos "desacoplados" y tenemos:

$$\boxed{\langle\vec p_1', \vec p_2'|S|\vec p_1, \vec p_2\rangle = \delta^3(\vec p_1'-\vec p_1)\delta^3(\vec p_2'-\vec p_2)-2\pi i \ t(\vec p_{rel}'-\vec p_{rel})\delta(\sum E_{i}'-\sum E_{i})\delta^3(\sum \vec p_i'-\sum \vec p_i) }$$

**Interpretación:** el primero de los dos términos es la amplitud de que las partículas no se dispersen. El segundo es la amplitud de que sí se dispersen, en cuyo caso se conservan energía total y momento lineal total, pero no los momentos individuales de las partículas

**Def.:** se define la amplitud de scattering de dos partículas como la amplitud de scattering de la partícula efectiva (y por tanto va con la masa efectiva):

$$\boxed{f(\vec p_{rel}'\leftarrow \vec p_{rel})=-(2\pi)^2\mu \ t(\vec p_{rel}'\leftarrow \vec p_{rel})}$$

# 3. Sección eficaz diferencial en distintos inerciales

- En un experimento de scattering de 1 sola partícula, podíamos ponernos en el inercial de laboratorio (esto es, el del centro de scattering, que es fijo) y este era el único inercial que tendría sentido considerar en la práctica (no sería conveniente ponernos en el inercial de la partícula, de modo que $\vec p=0$ siempre, y lo que se moviera fuera el centro de scattering). Sin embargo, en un experimento de scattering de 2 partículas tenemos ahora varios inerciales en los que podemos ponernos

1. Inercial de laboratorio: la partícula target se encuentra inicialmente en reposo y le disparamos otra partícula
2. Inercial del CM: aquel en el que $\langle \vec P_{CM}\rangle = 0$
3. Arbitrario: aquel en el que $\vec p_1$ y $\vec p_2$ pueden ser lo que sea (se emplea por ejemplo en el estudio de colisiones moleculares en un gas)

**Def.:** definimos la sección eficaz pidiendo que sea válida **en todo inercial** la relación conocida para 1 sola partículas se mantenga:

$$\boxed{N_{sc}(\Delta \Omega \leftarrow \vec p_1, \vec p_2)=n_{inc}(\vec p_1, \vec p_2)\sigma(\Delta \Omega \leftarrow \vec p_1, \vec p_2)}$$

En general, si alguna de las magnitudes no es invariante bajo cambio de inercial, la sección eficaz también se deberá transformar.

**Resultado 1:** el número de cuentas de partículas en un mismo ángulo sólido ha de ser invariante bajo cambio de inercial (porque es lo que registra la máquina). De este modo, si en el CM veo $\Delta\Omega$ y momentos $\vec p_1, \vec p_2$ (con $\vec p_1=-\vec p_2$) y en otro inercial cualquiera veo $\Delta \Omega'$ con momentos $\vec p_1', \vec p_2'$, entonces $N_{sc}(\Delta \Omega \leftarrow \vec p_1, \vec p_2)=N_{sc}(\Delta \Omega' \leftarrow \vec p_1', \vec p_2')$

**Resultado 2:** la densidad de partículas incidentes, entendida como aquella definida en el plano perpendicular al movimiento relativo de las partículas, es invariante bajo cambio de inercial

**Resultado 3:** como estas dos magnitudes son invariantes, la sección eficaz es asimismo **invariante bajo cambios de inercial**

$$\boxed{\sigma(\Delta\Omega \leftarrow \vec p_1, \vec p_2) = \sigma(\Delta\Omega' \leftarrow \vec p_1', \vec p_2')}$$

**Nota:** sin embargo, la sección eficaz diferencial sí que se transforma bajo cambios de inercial. Consideremos el límite $\Delta \Omega \to d\Omega$, de modo que $\sigma(d\Omega \leftarrow \vec p_1, \vec p_2)=\left(\frac{d\sigma}{d\Omega}\right)d\Omega$

En otro inercial, tendremos $\sigma(d\Omega' \leftarrow \vec p_1', \vec p_2')=\left(\frac{d\sigma}{d\Omega}\right)'d\Omega'$, pero como ambas magnitudes son iguales, tenemos entonces que la sección eficaz diferencial se transforma con el cambio de inercial de la forma:

$$\boxed{\left(\frac{d\sigma}{d\Omega}\right)'=\left(\frac{d\sigma}{d\Omega}\right)\frac{d\Omega}{d\Omega'}}$$

- La sección eficaz diferencial más conveniente para cálculos teóricos es la del CM, pero en la práctica la que se mide es la del inercial del laboratorio. La transformación entre estas dos es como sigue:

$$\boxed{\left(\frac{d\sigma}{d\Omega}\right)_{lab}=\left(\frac{d\sigma}{d\Omega}\right)_{CM}\frac{1+2\frac{m_1}{m_2}\cos\theta_{CM}+(m_1/m_2)^2)^{3/2}}{|1+\frac{m_1}{m_2}\cos\theta_{CM}|}}$$

donde $\theta_{CM}$ es el ángulo de scattering del CM

**Nota:** esta transformación de la sección eficaz diferencial es también válida en scattering relativista

# 4. Sección eficaz en el centro de masas

- Se puede demostrar (de todos modos lo dicta la intuición) que, como el problema de scattering de 2 partículas se puede reducir al de scattering de 1 sola partícula por un potencial fijo, la sección eficaz diferencial se puede calcular a través de a amplitud de scattering. Tenemos entonces que, en el centro de masas:

$$\boxed{\frac{d\sigma}{d\Omega}=|f(\vec p \leftarrow \vec p_0)|^2}$$

**Nota:** he escrito $f(\vec p \leftarrow \vec p_0)$ para aligerar la notación, realmente deberíamos hacer alusión a los momentos de las dos partículas, de modo que $f(\vec p \leftarrow \vec p_0, -\vec p_0)$. Con todo, **en el inercial del CM, el momento relativo entre las partículas, $\vec p_{rel}$, coincide con $\vec p_1$**, de modo que podemos estudiar la partícula efectiva, todas las magnitudes de la cual nos darán las relativas del sistema completo

- Ahora vemos la ventaja teórica que tiene calcular la sección eficaz diferencial en el inercial del CM: podemos hacerlo con las herramientas que ya sabemos, al bajo precio de cambiar $m\to \mu$.
