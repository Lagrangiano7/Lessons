# 1. El operador de Green

**Def.:** dado un operador funcional $F$, se define el operador de Green como aquel operador dependiente de un parámetro complejo $z$ tal que:

$$G_F(z)=(z-F)^{-1}; \ \ z\in D\subseteq \mathbb C$$

de tal forma que se cumpla la identidad: $G_F(z)(z-F)=I; \ \ \forall z \in D$

**Propiedad:** tal y como se construye el operador de Green, es evidente que si el operador $F$ es autoadjunto, entonces:

$$G(z^*)=G^\dagger (z)$$

## Analiticidad

**Def.:** diremos que un operador $G(z)$ es analítico si la función compleja correspondiente $\langle \chi |G(z)|\psi\rangle$ es analítica para cualesquiera estados $|\chi\rangle, |\psi\rangle$

**Observación:** vemos que, por fuerza, el operador de Green nunca será analítico en todo $\mathbb C$, porque cuando $z$ coincida con algún autovalor del operador, habrá un polo simple. Como en general el Hamiltoniano (que es el único operador para el que nos interesa usar el operador de Green) tiene espectro ligado (discreto) y espectro de scattering (continuo, con $E>0$), tendremos entonces que el resolvente correspondiente tendrá polos simples en $z=\{E_n\}$ y en $z=E>0$, de modo que se verá así:

![](Polos_G.png)

**Observación:** el operador de Green de cualquier Hamiltoniano con espectro discreto y continuo presenta polos simples cuando $z$ es igual a una energía del espectro. Esto implica naturalmente que posee un rayo en $z\in (0,\infty)$, que corresponde al espectro continuo

**Resultado:** no da lo mismo acercarse "por debajo" de la recta real que "por encima". Concretamente

$$\langle \chi |G_0(E_0+i0)|\psi\rangle - \langle \chi |G_0(E_0-i0)|\psi\rangle = -2\pi i\displaystyle\sum_{l,m}\langle \chi|E_0,l,m\rangle\langle E_0,l,m|\psi\rangle$$

**Nota:** véase que esto no es la identidad porque solo aparece un valor de la energía

## Ecuación de Lippman-Schwinger

**Observación:** un conocimiento completo del operador de Green asociado a un Hamiltoniano $H$ nos daría un conocimiento completo del espectro de dicho Hamiltoniano (basta con ver dónde tiene los polos $G(z)$, y ese será el espectro de $H$). Sin embargo, en la práctica es tan difícil el conocimiento total de los polos de $G(z)$ como lo es el del espectro de $H$. Por tanto, nos interesa conectar este operador con alguno otro que conozcamos completamente: $G_0(z)$

Partiendo de la identidad:

$$A^{-1}=B^{-1}+B^{-1}(B-A)A^{-1}$$

y haciendo $A=z-H=G^{-1}(z)$ y $B=z-H_0=G_0^{-1}(z)$, tenemos inmediatamente que:

$$\boxed{G(z)=G_0(z)+G_0(z)VG(z)}$$

o, identificando los operadores al revés:

$$\boxed{G(z)=G_0(z)+G(z)VG_0(z)}$$

# 2. El operador $T$

**Def.:** se define el operador $T$ como sigue:

$$\boxed{T(z):=V+VG(z)V}$$

**Propiedad:** como el potencial es real, encontramos que se cumple la misma propiedad que en $G(z)$:

$$T(z^*)= T^\dagger (z)$$

**Observación:** como $V$ es conocido, toda la información de $G(z)$ está contenida en $T(z)$ y viceversa

- Aplicando la ec. de Lipman-Schwinger, es inmediato encontrar una relación de $T(z)$ con el resolvente libre y el completo:

$$\boxed{G_0(z)T(z)=G(z)V}$$
$$\boxed{T(z)G_0(z)=VG(z)}$$

- Y una expresión de $G(z)$ en términos de $T(z)$:

$$\boxed{G(z)=G_0(z)+G_0(z)T(z)G_0(z)}$$

- Combinando estas expresiones, llegamos a una ecuación para $T(z)$ en términos de sí misma como teníamos con $G(z)$:

$$\boxed{T(z)=V+VG_0(z)T(z)}$$

**Nota:** esta ecuación se conoce como "ecuación de Lipppman-Schwinger para $T(z)$". Es el comienzo de muchos métodos para calcular $T(z)$. Por ejemplo, podemos resolver la ecuación por recurrencia, de tal forma que obtenemos la conocida como **serie de Born**:

$$T(z)=V+VG_0(z)V + \ldots$$

que puede converger o no. Luego esta serie se puede truncar en el n-ésimo término en mejor o peor aproximación dependiendo de lo débil que sea el potencial y obtener posteriormente una forma aproximada de la amplitud de scattering. En aproximación de Born, nos quedamos solo con el primer término, de modo que $T(z)=V$

**Problema que veo yo:** si admitimos que $T(z)=V$, estamos admitiendo que el operador $T(z)$ (estrictamente, cualquier función que se construya tomando elementos de matriz de este operador) no tiene polos en la recta real, cosa que sabemos que es falsa porque el Hamiltoniano completo tiene espectro. Supongo que será por esto por lo que el teorema óptico falla en aprox. de Born

**Idea:** podríamos quedarnos con aprox. de Born e introducir el siguiente término como una perturbación

# 3. Relación de $S$ con $T(z)$

- Deseamos ahora entender la relación entre el operador $T(z)$ (o equivalentemente por Lippman-Schwinger, el resolvente completo $G(z)$) con el operador de scattering, $S$

**Expresando los operadores de Moller en función de $T(z)$:**

- **Recuerda:** los operadores de Moller son aquel que coge el estado de la partícula en $t\to-\infty$ y lo lleva a $t=0$ ($\Omega_-$) y aquel que coge el estado de la partícula en $t=0$ y lo lleva a $t\to \infty$ ($\Omega_+$)
- Escribiendo la acción de los operadores de Moller como la integral de su derivada respecto del tiempo, llegábamos a que:

$$\Omega_-|\phi\rangle = |\phi\rangle +i\left[\int_0^\infty  U^\dagger(t)VU_0^\dagger(t) dt\right]|\phi\rangle$$

- Encontrábamos que esta integral era absolutamente convergente, en tanto que su norma también es convergente
- **Nota:** es sencillo ver que si una integral del tipo $\int_0^\infty |\Psi\rangle dt$ es absolutamente convergente, entonces $\displaystyle\lim_{\varepsilon \to 0^+}\int_0^\infty |\Psi\rangle e^{-\varepsilon t}dt$ también lo es

- Entonces, podemos reescribir la acción del operador de Moller $\Omega_-$ sobre cualquier ket $|\psi\rangle$ de esta forma:

$$\Omega_-|\psi\rangle = |\psi\rangle + i\left[\displaystyle\lim_{\varepsilon\to 0^+}\int_0^\infty U^\dagger(t)Ve^{-\varepsilon t}U_0^\dagger(t)dt\right]|\psi\rangle$$

**Observación:** este planteamiento es equivalente a sustituir el potencial $V$ por otro dependiente del tiempo $Ve^{-\varepsilon t}$. Sin embargo, como la dependencia temporal es exponencial con $\varepsilon\to 0^+$, el potencial varía muy lentamente. Tan lentamente que en la práctica, para cuando la partícula ha abandonado la zona donde el potencial es apreciable, a este no le ha dado tiempo de cambiar apreciablemente (aprox. adiabática es procedente con precisión arbitrariamente alta)

- Introduciendo la identidad de momentos en la igualdad e integrando en el tiempo, llegamos finalmente a que:

$$\Omega_-|\psi\rangle = |\psi\rangle + \displaystyle\lim_{\varepsilon\to 0^+}\int d^3\vec p G(E_p-i\varepsilon)V|\vec p\rangle \langle \vec p|\psi\rangle$$

Asimismo, en el caso del otro operador de Moller, tenemos que solo cambia el signo de $i\varepsilon$:

$$\Omega_+|\psi\rangle = |\psi\rangle + \displaystyle\lim_{\varepsilon\to 0^+}\int d^3\vec p G(E_p+i\varepsilon)V|\vec p\rangle \langle \vec p|\psi\rangle$$

**Nota:** este es el motivo de los subíndices de los operadores de Moller: coincide con el signo que lleva el término $i\varepsilon$. Véase que $\Omega_-$ se acerca a la recta real por debajo, mientras que $\Omega_+$ lo hace por arriba

- Ahora que conocemos cómo actúan los operadores de Moller en términos de $G(z)$ (alternativamente, a través de la ecuación de Lippman-Schwinger, de $T(z)$), podemos entender cómo son los elementos de matriz de $S$ en términos de dicho operador. Obtenemos lo siguiente:

$$\boxed{\langle \vec p'|S|\vec p\rangle = \delta^3(\vec p'-\vec p)-2\pi i\delta(E_{p'}-E_p)\displaystyle\lim_{\varepsilon\to 0^+}\langle \vec p'|T(E_p+i\varepsilon)|\vec p\rangle}$$

- Comparando con la expresión ya conocida de los elementos de matriz de $S$, deducimos entonces que **la matriz T on shell, $t(\vec p'\leftarrow \vec p)$, no es más que el elemento de matriz correspondiente a dichos momentos del operador $T$ cuando nos acercamos desde arriba a la recta real cuando $E_{p'}=E_p$**:

$$\boxed{t(\vec p'\leftarrow \vec p)=\displaystyle\lim_{\varepsilon\to 0^+}\langle \vec p'|T(E_p+i\varepsilon)|\vec p\rangle}$$

**Observación:** en el primer tema solo sabíamos que había una función analítica en los momentos que llamábamos "matriz T on shell" que apparecía en los elementos de matriz (de momentos) de $S$, y que estaba íntimamente relacionado con la amplitud de scattering. Sin embargo, solo conocíamos sus valores físicamente realizables. Nos faltaba información sobre dicha función en tanto que no sabíamos de dónde provenía. Ahora conocemos el operador entero del que proviene, así como sus propiedades y, lo más importante, cómo calcularlo (Lippman-Schwinger para $T$). A partir de ahí, podemos tomar elementos de matriz y evaluar el límite en el que $T$ pasa a estar on shell. La importancia de esto es que ahora tenemos una forma de calcular la amplitud de scattering, que hasta ahora habíamos dado por hecho

**Nota:** véase que el uso de $T$ solo atañe al caso en el que tomamos elementos de matriz en base de momentos. Si nuestro problema tuviera simetría de revolución, sería tanto más conveniente trabajar con ondas parciales, y nos evitaríamos tener que trabajar con la matriz T on shell porque $S$ sería entonces diagonal en la base de ondas esféricas
