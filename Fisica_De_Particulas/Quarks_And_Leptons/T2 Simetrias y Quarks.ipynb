# 1. Grupos y sus propiedades

**Def.:** un grupo es aquella estructura algebraica formada por un conjunto no vacío y una operación interna binaria definida dentro del conjunto tal que se cumplen 4 propiedades fundamentales:

1. El producto (aplicación de la operación interna) de cualesquiera dos elementos del grupo da otro elemento del grupo (cerrado bajo dicha operación interna)
2. Existe elemento neutro
3. Existe elemento inverso
4. El producto es asociativo

- Si además el producto es conmutativo, se dice que el grupo es abeliano

- La teoría de grupos sirve para tratar las simetrías, porque una operación de simetría (rotación, traslación, etc.) cumple exactamente las 4 propiedades de los grupos (i.e: la composición de dos rotaciones se puede expresar mediante una única rotación $R_3=R_2R_1$, el elemento neutro es no rotar, el elemento inverso es la rotación inversa y las rotaciones son asociativas. Sin embargo, rotaciones en torno a distintos ejes no conmutan y por ende el grupo de todas las rotaciones no es abeliano)


**Def.:** un grupo de Lie es aquel en el que las operaciones de grupo son funciones diferenciables (si el parámetro es real) o analíticas (si es complejo). Entiendo que al no especificar región de diferenciabilidad / analiticidad será en todo su dominio

- De este modo, las operaciones de simetría que conforman grupos de Lie se pueden expresar como sucesiones de operaciones infinitesimalmente pequeñas (i.e: podemos construir una rotación finita a partir de muchas rotaciones arbitrariamente pequeñas e incluso infinitesimales)

**Def.:** sea un grupo, por ejemplo el de rotaciones $R_1, R_2, \ldots$. Entonces, se llama "representación unitaria" de dicho grupo al grupo de operadores unitarios asociados al grupo de rotaciones, y que actúan sobre un espacio de Hilbert. La representación unitaria de un grupo es a su vez un grupo y tiene la misma estructura que el original. En el caso de las rotaciones, la representación unitaria de dicho grupo es el grupo de las matrices de Wigner

## Un sistema invariante bajo cierta operación de simetría conserva su generador

- Sea un sistema invariante bajo cierta operación de simetría (de Lie) $U(\varepsilon)$. Entonces, se cumple:

$$[H, U(\varepsilon)]=0; \ \ \forall \varepsilon$$

en todo el dominio del operador, que usualmente será la recta real

- En la imagen de Heisenberg, tenemos que:

$$i\frac{d}{dt}U=[H, U]=0$$

tomando valor esperado en ambos lados:

$$\frac{d}{dt}\langle \Psi(t)| U(\varepsilon)|\Psi(t)\rangle = 0$$

**Resultado:** el valor esperado de la operación de simetría es una constante del movimiento (ya lo sabíamos, puesto que conmuta con $H$, y admitimos que no depende explícitamente del tiempo)

Si en particular los estados que tomamos son los autoestados del generador, entonces:

$$\frac{d}{dt}\langle lm|e^{-iJ_3\theta}|lm\rangle = 0$$

y vemos que los autoestados del generador del operador de simetría se conservan.

**Resultado:** si un sistema es invariante bajo cierta operación de simetría, los autovalores del generador serán constantes del movimiento.

## Operadores de Casimir

**Def.:** un "operador de Casimir / invariante" es una función no-lineal de los generadores que conmuta con todos los generadores. En el caso de las rotaciones, solo existe uno, que es el módulo al cuadrado del momento angular:

$$J^2=J_1^2+J_2^2+J_3^2$$

tal que:

$$[J^2, J_i]=0; \ \ i=1, 2, 3$$

**Def.:** el número de generadores de un grupo que pueden ser diagonales simultáneamente se llama el "rango" del grupo. En el caso de $SU(2)$, como se ve en las matrices de Pauli, el rango es 1. En $SU(3)$, el rango es 2

**Teorema:** un grupo tiene tantos operadores de Casimir como rango tenga. Por eso $SU(2)$ tiene 1 solo operador de Casimir

# 2. El grupo $SU(2)$

- En la representación de menor dimensión (pero no trivial) del grupo de rotaciones ($j=1/2$), podemos escribir los generadores de dicho grupo como las matrices de Pauli:

- De este modo, los operadores de la representación unitaria de las rotaciones cuando $j=1/2$ se pueden escribir como:

$$U(\theta)=e^{-i\theta\sigma_i/2}$$

- Y naturalmente son unitarias. Sin embargo, como las matrices de Pauli tienen asimismo traza nula, se puede demostrar que tienen determinante unidad (que en $j=1/2$ es trivial de hacer cogiendo directamente las matrices de Pauli), de modo que dichas matrices pertenecen a un subgrupo de $U(2)$, a saber: $SU(2)$

**Def.:** el grupo $SU(2)$ es aquel que comprende todas las matrices 2x2 unitarias y con determinante unidad

**Fíjate:** existen infinitas representaciones de un mismo grupo, y lo que cambia es la dimensión. Por ejemplo, el grupo $SU(2)$ tiene representaciones en dimensión 1, 2, 3, 4, ... para, respectivamente, $j=0,1/2, 1, 3/2, \ldots$

**Def.:** la representación en dimensión 2 del grupo $SU(2)$ es su "representación fundamental", porque todas las demás representaciones se pueden construir a partir de ella


# 3. Combinando representaciones

- El producto de dos representaciones irreducibles de dimensión $2j_1+1$ y $2j_2+1$ puede asimismo expresarse como la suma de otras tantas representaciones irreducibles de dimensión $J$ (esencialmente, esto nos permite pasar de base desacoplada a base acoplada cuando acoplamos momentos angulares)
- Por ejemplo, si tenemos un sistema compuesto por dos subsistemas de momento angular $|j_1,m_1\rangle$ y $|j_2, m_2\rangle$, podemos expresar el estado del sistema completo como: $|j_1, m_1\rangle\otimes |j_2, m_2\rangle$, pero también podríamos expresarlo como suma de estados en la base acoplada, donde los coeficientes son los de Clebsch-Gordan

**Forma simbólica de expresar esta la equivalencia de producto y suma de representaciones irreducibles:** sean dos partículas de spin 1/2. Entonces, podemos escribir la descomposición de la base desacoplada en términos de la base acoplada de forma simbólica como:

$$2\otimes 2=3\oplus 1$$

donde los números hacen alusión a la dimensión de cada subespacio. A la izquierda está lo que sería la base desacoplada, donde cada partícula tiene asociado un subespacio de dimensión 2. Al pasar a la base acoplada, obtenemos 3 estados triplete y 1 singlete de spin (donde estos términos hacen referencia a la dimenión de cada nuevo subespacio, ahora en la base acoplada)

**Ejemplo:** si añadimos una tercera partícula de spin 1/2, entonces tenemos que: al acoplar las dos primeras, $J\in \{0,1\}$. Al acoplar el spin de la tercera partícula a cada uno de estos dos posibles valores del módulo del momento angular tenemos: $J=0+1/2=1/2$ y $J\in\{1/2, 3/2\}$. Por ende:

$$(2\otimes 2)\otimes 2 = 4\oplus 2\oplus 2$$

Es decir, un sistema de 3 partículas de spin $1/2$ da lugar a (1) un cuadruplete de spin 3/2 y (2) dos dobletes de spin 1/2

# 4. Grupos finitos: $\Pi$, $C$

- En la definición de grupo, en ningún momento se dice que el conjunto sobre el que se construye ha de ser infinito. De hecho, puede contener únicamente 2 elementos: el elemento identidad $e$, requerido por las propiedades definitorias del grupo, y un elemento $g$ que sea su propia inversa, tal que $g^2=e$ (or ejemplo, $g$ podría ser inversión de paridad o conjugación de partícula-antipartícula)
- Por esta definición de $g$, la representación unitaria del grupo contará únicamente con dos operadores: $I$ y $U(g)$. Además, se cumplirá:

$$U^2=I$$

- Porque la representación unitaria de un grupo tiene su misma estructura

**Propiedades:** $U$ es entonces (1) unitaria y (2) hermítica

## Espectro de $U$

- $U$ solo puede tener 2 autovalores: $\pm 1$

**Dem.:**

Sea un autovector de $U$ con autovalor $p$. Entonces, $U^2|p\rangle = p^2|p\rangle = I|p\rangle$. Pero entonces necesariamente: $p^2=1$. Resolviendo, tenemos que los autovalores son $p=\pm 1$

## Transiciones a estados de autovalor opuesto

- Si el sistema es invariante bajo cierta transformación "binaria" como la que hemos descrito, entonces no se permiten las transiciones a estados con autovalor opuesto (por ejemplo, en un sistema invariante bajo inversión de paridad, no se permiten las transiciones a 

**Dem.:** 

- Sean $|\phi_p\rangle, |\phi_i\rangle$ dos autoestados del operador paridad, donde los subíndices representan "par" e "impar", respectivamente. Como el sistema es invariante bajo inversión de paridad, tenemos $[H, \Pi]=0$, de modo que además los autoestados de la paridad lo son también del Hamiltoniano. Evaluemos el elemento de matriz:

$$\langle \phi_p|[H, \Pi]|\phi_i\rangle=0$$

Si lo desarrollamos, haciendo actuar el operador paridad donde toque, tenemos que:

$$E_i\langle \phi_p|\phi_i\rangle = 0 \Longrightarrow \text{Solo transiciones entre estados de igual paridad}_\blacksquare$$

## Interacciones y sus invariancias

- Los sistemas con interacción fuerte y EM son invariantes bajo inversión de paridad y bajo conjugación de carga (partícula --> antipartícula), y por ende se conservan sus números cuánticos, que son multiplicativos. Esto quiere decir que todo proceso EM o fuerte en el que la paridad total del sistema a la izda. no coincida con la paridad total a la dcha. es automáticamente prohibido y no sucede
- Los sistemas con interacción débil no son invariantes bajo ninguna de estas operaciones de simetría, como mostró el experimento de Wu para la inversión de paridad

# 5. El grupo $SU(2)$ de isospin

- Si solo estamos interesados en la interacción fuerte, entonces podemos pensar en el protón y el neutrón como dos posibles estados de una misma partícula (nucleón), debido a que sus masas son muy parecidas
- Considerarlas diferentes sería como distinguir con dos partículas distintas un electrón con spin up y otro con spin down
- Como en muy buena aprox. la fuerza nuclear fuerte (no tiene sentido de interacción fuerte porque los quarks ya están confinados) no distingue entre protones y neutrones, podemos formularlo como si fueran la misma partícula con dos estados posibles distintos: estado protón y estado neutrón

$$|p\rangle = \begin{pmatrix}1 \\ 0\end{pmatrix}; \ \ |n\rangle = \begin{pmatrix}0 \\ 1\end{pmatrix}$$

- Entonces, todo el formalismo de isospin es una copia exacta del formalismo de spin y recuperamos simetría $SU(2)$

- Entonces, es de suponer que también recuperamos la estructura de dicho grupo, de modo que el isospin tiene 3 generadores de grupo, con las relaciones de conmutación típicas de $SU(2)$:

$$[I_j, I_k]=i\varepsilon_{jkl}I_l$$

y donde $I_i=\frac{1}{2}\sigma_i$, como es de esperar

**Nota:** véase que esta representación del isospin totalmente equivalente al spin $1/2$ (que corresponde a la representación fundamental de $SU(2)$ por ser dimensión 2) es solamente posible porque el protón y el neutrón tienen asimismo $I=1/2$. A la partícula de mayor carga se le asigna por convenio el mayor valor de la tercera componente de spin y se va bajando

## Isospin para antipartículas

- Deseamos que los estados de antipartículas se transformen del mismo modo que los estados de partículas "ordinarias" bajo transformaciones de isospin, para poder después combinar estados de partículas ordinarias con antipartículas usando los mismos Clebsch-Gordan
- Podríamos pensar que para hacer esto basta con ver cómo se transforman los estados de partículas ordinarias, aplicar conjugación de carga y obtendríamos cómo se transforman los estados de antiprtículas. Hagámoslo. Rotemos por ejemplo el isospin un ángulo $\pi$ en torno al eje 2 para un estado arbitrario de isospin (partículas ordinarias):

$$\begin{pmatrix}p'\\ n'\end{pmatrix}= e^{-i\pi\sigma_2/2}\begin{pmatrix}p\\ n\end{pmatrix}=\begin{pmatrix}0&-1\\1&0\end{pmatrix}\begin{pmatrix}p\\ n\end{pmatrix}$$

- Si ahora aplicamos conjugación de carga $C$ a cada lado, que actúa asignando a cada partícula su antipartícula:

$$C|p\rangle = |\bar p\rangle; \ \ C| n\rangle = |\bar n \rangle$$

- Tenemos entonces que la expresión queda:

$$\begin{pmatrix}\bar p'\\ \bar n'\end{pmatrix}= e^{-i\pi\sigma_2/2}\begin{pmatrix}\bar p\\ \bar n\end{pmatrix}=\begin{pmatrix}0&-1\\1&0\end{pmatrix}\begin{pmatrix}\bar p\\ \bar n\end{pmatrix}$$

- Sin embargo, como la conjugación de carga invierte todos los números cuánticos asociados a una partícula, el estado de mayor tercera componente de isospin no es $\bar p$ sino $\bar n$. Entonces, debemos cambiar el orden de la matriz columna para que el primer ket de la base sea, como en el caso de las partículas ordinarias, el de mayor tercera componente de isospin. A eso nos referimos con que pedimos que partículas y antipartículas se comporten igual bajo rotaciones de isospin. También añadimos un signo "-" al coeficiente del primer ket de la base para que la transformación quede igual en ambos casos:

$$\begin{pmatrix}-\bar n'\\ \bar p'\end{pmatrix}= e^{-i\pi\sigma_2/2}\begin{pmatrix}-\bar n\\ \bar p\end{pmatrix}=\begin{pmatrix}0&-1\\1&0\end{pmatrix}\begin{pmatrix}-\bar n\\ \bar p\end{pmatrix}$$

**Nota:** véase que, independientemente de si hacemo este cambio en orden y signo o no, obtenemos las mismas ecuaciones. La clave está en ver que es el orden de los kets de la base el que intentamos respetar. Tendremos los mismos Clebsch-Gordan si los estados siempre tienen la forma:

$$\begin{pmatrix}\text{$\alpha(|I_{3,max}\rangle)$}\\ \text{$\beta(|I_{3, min}\rangle)$}\end{pmatrix}$$

**Nota:** esto se puede hacer en $SU(2)$ pero no en $SU(3)$. No hay forma de reordenar y manipular los signos de un estado en $SU(3)$ para que las antipartículas se transformen bajo rotaciones de isospin de igual manera que las partículas

# 6. El grupo $SU(3)$

- El grupo $SU(3)$ es aquel grupo comprendido por todas las matrices unitarias 3x3 con determinante unidad
- **Importante:** sus generadores son cualesquiera **8** matrices (1) hermíticas, (2) de traza nula y (3) linealmente independientes entre sí (esto se debe a que empiezas con una matriz 3x3 general y tienes 18 parámetros reales, 2 por cada una de las 9 entradas porque en general estas serán complejas. Impones hermiticidad y traza nula y se te quedan 8 parámetros reales, así que necesitas 8 generadores lin. indep. con esas 2 propiedades)
- Estos generadores se suelen tomar como las matrices de Gell-Mann, que son el análogo de SU(3) de las matrices de Pauli en SU(2). Hay 2 diagonales y se suelen tomar la tercera y la octava:

$$\lambda_i=\begin{pmatrix}1 & & \\ & -1 & \\ & & 0\end{pmatrix}; \ \ \lambda_8=\begin{pmatrix}1 & & \\ & 1 & \\ & & -2\end{pmatrix}$$

**Nota:** las 3 cargas de color: R, G, B forman la representación fundamental del grupo $SU(3)$. Tomamos la base como:

$$R=\begin{pmatrix}1 \\ 0 \\ 0\end{pmatrix}; \ \ G=\begin{pmatrix}0 \\ 1 \\ 0\end{pmatrix}; \ \ B=\begin{pmatrix}0 \\ 0 \\ 1\end{pmatrix}$$

- Las matrices de Gell-Mann 1, 2 y 3 coinciden con las matrices de Pauli (con un 0 en el subespacio restante) y por ende reflejan un subgrupo $SU(2)$ de $SU(3)$
- **Nota:** las combinaciones lineales de las matrices de Gell-Mann no diagonales dan los equivalentes a los operadores escalera, solo que ahora podemos "movernos en 2 direcciones" porque tenemos 3 estados en la base. Esto es, con la "partícula nucleón" o con una sola partícula con spin, aplicar el operador escalera correspondiente a un elemento de la base nos da el otro y ya hemos acabado. Ahora, sin embargo, podemos pasar de G --> B ó R etc. como se ve en la imagen:



**Importante:** las relaciones de conmutación de las matrices de Gell-Mann son:


$$[\lambda_i, \lambda_j]=2i\displaystyle\sum_kf_{ijk}\lambda_k$$

donde los coeficientes $f_{ijk}$ reciben el nombre de "constantes de estructura". En el caso de las matrices de Pauli, serían los coeficientes del símbolo de Levi-Civita en 3 dimensiones

**Propiedad:** las constantes de estructura de SU(3) son completamente antisimétricas (como Levi-Civita) y aquellas que no se anulan son permutaciones de los índices de:

$$f_{123}=1; \ \ f_{458}=f_{678}=\sqrt 3/2; \ \ f_{147}=f_{165}=f_{246}=f_{257}=f_{345}=f_{376}=1/2$$

# 7. Extrañeza y el grupo $SU(3)$ de sabor

- En 1947 se descubre el pion y con él el nucleón pierde el rol central que tenía en física de partículas: conocíamos ahora un segundo hadrón (se entiende hadrón como una partícula que siente la interacción fuerte)
- Algunos de los nuevos hadrones descubiertos tenían una vida media sorprendentemente alta para ser partículas que se desintegran a través de la interacción fuerte, ya que estas vidas medias suelen estar en torno a $10^{-23}(s)$ (por regla general, cuanto más fuerte es una interacción, menos viven las partículas que por ella decaen)
- **Solución (Gell-Mann, Nishijima):** existe un nuevo número cuántico aditivo además de $I_3$ que llamaremos "extrañeza" $S$ que debe ser conservado en las interacciones fuertes y EM, pero no en la débil
- De este modo, el barión $\Sigma$ se puede crear a través de la interacción fuerte conservando extrañeza: $\pi^-+p\longrightarrow K^++\Sigma^-$, pero no se puede desintegrar a través de la interacción fuerte conservando extrañeza: $\Sigma^-\longrightarrow \Delta+\pi^-$ porque está cinéticamente prohibido. En su lugar, solo puede desintegrarse a través de la interacción débil, a la que le da igual la extrañeza (la viola) y esto "explica" por qué $\Sigma$ vive tanto tiempo: no se desintegra por la interacción fuerte como pensábamos, solo por la débil

- El siguiente paso lógico era entonces expandir el grupo $SU(2)$ de isospin para acomodar este nuevo grado de libertad que era la extrañeza. Había que pasar a un grupo de rango 2
- Se propuso (y 3 años después se confirmó) que el candidato correcto era $SU(3)$, de modo que podíamos agrupar mesones y bariones (por separado) con propiedades similares (en esencia, masas similares) dentro de representaciones en octetes de $SU(3)$
- Sin embargo, las variaciones de masa en este octete son de $400MeV$ (lógicamente, si la simetría fuera exacta, entonces todos tendrían la misma masa, al igual que un electrón (libre) no gana ni pierde energía por tener spin up o down)
- Ahora que tenemos la teoría quark sabemos que la simetría $SU(2)$ de isospin es tan buena (si bien tampoco es exacta, ya que protón y neutrón tienen masas que no son idénticas) porque la masa de los quarks $u$ y $d$ es parecida. Sin embargo, al introducir el nuevo quark $s$, más pesado por ser de la generación siguiente, la simetría ya no es tan buena
- Usamos el grupo $SU(3)$ de sabor para catalogar los distintos hadrones, y lo llamamos "grupo SU(3) de sabor". No guarda relación con el grupo SU(3) de color que hemos visto antes, y este último sí que se cree que es una simetría exacta (un quark rojo tiene la misma masa que uno azul o uno verde)

# 8. Estados ligados de quarks

- Hemos dicho entonces que el grupo de los hadrones conocidos por aquel entonces tenía simetría SU(3) de sabor aunque esta no se sostuviera bien
- La representación fundamental de SU(3), a partir de la cual todas las otras representaciones (como octetes, etc.) se pueden formar es un triplete (como en SU(2) era el doblete p/n), y son precisamente los quarks u, d, sç
- **Def.:** se define la cantidad aditiva "hipercarga" como:

$$Y=B+S$$

donde:

- $Y$: hipercarga
- $B$: número bariónico
- $S$: extrañeza

Hablar de hipercarga en lugar de extrañeza no tiene ninguna relevancia física, sencillamente sitúa el triángulo del triplete en el origen cuando lo representamos con respecto al eje $I_3, Y$

- La carga de cada quark es entonces:

$$Q=I_3+\frac{Y}{2}$$

**Nota:** los quarks mantienen su identidad en las interacciones EM y fuerte. Transiciones del tipo $s\to u+\text{leptones}$ solo se dan a través de la interacción débil

## 8.1 Mesones

**Def.:** un mesón es un estado ligado de quark-antiquark

## El nonete de mesones

- Como estamos considerando los 3 quarks más ligeros, podemos preguntarnos cuántos mesones podemos construir con ellos
- Haciendo una tabla con filas $uds$ y columnas $\bar u\bar d\bar s$, tenemos 9 combinaciones distintas
- Tambiñén podíamos haber cogido el triángulo de antiquarks y situado su baricentro sobre cada vértice del triángulo de quarks ordinarios, formando así el famoso hexágono que da lugar al nonete de mesones
- Obtenemos un octete de SU(3) y un singlete de SU(3), lo que significa que podemos relacionar 8 de los 9 estados entre sí a través de transformaciones de isospin, pero el noveno está desconectado de todos los demás
- **Nota:** los hadrones que observamos han de ser superposiciones de los distintos estados de quarks (por ejemplo, si se acoplan a spin 0 ó 1, se consideran diferentes hadrones)

## Paridad del mesón

- Si consideramos el mesón libre, entonces su parte angular espacial irá con un armónico esférico y por ende la inversión de paridad le afectará como:

$$\boxed{\Pi|q\bar v\rangle = -(-1)^L|q\bar v\rangle}$$

donde el "-" de delante viene de que los quarks y antiquarks tienen paridades opuestas

**Resultado:** un estado mesónico (libre) es autoestado del operador de inversión de paridad

## Conjugación de carga en mesones

**Observación:** si un mesón está compuesto por un quark y su correspondiente antiquark, entonces es un autoestado del operador de conjugación de carga (i.e: es su propia antipartícula). El autovalor correspondiente será:

$$\boxed{C|q\bar q\rangle = (-1)^{L+S}|q\bar q\rangle}$$

El motivo de esto es el siguiente: el operador conjugación de carga pasa una partícula a su antipartícula correspondiente, pero en esencia lo que está haciendo es permutar dos partículas. Por ende, sale (1) un signo "-" por permutar dos fermiones, (2) un signo $(-1)^{S+1}$ por la permutación de dos spines 1/2, donde el triplete de spin $S=1$ es simétrico y no saca signo y el singlete de spin $S=0$ sí lo saca, y finalmente la parte de inversión espacial. Por ende, tenemos $-(-1)S^{S+1}(-1)^L=(-1)^{L+S}$


Si lo quieres ver así, puedes construir el operador de conjugación de carga actuando sobre un estado $|q\bar q\rangle$ invirtiendo paridad y permutando spines, de modo que tienes la imagen espejada tanto en el espacio como en spines, que es lo que sucedería si pasásemos el quark ordinario a antiquark y el antiquark a quark ordinario:

$$C|q\bar q\rangle \equiv P_S\Pi|q\bar q\rangle$$

## 8.2 Bariones

**Def.:** un barión es el estado ligado conformado por 3 quarks (ordinarios)

La descomposición en sabor es ahora:

$$3\otimes 3\otimes 3=10\oplus 8\oplus 8\oplus 1$$

donde los bariones más ligeros se acomodan en el octete (spin 1/2) y decuplete (spin 3/2, totalmente simétrico)

**Problema:** la existencia del decuplete parece que viola el principio de exclusión de Pauli, porque predice la existenia de bariones donde 2 ó incluso los 3 quarks que los componen se encuentran en el mismo estado cuántico. Por ejemplo: $\Delta^{++}=u\uparrow u\uparrow u\uparrow$

**Solución:** los quarks tienen un grado de libertad adicional llamado "color". OJO, esto no los hace distinguibles, al igual que los 3 quarks tienen el spin como grado de libertad y no por ello son distinguibles. Lo que pasa es que siguen siendo indistinguibles pero los quarks están en distintos estados de color (hay 3)

**IMPORTANTE:** todos los hadrones son "incoloros" (o se cancela el color de los dos quarks como en mesones o tenemos quarks d los 3 colores como en bariones). Por tanto, los bariones están en un estado singlete de color:

$$\boxed{|qvw\rangle_{color}=\frac{1}{\sqrt 6}(RGB-RBG+BRG-BGR+GBR-GRB)}$$

**Observación:** el singlete de color es totalmente antisimétrico, de modo que el resto del vector de estado del barión: sabor+spin+espacio ha de ser en total, totalmente simétrico

**Nota:** como todos los bariones son singletes de color, todos tienen el mimo ket de color, así que lo omitimos a la hora de dar el vector de estado, pero recordamos que solo valen aquellos vectores de estado que sean simétricos para la completitud de los demás grados de libertad

- Para saber cuántos estados de bariones podemos generar, debemos recordar que sus vectores de estado tendrán esencialmente parte de sabor y parte de spin, de modo que $|barion\rangle = |sabor\rangle|\otimes|spin\rangle$. Como sabemos que la parte de sabor obedece $SU(3)$ y la parte de spin $SU(2)$, tenemos que nuestros posibles vectores de estado se descompondrán según:

$$(10+8+8+1), \ \ (4+2+2)$$

donde la primera parte da cuenta de $3\otimes3\otimes 3$ del sabor y la segunda de $2\otimes 2\otimes 2$ del spin 1/2 de cada quark

**OJO!:** aunque un quark sea down y otro sea up, ambos siguen siendo indistinguibles! Es solo que uno está en estado down y otro en up, como puede ser el caso del spin up y down. Por ende, los vectores de onda que nos queden han de ser simétricos bajo intercambio de los 3 quarks. Si uno fuera indistinguible por ser d y los otros dos u, por ejemplo, entonces sería $|uu\rangle\otimes|d\rangle$

**Nota:** los estados totalmente simétricos (que son los físicamente realizables) son aquellos correspondientes a los vectores de onda simétricos. Como el decuplete de sabor es totalmente simétrico y el cuadruplete de spin 3/2 también, una posibilidad es tener un decuplete de bariones de spin 3/2, como se observa. La otra posibilidad para dar estados totalmente simétricos es juntar un octete de bariones con cierta simetría mixta con el doblete de spin de igual simetría y sumarlos, de modo que queda:

$$|\text{octete de bariones}\rangle = \frac{1}{\sqrt{2}}[(8,2)+(8,2)]$$

donde el primer $(8,2)$ corresponde a combinaciones del octete simétrico mixto y al doblete de spin simétrico mixto, y el segundo a sus equivalentes antisimétricos mixtos (recuerda que son simétricos / antisimétricos bajo intercambio de la primera y segunda partícula, pero la tercera le sigue al sumarlos)


**Ejemplo: vector de estado de protón con spin up (sin parte espacial)**

- Se trata de una partícula del octete de bariones (spin 1/2), por lo que:

$$|p\uparrow \rangle=\frac{1}{\sqrt 2}(p(M_S)\chi(M_S)+p(M_S)\chi(M_S))$$

desarrollando, se nos queda que:

$$|p\uparrow\rangle=\frac{1}{\sqrt 18}[uud(\uparrow\downarrow\uparrow +\downarrow\uparrow\uparrow - 2\uparrow \uparrow \downarrow)+udu(\uparrow \uparrow \downarrow + \downarrow\uparrow\uparrow - 2\uparrow \downarrow \uparrow)+duu(\uparrow \downarrow \uparrow+\uparrow \uparrow \downarrow -2\downarrow\uparrow\uparrow )]$$

de forma más concisa:

$$|p\uparrow\rangle = \frac{1}{\sqrt 18}[u\uparrow u\downarrow d\uparrow + u\downarrow u\uparrow d\uparrow -2u\uparrow u\uparrow d\downarrow + \text{ permutaciones}]$$

- Naturalmente, el estado de spin se ha conseguido buscando aquellas combinaciones de 3 spines 1/2 que den spin neto 1/2. Se me ocurre que esto es fácil acoplando primero 2 spines 1/2 para obtener el famoso triplete de spin 1 y el singlete de spin 0 y después volviendo a acoplar, y quedándome solo con aquellos estados que tengan tercera componente hacia arriba. En la parte de sabor sería exactamente igual pero cambiando $\uparrow \to u, \downarrow \to d$ porque estamos trabajando en una representación de $SU(2)$ dentro de $SU(3)$ (esto quiere decir que estamos usando los quarks u,d pero no s)
  

# 9. Momentos magnéticos

- El momento magnético de una partícula puntual de carga $Q$ es:

$$\mu = \frac{Q}{2m}$$

- El momento magnético es una magnitud aditiva, de modo que el del protón será sencillamente:

$$\mu(p)=\displaystyle\sum_{i=1}^3 \langle p\uparrow |\mu_i|p\uparrow\rangle$$

donde por convenio el momento magnético se mide en estados de tercera componente de spin máxima ($M_J=+J$)

- Usando la función de onda explícita del vector de estado del protón y habida cuenta que el spin down provoca que el momento magnético esté invertido, tenemos:

$$\mu(p)=3\times\frac{1}{18}[(\mu_u-\mu_u+\mu_d) + (-\mu_u+\mu_u+\mu_d) + 4(\mu_u+\mu_u-\mu_d)]$$

donde el factor 3 está porque las permutaciones dan las mismas contribuciones que la parte del vector de estado mostrada

Cancelando términos:

$$\mu_p=\frac{1}{3}(4\mu_u-4\mu_d)$$

**Nota:** el momento magnético del neutrón se obtiene sencillamente cambiando $u$ por $d$ porque su composición de quarks también se puede obtener así a partir de la del protón

- Si hacemos la aprox. $m_p=m_n$ (en el fondo, $m_u=m_d$) tenemos que $\mu_u=-2\mu_d$ por el tema de las cargas, de modo que llegamos a la sencilla relación entre el momento dipolar del neutrón y el del protón:

$$\boxed{\frac{\mu_n}{\mu_p}\approx -\frac{2}{3}}$$

que está en buen acuerdo con los resultados experimentales: $-0.68497945\pm 0.00000058$ pero también refleja que la masa de ambos quark no es idéntica. Además, el momento magnético así calculado solo vale en el límite no relativista, y como los quarks que componen al protón son los más ligeros, su estado ligado es bastante relativista

# 10. Masa de los hadrones

## Masa efectiva de los quarks

- Podemos dar una estimación de la masa de los quarks constituyentes de un hadrón repartiendo su masa entre estos y sabiendo que $m_u\approx m_d$ (la simetría de isospin es muy buena, no lo sería si los dos quarks no tuvieran masas muy próximas. Puedes pensar que es porque las masas próximas son las del neutrón y la del protón y estarías en lo cierto, pero como la interacción fuerte no distingue entre quarks up y down, podemos decir que son las masas de los quarks las que se parecen)
- De este modo, tenemos las siguientes **masas de quarks constituyentes** (se llaman así porque realmente estas son sus masas efectivas cuando se encuentran en estado ligado formando un hadrón, no tiene por qué coincidir con sus masas libres, si siquiera es posible liberarlos)

$$m_u\approx m_d\approx 0.39GeV$$

$$m_s\approx 0.51GeV$$

$$m_c\approx 1.6GeV$$

## Confirmación de que son partículas puntuales

- Si ahora repetimos el cálculo de las masas a partir de la expresión de los momentos magnéticos calculados previamente y haciendo la hipótesis de que son partículas puntuales, de modo que su momento magnético es $\mu_i=Q_i/{2m_i}$, obtenemos masas muy similares a las obtenidas por el otro procedimiento:

$$m_u\approx 0.34GeV$$
$$m_s\approx 0.51GeV$$

Vemos que conforme los quarks se van haciendo más pesados, tanto mejor es la estimación de su masa

- Como obtenemos los mismos valores (en muy buen acuerdo) mediante ambos métodos, reforzamos nuestro entendimiento de que los quarks son en efecto partículas puntuales (carentes de estructura y por ende fundamentales)

**Nota:** estas masas que hemos estimado se deben entender como la masa de los quarks en estado fundamental cuando se encuentran ligados por un potencial como el de la interacción fuerte. Este potencial ha de ser Coulombiano a cortas distancias, porque la QCD se comporta como la QED a altas energías, y confinante a largas distancias (1 fermi) porque los quarks no pueden escapar ($\neq E \ \text{tq.} E>V(r) \ \forall r>0$). Entonces, debe ser como:

$$V(r)=-\frac{4}{3}\frac{\alpha_S}{r}+\alpha r$$

- Donde $4/3$ es el factor de color asociado con el factor de acoplamiento quark-gluon $\alpha_S$
- **Nota:** el potencial de la interacción fuerte se estudia con el charmonio, viendo qué niveles están degenerados y a qué clase del potencial le pertenece dicho espectro en EM (el charmonio, por ser el quark charm tan pesado, es un estado ligado no relativista)

## ¿Por qué dos hadrones de igual contenido en quarks pero con distinto spin tienen distinta masa?

- Esto se debe a que, al igual que en QED, en QCD las fuerzas son dependientes del spin, y en QCD también
- Por ende, existe una corrección de estructura hiperfina (un acoplo entre los spines de los quarks) que produce una contribución adicional a la masa además de la efectiva de los quarks, y que puede añadir o restar masa
- En el caso del átomo de hidrógeno:

$$\Delta E_{hf}=\frac{2\pi\alpha}{3}\frac{\vec \sigma_1\cdot \vec \sigma_2}{m_1m_2}|\Psi(0)|^2$$

- Podemos copiar este resultado en QCD pasando de la constante de estructura fina a la constante de acoplo de QCD:

$$\alpha \to \begin{cases}\frac{4}{3}\alpha_S \ \ (\text{mesones}) \\ \frac{2}{3}\alpha_S \ \ (\text{bariones})\end{cases}$$

donde los factores numéricos se llaman "factores de color", que aprenderemos a calcular en el apartado siguiente

Suponiendo que:

1. El confinamiento de los quarks es independiente de sus spines y masas
2. A cortas distancias, $\alpha_S$ es suficientemente pequeño como para que la corrección de estructura hiperfina sea relevante
3. La única ruptura de simetría viene de que los quarks de distinto sabor tienen distinta masa

Entonces podemos calcular la masa de los hadrones como la suma de las masas de los quarks constituyentes y la contribución de estructura hiperfina:

$$m(q_1\bar q_2)=m_1+m_2+a\frac{\vec\sigma_1\cdot \vec \sigma_2}{m_1m_2}$$
$$m(q_1q_2q_3)=m_1+m_2+m_3+\left[\frac{a'}{2}\displaystyle\sum_{i>j}\frac{\vec \sigma_i\cdot \vec \sigma_j}{m_im_j}\right]$$

# 11. Factor de color

- En QED, la intensidad del acoplo EM entre dos quarks viene dada por el factor $e_1e_2\alpha$. Es esperable que, siguiendo con nuestra analogía, en QCD la intensidad del acoplo venga dada por el producto de dos factores intrínsecos del quark ($c_1, c_2$) y una constante de acoplo ($\alpha_S$)
- **Def.:** el "factor de color" es:

$$C_F=\frac{1}{2}|c_1c_2|$$

- La intensidad del acoplo fuerte entre dos quarks vendrá dada en analogía a QED por:

$$\frac{1}{2}c_1c_2\alpha_S$$

## Cálculo del factor de color en distintas interacciones

- El factor de color se calcula a partir de los posibles diagramas de Feynman a primer orden en QCD
- **Recuerda:** los gluones forman un octete de SU(3). Hay 8 gluones, y el noveno que esperaríamos obtener (recuerda que un gluón tiene color y anticolor, de modo que hay 9 combinaciones a priori) no tiene color, de modo que no puede representar un gluón. Los gluones posibles son los siguientes:

$$R\bar G, R\bar B, G\bar R, B\bar R, B\bar G, \frac{1}{\sqrt 2}(R\bar R-G\bar G), \frac{1}{\sqrt 6}(R\bar R+G\bar G-2B\bar B)$$

El estado singlete de color es aquel que representa todos los colores por igual y por ende no tiene color:

$$\frac{1}{\sqrt 3}(R\bar R+G\bar G+B\bar B)$$

- Ahora que sabemos qué quarks podemos usar, ya podemos ponernos a calcular factores de color

## Interacción de 2 quarks azules (B)

**Clave:** como el color se conserva en cada vértice, solo podemos seleccionar aquel gluón que, en este caso, tenga $B\bar B$. El factor de color será entonces el coeficiente al cuadrado de la parte del gluón con los colores deseados. Como el nuestro tiene factor $\frac{-2}{\sqrt 6}$, tenemos que:

$$c_1c_2=\frac{2}{3}$$

**Importante:** habrá veces en las que más de un diagrama de Feynman contribuya al cálculo a primer orden. En este caso, las coeficientes $c_1c_2$ calculados con distintos diagramas se sumarán si la función de onda **de color** es simétrica bajo intercambio de partículas y se restarán si es antisimétrica. Esto se puede reducir a una sencilla fórmula:

$$\boxed{c_1c_2=P-\frac{1}{3}}$$

donde $P= 1$ si la función de onda de color es simétrica bajo intercambio de partículas y $P=-1$ si es antisimétrica

**Nota:** esta fórmula solo sirve para sistemas con 2 quarks ordinarios (o 2 antiquarks si le meto un signo "-", entiendo yo). Los antiquarks introducen un signo "-" en los vértices y por ende requieren un trato más delicado

## Interacción de 2 quarks rojos (R)

- En este caso tenemos 2 posibles gluones: $\frac{1}{\sqrt 2}(R\bar R-G\bar G), \frac{1}{\sqrt 6}(R\bar R+G\bar G-2B\bar B)$
- Sin embargo, veremos que el factor de color sale el mismo que en la interacción de 2 quarks azules, como dicta la simetría de color
- Como se trata de un sistema de 2 quarks de igual color, sabemos que la parte de color de la función de onda es simétrica (es un caso teórico, en la práctica no se dará porque no está en un singlete de color)
- Tenemos entonces las siguientes amplitudes:

$$(c_1c_2)_1=\frac{1}{\sqrt 6}\frac{1}{\sqrt 6}; \ \ \ (c_1c_2)_2=\frac{1}{\sqrt 2}\frac{1}{\sqrt 2}$$

Como la función de onda de color es simétrica bajo el intercambio de los dos quarks, las amplitudes se suman, y tenemos:

$$c_1c_2=(c_1c_2)_1(c_1c_2)_2=\frac{2}{3}$$

Como obteníamos en el caso de la interacción de 2 quarks azules

## Factor de color de un mesón (interacción quark-antiquark)

- Necesariamente, el quark y el antiquark, que forman un mesón, se deben encontrar en un singlete de color
- **IMPORTANTE:** el vértice de la antipartícula introduce un signo "-". En QED es más familiar porque sabemos que la carga de la antipartícula es la opuesta a la de la partícula ordinaria, pues enn QCD igual pero con los coeficientes $c$
- Como en el mesón todos los colores tienen lugar por igual: $|\chi_{color}\rangle = \frac{1}{\sqrt 3}(R\bar R+G\bar G+B\bar B)$, tenemos que evaluar 9 diagramas a primer orden en total ($B\bar B, B\bar R, B\bar G$ y los mismos para los otros 2 colores)
- Sin embargo, por la simetría de color, basta con estudiar los 3 diagramas correspondientes a cualquiera de los colores (por ejemplo B) y multiplicar la amplitud obtenida por 3, porque sabemos que obtendremos lo mismo
- Los 3 diagramas de Feynman a evaluar son (recuerda: cada vértice conserva color):

![Amplitud de interacción de mesones](Interaccion_QCD_mesones.jpg)

- Por ser la parte de color simétrica (ha de serlo porque es un mesón), tenemos entonces que las amplitudes se suman:

$$c_1c_2=3\times \frac{1}{\sqrt 3}\frac{1}{\sqrt 3}(-\frac{2}{3}-1-1)=-\frac{8}{3}$$

donde las dos raíces cuadradas vienen de la función de onda del propio mesón. Hasta ahora no habían aparecido términos relacionados con la función de onda porque siempre habían sido casos de un mismo color, no superposiciones de varios

**Resultado:** el factor de color de la interacción a primer orden de un mesón es:

$$C_F=\frac{4}{3}$$

## Factor de color de un barión (interacción quark-antiquark en un barión)

- El problema ya no es el mismo, porque ahora la función de onda de color del par quark-antiquark ha de ser distinta para que al acomodar otro quark la función de onda de color siga quedando simétrica (recuerda que es un barión y por tanto incoloro)
- Aprovechando que ahora tenemos la interacción de 2 quarks, podemos usar la útil fórmula que hemos presentado antes, con $P=-1$ porque ahora la función de onda de los 2 quarks ha de ser antisimétrica (para conseguir una simétrica al meter el tercer quark) y por ende el factor de color de los bariones es:

$$C_F=\frac{2}{3}$$
