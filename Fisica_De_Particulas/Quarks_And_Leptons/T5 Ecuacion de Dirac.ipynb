# 1. Desarrollando la ecuación de Dirac

- Recordemos que Dirac trataba de solventar el problema de la energía negativa construyendo una ecuación que fuera lineal en las derivadas (si lo es la temporal lo tiene que ser también en la espacial para que siga siendo covariante). En general, tenemos entonces una ecuación de la forma:

$$H\Psi=(\vec \alpha\cdot \vec P+\beta m)\Psi$$

donde los coeficientes $\vec \alpha$ y $\beta$ podrían en general matrices

Sabemos que las partículas deben cumplir la ecuación de la energía de Einstein (si se quiere, siempre podemos asignarle un cuadrimomento a una partícula dada), de modo que al aplicar 2 veces el Hamiltoniano debemos recuperar:

$$H^2\Psi=(\vec P^2+m^2)\Psi$$

Esto nos permite determinar quiénes son los coeficientes de nuestro modelo de ecuación:

$$H^2\Psi=(\alpha_i^2P_i^2+(\alpha_i\alpha_j+\alpha_j\alpha_i)P_iP_j+(\alpha_i\beta+\beta\alpha_i)P_i m+\beta^2m^2)\Psi$$

donde se ha usado el convenio de sumas con índices y donde en el segundo término $i>j$. Comparando, tenemos:

1. $\alpha_1, \alpha_2, \alpha_3, \beta$ anticonmutan todos entre sí
2. $\alpha_1^2= \alpha_2^2=\alpha_3^2=\beta^2=1$

Observamos que efectivamente los coeficientes no conmutan, como sin duda harían si fueran numéricos, de modo que deben ser matrices


**Observación:** las matrices de menor dimensión que cumplen estas 2 propiedades son 4x4 incluyendo las matrices de Pauli o la identidad. Hya varias matrices que cumplen estas propiedades, y todas se pueden meter en la ecuación de Dirac, pero la Física no cambia porque lo único que necesitamos es que nuestra ecuación de Dirac cumpla la condición de recuperar la energía de Einstein cuando se aplica 2 veces. Esto da lugar a varias "representaciones":

- En la representación de **Dirac-Pauli** tenemos que los coeficientes de la ecuación de Dirac son de la forma:

  $$\vec \alpha=\begin{pmatrix}0 & \vec \sigma \\ \vec \sigma & 0\end{pmatrix}; \ \ \beta = \begin{pmatrix}I & 0 \\ 0 & -I\end{pmatrix}$$

- En la representación de Weyl, los coeficientes toman la forma:

  $$\vec \alpha=\begin{pmatrix}-\vec \sigma & 0 \\ 0 & \vec \sigma\end{pmatrix}; \ \ \beta = \begin{pmatrix}0 & I \\ I & 0\end{pmatrix}$$

**Nota:** siempre usaremos la representación de Dirac-Pauli

**Resultado:** las soluciones de la ecuación de Dirac han de ser matrices columna de 4 elementos! A dichas matrices columna se les llama **espinores**



# 2. Forma covariante de la ecuación de Dirac y matrices $\gamma$ de Dirac

- Si multiplicamos por $beta$ a ambos lados de la ecuación de Dirac y hacemos la sustitución de siempre: $H\to i\frac{\partial}{\partial t}; \ \ \vec p\to -i\vec \nabla$, tenemos que la ecuación de Dirac queda de la forma:

$$i\beta\frac{\partial\Psi}{\partial t} = -i\beta\vec\alpha\cdot \vec \nabla \Psi+m\Psi$$

pasando a la izquierda el primer término de la derecha, podemos escribir la ecuación de Dirac en su forma habitual:

$$\boxed{(i\gamma^\mu\partial_\mu-m)\Psi=0}$$

done se ha comprimido la notación usando las matrices $\gamma$ de Dirac

**Def.:** $\boxed{\gamma^\mu = (\beta, \beta\vec\alpha)}$

Vemos que su estructura se parece mucho a la de un cuadrivector, pero ahora los valores que tiene este "cuadrivector" son matrices. Por ende, las matrices $\gamma$ de Dirac no son propiamente un cuadrivector. En la sección siguiente explicaremos en qué sentido se pueden entender como un cuadrivector

**Notación:** habitualmente, en lugar de poner explícitamente el producto escalar $\gamma^\mu \partial_\mu$ se escribe $\cancel\partial$. De este modo, la ecuación de Dirac se suele presentar de forma incluso más concisa, simbólica, como:

$$(i\cancel\partial-m)\Psi=0$$

## Propiedades de las matrices $\gamma$

1. Satisfacen la siguiente relación de anticonmutación:

$$\gamma^\mu\gamma^\nu+\gamma^\nu\gamma^\mu=2g^{\mu\nu}$$

2. Como $\gamma^0=\beta$, que es una matriz con identidades, se cumple que:

$$(\gamma^0)^\dagger =\gamma^0; \ \ \ (\gamma^0)^2=I$$

3. Las otras 3 componentes del "cuadrivector" $\gamma^\mu$ no son autoadjuntas sino antiautoadjuntas:

$$(\gamma^k)^\dagger=-(\gamma^k)^\dagger; \ \ \ (\gamma^k)^2=-I$$

4. Podemos comprimir las propiedades de (anti)hermiticidad de 2 y 3 en una sola:

$$(\gamma^\mu)^\dagger = \gamma^0\gamma^\mu\gamma^0$$

**Nota:** la propiedad 4 es fácil de comprobar: para $\gamma^0$, los primeros dos dan una identidad que deja el último de los 3 como está y tenemos que $\gamma^0$ es hermítico. Para las otras 3 componentes "spacelike", tenemos que anticonmutan con $\gamma^0$ y por ende al moverlas al final sacan un "-". Luego el razonamiento es el mismo: los dos primeros $\gamma^0$ dan una identidad y dejan que las componentes spacelike del "cuadrivector" $\gamma^\mu$ son antihermíticas

## Proposición: cada componente del spinor de Dirac cumple una ecuación de Klein-Gordon

**Dem.:**

Partamos de la ecuación de Dirac covariante que acabamos de hallar y apliquémosle por la izquierda el operador $(i\cancel \partial +m)$, de modo que queda:

$$(i\gamma^\mu\partial_\mu+m)(i\gamma^\nu\partial_\nu-m)\Psi=0$$

Desarrollando:

$$[-\gamma^\mu\gamma^\nu\partial_\mu\partial_\nu-m^2]\Psi=0$$

Como los Christoffel son 0 porque la métrica de Minkowski es constante, las derivadas covariantes conmutan y podemos reescribir esta expresión como:

$$[\frac{1}{2}(\gamma^\mu\gamma^nu+\gamma^\nu\gamma^\mu)\partial_\mu\partial_\nu+m^2]\Psi=0$$

Si ahora recordamos que el anticonmutador de las matrices gammma de Dirac da 2 veces el tensor métrico (contravariante), tenemos:

$$[g^{\mu\nu}\partial_\mu\partial_\nu+m^2]\Psi=0$$

Y finalmente contrayendo los índices de las derivadas covariantes recuperamso el D'Alembertiano, de modo que queda:

$$(\square^2+m^2)\Psi=0$$

**Nota:** $\Psi$ sigue siendo un spinor. Sin embargo, como el operador de la ecuación de Klein-Gordon es escalar, se aplica a cada componente del spinor por separado, QED.

**Interpretación:** la interpretación de esto no es que la ecuación de Dirac sean sencillamente 4 ecuaciones de Klein-Gordon empaquetadas en la notación, como veremos tiene más significado que eso (permite soluciones para partículas con spin y antipartículas). El hecho de que podamos recuperar una ecuación de Klein-Gordon para cada componente del spinor indica que la ecuación de Dirac es consistente con la fórmula de la energía de Einstein

# 3. Ecuación autoadjunta de Dirac y densidad de corriente del electrón

- Si deseamos construir la corriente, podemos proceder como se hace también en mecánica cuántica no relativista: partimos de la ecuación pertinente (Schrödinger en ese caso) y multiplicamos por el complejo conjugado de la función de onda solución, de modo que llegamos a una ecuación de continuidad para la densidad de probabilidad de presencia. En el caso que nos atañe, como las soluciones de la ecuación de Dirac son spinores y los coeficientes matrices, debemos multiplicar por el complejo conjugado del spinor solución, pero el procedimiento es el mismo en espíritu
- Tomando el complejo conjugado de la ecuación de Dirac:

$$-i\frac{\partial \Psi^\dagger}{\partial t}\gamma^0-i\frac{\partial \Psi^\dagger}{\partial x^k}(-\gamma^k)-m\Psi^\dagger =0$$

- Para recuperar una ecuación en forma covariante (recuerda, esto es que se su forma se conserva bajo transformaciones de Lorentz, siempre que los coeficientes y las soluciones se transformen de manera adecuada), debemos multiplicamos por $\gamma^0$ y definimos un nuevo spinor (ahora matriz fila, no columna):

$$\bar \Psi = \Psi^\dagger \gamma^0$$

- De este modo, la ecuación que nos queda en términos de este spinor fila tiene ya forma covariante y es:

$$\boxed{i\partial_\mu \bar \Psi \gamma^\mu+m\bar \Psi=0}$$

- Finalmente, multiplicando la ecuación de Dirac por $\bar \Psi$ por la izquierda y $\Psi$ por la derecha a ambos lados y sumando, recuperamos una ecuación de continuidad para la densidad de probabilidad:

$$\partial_\mu j^\mu=0$$

donde la densidad de corriente es:

$$j^\mu = \bar \Psi\gamma^\mu \Psi$$

**Resultado:** al proceder así, la densidad de probabilidad de presencia queda ahora definida positiva, y Dirac resolvió uno de los 2 "problemas" que presentaba la ecuación de Klein-Gordon:

$$\rho = j^0=\bar \Psi \gamma^0\Psi = \displaystyle\sum_{i=1}^4|\Psi_i|^2$$

**Cuadrivector de densidad de corriente de carga del electrón:** hemos dicho que Pauli y Weisskopf indicaron que el cuadrivector de densidad de corriente del electrón realmente había de ser de carga, no de presencia, de modo que permitiera signos negativos. Entonces, definimos el cuadrivector de densidad de corriente de carga del electrón como:

$$\boxed{j^\mu=-e\bar \Psi\gamma^\mu \Psi}$$

# 2. Spinores de una partícula libre

- Deseamos ahora encontrar los autoestados del Hamiltoniano de Dirac para la partícula libre (las energías pueden ser cualesquiera porque el espectro es continuo), de modo que buscamos solucionar el problema de autovectores:

$$(\vec \alpha \cdot \vec P+\beta m)\Psi=E\Psi$$

## Partícula en reposo

- Supongamos que estamos en el inercial propio de la partícula (o que esta se encuentra en reposo en el laboratorio), entonces tenemos que $\vec p=0$ y la ecuación de autovalores queda de la forma (representación de Pauli-Dirac):

$$\begin{pmatrix}mI & 0 \\ 0 & -mI\end{pmatrix}\Psi=E\Psi$$

donde vemos que existen 2 energías positivas y 2 negativas. Las dos primeras atañen a la partícula ordinaria y las dos últimas a su antipartícula

Como hay 4 autovalores, hay 4 soluciones, que son las triviales porque la matriz ya es diagonal
