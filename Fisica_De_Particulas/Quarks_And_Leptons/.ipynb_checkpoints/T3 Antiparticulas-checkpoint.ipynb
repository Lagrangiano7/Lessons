**Nota:** los procesos de interés dentro de la física de partículas suelen ser relativistas. La teoría no relativista no predice la existencia de antipartículas ni contempla las velocidades de las partículas colisionando en experimentos habituales. Además, la física relativista tiene como postulado que las leyes de la física son las mismas en todos los inerciales (se dice que las ecuaciones son Lorentz-covariantes). De este modo, es natural que desarrollemos una teoría de partículas relativista

# 1. Relatividad especial

**Def.:** un evento tiene lugar en un punto del espacio y del tiempo $(t, \vec x)$. A este objeto, también denotado en notación tensorial $x^\mu$ **y a todo aquel que se transforme como él** se le denomina **cuadrivector**

- Un cuadrivector es por ejemplo aquel conformado por la energía de una partícula y su momento lineal:

$$p^\mu = (E/c, \vec p)$$

**Def.:** a la primera componente de un cuadrivector se le denomina "timelike" porque en el cuadrivector de un evento la primera coordenada es el tiempo. A las otras 3 componentes se les denomina "spacelike" por el mismo motivo

**Def.:** trabajaremos con la métrica de Minkowski:

$$g_{00}=1; g_{11}=g_{22}=g_{33}=-1$$

**Nota:** el módulo (calculado de acuerdo a esta métrica) de un cuadrivector es Lorentz-invariante

$$x_\mu x^\mu = c^2t^2-\vec x^2=inv.$$

$$\frac{E^2}{c^2}-\vec p^2=inv.=m^2c^2$$

donde la famosa ecuación de Einstein no es sino un reflejo de que el cuadrivector de momento es Lorentz-invariante (si bien la parte de la masa no es del todo evidente y trasciende a lo de Lorentz-invariante, porque esta invariante no nos dice el valor del invariante)

**Nota:** las componentes espaciales de un cuadrivector contravariante son $\vec A$ y las de uno covariante son $-\vec A$. La única excepción son las derivadas covariante y contravariante, que llevan el signo de la parte espacial al revés:


$$\partial^\mu = \left(\frac{\partial }{\partial t}, -\vec \nabla\right); \ \ \partial_\mu = \left(\frac{\partial }{\partial t}, \vec \nabla\right)$$

**Nota:** las magnitudes Lorentz-invariantes son, naturalmente, escalares (da igual lo que le transformes, no va a cambiar). Por ende, un invariante Lorentz se puede construir tomando el producto escalar de dos cuadrivectores, como es en el caso del cuadrivector posición o el de momento. Para la derivada, obtenemos el operador D'Alembertiano, que es un operador Lorentz-invariante:

$$\square^2=\partial_\mu \partial^\mu=\frac{\partial^2}{\partial t^2}-\nabla^2$$

# 2. Ecuación de Klein-Gordon

- A la hora de hallar la ecuación de Scrhödinger, podemos partir de la ecuación habitual del Hamiltoniano clásico:

$$E=\frac{p^2}{2m}+V$$

e identificamos $H\to i\frac{\partial}{\partial t}; \ \ \vec p\to -i\vec \nabla$ de modo que en lugar de funciones dinámicas ahora tenemos operadores. Haciendo ambos lados actuar sobre una función de onda arbitraria, tenemos que la ecuación de consr

$$i\frac{\partial}{\partial t}\Psi(\vec x, t)=-\nabla^2\Psi(\vec x, t)+V\Psi(\vec x, t)$$

que es la ecuación de Schrödinger

- Podemos hacer lo mismo con la versión relativista de esta expresión (recordemos que en sistemas donde la energía se conserva, que será siempre el caso en nuestro campo de estudio, $H\equiv E$ clásicamente). Probemos a ver qué pasa:

$$E^2=p^2c^2+m^2c^4$$

**Nota:** utilizamos unidades naturales: $c=1$ además de que $\hbar=1$ que ya lo estábamos usando

De este modo, haciendo $E\to i\frac{\partial}{\partial t}; \ \ \vec p\to -i\vec \nabla$ y haciendo actuar ambos lados sobre una función de onda arbitraria tenemos que:

$$-\frac{\partial^2\Psi}{\partial t^2}+\nabla^2\Psi=m^2\Psi$$

Que es una forma de expresar la ecuación de Klein-Gordon haciendo una distinción entre coordenadas espaciales y temporales, como se viene haciendo en clásica. Notamos que es una ecuación de onda clásica pero con un término inhomogéneo constante

Si deseamos poner espacio y tiempo al mismo nivel, podemos identificar el operador D'Alembertiano en la expresión de arriba y llegamos a la forma más habitual de la ecuación de Klein-Gordon:


$$\boxed{\square^2+m^2\Psi=0}$$

## Estados de energía negativa

- De la ecuación de Einstein, que se puede tomar como la ecuación de Klein-Gordon independiente del tiempo, vemos que esta admite tanto energías positivas como negativas, a saber:

$$E=\pm \sqrt{p^2+m^2}$$

- Al principio, uno podría no alarmarse y pensar que sencillamente se está cogiendo el origen en energías inferiores a las consideradas, pero véase que esto no puede ser, porque al ser una partícula libre, la energía no puede ser nula. Como mínimo, para una partícula en reposo, será $E=mc^2$. Sin embargo, ahora tenemos la posibilidad de partículas con energía (y por ende densidad de probabilidad de presencia, porque ambas salen proporcionales) negativas
- Tampoco podemos descartar las energías negativas como "no físicas" y trabajar solo con las positivas, porque tenemos que trabajar con un conjunto completo de estados. De lo contrario, no podríamos por ejemplo construir la condición de clausura, porque no estaríamos usando todos los estados disponibles del espacio de Hilbert que conforman la base de autoestados del Hamiltoniano

- Si de verdad hubiera energías negativas, entonces no existe tal cosa como un estado fundamental y las partículas estarían en una sempiterna transición hacia niveles más bajos, incluso si fuera libre (tendría que hacer la cuenta con partículas sometidas a un potencial, pero da igual porque las partículas que estudiamos son libres, salvo un breve momento de interacción)

## Intento de solución de Dirac

- Dirac trató de solucionar este problema escribiendo una ecuación que fuera lineal en $\partial/\partial t$ y $\vec \nabla$. Lo consiguió, y esto le permitió obtener una densidad de probabilidad de presencia positiva, además de que su ecuación describía partículas de spin 1/2, mientras que la ecuación de Klein-Gordon no contempla spin. Sin embargo, el problema de la energía negativa perduraba en su ecuación
- **Su propuesta de solución:** el espacio está lleno de un número infinito de electrones que ya ocupan todos los posibles estados de energía negativa (exclusión de Pauli), formando un "mar de electrones", y por eso los electrones no pueden seguir bajando indefinidamente en el espectro. Lo que nosotros llamamos energía 0 es sencillamente el borde de ese mar de electrones, hasta donde llega. Esto también explica la producción de pares: un fotón interacciona con un electrón del mar y lo hace promocionar a un estado de energía positiva, dejando un hueco en el mar. Dicho hueco es el positrón, y el electrón que ha promocionado, el electrón. Esto se parece bastante a teoría de bandas, como si el vacío fuera una gran banda

**Problema que le veo yo:** si esta solución es solo para electrones, qué hay del resto de partículas? En particular, qué hay de los bosones, que no obedecen exclusión de Pauli?

## Intento de solución de Pauli y Weisskopf

- La ecuacón de Klein-Gordon es la correcta, basta con meter un factor $-e$ en el flujo de probabilidad de presencia, donde ahora se vuelve flujo de corriente del electrón. Lo mismo sucede con la densidad de probabilidad de presencia, que ahora pasa a ser densidad de carga y por ende no hay problema en que sea negativa
- Sin embargo, la interpretación de las energías negativas sigue siendo una incógnita

## Interpretación de Feynman-Stückelberg

**Clave:** una solución de energía negativa es una partícula que se propaga de forma opuesta a cómo lo haría una partícula ordinaria. Alternativamente, es una antipartícula de energía positiva

**Nota:** no puede ser sencillamente inversión temporal del movimiento porque este no le mete un signo "-" a la energía. Sin emabrgo, dudo que realmente haya nada viajando hacia atrás en el tiempo, sino que es nuestra forma matemática de comprender una antipartícula con $E>0$ viajando hacia delante en el tiempo

**Explicación de la interpretación FS:**

Sea un electrón de energía $E$ y momento $\vec p$. Su cuadrivector de corriente será:

$$j^\mu(e^-) = -2e|N|^2(E, \vec p)$$

En el caso de la antipartícula, el único número que cambia su signo de aquí es la carga, de modo que el cuadrivector de corriente del positrón será la misma con el signo cambiado:

$$j^\mu(e^+) = +2e|N|^2(E, \vec p)$$

Pero poniendo un "-" delante de la carga y metiendo otro en el cuadrivector para cancelarlo, tenemos:

$$j^\mu(e^+) = -2e|N|^2(-E, -\vec p)$$

Esta corriente de carga es la misma que la que tendría un electrón de energía $-E$ y momento inverso al original. Vemos que tiene energía negativa, pero si consideramo la antipartícula como tal, esta tiene energía positiva. Hemos resuelto la dificultad de las energías negativas: no son más que la manifestación de una antipartícula, pero realmente esta tiene energía positiva

- Otra forma de verlo es que, como la partícula y antipartícula han de propagarse igual, entonces tenemos que introducir un signo "-" adicional en el tiempo para que el "-" de la energía se cancele y recuperemos:

$$e^{i(-E)(-t)}=e^{iEt}$$

**Nota:** el los diagramas de Feynman podemos trabajar siempre con antipartículas sencillamente cambiando el sentido en el que se propagan en el tiempo (si se quiere, esto invierte su momento) y cambiando el signo de su energía para pasar a considerarlas partículas ordinarias. Hecho este cambio, siempre trabajaremos con partículas ordinarias en los diagramas de Feynman (aunque tendrán energía negativa si representan antipartículas, pero esto permite la conservación de energía en las deltas y al final sale lo mismo que si todo fueran partículas ordinarias normales)

**Nota:** las partículas virtuales (los pasos intermedios) no tienen por qué conservar energía. Da igual porque los pasos intermedios no son físicos (recuerda que no son sino introducciones de la identidad en momentos en la serie de Born y que luego integramos en todos estos momentos), de modo que la energía no se conserva en los vértices que involucren partículas virtuales, pero en el proceso completo (que es el que sucede) sí
