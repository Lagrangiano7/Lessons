**Nota:** para familiarizarnos con los diagramas de Feynman, seleccionamos la QED por ser la teoría gauge más sencilla, en tanto que solo tiene una partícula gauge, que es el fotón. Consideraremos partículas puntuales como el electrón para nuestras disquisiciones de este capítulo e ignoraremos que tiene spin, de modo que trabajaremos con "leptones de spin 0" ficticios solo para ver cómo funcionan los diagramas de Feynman. El formalismo con spin se desarrolla en los Temas 5 y 6

# 1. Procedimiento con partículas indistinguibles

- Si estamos trabajando con 2 partículas distinguibles, entonces solo hay 1 diagrama de Feynman a menor orden. Sin embargo, si las 2 partículas son indistinguibles (i.e: 2 electrones), no sabemos si el primer electrón al interactuar con el segundo ha salido por el camino A y el segundo por el B o al revés, de modo que tenemos 2 contribuciones posibles (2 diagramas a menor orden), donde las partículas se intercambian
- Sumamos amplitudes porque todas ellas son válidas y representan el mismo proceso físico, y porque al ser las 2 partículas indistinguibles, la amplitud del proceso total ha de ser invariante bajo intercambio de las 2 partículas (i.e: de sus cuadrimomentos momentos)
- **IMPORTANTE:** si estás trabajando con electrón-positrón, podrías pensar que las 2 partículas son distinguibles y por tanto solo hay 1 contribución. Sin embargo, recuerda que las antipartículas se tratan como partículas ordinarias la interpretación FS y por ende a la interacción de una partícula y su antipartícula le corresponden asimismo 2 diagramas a menor orden!

## Crossing

- La amplitud de la interacción de una partícula con su antipartícula es la misma que la de la partícula original con otra del mismo tipo (indistinguible) pero cambiando el signo del cuadrimomento de las antipartículas
- Esto viene directamente de que las antipartículas se pueden ver como partículas ordinarias de cuadrimomento opuesto al de las ordinarias, y a los diagramas de Feynman qué más les da el signo el cuadrimomento. Podemos recuperar uno a partir del otro porque lo único que cambia en las matemáticas es el signo del cuadrimomento

## OFPT y cálculo de Feynman

- La Old Fashioned Perturbation Theory (OFPT) es como se hacía teoría de perturbaciones (en casos relativistas) antes: se cogía la teoría de perturbaciones no relativista y se construían los distintos diagramas de la interacción. Ellos por separado no eran Lorentz-invariantes, pero la suma de todos ellos sí lo era. En los vértices se conservaba el momento lineal pero no la energía, y las partículas intermedias (vituales) estaban en su capa de masas
- Ahora, con la teoría de perturbaciones moderna, lo que se conserva en cada vértice es el cuadrimomento. El precio a pagar es que las partículas virtuales no se encuentran en su capa de masas, pero es da igual porque lo único físico es el inicio y el final del diagrama (las partículas reales)

**Nota:** decimos que una partícula se encuntra en su capa de masas si cumple $p^2=m^2$, donde $p^2=p_\mu p^\mu$ es el módulo del cuadrivector momento. Naturalmente, todas las partículas reales se encuentran en su capa de masas porque tienen que cumplir la ecuación de Einstein, pero las partículas virtuales pueden escaquearse de ello porque no son partículas reales

# 2. Variables invariantes

- En procesos de scattering del tipo $AB\to CD$, esperamos tener 2 variables dinámicas independientes (por ejemplo, energía y ángulo de scattering) para la amplitud de scattering. Sin embargo, es deseable (y posible) expresar la amplitud de scattering en términos de variables que sean Lorentz-invariantes. Las construimos a partir de productos escalares de cuadrivectores porque sabemos que así se construyen los invariantes Lorentz, y escogemos el cuadrimomento porque es lo que siempre nos sale (véase que como las partículas reales están en su capa de masas, los módulos de los cuadrimomentos de las 4 partículas ya están determinados: $p_i^2=m^2$). Además, como se conserva el cuadrimomento een cada vértice, solo 3 de los 4 cuadrimomentos son independientes.

**Def.:** las variables de Mandelstam son las variables Lorentz-invariantes con las que podemos expresar la amplitud de scattering:

$$s=(p_A+p_B)^2$$
$$t=(p_A-p_C)^2$$
$$u=(p_A-p_D)^2$$

# 3. Sección eficaz

- Una vez hemos calculado la amplitud de scattering de un proceso, podemos hallar su sección eficaz total (y con unos pocos cálculos más, también la diferencial):

$$\boxed{d\sigma=\frac{|\mathfrak M|^2}{F}dQ}$$

donde:

- $F=4\sqrt{(p_A\cdot p_B)^2-m_A^2m_B^2}$ es el flujo incidente medido en el sistema de laboratorio
- $dQ=(2\pi)^4\delta^4(p_C+p_D-p_A-p_B)\frac{d^3p_C}{(2\pi)^32E_C}\frac{d^3p_D}{(2\pi)^32E_D}$ es el factor Lorentz-invariante del espacio de fases

**Útil:** se puede demostrar que, para un proceso $AB\to CD$, en el inercial del CM $dQ$ y $F$ toman la forma:

$$dQ=\frac{1}{4\pi^2}\frac{4}{\sqrt s}D\Omega$$

$$F=4p_i\sqrt s$$

## Sección eficaz diferencial para $AB\to CD$

- En el inercial del CM, se puede demostrar que la sección eficaz diferencial de un proceso es:

$$\boxed{\left.\frac{d\sigma}{d\Omega}\right |_{CM}=\frac{1}{64\pi^2s}\frac{p_f}{p_i}|\mathfrak M|^2}$$

- Este resultado es de gran utilidad porque permite calcular la sección eficaz diferencial de cualquier proceso de 2-2 partículas. Si deseamos mayor precisión, solo tenemos que considerar diagramas de Feynman a orden superior, de modo que todo el trabajo está en la amplitud de scattering $\mathfrak M$, y la fórmula de la sección eficaz diferencial no se ve alterada en absoluto

# 4. Ritmo de desintegración para $A\to 1+2+\ldots+n$

- Sea el proceso de desintegración de una partícula a sus $n$ hijos, entonces el diferencial del ritmo de desintegración se puede calcular con la amplitud de scattering:

$$d\Gamma = \frac{|\mathfrak M|^2}{2E_A}\frac{d^3p_1}{(2\pi)^32E_1}\ldots \frac{d^3p_n}{(2\pi)^32E_n}(2\pi)^4\delta^4(p_A-p_1-\ldots-p_n)$$

## Ritmo de desintegración de $A\to 1+2$

- En el caso de desintegración más sencillo, una partícula pasa a dar dos (si diera una no decaería). En el inercial propio de la partícula madre (esto es, en el inercial del CM), el rito de desintegración es:

$$\boxed{\Gamma(A\to 1+2)=\frac{p_f}{32\pi^2m_A^2}\int |\mathfrak M|^2d\Omega}$$

**Nota:** este ritmo de desintegración es el relativo a un cierto canal (i.e: desintegración por EM). Si la partícula tiene más de un modo de desintegración, entonces los ritmos se suman para dar el ritmo total

**Def.:** se define la "vida media" (lifetime) de una partícula como la inversa de su ritmo de desintegración total:

$$\tau=\frac{1}{\Gamma}$$
