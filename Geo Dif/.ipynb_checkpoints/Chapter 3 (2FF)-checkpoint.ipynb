# 1. Second Fundamental Form

**Def.:** the "second fundamental form" $(b_{\alpha\beta})$ is a rank 2 pseudotensor fulfilling the following properties:

1. It is invariant under congruences
2. It is invariant under reparametrizations (and therefore tensor) if and only if the reparamtrization is orientation-preserving ($D>0$)
3. It is symmetric

$$\boxed{b_{\alpha\beta}=\vec x_{\alpha\beta}\cdot \vec n=-\vec x_\alpha\cdot \vec n_\beta}$$

- If you like, its geometrically interpretation can be "how a surface pulls away from its tangent plane in a given point". Nonetheless, this is a naïve way of saying the second fundamental form accounts for the local behaviour of a surface
- Given a point on the surface, you can calculate how much a second point separates from the tangent plane of the original point like so:

$$d=(\vec x(u^1+n^1, u^2,n^2)-\vec x(u^1,u^2))\cdot \vec n$$

- If the displacement is small, then we can expand this in Taylor series and obtain:

$$\boxed{d=\frac{1}{2}n^\alpha n^\beta \vec x_{\alpha\beta}\cdot \vec n+\ldots}$$

- Where we see we recover the second fundamental form in a natural way

## 1.1 Classification of surfaces according to the sign of $b$

- The determinant of the second fundamental form is denoted $b=det(b_{\alpha\beta})$. It is not a scalar, and it transforms just like $g$ but introducing an additional sign depending on the dimension we are working on:

$$\bar b=(-1)^dD^2b$$

- Since we shall always be working with surfaces, we need not worry about this sign
**Note:** the sign of $b$ is **invariant** under coordinate changes, so we can use it to classify surfaces based on "how they look locally"

### Case 1: Eliptical paraboloid ($b>0$)

- It includes spheres, eliptical / circular paraboloids, etc.

### Case 2: Hyperbolic paraboloid ($b<0$)

### Case 3: $b=0$

- If $(b_{\alpha\beta})\not\equiv 0$ in that point, then we say it's a "parabolic point". The surface could be a parabolic cylinder but we do not know

- If $(b_{\alpha\beta})\equiv 0$ in that point, then we say it's a "planar point" and we do not know what happens in that point

- In both cases we would need to calculate higher orders of $d$

# 2. Curvatures of curves contained in a surface

- Given a curve contained in a surface, we can decompose it into its projection on the normal vector to the surface and its orthogonal complement
- **Def.:** the "curvature vector" of a curve is:

$$\vec \kappa:=\ddot {\vec x}=\vec\kappa_n+\vec\kappa_g$$

**Def.:** the "normal curvature" $\kappa_n$ is the projection of the curvature vector of a curve onto the normal vector to the surface. We can think of it as "the curvature every curve needs to pay in order to not pop out of the surface".

$$\boxed{\kappa_n=b_{\alpha\beta}\dot u^\alpha\dot u^\beta = \frac{b_{\alpha\beta}u'^\alpha u'^\beta }{g_{\mu\nu}u'^\mu u'^\nu}}$$

**Result:** **any curve contained in a surface and going through a point $P\in S$ with a given direction has the same normal curvature**

**Def.:** the "geodesic curvature" is the projection of the curvature vector onto the orthogonal complement of $\vec n$ (that is to say, onto the tangent space spanned by $\vec x_\alpha$). It can be though of as "the freedom a curve has to go left and right, independent of the surface".

## 2.1 Normal curvature

### Normal sections

**Def.:** a "normal section" of a given surface with a curve on it is the intersection of said surface with the plane $\vec t\vec n$ in the considered point

- Utility: any 2 curves contained in a surface passing through the same point of said surface with the same tangent vector at that point have the same normal curvature. In that point, given the curve of interest, we can build another different curve taking the intersection of the $\vec t\vec n$ plane with the surface. By construction, these two curves will have the same normal curvature, and in fact the normal curvature of the newly created curve will be all of its curvature. Therefore, we can then calculate the whole curvature of the first curve with:

$$\kappa\cos\gamma=\kappa_n$$

where $\kappa_n$ is the normal curvature of both curves and $\vec p\cdot \vec n=\cos\gamma$ (p for the original curve)

- For instance, when working on a sphere, if our curve of interest is the small circumference and we wish to determine its curvature, we can trace the normal section in that point --> we get the big circumference with $\kappa_n\equiv\kappa_B=\frac{1}{R}$ and then:

$$-\frac{1}{R_S}\sin\theta=-\frac{1}{R}\Longrightarrow R_S=R\sin\theta$$

![](Normal_Curvature.png)

**Def.:** a curve for which $\kappa_n=0$ in every point is called "asymptotic curve". All straight lines (like the generatrix of some revolution surface) are asymptotic curves. The equation for them is eviently:

$$b_{\alpha\beta}u'^\alpha u'^\beta=0$$

## 2.2 Principal curvatures

**Def.:** the principal curvatures $\kappa_1,\kappa_2$ of a given surface are the max. and min. values of the normal curvatures a curve can have in said surface. They are intrinsic to the surface (since we "iterate" over all the possible curves contained in the surface)

- For instance, in a sphere, both the max. and the min. normal curvatures shall be $-\frac{1}{R}$. You could think smaller circumferences might have smaller normal curvature but for every small circumference and for every point of it there is one great circumference which which has $\kappa_n=-1/R$ and therefore the small circumference also has it.

- For a plane, $\kappa_1=\kappa_2=0$

- For a circular cylinder, $\kappa_1=0, \kappa_2=-1/R$

- Principal curvatures can be obtained from the eigenvalue problem (in a non-orthogonal basis):

$$(b_{\alpha\beta}-\kappa_ng_{\alpha\beta})l^\beta=0$$

**Observation:** if both $(g_{\alpha\beta})$ and $(b_{\alpha\beta})$ are diagonal, you can read off the principal curvatures from this eigenvalue problem, and obtain:

$$\boxed{\kappa_1=\frac{b_{11}}{g_{11}}; \ \ \kappa_2=\frac{b_{22}}{g_{22}}}$$

- If this is not the case, it is simpler in practice to solve an orthogonal eigenvalue problem:

$$\boxed{b^\mu_{\ \beta}-\kappa_n\delta^\mu_{\ \beta}=0}$$

**Def.:** $b^\mu_{\ \beta}=g^{\mu\alpha}b_{\alpha\beta}$ is called the "Weingarten operator / shape operator"

## 2.3 Gaussian and mean curvatures

- From the two principal curvatures we define 2 of the most important quantities in differential geometry

1. Mean curvature: $H=\frac{1}{2}(\kappa_1+\kappa_2)$
2. Gaussian curvature: $K=\kappa_1\kappa_2$

- **Note:** if we change orientation, $H$ changes sign but $K$ does not. Furthermore, a plane and a circular cylinder have the same Gaussian curvature. This is because it is connected to the first fundamental form

- **Def.:** surfaces with $H=0$ are called "minimal surfaces" because they have minimal area (soap bubbles and in general any system trying to minimize surface tension adopt minimal surface forms)

- How do we obtain these curvatures in practice?

- From the characteristic polynomial of the eigenvalue problem for principal curvatures, we deduce:

$$H=\frac{1}{2}Tr(b^\alpha_\beta)$$
$$K=\frac{b}{g}$$

# 3. Gauss-Weingarten equations

- They are the analogue of the Frenet equations but for the Gauss trihedron (i.e: they allow us to do geometry on surfaces)

$$\vec x_{\alpha\beta}=\Gamma_{\alpha\beta}^{\ \ \ \ \ \gamma}\vec x_\gamma+b_{\alpha\beta}\vec n$$
$$\vec n_\alpha=-b_\alpha^{\ \ \gamma}\vec x_\gamma$$

- where $\Gamma_{\alpha\beta}^{\ \ \ \ \ \gamma}$ are called the "Christoffel symbols of the second kind". If we lower the first index, they are "of the first kind" 

## Christoffel symbols

### Properties

1. They are **not** tensors since they are not invariant under reparametrizations (they get an additional term called "inhomogeneous term"). They transform as follows:

$$\bar\Gamma_{\alpha\beta}^{\ \ \ \ \ \lambda}=\Gamma_{\gamma\mu}^{\ \ \ \ \ \sigma}\frac{\partial u^\mu}{\partial \bar u^\beta}\frac{\partial u^\gamma}{\partial \bar u^\alpha}\frac{\partial \bar u^\lambda}{\partial  u^\sigma}+\frac{\partial \bar u^\lambda}{\partial u^\sigma}\frac{\partial^2 u^\sigma}{\partial \bar u^\alpha\partial \bar u^\beta}$$

**Note:** if we restrict ourselves to using linear diffeomorphisms, then the Christoffel symbols are tensors since the second derivative vanishes

2. Symmetric in lower indices: $\Gamma_{\alpha\beta}^{\ \ \ \ \ \gamma}=\Gamma_{\beta\alpha}^{\ \ \ \ \ \gamma}$
3. They can be built using only the 1FF

$$\Gamma_{\alpha\beta}^{\ \ \ \ \ \lambda}=\frac{1}{2}g^{\mu\lambda}\left(\frac{\partial g_{\alpha\lambda}}{\partial u^\beta}+\frac{\partial g_{\lambda\beta}}{\partial u^\alpha}-\frac{\partial g_{\beta\alpha}}{\partial u^\lambda}\right)$$

**Note:** notice this doesn't involve $\vec x$ whatsoever, so it's a general result for Riemannian manifolds!

4. If the coordinate system is orthogonal ($g_{12}=g_{21}=0$), then we have explicit expressions for the Christoffel symbols

$$\Gamma_{11}^{\ \ \ \ \ 1}=\frac{1}{2g_{11}}\frac{\partial g_{11}}{\partial u^1}$$
$$\Gamma_{11}^{\ \ \ \ \ 2}=-\frac{1}{2g_{22}}\frac{\partial g_{11}}{\partial u^2}$$
$$\Gamma_{12}^{\ \ \ \ \ 2}=\Gamma_{21}^{\ \ \ \ \ 2}=\frac{1}{2g_{22}}\frac{\partial g_{22}}{\partial u^1}$$
$$\Gamma_{12}^{\ \ \ \ \ 1}=\Gamma_{21}^{\ \ \ \ \ 1}=\frac{1}{2g_{11}}\frac{\partial g_{11}}{\partial u^2}$$
$$\Gamma_{22}^{\ \ \ \ \ 1}=-\frac{1}{2g_{11}}\frac{\partial g_{22}}{\partial u^1}$$
$$\Gamma_{22}^{\ \ \ \ \ 2}=\frac{1}{2g_{22}}\frac{\partial g_{22}}{\partial u^2}$$

# 4. PMC equations and the Riemann tensor

- When working with curves, if given a positive-definite function $\kappa$ and another function $\tau$ we could state they represent a single curve (except for congruences) and we could reconstruct it using Frenet equations
- Can we do the same for surfaces if we are provided with a positive-definite bilinear and symmetric form $(g_{\alpha\beta})$ and a bilinear and symmetric form $(b_{\alpha\beta})$?
- In general, **NO!** Not every pair of these forms induces a surface where one of them is the 1FF and the other is the 2FF (that is to say, the Gauss-Weingarten equations do not have a solution for every pair of these two forms)

**(Sufficient) condition for existent and (locally) unique solution:** (Peterson-Mainardi-Codazzi equations)

$$\vec x_{\alpha\beta\lambda}=\vec x_{\alpha\lambda\beta}$$

- We get conditions for the $\vec n$ part and the $\vec x_{\alpha}$ parts, of course. In practice, we get 3 conditions:

$$\Gamma_{\alpha\beta}^{\ \ \ \ k}b_{k\lambda}-\Gamma_{\alpha\lambda}^{\ \ \ \ k}b_{k\beta}+\frac{\partial b_{\alpha\beta}}{\partial u^\lambda}-\frac{\partial b_{\alpha\lambda}}{\partial u^\beta}=0; \ \ (\alpha,\beta,\lambda)=(112),(212)$$

$$R_{1212}=b$$

- Where $R_{1212}$ is the corresponding matrix element of the 0-4 Riemann tensor

## 4.1 Riemann tensor

- The 1-3 Riemann tensor is defined as follows:

$$ {\displaystyle {R^{\rho }}_{\sigma \mu \nu }=\partial _{\mu }\Gamma _{\nu \sigma }^{\rho }-\partial _{\nu }\Gamma _{\mu \sigma }^{\rho }+\Gamma _{\mu \lambda }^{\rho }\Gamma _{\nu \sigma }^{\lambda }-\Gamma _{\nu \lambda }^{\rho }\Gamma _{\mu \sigma }^{\lambda }}$$

- The 0-4 tensor is obtained just by lowering the first index

$$R_{\rho\sigma\mu\nu}=g_{\alpha\rho}R^\alpha_{\ \ \sigma\mu\nu}$$

**Note:** if we are working with a given surface (which we already know that has a valid "combination" of 1FF and 2FF since it is a surface and therefore a solution of the GW equations), then the PMC conditions are satisfied and we can obtain the 0-4 Riemann tensor directly from the second fundamental form:

$$R_{\sigma\alpha\lambda\beta}=b_{\alpha\beta}b_{\lambda\sigma}-b_{\alpha\lambda}b_{\beta\sigma}$$

- The 1-3 Riemann tensor can be built by raising the first index ($\sigma$) in both sides

**Properties of the 0-4 Riemann tensor:** (0-4 because we can then compare all 4 indices)

1. $R_{(\sigma\alpha)(\lambda\beta)}=R_{(\lambda\beta)(\sigma\alpha)}$
2. $R_{\sigma\alpha(\lambda)(\beta)}=-R_{\sigma\alpha(\beta)(\lambda)}$
3. $R_{(\sigma)(\alpha)\lambda\beta}=-R_{(\alpha)(\sigma)\lambda\beta}$
4. $R_{\sigma\alpha\lambda\beta}+R_{\sigma\lambda\beta\alpha}+R_{\sigma\beta\alpha\lambda}=0$ (Bianchi identity)

**Note:** in 2 dimensions, the Bianchi identity is a direct consequence of the former 3 properties. In higher dimensions, it still holds but it is no longer a consequence of them

**Note:** due to the antisimmetry property of the 0-4 tensor, only 4 out of the 16 matrix elements are nonzero. Furthermore, only 1 of them is independent from the other 3. We take this to be $R_{1212}$. For surfaces, we can use the formula for the Riemann tensor with the 2FF and get:

$$R_{1212}=b_{22}b_{11}-b_{12}b_{21}$$

as we knew we would get, since this is one of the 3 conditions the PMC equations impose

**Note:** in dimension $n$, the Riemann tensor has $\frac{1}{12}n^2(n^2-1)$ independent terms (for $n=2$, just 1)

## 4.2 Theorema egregium

- "The Gaussian curvature $K$ of a surface is independent of the 2FF but depends only on the 1FF"

$$K=\frac{R_{1212}}{g}$$

- Since we already knew $K=b/g$. The justification of this theorem is that $R_{1212}$ can be calculated only from Christoffel symbols, which at the same time can be obtained only from the 1FF

**Result:** we cannot determine the 2FF from the 1FF but its determinant $b$, yes

**Result:** since the Gaussian curvature only depends on the 1FF and the Riemann tensor as well, we can calculate the Riemann tensor directly from the 1FF is we know the Gaussian curvature of the surface:

$${\displaystyle R_{abcd}=K\left(g_{ac}g_{db}-g_{ad}g_{cb}\right).}$$

## 4.3 Theorem for locally isometric surfaces

- "Two surfaces are locally isometric (same $g_{\alpha\beta}$) $\Longrightarrow$ they have the same Gaussian curvature"

- This explains why the plane and the cylinder have the same Gaussian curvature!

**NOTE:** if the Gaussian curvature is constant, then the implication also goes in the other way

## Information of a manifold

- In $d=2$, all of the important information of the manifold is contained in the Gaussian curvature, $K$
- In $d=3$, it is contained in the Ricci tensor, $\boxed{R_{\nu\sigma}=g^{\mu\lambda}R_{\mu\nu\lambda\sigma}}$
- In higher dimensiones, it is contained in the Riemann tensor

**Note:** this is why we can recover the Riemann tensor completely from these magnitudes in each dimension (i.e: in $d=2$, we can build it just from the Gaussian curvature). In $d=3$, we can express the Riemann tensor in terms of the Ricci tensor as follows:

$$\boxed{R_{\mu\nu\lambda\sigma}=(g_{\mu\lambda} R_{\nu\sigma}-g_{\nu\lambda}R_{\mu\sigma}-g_{\mu\sigma}R_{\nu\lambda}+g_{\nu\sigma}R_{\mu\lambda})+\frac{R}{2}(g_{\mu\sigma} g_{\nu\lambda}-g_{\mu\lambda}g_{\nu\sigma})}$$
