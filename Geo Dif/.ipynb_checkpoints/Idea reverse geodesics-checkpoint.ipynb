**Idea:** the equations for geodesics in a given metric are such that, given the Christoffel symbols of said metric, one can solve a 2-equation system of 2nd order ODEs to obtain the map of the geodesic, $(u^1(s), u^2(s))$. It is guaranteed that said solutions will be geodesics in the given metric. Nonetheless, what if we think of the problem in reverse? Given a certain set of coordinates $(u^1(s), u^2(s))$, find the metric(s) in which said curve is a geodesic.

**Procedure sketch:** Instead of knowing the Christoffel symbols and solving for $(u^1(s), u^2(s))$, we now treat $(u^1(s), u^2(s))$ as given functions and solve the two-equation EDP system for $g_{ij}$. In order to simplify the procedure, we shall only deal with orthogonal coordinate systems.

**Intuition:** a circumference $(u^1(s), u^2(s))=(\cos s, \sin s, 0)$ is not a geodesic on a plane. Nonetheless, if it is a great circumference of a sphere, then it is indeed a geodesic. We therefore observe that whether a curve is or not a geodesic depends on the metric being considered (i.e: the surface in which the curve is contained).

We now formulate the general equations for geodesics (expanding the Christoffel symbols) for a general orthogonal metric. These are the equations we shall be solving for $g_{11}$ and $g_{22}$.

$$\begin{cases}\ddot u^1(s)+\frac{1}{2g_{11}}\frac{\partial g_{11}}{\partial u^1}(\dot u^1)^2+\frac{1}{g_{11}}\frac{\partial g_{11}}{\partial u^2}\dot u^1\dot u^2-\frac{1}{2g_{11}}\frac{\partial g_{22}}{\partial u^1}(\dot u^2)^2=0
\\
\ddot u^2(s)+\frac{1}{2g_{22}}\frac{\partial g_{22}}{\partial u^2}(\dot u^2)^2+\frac{1}{g_{22}}\frac{\partial g_{22}}{\partial u^1}\dot u^1\dot u^2-\frac{1}{2g_{22}}\frac{\partial g_{11}}{\partial u^2}(\dot u^1)^2=0\end{cases}$$

**Interesting property (hopefully helpful) about this system:** we can pass from the first equation to the other and back by changing the indices: $1\leftrightarrow 2$

**Big problem:** we cannot get rid of the dependence in $s$, and the metric should depend on $u^1,u^2$, not on $u^1(s), u^2(s)$.

**Example:** given the map of a curve $(u^1(s), u^2(s))=(s,s)$, find a metric in which this curve is a geodesic (we should obtain the plane, that is, all coefficients of the metric should be constant)

- We calculate: $\dot u^1=\dot u^2=1$;  $\ddot u^1=\ddot u^2=0$, and we substitute:

$$\begin{cases}\frac{1}{2g_{11}}\frac{\partial g_{11}}{\partial u^1}+\frac{1}{g_{11}}\frac{\partial g_{11}}{\partial u^2}-\frac{1}{2g_{11}}\frac{\partial g_{22}}{\partial u^1}=0
\\
\frac{1}{2g_{22}}\frac{\partial g_{22}}{\partial u^2}+\frac{1}{g_{22}}\frac{\partial g_{22}}{\partial u^1}-\frac{1}{2g_{22}}\frac{\partial g_{11}}{\partial u^2}=0\end{cases}$$

- We multiply the first equation by $2g_{11}$ and the second one by $2g_{22}$
- We shall also perform a change of notation for a more convenient presentation of the equations, to wit:

$$u^1\to x$$
$$u^2\to y$$
$$g_{11}(u^1, u^2)\to f(x,y)$$
$$g_{22}(u^1, u^2)\to g(x,y)$$


- We can now rewrite the equations in a more compact manner:

$$\begin{cases}f_x+2f_y-g_x=0
\\
g_y+2g_x-f_y=0\end{cases}$$

- After feeding this system of equations to chatgpt (mind me, I couldn't uncouple this system even though I tried), I get:

$$f=cte., g=cte.$$

as expected. I therefore conclude:

The curve whose map is $(u^1(s), u^2(s))=(s,s)$ is a geodesic in any metric such that:

$$(g_{\alpha\beta})=\begin{pmatrix}c&0\\0&d\end{pmatrix}; \ c,d\in \mathbb R$$

(the constants cannot be 0 at the same time, since the metric would be nil)

- Just as expected, the straight line is a geodesic in the plane. Nonetheless, according to this result, the straight line is also a geodesic in the cylinder (of course, if we draw a $45^\circ$ line in a plane and fold it into a cylinder, we get a helix!)

- I am therefore moved to end this short dissertation with the following theorem

## Theorem

- If a given map $(u^1(s), u^2(s))$ represents a geodesic in a given surface, then the same map also represents a geodesic in any and every surface that is locally isometric to the original one

**Proof:** the proof of this theorem is immediate: since the mapping of the curve is previous to the selection of the surface, and since two locally isometric surfaces share, by definition, the same metric (in the appropriate set of coordinates), the equations of the geodesic remain unchanged, and therefore their solution too.
