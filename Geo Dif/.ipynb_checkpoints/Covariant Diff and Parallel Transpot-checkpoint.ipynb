# 1. Covariant Derivative

## Definition

- When differentiating a vector with respect to a coordinate of the surface, we should differentiate the vector as a mathematical object, not just its tensor components. We want to compute:

$$\frac{\partial}{\partial u^\beta}\vec v=\frac{\partial}{\partial u^\beta}(v^\alpha\vec x_\alpha)$$

Applying Leibniz's rule and introducing the GW equations in the term with $\vec x_{\alpha\beta}$, we get:

$$\frac{\partial}{\partial u^\beta}\vec v=\left(\frac{\partial v^\lambda}{\partial u^\beta}+\Gamma_{\alpha\beta}^\lambda v^\alpha\right)\vec x_\lambda+b_{\alpha\beta}v^\alpha\vec n$$

So we see that, in general, the derivative of a vector of the tangent space has a component on the tangent space and another component on the normal axis.

**Def.:** for a vector, we define the "covariant / absolute" derivative as the tangent space part of the partial derivative of said vector with respect to the coordinates of the surface. We denote the former with $;\mu$ and the latter with $,\mu$, though it can also be seen in the literature as $\nabla_\mu$:

$$\boxed{\nabla_\mu v^ \lambda = v^\lambda_{; \mu}=v^\lambda_{,\mu}+\Gamma_{\alpha\beta}^\lambda v^\alpha}$$

**Note:** it's good practice to always sum with the 1st index of the Christoffel symbols. Although they are symmetric in the two lower indices, there is a generalization of said symbols in which they are no longer symmetric and therefore summations are carried out by convention with respect to the first (lower) index

## How does it act on higher order tensors?

- **Key:** for every contravariant index, it introduces a term $+\Gamma$ and for every contravariant index it introduces a term $-\Gamma$

**Scalars:**

$$\phi_{;\mu}=\phi_{,\mu}$$

**1st order tensors:**

- Contravariant components:

  $$v^\lambda_{; \mu}=v^\lambda_{,\mu}+\Gamma_{\alpha\beta}^\lambda v^\alpha$$

- Covariant components:

$$v_{\lambda; \mu}=v_{\lambda,\mu}-\Gamma_{\lambda\beta}^\alpha v_\alpha$$

**2nd order tensors:**

- Twice-covariant components:


$$t_{\alpha\beta; \mu}=t_{\alpha \beta, \mu}-\Gamma_{\alpha\mu}^\lambda t_{\lambda\beta}-\Gamma_{\beta\mu}^\lambda t_{\alpha\lambda}$$

- Twice-contravariant components:

$$t^{\alpha\beta}_{;\mu}=t^{\alpha\beta}_{,\mu}+\Gamma_{\lambda\mu}^\alpha t^{\lambda\beta}+\Gamma_{\lambda\mu}^\beta t^{\alpha\lambda}$$

- Mixed components:

$$t^{\alpha}_{\beta;\mu}=t^{\alpha}_{\beta,\mu}+\Gamma_{\lambda\mu}^\alpha t^{\lambda}_\beta-\Gamma_{\beta\mu}^\lambda t^\alpha_{\lambda}$$

## Properties of the covariant derivative

1. It is a tensor (such is the motivation to define it)
2. It is a linear operator
3. It obeys the Leibniz rule
4. $g_{\alpha\beta}, g^{\alpha\beta}, \delta^\alpha_\beta$ have null covariant derivative ("covariantly constant")
5. It commutes with contractions: $(t^{\alpha\lambda}_\lambda)_{;\mu}=(t^{\alpha\lambda}_{\beta; \mu})\delta^\beta_\lambda$ (because the derivative of the delta is 0 and because Leibniz's rule holds)
6. Covariant derivatives do not commute

- The "commutator" of these derivatives involves Riemann's tensor. It introduces a term $+R$ for every covariant index and $-R$ for every contravariant index:

**Scalars:** (they do commute so long as the Schwartz lemma holds)

$$\phi_{;\sigma;\mu}=\phi_{;\mu;\sigma}$$

**1st order tensors:**

- Covariant components:

$$v_{\alpha; \sigma;\mu}-v_{\alpha; \mu; \sigma}=v_\lambda R^\lambda_{\alpha\sigma\mu}$$

- Contravariant components:

$$v^\alpha_{;\sigma;\mu}-v^\alpha_{;\mu;\sigma}=-t^\lambda R^\alpha_{\lambda\sigma\mu}$$

**2nd order tensors:**

- Twice-covariant components:

$$t_{\alpha\beta; \sigma;\mu}-t_{\alpha\beta; \mu;\sigma}=t_{\lambda\beta}R^\lambda_{\alpha\sigma\mu}+t_{\alpha\lambda}R^\lambda_{\beta\sigma\mu}$$

- Twice-contravariant components:

$$t^{\alpha\beta}_{;\sigma;\mu}-t^{\alpha\beta}_{;\mu;\sigma}=-t^{\lambda\beta}R^\alpha_{\lambda\sigma\mu}-t^{\alpha\lambda}R^\beta_{\lambda\sigma\mu}$$

- Mixed components:

$$t^\alpha_{\beta; \sigma; \mu}-t^\alpha_{\beta; \mu; \sigma}=t^\alpha_\lambda R^\lambda_{\beta\sigma\mu}-t^\lambda_\beta R^\alpha_{\lambda\sigma\mu}$$

## 2nd Bianchi Identity

- This is a property of the covariant derivatives of the Riemann tensor: the last 3 lower indices (including the covariant derivative) are cyclic:

$$\boxed{R^\sigma_{\alpha ab;c}+R^\sigma_{\alpha bc;a}+R^\sigma_{\alpha cb;a}=0}$$

## Covariant Derivative of a Vector Field alongside a Curve

- Let a vector field $\vec a$ be defined on the points of some curve $C$ lying on a surface $S$ (the field is defined on the tangent space of the surface). Since it is defined **exclusively** along this curve, it only depends on the parameter of said curve, and we can write the vector field as:

$$\vec a(t)=a^\lambda \vec x_\lambda$$

- If we take its ordinary derivative with respect to $t$, we find:

$$\frac{d\vec a}{dt}=\left[a'^\lambda(t)+\Gamma_{\alpha\beta}^\lambda  a^\alpha u'^\beta\right]\vec x_\lambda + b_{\alpha\beta}a^\alpha u'^\beta \vec n$$

**Def.:** in analogy to the definition of covariant derivative, we take the covariant derivative of a vector field along a curve to be the tangent space part of the "full" derivative $\frac{d\vec a}{dt}$

Therefore, we denote:

$$\boxed{\frac{\nabla a^\lambda}{dt}=a'^\lambda(t)+\Gamma_{\alpha\beta}^\lambda  a^\alpha u'^\beta}$$

**Note:** everything is now evaluated on the coordinates of the curve, including the Christoffel symbols. When you are done with the calculation, you should get only a function of $t$ since $\vec a(t)$ is also only a function of $t$

# 2. Parallel transport

**Def.:** a vector field $\vec a(t)$ along a curve $C$ is said to be "parallel along the curve" if:

$$\boxed{\frac{\nabla \vec a}{dt}=0}$$

along said curve

**How I see it:** picture yourself on a car, taking a turn. Picture arrows periodically located on the side of the road. You would say the arrows form a vector field (if there were an arrow for every point of the road, but this is just a visualization) if, to you, it doesn't look like the arrows are moving. You always see the same arrow, despite the fact that an external observer would clearly see how not all the arrows point in the same direction, but rather they "follow" the curve

**Observation:** if a vector field is parallel along a curve (any curve along the surface, really) then it means its ordinary derivative with respect to $t$ is normal to the surface, as we are precisely imposing its tangent part to be null

## Parallel transport

- **Def.:** let a vector $\vec v$ lie on a point of a surface $S$. Let a curve $C$ of $S$ pass through that point. Then, we call "parallel transporting $\vec v$" if we find a vector field defined along the curve $C$ such that $\vec a(0)=\vec v$ and such that it is parallel along the curve

**In practice:** we take $\vec a(0)=\vec v$ as an initial condition and solve the equations for parallel transport:

$$\begin{cases}\frac{da^\lambda}{dt}+\Gamma_{\alpha\beta}^\lambda a^\alpha u'^\beta=0 \\ \vec a (0)=\vec v\end{cases}$$

**Elementary property of parallel transport:** the scalar product of any two vector fields defined along some curve is invariant if we parallel transport them along said curve

**Consequence 1:** the length of a vector remains invariant under parallel transport

**Consequence 2:** the angle between any 2 vectors remains invariant under parallel transport

**Consequence 3:** if two surfaces $S_1,S_2$ are tangent along a curve $C$, then parallel transport on $S_1$ along said curve is the same as parallel transport on $S_2$ along the same curve (this makes sense because, by definition of tangency, they share the normal vector on every point of the curve. Making the ordinary derivative parallel to said vector, we are done parallel-transporting)

## Geodesics and parallel transport

- **Def.:** a geodesic curve is that which parallel-transports its velocity

This is to say, if we take $\vec a^\alpha(t)=u'^\alpha(t)$ and introduce it in the parallel-transport equations, we recover the 2nd order ODE equations for geodesics


**In practice, how does one perform parallel transport?:** if it is along geodesics, it's very simple: since the parallel-transport their velocity vector, and since parallel transport conserves the angle of 2 parallel-transported vectors, we only have to determine the angle of the "initial" vector with respect to the velocity of the curve in said point and then trace on each point a vector of same length as the initial one and fornming the same angle with the curve's velocity in said point. If the curve is not a geodesic, then we have to solve the equations

**Note:** in general, parallel transport "has memory" (i.e: parallel-transporting a vector alongside a closed curve on a surface need not yield the same vector as before. Of course, its length will be the same, but it will get back to the point of departure pointing in a different direction)

**Note:** for parallel-transport **on  a surface** to be invariant, it is a sufficient condition that $K=0$ on that surface. In higher dimensional manifolds, the condition is more demanding, and one has to ask that the Riemann tensor is identically 0
