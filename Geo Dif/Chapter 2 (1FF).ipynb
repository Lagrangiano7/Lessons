# 0. Definitions

**Def.:** a regular surface is a subset $S\subset \mathbb{R}^3$ in which for each point $P\in S$ there exists an open set $U\subset \mathbb{R}^2$, an open set $V\subset \mathbb{R}^3$ and a map $\vec x: U\to V\cap S$ such that:

1. $\vec x \in C^\infty (U)$
2. $\vec x$ is homomorphsim (bijective, continuous and its inverse is also continuous)
3. $\vec x$ is regular (Jacobian rank 2, avoids degeneracy)

**Def.:** the curve delimiting a surface of revolution is called "generatrix / meridian"

**Def.:** the lines forming circumferences in a surface of revolution are called "parallels"

- We shall be working on the tangent space of a surface, where the basis for each point is $B:=\{\frac{\partial \vec x}{\partial u}\Big |_{P}, \frac{\partial \vec x}{\partial v}\Big |_{P}\}$. Or, in a more compact notation, $B:=\{\vec x_\alpha\}$. **This basis is not orthogonal in general**

**Def.:** the "Gauss trihedron" $\{\vec x_1, \vec x_2, \vec n\}$ is the trihedron we shall place on top of each point of the surface

**Def.:** a "tensor" is a mathematical object which remains invariant under changes in the coordinate system

# 1. First fundamental form

**Def.:** given a parametrized surface, the "first fundamental form" is a mathematical object fulfilling the following properties:

1. It is a tensor
2. It is positive-definite (therefore, $det (g_{\alpha\beta})>0$)
3. It is bilinear
3. It is symmetric

$$\boxed{g_{\alpha\beta}=\vec x_\alpha\cdot\vec x_\beta}$$

- If we have our surface parametrized, then we can work out its basis in every point and therefore we can think of the first fundamental form as the Gram matrix of the surface

**NOTE:** if two surfaces are "locally isometric", there exists a coordinate system for either one of them for which the first fundamental form of both surfaces is the same. This happens, for instance, for the plane in cartesian coordinates and the circular cylinder in cylindrical coordinates. This is equivalent to claiming there exists a change of coordinates that takes us from a map to the other

**Def.:** surfaces locally isometric to the plane (i.e: $K=0$ everywhere on the surface) are said to be "developable" since we can unfold them into a plane

## 1.1 Scalar product with 1FF

- The scalar product of any two vectors $v=v^\alpha\vec x_\alpha$, $\vec w=w^\beta\vec x_\beta$ can be computed as:

$$\vec v\cdot \vec w=g_{\alpha\beta}v^\alpha w^\beta$$

## 1.2 Area of a surface with 1FF

- The area of a given surface (or a portion of it) can be calculated as follows:

$$\boxed{A=\int_S\sqrt g du^1du^2}$$

**Note:** the element of surface $dA\sqrt g du^1du^2$ is invariant under coord. transformations

## 1.3 Transformation of mathematical objects under change of coordinates

- **Note:** tensors are by definition invariant under change of coordinates. Nonetheless, we shall see their components do change. This is due to the fact that, since the vectors of the basis change, the components need to change as well in order to keep the tensor immutable

### Basis vectors

$$\bar{\vec x}_\beta=\frac{\partial u^\alpha}{\partial \bar u^\beta}\vec x_\alpha$$

**Def.:** if a tensor's components change like the basis vectors do, it shall be called "covariant". If it changes inversely to how the basis vectors do it, then it shall be called "contravariant".

### First fundamental form

$$\bar g_{\mu \nu}=\frac{\partial u^\alpha}{\partial \bar u^\mu}\frac{\partial u^\beta}{\partial \bar u^\nu}g_{\alpha\beta}$$

- We therefore say it is "twice covariant". In practice, once we know an object is a tensor, we shall only have to look at where its indexes are located in order to know how many times it is covariant and how many it is contravariant

### Vectors

- Since they are linear combinations of the basis vectors, and since vectors are invariant under reparametrizations, they must "oppose" the change of the basis vectors (this can be shown formally by doing the direct computation)

$$\bar v^\mu = \frac{\partial \bar u^\mu}{\partial u^\alpha}v^\alpha$$

**Note:** we may speak about "covariant / contravariant vectors" but this is wrong. Vectors are invariant under reparametrizations because they are tensors. Nonetheless, their componentes may be covariant or contravariant, but to keep it short we associate this property loosely to the vector itself

### $g=det(g_{\alpha\beta})$

$$\bar g=D^2g$$

**Def.:** a "scalar" (or rank 0 tensor) is anything that transforms as $\bar a=a$. That is to say, anything we would call "scalar" in linear algebra that is parametrization independent

- We notice the determinant of the first fundamental form is not a scalar, since it transforms with $D^2$, where $D$ is the determinant of the Jacobian matrix representing the change of coordinates

### Surface element

$$d\bar A=dA$$

- The element of surface $dA=\sqrt gdu^1du^2$ is indeed a scalar since it remains invariant under reparametrizations. This is due to the fact that $\sqrt g$ introduces a Jacobian but the differentials introduce its inverse

# 2. Orientability of a surface

**Theorem:**

- A surface $S$ is said to be "orientable" if it can be covered by a family of charts such that for each point $P$ belonging to two charts, the change of coordinates has a positive determinant (i.e: is orientation preserving)

**Def.:** an "elementary surface" is that which can be covered by a single chart. These are all orientable

**Theorem:**

- A surface that can be covered by 2 charts whose intersection is connected is orientable (i.e: sphere)

# 3. Dual basis

- We'd like to obtain the coefficients of a vector expressed in the basis of the tangent space
- We build another basis of the tangent space called "dual basis" which is, by construction, orthonormal to the original basis, like so:

$$\vec x^\alpha\cdot \vec x_{\beta}=\delta^\alpha_{\ \beta}$$

- This means we can now get the coefficients of a vector expressed in the original basis just by taking its scalar product with the corresponding dual basis vector:

$$v^1=\vec x^\alpha\cdot \vec v$$

- Building the dual basis could not be any easier:

$$\vec x^\alpha=g^{\alpha\beta}\vec x_\beta$$

- This often receives the name of "lowering / raising indices" or "index juggling". The first fundamental form can be used to hide the metric and make the scalar product of two vectors resemble the usual scalar product defined in euclidean spaces:

$$\vec v\cdot \vec w=v_1w^1 + v_2w^2$$

**Note:** in the dual basis, everything transforms opposite to how it would transform in the ordinary basis!

# 4. Riemannian Manifolds

- For a given first fundamental form, there may not exist any surfaces that induce it. That is to say, if I am given, for instance, $(g_{\alpha\beta})=\begin{pmatrix}\frac{1}{v^2}&0\\0&\frac{1}{v^2}\end{pmatrix}; v>0$ (Poincaré semiplane), there is no parametrized surface $\vec x(u,v)$ such that it induces this first fundamental form

**Def.:** a Riemannian manifold is an open set $U=\{(u^1,\ldots,u^n)\}\subset \mathbb{R}^n$ in which we have defined a form $(g_{\alpha\beta})$ called "metric" with the properties the first fundamental form would have:

1. Bilinear
2. Positive-definite
3. Symmetric
4. Twice-covariant

- Riemannian manifolds are a generalization of surfaces, and therefore every surface is a Riemannian manifold. In fact, every first fundamental form is a metric, but not all metrics can be first fundamental forms, since these arise from a parametrized surface

- The point of Riemannian manifolds is they take $\vec x$ out of the equation. Our mathematics only depends on the metric, which is no longer induced by the surface, but it is rather the fundamental element of our theory now

# 5. Special tensors

### 1. Kronecker delta (tensor)

- It is the identity tensor

$$\delta^\alpha_{\ \beta}v^\beta=v^\alpha$$

### 2. Levi-Civita pseudotensor

We can ask ourselves whether the Levi-Civita symbols, $\varepsilon_{\alpha\beta}$, are tensors. Said symbols are defined as:

$$\varepsilon_{11}=\varepsilon_{22}=0; \varepsilon_{12}=1; \varepsilon_{21}=-1$$

in **every** coordinate system.

If these were tensors, they should transform in a twice-covariant manner, so:

$$\bar \varepsilon_{\mu\nu}=\frac{\partial u^\alpha}{\partial \bar u^\mu}\frac{\partial u^\beta}{\partial \bar u^\nu}\varepsilon_{\alpha\beta}$$

Therefore, we have:

$$\bar \varepsilon_{11}=\bar\varepsilon_{22}=0; \bar\varepsilon_{12}=D; \bar\varepsilon_{21}=-D$$

and therefore they no longer fulfill their definition! So the Levi-Civita symbols are **not** tensors.

We can ask ourselves if there is something alike these symbols which does behave like a tensor. Consider the following definition:

$$\epsilon_{\alpha\beta}:=\sqrt g\varepsilon_{\alpha\beta}$$

These is called the Levi-Civita pseudotensor. It is defined as follows:

$$\epsilon_{11}=\epsilon_{22}=0; \epsilon_{12}=\sqrt g; \epsilon_{21}=-\sqrt g$$

where $\sqrt g$ is a term depending on the coordinate system! So they adopt different values depending on the coordinate system. Let us check if they transform like tensors by computing:

$$\bar \epsilon_{\mu\nu}=\frac{\partial u^\alpha}{\partial \bar u^\mu}\frac{\partial u^\beta}{\partial \bar u^\nu}\epsilon_{\alpha\beta}$$

we therefore get:

$$\bar\epsilon_{11}=\bar \epsilon_{22}=0; \bar \epsilon_{12}=+\sqrt g D; \bar\epsilon_{21}=-\sqrt g D$$

where $g$ is in the original coordinate system! This is key, because we can then introduce the term $D$ in the square root, and since $\bar g=D^2g$, we have:

$$\bar\epsilon_{11}=\bar \epsilon_{22}=0; \bar \epsilon_{12}=+\sqrt{\bar g} \cdot \text{sgn(D)}; \bar\epsilon_{21}=-\sqrt{\bar g} \cdot \text{sgn(D)}$$

So the Levi-Civita pseudotensors do transform like tensors as long as the transformation is such that the Jacobian's determinant is positive!

**Important note:** tensors shall always be tensors. Nonetheless, a pseudotensor could be a tensor if we restrict ourselves to a given transformation group. For instance, if our coordinate transformations are such that $D>0$, then the Levi-Civita pseudotensors are tensors. **The product of 2 pseudotensors is always a tensor**

**What do the Levi-Civita pseudotensors do?**

- Given two vectors of the tangent space of the surface, they yield the area formed by them (i.e: cross product)

$$\boxed{\epsilon_{\alpha\beta}v^\alpha w^\beta=(\vec v\times \vec w)\cdot \vec n=(\pm)||\vec v\times \vec w||}$$

- This can be generalized to an arbitrary dimension. For dimension 3 (the symbols / pseudovectors gain one index) everything is defined in the same way, the only difference being we must feed the Levi-Civita pseudotensors 3 vectors to get the volume they conform, that is why, in order to get the volume of a parallelepiped, we calculate the mixed product of its 3 generating vectors (i.e: we take their determinant)

### 3. First fundamental form:

- The first fundamental form obeys the following contraction rule:

$$g_{\alpha\beta}g^{\alpha\mu}=\delta^\mu_{\ \beta}$$

- That is to say, the first fundamental form with both of its two indices raised is the inverse of the first fundamental form with its two indices lowered
