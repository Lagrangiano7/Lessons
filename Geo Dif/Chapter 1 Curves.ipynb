# 1. Geometric property

**Def.:** a geometric property is one that is invariant under congruences and reparametrizations

**Note:** a curve is defined as a one-parameter vector function $\vec{x}(t):I\to \mathbb{R}^3$

## 1.1 Congruences

- All the bases we use shall be ordered (orientation matters)
- If you change bases, be sure the change of basis matrix $P$ fulfils $det(P)>0$

**Def.:** a transformation $T$ is an $\underline{\text{isometry}}$ if it conserves the distance between the transformed points. In particular, in $\mathbb{R}^N$ all isometries are linear transformations

$$(T\vec x, T\vec y)=(\vec x, \vec y) ;\ \ \forall \vec x, \vec y$$

An isometry takes on the general form:

$$T\vec x = R\vec x+\vec a$$

where $R$ is an orthogonal matrix and $\vec a$ is a constant translation vector. In order to preserve orientation, we shall only use $R \in SO(3)$ (rotations)

## 1.2 Diffeomorphisms

**Def.:** a diffeomorphism $\phi: J\to I$ is a one-variable, scalar function, which fulfils 3 properties:

1. $\phi$ is a bijection
2. $\phi \in C^\infty(J)$
2. $\phi^{-1}\in C^\infty (I)$

Diffeomorphisms are the perfect candidates to use for reparametrizing curves because they have all the properties a reparametrization needs (you can go back and forth between two parametrizations of the same curve, as well as for all of their derivatives)

**How do we reparametrize a curve?**

Let $\tilde{\vec x}(\bar t)$ be a curve, and let $t=\phi(\bar t)$ be a diffeomorphism. Then, the reparametrized curve is:

$$\tilde{\vec x}(\bar t)=\vec x(\phi(\bar t)) \equiv \vec x(t)$$

The derivative with respect to either curve's natural parameter will follow the chain rule:

$$\tilde{\vec x}(\bar t)'=\vec x(\phi(\bar t))'\phi'(\bar t)$$

## Length of a curve

**Def.:** the length of a portion of a parametrized curve is calculated as follows.

$$l = \int_{t_0}^{t_1} ||\vec x'(t)||dt$$

**The length of a curve is a geometric property**

## 1.3 Natural / arc length parameter

**Def.:** the arc length parameter specific to a curve is defined as follows:

$$s\equiv \int_{t_0}^t||\vec{x}'(u)||du$$

**Notation:** $\frac{d\vec x(t)}{dt}= \vec x'(t)$ but $\frac{d\vec x(s)}{ds}=\dot{ \vec x}(s)$

Of course, the derivative of $s$ with respect to $t$ is given by:

$$\frac{ds}{dt}=||\vec x'(t)||$$

Using the chain rule, one can compute the derivative of a curve with arbitrary parameter as follows:

$$\vec x(t)=\dot{\vec x}(s)||\vec x'(t)||$$

**Note:** the arc length parameter can be interpreted as the length of a curve starting to measure from $t_0$ and up to $t$

**Unit-speed curves:** a curve parametrized with its natural parameter is also known as "unit-speed curve" because it fulfils:

$$||\vec x(s)||=1$$

**Property: velocity and acceleration are orthogonal when using $s$:** $$\dot{\vec x}(s)\cdot \ddot{\vec x}(s)=0$$

# 2. Frenet curves

**Def.:** a regular curve is that whose velocity is never zero ($\vec v(t)\neq 0; \ \ \forall t\in I$). This depends on the parametrization of the curve (if a curve is not regular, it could have cusp points)

**Def.:** a Frenet curve is one whose $N-1$ first derivatives are linearly independent (where $N$ is the dimension we are working on)

**Def.:** a Frenet Reference Frame (FRF) is a set of $N-1$ (ordered) unitary orthogonal vectors. There is one such set for every point of the curve

**Properties of the FRF**

1. It forms an orthonormal base
2. The span of the base vectors is the same as the span of the $N-1$ first derivatives of the curve and they have the same orientation as these
3. The base is positively oriented
4. Every regular curve in $\mathbb{R}^2$ is a Frenet curve (because they only need their first derivative to be linearly independent, which means being nonzero for just one element, and this is precisely the meaning of "regular curve")

The elements of the FRF, $\{\vec e_i\}$, are calculated applying Gram-Schmidt to the N-1 first derivatives of the curve, and the last one, $\vec e_N$, is determined by finding the last orthogonal (unit) vector to the other $N-1$ we have already found

**WATCH OUT:** straight lines are Frenet curves in $\mathbb{R}^2$ but NOT in $\mathbb{R}^3$ (every vector is linearly dependent with 0, which is its second derivative)

## 2.1 Curvatures

**Def.:** to every Frenet curve we have associated $N-1$ functions called "curvatures". They are calculated as follows:

$$\kappa_i = \frac{\vec e_i'(t)\cdot \vec e_{i+1}(t)}{||\vec x'(t)||}$$

**Properties of curvatures:**

1. They are geometric properties
2. They are positive-definite safe for the last one
3. In 3 dimensions, the first curvature is called "curvature" ($\kappa$) and the second one is "torsion" ($\tau$)

## 2.2 Frenet equations

One can calculate the derivatives of the elements of the FRF as follows:

$$\begin{pmatrix}\vec e_1'\\ \vdots \\ \vec e_N'\end{pmatrix}||\vec x'(t)||\begin{pmatrix}0&\kappa_1(t)&\ldots & 0\\ -\kappa_1(t) & 0 & \ldots & 0\\ \vdots & \vdots & \vdots & \vdots\end{pmatrix}\begin{pmatrix}\vec e_1\\ \vdots \\ \vec e_N\end{pmatrix}$$

There are only elements in the two adjacent diagonals to the main diagonal

## 2.3 Fundamental theorem of curve theory

If two curves $\vec x(t)$ and $\tilde{\vec x}(t)$ (with the same parametrization) fulfill:

1. They are Frenet curves
2. They have the same speeds (modulus)
3. They have the same curvatures

Then there exists a **unique** congruence relating both curves:

$$\tilde{\vec x}(t)=R\vec x(t)+\vec a$$

## 2.4 Theorem of curvatures

Let $\{\kappa_i\}_{i=1}^{N-1}$ be $N-1$ functions with good behaviour. If all the functions, except for the last one, are positive definite, then they define a **single** curve, and they are its curvatures

## 2.5 Inflection points

**Def.:** an $\underline{\text{inflection point}}$ (defined in 2 and 3 dimensions) is a point in the parametrization $t_0$ such that:

$$\kappa(t_0)=0$$

## 2.6 Check if a curve is contained in a plane ("curva alabeada")

For a Frenet curve, all 3 statements are equivalent:

1. $\vec b=cte.$
2. The curve lies on a plane
3. $\tau(s)=0$ in some (finite) interval where the curve is defined

## 2.7 Reconstruction of a curve from its curvatures (only $\mathbb{R}^2$)

The velocity vector $\dot{\vec x}(s)$ is always tangent to the curve. It therefore forms some angle $\theta(s)$ with the parallel to the X axis in every point of the parametrization. If the velocity is parametrized with the natural parameter, then its modulus is unity, so we can write:

$$\dot{\vec x}(s)=(\cos(\theta(s)), \sin(\theta(s)))$$

Which leads to:

$$\boxed{\dot \theta(s)=\kappa(s)}$$

We can now reconstruct the curve as follows:

1. $\theta(s)=\int\kappa(s)ds+\theta_0$
2. $\begin{cases}x(s)=\int\cos(\theta(s))ds +x_0 \\ y(s)=\int\sin(\theta(s))ds+y_0\end{cases}$

We have 3 degrees of freedom (one for rotation and 2 for translation) which we can set to 0 and manipulate the curve after all the calculations.

**Observation:** a curve is completely determined by its curvatures, except for the aforementioned degrees of freedom. The curvatures give us the identity of the curve

## 2.8 Curves in $\mathbb{R}^3$ with natural parameter

**Frenet trihedron in 3 dimensions: (for s)**

$$\begin{cases}\vec t=\dot{\vec x}(s) \\ \vec p = \frac{\ddot{\vec x}(s)}{||\ddot{\vec x}(s)||} \\ \vec b = \vec t\times \vec p\end{cases}$$

**Frenet trihedron in 3 dimensions: (for arbitrary param. t)**

$$\begin{cases}\vec t=\frac{\vec x'(t)}{||\vec x'(t)||} \\ \vec p = \vec b\times \vec t \\ \vec b = \frac{\vec x'\times \vec x''}{||\vec x'\times \vec x''||}\end{cases}$$

**Names:**
- $\vec t$: tangent vector
- $\vec p$: principal normal vector
- $\vec b$: binormal vector
- $\vec t\vec p$ plane: osculating plane
- $\vec t\vec b$ plane: rectifying plane
- $\vec p\vec b$ plane: normal plane

**Frenet equations in 3 dimensions:**

$$\begin{pmatrix}\dot{\vec t}\\ \dot{\vec p} \\ \dot{\vec b}\end{pmatrix}=\begin{pmatrix}0 & \kappa & 0 \\ -\kappa & 0 & \tau \\ 0 & -\tau & 0\end{pmatrix}\begin{pmatrix}\vec t \\ \vec p \\ \vec b\end{pmatrix}$$

**Def.:** the curvature vector is given by:

$$\vec \kappa(s)=\kappa(s)\vec p(s)$$

**Def.:** the center of curvature / evolute of a curve is the locus of all its centers of curvature. It is given by:

$$\vec x_C(s) = \vec x+\frac{\vec p(s)}{\kappa(s)}$$

**Def.:** the radius of curvature of a curve is the distance between a point of the curve and its corresponding center of curvature. It is inmediate from the definition of evolute of a curve that the radius of curvature is the inverse of the curve's curvature:

$$\boxed{R(t) = ||\vec x_c(t)-\vec x(t)||=\frac{1}{\kappa(t)}}$$

### 2.8.1 Curvature and torsion in $\mathbb{R}^3$ curves:

**For s:**

$$\kappa(s)=||\ddot{\vec x}(s)||$$

$$\tau(s)=\frac{\text{det}(\dot{\vec x}(s), \ddot{\vec x}(s), \dddot{\vec x}(s))}{\kappa^2(s)}$$

**For arbitrary param. t:**

$$\kappa(t) = \frac{||\vec x'\times \vec x''||}{||\vec x'||^3}$$

$$\tau(t) = \frac{\text{det}(\vec x', \vec x'', \vec x''')}{||\vec x'\times \vec x''||^2}$$

## 2.9 Local form of a Frenet curve in $\mathbb{R}^3$

Let us expand an arbitrary Frenet curve parametrized with is natural parameter $s$ in the neighbourhood of a point $s_0=0$:

$$\vec x(s)=\vec x(0)+s\vec t(0)+\frac{s^2}{2}\kappa(0)\vec p(0)+\frac{s^3}{6}\kappa(0)\tau(0)\vec b(0)$$

In a more compact manner:

$$\vec x(s)-\vec x(0)=\left(s, \frac{s^2}{2}\kappa, \frac{s^3}{6}\kappa\tau\right) \to \text{Twisted cubic curve}$$

**Locally, every Frenet curve is a twisted cubic curve**

## 2.10 Types of curves in $\mathbb{R}^3$ according to curvature and torsion

1. Straight line: $\kappa(s)=0$
2. Plane curve: $\kappa(s)>0$ and $\tau(s)= 0$
3. Circumference: $\kappa(s)>0=cte.$ and $\tau(s)=0$
4. Cylindrical Helix: $\kappa(s)=cte.$ and $\tau(s)=cte.\neq 0$

Curvature of circular helix with $\omega=1$: $\kappa=\frac{a}{a^2+b^2}$ where $\vec x(t)=(a\cos(t),a\sin(t),bt)$
