# 1. Geodesic curvature

**Def.:** "geodesic curvature" is the projection of a curve's curvature vector $\vec \kappa$ onto the tangent space, which is the orthogonal complement of $\vec n$ by definition. Recall:

$$\vec \kappa=\vec \kappa_n+\vec \kappa_g$$

It can be thought of as the "freedom" a curve has to curve itself once it has paid the "toll" of the normal curvature for not popping out off the surface (the toll is the same for every curve with a given $\vec t$ on a point P)

**Note:** the geodesic curvature of a curve depends not only on the curve but also on the surface it is lying on (for instance, a great circle on a sphere has $\kappa_g=0$, but the same circumference on a plane has $\kappa_g\neq 0$)

## 1.1 Formulae in practice

Since the curvature vector is normal to $\vec t$ and since the normal projection is already accounted for by the normal curvature, and since the normal curvature and geodesic curvature vectors are, by construction, orthogonal, we have:

$$\vec \kappa_g\propto \vec n\times \vec t$$

More precisely, it is the projection of $\vec\kappa$ onto this direction. Therefore, we obtain:

$$\kappa_g=\vec\kappa\cdot(\vec n\times \vec t)$$

**With natural parameter (s):**

$$\boxed{\kappa_g = det(\vec t, \dot{\vec t}, \vec n)=det(\dot{\vec x}, \ddot{\vec x}, \vec n)}$$

**With arbitrary parameter (t):**

$$\boxed{\kappa_g = \frac{det(\vec x', \vec x'', \vec n)}{||\vec x'(t)||^3}}$$

**Note:** the geodesic curvature of a curve is inherent to the surface and the curve, so we only need the 1FF (and the curve) to compute it. It does not depend on the 2FF even if it carries the normal vector $\vec n$ (we shall see it can be calculated without the need of introducing the 2FF or even the normal vector)

## 1.2 Formulae in terms of the Christoffel symbols

- The curvature vector of a given curve is equal to $\ddot{\vec x}$. Let us decompose this vector in the Gauss trihedron:

$$\vec \kappa = \ddot{\vec x}=(\Gamma_{\beta\gamma}^\lambda \dot u^\beta \dot u^\gamma+\ddot u^\lambda)\vec x_\lambda + b_{\beta\gamma}\dot u^\beta\dot u^\gamma\vec n$$

- We have managed to decompose the curvature vector in terms of a normal component (normal curvature vector) and one that is orthogonal to the normal component, so it must be the geodesic curvature vector! We therefore have yet another expression for the geodesic curvature of a given curve:

$$\boxed{\kappa_g=\sqrt g\left(\Gamma_{\beta\gamma}^2\dot u^1\dot u^\beta\dot u^\gamma+\dot u^1\ddot u^2-\Gamma_{\beta\gamma}^1\dot u^2\dot u^\beta\dot u^\gamma-\dot u^2\ddot u^1\right)}$$

**Note:** for arbitrary parameter, just include an additional $||\vec x'(t)||^3$ in the denominator and switch dots to primes

# 2. Geodesics

**Def.:** a "pregeodesic" is a curve with $\kappa_g=0$

**Def. 1:** a "geodesic" is a pregeodesic parametrized with a parameter proportional to its natural parameter

**Def. 2:** a "geodesic" is a curve whose acceleration vector is normal to the surface in every point of the curve

**Note:** from this last definition, you could think a normal section is always a geodesic of the surface you are working with, since its acceleration is parallel to $\vec n$ in the point where you are taking the vector $\vec t$ to build the normal section. Nonetheless, this need not be the case for the rest of points of the curve, so normal sections, in general, are not geodesics

## 2.1 Equation for geodesics

- Since we already know the expression for the geodesic curvature, we only have to impose that it be 0 in order to obtain the equation governing geodesics:

$$\boxed{\ddot u^\lambda+\Gamma_{\beta\gamma}^\lambda \dot u^\beta \dot u^\gamma=0}$$

### Theorem of unicity of a geodesic

- Since the equations governing geodesics are a 2nd order ODE system, we can apply the (local) existence and unicity theorem of ODEs to said system. In essence, we have:

"Given a point and a direction, there exists only one geodesic which passes through that point"

**Note:** the existence and unicity of this geodesic is local, which means after some "distance" has passed, this geodesic could intersect, overlap, etc. with other geodesics. This is, that geodesic need not be unique for all of its domain, just in a neighbourhood of the point

**Note:** specifying a point means fixing initial position and specifying a direction means fixing an initial velocity. Therefore, these are the 4 initial conditions our 2nd order 2-ODE system --> unique solution (locally, since they are nonlinear. If they were linear, the unicity would hold for all of their domain)

### Corollary: straight lines

"A straight line in any given surface is always a geodesic in said surface"

- Recall they are also asymptotic lines, too, since their curvature is identically 0.

## 2.2 Geodesic curvature of coordinate axes

- In the plane, a circumference is not a geodesic curve. Nonetheless, the same circumference can be a geodesic when lying on a sphere if it is a great circumference of said sphere. Let us compute the geodesic curvature of the coordinate axes of a general surface
- We shall do this by letting the other component be constant

$$(\kappa_g)_{u_1}=-\Gamma_{22}^1\sqrt{\frac{g}{(g_{22})^3}}$$
$$(\kappa_g)_{u_2}=\Gamma_{11}^2\sqrt{\frac{g}{(g_{11})^3}}$$

### Corollary:

"A given coordinate axis is a geodesic if and only if the corresponding Christoffel symbol is 0"

### 2.2.1 Example: sphere

- The first fundamental form of a sphere in spherical coordinates is:

$$(g_{\alpha\beta})=\begin{pmatrix}1&0\\0&\sin^2\theta\end{pmatrix}$$

1. $\theta=cte.$ (parallels of the sphere)

$$(\kappa_g)_{\theta}=\text{cotan}(\theta)$$

- Therefore, the parallels are only geodesics if $\theta=\pi/2$ (of course, it is then a great circumference)

2. $\varphi=cte.$ (meridians of the sphere)

$$(\kappa_g)_{\varphi}=0$$

- Therefore, the meridians of the sphere are always geodesics of the sphere (of course, they are always great circumferences)

## Theorem (geodesics in 2 different surfaces)

"Let there be two surfaces, tangent to one another in a curve $C$. Then, $C$ is either a geodesic in both surfaces or it is not a geodesic in either surface"

**Justification:** the definition of 2 surfaces being tangent in a curve is they share the same normal vector in every point of the curve. Therefore, if the acceleration of $C$ is parallel to the normal of one of the surfaces, it is also normal to that of the other surface, since they share a normal in the intersection, and therefore it is either a geodesic in both surfaces or in none

# 3. Geometric Lagrangian

## Theorem of minimal arc length

"If there exists an arc of minimal length between two points on a surface, then it is an arc of a geodesic"

- Therefore, the arc length must be minimal!

$$\delta l = \delta \int_{t_0}^{t_1}\sqrt{g_{\alpha\beta}u'^\alpha u'^\beta}dt=0$$

This reminds us of the minimal action principle, where the integrand is the Lagrangian. Indeed, it can be shown that the Lagrangian:

$$\mathcal L = \frac{1}{2}g_{\alpha\beta}u'^\alpha u'^\beta$$

fulfills the Euler-Lagrange equations (arbitrary parameter $t$):

$$\frac{\partial\mathcal L}{\partial u^\mu}-\frac{d}{dt}\left(\frac{\mathcal L}{\partial \dot u^\mu}\right)=0$$

**Note:** the solutions to these equations are, by definition, pregeodesics. We can therefore reformulate the theorem of minimal length as follows: "in order for an arc $C$ of class $\geq 2$ on a surface $S$ with endpoints $P_1,P_2$ have minimum arc length among all neighbouring arcs on $S$ joining $P_1P_2$, $C$ must necessarily satisfy the Euler-Lagrange equations

- The explicit form of these equations are the equations for pregeodesics (with arbitrary parameter $t$):

$$u''^\mu+\Gamma_{\nu\sigma}^\mu u'^\nu u'^\sigma=\frac{s''}{s}u'^\mu$$

- Nonetheless, this includes an inhomogeneous term in the equation which complicates matters. Luckily, since we are interested in geodesics, we can solve the homogeneous equation for natural parameter and get the geodesics directly, so the problem is actually simpler

- If we expand the Euler-Lagrange equations for natural parameter, we recover the equations for geodesics, as expected

## 3.1 Conserved quantity

**Notice:** since we are working with a parameter that is at least proportional to the natural parameter, we can therefore state our geodesics shall be "constant speed" curves (maybe just not 1), so:

$$||\dot{\vec x}(s)||^2=g_{\alpha\beta}\dot u^\alpha\dot u^\beta=cte.$$

where the constant is fixed by the initia conditions of the geodesic. Therefore, this is just as a first integral / conserved quantity in Mechanics, and we can treat it as such. In fact, we need not solve the 2 equations for the geodesic's components, we can throw out one, since this conserved quantity, when derived, gives a linear combination of these 2 equations

## 3.2 Calculating Christoffel symbols in non-orthogonal coord. systems

- The equation for geodesics involves some Christoffel symbols, so if we want to calculate said symbols for a given coordinate system, we can simply formulate its Lagrangian, apply the Euler-Lagrange equations and compare the resulting equation with that of geodesics, so we cann read off the Christoffel symbols

- Let us exemplify this with the plane in polar coordinates:

$$(g_{\alpha\beta})=\begin{pmatrix}1&0\\0&r^2\end{pmatrix}$$

that is, $u^1=r; u^2=\theta$.

- Therefore, the Lagrangian is:

$$\mathcal L =\frac{1}{2}g_{\alpha\beta}\dot u^\alpha\dot u^\beta = \frac{1}{2}(g_{11}\dot r^2+g_{22}\dot\theta^2)=\frac{1}{2}\dot r^2+\frac{1}{2}r^2\dot \theta^2$$

- Now this has become a Classical Mechanics problem. Applying the Euler-Lagrange equations to this Lagrangian for both variables, we get

$$\ddot r-r\dot\theta\dot\theta=0$$
$$\ddot \theta+\frac{2}{r}\dot r\dot \theta=0$$

- From the first equation, it is clear that: $\Gamma_{22}^1=\Gamma_{\theta\theta}^1=-r$
- From the second one, we could have a Christoffel symbol with 2 different lower indices, so how can we decide which one it belongs to? Well, since they must be symmetric, we must divide it in half! --> $\Gamma_{12}^2=\Gamma_{21}^2=\frac{1}{r}$
- The rest of the Christoffel symbols for this coordinate system are identically 0

# 4. Gauss-Bonnet theorem and connection of geometric-topologic properties of manifolds

**Note:** the Gauss-Bonnet theorem is just for closed regions of a given surface. We can then triangulate the whole surface and apply the theorem to each triangle, obtaining a connection between the integral curvature of the surface (geometric property) and the topology of the surface

- We begin by stating the Gauss-Bonnet theorem

## 4.1 Gauss-Bonnet theorem

- Let $R$ be a simply connected portion of a surface $S$
- If there exists a representation $\vec x(u^1, u^2)$ of class $\geq 3$ for said surface $S$
- If the boundary of the region is a curve $C$ is a simple closed curve with representation $\vec x(u^1(s), u^2(s))$ of class $\geq 2$

Then the following equality holds in $R$:

$$\boxed{\int_C\kappa_g ds+\iint_R KdA=2\pi}$$

### Generalization

- If the curve delimiting the region $R$ in the surface is not smooth but instead it has $n$ vertices with their respective angles $\{\alpha_j\}$, then the theorem is generalized to the following:

$$\boxed{\int_C\kappa_g ds+\iint_R KdA=\displaystyle\sum_{j=1}^n\alpha_j+(2-n)\pi}$$

**Note:** of course, if we ask for the curve to be smooth (no vertices), we recover the original formula

**Note:** the sense of integration along $C$ follows the right-hand rule. In practice, integrate always anticlockwise along $C$ so that the normal vector of the region points towards you

### Excess angle and interpretation of Gaussian curvature

- Figure a surface $S$ on which we draw a geodesic triangle (a triangle whose 3 sides are arcs of geodesics)
- We therefore have:

$$\iint_T KdA=\alpha_1+\alpha_2+\alpha_3-\pi=\Delta$$

- Where $\Delta$ is the excess angle
- The GB theorem does not impose any constraints on the area of the triangle, so we can make it infinitely small, and therefore $K\approx cte.$ in that region. Therefore:

$$K=\displaystyle\lim_{A\to 0}\frac{\Delta}{A}$$

and we can understand the Gaussian curvature as a measure of the excess angle of a geodesic triangle on said surface. Therefore, a region with $K=0$ will "behave" like a plane (the angles of the infinitely small triangle add up to $180^\circ$ as they do on a plane), and a region with $K>0$ will have a positieve excess angle (the angles of the infinitely small triangle will add up to more than they do on the plane). Finally, for a region where $K<0$, they will add up to less

- We can therefore understand Gaussian curvature precisely as how a surface bends locally

## Theorem

"On a surface $S$ of class $\geq 3$ for which $K\leq 0$ everywhere, there do not exist closed geodesics such that one of them constitutes the entire boundary of a bounded portion of $S$"

- For instance, in the plane the geodesics are straight lines and therefore we cannot enclose a closed region $R$ with a single geodesic. On the other hand, on the sphere we can, just take a great circle and you have enclosed the two halves of the sphere

## 4.2 Application to closed surfaces

**Idea:** the Gauss-Bonnet theorem applies to regions on a surface. Nonetheless, we can triangulate the entire surface and apply the theorem to each triangle (or triangulation element) and then add up all the contributions so as to try and obtain a result that is general for the surface. We shall also find that it is independent of the triangulation carried out 

**Def.:** a "closed surface" is one that is closed and bounded (what you would ordinarily call "compact region")

**Theorem:** any closed surface can be triangulated

**Theorem:** any closed orientable surface is homeomorphic to one of the infinite normal types of closed orientable surfaces, and these can be obtained from attaching $p$ "handles" or making $p$ holes to a sphere

**Note:** a "homeomorphism" is a transformation leading from one topological space to another such that it is (1) bijective, (2) continuous and (3) its inverse is also continuous. In practice, you can morph one topological object into another one continously and reversibly

- Applying the GB theorem to each element of the triangulation and adding them all up, we reach the following result:

**Note:** the "genus" ($p$) of a surface is the number of holes it has. For instance, a sphere has $p=0$, a toroid has $p=1$, etc.

## Theorem (genus)

"Any closed orientable surface $S$ of genus $p$ for which a representation of class $\geq 3$ exists has the integral curvature given by:"

$$\boxed{\iint_S KdA=4\pi(1-p)}$$

- Where we now integrate over all the (closed) surface, $S$, in opposition to the GB theorem, where we integrate over a closed region of $S$

**Note:** alternatively, we can formulate this theorem in terms of the Euler characteristic

## Theorem (Euler characteristic)

"Any closed orientable surface $S$ for which a representation of class $\geq 3$ exists has the integral curve given by:"

$$\boxed{\iint_SKdA=2\pi\chi(S)}$$

- where $\chi(S)=V-E+F$ is the "Euler characteristic" of a surface
- $E$: number of independent edges employed in the triangulation
- $V$: number of independent vertices employed in the triangulation
- $F$: number of faces employed in the triangulation

**Note:** the Euler characteristic of a surface is independent of which triangulation we choose

**Note:** the Euler characteristic of a surface and its genus are closedly related:

$$\boxed{\chi(S)=2(1-p)}$$

### Corollary

"Every 3D polyhedron with no holes in it has the same Euler characteristic, which is $\chi = 2$"

**Proof:** they are all homeomorphic to the sphere, which has said Euler characteristic, and therefore they have it too. This means for every 3D polyhedron we can imagine (with no holes in it), if we add up its vertices and faces and subtract the edges, we get 2
