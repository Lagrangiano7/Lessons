# 0. Introducción

- En el tema anterior hemos estudiado las interacciones entre átomos de la red
- Ahora, vamos a estudiar cómo se comportan los electrones del sólido
- Los electrones del sólido están sometidos a 3 interacciones distintas:

1. Atracción electrostática de los átomos de la red
2. Repulsión electrostática con otros electrones
3. Interacciones con fonones de la red

- En este tema solo consideraremos la primera interacción, o ninguna. Veremos 3 modelos:

1. Aproxiación de electrones libres
2. Aproximación de electrones casi libres
3. Aproximación de enlace fuerte

**Objetivos principales:**

1. Distinguir entre metales, semiconductores y aislantes en base a su estructura de bandas y la disposición de los electrones en ellas
2. Predecir propiedades físicas de los conductores (conductividad eléctrica, térmica y capacidad calorífica)

**IMPORTANTE:** a lo largo de todo este tema, admitiremos que los iones se encuentran estáticos en sus posiciones de equilibrio, esto es, no oscilan como en el tema anterior. Esto complicaría mucho los cálculos y la contribución de las oscilaciones de la red a las magnitudes macroscópicas ya se ha tenido en cuenta en el tema anterior (estrictamente, habría que tenerlo en cuenta en el potencial que sienten los electrones, porque los iones no están estáticos así que el potencial tampoco lo estará, pero admitimos que las oscilaciones son suficientemente pequeñas como para considerarlos estáticos)

# 1. Aproximación de electrones libres

**Concepto:** los electrones del sólido no interactúan con la red ni con los otros electrones (son libres, y corresponden a los electrones de valencia por ser los menos ligados). Sistema de $N$ fermiones de spin $s=1/2$ confinados en un potencial de caja tridimensional

**En la práctica:** modelamos los electrones de un conductor como un gas de Fermi

**Predicciones:**

1. Capacidad calorífica de un metal
2. Conductividad eléctrica de un metal
3. Relación entre conductividad térmica y eléctrica

**Problemas:**

1. No explica por qué a veces parece que los portadores de carga son positivos (huecos, necesita teoría de bandas)
2. Solo explica metales, y sus predicciones se desvían considerablemente de las medidas experimentales

## 1.1 Energía de Fermi $\epsilon_F$

**Def.:** la "energía de Fermi" ($\varepsilon_F$) es la energía del nivel más alto ocupado en un sistema de fermiones indistinguibles (formalismo de números de ocupación). Véase que este nivel es para 1 sola partícula, no es la energía del nivel considerando todas las partículas

**Def.:** la estadística de Fermi-Dirac es una distribución que tiene en cuenta el principio de exclusión de Pauli y que se aplica por tanto a fermiones. Se interpreta como la probabilidad de que un nivel dado esté ocupado.

$$f(E, T)=\frac{1}{z^{-1}e^{\beta E}+1}$$

- Cumple la propiedad de que a $T=0(K)$ se vuelve una función escalón con límite en $E=\varepsilon_F$

## 1.3 Relación de dispersión $E(\vec k)$

**Def.:** la "relación de dispersión" de un sistema condensado se entiende en general como la relación entre la frecuencia de sus modos de vibración y sus longitudes de onda. No obstante, en sistemas cuánticos extendemos la definición a la relación entre la energía de una onda plana y su vector de onda correspondiente

- El Hamiltoniano de una sola partícula en un potencial de caja tridimensional coincide con su energía cinética, de modo que:

$$H=\frac{\vec p^2}{2m}$$

- Esto da lugar a un espectro cuantizado tal que:

$$\boxed{E=\frac{\hbar^2k^2}{2m}; \ \ k_i=\frac{2\pi n}{L}}$$

- Además, como el Hamiltoniano solo depende del momento, $[H,p_i]=0$ así que los autoestados del Hamiltoniano serán ondas planas

**Def.:** se define la "superficie de Fermi" como el espacio geométrico en el espacio de las k's tal que dé lugar a la energía de Fermi. La superficie de Fermi contiene todas las demás superficies correspondientes a energías menores. En aproximación de electrones libres, mirando la relación de dispersión vemos que se trata de una esfera

- Así como se define la energía de Fermi, se definen otras tantas magnitudes a partir de esta:

1. Temperatura de Fermi: $T_F=\frac{\varepsilon_F}{k_B}$
2. Número de onda de Fermi: $k_F=\sqrt{\frac{2m\varepsilon_F}{\hbar^2}}$
3. Momento de Fermi: $p_F=\hbar k_F$
4. Velocidad de Fermi: $v_F=\frac{\hbar}{m}k_F$ ó $\varepsilon_F=\frac{1}{2}mv_F^2$

## 1.2 Densidad de estados $g(E)$

**Nota:** el número de partículas del sistema se debe mantener constante si el sistema es cerrado (en este modelo, el sistema no puede perder partículas porque están sometidas a un potencial de caja). Por tanto, el número de partículas del sistema a cualquier temperatura debe coincidir con el que hay a 0 absoluto. Como en 0 absoluto la función de Fermi da 1 si el nivel está ocupado y 0 si no lo está, el número total de partículas a 0 absoluto (y a cualquier otra $T$) será:

$$N=2\displaystyle\sum_{k_x,k_y,k_z}^\infty \frac{1}{z^{-1}e^{\beta E}+1}\approx 2\int \frac{d^3\vec r}{(\Delta k)^3}f(E, T)$$

- donde $\Delta k=\frac{2\pi}{L}$ y el factor $2$ viene de la degeneración de spin (me cabe en cada nivel el doble de fermiones que si los considerarse como los trata la estadística de Fermi, donde solo cabe uno por nivel, por construcción). Tomando este paso al continuo y sabiendo que la relación de dispersión solo depende del módulo de $\vec k$, podemos integrar en esféricas

$$N\approx 2\frac{L^3}{8\pi^3}\int 4\pi^2k^2dkf(k,T)$$

- Empleando la relación de disprsión, sabemos que $k=\sqrt{\frac{2m}{\hbar^2}}\sqrt E$, por lo que:

$$N\approx V\int_0^\infty \frac{\sqrt{2m^3}}{\pi^2\hbar^3}\sqrt Ef(E, T)dE$$

**Def.:** se define la "densidad de estados" $g(E)$ como el número de estados existentes cuya energía está entre $E$ y $E+dE$ (por unidad de volumen):

$$\boxed{g(E):=\frac{\sqrt{2m^3}}{\pi^2\hbar^3}\sqrt E}$$

- Entonces, el número de partículas queda (haciendo $T=0$):

$$N=V\int_0^{\varepsilon_F}g(E)dE$$

Así que la energía de Fermi se puede expresar como función de la concentración de los electrones $n=N/V$:

$$\boxed{\varepsilon_F = \frac{\hbar^2}{2m}(3\pi^2n)^{2/3}}$$

- Que recuerda a la frecuencia de Debye. En ambos casos, indican el nivel máximo que se puede ocupar, y solo dependen de la concentración

- La densidad de energía a $T=0(K)$ es:

$$\boxed{u=\frac{3}{5}n\varepsilon_F}$$

### ¿Qué pasa cuando $T\neq 0$?

**Def.:** se define la "temperatura de Fermi" como:

$$T_F=\frac{\varepsilon_F}{k_B}\sim 10^4(K)$$

- Si $T>>T_F$, la estadística de Fermi-Dirac tiende a la de Maxwell-Boltzmann. Los electrones pierden su comportamiento fermiónico (fuera exclusión de Pauli porque están demasiado alejados) y el gas de electrones pasa a comportarse de forma idéntica a un gas ideal a todos los efectos (solo que cargado, pero da igual porque no consideramos interacciones entre electrones)

- Si $T<< T_F$, podemos hacer desarrollo de Sommerfeld en torno a $E=\varepsilon_F$ porque al final es el único sitio donde la distribución de Fermi varía apreciablemente (sigue siendo en buena aproximación una funcióne escalón)

En este último caso, el potencial químico ya no coincide con la energía de Fermi como en $T=0$, sino que hay una corrección (pequeña):

$$\mu(T)=\varepsilon_F\left(1-\frac{\pi^2}{12}\left(\frac{T}{T_F}\right)^2\right)$$

## 1.4 Capacidad calorífica $C_V(T)$

- De acuerdo a la distribución clásica de Maxwell-Boltzmann, si los electrones son partículas libres, a cada uno le corresponde una contribución a la capacidad calorífica de $\frac{3}{2}k_B$ (teorema de equipartición)
- Experimentalmente, a temperatura ambiente se registra que la contribución de los electrones es un $1\%$ de esto

**Concepto clave:** clásicamente no se considera el principio de exclusión de Pauli, de modo que se entiende que todos los electrones son libres de conducir la electricidad, promocionando a niveles de energía superiores. Sin emabargo, si reparamos en la forma de la función de Fermi, veremos que solo los electrones en los niveles de energía $E\in\varepsilon_F\pm k_BT$ pueden promocionar, de modo que solo estos adquirirán ese paquete de energía del orden de $k_B T$ al calentar la muestra, no todos ellos.

- El número de electrones excitados en un gas de electrones a temperatura $T$ es:

$$\boxed{n_{exc}=\frac{3}{2}n\frac{T}{T_F}}$$

- La capacidad calorífica de un gas de electrones a temperatura $T<<T_F$ es:

$$\boxed{C_{V, e^-}=\frac{\pi^2}{2}Nk_B\frac{T}{T_F}}$$

- A temperaturas muy por debajo de $\theta_D$, aplica la ley $T^3$ de Debye. A temperaturas muy por debajo de $T_F$, aplica la expresión que acabamos de hallar. Entonces, como la capacidad calorífica es extensiva, a temperaturas muy por debajo de estas dos, tenemos que:

$$C_{V,sol}=AT+BT^3$$

- Si graficamos $C_V/T$ frente a $T^2$, veremos que obtenemos una recta. La pendiente da la contribución del sólido a la capacidad calorífica del sólido y la ordenada de origen da la de los electrones

**Nota:** el modelo de electrones libres solo vale para conductores. Los aislantes, por tanto, tendrán ordenada de origen nula en este gráfico, porque los electrones no conducen electricidad, de modo que no se mueven y por tanto no recogen ese paquete de energía del orden de $k_BT$. Sin embargo, a temperaturas muy bajas, son los electrones los que predominan en la capacidad calorífica de los conductores (para ellos no se va a 0 en T=0 como predecía el modelo de Debye por sí solo)

**Problema:** la capacidad calorífica de los electrones en este modelo es del orden de magnitud de lo que dan las mediciones experimentales, pero en general discrepan

**Def.:** se define la "masa térmica efectiva" ($m_t$) como el cociente entre las ordenadas de origen experimental y teórica por la masa del electrón:

$$m_t:=\frac{A_{exp}}{A_{teo}}m_e$$

- El propósito de esto es entenderlo como que las partículas del sistema no tienen exactamente la masa del electrón sino una masa térmica efectiva, de modo que este modelo da la contribución correcta de la capacidad calorífica de los electrones si se considera que estos tienen masa $m_t$. Normalmente es $m_t\sim 1.2m_e$ pero varía bastante entre $0.5$ y $2$

## 1.5 Conductividad eléctrica de un gas de electrones ($\sigma$)

- Si sometemos un conductor a un campo eléctrico externo, las cargas se verán aceleradas, pero a su vez colisionarán con las impurezas de la red y los fonones, dando lugar a un estado estacionario en el que las cargas, en promedio, viajan en grupo a velocidad constante

- Si el tiempo de colisión (tanto con impurezas como con fonones) es $\tau$, entonces tenemos:

$$m\frac{d\vec v}{dt}=-e\vec E$$

- En un intervalo pequeño de tiempos $\Delta t=\tau$, se cumple entonces que:

$$\vec v=-\frac{e}{m}\vec E\tau$$

- Como la densidad de corriente es:

$$\vec J_e=-ne\vec v=\frac{ne^2\tau}{m}\vec E$$

- De modo que, considerando el conductor óhmico, podemos extraer su conductividad eléctrica:

$$\boxed{\sigma=\frac{ne^2\tau}{m}}$$

## 1.6 Conductividad térmica de un gas de electrones ($\kappa$)

- Un sistema cuyas partículas se desplazan a velocidad $v$ con capacidad calorífica $C$ y recorrido libre medio $l$ tiene asociada una conductividad térmica:

$$\kappa=\frac{1}{3}\frac{C}{V}vl$$

- En el gas de electrones, el recorrido libre medio es $l=v_F \tau$, de modo que el gas de electrones tiene asociada una conductividad térmica:

$$\boxed{\kappa=\frac{\pi^2nk_B^2T\tau}{3m}}$$

## 1.7 Relación $\sigma-\kappa$ (Ley de Wiedemann-Franz)

- En un gas de electrones, a $T$ no demasiado baja ($0^\circ C-100^\circ C$), la conductividad térmica y la eléctrica se relacionan de la forma:

$$\boxed{\frac{\kappa}{\sigma}=\frac{\pi^2}{3}\left(\frac{k_B}{e}\right)^2T}$$

- Esta relación es independiente de cualquier propiedad distintiva de los metales, de modo que es una ley universal para todos aquellos que se adhieran bien al modelo de gas de electrones

**Def.:** se define el "número de Lorentz" como:

$$L:=\frac{\kappa}{\sigma T}=\frac{\pi^2}{3}\left(\frac{k_B}{e}\right)^2$$

# 2. Aproximación de electrones casi libres

**Concepto clave:** admitimos gas de electrones **libres** pero introducimos la interacción de los electrones con los iones de la red como un término perturbativo. NO consideramos la interacción e-e

## Aparición de bandas y gaps

**Modelo del potencial:** admitimos que el potencial originado por los iones de la red es periódico en la red directa, porque la posición de los iones en la red también lo es. Por ende, se puede descomponer en serie de Fourier en términos de los vectores de la red recíproca:

$$V(\vec r+\vec R)=V(\vec r)\Longrightarrow V(\vec r)=\displaystyle\sum_{\vec G}V_{\vec G}e^{i\vec G\cdot \vec r}$$

con $V_{\vec G}=\frac{1}{V_{celda}}\int_{celda}e^{-i\vec G\cdot \vec r}d^3\vec r$

### Teorema de Bloch

"Para un potencial periódico, las soluciones de la ecuación de Schrödinger independiente del tiempo cumplen la propiedad":

$$\Psi_{\vec k}(\vec r+\vec R)=e^{i\vec k\cdot \vec R}\Psi_{\vec k}(\vec r)$$

- Esto es, no son periódicas, pero casi. Podemos definir las funciones de Bloch de la forma:

$$u_{\vec k}=e^{-i\vec k\cdot \vec r}\Psi_{\vec k}(\vec r)$$

- Y estas sí son periódicas. Por tanto, podemos reformular el teorema de Bloch como sigue:

"Para un potencial periódico, las soluciones de la ecuación de Schrödinger independiente del tiempo son funciones de Bloch moduladas por una onda plana:"

$$\Psi_{n, \vec k}(\vec r)=e^{i\vec k\cdot \vec r}u_{n, \vec k}(\vec r)$$

- donde el subíndice $n=1,2,3,\ldots$ hace referencia a la banda en la que nos encontramos. Las bandas son el análogo de las ramas en teoría de fonones

### Teoría de perturbaciones

- A orden 0, los electrones son del todo libres y volvemos al modelo anterior, de modo que:

$$E^{(0)}=\frac{\hbar^2k^2}{2m}$$

$$\Psi^{(0)}_{\vec k}(\vec r)=\frac{1}{\sqrt V}e^{i\vec k\cdot \vec r}$$

- Hagamos ahora teoría de perturbaciones a primer y segundo orden para ver los efectos de la corrección introducida por nuestro potencial periódico de iones de la red

- A orden 0, en cualquier sección la relación de dispersión forma una parábola. Esto quiere decir que la energía en un punto y en dicho punto desplazado cualquier vector de la red recíproca $\vec G$ tendrán energías muy diferentes y no habrá degeneración. Sin embargo, en los límites de la 1ZB se cumple la condición de Bragg, y se puede ver que la energía es idéntica a uno y otro extremo de 1ZB --> caso degenerado

**Caso no-degenerado (lejos de los límites de 1ZB):**

$$E_{\vec k}\approx E^{(0)}_{\vec k}+\langle \vec k|V|\vec k\rangle+\displaystyle\sum_{\vec k'}\frac{|\langle \vec k|V|\vec k'\rangle|^2}{E_{\vec k}^{0}-E_{\vec k'}^{(0)}}$$

- La relación de dispersión lejos de los límites de 1ZB queda entonces de la forma (tomando el origen de energías en el término a primer orden):

$$\boxed{E_{\vec k}=E_{\vec k}^{(0)}+\displaystyle\sum_{\vec G\neq 0}\frac{|V_{\vec G}|^2}{E_{\vec k}^{(0)}-E_{\vec k-\vec G}^{(0)}}}$$

**Caso degenerado (cerca de los límites de 1ZB):**

- La degeneración tiene lugar para los estados $| -\vec G/2\rangle, |\vec G/2\rangle$

$$\boxed{E_{\vec k}=E_{\vec G/2}^{0}\pm |V_{\vec G}|}$$

**Observación:** en los límites de 1ZB, el espectro se desdobla en 2. Esto es, hay dos "ramas" en la relación de dispersión de la energía (en general, infinitas). Entre ellas, existe un gap (región de energía no permitida) de anchura:

$$\boxed{\Delta E_{gap, \vec G}=2|V_{\vec G}|}$$

## ¿Por qué aparecen bandas y gaps según este modelo? (Explicación en 1D)

**Concepto clave:** el desdoblamiento lo da la perturbación, de modo que la clave debe estar en el potencial de los iones de la red. Cuando ponemos muchos átomos juntos, sus niveles de energía se solapan para dar lugar a bandas (prácticamente) continuas de energía permitida separasas por gaps, donde para un solo átomo había ciertos valores de energía permitidos y el resto de intervalos entre ellos eran prohibidos (espectro discreto ya no lo es)

- En la 1ZB se cumple la condición de Bragg, de modo que las ondas (de cualquier naturaleza, incluidos los electrones) que pasen por ella se verán difractados
- En cristal unidimensional, esto equivale a que se vean reflejados --> función de onda entrante interfiere con función de onda saliente --> patrón de ondas estacionarias

- Cuando las ondas entrante y saliente interfieran constructivamente en 1ZB, $|\Psi_+|^2\propto \cos^2\left(\frac{\pi}{a}x\right)$
- Cuando interfieran destructivamente, $|\Psi_+|^2\propto \sin^2\left(\frac{\pi}{a}x\right)$
- En el primer caso, la probabilidad de presencia de los electrones tendrá máximos sobre los puntos de la red, que es donde están los iones --> potencial Coulombiano de carga opuesta --> energía menor
- En el segundo caso, estos máximos de probabilidad de presencia estarán entre dos puntos de la red --> energía mayor

**Resultado:** dos funciones de onda con el mismo vector de onda originan dos energías distintas (porque en un caso interfieren constructivamente al reflejarse y en el otro lo hacen destructivamente)
