# 1. Fuerzas de dispersión

- No hace falta que haya carga neta, son fuerzas de dipolo instantáneo inducido
- Presente en todas las interacciones pero es la más débil
- Es predominante en gases nobles porque no tienen ningún otro tipo de interacción (orbitales completos)

**Def.:** la "energía de cohesión" es la energía que debemos suplir al sólido cristalino para disociarlo en sus átomos **neutros** constituyentes y llevarlos al infinito (en reposo) manteniendo su configuración electrónica

**Def.:** la "energía reticular / de red" es la energía que debemos suplir al sólido cristalino para disociarlo en sus iones constituyentes y llevarlos al infinito (en reposo)

$$U_{coh}=2N_T U_0 \left[c_{12}\left(\frac{\sigma}{R}\right)^{12}-c_6\left(\frac{\sigma}{R}\right)^6\right]$$

- $c_n=\displaystyle\sum_{i\neq j}^\infty d_{ij}^{-n}$

**Fíjate:** estos coeficientes solo dependen del tipo de red de Bravais considerado

- Estos coeficientes se pueden calcular por ordenador y obtenemos
    - FCC: $c_6 = 14.45293$   ;   $c_{12}=12.13188$
 
 **Posición de equilibrio:** es la posición de mínima energía $\frac{dU}{dR}\Big |_{R_0}=0$
 
 $$R_0 = 1.09\sigma$$
 
 **Módulo de compresibilidad:** $B=V\frac{\partial^2U}{\partial V^2}$
 
 para FCC: $B = 75.15\frac{U_0}{\sigma^3}$

# 2. Enlace iónico

- Solo estudiaremos cristales iónicos con estequiometría 1:1 (i.e: halógenos con metales alcalinos)
- Se forman porque es energéticamente favorable tener la capa de valencia llena, y como uno necesita soltar un electrón y otro necesita añadir un electrón, se juntan y forman un cristal
- Los sólidos iónicos se mantienen juntos por la atracción EM de los iones (cada ion está rodeado de iones de carga opuesta)
- El potencial ha de tener en cuenta la atracción EM y la repulsión EM + otras interacciones como la repulsión por exclusión de Pauli. Hay dos formas de tener este último factor repulsivo en cuenta:

1. $E_r(R)=\lambda e^{-R/\rho}$;   $\rho\sim 0.3 A$
2. $E_r(R)=\frac{\lambda}{R^n}$;   $n=9,12,\ldots$

**Nota:** el factor repulsivo decae tan rápido (Pauli) que basta con tenerlo en cuenta solo para primeros vecinos

$$U_{ij}=\begin{cases}E_r(R)-\frac{q^2}{R} \text{ si primeros vecinos} \\ \pm\frac{q^2}{Rp_{ij}} \text{ para el resto}\end{cases}$$

La energía total vendrá de sumar la energía de cada partícula:

$$U_T = N\left(z\lambda e^{-R/\rho}-\alpha\frac{q^2}{R}\right) = -\frac{N\alpha q^2}{R_0}\left(1-\frac{\rho}{R_0}\right)$$

**OJO!:** $N$ es solamente el número de partículas de una especie de las dos que componen el cristal! $N=\frac{1}{2}N_T$

- $z$: número de primeros vecinos
- $\alpha = \displaystyle\sum_{i\neq j}\frac{(\pm)}{p_{ij}}$: cte. de Madelung. Signo + si tienen distinto signo

# 3. Enlace covalente

- Es un enlace formado cuando se "comparten electrones" entre 2 átomos (las funciones de onda están extendidas como una sola a todos los átomos que participan del enlace)
- El ejemplo más simple de enlace covalente es H$_2$, y hay 2 tipos:

1. Enlazante: spines antiparalelos (nube electrónica con interf. constructiva entre los dos núcleos) (energéticamente favorable)
2. Antienlazante: spines paralelos (interf. destructiva entre los dos núcleos)

## Hibridaciones

El solapamiento de los orbitales enlazantes es máximo si se produce la hibridación de los $s$ y $p$

1. $sp^2$: simetría hexagonal
2. $sp^3$: forma tetraédrica, más difícil de obtener que la $sp^2$ pero mucho más fuerte (diamante vs. grafito / grafeno)

# 4. Enlace metálico

- Enlace no-direccional entre átomos tanto idénticos como diferentes, siempre que sean metálicos (esto permite las amalgamas)
- Electrones están deslocalizados en el cristal
- 2 modelos:

1. Modelo de electrones libres (ningún electrón se ve afectado por los demás)
2. Modelo de enlace fuerte (la función de onda de los electrones es comb. lineal de los orbitales atómicos)
