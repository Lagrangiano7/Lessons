# 1. Red directa

**Def.:** la estructura cristalina se compone de (1) una red cristalina y (2) una base

**Def.:** la red cristalina (red directa) es un patrón periódico de extensión infinita de puntos matemáticos

**Def.:** la base son las especies químicas que componen el sólido estudiado, que se sitúan en los puntos de la red. La base puede ser única, doble, triple, etc. De haber varias especies químicas, los puntos de la red vienen ocupados por UNA sola de las especies químicas y las otras especies de la base se salen de los puntos de la red directa

**Nota:** la red cristalina se compone de 6 parámetros de red: los módulos de los 3 vectores que conforman su base y los ángulos que forman entre sí

**¿Cómo especifico dónde van los elementos que componen el cristal?:** Tengo que dar 1 posición por cada elemento de la base. Uno suele ser el $(0,0,0)$ y el otro uno en general con coordenadas racionales. Para construir la celdad unidad, basta con situar el primer elemento en todos los puntos de dicha celda y luego equidistante el segundo elemento de la base.

**OJO!:** la base del cristal, una vez tomemos un punto de la red cristalina como origen, se define como $\vec a_i=a\hat i+b\hat j+c\hat k$, donde $a,b,c$ son las constantes de red del cristal (las dimensiones del paralelepípedo que conforma la celda unidad). Las posiciones después se dan en esta base, en coordenadas, de modo que no incluimos nuevamente los parámetros de red (e.g: $\vec r=(1,2,0)\equiv 1\cdot \vec a_1+2\cdot\vec a_2$)

**Propiedades de simetría:** la red directa de un sólido cristalino ideal tiene algunas propiedades relacionadas con su simetría. Las hay de 2 tipos: de traslación y puntuales:

1. Simetría de traslación: la red cristalina se construye mediante la repetición periódica de una celda unidad. Por tanto, es de esperar que quede inalterada bajo la traslación de cualquier vector de red $\vec R_{mnp}$

2. Operaciones puntuales de simetría:

- Rotación en torno a un eje (los únicos órdenes de rotación compatibles con la simetría de traslación y por tanto permitidos son 1, 2, 3, 4, 6). Un orden de rotación es el número de veces que podríamos hacer una rotación hasta que el sólido estuviera otra vez en la configuración inicial si pusiéramos etiquetas a los puntos de la red (por ejemplo, un cuadrado tiene orden 4 en torno a su eje central)
- Reflexión con un plano
- Simetría de inversión (con respecto a un centro de simetría) ($\vec r\to -\vec r$ con respecto a algún punto, no necesariamente punto de la red)
- Rotación impropia (rotación + inversión)

## 1.1 Tipos de redes (14 redes de Bravais)

**7 tipos de celdas unidad:**

1. Cúbica $\to \begin{cases}\text{Cúbica simple (SC)} \\ \text{Cúbica centrada en el cuerpo (BCC)}\\ \text{Cúbica centrada en las caras (FCC)}\end{cases}; \ \ \  \boxed{a=b=c; \ \ \alpha=\beta=\gamma=90^\circ}$
2. Tetragonal   $\boxed{a=b\neq c; \ \ \alpha=\beta=\gamma=90^\circ}$
3. Ortorrómbica   $\boxed{a\neq b\neq c; \ \ \alpha=\beta=\gamma=90^\circ}$
4. Trigonal / Romboédrica  $\boxed{a=b=c; \ \ \alpha=\beta=\gamma < 120^\circ \neq 90^\circ}$
5. Hexagonal  $\boxed{a=b\neq c; \ \ \alpha=\beta=90^\circ; \gamma=120^\circ}$
6. Monoclínica $\boxed{a\neq b\neq c; \ \ \alpha=\gamma=90^\circ; \beta\neq 90^\circ}$
7. Triclínica  $\boxed{a\neq b\neq c; \ \ \alpha\neq 90^\circ; \beta\neq 90^\circ; \gamma\neq 90^\circ}$

Todos los subtipos de celda unidad en que se desglosan estos 7 tipos principales componen las 14 redes de Bravais

![REDES](redes_de_bravais.png)

**Def.:** una celda primitiva es aquella celda unidad que contiene 1 solo punto de la red directa. Hay en general varios tipos de celdas primitivas. Su volumen es el de la celdad unidad dividido por el número de puntos que contiene, y esto es así porque el cristal se considera homogéneo (la densidad es constante)

Por ejemplo, como SC solo contiene 1 punto, ya es celda primitiva, de volumen $a^3$ ($a$ parámetro de red. Equiespaciado entre los puntos de la red directa)

En FCC sería $V_{primitivo} = \frac{1}{4}a^3$ y en BCC, $V_{primitivo}=\frac{1}{2}a^3$

**Estructuras de principal importancia:** los gases nobles, a temperatura suficientemente baja, forman sólidos cuasiperfectos, y como sus átomos constituyentes tienen el octeto completo, se aproximan muy bien a esferas rígidas. Entonces, es de suponer que se distribuyan como tales en el sólido, y adoptan 2 configuraciones posibles:

1. FCC
2. HCP (empaquetamiento hexagonal compacto): NO ES una celda unidad. Es una celda hexagonal con base doble (por tanto 2 átomos por celda) situados en $(0,0,0)$ y en $(2/3, 1/3, 1/2)$.

**Construcción por capas:** Si ponemos una capa de esferas y luego otra encima de estas en los huecos que forman, entonces podemos obtener: (1) red FCC si ponemos una tercera capa sobre donde estarían los huecos no ocupados de la primera capa o (2) HCP si ponemos una tercera capa directamente sobre donde estarían las esferas de la primera capa

## 1.2 Tratamiento y notación en la red directa

### Para identificar planos: índices de Miller

**Pasos para hallar los índices de Miller de un plano:**

1. Encuentro intersecciones del plano sobre los ejes de la celda unidad (primitiva o no)
2. Invierto coeficientes obtenidos
3. Multiplico por el menor número que los haga enteros
4. Estos son los índices de mi plano, $(hkl)$ (si quedara algún coeficiente negativo, señalizo su signo con una barra encima del número. e.g: $(\bar 231)$)

**OJO!:** como por construcción de índices de Miller cogemos la terna de enteros más baja, estamos definiendo infinitos planos (paralelos) con la misma terna. La distancia entre dos planos paralelos que comparten los índices de Miller depende de la celda de Bravais que tengamos:

1. Cúbica: $d_{hkl}^2=\frac{a^2}{h^2+k^2+l^2}$
2. Tetragonal: $d_{hkl}^2=\left[\frac{h^2+k^2}{a^2}+\frac{l^2}{c^2}\right]^{-1}$
3. Ortorrómbico: $d_{hkl}^2=\left[\left(\frac{h}{a}\right)^2+\left(\frac{k}{b}\right)^2+\left(\frac{l}{c}\right)^2\right]^{-1}$

### Para identificar direcciones (direcciones cristalográficas)

Denoto por $[uvw]$ una dirección (son las coordenadas del vector de red que señaliza esa dirección). Han de ser las coordenadas más bajas posibles, pero que sigan siendo números enteros

Denoto por $<uvw>$ todas las direcciones equivalentes a la dirección $[uvw]$ en un cristal. Equivalentes quiere decir que son geométricamente equivalentes. Por ejemplo, $<100>$ representa 6 direcciones equivalentes (semiejes X, Y y Z positivos y negativos)

## 1.3 Celda primitiva de Wigner-Seitz

Es un procedimiento general para encontrar una celda primitiva en cualquier tipo de red directa. Sin embargo, es mucho más relevante en la red recíproca, donde define la primera zona de Brillouin.

**Pasos:**

1. Me fijo en un punto de la red directa y lo uno con todos los puntos aledaños
2. Dibujo planos (rectas en 2D) en el punto medio de las líneas que unen mi punto con los demás
3. Selecciono el menor volumen (área en 2D) definido por estas rectas

# 2. Red recíproca

**Def.:** la red recíproca es el espacio dual de la red directa, módulo $2\pi$. Naturalmente, su base ($\{\vec b_i\}$) es la base dual de la red directa, de modo que se cumple:

$$\boxed{\vec a_i\cdot \vec b_j = 2\pi \delta_{ij}}$$

**Propiedades:**

1. El vector de la red recíproca con coordenadas $(h,k,l)$ es perpendicular a los planos de la red directa cn índices de Miller $(h,k,l)$

2. La distancia entre dos planos paralelos de índices de Miller $(hkl)$ en la red directa es inversamente proporcional al módulo del vector de la red recíproca con coordenadas $(h,k,l)$: $d_{hkl}=\frac{2\pi}{||\vec G_{hkl}||}$

## 2.3 Construyendo la base de la red recíproca

La base de la red recíproca se construye a partir de la base de la red directa de la forma:

$$\vec b_1 = 2\pi\frac{\vec a_2\times\vec a_3}{V}; \ \ \vec b_2 = 2\pi\frac{\vec a_3\times\vec a_1}{V}; \ \ \vec b_3 = 2\pi\frac{\vec a_1\times\vec a_2}{V}$$

con $V=[\vec a_1, \vec a_2, \vec a_3] = \vec a_1\cdot (\vec a_2\times \vec a_3)$ el volumen de la celda unidad de la red directa

**Nota:** si la celda unidad de la red directa es (1) cúbica, (2) tetragonal o (3) ortorrómbica, entonces la relación entre módulos de vectores de los vectores de la base de la red directa y la recíproca es muy sencilla (ejes son paralelos a los de la red directa):

$$b_i = \frac{2\pi}{a_i}$$

**Relación entre vectores de la red directa y la recíproca:**

Es evidente que se cumple:

$$\boxed{\vec R\cdot \vec G = 2\pi k; \ \ k\in \mathbb{Z}}$$

## 2.2 Cond. de difracción

Supondremos que las colisiones de nuestros proyectiles (fotones, electrones, etc.) contra las partículas que componen el sólido son elásticas (no se pierde energía), de modo que solamente se redireccionan dichos proyectiles. Si inicialmente incidían como una onda plana con vector de onda $\vec k$, entonces saldrán reflejados en la dirección de onda $\vec k'$, que cumple $||\vec k'||=||\vec k||$ por conservación de energía.

Habrá máximo de difracción si y solo si el vector $\Delta \vec k$ coincide con algún vector de la red recíproca:

$$\boxed{\Delta\vec k=\vec G} \Longleftrightarrow \text{Hay un máximo de difracción en dirección $\vec k'$}$$

**OJO!:** no hemos tenido en cuenta la acción de los átomos que componen la base. Estos podrían anular la difracción en algunas direcciones en las que, en principio, debería haberla. A esto se le llaman **extinciones**.

**Ley de Bragg**

Otra forma de dar cuenta de las direcciones donde habrá máximos de difracción es empleando la ley de Bragg. Si el haz (plano y armónico) indice con ángulo $\theta$ con respecto a la superficie y con longitud de onda $\lambda$, y los planos del cristal están equiespaciados una distancia $d$, entonces se cumple:

$$\boxed{2d\sin\theta=k\lambda; \ \ k\in \mathbb{Z}}$$

**Esfera de Ewald**

La tercera forma (esta es gráfica) de determinar en qué direcciones habrá máximos de difracción es coger la red recíproca, centrarnos en un punto de esta y trazar una esfera con radio $R=||\vec k||$. Todos los puntos que caigan sobre la superficie de esta esfera podrán unirse al central mediante un vector $\vec k'$ y por tanto cualesquiera dos puntos de la superficie cumplirán (1) pueden unirse mediante un vector de la red recíproca $\vec G = \vec k'-\vec k=\Delta\vec k$ y (2) tendrán el mismo módulo. Por tanto, habrá un máximo de difracción en la dirección $\vec k'$

**OJO!:** el ángulo que forman $\vec k$ y $\vec k'$ es $2\theta$, donde $\theta$ es el ángulo (con respecto a la superficie) con el que incide el haz entrante.

## 2.3 Factor de forma y dispersión. Extinciones

Aunque hasta ahora hemos tratado la difracción sin tener en cuenta los elementos dispersores (los átomos que componen el cristal), es importante admitir que son ellos los que provocan la difracción de las ondas incidentes.

**Def.:** el $\underline{\text{factor de forma / de scattering / atómico}}$ de un átomo / ion viene dado por:

$$\boxed{f=\int_{\mathbb{R}^3}\rho_{e}(\vec r)e^{i\Delta\vec k\cdot \vec r}d^3\vec r}$$

**Def.:** el $\underline{\text{factor de estructura}}$ de una celda unidad viene dada por la suma de los factores de forma de los átomos que contiene (ojo porque son los puntos de red que contiene la celda unidad, multiplicados por el número de átomos de la base. Por ejemplo, como el diamante tiene estructura FCC con base doble, habría que dar cuenta de 8 átomos en total). El factor de estructura de una celda unidad viene dado por:

$$\boxed{F = \displaystyle\sum_{\alpha=1}^Afe^{i\Delta \vec k\cdot \vec r_{\alpha}}}$$

donde:

- $A$: número total de elementos dispersores en una celda unidad
- $\Delta \vec k$ es la diferencia entre el vector de onda del haz saliente con el del entrante
- $\vec r_\alpha$ es la posición dentro de la celda unidad del elemento dispersor correspondiente

**OJO!:** los máximos de difracción se observarán cuando $\Delta \vec k=\vec G$ y habrá que sustituir

Una vez hayamos calculado $F$, podemos ver que se anula para algunas combinaciones de $h,k,l$. Entonces, en las direcciones inducidas por estas componentes (esto es, en $\vec k'=\vec k-\vec G_{hkl}$) no habrá el máximo de difracción que debería haber. Algunos ejemplos concretos siguen.

**Extinciones en FCC ($F_1$):** si los 3 índices $h,k,l$ no tienen todos la misma paridad, hay extinción

**Extinciones en BCC ($F_2$):** si $h+k+l$ impar, extinción. Máximo si $h+k+l$ par.

**Extinciones por base doble ($F_3$) :** si se cumple $h+k+l=4n+2$, hay extinción

las extinciones por diferentes motivos son multiplicativas. Entonces, el diamante, como tiene estructura FCC con base doble, cumplirá que su factor de estructura viene dado por:

$$F_{diamante}=fF_1F_3$$

donde $f$ es el factor de forma de los átomos de Carbono.

**Factor de estructura del sólido completo:** como el sólido, por definición, se construye trasladando la celda unidad en el espacio sin dejar huecos, entonces el factor de estructura del sólido vendrá dado por la suma de todos los factores de estructura de las celdas unidad que lo componen (modulado por la fase $\vec G\cdot \vec R$):

$$\boxed{F_T=\displaystyle\sum_{m,n,p=-\infty}^\infty Fe^{i\vec G_{hkl}\cdot \vec R_{mnp}}}$$

La intensidad cumple: $I\propto |F|^2$, de modo que el factor de estructura determina de manera unívoca los planos (hkl) de la red directa que dan lugar a máximos y mínimos de difracción

## 2.4 Zonas de Brillouin

**Def.:** la primera zona de Brillouin es la celda primitiva de Wigner-Seitz de la red recíproca. Es muy importante porque cualquier vector que parta del punto central de esta celda y llegue al borde de la misma cumple automáticamente la condición de difracción.

La dirección en la que el haz sale difractado es entonces $\vec k'=\vec G+\vec k$. Esto es, una vez tengamos los vectores $\vec k$ que parten del punto central de la primera zona de Brillouin y llegan a su frontera, cualquier vector $\vec k'$ construido sumándole al vector $\vec k$ cualquier vector de la red recíproca es dirección de máximo de difracción.

**Nota:** la segunda, tercera, etc. zonas de Brillouin se construyen como la primera, y el procedimiento de Wigner-Seitz es el mismo, pero en lugar de usar los primeros vecinos se cogen los segundos, terceros, etc.
