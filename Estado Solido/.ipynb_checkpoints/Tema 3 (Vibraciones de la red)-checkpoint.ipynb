- Hemos visto la energía de un cristal cuando todos los átomos están perfectamente estáticos, situados en los correspondientes puntos de la red
- Esto no es lo que sucede en realidad, sino que los átomos oscilan en torno a estas posiciones de equilibrio
- Además, los electrones también interaccionan entre sí, y eso no lo hemos tenido en cuenta
- Por último, los electrones interaccionan también con los iones, y esto tampoco lo hemos tenido en cuenta

# 0. Aproximación de Born-Oppenheimer

- Existe una diferencia tan grande entre la masa de los iones y la de los electrones que sus dinámicas están prácticamente desconectadas: los electrones, que se mueven rápido, se adaptan al potencial de los iones de forma instantánea. Los iones, que se mueven lento, ven una distribución de carga estática en el tiempo (promedio)

- Por tanto, el Hamiltoniano de todo el cristal tiene 3 contribuciones:

$$H = H_{ion-ion}+H_{e-e}+H_{ion-e}$$

- En este tema nos ocuparemos del primer término, que describe las interacciones entre iones (vibraciones de la red)

# 1. Relación de dispersión

**Def.:** en un medio por el que se propagan ondas, la relación de dispersión es la relación entre la frecuencia de una onda y su vector de onda: $\omega(\vec k)$

## 1.1 Modelo clásico de vibraciones y solución

- Todos los potenciales de red que hemos visto en el Tema 2 tienen un mínimo (posición de equilibrio)
- Los iones de la red no están fijos en los puntos de la red sino vibrando alrededor de ellos
- Podemos hacer aprox. armónica al potencial (equivale a hacer aprox. de muelles porque solo considera interacción a primeros vecinos por construcción)

**Notación:**
- $\vec R_i$: posición del punto de la red i-ésimo (pos. de equilibrio del ion i-ésimo)
- $\vec r_i$: posición del ion i-ésimo
- $\vec u_i=\vec r_i-\vec R_i$: desplazamiento del ion i-ésimo de su pos. de equilibrio

### 1.1.1 Modos normales

- **Def.:** un modo normal es un patrón de movimiento en el cual todos los componentes de un sistema oscilan cumpliendo una serie de propiedades

1. Todas las partículas oscilan con la misma frecuencia ("frecuencia normal del modo")
2. Amplitud cte. en el tiempo
3. Relación cte. entre sus fases

- Los modos normales son una base ortogonal de funciones de un sistema oscilante (cuando tengamos las ecuaciones del movimiento, podremos introducir los modos normales porque sabemos que las soluciones de nuestro sistema vibrante serán combinaciones lineales de modos normales, de modo que un modo normal también ha de ser solución)

### 1.1.2 Red unidimensional (monoatómica)

**Def.:** las condiciones de contorno de "Born-von Karman" son condiciones de contorno periódicas que garantizan la simetría de traslación en la red. Para una solución $u(\vec r)$, se cumple:

$$\vec u(\vec r+N_i\vec a_i)=u(\vec r)$$

- Si nuestro cristal es un sistema finito (pero con muchas partículas) de dimensiones $N_1a_1$ de largo, $N_2a_2$ de ancho y $N_3a_3$ de alto, entonces es como si tuviéramos infinitos cristales "stacked" con el nuestro, de modo que al avanzar $N_i\vec a_i$ nos iríamos al mismo punto del siguiente cristal "copia"

- Como hacemos aprox. armónica al potencial, tenemos entonces:

$$U_T \approx U_0+U_2$$

donde $U_0$ es la energía de cohesión (Tema 2) y $U_2$ es el término armónico. Hagamos Taylor en $U_T$:

$$U_T(\{x_i\})=U(R_i-u_i)\approx U_T(R_i)-\displaystyle\sum_{i}u_i\frac{\partial U}{\partial x_i}\Big |_{R_i}+\frac{1}{2}\displaystyle\sum_{i,j}u_i\frac{\partial^2U}{\partial x_i\partial x_j}u_j$$

- El término lineal se anula por definición de mínimo. Si consideramos solo interacción entre primeros vecinos, tenemos un sistema de N partículas conectadas por un muelle de constante $K\equiv \frac{\partial^2U}{\partial x^2}\Big |_{eq}$ (fíjate que es pos. de equilibrio de potencial entre 2 partículas nada más), de modo que el Hamiltoniano del sistema es de la forma

$$U\approx U_0+\frac{1}{2}K\displaystyle\sum_{n=0}^{N-1}\left[u_n-u_{n+1}\right]^2$$

Aplicamos ecs. de Hamilton y tenemos las ecuaciones de movimiento de cada partícula:

$$m\ddot u_n = K(u_{n+1}+u_{n-1}-2u_n)$$

Introducimos modos normales como ansnatz (sabemos que funcionan porque son una base de un sistema oscilante)

$$\boxed{u_n=Ae^{i(kna-\omega t)}}$$

Imponemos cond. de contorno de Born-von Karman y obtenemos que solo valen algunos valores de $k$:

$$u_{n+N}=u_n \Longrightarrow \boxed{k=\frac{2\pi l}{Na}; \ \ l\in \left\{-\frac{N}{2}, \ldots, \frac{N}{2}\right\}}$$


- Así obtenemos base de soluciones $u_{n_l}$. Tenemos $N$ partículas, de modo que hay $N$ modos normales (espacio de soluciones tiene dimensión $N$)

**Relación de dispersión:**

- Finalmente introducimos base de soluciones en las ecuaciones del movimiento y obtenemos relación entre $\omega$ y $k$:

$$\boxed{\omega(k)=2\sqrt{\frac{K}{m}}\left|\sin\left(\frac{ka}{2}\right)\right|; \ \ k\in\left\{-\frac{\pi}{a},\ldots, \frac{\pi}{a}\right\}}$$

**Fíjate:** $k$ son todos aquellos vectores de onda contenidos en la primera zona de Brillouin del cristal. La relación de dispersión es periódica en este intervalo de modo que toda la física se encuentra contenida en esta zona y podemos prescindir del resto. Esto se podía deducir evaluando directamente los límites de $k$ en su expresión correspondiente

**Nota:** si bien los valores de $k$ están discretizados, en la práctica hay tantos y tan juntos que tomamos $k$ como variable real

![DISPERSION](Rel_dispersion_1D.png)

- Las zonas rojas delimitan la primera zona de Brillouin de la red unidimensional

**Def.:** la velocidad de fase se define como:

$$v_f\equiv c = \frac{\omega}{k}=2\sqrt{\frac{K}{m}}\frac{\left|\sin\left(\frac{ka}{2}\right)\right|}{k}$$

**Def.:** la velocidad de grupo se define como:

$$v_g=\frac{d\omega}{dk}=a\sqrt{\frac{K}{m}}\left|\cos\left(\frac{ka}{2}\right)\right|$$

![VEL_GRUPO_1D](Vel_grupo_1D.png)

**Casos límite de la relación de dispersión:**

1. Rango de ondas acústicas: $|k| << \pi/a$ ($\lambda >>a$)

    - $\omega = a\sqrt{\frac{K}{m}}|k|$ (régimen lineal)
    - $c=v_g = a\sqrt{\frac{K}{m}}$

2. Frontera de la 1ZB: $|k|\to \frac{\pi}{a}$ ($\lambda \to 2a$)

    - $\omega=2\sqrt{\frac{K}{m}}$
    - $c=\frac{2a}{\pi}\sqrt{\frac{K}{m}}$
    - $v_g = 0$
    
**Nota:** la velocidad de grupo es 0 en la frontera de 1ZB porque las ondas de sonido se difractan con un ángulo de $\pi$ radianes, de modo que interfieren con las incidentes en la región y se crea un patrón de ondas estacionarias en la frontera de 1ZB. Se comprueba fácilmente con la ley de Bragg tomando distancia entre planos $d=a$ y $\lambda = 2a$

### 1.1.3 Red unidimensional (diatómica)

- Si tienes $p$ especies en la base, tendrás $D\cdot p$ ramas, donde $D$ es la dimensión del sistema. En este caso, entonces, obtendremos 2 ramas
- Sea ahora un cristal monoatómico de base doble. La constante elástica de los átomos / iones de tipo 1 con los de tipo 2 es $K$ y la de tipo 2 con el siguiente tipo 1 será $G$
- Por simplicidad, diremos que ambos tipos de partícula tienen la misma masa $m_1=m_2\equiv m$

- Los átomos de tipo 1 tendrán amplitud de oscilaciones $u_{n,1}$ y los de tipo 2 tendrán $u_{n,2}$. Entonces, tenemos ahora:

$$H_2 = T+\frac{1}{2}K\displaystyle\sum_n(u_{n,1}-u_{n,2})^2+\frac{1}{2}G\displaystyle\sum_n(u_{n+1,1}-u_{n,2})^2$$

- Podemos aplicar directamente la mecánica Newtoniana y obtenemos una ecuación del movimiento para cada tipo de partícula

$$\begin{cases}m\ddot u_{n,1}=-K(u_{n,1}-u_{n,2})-G(u_{n,1}-u_{n-1,2}) \\ m\ddot u_{n,2}=-K(u_{n,2}-u_{n,1})-G(u_{n,2}-u_{n+1,1}) \end{cases}$$

- Proponemos nuevamente soluciones de modos normales (uno por cada tipo de partícula)

$$u_{n,1}^l=A_1e^{i(k_lna-\omega t)}$$
$$u_{n,2}^l=A_2e^{i(k_lna-\omega t)}$$

con $k_l=\frac{2\pi l}{Na}$

- Introducimos estas soluciones en las ecuaciones del movimiento y obtenemos un sistema de ecuaciones lineales homogéneo ==> para evitar la solución trivial ($A_1=A_2=0$), el determinante debe ser nulo. Esto nos da la relación de dispersión:

$$\omega^2(k)=\frac{K+G}{2}\pm \frac{1}{m}\sqrt{K^2+G^2+2KG\cos(ka)}$$

- Obtenemos así dos ramas (fíjate que $\omega>0$ por construcción, de modo que no podemos obtener 4 por un doble +/-)

![REL_DISPERSION_BASEDOBLE_1D](Rel_dispersion_basedoble_1D.png)

**Nota:** fíjate en que hay una banda de frecuencias prohibidas!

**Rama óptica**

- La observamos cuando las partículas de una misma celda del sistema oscilan "más en antifase que en fase" (desfase > $\pi/2$)
- Puede no aparecer en las relaciones de dispersión si las partículas de una misma celda están muy ligadas entre sí (entonces actúan como un solo cuerpo, de modo que recuperamos el caso de base simple y solo tendríamos rama acústica)
- Si hay rama óptica debe haber rama acústica, pero lo contrario puede no suceder
- Se llama "rama óptica" porque los fonones de esta rama interaccionan con los fotones

**Rama acústica**

- La observamos cuando las partículas de una misma celda oscilan "más en fase que en antifase" (desfases < $\pi/2$)
- Se estudia con espectroscopía de Brillouin. Se irradia el cristal con ondas EM de gran $\lambda$ de modo que altera la densidad del sólido de forma periódica y este actúa como una red de difracción

**Nota:** rama determina desfase entre partículas de una misma celda. $k$ determina desfase entre celdas ($k\to 0$: en fase, $k\to\pi/a$: en antifase). La relación entre la amplitud de dos partículas de distinto tipo dentro de la misma celda viene dada por:

$$\frac{A_2}{A_1}=\mp\frac{k+Ge^{ika}}{|K+Ge^{ika}|}$$

- $A_2=-A_1$: rama óptica (antifase)
- $A_2=A_1$: rama acústica (en fase)

### 1.1.3 Red tridimensional (Bravais)

- Vamos a estudiar el mismo modelo pero en 3 dimensiones
- Al hacer aproximación armónica, tenemos ahora que el potencial es:

$$U_T\approx U_0+U_2=U_0+\frac{1}{2}\displaystyle\sum_{\vec R,\vec R'}\displaystyle\sum_{\mu,\nu=x,y,z}u_\mu(\vec R)D_{\mu\nu}(\vec R-\vec R')u_\nu(\vec R')$$

con $D_{\mu\nu}=\frac{\partial^2U}{\partial x_\mu\partial x_\nu}$

**Propiedades de D:**

1. Simétrica: $(D_{\mu\nu})=(D_{\nu\mu})$
2. $D(\vec R-\vec R')=D(\vec R'-\vec R)$
3. $\displaystyle\sum_{\vec R\neq 0}D(\vec R)=0$

**Ecuaciones del movimiento (base simple):**

$$m\ddot{\vec u}(\vec R)=-\displaystyle\sum_{\vec R'}D(\vec R-\vec R')\vec u(\vec R')$$

- Proponemos ansnatz de modos normales en 3 dimensiones:

$$\vec u(\vec R, t)=\vec Ae^{i(\vec k\cdot \vec R-\omega t)}$$

- Imponemos condiciones de Born-von Karmann y obtenemos una discretización de los vectores de onda posibles:

$$\boxed{\vec k=\frac{n_1}{N_1}\vec b_1+\frac{n_2}{N_2}\vec b_2+\frac{n_3}{N_3}\vec b_3}; \ \ n_i\in[-N_i/2,N_i/2]$$

**Simetría de traslación en red espacial ==> periodicidad de las soluciones en red recíproca**

- Introducimos el ansnatz en las soluciones para obtener la relación de dispersión, que queda como un problema de autovalores:

$$m\omega^2\vec A=\mathscr{D}(\vec k)\vec A$$

con $\mathscr{D}(\vec k)=\displaystyle\sum_{\vec R\neq 0}D(\vec R)e^{-i\vec k\cdot \vec R}$

- A $\mathscr{D}(\vec k)$ se le llama **matriz dinámica**, y $D$ es su transformada de Fourier. Aprovechando las propiedades de $D$, también podemos escribirla de la forma:

$$\mathscr{D}(\vec k)=-2\displaystyle\sum_{\vec R\neq 0}D(\vec R)\sin^2\left(\frac{\vec k\cdot \vec R}{2}\right)$$

**Propiedades de $\mathscr{D}(\vec k)$:**

1. Real
2. Simétrica
3. $\mathscr{D}(\vec k)=\mathscr{D}(-\vec k)$
4. Propiedad 3 implica que los autovalores también cumplirán propiedad 3 y por tanto $\omega(\vec k)=\omega(-\vec k)$

- Por ser $\mathscr{D}$ real y simétrica, existen 3 autovectores reales (**vectores de polarización**) que cumplen la ecuación de autovalores que nos da la relación de dispersión en 3 dimensiones:

$$\mathscr{D}(\vec k)\vec A_s=m\omega_s^2(\vec k)\vec A_s$$

donde $s=1,2,3$ es el número de la rama correspondiente (las denotaremos L1, T1, T2. Puede que alguna de ellas no aparezca, en cuyo caso diremos que la rama está degenerada porque solapa con otra)

**Nota:** los 3 vectores de polarización siempre serán perpendiculares entre sí (podemos hacer Gram-Schmidt si hubiera uno o más autovalores degenerados), de modo que conforman un triedro

$$\vec A_s\cdot \vec A_s'=\delta_{ss'}$$

- Uno de los autovectores irá en la dirección de propagación de la perturbación (1 modo longitudinal), y los otros dos serán perpendiculares a ella (2 modos transversales)

**Nota:** a baja $\vec k$, los medios tienen dispersión lineal, pero diferente en cada rama. Esto se comprueba haciendo Taylor en la matriz dinámica e introduciendo en la ecuación de autovalores, de donde queda:

$$\omega_s=c_sk$$

**Proposición:** en un medio anisótropo (cristal), si $\vec k$ está orientado paralelo a un eje de rotación de orden 3, 4 ó 6 entonces los dos modos transversales están degenerados

## 1.2 Modelo cuántico de vibraciones

- Sea un cristal tridimensional en aproximación armónica. Se compone entonces de 3N osciladores unidimensionales
- Cuantizamos las vibraciones de la red con una cuasipartícula que llamaremos "fonón"
- Es una cuasipartícula porque necesita de un medio material para existir, no puede ser por sí misma
- Es del todo análogo al fotón en el vacío, de modo que es un bosón y como tal le corresponde la estadística de Bose-Einstein
- La energía del nivel energético correspondiente a los fonones de la rama $s$ viajando en la dirección $\vec k$ será:

$$E_{\vec k, s}=\hbar\omega_s(\vec k)\left(n_{\vec k, s}+\frac{1}{2}\right)$$

donde $n_{\vec k, s}=\frac{1}{e^{\beta\hbar\omega_s(\vec k)}-1}$ es el número de ocupación de este nivel

- Alternativamente, también podemos usar el formalismo de segunda cuantización para crear cualquier estado de vibración aplicando el operador creación de los niveles necesarios al estado vacío

# 2. Propiedades térmicas de los cristales

- Queremos ahora predecir algunas propiedades. La más importante de todas será la capacidad calorífica de un cristal

## 2.1 Ley de Dulong-Petit

- En 1819, Dulong y Petit midieron la capacidad calorífica molar (a volumen constante) a temperatura ambiente de muchos sólidos y vieron que la gran mayoría de ellos tenía una curva constante:

$$\boxed{C_V/N=3N_Ak_B}$$

- Esto se deduce fácilmente con la colectividad canónica clásica, donde el sistema se ve como $3N$ osciladores armónicos clásicos y por el teorema de equipartición cada partícula tiene 6 grados de libertad, cada uno contribuyendo con $1/2k_BT$, de modo que $E=3Nk_BT$

- Cuando se tuvo acceso a temperaturas más bajas, se observó que la capacidad calorífica de los cristales se desvía drásticamente de su tendencia lineal y tiende a 0

## 2.2 Densidad de modos

- Todo el rato estamos trabajando con sumatorios sobre los vectores $\vec k$ de la primera zona de Brillouin del sistema
- Sin embargo, quisiéramos hacer un paso al continuo e integrar en frecuencias para así poder esconder la relación de dispersión

### 2.2.1 ¿Podemos pasar al continuo las sumatorias en $\vec k$?

- Según hemos visto:

$$\vec k=\frac{2\pi n_1}{L_1}\hat i+\frac{2\pi n_2}{L_2}\hat j+\frac{2\pi n_3}{L_3}\hat k; \ \ n_i\in[-N_i/2,N_i/2]$$

donde $N_i$ es el número de celdas que tiene un cristal en un la dimensión $i$

- El volumen del espacio recíproco que ocupa cada vector $\vec k$ será entonces:

$$\delta^3\vec k=\frac{(2\pi)^3}{V}$$

- Por otro lado, el volumen que ocupa toda la 1ZB será:

$$\Omega_{1ZB}=N\delta^3\vec k=\frac{(2\pi)^3}{V}N$$

con $N=N_1N_2N_3$, pero $V/N$ es el volumen de una celda de la red directa ($a_1a_2a_3$), de modo que:

$$\Omega_{1ZB}=\frac{(2\pi)^3}{a_1a_2a_3}>>\delta^3\vec k$$

- De este modo, podemos imaginarnos la 1ZB como una malla discretizada con celdas tan pequeñas que no se comete un error significativo al pasar al continuo
- Haciendo $\frac{1}{V}=\frac{\delta^3\vec k}{8\pi^3}$, podemos pasar de sumatorios en $\vec k$ a integrales, de la forma:

$$\displaystyle\sum_s\displaystyle\sum_{\vec k}\frac{1}{V}f(\vec k, s)\to \displaystyle\sum_s\int_{1ZB}\frac{\delta^3\vec k}{8\pi^3}f(\vec k, s)$$

### 2.2.2 Pasando a frecuencias

- Ya hemos pasado al continuo. Ahora queremos trabajar con frecuencias, de modo que necesitamos definir una densidad de modos
- **Def.:** se define la densidad de modos $g(\omega)$ como el número de modos con frecuencia contenida entre $\omega$ y $\omega+d\omega$ dividida por el volumen del sólido

$$g(\omega):=\frac{N(\omega)}{V}$$



- **Def.:** una "superficie equifrecuencial" es aquella superficie de la primera zona de Brillouin que da lugar a una misma frecuencia. No es más que la representación geométrica de la relación de dispersión para una frecuencia dada: $\omega(\vec k)=\omega_0$
- En el modelo de Debye, las superficies equifrecuenciales son esferas:

$$\omega=ck\Longrightarrow k_x^2+k_y^2+k_z^2=\left(\frac{\omega}{c}\right)^2$$

- La densidad de modos tiene la forma integral:

$$\boxed{g(\omega)=\frac{1}{(2\pi)^3}\int_S\frac{dS}{||\vec v_g||}; \ \ \vec v_g = \vec \nabla_k\omega(\vec k)}$$

## 2.1 Modelo de Einstein

- Considérense los $3N$ osciladores cuánticos en lugar de clásicos
- **Suposición clave:** los $3N$ osciladores tienen la misma frecuencia $\omega_E$
- Recupera bien ley de Dulong-Petit para altas temperaturas, y también tiende a 0. Funciona bien pero tiende a 0 de forma exponencial
- Un año más tarde, Kammerlingh Onnes llegaró a temperaturas más bajas, donde se vio que la tendencia de $C_V$ hacia 0 no era exponencial sino cúbica --> el modelo de Eisntein estaba mal
- **Nota:** el modelo de Einstein aún sirve para la rama óptica, porque su relación de dispersión es una recta cte. en $\omega_E$, que se asemeja bastante a la curva de esta rama

- Su densidad de modos será $g(\omega)\propto \delta(\omega-\omega_E)$

## 2.2 Modelo de Debye

- **Suposición clave:** la relación de dispersión de los sólidos es lineal $\boxed{\omega=ck}$ y válida para todas las ramas (todas tienen la misma c)
- Sirve bien para la rama acústica con $k$ pequeño (fíjate en el dibujo)
