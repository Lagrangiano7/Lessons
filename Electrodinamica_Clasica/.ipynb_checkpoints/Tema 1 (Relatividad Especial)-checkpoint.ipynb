**Objetivos:**

1. Entender postulados de la relatividad especial
2. Formular relatividad especial geométricamente
3. Entender el grupo de Poincaré (el grupo de isometrías de la relatividad especial)
4. Formular una dinámica relativista de la partícula

# 0. Definiciones preliminares

**Def.:** un "sistema de referencia" es un sistema de coordenadas espaciales + un reloj, de modo que podemos señalar un evento que ha sucedido en un instante $t$ según nuestro reloj y en la posición $\vec x$ según nuestro sistema de coordenadas con la 4-tupla: $(t, \vec x)$

**Def.:** un "sistema de referencia **inercial**" es aquel sistema de referencia en el que las partículas libres (que no están sometidas a ninguna fuerza en el sentido clásico) se mueven en línea recta

# 1. Postulados

1. **Principio de relatividad:** en ausencia de fuerzas gravitatorias, todas las leyes de la Física son las mismas en todo sistema de referencia inercial
2. **Universalidad de $c$:** la velocidad de la luz en el vacío es constante e independiente del estado de movimiento del emisor

**Interpretación de (1):** cuando formulemos una acción para nuestro sistema, esta deberá mantener su forma bajo aquellas transformaciones que nos lleven de un inercial a otro (no tiene por qué valer lo mismo, pero como las leyes de la Física son las mismas en todo inercial, deben tener la misma forma). Esto también querrá decir que es de esperar que las ecuaciones de Euler-Lagrange que saquemos sean covariantes

**Interpretación de (2):** la Naturaleza obedece el principio de causalidad: el futuro no puede afectar al pasado, porque la velocidad de propagación de cualquier señal está acotada superiormente (y obviamente inferiormente por ser positiva, sino tendríamos más de lo mismo, la señal viajaría al pasado)

# 2. Formulación geométrica

## Intervalo espaciotemporal es invariante: consecuencia de postulado 2

- Consideremos una consecuencia directa de la universalidad de la velocidad de la luz. Nos situamos en un inercial cualquiera, $S$
- Sea un haz de luz que se emite en el instante $t_1$ en la posición $\vec x_1$ y que llega a un receptor situado en $\vec x_2$ en el instante $t_2$
- Como la velocidad de la luz es constante, de acuerdo a nuestro observador el haz ha recorrido una distancia $|\vec x_2-\vec x_1|$ en un tiempo $t_2-t_1$ a una velocidad $c$, es decir:

$$|\vec x_2-\vec x_1|=c(t_2-t_1)$$

Elevando al cuadrado y pasándolo todo al mismo lado, tenemos una cantidad que se anula:

$$-c^2(t_2-t_1)^2+|\vec x_2-\vec x_1|^2=0$$

En otro sistema de referencia también inercial, en general expresaremos las posiciones del emisor y el receptor en otras coordenadas, y nuestro reloj medirá otros tiempos, pero como la velocidad de la luz es universal, tendremos que se sigue cumpliendo la misma igualdad:

$$-c^2(t_2'-t_1')^2+|\vec x_2'-\vec x_1'|^2=0$$

**Proposición:** en particular, para cualesquiera dos eventos $(t_1, \vec x_1), (t_2, \vec x_2)$ descritos en el mismo inercial, no necesariamente conectados por un haz de luz, se cumple que esta magnitud es **invariante** (es la misma en cualquier inercial), pero no tiene por qué ser 0

## El cono de Minkowski

- Tenemos ahora un invariante obtenido como consecuencia directa del segundo postulado, que es un hecho experimental, de modo que para cualesquiera dos eventos registrados desde un mismo inercial:

$$\Delta s^2=-c^2(\Delta t)^2+(\Delta \vec x)^2$$

donde $\Delta s^2$ recibe el nombre de **intervalo espaciotemporal**. Si en particular los dos eventos están conectados a través de un haz de luz, entonces $\Delta s^2=0$. Si solo tomamos 2 variables espaciales, podemos representar esta relación con el tiempo en el eje $z$, y queda: $|\Delta t|= \sqrt{\Delta x^2+\Delta y^2}$, definiendo un "cono de luz" o "cono de Minkowski":

![Cono_de_luz.png](Cono_de_luz.png)

De forma más intuitiva, si tomamos el evento $(t_1, \vec x_1)$ como referencia, el cono de luz representa las áreas del espacio que se vuelven accesibles para la señal emitida en dicho instante y en dicha posición conforme va pasando el tiempo. Lógicamente, lo máximo a lo que aspira es a desbloquear las regiones nuevas del espacio al mismo ritmo que una señal luminosa.

Esto permite clasificar de forma natural cualesquiera dos eventos vistos en un mismo inercial a partir del signo de su intervalo espaciotemporal (pero la clasificación es válida para todo inercial porque el intervalo espaciotemporal es invariante Lorentz):

1. $\Delta s^2 < 0$: intervalo de **género tiempo** / eventos **separados temporalmente**: existe inercial en el que suceden en el mismo lugar pero en distintos instantes
2. $\Delta s^2 > 0$: intervalo de **género espacio** / eventos **separados espacialmente**: existe inercial en el que suceden en el mismo tiempo pero en distinto lugar: son simultaneos. Un intervalo de tipo espacio es **acausal** porque necesitaríamos una señal con velocidad superior a la de la luz para conectarlos en el intervalo de tiempo dado
3. $\Delta s^2 = 0$: intervalo de **género luz** / eventos **separados por señal luminosa**

## Elemento de línea, variedad de Minkowski y tensor métrico

- **Def.:** el espaciotiempo es una variedad diferenciable Lorentziana de 4 dimensiones, cuyas variables son el tiempo y las tres dimensiones del espacio.

- **Def.:** un "evento" en el espaciotiempo es un punto sobre la variedad, tal que $p\equiv (t, \vec x)\in M$

- Podemos hacer el intervalo espaciotemporal tan pequeño como nosotros queramos. Esto es, como admitimos que el espaciotiempo es un continuo, podemos considerar eventos arbitrariamente cercanos en espacio y en tiempo. Entonces, tenemos el **elemento de línea:**

$$ds^2=-dt^2+d\vec x^2$$

- Esto nos induce automáticamente una métrica en la variedad:

$$\eta_{\mu\nu}=(-+++)$$

tal que:

$$ds^2=\eta_{\mu\nu}dx^\mu dx^\nu$$

**Proposición:** las componentes de la métrica son universales (las mismas en todos los inerciales, de modo que $\eta_{\mu\nu}'=\eta_{\mu\nu}$)

**Dem.:** 

Sabemos que el elemento de línea es invariante, y que de él podemos leer la métrica. Por ende, en un inercial $S$ tendremos $ds^2=-dt^2+d\vec x^2\Longrightarrow \eta_{\mu\nu}=(-+++)$. En otro inercial $S'$ tendremos $ds^2=-dt'^2+d\vec x'^2\Longrightarrow \eta'_{\mu\nu}=(-+++)_\blacksquare$

# 3. El grupo de Poincaré

- Queremos ahora estudiar qué transformaciones dejan invariante el elemento de línea, porque estas serán las simetrías de nuestro espaciotiempo
- Queremos transformaciones tales que:

$$(ds')^2=ds^2$$

## Traslaciones espaciotemporales

$$x^\mu \to x'^\mu = x^\mu + \alpha^\mu$$

## Transformaciones de Lorentz propias ortocronas ($det\Lambda=1$)

**Def.:** una transformación propia es aquella que no invierte la orientación del espacio

**Def.:** una transformación ortocrona es aquella que no invierte la dirección del tiempo ($\Lambda^0_0\geq 1$)

**Def.:** el grupo de Lorentz (propio ortocrono) es el subgrupo del grupo de Poincaré compuesto por rotaciones y boosts de Lorentz

$$x^\mu\to x'^\mu = \Lambda^\mu_\nu x^\nu$$

Vemos que en general mezcla las componentes del espaciotiempo entre sí. Veremos que se pueden expresar como rotaciones hiperbólicas

**Proposición:** el tensor métrico cumple la relación: $\boxed{\eta_{\alpha\beta}=\Lambda^\mu_\alpha\Lambda^\nu_\beta\eta_{\mu\nu}}$

O bien en forma matricial:

$$\eta = \Lambda^T\eta\Lambda$$

**Nota:** podemos llegar a esta expresión por dos caminos: admitiendo a priori que el tensor métrico es efectivamente un tensor dos veces covariante y escribiendo su ley de transformación, para después admitir que como es universal le puedo quitar la prima. También puedo hacerlo a partir del elemento de línea, y esto me da que el tensor métrico es efectivamente un tensor dos veces covariante. Le quito la prima y me vuelve a dar esto.

**Interpretación:** las transformaciones de Lorentz son aquellas que dejan invariante la métrica

**Dem.:** 

Imponemos que las transformaciones que hagamos dejen invariante el elemento de línea. Las traslaciones lo cumplían inmediatamente. Veamos qué tienen que cumplir las transformaciones $\Lambda$ para que esto sea cierto:

$$ds'^2= dx'_\mu dx'^\mu=\eta_{\mu\nu}dx'^\nu dx'^\mu$$

donde hemos usado que la métrica es universal. Expresemos este elemento de línea ahora en términos de las variables viejas:

$$ds'^2=\eta_{\mu\nu}\Lambda^\nu_\alpha dx^\alpha \Lambda^\beta_\mu dx^\beta$$

Pero imponemos que el elemento de línea es invariante bajo estas transformaciones, así que:

$$ds'^2=ds^2 \Longrightarrow \eta_{\mu\nu}\Lambda^\nu_\alpha  \Lambda^\mu_\beta dx^\alpha dx^\beta=\eta_{\alpha\beta}dx^\alpha dx^\beta$$

**Conclusión:** $\eta_{\alpha\beta}= \Lambda^\nu_\alpha  \Lambda^\mu_\beta\eta_{\mu\nu\blacksquare}$

**Propiedad:** $\boxed{\Lambda^\mu_{\ \ \ \nu}\Lambda_{\mu}^{\ \ \ \nu}=\delta^\mu_\nu}$, donde al parecer son objetos diferentes porque el primero transforma vectores y el segundo covectores. Yo creo que da un poco igual porque se entiende por el contexto, mirando qué índice se está contrayendo. Si lo necesitas, el inverso de uno se obtiene a partir del otro subiendo y bajando índices:

$$\Lambda_\mu^{\ \ \ \nu}=\eta_{\mu\sigma}\eta^{\nu\rho}\Lambda^\sigma_{\ \ \ \rho}$$

### Rotaciones espaciales

- Las rotaciones espaciales actúan como:

$$t'=t$$
$$\vec x'=R\vec x$$

Por ende, sus matrices son de la forma:

$$\boxed{\Lambda = \left(
\begin{array}{c|ccc}
1 & 0 & 0 & 0 \\ \hline
0 &   &   &   \\
0 &   & R &   \\
0 &   &   &  
\end{array}
\right)}$$

tal que:

$$\begin{pmatrix}t'\\ \vec x'\end{pmatrix}= \left(
\begin{array}{c|ccc}
1 & 0 & 0 & 0 \\ \hline
0 &   &   &   \\
0 &   & R &   \\
0 &   &   &  
\end{array}
\right)\begin{pmatrix}t\\ \vec x\end{pmatrix}$$

### Boosts de Lorentz

- Sean dos inerciales $S, S'$ tales que tienen la misma orientación en el espacio, no están trasladados, pero que no viajan a la misma velocidad
- Sea $\vec v$ la velocidad de $S'$ que mide un observador en $S$. Entonces, un observador en $S'$ medirá el espacio y el tiempo de forma distinta que uno en $S$, pero existe una relación entre ellas:

$$t'=\gamma\left(t-\frac{\vec v\cdot \vec x}{c^2}\right)$$
$$\vec x'= \vec x + (\gamma-1)(\vec v\cdot \vec x)\frac{\vec v}{v^2}-\gamma\vec vt$$

**Importante / Útil:** esto es cómo se transforman las coordenadas bajo un boost. Como los cuadrivectores son aquellos objetos que se transforman igual que las coordenadas, esta expresión sirve para cualquier cuadrivector (por ejemplo, para el cuadrimomento)

- En forma matricial, tenemos que:

$$\boxed{\begin{pmatrix}ct'\\ x_i'\end{pmatrix} = \begin{pmatrix}\gamma & -\gamma\beta_i \\ -\gamma\beta_i & \delta + (\gamma-1)\frac{\beta_i^2}{\beta^2}\end{pmatrix} \begin{pmatrix}ct\\ x_i\end{pmatrix}}$$

donde:

- $\beta_i=v_i/c$
- $\delta$ es 1 si estamos en la misma coordenada (x' con x, etc.) y 0 si no

### Transformaciones de Lorentz como rotaciones hiperbólicas

- Sea un inercial $S$ y otro inercial $S'$ que se aleja de él con una velocidad $\vec v$ (en línea recta). Tomemos el eje X en la dirección de esta velocidad, de modo que las coordenadas perpendiculares no se transforman: $y'=y; z'=z$. Tenemos entonces que la relación entre lo que miden observadores en estos dos inerciales es la transformación de Lorentz:

$$\begin{pmatrix}ct'\\ x'\end{pmatrix} = \begin{pmatrix}\gamma & -\gamma\beta \\ -\gamma\beta & \gamma\end{pmatrix} \begin{pmatrix}ct\\ x\end{pmatrix}$$

- Recordemos que $\gamma\in [1, \infty)$ y que $\beta = v/c\in [0, 1)$. Por ende, podemos asociarles funciones con los mismos recorridos. Hacemos entonces la identificación:

$$\gamma=\cosh(\varphi)$$
$$\beta = \tanh(\varphi)$$

donde el parámetro $\varphi$ se llama **rapidity**. La transformación de Lorentz pura es entonces una rotación hiperbólica en el espaciotiempo. Este planteamiento permite resolver problemas donde un sistema de referencia está acelerado con respecto del otro de forma sencilla, porque la rapidity se comporta linealmente con la aceleración

## Transformaciones de Lorentz infinitesimales (repr. vectorial)

- Deseamos hallar los generadores del grupo de Lorentz en la representación vectorial del grupo (esto es, cogemos aquella representación del grupo de Lorentz que a cada elemento del grupo le asocie un endomorfismo regular que actúe sobre los vectores de la variedad)
- Consideramos la transformación de Lorentz infinitesimal:

$$\Lambda = I+\delta\omega \longrightarrow \Lambda^\mu_\nu = \delta^\mu_\nu+\delta \omega^\mu_\nu$$

**Proposición:** el objeto $\delta \omega_{\mu\nu}=\eta_{\mu\rho}\delta \omega^\rho_\nu$ es antisimétrico

**Dem.:**

Introduciendo en la condición definitoria de las matrices de Lorentz (aquellas que dejan invariante la métrica): $\eta_{\alpha\beta}=\Lambda^\mu_\alpha\Lambda^\nu_\beta \eta_{\mu\nu}$. Sustituyendo y quitando el término $\mathcal O(\delta\omega^2)$, porque estamos a primer orden, sale que este objeto es antisimétrico$_\blacksquare$

**Proposición:** una transformación infinitesimal se puede expresar en términos de los generadores del grupo de Lorentz, $S_{\rho\sigma}$, como:

$$\boxed{\delta v^\mu = \frac{i}{2}\delta\omega^{\rho\sigma}(S_{\rho\sigma})^\mu_\nu v^\nu}$$

donde $(S_{\rho_\sigma})^\mu_\nu = -2i\delta^\mu_{[\rho} \ \eta_{\sigma]\nu}$

**Dem.:**

Los elementos del grupo de Lorentz se pueden escribir en términos de sus generadores como:

$$\Lambda = e^{\frac{i}{2} \omega^{\rho\sigma}S_{\rho\sigma}}$$

Si la transformación es pequeña, entonces podemos desarrollar a primer orden:

$$\Lambda \approx I+\frac{i}{2}\delta\omega^{\rho\sigma}S_{\rho\sigma}$$

Esto son matrices, porque los elementos del grupo de Lorentz son matrices 4x4. Si tomamos componentes, tenemos $\Lambda^\mu_\nu$. Basta entonces con tomar $\delta v^\mu = v'^\mu-v^\mu = \frac{i}{2}\delta^{\rho\sigma}(S_{\rho\sigma})^\mu_\nu v^\nu_\blacksquare$

La expresión concreta de los generadores viene de imponer que los elementos del grupo de Lorentz generados por estos $S_{\rho\sigma}$ sigan cumpliendo su propiedad definitoria

### Generadores como trivectores

- Vemos ahora que los generadores del grupo de Lorentz (que son 6) se pueden separar de forma clara en 3 de rotaciones y 3 de boosts de Lorentz
- Partimos de la expresión general (intrínseca):

$$\delta \mathbf{v} = \frac{i}{2}\delta\omega^{\rho\sigma}\hat S_{\rho\sigma}\mathbf v$$

donde la negrita indica que se trata de un campo vectorial (sin especificar base, por eso no trabajo en componentes) y el gorrito de $\hat S$ es para recalcar que es una matriz.

- Tenemos que la contracción de dos objetos antisimétricos da uno simétrico, así que:

$$\delta \omega^{\rho\sigma}S_{\rho\sigma}=2\displaystyle\sum_{\sigma < \rho} \delta\omega^{\rho\sigma}S_{\rho\sigma}$$

$$=2(\delta\omega^{10}S_{10} + \delta\omega^{20}S_{20} + \delta\omega^{30}S_{30}+\delta\omega^{21}S_{21}+\delta\omega^{31}S_{31}+\delta\omega^{32}S_{32})$$

$$=2(\delta\omega^{i0}S_{i0}+\frac{1}{2}\delta \omega^{ij}S_{ij})$$

Introduciendo en la expresión, tenemos:

$$\delta \mathbf{v} = i(\delta\omega^{i0}\hat S_{i0}+ \frac{1}{2}\delta\omega^{ij}\hat S_{ij})\mathbf v$$

Haciendo los cambios de variable: $\delta\xi^i=-\delta \omega^{i0}$, $-\delta \omega^{ij}=\epsilon^{ijk}\delta\theta_k$, tenemos:

$$\delta \mathbf{v} = i(\delta\xi^i\hat S_{i0}+ \frac{1}{2}\epsilon^{ijk}\delta\theta_k\hat S_{ij})\mathbf v$$

Finalmente, definiendo:

$$\boxed{\hat K_i=\hat S_{i0}; \ \ \ \hat S^i = \epsilon^{ijk} \hat S_{jk}}$$

Tenemos que los vectores de la variedad se transforman bajo transformaciones de Lorentz infinitesimales como:

$$\boxed{\delta \mathbf v = -i(\delta\vec \xi\cdot \vec{\hat K}+\delta \vec \theta\cdot \vec{\hat S})\mathbf v}$$

donde estos trivectores, que son los generadores del grupo de Lorentz expresados de forma alternativa, cumplen las relaciones de conmutación:

$$\boxed{\begin{gather}[\hat S_i, \hat S_j]=i\epsilon_{ijk}\hat S^k \\ [\hat S_i, \hat K_j]=i\epsilon_{ijk}\hat K^k \\ [\hat K_i, \hat K_j]=-i\epsilon_{ijk}\hat S^k\end{gather}}$$

**Consecuencias:**

1. Los generadores de las rotaciones, $\vec{\hat S}$, son momentos angulares
2. Los generadores de los boosts de Lorentz, $\vec{\hat K}$, son vectores (espaciales, en sentido tridimensional)
3. El resultado de aplicar dos boosts de Lorentz en un orden dado y de aplicarlos en el orden inverso difiere en una rotación en el espacio

### Generalización a representaciones tensoriales

- Se puede demostrar que es posible generalizar esta separación de los generadores del grupo de Lorentz en dos trivectores para cualquier representación de dicho grupo. La expresión intrínseca es cierta para cualquier tensor, independientemente de su tipo:

$$\delta \mathbf{T} = \frac{i}{2}\delta\omega^{\rho\sigma}\hat S_{\rho\sigma}\mathbf T$$

De modo que también es cierto que:

$$\boxed{\delta \mathbf T = -i(\delta\vec \xi\cdot \vec{\hat K}+\delta \vec \theta\cdot \vec{\hat S})\mathbf T}$$

Si lo queremos en componentes, podemos servirnos de la útil propiedad:

$$\boxed{(S_{\rho\sigma})^{a\ldots z}_{a'\ldots z'}=(S_{\rho\sigma})^a_{a'}\delta^{b}_{b'}\ldots \delta^z_{z'} + \delta^a_{a'}(S_{\rho\sigma})^b_{b'}\ldots \delta^z_{z'}+\delta^a_{a'}\ldots (S_{\rho\sigma})^z_{z'}}$$

Por ejemplo, para un tensor dos veces contravariante:

$$(\hat ST)^{\mu\nu}=(S_{\rho\sigma})^\mu_\alpha T^{\alpha\nu} + (S_{\rho\sigma})^\nu_\alpha T^{\mu\alpha}$$

### Nota sobre vectores en la variedad

**Def.:** llamaremos "vector" a todo aquel objeto que se transforme con la representación fundamental del grupo de Lorentz. Esto se generaliza para tensores n-veces contravariantes y m-veces covariantes (no quiero decir que estos sean vectores, son comb. lineal de productos tensoriales de varios vectores y covectores)

**Género de un vector:** sea un vector cualquiera $v^\mu$, diremos que es de género tiempo, espacio o luz con el mismo criterio que con el intervalo espaciotemporal, ahora aplicado a su magnitud al cuadrado: $v^2=v_\mu v^\mu$. Podemos ver que el intervalo espaciotemporal no es más que $\Delta s^2 = x_\mu x^\mu$, pero esto no quiere decir que la posición sea un vector contravariante. Más bien, los vectores contravariantes son aquellos que se transforman como las coordenadas (la posición)

# 4. Geometría diferencial en la variedad

## Dual de Hodge

**Observación:** en el grupo de Lorentz, los pseudotensores de Levi-Civita son tensores, porque $det\Lambda=1$

**Def.:** dado un tensor antisimétrico, se define su dual de Hodge como el pseudotensor (en el grupo de Lorentz, tensor) resultante de contraer todos sus índices disponibles con el pseudotensor de Levi-Civita: $\epsilon_{\mu\nu\rho\sigma}$, pesado con $1/n!$ donde $n$ es el número de índices:

$$Q\to ^*Q_{\mu\nu\rho\sigma}:=\frac{1}{0!}\epsilon_{\mu\nu\rho\sigma}Q$$
$$R^\sigma \to ^* R_{\mu\nu\rho}=\frac{1}{1!}R^\sigma\epsilon_{\mu\nu\rho\sigma}$$

**Propiedad:** como su nombre indica, esta operación es dual: el dual de Hodge de un dual de Hodge da el tensor original (tal vez con un signo "-"):

$$^{**}T=(-1)^{1+n(4-n)}T$$

**Útil: contracciones de los Levi-Civita**

1. $\epsilon^{\mu\nu\rho\sigma}\epsilon_{\mu\nu\rho\sigma}=-4!$
2. $\epsilon^{\mu\nu\rho\sigma}\epsilon_{\alpha\nu\rho\sigma}=-3!1!\delta^\mu_\alpha$
3. $\epsilon^{\mu\nu\rho\sigma}\epsilon_{\alpha\beta\rho\sigma}=-2!2!\delta^\mu_{[\alpha}\delta^\nu_{\beta]}$
4. $\epsilon^{\mu\nu\rho\sigma}\epsilon_{\alpha\beta\gamma\sigma}=-1!3!\delta^\mu_{[\alpha}\delta^\nu_\beta \delta^\rho_{\gamma]}$

Regla mnemotécnica:

1. Traer adelante todos los índices que no se van a contraer
2. Poner "-" y el factorial de todos los índices que sí se contraen multiplicado por el factorial de todos los que no se contraen
3. Poner delta de las variables que no se contraen, totalmente antisimetrizada en los índices inferiores

### Generalización de Teoremas de Gauss y de Stokes

- En cursos de cálculo, se estudia el teorema de Gauss en $\mathbb R^3$, que es una variedad diferenciable. Sin embargo, también podemos generalizarlo a otras variedades, incluso si tienen dimensión distinta. Vemos que en dimensiones superiores, el teorema de Gauss y el de Stokes dicen lo mismo: toda la información contenida en el interior de una región se encuentra asimismo en su frontera
- Sea $\Omega\subset M$ una región de una cierta variedad diferenciable, y sea $T^\mu$ un campo vectorial sobre dicha variedad. Entonces, tenemos que:

$$\boxed{\int_\Omega  (\partial_\mu T^\mu)d^4x=\int_{\partial\Omega}T^\mu d\sigma_\mu}$$

donde $\sigma_\mu$ es el vector perpendicular a la región

- Sea ahora una región tridimensional $\Sigma\subset M$ y $T^{\mu\nu}$ un campo tensorial **antisimétrico**, entonces:

$$\boxed{\int_\Sigma(\partial_\nu T^{\mu\nu})d\sigma_\mu = \frac{1}{2}\int_{\partial\Sigma}T^{\mu\nu}dS_{\mu\nu}}$$

- Sea ahora una región bidimensional $S\subset M$ y $T^\mu$ un campo vectorial, entonces:

$$\boxed{\int_S  (\epsilon^{\mu\nu\rho\sigma}\partial_\rho T_\sigma)dS_{\mu\nu}=2\int_{\partial S}T_\mu dx^\mu}$$

# 5. Dinámica de la partícula

## Cinemática

### Sistemas de referencia propios y comóviles

**Def.:** dada una partícula en movimiento, en general acelerada, un "sistema de referencia propio" de la partícula será todo aquel que se mueva solidario a ella para todo instante de tiempo (i.e: "nos sentamos sobre la partícula"). Dos propios solo pueden diferir en una rotación y/o una traslación

**Def.:** dada una partícula en movimiento, en general acelerada, un "inercial comóvil" es todo aquel sistema de referencia **inercial** que, para un instante concreto, se mueve solidario a la partícula (esto es, se mueve con su velocidad instantánea). Lógicamente, al "avanzar un instante $dt$", este inercial dejará de ser comóvil porque la velocidad de la partícula habrá cambiado

### Tiempo propio

**Def.:** el "tiempo propio" ($\tau$) de una partícula es el tiempo que mide un observador en el sistema de referencia propio de esta. **El tiempo propio es un invariante relativista**

**Dem.:**

El elemento de línea en el sistema de referencia propio será:

$$ds^2=-d\tau^2+d\vec x^2$$

pero como estamos situados sobre la partícula, esta se encuentra en el origen (o a una distancia fija) del sistema de referencia propio, de modo que tenemos $d\vec x= 0$. Por ende:

$$ds^2=-d\tau^2$$

Y tenemos que el tiempo propio es un invariante relativista$_\blacksquare$

**Nota:** esto ya se podía haber previsto, porque es el tiempo medido en un sistema de referencia concreto. De este modo, si me voy a otro sistema de referencia, cambia cómo mido el tiempo yo, pero cómo lo mide el propio no, de modo que es lógico que sea un invariante.

**Proposición:** el tiempo medido en el sistema de referencia propio coincide con el tiempo medido en los comóviles. Esto se debe a que, en cada instante de tiempo, son inerciales en reposo uno con respecto al otro y por ende miden el mismo tiempo. 

**Nota:** Véase sin embargo que hay un matiz conceptual: para cada instante, tenemos que cambiar de comóvil, de modo que realmente son muchos tiempos medidos desde un sistema de referencia inercial distinto cada uno

### Cuadrivelocidad

**Def.:** sea una trayectoria en el espaciotiempo parametrizado con el tiempo propio de la partícula, $\gamma(\tau)\equiv x^\mu(\tau)$, entonces se define su cuadrivelocidad como:

$$\boxed{u^\mu(\tau) = \frac{dx^\mu}{d\tau}}$$

**Proposición:** la cuadrivelocidad es un tensor (concretamente un vector contravariante)

**Dem.:**

Esto se debe a que el tiempo propio es un invariante Lorentz, de modo que la derivada con respecto a este conmuta con las transformaciones de Lorentz y, como lo que tenemos en el argumento de la derivada son las coordenadas, la cuadrivelocidad se transforma como ellas y es por ende un vector contravariante$\blacksquare$

**OJO!:** si parametrizásemos con cualquier otro tiempo, este no sería invariante relativista, de modo que no sería un tensor porque no se transformaría bien

- **Nota:** que parametricemos la trayectoria de la partícula con su tiempo propio no quiere decir que solo podamos medir la cuadrivelocidad en el sistemma de referencia propio de esta. De hecho, esto tendría poco sentido, porque la gracia de que sea un tensor es que podamos conocer su valor en cualquier sistema de referencia (inercial, aunque se puede generalizar haciendo que las velocidades no sean constantes). En particular, para un inercial arbitrario $S$, tenemos que la trayectoria de la partícula vista desde ese inercial será:

$$x^\mu(\tau)=(ct(\tau), \vec x(\tau))$$

De este modo, la cuadrivelocidad de una partícula medida desde un inercial $S$ cualquiera será:

$$v^\mu(\tau)=(c\gamma, \vec v\gamma)$$

**Nota 1:** he usado que $\boxed{dt=\gamma d\tau}$, que se deduce de tomar la regla de dilatación temporal y hacer los intervalos temporales infinitesimalmente pequeños

**Nota 2:** $\vec v$ es la velocidad que mide el observador de $S$ en cada instante de su reloj

**Nota 3:** el factor de Lorentz $\gamma$ depende de la velocidad de la partícula **en el inercial $S$**: $\gamma=\gamma(v^2)$. En las transformaciones de Lorentz, depende de la velocidad con la que se está alejando el inercial $S'$ de nuestro inercial de referencia $S$, pero como en este caso $S'$ es el propio de la partícula, la velocidad del propio coincide con la de la partícula y es por eso que solo nos sale la velocidad de la partícula en $\gamma$, naturalmente como la medimos desde $S$

**Corolario:** las trayectorias parametrizadas con su tiempo propio son **curvas de velocidad unidad**. Además, su cuadrivelocidad es **de tipo tiempo + invariante Lorentz + constante**, y vale:

$$\boxed{v^2=-c^2}$$

**Nota:** esto era esperable, porque hemos visto que el tiempo propio es esencialmente el elemento de linea de la curva

**Dem.:**

Es un resultado conocido de teoría de curvas que una curva parametrizada con su parámetro natural (longitud de arco) es de módulo unidad:

$$\dot{\vec x}(s)^2 = \left(\frac{d\vec x}{dt}\frac{ds}{dt}\right)^2$$

donde $s=\int_{t_0}^t|\vec x'(\sigma)|d\sigma \Longrightarrow \frac{d\vec x}{dt}=|\vec x'(t)|$ de modo que la expresión queda en:

$$\dot{\vec x}(s)=\left|\frac{d\vec x}{dt}\right|^2\left|\frac{d\vec x}{dt}\right|^{-2}=1$$

A pesar de que este resultado es propio de la teoría de curvas en $\mathbb R^n$, podemos extrapolarlo a variedades Lorentzianas. Es fácil de demostrar: sea la trayectoria en el sistema de referencia propio de la partícula:

$$x^\mu(c\tau, \vec 0)$$

donde hemos cogido cualquier propio que esté sobre la partícula (i.e: que esta se encuentre en su origen). Entonces, la cuadrivelocidad en el propio será:

$$v^\mu = (c, \vec 0)$$

Y tomando ahora la magnitud al cuadrado:

$$v^2 = v_\mu v^\mu = -c^2=cte.+inv._\blacksquare$$

### Cuadriaceleración

**Def.:** la cuadriaceleración es la derivada de la cuadrivelocidad con respecto del tiempo propio de la partícula

$$b^\mu(\tau)=\frac{dv^\mu}{d\tau}=\frac{d^2x^\mu}{d\tau^2}$$

**Proposición:** la cuadriaceleración es un tensor (concretamente, un vector contravariante)

**Proposición:** la cuadriaceleración es ortogonal a la cuadrivelocidad en todo inercial

**Dem.:**

Este también es un resultado conocido de teoría de curvas. Sea una curva parametrizada con su parámetro natural, $\vec x(s)$, entonces tenemos que:

$$\dot{\vec x}\cdot \dot{\vec x}=1 \Longrightarrow \frac{d}{ds}(\dot{\vec x}\cdot \dot{\vec x})=0=2\ddot{\vec x}\cdot \dot{\vec x}$$

Y por ende $\ddot{\vec x}\cdot \dot{\vec x}=0$. Veamos ahora que esto se cumple asimismo en curvas en una variedad Lorentziana. El argumento es exactamente el mismo:

$$v^2=-c^2\Longrightarrow \frac{d}{d\tau}(v^2)=2\eta_{\mu\nu}b^\mu v^\nu = 0\Longrightarrow b_\mu v^\mu=0_\blacksquare$$

**Proposición:** la cuadriaceleración es un vector de tipo espacio. Para cualquier inercial arbitrario, tenemos que:


$$\boxed{b^2=\gamma^4\left[\gamma^2\frac{(\vec v\cdot \vec a)^2}{c^2} + \vec a^2\right] \geq 0}$$

Asimismo, sus componentes en un inercial arbitrario son:

$$\boxed{b^0=\gamma^4\frac{\vec v\cdot\vec a}{c}; \ \ \vec b=\gamma^4\left(\frac{(\vec v\cdot \vec a)\vec v}{c^2}\right)+\gamma^2\vec a}$$

**Leyes de transformación de la tri-aceleración:**

$$\boxed{\begin{gather}a'_{||}= \frac{a_{||}}{\gamma^3(1-vV_{||}/c^2)^3} \\ \vec a'_\perp = \frac{\vec a_\perp}{\gamma^2(1-vV_{||}/c^2)^2} + \frac{va_{||}\vec V_\perp}{c^2\gamma^2(1-vV_{||}/c^2)^3}\end{gather}}$$

**Nota:** la descomposición está hecha con respecto de la velocidad que medimos de la partícula desde $S$: $\vec a=a_{||}\frac{\vec v}{v^2}+\vec a_\perp$ (e ídem para $\vec V$, que es la velocidad a la que $S'$ se aleja de $S$)

## Ecuaciones de Euler-Lagrange

- Para un sistema dado, construimos su acción como la magntiud más general que refleje las simetrías que sabemos que tiene nuestro sistema (por ejemplo, debería tener la misma forma bajo transformaciones de Lorentz si queremos describir un sistema relativista, para respetar el postulado de relatividad)
- Para un parámetro temporal arbitrario $\sigma$, tenemos que la acción del sistema será:

$$S=\int_{\sigma_0}^{\sigma_1}d\sigma L(x^\mu, \mathring x^\mu, \sigma)$$

- En virtud del principio de acción extremal, la partícula seguirá una trayectoria en el espacio de fases en velocidades dado por las soluciones de las ecuaciones de Euler-Lagrange:

$$\boxed{\frac{\partial L}{\partial x^\mu}-\frac{d}{d\sigma}\left(\frac{\partial L}{\partial \mathring x^\mu}\right)=0}$$

**Proposición:** el Lagrangiano es un invariante Lorentz

**Proposición:** las ecuaciones de Euler-Lagrange son covariantes

## Primer Teorema de Noether

- Sea una transformación de las trayectorias dependiente de $N$ **parámetros** tal que $\delta x^\mu = X^\mu_a \delta\epsilon^a$
- Sea un Lagrangiano $L$ que, en el entorno de cualquier trayectoria (no solo las físicas), varíe únicamente en la derivada temporal total de una función $f(x^\mu, \mathring x^\mu, \sigma)$ (una por cada parámetro) cuando le apliquemos estas transformaciones:

$$\delta L = \mathring f_a\delta\epsilon^a$$

- Entonces, diremos que esta transformación es una **simetría del sistema** y tenemos $N$ cantidades conservadas:

$$\boxed{Q_a=\theta_a-f_a=cte.}$$

con $\theta_a = \frac{\partial L}{\partial \mathring x^\mu}X^\mu_a$

**Nota:** el segundo teorema de Noether es para el caso de que los parámetros de las transformaciones sean funciones sobre la variedad (i.e: decimos que la transformación es local). Entonces, lo que nos dice el segundo teorema de Noether es que si nuestro Lagrangiano es invariante bajo dichas transformaciones, no es que se conserve nada, es que las ecuaciones del movimiento (tanto sobre trayectorias físicas como las que no lo son) son redundantes: estamos usando demasiados grados de libertad. Es por esto que en teorías de campos, cuando imponemos cierta invariancia gauge, tenemos por hipótesis una libertad de gauge, lo que quiere decir que tenemos demasiados grados de libertad (i.e: podemos fijar un gauge concreto y la dinámica del sistema permanece inalterada--> no necesitábamos ese grado de libertad). Sin embargo, son pocas las veces que realmente podemos prescindir del grado de libertad adicional, porque la descripción del sistema depende de él

## Conservación del cuadrimomento

- Sea un sistema invariante bajo traslaciones espaciotemporales: $\delta x^\mu = \delta \alpha^\mu$ (concretamente, supongamos que la variación del Lagrangiano es 0 bajo estas transformaciones)
- Entonces, tenemos que:
    -  $X_a^\mu=\delta^\mu_a$
    - $\delta\epsilon^a=\delta\alpha^a$
    - $f_a=0$

Por ende, la carga Noether conservada será:

$$Q_a=\frac{\partial L}{\partial\mathring x^\nu}\delta^\nu_a$$

O bien poniéndolo ya con índices griegos:

$$Q_\mu = p_\mu=cte.$$

donde $p_\mu=\frac{\partial L}{\partial\mathring x^\nu}$ es el "cuadrimomento canónico conjugado de $x^\mu$"


### Corolario: conservación de la energía y el trimomento lineal

- Supongamos que hacemos una transformación en cierta dirección del espaciotiempo, señalada por el vector unitario $n^\mu$. Entonces, tenemos que la transformación es: $\delta x^\mu=n^\mu\delta\alpha$. Si el Lagrangiano es invariante bajo dicha transformación, entonces tenemos que se conserva la proyección del cuadrimomento en esa dirección:

$\delta x^\mu = \delta\alpha n^\mu \Longrightarrow \delta\epsilon=\delta\alpha; X^\mu = n^\mu$

$$Q=n^\mu p_\mu=cte.$$

**Resultado:** si la dirección espaciotemporal es de tipo tiempo, existe un inercial en el que lo que se conserva es la energía del sistema, $E=p^0$. Si la dirección es de tipo espacio, existe un inercial en el que lo que se conserva es el trimomento del sistema. Si la dirección es de tipo luz, no existe ningún inercial donde se conserve solo una de las dos magnitudes

**OJO!:** esta carga será conservada en cada inercial, pero podrá variar de uno a otro (i.e: no es invariante) porque está construida como la contracción de un tensor con algo que en general no es un tensor

### Condición de invariancia

- Un sistema es invariante bajo traslaciones en cierta dirección del espaciotiempo cuando:

$$\boxed{\frac{\partial L}{\partial x^\mu}n^\mu=0}$$

En particular, será invariante bajo traslaciones espaciotemporales arbitrarias si no depende explícitamente de las coordenadas $x^\mu$

**Dem.:**

$$\delta L =L'-L=L(x^\mu+n^\mu\delta\alpha, \mathring x^\mu, \sigma)-L(x^\mu, \mathring x^\mu, \sigma)\approx \frac{\partial L}{\partial x^\mu}n^\mu\delta\alpha=0$$

donde igualamos a 0 por hipótesis, porque partimos de que es invariante y queremos buscar qué condición ha de cumplir para ello. Si el Lagrangiano es invariante bajo traslaciones espaciotemporales en cualquier dirección, vemos entonces que esta relación vale para $n^\mu$ arbitrario, y por ende lo que se debe anular es la derivada (es una condición más restrictiva, como cabía esperar, porque la simetría es mayor)$_\blacksquare$

## Conservación del cuadrimomento angular

- Sea un sistema invariante bajo transformaciones de Lorentz. Entonces, la carga conservada es el "cuadrimomento angular":

$$\boxed{J_{\rho\sigma}=x_\rho p_\sigma - x_\sigma p_\rho}$$

### Corolario: conservación de trimomentos

- Si el sistema es solo invariante bajo rotaciones en el espacio en torno a algún eje $\hat n$, entonces lo que se conserva es la proyección del momento angular (normal, entendido como generador de las rotaciones) en esa dirección:

$$Q=\hat n\cdot \vec J=cte$$

con $J^i=\frac{1}{2}\epsilon^{ijk}J_{jk}$

- Si el sistema es solo invariante bajo boosts de Lorentz en la dirección $\hat n$, entonces lo que se conserva es la proyección de los generadores de dichos boosts en esa dirección:

$$Q=\hat n\cdot \vec K$$

con $K_i=J_{i0}$

### Condición de invariancia

- Un sistema es invariante Lorentz si su Lagrangiano está construido únicamente con escalares Lorentz, ya que estos no se transforman bajo transformaciones Lorentz

## Invariantes de Casimir

**Def.:** un "invariante de Casimir" de un grupo de Lie es aquel elemento del álgebra universal del grupo que conmuta con todos los elementos del grupo (**invariante** bajo todo el grupo de transformaciones). Esto es, todos los observadores están de acuerdo en cuánto vale el número, independientemente del inercial desde el que lo midan

- Los únicos dos invariantes de Casimir del grupo de Poincaré son:

1. La masa: $p^2=-m^2$
2. La magnitud del spin al cuadrado (véanse apuntes de Garay, vector de Pauli-Lubanski)

**Fíjate:** para que sea conservada, el Lagrangiano ha de ser invariante bajo cierta simetría, pero es que el Casimir también es invariante!

**Def. (matemática):** un "invariante Casimir" de un grupo de Lie es aquel elemnto del álgebra universal del grupo que conmuta con todos los generadores de dicho grupo

**Def.:** el álgebra universal de un grupo es el álgebra compuesto por todos los polinomios de los elementos del álgebra (normal) y sus combinaciones lineales

**Interpretación de la masa:** el hecho de que la masa de un sistema siempre sea la misma en cualquier inercial nos da una interpretación (no-cuántica) de esta: la masa (inercial) de una partícula es el mínimo de la energía que puede tener, y este mínimo se observa, lógicamente, en su sistema de referencia propio (equivalentemente, en sus comóviles)

## Segundo Teorema de Noether

- "Si el Lagrangiano es invariante (hasta la derivada temporal total de cierta función) bajo una transformación dependiente de las coordenadas de la variedad (i.e: función, no parámetro), entonces a nuestro sistema le sobran grados de libertad (tenemos libertad de gauge) y **las ecuaciones del movimiento para distintos grados de libertad no son independientes**"

# 6. La partícula libre

- Sabemos que la trayectoria de una partícula a través de cierta variedad diferenciable traza una geodésica, y sabemos que las geodésicas cumplen que su longitud de arco entre cualesquiera dos puntos de la variedad por los que pasen es extremal, de modo que:

$$l=\int_{s_0}^{s_1} ds = \int_{\sigma_0}^{\sigma_1} \sqrt{|\eta_{\mu\nu}\mathring x^\mu \mathring x^\nu|}d\sigma$$

- **OJO!:** si en particular escojo como parámetro temporal el tiempo propio de la partícula en movimiento ($\sigma=\tau$), me sale la magnitud al cuadrado de la cuadrivelocidad, que es constante. Esto es una ligadura del sistema, y precisamente escojo construir la acción con un parámetro temporal arbitrario porque la alternativa era construirla con $d\tau$, pero es incorrecto derivar las ecuaciones E-L tras imponer la ligadura. Estrictamente, habría que introducirla durante el cálculo de E-L con un multiplicador de Lagrange, pero para partícula libre y para partícula acoplada a campo EM funciona. Con todo, construyo acción inv. bajo reparam. temporales para no tener que meter ligaduras y luego ya hago $\sigma=\tau$ y respeto ligaduras a posteriori
