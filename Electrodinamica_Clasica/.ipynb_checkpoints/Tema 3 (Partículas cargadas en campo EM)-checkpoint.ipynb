**Objetivos:**

- Derivar las ecuaciones del movimiento de una partícula puntual sometida a campo EM externo
- Generalizarlas a distribuciones de masa y carga
- Entender invariantes adiabáticos

# 1. Partícula puntual en campo EM externo: ecs. del movimiento

## Lagrangiano

Deseamos una acción que sea invariante bajo:

1. Lorentz (principio de relatividad, postulado 1 de rel. especial)
2. Reparametrizaciones temporales (tiene que valer para cualquier tiempo con el que parametrice las trayectorias, ningún inercial es especial)
3. Gauge de cuadripotencial (campo EM es invariante bajo $A_\mu \to A_\mu +\partial_\mu\alpha$ y por ende la dinámica de la partícula también ha de serlo)

Para ello, tengo los siguientes ingredientes:

- $\mathring{\mathbf z}(s)ds,\mathbf A, \mathbf \eta, \mathbf \epsilon$

Donde hemos cogido el diferencial de la trayectoria y no la trayectoria porque así conseguimos algo invariante bajo reparametrizaciones temporales. Las únicas cantidades Lorentz-invariantes que podemos construir con estos tensores son:

- $\mathbf A^2, \mathbf A\cdot \mathring{\mathbf z}(s)ds, \mathring{\mathbf z}^2(s)(ds)^2$

- La acción deberá por tanto ser una función en principio arbitraria de estas cantidades, si bien ha de ser homogénea en las velocidades para que sea invariante bajo reparametrizaciones temporales:


$$S=\int ds\sqrt{-\mathring{\mathbf z}^2} \ \times f(\mathbf A^2, \mathbf A\cdot \mathring{\mathbf z}/\sqrt{\mathring{-\mathbf z}^2})$$

Como además debe haber invariancia gauge, solamente podemos usar dos términos:

- $\sqrt{-\mathring{\mathbf z}^2}$ (no depende de $\mathbf A$)
- $\mathbf A\cdot \mathring{\mathbf z}$

El segundo término es fácil ver que deja la acción invariante gauge porque al hacer la transformación gauge la acción solo cambia en una derivada temporal total:

$$\int dsA_\mu \mathring z^\mu \to \int ds (A_\mu+\partial_\mu\alpha)\mathring z^\mu$$

Pero como cuando trabajamos con partículas puntuales, los campos están evaluados sobre su trayectoria (lógicamente, porque es lo único que ven, al menos en teorías locales), tenemos que $\partial_\mu\alpha \mathring z^\mu = \frac{d}{ds}\alpha(z(s))$

Por tanto, la acción más general posible que cumpla las 3 invariancias que hemos pedido es:

$$\boxed{S=-mc\int ds\sqrt{-\mathring{\mathbf z}^2}+q\int ds\mathbf A\cdot \mathring{\mathbf z}}$$

**Nota:** no sería correcto introducir directamente la ligadura de que las curvas parametrizadas con tiempo propio son curvas unidad, porque esto es una ligadura y como tal habría que meterla con multiplicadores de Lagrange. Por ende, lo correcto es aplicar ecuaciones de E-L con parámetro temporal general y ya después, cuando hemos particularizado trayectorias clásicas (por haber usado E-L), entonces podemos aplicar la ligadura, porque esta solo se cumple sobre trayectorias clásicas.

## Ecuaciones del movimiento

Las ecuaciones (parametrizadas con tiempo propio para que sean ecuaciones entre tensores) son:

$$\boxed{m\dot u_\mu = qu^\nu F_{\mu\nu}}$$

**Observación:** la 4-fuerza $f_\mu =qu^\nu F_{\mu\nu}$ es ortogonal a la cuadrivelocidad (claro, ya habíamos dicho que la cuadriaceleración (que es lo que hay en la izda.) y la cuadrivelocidad son ortogonales)

**Proposición:** en dinámica relativista, toda 4-fuerza ha de ser ortogonal a la cuadrivelocidad: $\boxed{f^\mu u_\mu=0; \ \ \forall f}$, y toda fuerza ha de ser un vector de tipo espacio

**Dem:**

- La segunda ley de Newton relativista se escribe como: $m\dot u^\mu = f^\mu$. Cuando parametrizamos las trayectorias con tiempo propio, tenemos que estas son curvas de velocidad unidad, de modo que $u^2=-1$. Derivando con respecto de $\tau$ a ambos lados, tenemos que $u^\mu\dot u_\mu=0$, y por ende cuadrivelocidad y cuadriaceleración son ortogonales, y por ello la 4-fuerza también es ortogonal a la cuadrivelocidad. Asimsmo, como todo cuadrivector ortogonal a uno de tipo tiempo es de tipo espacio, toda 4-fuerza ha de ser de tipo espacio$_\blacksquare$

En un inercial dado:

$$\boxed{\begin{cases}\frac{d\vec p}{dt}=q(\vec E+\vec v\times \vec B) \\ \frac{dE_M}{dt}=q\vec v\cdot \vec E\end{cases}}$$

donde $\vec p=\gamma m\vec v$ y $E_M=\gamma mc^2=\sqrt{m^2c^4+\vec p^2c^2}$. Véase que en general $E_M$ (energía mecánica) ya no será la energía total del sistema, porque la segunda ecuación se interpreta como "el campo eléctrico bombea energía al sistema". Véase que el campo magnético no puede hacer esto

**Proposición:** si el potencial vector no depende explícitamente del tiempo, la energía total se conserva: $\boxed{\frac{\partial\vec A}{\partial t}=0\Longrightarrow E=E_M+\varphi=cte.}$

**Dem:**

El campo eléctrico es en general: $\vec E=-\vec \nabla \varphi-\frac{\partial\vec A}{\partial t}$. Si la última componente se anula, entonces podemos escribir la segunda ec. como:

$$\frac{dE_M}{dt}=-q\vec v\cdot \vec\nabla\varphi$$

Sin embargo, podemos escribir el término RHS como una derivada temporal total: $\frac{d}{dt}\varphi(\vec x(t))=\vec\nabla\varphi\frac{d\vec x}{dt}$. Por ende, podemos meter todo bajo el signo de la derivada y tenemos:

$$E=E_M+q\varphi=cte._\blacksquare$$

## Generalización: distribuciones de carga / masa

Ya sabemos cómo es la dinámica de una partícula sometida a un campo EM de fondo (no hemos considerado la dinámica del campo, es decir, la contribución al campo EM que hace la propia carga)

Vamos a estudiar ahora qué pasa si en vez de tener una partícula puntual tenemos una distribución de masa con cierta distribución de carga, sometida a un campo EM externo. La forma en la que haremos esto será relacionando el concepto de partícula puntual con el de distribución, ya que conocemos la dinámica de la primera

**Def.:** sea la distribución de masa de una partícula puntual: $\mu(\vec x)=m\delta^3(\vec x-\vec z(t))$, y la distribución de carga: $\rho(\vec x)=q\delta^3(\vec x-\vec z(t))$

**Proposición:** estas densidades no son Lorentz-invariantes, pero $\frac{1}{\gamma}\mu$ y $\frac{1}{\gamma}\rho$ sí lo son

**Dem:**

Sabemos que la masa es una cantidad Lorentz-invariante, de modo que al hacer la integral: $\int_{\mathbb R^3}d^3\vec x\mu(\vec x)=m$ deberíamos obtener lo mismo en todo inercial. Si escogemos la medida de integración invariante $\gamma d^3\vec x$, entonces tenemos que dividir la densidad por $\gamma$, y como tanto $m$ como la medida de integración son Lorentz-invariantes, la densidad dividida por $\gamma$ también ha de serlo. Esto aplica también para la densidad de carga$_\blacksquare$

**Def.:** sea la 4-corriente de carga asociada a la partícula:

$$j^\mu=(c\rho, \vec j); \ \ \vec j=\rho\vec v$$

Es un cuadrivector, y se puede escribir en términos de la trayectoria de la partícula de la forma:

$$j^\mu = qc\int d\tau \dot z^\mu\delta^4(\mathbf x-\mathbf z(\tau))$$

Entonces, la ecuación (covariante) del movimiento de la partícula toma la forma:

$$\boxed{\frac{\mu}{\gamma}\dot u^\mu=j_\nu F^{\mu\nu}}$$

**OJO!:** lo que antes era una cuadrivelocidad ahora es un campo vectorial, porque en cada punto del espacio-tiempo el fluido tendrá una cuadrivelocidad dada. Por ende, esta generalización se hace de forma un poco sloppy aunque al final se llega al resultado correcto. Este es ahora aplicable a cualquier distribución de carga y masa

# Campo EM y transformaciones de Lorentz

- La forma explícita (en la base natural) del tensor $F_{\mu\nu}$ y de su dual de Hodge es:

$$F_{\mu\nu}={\begin{pmatrix}0 & -E^1/c & -E^2/c & -E^3/c \\ E^1/c & 0 & B^3 & -B^2 \\ E^2/c & -B^3 & 0 & B^1 \\ E^3/c & B^2 & -B^1 & 0\end{pmatrix}}$$
$$^*F_{\mu\nu}={\begin{pmatrix}0 & B^1 & B^2 & B^3 \\ -B^1 & 0 & E^3/c & -E^2/c \\ -B^2 & -E^3/c & 0 & E^1/c \\ -B^3 & E^2/c & -E^1/c & 0\end{pmatrix}}$$


**Proposición:** sea un inercial S' moviéndose con velocidad $\vec v$ con respecto de $S$. Entonces, los campos eléctrico y magnético observados en este nuevo inercial serán:

$$\boxed{\begin{gather}\vec E'=\gamma(\vec E+\vec v\times \vec B)-\frac{\gamma^2}{\gamma+1}(\vec v\cdot \vec E)\frac{\vec v}{c^2} \\ \vec B'=\gamma(\vec B-\vec v\times \vec E/c^2)-\frac{\gamma^2}{\gamma+1}(\vec v\cdot \vec B)\frac{\vec v}{c^2} \end{gather}}$$

**Proposición:** como $F_{\mu\nu}$ es invariante gauge, los campos EM también lo son

## Invariantes

**Proposición:** las cantidades $\boxed{F_{\mu\nu}F^{\mu\nu}=\frac{-2}{c^2}(\vec E^2-c^2\vec B^2)}$ y $\boxed{^*F^{\mu\nu}F_{\mu\nu}=\frac{2}{c}\vec E\cdot \vec B}$ son invariantes. El primero es Poincaré invariante y el segundo solo es invariante para transformaciones de Lorentz propias ortocronas

## Invariantes adiabáticos

**Def.:** en un sistema perturbado, un "invariante adiabático es una cantidad que en el sistema sin perturbar se conservaría, y que en el perturbado solo varía en órdenes superiores

**Proposición:** las "integrales de acción" son invariantes adiabáticos:

$$J_a=\oint p_adq_a$$

**Caso de interés: partícula sometida a campo $\vec B$ uniforme y cte.**

En este caso, tenemos que un invariante adiabático es:

$$J=\frac{1}{2\pi}\oint P_\varphi d\varphi$$

Sin embargo, el integrando también se puede escribir como $\vec P_\perp\cdot d\vec l_\perp$, donde $\vec P$ es el momento canónico conjugado del sistema y $d\vec l_\perp$ es el elemento de línea de la circunferenci que describe la partícula en el plano perpendicular a $\vec B$. Entonces, tenemos que $\vec P_\perp\cdot d\vec l_\perp=P_\varphi \frac{\vec e_\varphi}{r}\cdot rd\varphi\vec e_\varphi=P_\varphi d\varphi$. Por tanto, tenemos:

$$J=\frac{1}{2\pi}\oint \vec P_\perp \cdot d\vec l_\perp$$

Como $\vec P=\vec p+q\vec A$, podemos escribir:

$$J=\frac{1}{2\pi}\oint (\vec p_\perp \cdot d\vec l_\perp)+\frac{1}{2\pi}q\oint\vec A\cdot d\vec l_\perp$$

Como $\vec p_\perp || d\vec l_\perp$, y además es de módulo constante $p_\perp = qBr_B$, tenemos que la primera integral da $p_\perp r_B$, equivalentemente $\frac{p_\perp^2}{qB}$. La segunda integral se puede calcular mediante el teorema de Stokes, porque es una integral de línea cerrada, así que:

$$\oint_{\partial S}\vec A\cdot d\vec l_\perp = \int_S(\vec \nabla \times \vec A)\cdot d\vec S = \int \vec B\cdot \hat nd^2x_\perp=-B\pi r_B^2$$

De este modo, concluimos que el invariante adiabático en este sistema será:

$$J=\frac{p_\perp^2}{2qB}=cte.$$

Si ahora perturbamos el sistema, ya sea en el tiempo, el espacio o en ambas, de tal forma que $\left|\frac{\dot \omega_B}{\omega_B}\right|<<1$, entonces este invariante adiabático se mantendrá prácticamente constante.


## Aplicación: botella magnética

- Sea una partícula cargada sometida a un campo $\vec B$ que es estático pero ligeramente inhomogeneo en el espacio: $\vec B=B_z(z)\vec e_z+\vec B_\perp$, donde $\vec B_\perp$ muy pequeño y se ha añadido para que la divergencia del campo sea nula
- Como el campo magnético es estático, no da lugar a campo eléctrico y por ende la energía del sistema se conserva, de modo que $E^2=cte. \Longrightarrow \vec p^2 = \vec p_\perp^2+\vec p_{||}^2=cte.$ Podemos escribir la componente paralela en función de nuestro invariante adiabático:

$$p_{||}^2=\vec p^2-2qJB_z(z)$$

También podemos expresar el radio de las rotaciones en el plano perpendicular al campo en términos del invariante:

$$r_B=\sqrt{\frac{2J}{qB(z)}}$$

De este modo, vemos que en las regiones donde el campo sea muy intenso, el radio irá a 0. Metiendo esto en la ecuación del movimiento, tenemos que:


$$\ddot z=-\frac{qJ}{m^2\gamma^2}\partial_z B_z(z)$$

De modo que la partícula irá trazando circunferencias de radio $r_B(z)$ hasta que llegue un punto en el que $\vec p_{||}^2=0$ y entonces volverá a bajar. Es decir, hemos conseguido "atrapar" a la partícula en una región finita del espacio. Esto es una "botella magnética"
