# 1. Transformaciones de campos

**Def:** sea un campo $\Phi_A(x)$ una aplicación definida sobre cierta variedad tal que a cada punto de esta le asocia $N$ valores ($\Phi_1(p), \ldots, \Phi_N(p)$). Sea una transformación activa en general tanto del campo como de las coordenadas, de modo que el nuevo campo tras la transformación es $\Phi'_A(x')$. Entonces, se define la "variación general" del campo como:

$$\bar \delta \Phi(x):=\Phi'_A(x')-\Phi_A(x)$$

**Proposición:** una transformación general de un campo cualquiera tiene dos contribuciones (si la transformación de las coordenadas es infinitesimal): la **transformación intrínseca** ($\delta\Phi_A$) y el **término de transporte**:

$$\boxed{\bar \delta \Phi_A(x)=\delta\Phi_A(x)+\delta x^\mu \partial_\mu \Phi_A(x)}$$

donde $\delta\Phi_A(x)=\Phi'_A(x)-\Phi_A(x)$ se llama "transformación intrínseca", porque solo cuantifica el cambio del campo en su forma funcional

**Dem.:**

Consideremos una transformación infinitesimal de las coordenadas. Entonces, tenemos que la variación general del campo es:

$$\bar \delta\Phi_A(x)=\Phi'_A(x^\mu+\delta x^\mu)-\Phi_A(x)=\Phi'(x)+\delta x^\mu \partial_\mu \Phi'_A(x)-\Phi_A(x)$$

esto ya es lo de arriba, salvo porque tenemos un término de transporte con el campo primado. Sabiendo que $\Phi'_A(x')=\bar \delta\Phi_A + \Phi_A(x)$, tenemos entonces que $\delta x^\mu \partial_\mu \Phi'_A(x) =\delta x^\mu \partial_\mu \Phi_A(x)+\mathcal O(\delta^2)$. Por ende, da igual si escribimos el término de transporte con o sin prima porque a primer orden son lo mismo, y como las transformaciones son infinitesimales por hipótesis, ambas valen. Sustituyendo, hemos acabado$_\blacksquare$

**Proposición:** como la variación intrínseca solo afecta a la forma funcional de los campos, conmuta con las derivadas:

$$\boxed{[\delta, \partial_\mu]=0}$$

**Dem.:**

La clave está en que, como la variación intrínseca solo se preocupa de las variaciones en la forma funcional de los campos y por ende evalúa todo en el mismo punto, las derivadas de todo se van a cancelar debido a la linealidad de las derivadas. Lo vemos:

$$[\delta, \partial_\mu]\Phi_A(x)=\delta(\partial_\mu\Phi_A(x))-\partial_\mu(\delta \Phi_A(x))$$

Aplicando la definición de variación intrínseca:

$$[\delta, \partial_\mu]\Phi_A(x)=\partial_\mu\Phi'_A(x)-\partial_\mu\Phi_A(x)-\partial_\mu(\Phi'_A(x)-\Phi_A(x))=0_\blacksquare$$

## Invariancia de un campo bajo cierta transformación

**Def.:** diremos que un campo $\Phi_A(x)$ es "invariante bajo una transformación" si su variación **general** bajo dicha transformación es 0. Esto es, si se cumple:


$$\Phi_A'(x')=\Phi_A(x)$$

# 2. Transformación de campos bajo el grupo de Poincaré

## 1. Traslaciones

**Proposición:** todo campo local (que depende solo de un punto $x$) es invariante bajo traslaciones $x'^\mu = x^\mu+\alpha^\mu$, de modo que:

$$\Phi_A'(x')=\Phi_A(x)$$

Esta expresión se lee como "el valor del nuevo campo (el trasladado) en el nuevo punto (trasladado) es el valor del campo original en el punto original". Esto es, el campo "se mueve" con el sistema coordenado

**Ejemplo:** sea una función de onda que depende solo de la dimensión espacial, $\Psi(x)$. Si la trasladamos un tramo $a$, entonces tenemos que:

$$\Psi(x)\xrightarrow[T_a]{}T_a\Psi(x)=\Psi(x-a)$$

Por ende, vemos que (1) la forma funcional no cambia: $\Psi'=\Psi$ y (2) la nueva variable es $x'=x-a$, de modo que efectivamente $\Psi(x)=\Psi'(x')$

## Operador momento

- La variación intrínseca de un campo local bajo traslaciones será entonces:

$$\boxed{\delta\Phi_A(x)=-\delta \alpha^\mu \partial_\mu \Phi_A(x)}$$

**Def.:** se define el **operador** momento como $\boxed{\hat P_\mu = -i\partial_\mu}$. La variación intrínseca bajo traslaciones se escribe entonces como:

$$\boxed{\delta\Phi_A(x)=-i\delta \alpha^\mu \hat P_\mu \Phi_A(x)}$$

**Conclusión:** el operador momento es el generador de las traslaciones.

**Propiedad:** como el grupo de traslaciones es abeliano, sus generadores conmutan todos con todos:

$$\boxed{[\hat P_\mu, \hat P_\nu]=0}$$

Podemos integrar formalmente la variación intrínseca para encontrar la nueva forma funcional del campo en términos de la variable vieja $\Phi_A'(x)$:

$$\Phi'_A(x)=e^{-i\alpha^\mu \hat P_\nu}\Phi_A(x)$$

## 2. Lorentz

### Operador $\hat L_{\mu\nu}$

- Para transformaciones de Lorentz: $\delta x^\mu = \delta\omega^\mu_\nu x^\nu$, por lo que la variación general del campo es:

$$\bar \delta\Phi_A(x)=\delta\Phi_A(x)+\delta \omega^\mu_\nu x^\nu\partial_\mu\Phi_A(x)$$

**Def.:** se define el "operador momento angular orbital" como:

$$\boxed{\hat L_{\mu\nu}=x_\mu \hat P_\nu -x_\nu \hat P_\mu}$$

Ahora la variación general se puede escribir en términos de este operador como hacíamos en relatividad especial para la partícula, solo que ahora lo que tenemos son campos:

$$\bar \delta\Phi_A(x)=\delta\Phi_A(x)-\frac{i}{2}\delta\omega^{\mu \nu}\hat L_{\mu\nu}\Phi_A(x)$$

O bien con trivectores (ahora son operadores):

$$\bar \delta\Phi_A(x)=\delta\Phi_A(x)+i(\hat{\vec K}\cdot d\vec \xi+\hat{\vec L}\cdot d\vec \theta)\Phi_A(x)$$

**Def.:** hasta ahora, definíamos un "vector" como todo aquel objeto que se transformase como las coordenadas (o inversamente) bajo una transformación de Lorentz. Esto es, $v^\mu \to v'^\mu = \Lambda v^\mu$. Esto es equivalente a decir que $\boxed{[\hat L_{\rho\sigma}, \hat v_\mu]=i(\eta_{\mu\rho}v_\sigma-\eta_{\mu\sigma}v_\rho)}$. Esta relación, junto con $[\hat P_\mu, \hat P_\nu]$ y la de $[\hat L_{\rho\sigma}, \hat L_{\mu \nu}]$ completan el álgebra de Poincaré, porque nos dan las relaciones de conmutación entre todos los elementos de la base del álgebra

### Operador $\hat S_{\mu\nu}$

**Proposición:** podemos escribir la variación general de un campo bajo una transformación de Lorentz en términos de $\delta\omega$ y un operador, que llamaremos "operador de spin", de la forma:

$$\bar \delta\Phi_A(x)=\frac{i}{2}\delta\omega^{\mu\nu}(\hat S_{\mu\nu}\Phi)_A$$

### Operador $\hat J_{\mu\nu}$

**Proposición:** la variación intrínseca de un campo se puede escribir en términos de un operador "momento angular total" $\hat J_{\mu\nu}$ como:

$$\delta\Phi_A(x)=\frac{i}{2}\delta\omega^{\mu\nu}(\hat J_{\mu\nu}\Phi)_A$$

donde $\boxed{\hat J_{\mu\nu}= \hat L_{\mu\nu}+\hat S_{\mu\nu}}$

**Dem.:**

Se llega a esta expresión directamente introduciendo en la expresión de $\bar \delta\Phi_A$ en función de $\hat L_{\mu\nu}$ la expresión en términos del operador de spin$_\blacksquare$

### Campos escalares

**Def.:** un campo "escalar" bajo cierta transformación es aquel que queda invariante bajo dicha transformación, y por ende $\bar \delta\Phi_A(x)=0$

**Consecuencia:** los campos escalares bajo las transformaciones de Lorentz "no tienen spin": $\hat S_{\mu\nu}\phi=0$. Por ende, ya sabemos cómo se transforman los campos escalares bajo transformaciones de Lorentz:

$$\delta\phi(x)=\frac{i}{2}\delta\omega^{\mu\nu}\hat J_{\mu\nu}\phi(x) = \frac{i}{2}\delta\omega^{\mu\nu}\hat L_{\mu\nu}\phi(x)$$

**Nota:** estos operadores ya los conocemos, porque son los generadores del grupo de Lorentz, que se pueden obtener con teoría de grupos de Lie

### Campos vectoriales

**Def.:** un campo vectorial bajo transformaciones de Lorentz es aquel que se transforma según: $V^\mu \to V'^\mu = V^\mu +\delta\omega^\mu_\nu V^\nu$

Su variación general se puede escribir intrínsecamente como:

$$\bar \delta \mathbf{V}(x)=\frac{i}{2}\delta\omega^{\mu\nu}\hat{\mathbf S}_{\mu\nu}\mathbf V$$

Si queremos tomar coordenadas, tendremos entonces que $(\hat{\mathbf S}_{\rho\sigma}\mathbf V)^\mu=(\hat{\mathbf S}_{\rho\sigma})^\mu_\nu V^\nu$

### Campos tensoriales

- Para campos tensoriales, tenemos exactamente las mismas leyes de transformación, porque las hemos dado de forma intrínseca. Sin embargo, hay que tener en mente que ahora el tensor de spin tendrá más componentes. Concretamente, para un tensor n-veces contravariante $T^{\alpha_1\ldots\alpha_n}$ tenemos:

$$(\hat{\mathbf S}_{\rho\sigma}\mathbf T)^{\alpha_1\ldots\alpha_n}=(\hat{\mathbf S})^{\alpha_1\ldots\alpha_n}_{\beta_1\ldots\beta_n}T^{\beta_1\ldots \beta_n}$$

## Casimires de un grupo

**Def.:** un "operador de Casimir" es aquel elemento del álgebra universal del grupo (operador) que conmuta con todos los elementos del grupo (esto es, aquel operador asociado al grupo que es invariante bajo todas sus transformaciones)

- El operador momento $\hat P_\nu$ es invariante bajo traslaciones, pero no lo es bajo transformaciones de Lorentz porque es un (co)vector. Sin embargo, su magnitud $\hat P^2=\hat P_\mu \hat P^\mu$ es un escalar Lorentz por construcción, y sigue siendo invariante bajo traslaciones, de modo que es invariante bajo todo el grupo de Poincaré!--> Es un Casimir del grupo. Concretamente:

$$\boxed{-\hat P^2=\square}$$

**Interpretación de sus autovalores:** tiene un espectro continuo de autovalores positivos, y estos dan la masa (inercial) del sistema. Al igual que en relatividad especial para una partícula, la interpretamos como la energía mínima que podemos medir para el sistema. Esta medida la hará un observador en el centro de inercia del sistema

**Nota 1:** se puede demostrar que el grupo de Poincaré tiene un total de dos Casimires (la magnitud del operador momento y la magnitud del operador de Pauli-Lubanski) si la masa es $m\neq 0$. Si no, ambos son proporcionales, y el factor de proporcionalidad es el operador de helicidad

**Nota 2:** los Casimires de un grupo de simetrías de un sistema dan información intrínseca de este sistema, porque no dependen del inercial en el que se consideren. En particular, para el grupo de Poincaré, tenemos masa y spin

# 3. Ecuaciones de Euler-Lagrange

## Construyendo la acción

Consideramos algunas propiedades ad hoc que ha de cumplir nuestra acción:

1. **Localidad** (la dependencia en el espacio-tiempo es puntual) $\to$ Causalidad (la información tarda un tiempo finito en llegar de un punto al otro del espacio)
2. Debe contener como máximo **derivadas de segundo orden** $\to $ Causalidad (si hay derivadas superiores, el modelo suele resultar ser no causal)
3. **Real** $\to$ Evitar extinciones: si la acción es compleja, da lugar a extinciones (pérdida de materia). Esto se ve bien en la integral de camino de Feynman, pero estamos en la teoría clásica
4. Es conveniente que **refleje las simetrías** del sistema que describe (i.e: si el sistema es invariante bajo transformaciones de Lorentz, la acción también, etc.) $\to$ Facilita teorema de Noether

- Podemos construir la acción a partir de la densidad Lagrangiana de la forma:

$$\boxed{S=\int_\Omega d^4x \mathcal L(\Phi_A, \partial_\mu\Phi_A, x)}$$

donde $\Omega\subseteq M$. Véase que la acción así definida ya cumple todas las propiedades que hemos pedido

## Variando la acción

- **Princpio variacional:** las trayectorias físicas de campos que tienen condiciones de contorno fijas en $\partial\Omega$ son aquellas que hacen extremal la acción

- Supongamos conocida la "trayectoria" (configuración en cada instante del tiempo) clásica de nuestro campo, y variemos un poco las trayectorias en torno a las clásicas. Entonces, tenemos que la acción varía de la forma:


$$\delta S = \int_\Omega d^4x \delta \mathcal L(\Phi_A, \partial_\mu\Phi_A, x) = \int_\Omega d^4x\left[\frac{\partial\mathcal L}{\partial\Phi_A}\delta\Phi_A+\frac{\partial \mathcal L}{\partial (\partial_\mu \Phi_A)}\delta(\partial_\mu\Phi_A)\right]$$

**Recuerda:** las variaciones intrínsecas de campos no alteran el espacio-tiempo así que conmutan con las derivadas: $[\delta,\partial_\mu]\Phi_A(x)=0$ y por ende podemos intercambiar estos dos símbolos en el segundo término. Usando la regla de la cadena como siempre en este tipo de cálculos variacionales, podemos reescribir la expresión como:

$$\delta S = \int_\Omega d^4x\left[\frac{\partial\mathcal L}{\partial\Phi_A}\delta\Phi_A+\partial_\mu\left(\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\delta\Phi_A\right)-\partial_\mu\left(\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\right)\delta\Phi_A\right]$$

Agrupando:

$$\delta S = \int_\Omega d^4x\left[\frac{\partial\mathcal L}{\partial\Phi_A}-\partial_\mu\left(\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\right)\right]\delta\Phi_A+\int_\Omega d^4x\partial_\mu\left(\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\delta\Phi_A\right)$$

Centrémonos en la segunda integral. Aplicando el teorema de Gauss en 4 dimensiones, sabemos que la integral sobre una región de $M$ de la cuadridivergencia de un campo vectorial es igual a la integral de superficie sobre la frontera de dicha región, $\partial\Omega$. Por ende:

$$\int_\Omega d^4x\partial_\mu\left(\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\delta\Phi_A\right) = \int_{\partial\Omega}d^3\sigma n_\mu \left(\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\right)\delta\Phi_A$$

Pero como por hipótesis los campos se anulan en la frontera de la región considerada, entonces $\delta\Phi_A(\partial\Omega)=0$ y la integral también. Así, variando infinitesimalmente la acción sobre configuraciones físicas y considerando que los campos se anulan en la frontera de una cierta región, tenemos las ecuaciones de Euler-Lagrange para campos:

$$\boxed{\frac{\partial\mathcal L}{\partial\Phi_A}-\partial_\mu\left(\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\right)=0}$$

## Libertad de gauge de $\mathcal L$

**Proposición:** dos densidades Lagrangianas que solo difieren en la divergencia de un campo vectorial inducen la misma acción (solo cambian las condiciones de contorno)

$$\mathcal L'=\mathcal L +\partial_\mu f^\mu(\Phi_A,\partial_\nu\Phi_A); \ \ \ \ \ f^\mu=f^\mu(\Phi_A,\partial_\mu\Phi_A, x)$$

# 4. Primer Teorema de Noether

- Sea un conjunto de transformaciones intrínsecas infinitesimales de los campos, que vienen caracterizadas por una familia de parámetros continuos $\delta\epsilon^a$ de forma que $\boxed{\delta\Phi_A(x)=F_{Aa}\delta\epsilon^a}$
- Si la variación de la densidad Lagrangiana bajo estas transformaciones es la divergencia de algún campo vectorial, entonces diremos que estas transformaciones son una **simetría** del sistema, y existe una corriente Noether conservada:

$$\delta\mathcal L = \partial_\mu f^\mu_a\delta\epsilon^a \Longrightarrow \boxed{j^\mu_a = \theta^\mu_a-f^\mu_a}$$

donde $\boxed{\theta^\mu_a = \frac{\partial \mathcal L}{\partial(\partial_\mu\Phi_A)}F_{Aa}}$

**OJO!:** aquí $a$ pueden ser cualesquiera índices, por ejemplo $a=\nu$ ó $a=\rho\sigma$, etc. Todo depende de cómo parametricemos la transformación

## Corrientes y Cargas Noether

**Def.:** diremos que una corriente es una corriente Noether si proviene de una corriente del teorema de Noether. Decimos que es conservada en el sentido de que su divergencia es 0 sobre trayectorias clásicas:

$$\boxed{\partial_\mu j^\mu_a = -[EL]^A F_{Aa}}$$

**Def.:** una "carga Noether" $Q$ es la integral espacial de la primera componente de la corriente de Noether, y es análoga a la carga que sale de integrar una ecuación de continuidad:

$$\int d^4x\partial_\mu j^\mu = 0 \Longrightarrow \int_{t_0}^{t_1}dt \int_{\mathbb R^3}d^3\vec x (\partial_t j^0_a+c\vec \nabla\cdot \vec j_a)=0$$

Definiendo $\boxed{Q=\int_{\mathbb R^3} d^3\vec xj^0_a}$ tenemos que:

$$0=\int_{t_0}^{t_1} dt\frac{dQ_a}{dt}+c\int_{t_0}^{t_1}dt\int_{\partial\Omega} d\vec A\cdot \vec j_a$$

Pero como admitimos que los campos se anulan en el infinito espacial y $\partial\Omega\to \infty$, tenemos entonces que la segunda contribución se anula y que la carga $Q_a$ así definida se conserva

**Proposición:** la adición de la divergencia de un tensor antisimétrico a la corriente Noether da otra corriente asimismo conservada.

$$\tilde j^\mu_a=j^\mu_a+\partial_\nu \kappa^{\mu\nu}_a$$

**Dem.:**

El término $\partial_\mu\partial_\nu\kappa^{\mu\nu}_a=0$ porque es contracción de tensor simétrico (derivadas conmutan) con antisimétrico$_\blacksquare$

## Invariancia bajo traslaciones: $\bar T^{\mu\nu}$

**Proposición:** un sistema es invariante bajo traslaciones si y solo si la densidad Lagrangiana no depende explícitamente del espacio-tiempo $x$

**Def.:** se define el **tensor de energía-momento** para cualquier sistema como:

$$\boxed{\bar T^\mu_\nu := -\frac{\partial \mathcal L}{\partial(\partial_\mu\Phi_A)}\partial_\nu\Phi_A+\mathcal L\delta^\mu_\nu}$$

**Proposición:** si un sistema es invariante bajo traslaciones, entonces su corriente Noether conservada es precisamente el tensor energía momento, y su carga Noether asociada es lo que se llama el "momento del sistema" $P_\nu$ (no es operador). Explícitamente, toma la forma:

$$\boxed{P_\nu := \int_{\partial\Omega}d^3\sigma \ n_\mu \bar T^\mu_\nu=\int_{\mathbb R^3}d^3\vec x \ \bar T^0_\nu}$$

**Dem 1:**

Tomemos la densidad Lagrangiana $\mathcal L(\Phi_A, \partial_\mu\Phi_A, x)$ y evaluemos su variación intrínseca bajo una traslación infinitesimal:

$$\delta\mathcal L = \frac{\partial\mathcal L}{\partial \Phi_A}\delta\Phi_A+\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\delta(\partial_\mu\Phi_A)+\bar \partial_\mu \mathcal L$$

donde la derivada con barra denota la derivada explícita en espacio-tiempo. Como los campos son locales, su variación intrínseca será: $\delta\Phi_A=-\delta\alpha^\mu\partial_\mu\Phi_A$. Introduciendo esta expresión en la variación de la densidad Lagrangiana:

$$\delta\mathcal L = -\delta\alpha^\mu\left[\frac{\partial\mathcal L}{\partial\Phi_A}\partial_\mu\Phi_A+\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\partial_\mu\partial_\nu\Phi_A\right]$$

Observamos que esta es precisamente la derivada de la densidad Lagrangiana (regla de la cadena), pero le falta la derivada explícita. Por ende:

$$\delta\mathcal L =-\delta\alpha^\mu(\partial_\mu\mathcal L-\bar\partial_\mu\mathcal L)$$

**Observación:** si $\bar\partial_\mu\mathcal L=0$, entonces la variación intrínseca de la densidad Lagrangiana es precisamente la de un campo invariante bajo traslaciones. Por ende, el sistema es invariante bajo traslaciones si y solo si la densidad Lagrangiana no depende explícitamente del punto del espaciotiempo$_\blacksquare$

**Dem 2:**

Supongamos ahora que el sistema es invariante bajo traslaciones en todas las direcciones del espacio. Calculemos la corriente Noether conservada asociada a esta invariancia continua. Por un lado, tenemos que la variación intrínseca de nuestros campos bajo traslacions es de la forma: $\delta\Phi_A=-\delta\alpha^\mu \partial_\mu\Phi_A=F_{Aa}\delta\epsilon^a$. Por ende, $a=\nu$ y hacemos la identificación del generador de la traslación: $\boxed{F_{A\nu}=-\partial_\nu\Phi_A}$.

Ahora ya podemos construir el primero de los dos términos de la corriente Noether: $\boxed{\theta^\mu_\nu =\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}(-\partial_\nu\Phi_A)}$. Vamos con el segundo término, que viene de cómo varía la densidad Lagrangiana bajo traslaciones. Como el sistema es invariante bajo traslaciones, la densidad Lagrangiana es un campo local y por ende, como hemos visto antes, $\delta\mathcal L=-\delta\alpha^\nu\partial_\nu \mathcal L=\partial_\mu f^\mu_\nu\delta\epsilon^\nu$. De aquí deducimos que $\boxed{f^\mu_\nu=-\mathcal L\delta^\mu_\nu}$.

Construyendo ahora la corriente Noether conservada como la resta de estas dos magnitudes, llegamos a que dicha corriente es:

$$j^\mu_\nu\equiv \bar T^\mu_\nu=\theta^\mu_\nu-f^\mu_\nu=-\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}\partial_\nu\Phi_A+\mathcal L\delta^\mu_\nu$$

## Invariancia Lorentz: $\bar J^{\mu\rho\sigma}$

**Proposición:** un sistema es invariante bajo transformaciones de Lorentz si y solo si la densidad Lagrangiana está construida con invariantes Lorentz (escalares y contracciones)

**Def.:** se definen las "corrientes de momento angular orbital y de spin" como:

$$\boxed{L^\mu_{\rho\sigma}:=x_\rho\bar T^\mu_\sigma-x_\sigma\bar T^\mu_\rho; \ \ S^\mu_{\rho\sigma}:=-i\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}(\hat S_{\rho\sigma}\Phi)_A}$$

La suma de estas dos corrientes da otra corriente: $\boxed{\bar J^\mu_{\rho\sigma}=L^\mu_{\rho\sigma}+S^\mu_{\rho\sigma}}$

**Proposición:** si un sistema es invariante Lorentz, entonces su corriente Noether conservada es precisamente el tensor $\bar J^\mu_{\rho\sigma}$, y su carga Noether asociada se llama "momento angular total" $J_{\rho\sigma}$. Explícitamente, esta carga toma la forma:

$$\boxed{J_{\rho\sigma} := \int_{\partial\Omega}d^3\sigma \ n_\mu \bar J^\mu_{\rho\sigma}=\int_{\mathbb R^3}d^3\vec x \ \bar J^0_{\rho\sigma}}$$

**Dem 1:**

Trivial, si la densidad Lagrangiana es exclusivamente función de cantidades invariantes, ella también será invariante

**Dem 2:**

Comprobamos ahora que, para un sistema invariante Lorentz, la corriente conservada es precisamente $\bar J^\mu_{\rho\sigma}$. En primer lugar, sabemos que los campos (que no tienen por qué ser invariantes Lorentz), tendrán variación intrínseca de la forma: $\delta\Phi_A=\frac{i}{2}\delta\omega^{\rho\sigma}(\hat J_{\rho\sigma}\Phi)_A$. De aquí extraemos que $F_{A\rho\sigma}=(\hat J_{\rho\sigma}\Phi)_A$. De este modo, construimos la primera parte de la corriente Noether:

$$\theta^\mu_{\rho\sigma}=\frac{i}{2}\frac{\partial\mathcal L}{\partial(\partial_\mu\Phi_A)}(\hat J_{\rho\sigma}\Phi)_A$$

Vamos ahora con la segunda contribución. Como admitimos que el sistema es Lorentz-invariante, la densidad Lagrangiana es un campo escalar Lorentz, y como tal no tiene spin, por lo que $\hat J_{\rho\sigma}=\hat L_{\rho\sigma}=-i(x_\rho\partial_\sigma-x_\sigma\partial_\rho)=-2ix_{[\rho}\partial_{\sigma]}$. De este modo, la variación intrínseca de la densidad Lagrangiana es:

$$\delta\mathcal L =\frac{i}{2}\delta\omega^{\rho\sigma}(\hat J_{\rho\sigma}\mathcal L)=\frac{i}{2}\delta\omega^{\rho\sigma}(\hat L_{\rho\sigma}\mathcal L)=\delta\omega^{\rho\sigma}x_{[\rho}\partial_{\sigma]}\mathcal L$$

Para hallar $f^\mu_{\rho\sigma}$, tenemos que dejar la expresión de la forma: $\delta\mathcal L = \partial_\mu f^\mu_{\rho\sigma}\delta\omega^{\rho\sigma}$. Ayuda escribirlo explícitamente: $x_\rho\partial_\sigma \mathcal L-x_\sigma\partial_\rho\mathcal L=\partial_\mu f^\mu_{\rho\sigma}$ y tenemos que hallar $f^\mu_{\rho\sigma}$. Es esperable que vaya con una delta que aproveche la regla de la cadena y que nos saque cuatro términos, dos de los cuales se anulen en la resta. Efectivamente, vemos que sale:

$$f^\mu_{\rho\sigma}=x_{[\rho}\delta^\mu_{\sigma]}\mathcal L$$

## Invariancia Poincaré: $T^{\mu\nu}$ y $J^{\mu\sigma\rho}$

**Proposición:** si el sistema es invariante Poincaré (traslaciones+Lorentz), entonces podemos aprovecharnos de la ambigüedad en la definición de la corriente Noether de las traslaciones para añadirle un término tal que la simetrice:

$$\boxed{T^{\mu\nu}=\bar T^{\mu\nu}+\partial_\rho\kappa^{\rho\mu\nu}}$$

donde $\boxed{\kappa^{\rho\mu\nu}=\frac{1}{2}(S^{\rho\mu\nu} + S^{\mu\nu\rho} - S^{\nu\rho\mu})}$. Este tensor se ha obtenido por el método de Bellinfante. Nótese la manifiesta similitud de esta expresión con la de los símbolos de Christoffel de primer orden en una conexión métrica, donde las corrientes de spin harían las veces de derivadas del tensor métrico. Aún no sé a qué se debe esta similitud.

**Nota:** $T^{\mu\nu}$ es simétrico. Por su parte, $\kappa^{\rho\mu\nu}$ es antisimétrico en sus dos primeros índices

**Proposición:** la corriente Noether asociada a la invariancia bajo el grupo de Poincaré es:

$$\boxed{J^{\mu\rho\sigma}=x^\rho T^{\mu\sigma}-x^\sigma T^{\mu\rho}}$$

Se puede comprobar que este solo difiere con $\bar J^{\mu\rho\sigma}$ en un término $\partial_\nu(\kappa^{\nu\mu\sigma}x^\rho)$. El argumento de la derivada es en esencia un tensor de rango 4 antisimétrico en los dos primeros índices, de modo que al tomar su derivada $\partial_\mu$, la cuenta se anula.

**Propiedad:** la corriente Noether de la invariancia Poincaré es antisimétrica en sus dos últimos índices: $J^{\mu\rho\sigma}=-J^{\mu\sigma\rho}$

**Proposición:** las cargas Noether asociadas a $T^{\mu\nu}$ son las mismas que las asociadas a $\bar T^{\mu\nu}$, y lo mismo aplica para las cargas de $J^{\mu\rho\sigma}$ y las de $\bar J^{\mu\rho\sigma}$

**Dem:**

Lo hacemos con la carga conservada del tensor energía-momento: los momentos $P^\nu$, y el argumento es idéntico para las cargas conservadas de la corriente de momento angular total. Si el sistema es invariante bajo traslaciones, entonces el tensor energía-momento es una corriente Noether conservada, y por ende: $\partial_\mu \bar T^{\mu\nu}=0$. Asimismo, como el sistema es por hipótesis invariante Lorentz, podemos simetrizar el tensor de energía-momento, y como este solo difiere en la derivada de un tensor antisimétrico con respecto del tensor energía-momento, tenemos que también es una corriente Noether conservada asociada a la misma simetría y por tanto: $\partial_\mu T^{\mu\nu}=0$. Asimismo, la resta de ambos también ha de ser 0:

$$\partial_\mu (T^{\mu\nu}-\bar T^{\mu\nu})=0 \Longrightarrow \int_{\partial\Omega}d^3\sigma \ n_\mu (T^{\mu\nu}-\bar T^{\mu\nu})=0$$

Pero estos dos términos son precisamente las definiciones de las cargas Noether conservadas asociadas a una y otra corriente:

$$P_\nu-\bar P_\nu=0 \Longrightarrow P_\nu = \bar P_\nu \text{$_\blacksquare$}$$

## Invariancia gauge (abeliana)

**Def.:** una transformación de gauge es aquella transformación de los campos que es **local**. Esto es, está parametrizada con funciones en lugar de parámetros reales. A su análisis le aplica el segundo Teorema de Noether. Atañe a grados de libertad internos de los campos, no a sus relaciones con el espaciotiempo

- Estudiaremos el caso de la invariancia gauge que subyace a la electrodinámica. Esta proviene de la invariancia bajo cambios de fase global en el campo. Para ello, consideraremos un campo escalar complejo, de modo que $\Phi_A=(\phi, \phi^*)$ como ya hemos tratado anteriormente en estos apuntes

## Invariancia bajo cambio de fase rígido

**Observación experimental:** los resultados experimentales no dependen de la fase global del campo considerado

**Consecuencia:** nuestro sistema debería ser invariante bajo cambios de fase globales, en particular rígidos (de fase constante). Consideremos entonces una transformación del tipo:

$$\phi'(x)=\phi(x)e^{-iq\alpha}$$

Vemos que esta transformación está descrita por un **parámetro** $\alpha$, y por tanto le corresponde todo el tratamiento del primer teorema de Noether PERO como no son transformaciones del grupo de Poincaré, en cada caso tendremos que derivar la expresión concreta de la varación intrínseca de los campos como podamos

**Proposición:** si la acción del sistema es tal que esta transformación es una simetría del sistema, hay una corriente Noether conservada. Si en particular la acción es de la forma:

$$\boxed{S=\int d^4x\left[-\partial_\mu\phi \partial_\mu \phi^* - V(|\phi|)\right]}$$

entonces la corriente Noether es:

$$\boxed{J^\mu= -iqc(\phi\partial^\mu\phi^*-\phi^*\partial^\mu\phi)}$$

y su carga Noether conservada es:

$$\boxed{Q=\frac{iq}{c}\int_{\mathbb R^3}d^3\vec x(\phi\partial_t\phi^*-\phi^*\partial_t\phi)}$$

**Dem:**

Para demostrar que la corriente Noether asociada a esta transformación es la citada, la calculamos directamente. Supongamos que el sistema en efecto sea invariante bajo transformaciones rígidas de fase global. Consideremos una transformación infinitesimal de fase de la forma: $\phi'(x)=\phi(x)e^{-iq\delta \alpha}$ e ídem para el campo complejo. Entonces, tenemos que la variación intrínseca de este campo es de la forma:

$$\delta\phi =\phi'(x)-\phi(x)=\phi(x)(1-iq\delta\alpha)-\phi(x)=-iq\phi\delta\alpha$$

Comparando con la expresión: $\delta\phi=F_\phi\delta\alpha$, encontramos que $\boxed{F_\phi=-iq\phi}$. Ídem para el campo complejo conjugado. Ya estamos en posición de calcular $\theta^\mu$:

$$\theta^\mu = -iq\frac{\partial\mathcal L}{\partial(\partial_\mu\phi)}\phi +iq\frac{\partial\mathcal L}{\partial(\partial_\mu\phi^*)}\phi^*=iq(\phi\partial^\mu\phi^*-\phi^*\partial_\mu\phi)$$

Ahora calculemos $f^\mu$. La variación intrínseca de la densidad Lagrangiana es:

$$\delta\mathcal L = \frac{\partial\mathcal L}{\partial\phi}\delta\phi+\frac{\partial \mathcal L}{\partial(\partial_\mu\phi)}\delta(\partial_\mu\phi) + \frac{\partial\mathcal L}{\partial\phi^*}\delta\phi^*+\frac{\partial \mathcal L}{\partial(\partial_\mu\phi^*)}\delta(\partial_\mu\phi^*)$$

Desarrollando, llegamos a que $\delta\mathcal L = 0$ y por tanto $f^\mu=0$. Así es que $J^\mu=\theta^\mu_\blacksquare$
