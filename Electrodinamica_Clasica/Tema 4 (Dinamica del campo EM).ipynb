# Campo EM libre

## Densidad Lagrangiana

Grados de libertad: $A_\mu$ (veremos que realmente sobran 2 de estos 4 grados)

- La única cantidad Lorentz-invarinte que podemos construir (y que no sea cuadri-divergencia de un vector) que podemos construir con las variables es $F_{\mu\nu}F^{\mu\nu}$, así que la densidad Lagrangiana ha de ser:

$$\boxed{\mathcal L = -\frac{1}{4\mu_0c}F_{\mu\nu}F^{\mu\nu}}$$

## Ecuaciones dinámicas

- Introduciendo esta densidad Lagrangiana en las ecuaciones de E-L, tenemos que el campo EM libre (en una región sin cargas) cumple:

$$\partial_\mu F^{\mu\nu}=0$$

**OJO!:** recuerda que, como $F^{\mu\nu}$ es el Riemann del fibre bundle, también debe cumplir la identidad de Bianchi o, equivalentemente, $\partial_\mu^*F^{\mu\nu}=0$. Esta ecuación es de consistencia, si F no cumple esto, no es un campo EM. Podemos romper la covariancia explícita y escribir estas ecuaciones como:

$$\begin{gather}\vec \nabla\cdot \vec E=0 \\ \vec \nabla \times \vec B=\frac{1}{c^2}\partial_t \vec E\end{gather}$$

$$\begin{gather}\vec \nabla\cdot \vec B=0 \\ \vec \nabla \times \vec E=-\partial_t \vec B\end{gather}$$

Las primeras 2 vienen de E-L, las segundas dos son de consistencia

## Tensor de energía-momento

**Observación:** la densidad Lagrangiana de esta teoría es (1) independiente de las coordenadas de forma explícita y (2) Lorentz-invariante. Por ende, podemos construir un tensor de energía-momento que será (1) corriente Noether conservada y (2) podemos simetrizarlo

$$T^{\mu\nu}_{can}=\frac{1}{c\mu_0}\left(-\frac{1}{4}F^2\eta^{\mu\nu}+F^{\mu\rho}\partial^\nu A_\rho\right)$$

Simetrizado:

$$\boxed{T^{\mu\nu}=-\frac{1}{c\mu_0}\left(\frac{1}{4}F^2\eta^{\mu\nu}+F^{\mu\rho}F_{\rho}^{\ \ \ \nu}\right)}$$

**Nota:** el orden de los índices es importante, si los intercambiamos sale un signo "-"

Por ser corriente Noether conservada, se cumple: $\boxed{\partial_\mu T^{\mu\nu}=0}$

## Interpretación de las componentes de $T^{\mu\nu}$

1. $T^{00}=\frac{1}{2c}(\varepsilon_0\vec E^2+\frac{1}{\mu_0}\vec B^2)=\frac{U}{c}$: densidad de energía
2. $T^{0i}=\frac{1}{c^2\mu_0}(\vec E\times \vec B)^i=\frac{S^i}{c^2}$: vector de Poyinting (flujo de energía en dirección espacial i-ésima)
3. $T^{ij}=-\frac{1}{c}(\varepsilon_0E^iE^j+\frac{1}{\mu_0}B^iB^j)+\frac{1}{2c}(\varepsilon_0\vec E^2+\frac{1}{\mu_0}\vec B^2)\eta^{ij}$: tensor de cizalladuras y presiones

**Nota:** véase que $T^{i0}$ no se menciona. Sería la densidad de momento lineal en la dirección i-ésima, pero como es $T^{i0}=T^{0i}$, el flujo de energía y de momento lineal son la misma cosa. Esto es lógico, porque en un fotón sabemos que ambas cantidades están relacionadas por $p=E/c$

Ahora veamos a qué ecuaciones da lugar el que $T^{\mu\nu}$ sea corriente Noether conservada

1. $\partial_\mu T^{\mu0}=0 \Longrightarrow \boxed{\partial_t U+\vec \nabla \cdot \vec S=0}$ (ecuación de continuidad para la energía)
2. $\partial_\mu T^{\mu i}=0\Longrightarrow \boxed{\partial_t S^i/c^2+c\partial_jT^{ij}=0}$ (conservación del momento lineal: el momento puede cambiar por cizalladuras o presiones)

## Ondas planas

### Ecuaciones de Helmholtz

Manipulando las ecuaciones de Maxwell en el vacío, podemos llegar a que el campo eléctrico y el magnético cumplen ecuaciones de onda:

$$\boxed{\begin{gather}\partial_t^2\vec E=c^2\vec\nabla^2\vec E \\ \partial_t^2\vec B=c^2\vec\nabla^2\vec B\end{gather}}$$

Estas ecuaciones a menudo reciben el nombre de "ecuaciones de Helmholtz" o directamente "ecuaciones de onda"

### Resolviendo ecs. de Helmholtz: ondas planas

**Proposición:** el campo $\vec E(\vec x, t)=E_0(\vec k)e^{i(\vec k\cdot \vec x-\omega t)}$ es solución de estas ecuaciones, siempre que cumpla la "relación de dispersión":

$$\boxed{\omega^2=c^2\vec k^2}$$

**Nota:** en QFT, diríamos que el fotón está on-shell, porque su energía está fijada y es la que debería ser

**Def.:** $\vec k$ es el "vector de onda", su módulo $k$ es el "número de onda" y $\omega>0$ es la frecuencia angular

**Proposición:** el campo con $-\omega$ también es solución

**Nota:** aunque empleemos campos complejos, esto no deja de ser notación, y el campo físico será siempre la parte real del campo complejo

- Introduciendo las soluciones de ondas planas en las ecuaciones de Maxwell, tenemos que además deben cumplir:

1. $\vec k\cdot \vec E=\vec k\cdot \vec B=0$
2. $\vec B=\hat k\times \vec E$

Conclusión: las ondas planas han de ser transversales y además los módulos de los dos campos están fijados

**Proposición:** el campo EM tiene 2 grados de libertad (en cada punto del espacio y del tiempo)

**Dem:** las soluciones en ondas planas vienen determinadas por 6 funciones, pero la condición (2) nos restringe 3 grados de libertad y la cond. (1) nos quita otro (el campo B ya se ha expresado en 2 en términos de E o viceversa, así que solo aplica una de las dos conds.) Esto nos deja con $6-3-1=2$ grados de libertad$_\blacksquare$

## Polarización de una onda plana

**Observación:** vemos que el campo $\vec E$ ha de ser perpendicular a $\vec k$, de modo que podemos descomponerlo en una base ortogonal a este vector (y por simplicidad cogemos la base ortonormal). De este modo, la onda plana de campo eléctrico más general vendrá dada por una descomposición arbitraria en dicha base:

$$\vec E(\vec x, t)=(E_1\vec\epsilon_1+E_2\vec\epsilon_2)e^{i(\vec k\cdot \vec x-\omega t)}$$

donde $\vec \epsilon_1\cdot \vec k=\vec\epsilon_2\cdot \vec k=0$ y donde $E_1, E_2$ son en general coeficientes complejos, de modo que:

$$E_1=|E_1|e^{i\theta_1}; \ \ E_2=|E_2|e^{i\theta_2}$$

**Def.:** llamamos "fase" del coeficiente i-ésimo a $\theta_i$

**Nota:** como el campo magnético viene completamente determinado por el eléctrico por $\vec B=\hat k\times\vec E$, nos restringimos a hablar del primero

**Def.:** se define el "vector de polarización" como el siguiente vector unitario y complejo:

$$\boxed{\vec\varepsilon=\frac{E_1\vec \epsilon_1+E_2\vec\epsilon_2}{\sqrt{|E_1|^2+|E_2|^2}}}$$

**Proposición:** si ambos coeficientes tienen igual fase, entonces el campo eléctrico apunta siempre en la misma dirección (paralelo / antiparalelo al vector de polarización) y además $\vec \varepsilon$ forma un ángulo con $\vec\epsilon_1$ tal que $\boxed{tan\theta=E_2/E_1}$

**Dem:**

De la definición del vector de polarización, tomemos el producto escalar y módulo a ambos lados con $\vec\epsilon_1$, de tal forma que nos sale $\cos\theta=\frac{|E_1|}{\sqrt{|E_1|^2+|E_2|^2}}$. Esta expresión recuerda mucho a la definición trigonométrica del coseno. En efecto, construyamos un triángulo rectángulo de base $|E_1|$ y lado $|E_2|$. Naturalmente, su hipotenusa será $\sqrt{|E_1|^2+|E_2|^2}$, de modo que el coseno del ángulo que forma la hipotenusa con la base coincidirá en esta construcción con el que tenemos nosotros. Entonces, tenemos inmediatamente que la tangente de ese mismo ángulo será $|E_2|/|E_1|$. Véase que, como los campos tienen la misma fase, no es necesario poner valores absolutos al dividirlos porque estas se cancelan$_\blacksquare$

**3 tipos de polarizaciones:**

1. Lineal: $E_1, E_2$ tienen igual fase
2. Elíptica: $E_1, E_2$ tienen distintas fases y distintos módulos
3. Circular: $E_1, E_2$ tienen igual módulo y sus fases difieren en $\pi/2$

El campo eléctrico en polarización circular es:

$$\vec E(\vec x, t)=E_0[\vec \epsilon_1\cos(\omega t-\vec k\cdot \vec x)\pm \vec \epsilon_2\sin(\omega t-\vec k\cdot \vec x)]$$

de modo que, en un punto fijo del espacio, vemos que va barriendo el plano con velocidad angular constante $\omega$

**Def.:** si el sentido de giro (con respecto de $\vec k$) es antihorario, diremos que el campo tiene **helicidad positiva**, y si es horario diremos que su helicidad es negativa. En términos de vector de polarización, $\vec\varepsilon=\frac{1}{\sqrt 2}(\vec \epsilon_1+i\vec \epsilon_2)$ es helicidad positiva y $\vec\varepsilon=\frac{1}{\sqrt 2}(\vec \epsilon_1-i\vec\epsilon_2)$ es helicidad negativa

## Densidad y flujo de energía en ondas planas

- Veamos la forma explícita del vector de Poyinting y de la densidad de energía de una onda plana EM

$$\mu_0\vec S = Re[\vec E]\times Re[\vec B]=\frac{1}{2}Re(\vec E\times \vec B)+\frac{1}{2}Re(\vec E\times \vec B^*)$$

Sin embargo, los campos EM típicamente oscilan tan rápido que no podemos medirlos en tiempo real, y lo que medimos es el vector de Poyinting promedio:

$$\boxed{\langle\vec S\rangle_t = \frac{1}{2}Re(\vec E\times\vec B^*)=\frac{1}{2\mu_0 c}(\vec E\cdot\vec E^*)\hat k}$$

Asimismo, la densidad de energía de una onda plana también se mide en promedio, y tenemos que:

$$\boxed{\langle U  \rangle_t = \frac{1}{4}(\varepsilon_0\vec E\cdot \vec E^*+\frac{1}{\mu_0}\vec B\cdot \vec B^*)=\frac{1}{2\mu_0c^2}\vec E\cdot \vec E^*}$$

# 2. Full interacting picture

- Estudiemos ahora qué pasa cuando la región del espacio y del tiempo considerara tiene cargas y corrientes presentes
- **Nota:** solo nos vamos a centrar en la dinámica del campo EM, pero si queremos la full-interacting picture, podemos considerar las ecuaciones de Maxwell en región con carga conjuntamente con las ecuaciones dinámicas de las cargas (Lorentz), y ya tendríamos la dinámica de ambos agentes acoplada


## Ecuaciones de Maxwell en presencia de fuentes

$$\boxed{\partial_\nu F^{\mu\nu}=\mu_0j^\mu; \ \ \partial_\nu^*F^{\mu\nu}=0}$$

Rompiendo la covariancia explícita (se puede demostrar que las ecs. de Maxwell son covariantes) tenemos:

$$\begin{gather}\vec \nabla\cdot \vec E=\frac{\rho}{\varepsilon_0} \\ \vec \nabla\times \vec B-\frac{1}{c^2}\partial_t\vec E=\mu_0\vec j \\ \vec \nabla\cdot \vec B=0 \\ \vec \nabla\times \vec E+\partial_t\vec B=0\end{gather}$$

Como siempre, tenemos las dos primeras ecuaciones, que provienen de E-L y las dos últimas, que son estructurales

## Tensor de energía-momento

**Proposición:** aunque el tensor energía-momento del campo EM y el de la partícula no son corrientes Noether conservadas, la suma de ambos sí, como es lógico, porque el sistema es cerrado cuando consideramos todos los agentes


$$T^{\mu\nu}_P=\frac{\mu(x)}{c\gamma}u^\mu u^\nu$$

$$T^{\mu\nu}=T_P^{\mu\nu}+T_{EM}^{\mu\nu}$$

## Conservación de la energía (total)

Exploremos ahora las consecuencias de $\partial_\mu T^{\mu\nu}=0$ rompiendo la covariancia

$$\boxed{\partial_t\mathcal E-\gamma c^2\partial_t \mu + \partial_t U + \vec \nabla\cdot \vec S=0}$$

Equivalentemente, podemos recordar que $\partial_\mu T_P^{\mu0}=j_\mu F^{0\mu}=\vec j\cdot \vec E/c^2$ y tenemos el teorema de Poyinting:

$$\boxed{\partial_t U + \vec \nabla\cdot \vec S+\vec j\cdot \vec E=0}$$

## Conservación del momento (total)

$$\boxed{\partial_t S^i/c^2+c\partial_j T_{EM}^{ij}+f^i=0}$$

donde $f^i$ es la "densidad de fuerza"

# 3. Ecuación de onda de $A_\mu$

- Revisitemos el caso de un campo EM libre, y resolvamos ahora las ecuaciones dinámicas sin romper la covariancia
- Partiendo de las ecs. de E-L para $F_{\mu\nu}$, podemos introducir la definición como cuadri-rotacional del potencial vector y nos queda una ecuación para esta cantidad, que al final es el grado de libertad de la teoría

$$\boxed{\square A^\mu-\partial^\mu(\partial_\nu A^\nu)=-\mu_0 j^\mu}$$

**Observación:** esta ecuación es gauge-invariante, lógicamente

## Sobre la fijación de gauge

**Proposición:** sea el conjunto de todos los potenciales vector (i.e: todas las soluciones de esta ecuación). Podemos construir una relación de equivalencia tal que dos potenciales sean equivalentes si inducen el mismo campo EM. Es decir, $A_1^\mu \sim A_2^\mu\Longleftrightarrow A_1^\mu = A_2^\mu+\partial^\mu \alpha$

**Observación:** la libertad de gauge en las ecuaciones nos dice que existe una ambigüedad en las mismas: podemos escoger cualquier miembro de una misma clase de equivalencia, que las ecuaciones siempre nos saldrán las mismas. Por ende, es deseable escoger un representante para cada clase de equivalencia. Sin embargo, a veces no podremos escoger un solo representante para cada clase de equivalencia, y en general tendremos un subconjunto de representantes de menor cardinal que el de la clase

**Proposición:** podemos escoger uno o varios representantes de cada clase de equivalencia imponiendo condiciones para los potenciales vector. Es decir, solo consideramos aquellos $A_\mu$ que cumplan nuestra "norma", y estos serán los representantes de su clase. A esto se le llama "fijación de gauge", porque escogemos uno concreto. Sin embargo, hay que cerciorarse de que esta fijación de gauge cubre todas las clases de equivalencia, porque si no hay clases que se quedan sin representante

**Def.:** cuando tengamos más de un representante por clase de equivalencia, diremos que existe una "libertad de gauge residual"

### Gauge de Lorenz

$$\boxed{\partial_\mu A^\mu=0}$$

**Nota:** siempre podemos imponer el gauge de Lorenz, porque dado un potencial vector que no cumpla esta condición, siempre existirá otro que sí la cumpla y que solo difiera del primero en el gradiente de una función:

$$A_{OK}^\mu=A^\mu + \partial^\mu\alpha \Longrightarrow \partial_\mu A_{OK}^\mu=0=\partial_\mu A^\mu+\square\alpha$$

Es decir, debemos encontrar una función $\alpha$ tal que sea solución de $\square\alpha=-\partial_\mu A^\mu$, pero esta ecuación siempre tiene solución y por ende el gauge de Lorenz es una fijación de gauge válida$_\blacksquare$

**Nota:** esta fijación de gauge deja cierta libertad residual, porque si $\alpha$ es solución de su ecuación, entonces $\alpha+\alpha_0$ tal que $\square\alpha_0=0$ también es solución. Por ende, esta fijación de gauge deja una libertad de gauge residual

### Otras fijaciones de gauge válidas

1. Gauge de Coulomb: $\vec\nabla\cdot \vec A=0$
2. Gauge temporal: $\varphi=0$
3. Gauge axial: $A_3=0$
