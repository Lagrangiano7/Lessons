# 1. Formalismo Hamiltoniano

## Hamiltoniano como transformada de Legendre de $\mathcal L$

**Teorema:** cuando la matriz Hessiana de un Hamiltoniano es no degenerada, entonces este proviene de un Lagrangiano (y viceversa). Decimos entonces que las dos formulaciones son equivalentes. El Hamiltonino puede calcularse entonces como la transformada de Legendre del Lagrangiano (y viceversa porque esta transformada es dual)

$$\boxed{\det\left(\frac{\partial^2H}{\partial p_\alpha \partial p_\beta}\right)\neq 0\Longleftrightarrow \det\left(\frac{\partial^2\mathcal L}{\partial \dot q^\alpha \partial \dot q^\beta}\right)\neq 0 \Longleftrightarrow H=p_\alpha \dot q^\alpha-\mathcal L}$$


## Ecuaciones de Hamilton

- El Hamiltoniano es la transformada de Legendre del Lagrangiano de un sistema
- Las ecuaciones del movimiento en este formalismo son:

$$\begin{cases}\dot q^\alpha = \frac{\partial H}{\partial p_\alpha} \\ \dot p_\alpha = -\frac{\partial H}{\partial q^\alpha}\end{cases}$$

## El espacio de fases

**Def.:** el espacio de fases es el fibrado cotangente del espacio de configuración: $T^*Q$. Este es el conjunto de todos los puntos de $Q$ y los espacios cotangentes asociados a cada uno de dichos puntos. El espacio de fases es a su vez una variedad diferenciable, y es sobre la que trabajaremos en este formalismo (esto es, nuestras variables básicas serán las q's y p's)

**Def.:** llamaremos **variable dinámica** a cualquier función definida sobre la variedad

**Proposición:** el conjunto de todas las variables dinámicas, $\mathcal F(T^*Q)$, tiene estructura de espacio vectorial

**Def.:** las trayectorias físicas serán curvas en la variedad, tales que su vector normal en cada punto esté determinado por las ecuaciones de Hamilton

## Variables canónicas

**Observación:** en el formalismo Hamiltoniano, posiciones y momentos están en igualdad de condiciones (esto quiere decir que se tratan igual desde el punto de vista de las ecuaciones de Hamilton, mientras que en las ecuaciones de E-L esto no sucedía)

**Def.:** introducimos una notación que nos permita expresar las variables de la variedad en igualdad de condiciones: $\xi^\alpha$. Estas se llamarán **variables canónicas**, y como la dimensión del espacio de fases es $2n$, hay $2n$ de ellas Escogeremos la ordenación: $\boxed{\xi^\alpha=q^\alpha; \alpha =1,\ldots,n \text{ y } \xi^\alpha=p_\alpha; \alpha=n+1,\ldots,2n}$

Con esta notación, las ecuaciones de Hamilton adquieren la forma:

$$\boxed{\dot \xi^\alpha = \Omega^{\alpha\beta}\partial_\beta H}$$

O bien:

$$\boxed{\Omega_{\alpha\beta}\dot \xi^\beta = \partial_\alpha H}$$


donde $\boxed{(\Omega)_{\alpha\beta} = \begin{pmatrix}
0_n & -I_n \\
I_n & 0_n
\end{pmatrix}}
$

O bien su inversa:

$$(\Omega)^{\alpha\beta}= \begin{pmatrix}
0_n & I_n \\
-I_n & 0_n
\end{pmatrix}$$

**Nota:** siempre contraigo con el segundo índice de la matriz simpléctica. Hay que tener cuidado porque es antisimétrica

**Nota:** podemos comprobar que, para cada valor de $\alpha$, la matriz $\Omega$ selecciona un único valor de $\partial_\beta H$, y que efectivamente recuperamos las ecuaciones de Hamilton. También podíamos haberlo visto como:

$$\begin{pmatrix}\dot Q \\ \dot P\end{pmatrix} = \begin{pmatrix}\partial H/\partial P \\ -\partial H/\partial Q\end{pmatrix} = \begin{pmatrix}
0_n & I_n \\
-I_n & 0_n
\end{pmatrix}\begin{pmatrix}\partial H/\partial Q \\ \partial H/\partial P\end{pmatrix}$$

que es la misma expresión que la recuadrada pero escrita en forma matricial.



## Matriz simpléctica ($\Omega$)

**Def.:** la matriz $\Omega$ se llama "matriz simpléctica". Es de gran importancia porque nos da las ecuaciones del movimiento (lleva metida la estructura de las ecuaciones de Hamilton).

**Propiedades:**

1. $\Omega^{\alpha\beta}\Omega_{\beta\gamma}=\delta^\alpha_\gamma$
2. Antisimétrica
3. Anti-ídempotente: $\Omega^2=-I$. En componentes: $\Omega^{\alpha\beta}=-\Omega_{\alpha\beta}$. En forma matricial: $\Omega^{-1}=-\Omega = \Omega^t$

## Trayectorias clásicas

Vemos que las trayectorias clásicas sobre el espacio de fases son las curvas integrales del campos vectorial $\Delta^\alpha_H=\Omega^{\alpha\beta}\partial_\beta H$:

$$\boxed{\frac{d\xi^\alpha}{dt}=\Delta^\alpha_H}$$

**Def.:** a este campo lo llamaremos **campo vectorial dinámico**

## Corchetes de Poisson

### Definición

**Def.:** dadas dos variables dinámicas $f, g\in \mathcal F(T^*Q)$, se define su corchete de Poisson $\{\cdot, \cdot\}: \mathcal F(T^*Q)\times \mathcal F(T^*Q)\to \mathcal F(T^*Q)$ como:

$$\boxed{\{f, g\}:=(\partial_\alpha f)\Omega^{\alpha\beta}(\partial_\beta g)}$$

En términos de q's y p's:

$$\{f, g\}=\frac{\partial f}{\partial q^\alpha}\frac{\partial g}{\partial p_\alpha}-\frac{\partial g}{\partial q^\alpha}\frac{\partial f}{\partial p_\alpha}$$

### Propiedades

1. Linealidad: $\{af +bg, h\}=a\{f, h\}+b\{g, h\}$ (e igual en el segundo argumento)
2. Antisimetría: $\{f, g\}=-\{g, f\}$
3. Identidad de Jacobi: $\{f,\{g, h\}\}+\{g,\{h, f\}\}+\{h,\{f, g\}\}=0$
4. Regla de Leibniz: $\{fg, h\}=f\{g, h\}+g\{f, h\}$
5. $\{\xi^\alpha, \xi^\beta\}=\Omega^{\alpha\beta}$

**Proposición:** el conjunto de todas las variables dinámicas con el corchete de Poisson definido es un álgebra de Lie: $\mathcal A_{\mathcal F}=(\mathcal F, \{\cdot, \cdot\})$

**Nota:** véase que el operador $\{\xi^\alpha, \cdot\}$ toma funciones y devuelve otras funciones, y además es lineal y obedece la regla de Leibniz. Por ende, es un campo vectorial. En particular, podemos considerar aquel operador que a cada variable dinámica $g$ le asigne un vector tal que:

$$X^\alpha_g=\{\xi^\alpha, g\}=\Omega^{\alpha\beta}(\partial_\beta g)$$

Que obviamente cumplirá: $X_g^\alpha(\partial_\alpha f)=\{f, g\}$

### Ecuaciones del movimiento

**Observación:** sabemos ya que las trayectorias clásicas en el espacio de fases vienen dadas por las curvas integrales del campo vectorial dinámico:

$$\frac{d\xi^\alpha}{dt}=X^\alpha_H=\Omega^{\alpha\beta}\partial_\beta H$$

Este campo vectorial se puede expresar con corchetes de Poisson como acabamos de describir: $X^\alpha_H=\{\xi^\alpha, H\}$, de modo que:

$$\frac{d\xi^\alpha}{dt}=\{\xi^\alpha, H\}$$

**Proposición:** en general, la evolución de cualquier variable dinámica se puede expresar como un corchete de Poisson:

$$\boxed{\frac{df}{dt}=\{f, H\}+\frac{\partial f}{\partial t}}$$

**Dem:**

Sea una variable dinámica $f(T^*Q, t)$ y sea $\xi^\alpha(t)$ la trayectoria clásica que sigue el sistema en el espacio de fases. Lo que nosotros observaremos serán los valores que toma la variable dinámica sobre aquellos valores de la variedad que queden encima de la curva, de modo que tendremos $f(\xi^\alpha(t))$. Entonces:

$$\frac{df}{dt}=\partial_\alpha f\frac{d\xi^\alpha}{dt}+\frac{\partial f}{\partial t}=(\partial_\alpha)\{\xi^\alpha, H\}+\frac{\partial f}{\partial t}=(\partial_\alpha) \Omega^{\alpha\beta}(\partial_\beta)+\frac{\partial f}{\partial t}=\{f, H\}+\frac{\partial f}{\partial t}_\blacksquare$$

## Teorema: evolución Hamiltoniana y corchetes de Poisson

- "Si una curva en el espacio de fases está generada por una evolución Hamiltoniana (i.e: es la curva integral de un vector dinámico) $\Longleftrightarrow$ el corchete de Poisson de cualesquiera dos variables dinámicas $f, g$ evoluciona de la forma":

$$\boxed{\frac{d\xi^\alpha}{dt}=\Delta^\alpha_H\Longleftrightarrow \frac{d}{dt}\{f, g\}=\{\dot f, g\}+\{f, \dot g\}}$$

Recuerda: $\Delta^\alpha_H=\Omega^{\alpha\beta}\partial_\beta H$

# 2. Geometría simpléctica

- Hasta ahora, hemos trabajado solo con coordenadas canónicas. Dado el Hamiltoniano en dichas coordenadas, podemos integrar la dinámica del sistema y hallar la curva que trazan las trayectorias clásicas sobre el espacio de fases
- Sin embargo, la estructura geométrica del espacio de fases ofrece mucho más, y es por eso que es interesante hacer una formulación geométrica (covariante o intrínseca) de la mecánica Hamiltoniana


## Contexto: formas diferenciales

### 1-formas

**Def.:** dado un espacio tangente $T_p (T^*Q)$, una 1-forma es un elemento del espacio dual $T^*_p(T^*Q)$. Naturalmente, $\sigma: T_pM\to \mathcal F(T^*Q)$

Se expanden en la base: $\mathbf \sigma=\sigma_\alpha d\zeta^\alpha$

**Def.:** un "campo de 1-formas" es la asingnación suave de una 1-forma a cada punto de la variedad

**Def.:** se define el operador $\iota\sigma:\chi(T^*Q)\to\mathcal F$ como aquel que sobre cada campo vectorial $X\in\chi(T^*Q)$ le hace actuar la 1-forma $\sigma$:

$$\iota_X\sigma = \sigma(X)=\sigma_\alpha X^\alpha$$

**Teorema:** una 1-forma se puede escribir como la diferencial de una función si y solo si $\sigma = df \Longleftrightarrow \partial_{[\alpha}\sigma_{\beta]}=0$

**Def.:** sea $\sigma$ un campo de 1-formas y sea $\gamma$ una curva sobre la variedad. Se define la integral de este campo a lo largo de dicha curva como:


$$\int_\gamma \mathbf{\sigma}:=\int_\gamma \sigma_\alpha d\zeta^\alpha(s)=\int_{s_0}^{s_1}\sigma_\alpha \dot \gamma^\alpha ds$$

**Nota:** esta integral no depende de la elección de coordenadas, ni tampoco de la parametrización escogida para la curva: es intrínseca.

### 2-formas

**Def.:** se define una 2-forma como una aplicación bilineal antisimétrica sobre vectores

**Base:** una base de las 2-formas es el conjunto de todos los productos tensoriales antisimétricos de la base de 1-formas: $\{d\zeta^\alpha\wedge d\zeta^\beta\}$, donde $d\zeta^\alpha\wedge d\zeta^\beta=d\zeta^{[\alpha}\otimes d\zeta^{\beta]}$, de modo que $d\zeta^\alpha\wedge d\zeta^\beta=-d\zeta^\beta\wedge d\zeta^\alpha$

**Def.:** se define el "conjunto de productos ordenados de elementos de la base" como: $\boxed{d\zeta^\alpha \dot \wedge d\zeta^\beta:=d\zeta^\alpha \wedge d\zeta^\beta}$ con $\alpha<\beta$

Tenemos entonces que podemos descomponer 2-formas en esta base como:

$$\boxed{\Omega = \Omega_{\alpha\beta}d\zeta^\alpha\dot \wedge d\zeta^\beta=\frac{1}{2}\Omega_{\alpha\beta}d\zeta^\alpha\wedge d\zeta^\beta}$$

**Nota:** también podemos escribir la forma simpléctica como $\boxed{\Omega=dp_\alpha\wedge dq^\alpha}$

**Dem:**

$$\Omega=\frac{1}{2}{\huge \omega}_{\alpha\beta}d\xi^\alpha d\xi^\beta=\frac{1}{2}{\huge \omega}_{\alpha\beta}(d\xi^\alpha \otimes d\xi^\beta-d\xi^\beta\otimes d\xi^\alpha)$$

donde hemos escogido coordenadas canónicas porque son más sencillas y además la forma canónica se reduce en ellas a la matriz canónica. Expandamos la expresión considerando que solo hay un grado de libertad, así que $\alpha,\beta=1,2$:

$$\Omega=\frac{1}{2}[\omega_{12}d\xi^1 d\xi^2+\omega_{21}d\xi^2d\xi^1-\omega_{12}d\xi^2d\xi^1-\omega_{21}d\xi^1d \xi^2]$$

Reconociendo en nuestra ordenación habitual de las coordenadas del espacio de fases que $q=\xi^1, p=\xi^2$ y recordando que $\omega_{12}=-1, \omega_{21}=+1$

$$\Omega=\omega_{12}d\xi^1 d\xi^2+\omega_{21}d\xi^2d\xi^1-\omega_{12}d\xi^2d\xi^1-\omega_{21}d\xi^1d \xi^2]$$

$$\Omega=\omega_{12}dqdp-\omega_{12}dpdq=dpdq-dqdp=dp\wedge dq$$

Es fácil entonces generalizar a un número arbitrario de grados de libertad, donde tendremos que:

$$\Omega=dp_\alpha\wedge dq^\alpha_\blacksquare$$

**Nota:** en un sistema de coordenadas cualquiera, la acción de una 2-forma sobre dos vectores se puede escribir como:

$$\Omega(X, Y)=\Omega_{\alpha\beta}X^\alpha Y^\beta$$

**Dem:**

Trabajamos en particular en variables canónicas y esta expresión será válida para cualquier otro sistema de coordenadas porque son tensores

$$\Omega(X,Y)=\frac{1}{2}\Omega_{\mu\nu}X^\alpha Y^\beta (d\xi^\mu\otimes d\xi^\nu-d\xi^\nu d\xi^\mu)X^\alpha Y^\beta$$
$$=\frac{1}{2}\Omega_{\mu\nu}X^\alpha Y^\beta (\delta^\mu_\alpha\delta^\nu_\beta-\delta^\nu_\alpha\delta^\mu_\beta)$$

Como $\Omega_{\mu\nu}$ antisimétrico, los dos términos dan lo mismo. Nos sale un factor dos, y concluimos, aplicando las deltas, que la aplicación en componentes es esa$_\blacksquare$

**Def.:** se define el operador "contracción" $\iota_X\Omega:T_\zeta^M\to T^*_\zeta M$ como aquella operación que a cada vector le asocia una 1-forma contrayendo con una 2-forma:

$$\iota_X\Omega = \Omega(X, \cdot) \longrightarrow \iota_X\Omega(Y)=\Omega(X, Y)$$

En coordenadas:

$$\boxed{(\iota_X\Omega)_\alpha = \Omega_{\alpha\beta}X^\beta}$$

**Def.:** diremos que una 2-forma es **no-degenerada** si y solo si el operador contracción es **inyectivo**. Esto es, si a cada vector se le asigna una 1-forma distinta. Equivalentemente, si y solo si $\iota_X\Omega=0 \Longleftrightarrow X=0 \Longleftrightarrow det(\Omega_{\alpha\beta})\neq 0$

**Proposición:** si $\iota_X\Omega$ es no-deg, entonces es un isomorfismo entre el espacio tangente y el cotangente en cada punto de la variedad (claro, porque es inyectivo y mapea todo el espacio imagen)

### k-formas

**Def.:** una k-forma es aquella forma k-lineal completamente antisimétrica. Todo es análogo al caso de las 2-formas. Podemos expresarlas en la base de productos ordenados:

$$\omega = \omega_{\alpha_1\ldots\alpha_k}d\zeta^{\alpha_1}\dot\wedge\ldots\dot\wedge d\zeta^{\alpha_k}=\frac{1}{k!}\omega_{\alpha_1\ldots\alpha_k}d\zeta^{\alpha_1}\wedge\ldots\wedge d\zeta^{\alpha_k}$$

**Def.:** el operador "contracción" de una k-forma se define como:

$$\iota_X\omega := \omega(X, \cdot, \ldots, \cdot)=\frac{1}{(k-1)!}\omega_{\alpha_1\ldots\alpha_k}X^{\alpha_1}d\zeta^{\alpha_2}\wedge\ldots\wedge d\zeta^{\alpha_k}$$

**Producto exterior:** sea $\omega$ una k-forma y sea $\rho$ una l-forma. Se define su producto exterior como su producto tensorial completamente antisimetrizado:

$$(\omega\wedge \rho)_{\alpha_1\ldots\alpha_k\beta_1\ldots\beta_l}=\frac{(k+l)!}{k!l!}\omega_{[\alpha_1\ldots\alpha_r}\rho_{\beta_1\ldots\beta_l]}$$

**Proposición:** el producto exterior de dos formas es otra forma, de "dimensión" $k+l$

**Contracción del producto exterior:**

$$\boxed{\iota_X(\omega\wedge \rho)=\iota_X(\omega)\wedge \rho+(-1)^k\omega\wedge \iota_X(\rho)}$$

## Contexto: derivada exterior

**Def.:** la "derivada exterior" es una aplicación que actúa sobre una k-forma para dar un (k+1)-forma que, esencialmente, es como la original pero con las componentes derivadas:

$$d\omega=\frac{1}{k!}d\sigma_{\alpha_1\ldots\alpha_k}d\zeta^{\alpha_1}\wedge\ldots\wedge d\zeta^{\alpha_k}$$

O bien:

$$\boxed{d\omega = \frac{1}{k!}\partial_\beta\omega_{\alpha_1\ldots\alpha_k}d\zeta^\beta d\zeta^{\alpha_1}\wedge\ldots\wedge d\zeta^{\alpha_k}}$$

O bien:

$$d\omega = (k+1)\partial_{[\beta}\omega_{\alpha_1\ldots\alpha_k]}d\zeta^\beta d\zeta^{\alpha_1}\dot\wedge\ldots\dot\wedge d\zeta^{\alpha_k}$$

**Regla de Leibniz "antisimetrizada":** $$\boxed{d(\omega\wedge \rho)=d\sigma\wedge \rho+(-1)^k\sigma \wedge d\rho}$$

**Def.:** diremos que una k-forma es **EXACTA** si se puede expresar como la derivada exterior de una (k-1)-forma

**Def.:** diremos que una k-forma es **CERRADA** si su derivada exterior es nula

**TEOREMA DE POINCARÉ:** una k-forma es exacta $\Longleftrightarrow$ es cerrada ($d^2=0$) (localmente)

## Contexto: derivada de Lie

**Def.:** la derivada de Lie es un operador que cuantifica cómo varía un campo tensorial a lo largo de las curvas integrales de un campo vectorial dado, $X$

**Identidad de Cartan:**

$$\boxed{\mathfrak L_X=\iota_Xd+d\iota_X}$$

**Nota:** fíjate que esta identidad solo vale para formas, porque la derivada exterior de un campo vectorial no está definida

**Nota:** para aplicar la identidad de Cartan a partir de 2-formas inclusive, es más fácil trabajar intrínsecamente para no tener que lidiar con bases antisimetrizadas. Para ello, vienen bien las siguientes expresiones:

$$\boxed{\begin{gather}
(\mathcal L_X \sigma)(Y)
= X\big(\sigma(Y)\big) - \sigma([X,Y]) \\
d\sigma(U,V)
= U\big(\sigma(V)\big) - V\big(\sigma(U)\big) - \sigma([U,V]) \\
(\mathcal L_X \Omega)(Y,Z)
= X\big(\Omega(Y,Z)\big)
- \Omega([X,Y],Z)
- \Omega(Y,[X,Z]) \\
\begin{aligned}
d\Omega(U,V,W)
&= U\big(\Omega(V,W)\big)
 + V\big(\Omega(W,U)\big)
 + W\big(\Omega(U,V)\big) \\
&\quad - \Omega([U,V],W)
 - \Omega([V,W],U)
 - \Omega([W,U],V)
\end{aligned}
\end{gather}}
$$

## La forma simpléctica

**Def.:** la "forma simpléctica" es una 2-forma definida sobre la variedad tal que:

1. Es no-degenerada ($\iota_X\Omega=0\Longleftrightarrow X=0$)
2. Es cerrada ($d\Omega=0$)
3. En las coordenadas canónicas, toma la forma: $(\Omega_{\alpha\beta})=\begin{pmatrix}0_n & -I_n \\ I_n & 0_n\end{pmatrix}$ tal que $\Omega = \frac{1}{2}\Omega_{\alpha\beta}d\xi^\alpha\wedge d\xi^\beta$

**Proposición:** por el teorema de Poincaré, como la forma canónica es cerrada, ha de poder expresarse como la derivada exterior de cierta 1-forma:

$$\boxed{\Omega = d\theta}$$

donde $\boxed{\theta = p_\alpha dq^\alpha}$ se llama **forma canónica**

## Ecuaciones de Hamilton intrínsecas

**Proposición:** las ecuaciones de Hamilton intrínsecas toman la forma:

$$\boxed{\iota_{\Delta_H}\Omega = -dH}$$



**Dem:**

- Partamos de las ecuaciones de Hamilton en coordenadas canónicas:

$$\Omega_{\alpha\beta}\dot \xi^\beta = \partial_\alpha H$$

Recordando que las trayectorias en el espacio de fases son precisamente las líneas de campo del vector dinámico: $\dot\xi^\beta=\Delta_H^\beta$, tenemos:

$$\Omega_{\alpha\beta}\Delta_H^\beta=\partial_\alpha H$$

**Observación:** el LHS es precisamente la contracción de la matriz simpléctica con el vector dinámico, expresado en la base de 1-formas asociada a las coordenadas canónicas, $\{d\xi^\alpha\}$, pero cambiado de signo. Es decir, el vector dinámico va en el segundo argumento de la forma simpléctica: $\Omega_{\alpha\beta}\Delta^\beta_H=\Omega(\cdot, \Delta_H)=-\Omega(\Delta_H, \cdot)=-\Omega_{\alpha\beta}\Delta^\alpha_H$

**Observación:** el RHS lo podemos entender como la derivada exterior de una 0-forma, que es una variable dinámica, de modo que:

$$dH=(0+1)\partial_{[\alpha]}H d\xi^\beta=\partial_\beta H d\xi^\alpha$$

Entonces, las ecuaciones de Hamilton realmente son una igualdad entre 1-formas:

$$-(\iota_{\Delta_H}\Omega)_\alpha d\xi^\alpha=\partial_\alpha d\xi^\alpha$$

En forma intrínseca:

$$-\iota_{\Delta_H}\Omega = dH_\blacksquare$$

**Teorema:** cada Hamiltoniano determina unívocamente su vector dinámico. Cada vector dinámico determina unívocamente (salvo factores aditivos independientes de las variables de la variedad) su Hamiltoniano. A esta ambigüedad en la determinación del Hamiltoniano se le asocia la ambigüedad de dónde se toma el origen de energías (para cada instante podemos tomar uno)

**Dem:**

Como $\iota_X\Omega$ es no-degenerada, es un isomorfismo entre los espacios tangente y cotangente situados en cada punto de la variedad. Entonces, a cada vector del espacio tangente en un punto dado, $\Delta_H$, le corresponde una y solo una 1-forma $dH$. Asimismo, a cada 1-forma $dH$ le corresponde un y solo un vector dinámico, $\Delta_H$. Nótese que el isomorfismo solo entiende de cosas de la variedad, de modo que podríamos añadirle al Hamiltoniano cierta cantidad que no dependa de las variables de la variedad y todo seguiría igual$_\blacksquare$

## Flujos Hamiltonianos

### Flujo de un campo vectorial

**Def.:** sea $X$ un campo vectorial definido sobre la variedad. Sea $\zeta\in T^*Q$ un punto de la variedad. Se define entonces el "flujo del campo vectorial $X$", $\varphi^X_s$, como una aplicación que al punto $\zeta$ le asigna el punto $\Phi(s)$, donde $\Phi$ es la curva integral de $X$ que pasa por $\zeta$. Un dibujito:

**Nota:** esencialmente, llamamos "flujo del campo vectorial $X$" a la aplicación que toma un objeto y lo desplaza una distancia paramétrica $s$ a lo largo de las curvas integrales de $X$ (lógicamente, parametrizadas con $s$)

**Nota:** el flujo de un campo vectorial tiene estructura de grupo uniparamétrico. En sí, a lo que llamamos "flujo" es a este grupo, lógicamente, no a la aplicación con una $s$ fija.

**GENERALIZACIÓN:** se generaliza el concepto de flujo Hamiltoniano para aplicarlo sobre tensores de la variedad. Este será aquella aplicación que avance el tensor a lo largo de las líneas integrales del campo vectorial que genere el flujo. Para un avance infinitesimal del tensor a lo largo de estas líneas, tenemos que:

$$\varphi^X_{\delta s}\mathbf T = \delta s\mathfrak L_X \mathbf T+\delta s\partial_s\mathbf T$$

$$\boxed{\therefore \varphi^X_{\delta s}=I+\delta s(\mathfrak L_X+\partial_s)}$$
**Proposición:** la variación general de un campo vectorial frente a estas transformaciones es de la forma:

$$\boxed{\bar\delta \mathbf W = \delta s \mathfrak L_X \mathbf W+\delta s \partial_s\mathbf W}$$

véase el término de transporte: $\delta x^\mu\partial_\mu \mathbf W=\delta s\partial_s\mathbf W$. La variación intrínseca no tengo tan claro de dónde sale

**Nota:** por ser el flujo un grupo de Lie uniparamétrico, podemos recuperar los elementos del grupo a partir del generador exponenciando:

$$\boxed{\varphi^X_s=e^{s(\mathfrak L_X+\partial_s)}}$$

**OJO!:** cuando queramos aplicar esta expresión en forma exponencial, debemos recordar que $e^{s\mathfrak L_X}f\neq e^{s\mathfrak L_Xf}$

### Campos vectoriales Hamiltonianos

- Como la forma simpléctica es cerrada, la contracción $\iota_X\Omega$ es un isomorfismo entre espacios tangentes y cotangentes, de modo que podemos explotarla para relacionar cada vector con un covector (en los espacios asociados a un punto de la variedad), es decir, ahora generalizamos el concepto de ecuaciones de Hamilton, donde solo cogíamos unos vectores muy concretos (los dinámicos). Ahora lo hacemos para todos los vectores.

**Def.:** dada una variable dinámica $f$, definimos su "campo vectorial Hamiltoniano" como el único campo vectorial $X_f$ que cumple:

$$\boxed{\iota_{X_f}\Omega = -df}$$

de modo que dicho campo se puede escribir como: $X_f = (\Omega^{\alpha\beta}\partial_\beta f) \partial_\alpha$

**OJO!:** no todos los campos vectoriales son Hamiltonianos, porque se debe cumplir que $\iota_X\Omega$ sea cerrada. A priori, podríamos pensar que se cumple para cualquier campo vectorial $X$ porque así hemos definido $\Omega$, pero hay algunos campos para los que $\iota_X\Omega$ no es cerrada. Esto es equivalente a que no se cumpla la condición de integrabilidad local, que es la que sostiene la equivalencia de los 3 formalismos de la Mecánica

**CRITERIO DE CAMPO HAMILTONIANO:** "un campo $X$ es Hamiltoniano si y solo si la forma $\iota_X\Omega$ es cerrada"

**Dem:**

$\Longrightarrow:$

Por hipótesis, sabemos que $X^\alpha=\Omega^{\alpha\beta}\partial_\beta f$ para alguna variable dinámica $f\in\mathcal F(T^*Q)$. Ahora podemos actuar de nuevo la forma simpléctica por la izquierda (siempre que $\Omega$ sea invertible **problema: entonces ya estoy admitiendo que es no-degenerada, que era lo que quería demostrar**)

$$\Omega_{\alpha \beta}X^\beta = \Omega_{\alpha\beta}\Omega^{\beta\sigma}\partial_\sigma f$$

Recordamos la propiedad: $\Omega^{\alpha\beta}\Omega_{\beta\sigma}=\delta^\alpha_\sigma$. En este caso, los índices están al revés en ambas, así que intercambiando índices en ambas formas salen dos signos "-" que se cancelan y queda "+". Si se quiere, bajamos y subimos los índices, aunque estrictamente no podemos hacer eso porque las variedades simplécticas no tienen definida una métrica. En todo caso, quedamos con la expresión:

$$\Omega_{\alpha \beta}X^\beta = \delta^\sigma_\alpha\partial_\sigma f=\partial_\alpha f$$

El LHS es precisamente la contracción (en el segundo argumento) de la forma simpléctica con el campo vectorial $X$, escrita en componentes: $\Omega_{\alpha \beta}X^\beta=\Omega(\cdot, X)=-\Omega(X, \cdot)$. El RHS no son más que las componentes de $df$. Entonces, tenemos en forma intrínseca que:

$$-\Omega(X, \cdot)=df$$

$\Longleftarrow:$

Ahora la hipótesis es que la contracción es una forma cerrada (exacta, por el teorema de Poincaré ambos conceptos son equivalentes, al menos localmente).

$$\iota_X\Omega \equiv \Omega(X, \cdot)=-df$$

Para algún $f\in \mathcal F(T^*Q)$. Escribiendo esta expresión en componentes de alguna carta:

$$\Omega_{\alpha\beta}X^\alpha=-\partial_\beta f \longrightarrow \Omega_{\beta\alpha}X^\alpha=\partial_\beta f$$

Actuando la forma simpléctica a ambos lados y recordando la condición de inversa:

$$\Omega^{\sigma\beta}\Omega_{\beta\alpha}X^\alpha = \Omega^{\sigma\beta}\partial_\beta f$$

$$\delta^\alpha X^\alpha=\Omega^{\sigma\beta}\partial_\beta f$$

$$X^\sigma=\Omega^{\sigma\beta}\partial_\beta f_\blacksquare$$

**Proposición:** sean $X_f, X_g$ dos campos vectoriales Hamiltonianos, entonces:

$$\boxed{X_{\{f, g\}}=-[X_f, X_g]}$$

### Flujos Hamiltonianos

**Def.:** un "flujo Hamiltoniano" es el flujo de un campo vectorial Hamiltoniano, es decir:

$$\boxed{\varphi^f\equiv \varphi^{X_f}=e^{s(\mathfrak L_{X_f}+\partial_s)}}$$

**Def.:** la variable dinámica a partir de la que sacamos el campo se llama "generatriz infinitesimal del flujo $\varphi^f$"

**Proposición:** en concreto para variables dinámicas, tenemos que la actuación del flujo Hamiltoniano de $f$ es:

$$\boxed{\varphi^f g = e^{s(\{\cdot, f\}+\partial_s)}g}$$

**Observación:** en este lenguaje intrínseco, el problema de integrar las ecuaciones del movimiento se reduce a: (1) dado el Hamiltoniano, hallar su campo vectorial Hamiltoniano correspondiente, que será el campo dinámico; (2) calcular las curvas integrales del campo dinámico, que nos darán las trayectorias clásicas sobre la variedad:

1. $X_H^\alpha\equiv \Delta^\alpha_H=\Omega^{\alpha\beta}\partial_\beta H \to \Delta^\alpha_H$
2. $\Delta^\alpha_H=\dot\zeta^\alpha \to \zeta^\alpha(t)$

**Evolución clásica de cualquier tensor:** el campo dinámico no solo nos da las trayectorias sobre la variedad, es más potente: también nos permite calcular la evolución temporal de cualquier tensor:

$$\boxed{\dot{\mathbf W} =\mathfrak L_{\Delta_H}\mathbf W+\partial_t\mathbf W}$$

**Nota:** esta expresión generaliza la expresión que teníamos para la evol. temporal de una variable dinámica $f$, que se puede entender como un escalar (tensor de rango 0)

**FLUJO HAMILTONIANO DE $H$:** en particular, podemos construir el flujo Hamiltoniano del propio Hamiltoniano, porque este también es una variable dinámica. Vemos que este nos da precisamente la evolución temporal de cualquier objeto:

$$\varphi^H_{\delta t} \mathbf W=\mathbf W+\delta t(\mathfrak L_{\Delta_H}+\partial_t)\mathbf W \Longrightarrow \dot{\mathbf W}\delta t=\delta \mathbf W$$

donde $\delta \mathbf W$ es la variación del tensor $\mathbf W$ al aplicarle el flujo Hamiltoniano del Hamiltoniano

**Nota:** el flujo Hamiltoniano del Hamiltoniano es precisamente el vector dinámico: $X_H\equiv \Delta_H$. Sabemos que las curvas integrales de dicho vector son las trayectorias clásicas del sistema. Por ende, si un tensor es invariante a lo largo de estas líneas de campo, es lo mismo que decir que su valor no se ve alterado al moverlo sobre trayectorias clásicas (i.e: dejar que el sistema evolucione en el tiempo) y por ende el tensor es una cantidad conservada

## Teorema de Noether

- Sea $f$ variable dinámica que no depende explícitamente del tiempo. Entonces:

$$\boxed{\dot f=0\Longleftrightarrow \text{ $H$ es invariante bajo el flujo de $X_f$ ($\mathfrak L_{X_f}H=0$)}}$$

**Nota:** Equivalentemente, podemos cambiar el rol de $f$ y $H$ en la invariancia:

- Sea $f$ variable dinámica que no depende explícitamente del tiempo. Entonces:

$$\boxed{\dot f=0\Longleftrightarrow \text{ $f$ es invariante bajo el flujo dinámico}}$$

**Dem:**

En el primer enunciado, tenemos que $H$ es invariante bajo el flujo generado por $f$. En particular, será invariante bajo flujos infinitesimales en el parámetro $s$ (que no es el tiempo, porque este parámetro se reserva para parametrizar el flujo del vector dinámico)

$$\varphi^f_{\delta s}H=H+\delta s(\mathfrak L_{X_f}+\partial_s)H$$

Salvo para evolución temporal, admitimos que las variables dinámicas no dependen del parámetro con el que se paramtriza el flujo, de modo que $\partial_s g=0; \forall g \in \mathcal F(T^*Q)$. Entonces, tenemos que $H$ invariante bajo el flujo generado por $f$ si y solo si $\mathfrak L_{X_f}H=\{f, H\}=\dot f=0$. Véase que si $f$ dependiera explícitamente del tiempo, esto no sería cierto, porque $\dot f=\{f, H\}+\partial_t f$.

El segundo enunciado lo he escrito yo pero es que es lo mismo, y es completamente natural una vez se entienda la proposición que se me ha ocurrido:

<span style="color: red" >**Proposición:**</span> sean $f, g\in \mathcal F(T^*Q)$ dos variables dinámicas cualesquiera. Si $f$ es invariante bajo el flujo generado por $g$, entonces $g$ es asimismo invariante bajo el flujo generado por $f$ (si no dependen explícitamente del parámetro del flujo de la otra. En concreto, si una de las dos es $H$, entonces la otra no puede depender explícitamente del tiempo)

- Esto hace que el teorema de Noether sea una equivalencia!: conservación $\Longleftrightarrow$ simetría porque $\varphi^f(H)=0\Longleftrightarrow \varphi^H(f)=\dot f =0$

**Dem:** la clave de esta demostración está en ver: (1) que las variables dinámicas no dependen del parámetro de los flujos y (2) que los corchetes de Poisson son antisimétricos. Supongamos por hipótesis que $\varphi^f g =g$. Entonces:

$$\varphi^f_{\delta s}g=g \Longrightarrow (\mathfrak L_{X_f}+\partial_s)g=0$$

pero como admitimos que $\partial_s g=0$, tenemos que:

$$\varphi^f g = g \Longleftrightarrow \mathfrak L_{X_f}g=0\Longleftrightarrow \{g, f\}=0$$

Consideremos por otro lado la acción del flujo generado por $g$ sobre $f$ y veamos qué da:

$$\varphi^g_{\delta s'}f = f+\delta s'(\mathfrak L_{X_g}+\partial_{s'})f=f+\delta s'\{f, g\}=f$$

porque ya hemos dicho que el corchete de Poisson (con el orden cambiado) es 0. Ahora entendemos que ambas variables dinámicas sean intercambiables. Como en el segundo enunciado del teorema admitimos que $\partial_t f=0$, podemos usar directamente esta proposición haciendo $g=H$ y vemos que ambas son equivalentes$_\blacksquare$

**Generalización a tensores (mía):**

- Sea $\mathbf W$ un tensor que no depende explícitamente del tiempo. Entonces:

$$\boxed{\dot{\mathbf W} = 0\Longleftrightarrow \mathfrak L_{\Delta_H}\mathbf  W =0}$$

**Conclusión:** los objetos conservados en la evolución temporal son aquellos para los que el flujo Hamiltoniano de $H$ es una simetría (claro, es que ese flujo es la evolución temporal)

**Conservación de la energía (mío):**

**Proposición:** si una variable dinámica no depende explícitamente del parámetro del flujo Hamiltoniano que induce, entonces esta variable dinámica se conserva a lo largo de su propio flujo:

$$\partial_s f =0 \Longleftrightarrow \varphi^f_sf =f$$

**Conservación de $H$:** en particular, si aplicamos esta proposición para $f=H$, tenemos que: "el Hamiltoniano de un sistema se conserva a menos que dependa explícitamente del tiempo":

$$\frac{dH}{dt}=\frac{\partial H}{\partial t}$$

**Dem.:**

Aplicando a la variable dinámica el flujo que induce: $\varphi^f_{\delta s}f = f+\delta s(\mathfrak L_{X_f}+\partial_s)f$, donde $\mathfrak L_{X_f}f=X_f(f)=\Omega^{\alpha\beta}(\partial_\beta f)\partial_\alpha f=\{f, f\}=0$. De este modo: $\varphi^f_{\delta s}f=f+\delta s\partial_s f$, así que si $f$ no depende explícitamente del parámetro de su flujo, entonces $f$ se conserva a lo largo de su propio flujo Hamiltoniano.

Para el caso particular $f=H$, tenemos entonces que: $\varphi^H_{\delta t}H=H+\delta t\partial_t H$, de modo que el Hamiltoniano se conserva en el tiempo (i.e: a lo largo de su propio flujo Hamiltoniano, que es precisamente el flujo del vector dinámico), si y solo si no depende explícitamente del tiempo$_\blacksquare$

## Variedades simplécticas (generalización del espacio de fases)

**Def.:** una "variedad simpléctica" es aquella variedad de dimensión par donde hay definida una 2-forma $\Omega$ (1) no-degenerada y (2) cerrada

- Toda la teoría de geometría simpléctica que hemos desarrollado es válida en variedades simplécticas. En particular, cualquier función $f\in \mathcal F((M,\Omega))$ puede definir la evolución Hamiltoniana de un sistema, como hemos visto:

1. Se construye el campo Hamiltoniano $X_f^\alpha=\Omega^{\alpha\beta}\partial_\beta f$
2. Se hallan las curvas integrales $\frac{d\zeta^\alpha}{dt}=X^\alpha_f$ y mediante condiciones iniciales seleccionamos una concreta. Esta será la trayectoria clásica del sistema para la función generadora de la evolución temporal $f$, que llamaremos "Hamiltoniano del sistema"

- El corchete de Poisson también se define igual

- Como es habitual en Física, la única diferencia con la teoría puramente matemática es que, de todas las variables dinámicas, existe una (el Hamiltoniano) que es privilegiada, en el sentido de que es esa la que induce lo que llamamos dinámica

### Teorema de Darboux

**Enunciado 1:** dada una variedad simpléctica $(M, \Omega)$, existe una carta de coordenadas $\xi^\alpha$ en el entorno de todo punto $p$ tal que la forma simpléctica adquiere la forma:

$$\Omega=\frac{1}{2}{\huge{\omega}}_{\alpha\beta}d\xi^\alpha d\xi^\beta$$

con ${\huge\omega}_{\alpha\beta}=\begin{pmatrix}0_n & I_n \\ -I_n & 0_n\end{pmatrix}$

**Enunciado 2:** localmente, todas las variedades simplécticas admiten coordenadas canónicas y por ende toda variedad simpléctica es localmente isomorfa al espacio de fases $T^*Q$. Como en este curso solo estudiaremos los aspectos locales de la formulación Hamiltoniana, para nosotros no existe ninguna diferencia entre una variedad simpléctica cualquiera y el espacio de fases

### Corolario: separación de la forma simpléctica

- Del teorema de Darboux se desprende que podemos escoger cualquier variable dinámica como coordenada canónica, aunque una vez escogida una las otras ya no son libres
- Escojamos una variable dinámica $f$ y digamos que es nuestra nueva variable dinámica. Entonces, debemos encontrar su conjugada canónica $g$, que cumplirá: $\{f, g\}=1$. Una vez hallada esta segunda variable dinámica, la forma simpléctica se puede separar como:

$$\Omega = dg\wedge df+\Omega|_{\mathcal L_g}$$

donde $(\mathcal L_g, \Omega|_{\mathcal L_g})$ es una variedad simpléctica de dimensión $2n-2$. Aplicando el mismo procedimiento en esta nueva variedad, podemos factorizar la forma simpléctica completamente

**Nota:** es muy importante que esta sub-variedad sea asimismo simpléctica, porque quiere decir que sigue aplicando la mecánica Hamiltoniana. Es decir, que hemos separado la dinámica del sistema en 1 grado de libertad (un par q,p) de todos los demás

## Reducción por simetría

- Es particularmente interesante escoger como variables canónicas aquellas variables dinámicas que sean conservadas, porque entonces **reducimos en 1 grado de libertad la dimensión del problema**. Esto no es nada nuevo, ya sabíamos que las primeras integrales nos quitaban un grado de libertad

**TEOREMA:** sean dos variables dinámicas conservadas $g_1, g_2$. La primera reducción se puede hacer con $g_1$ sin problema, pero en general $g_2$ dependerá de $f_1$, de modo que la condición suficiente para que podamos hacer una segunda reducción, esta vez con $g_2$, es que ambas cantidades conservadas estén "en involución" (i.e: que $g_2$ no dependa de $f_1$):

$$\boxed{\{g_1, g_2\}=0}$$

# 3. Transformaciones canónicas

## Definiciones (equivalentes)

**Def. 1:** una "transformación canónica" es toda aquella transformación en el espacio de fases que deja invariante la forma simpléctica. En lenguaje de campos, la variación general del campo de la forma simpléctica ha de ser nula bajo estas transformaciones:

$$\bar \delta \Omega=0$$

Como es un tensor, es obvio que la condición de invariancia ("el transformado es igual al original") se puede escribir como:

$$\Omega_{\gamma\delta}(\zeta')\partial_\alpha\zeta'^\gamma\partial_\beta \zeta'^\delta=\Omega_{\alpha\beta}(\zeta)$$

**Consecuencia (mía):** el determinante del Jacobiano de una transformación simpléctica ha de ser la unidad: $\det(J)=1$ (i.e: la transformación canónica es aquella que conserva el elemento de volumen en el espacio de fases)

**Def. 2:** una transformación generada por un vector $X$ es canónica si y solo si el flujo de $X$ es una simetría de la forma simpléctica:

$$\boxed{\varphi^X \text{ transf. canónica } \Longleftrightarrow \mathfrak L_X\Omega=0}$$

**Def. 3:** una transformación generada por un vector $X$ es canónica si y solo si $X$ es campo vectorial Hamiltoniano

$$\boxed{\varphi^X \text{ transf. canónica } \Longleftrightarrow \iota_X\Omega=-df}$$

**Def. 4:** una transformación es canónica si y solo si preserva los corchetes de Poisson entre las variables originales:

$$\boxed{\{\zeta^\alpha, \zeta^\beta\}_\zeta = \{\zeta^\alpha, \zeta^\beta\}_{\zeta'}}$$

**Observación:** el corchete de Poisson de las nuevas variables es el determinante de la matriz Jacobiana de la transformación

**Interpretación:** esta es la definición que mejor permite interpretar las transformaciones canónicas. El espacio de variables dinámicas es un álgebra de Lie, y el corchete de Poisson es su aplicación bilineal, de modo que el corchete de cualesquiera dos funciones ha de dar otra función. Además, como las funciones son vectores en este álgebra, el corchete de Poisson no depende de las coordenadas, en tanto que el corchete de dos vectores da otro vector, independientemente de qué coordenadas usemos después para describir las funciones. De este modo, una transformación simpléctica es aquella que, tras transformar, el corchete de dos vectores no cambia. Es decir, en este sentido es una transformación "aceptable", porque no tiene sentido que el corchete de dos vectores dependa de nada, es un hecho del álgebra. De este modo, si $\{f(\zeta), g(\zeta)\}=h(\zeta)$, al hacer la transformación estos elementos del álgebra serán $f'(\zeta'), g'(\zeta')$ y deben cumplir $\{f'(\zeta'), g'(\zeta')\}=h'(\zeta')$. En particular, si $\{f, g\}=1$, entonces tras transformar $\{f',g'\}=1$ porque la función constante no depende de las coordenadas

**Proposición:** el conjunto de todas las transformaciones canónicas junto con la operación de composición forman un grupo: el **grupo simpléctico**

**Interpretación:** las transformaciones canónicas son aquellas que relacionan un sistema mecánico (en el sentido de que se rige por las ecuaciones de Hamilton) con otro sistema mecánico. La idea es que faciliten la resolución del problema

## Corolario (útil)

Estos resultados aplican en particular a las transformaciones de coordenadas canónicas a otras coordenadas asimismo canónicas. Entonces, una transformación será canónica si y solo si (equiv.):

1. Preservan la **matriz** simpléctica
2. $\{\xi'^\alpha, \xi'^\beta\}=\Omega^{\alpha\beta}$ (matriz simpléctica)

## Función generatriz

- Una transformación general en el espacio de fases se dice completamente conocida si sabemos escribir las nuevas variables en términos de las viejas: $\zeta'=f(\zeta, t)$
- Como las transformaciones canónicas no son generales sino que tienen fuertes restricciones (deben preservar la forma simpléctica), es de esperar que necesitemos menos información para caracterizarlas

**Proposición:** una transformación canónica queda unívocamente caracterizada por una función del espacio de fases $F$, que llamaremos "función generatriz de la transformación canónica", y por el tipo de transformación. Dicha función es la solución a la ecuación:

$$\boxed{dF=p_\alpha dq^\alpha-p'_\alpha dq'^\alpha}$$

**Nota:** si conocemos la transformación canónica y queremos calcular $F$, esta está unívocamente determinada (salvo función temporal). Sin embargo, si conocemos la función generatriz $F$, esta induce varias transformaciones: una misma función generatriz induce varias transformaciones canónicas

**Dem:**

Por ser la transformación canónica, sabemos que debe preservar la forma simpléctica. Esto quiere decir que $\Omega=\Omega'$. Además, como la forma simpléctica se puede escribir como la derivada exterior de la forma canónica:

$$d\theta=d\theta' \Longrightarrow \theta-\theta'=dF$$

pero a su vez sabemos qué forma toma la forma canónica, así que: $dF=p_\alpha dº^\alpha-p_\alpha'dq'^\alpha$. Resolviendo la ecuación diferencial para las variables primadas, tenemos la transformación canónica$_\blacksquare$

## ¿Cuál es el nuevo Hamiltoniano?

- Hemos cambiado de variables: $\zeta\to \zeta'$. ¿Qué forma adquiere el Hamiltoniano del sistema en estas nuevas variables?

**Proposición:** el Hamiltoniano se transforma bajo una transformación canónica como:

$$\boxed{H'=H+\partial_t F+p_\alpha'\partial_tq'^\alpha}$$

**Nota:** para que una igualdad entre funciones sea consistente, todas deben estar expresadas en las mismas variables. De este modo, en esta igualdad podemos poner las funciones involucradas en las variables que queramos, pero todas en las mismas. En particular, si queremos el Hamiltoniano transformado expresado en las nuevas variables como es el caso típico, tendremos que poner todo el RHS en términos de las nuevas variables. En defintiva, esta igualdad nos da la variación funcional del Hamiltoniano tras la transformación, y tenemos que expresarlo en las nuevas variables para dar el Hamiltoniano nuevo, que cambiará en general en dependencia y en forma funcional como buen campo escalar

**TEOREMA:** la evolución temporal es una transformación canónica (porque en cada instante la mecánica es Hamiltoniana)

**Dem:** aplicando la def. 3 de "transformación canónica", este resultado es inmediato, porque efectivamente la forma $\iota_\Delta \Omega=-dH$ es cerrada, porque así definimos la forma simpléctica$_\blacksquare$

## 4 tipos principales de transformaciones canónicas

**Nota:** hemos visto que una función generatriz induce varias transformaciones canónicas, porque la solución de la ecuación generadora no es única. Sin embargo, si además fijamos el tipo de transformación, entonces $F$ sí que determina unívocamente la transformación canónica

### Tipo I $(q,q')$

**Def.:** una transformación canónica $(q^\alpha, p_\alpha)\to(q'^\alpha, p'_\alpha)$ es "de tipo I" si las coordenadas generalizadas $\{q^\alpha\}$ y las nuevas $\{q'^\alpha\}$ son independientes (si sirven para mapear localmente el espacio de fases)

Los viejos y nuevos momentos canónicos serán:

$$\begin{gather}p_\alpha = \frac{\partial F_1}{\partial q^\alpha} \\ p'_\alpha = -\frac{\partial F_1}{\partial q'^\alpha}\end{gather}$$


donde $F_1(q, q', t)=F(q, p(q, q', t), t)$

El Hamiltoniano se transforma entonces como:

$$\boxed{H'=H+\partial_t F_1}$$

**Ejemplo:** la transformación canónica $q'=p; p'=-q$ es de tipo I porque $q,q'$ son independientes. Por otro lado, la transformación también canónica $q'=q; p'=q+p$ no es de tipo I porque $q,q'$ no son independientes, y de hecho son iguales

**Criterio (mío):** para que una transformación canónica sea de tipo I, se debe cumplir: $det\left(\frac{\partial q'^\alpha}{\partial p_\beta}\right)\neq 0$

**Dem:**

La función generatriz de la transformación se puede expresar en cualesquiera variables (siempre que estas sean independientes). Si por ejemplo estamos en el caso 1, donde $(q,q')$ son independientes, entonces podemos expresar la función generatriz en términos de dichas variables, a saber:

$$dF=p_\alpha dq^\alpha -p_\alpha' dq'^\alpha$$

En este primer caso, tenemos suerte porque el diferencial de la función generatriz ya está expresado en términos de los diferenciales de estas variables. De este modo, identificamos: $F_1(q,q',t)=F(q,p(q,q',t),t)$. Comparando el diferencial de esta función con el de $F$, sacamos las expresiones de los momentos viejos y nuevos.
\\

Asimismo, la nueva forma del Hamiltoniano se demuestra de la siguiente manera. Partamos de la expresión general, válida para una función generatriz cualquiera. Por comodidad, adoptamos las variables viejas:

$$H'(q,p,t)=H(q,p,t)+\partial_t F(q,p,t)$$

pero sabemos que $F(q,p,t)=F_1(q,q'(q,p,t), t)$, de modo que: $\partial_t F|_{q,p}=\frac{\partial F_1}{\partial q'^\alpha}\partial_t q'^\alpha|_{q,p} + \partial_t F_1|_{q,p}$. Recordando que podemos obtener los momentos nuevos a partir de la función generatriz de tipo 1, reescribimos la expresión como: $\partial_t F|_{q,p}=-p_\alpha'\partial_t q'^\alpha + \partial_t F_1|_{q,p}$. Introduciendo en la expresión general del Hamiltoniano, los términos que van con los momentos se cancelan y queda $H'(q,p,t)=H(q,p,t)+\partial_t F_1|_{q,p} \ \blacksquare$

### Tipo II $(q,p')$

**Def.:** una transformación canónica $(q^\alpha,p_\alpha)\to(q'^\alpha, p'_\alpha)$ es "de tipo II" si las coordenadas generalizadas viejas y los nuevos momentos son independientes

Los viejos momentos y las nuevas coordenadas serán:

$$\begin{gather}p_\alpha = \frac{\partial F_2}{\partial q^\alpha} \\ q'^\alpha = \frac{\partial F_2}{\partial p'_\alpha}\end{gather}$$

donde $F_2(q,p',t)=F(q, p(q, p', t), t)+p'_\alpha q'^\alpha(q,p',t)$

El Hamiltoniano nuevo tiene la forma:

$$\boxed{H'=H+\partial_t F_2}$$

**Dem:**


Tenemos ahora que las variables independientes son $(q,p')$, así que podemos expresar el diferencial de la función generatriz en términos de dichas variables. Partimos de la definición:

$$dF=p_\alpha dq^\alpha -p_\alpha'dq'^\alpha$$

Ahora podemos completar el diferencial teniendo en cuenta que: $d(p_\alpha'q'^\alpha)=dp_\alpha' q'^\alpha + p'_\alpha dq'^\alpha$. Sustituyendo:

$$dF=p_\alpha dq^\alpha +q'^\alpha dp_\alpha -d(p_\alpha'q'^\alpha)$$

Por ende, tiene sentido definir:

$$F_2(q,p',t)=F(q,p(q,p',t),t)+p'_\alpha q'^\alpha(q,p',t)$$

Si ahora deseamos hallar las variables restantes (momentos viejos y coordenadas nuevas), podemos hacer:

$$dF_2=dF+d(p_\alpha q'^\alpha)=p_\alpha dq^\alpha +q'^\alpha dp_\alpha$$

De este modo:

$$q'^\alpha = \frac{\partial F_2}{\partial p'_\alpha};p_\alpha = \frac{\partial F_2}{\partial q^\alpha}$$

En cuanto a la forma del Hamiltoniano, partimos de:

$$H'(q,p,t)=H(q,p,t)+p_\alpha'\partial_t q'^\alpha+\partial_t F(q,p,t)$$

Ahora: $\partial_t F(q,p,t)=\frac{\partial}{\partial t}\left(F_2(q,p'(q,p,t),t)-p_\alpha'q'^\alpha\right) = \partial_t F_2|_{q,p}+\frac{\partial F_2}{\partial p'^\alpha}\partial_t p'^\alpha - \partial_t p_\alpha' q'^\alpha - p_\alpha'\partial_t q'^\alpha$. Un término se cancela con la derivada de $F_2$ y el otro cancela el término de qp del nuevo Hamiltoniano, y queda finalmente que $H'(q,p,t)=H(q,p,t)+\partial_t F_2|_{q,p} \blacksquare$

### Tipo III $(q',p)$

**Def.:** una transformación canónica es "de tipo III" si las coordenadas generalizadas nuevas y los viejos momentos son independientes

Las viejas coordenadas y los nuevos momentos serán:

$$\begin{gather}q^\alpha = -\frac{\partial F_3}{\partial p_\alpha} \\ p_\alpha'=-\frac{\partial F_3}{\partial q'^\alpha}\end{gather}$$

donde $F_3(q',p,t)=F(q(q',p,t), p, t)-p_\alpha q^\alpha(q',p,t)$

El nuevo Hamiltoniano tiene la forma:

$$H'|_{q,p}=H(q,p,t)+\partial_t(F_3|_{q,p}+q_\alpha p^\alpha)$$

### Tipo IV $(p,p')$

**Def.:** una transformación canónica es "de tipo IV" si los momentos viejos y nuevos son independientes

Las coordenadas viejas y nuevas serán:

$$\begin{gather}q^\alpha = -\frac{\partial F_4}{\partial p_\alpha} \\ q'^\alpha=\frac{\partial F_4}{\partial p'_\alpha}\end{gather}$$

donde $F_4(p,p',t)=F(q(p,p',t), p, t)-p_\alpha q^\alpha(p,p',t)+p_\alpha'q'^\alpha(p,p',t)$

El nuevo Hamiltoniano tiene la forma:

$$H'|_{q,p}=H(q,p,t)+\partial_t(F_4|_{q,p}+q_\alpha p^\alpha)$$

# Invariantes

## Elementos de volumen en una variedad

**Proposición:** dada una variedad de dimensión $2n$, la base de 2n-formas tiene un solo elemento, y en consecuencia todas las 2n-formas son proporcionales entre sí a través de alguna función sobre la variedad.

**Dem.:**

Podemos construir una base de 2n-formas como el conjunto de todos los productos antisimetrizados de la base de 1-formas, pero como siempre salen los mismos términos, estos elementos siempre son proporcionales. Es decir, la base de 2n-formas tiene un solo elemento independiente:

$$B:=\{d\zeta^1\wedge \ldots d\zeta^{2n}\}$$

Como solo hay un elemento de la base, todas las 2n-formas han de ser linealmente dependientes entre sí$_\blacksquare$

Por ende, las 2n-formas toman la expresión en coords:

$$\mathbf v=f(\zeta^\alpha)d\zeta^1\wedge \ldots \wedge d\zeta^{2n}$$

donde $f(\zeta^\alpha)$ a veces se llama "densidad de volumen / densidad Jacobiana"

**Proposición:** el resultado de aplicar una 2n-forma a 2n vectores de la variedad nos da el volumen del paralelepípedo que forman estos vectores (salvo un factor):

$$v(X_1, \ldots, X_{2n})=v\epsilon_{\alpha_1\ldots\alpha_{2n}}X^{\alpha_1}_1\ldots X^{\alpha_{2n}}_{2n}=v\det X^\beta_\alpha$$

donde $X^\beta_\alpha$ es la matriz formada poniendo cada vector en una fila

**Observación:** el volumen formado depende de la base del sistema coordenado que elijamos, porque las componentes de la 2n-forma lo hacen, y el volumen es proporcional a ellas

**Def.:** cada campo de 2n-formas define un "elemento de volumen", en tanto que el volumen de una región $U\subset M$ se calcula como:

$$V(U)=\int_U \mathbf v = \int_U f(\zeta^\alpha)d\zeta^1\wedge \ldots \wedge d\zeta^{2n}$$

**Observación:** esta definición es independiente del sistema coordenado seleccionado porque es intrínseca, solo depende de la 2n-forma elegida para computar el volumen y de la región de la variedad cuyo volumen se quiere calcular

**Def.:** se define el "elemento de volumen canónico" como el de aquella 2n-forma construida tomando $n$ veces el producto exterior de la forma simpléctica:

$$\mathbf v=\frac{1}{n!}\bigwedge_{i=1}^n\Omega=\bigwedge_{\alpha=1}^n dp_\alpha \wedge dq^\alpha$$

**Propiedad:** este elemento de volumen es invariante bajo transformaciones canónicas (porque $\Omega$ lo es)

**Propiedad:** el volumen de una región $U\subset T^*Q$ se mantiene invariante bajo transformaciones canónicas incluso si la región $U$ se ve alterada bajo estas (en punto de vista activo)

**Dem:**

En general, una transformación canónica activa deformará la región $U$ del espacio de fases, de modo que ahora será $U'=\varphi^f_sU$. El volumen de esta nueva región será:

$$V(U')=\int_{U'}\mathbf v$$

Sin embargo, esta transformación también se puede ver desde el punto de vista pasivo, como un cambio de coordenadas, de modo que:

$$V(U')=\int_{U'}\mathbf v=\int_U\varphi^f_{-s}\mathbf v=\int_U\mathbf v=V(U)_\blacksquare$$

## Teorema de Liouville (formulación de volumen)

- "El volumen de cualquier región del espacio de fases se mantiene invariante durante la evolución temporal"

**Nota 1:** en particular, el elemento de volumen se mantiene invariante, de modo que cualquier volumen finito también, porque estaremos sumando invariantes

**Nota 2:** en general, el elemento de volumen canónico se mantiene invariante bajo cualquier transformación canónica, porque está construido con la forma canónica. El teorema de Liouville así formulado particulariza este resultado para la evolución temporal, que es un tipo concreto de transformación canónica

## Distribuciones estadísticas

- Sea un sistema de $N>>1$ partículas, cada una de ellas con $n$ grados de libertad
- Nos interesa describir el comportamiento promedio del sistema, no la dinámica de cada partícula individual

**Def.:** sea la distribución de densidad $\rho$ definida en el espacio de fases, tal que a cada elemento de volumen le asocia una densidad de partículas (que se encuentran en el estado dinámico etiquetado por dicho elemento de volumen). La normalización es clara:

$$N=\int_{T^*Q}\rho \mathbf v$$

**Aplicación alternativa:** si lo que tenemos es un sistema cuyo estado dinámico no conocemos bien, podemos asignarle una densidad de probabilidad de que se encuentre en un estado dinámico dado. Esto define otra distribución en el espacio de fases, tal que:

$$1=\int_{T^*Q}\rho \mathbf v$$

### Ecuación de Liouville

Como en el caso (1) el número de partículas del sistema ha de mantenerse constante en el tiempo, y como en el caso (2) la probabilidad ni se crea ni se destruye, podemos tomar la derivada temporal y esta se iguala a 0. Como el elemento de volumen se mantiene constante en el tiempo:

$$0=\int_U\frac{d\rho}{dt}\mathbf v \Longrightarrow \frac{d\rho}{dt}=0$$

Pero como $\rho$ es una variable dinámica:

$$\frac{d\rho}{dt}=\{\rho, H\}+\partial_t\rho=0$$

De este modo, las distribuciones que satisfacen las descripciones de los sistemas dados arriba son las soluciones de la ecuación de Liouville:

$$\boxed{-\frac{\partial \rho}{\partial t}=\{\rho, H\}}$$

## Invariantes integrales de Poincaré

**Def.:** se define el p-ésimo invariante diferencial de Poincaré como el producto exterior de la forma simpléctica consigo misma $p$ veces: $\Omega^{\wedge_p}$. Véase que el 2n-ésimo invariante diferencial es precisamente el elemento de volumen canónico

**Def:.** el "primer invariante integral de Poincaré" es la superficie de una sección del espacio de fases, que viene de integrar la forma simpléctica:

$$I_1=\int_S \Omega$$

**Teorema:** podemos expresar el primer invariante integral en términos de la forma canónica a través del teorema de Stokes generalizado:

$$\boxed{J_\gamma =\oint_\gamma \theta}$$

Esta cantidad también recibe el nombre de "integral de Poincaré" y es invariante bajo transformaciones canónicas para cualquier curva cerrada definida en el espacio de fases. Por ende, en particular será una constante del movimiento

**Dem:**

$$\int_S\Omega=\int_Sd\theta = \oint_{\partial S}\theta$$

Independientemente de la superficie $S\subset T^*Q$ seleccionada, y por ende para cualquier curva $\gamma$ cerrada (y entiendo que simple)$_\blacksquare$

# Espacio de fases extendido: $T^*Q\times \mathbb R$

**Def.:** el "espacio de fases extendido" es aquella variedad construida integrando el tiempo como una coordenada más del espacio de fases. Las coordenadas serán ahora $\tilde \xi^\alpha=(q^\alpha, p_\alpha, t)$ y los vectores tendrán ahora la forma: $\tilde X=X+X^t\partial_t$. La derivada exterior también está definida: $\tilde d\omega=d\omega+\omega_t dt$

- Lógicamente, las trayectorias clásicas seguirán siendo las curvas integrales del vector dinámico, solo que ahora este tendrá además una componente para la dimensión temporal. Además, como ahora $t$ es una coordenada más, debemos parametrizar nuestras curvas con un parámetro arbitrario que llamaremos $\sigma$. Entonces, las ecuaciones de evolución temporal son:

$$\begin{cases}\mathring{\tilde \xi^\alpha}=\Delta_H^\alpha; \ \alpha=1,\ldots,2n \\ \mathring t=\Delta_H^t\end{cases}$$

## Teorema de Noether

- Sea una transformación tal que deja invariante el sistema, entonces existe una carga Noether conservada (on-shell):

$$\boxed{\delta S = \delta f|^{t_2}_{t_1}\Longrightarrow \delta Q = p_\alpha \delta q^\alpha -H\delta t-\delta f}$$

donde en general ahora la transformación ahora puedo afectar también a la coordenada temporal

**Dem:**

Sea $S=\int_{t_1}^{t_2} dt L=\int_{t_1}^{t_2} dt(p_\alpha \cdot q^\alpha -H)$. Ahora estudiemos la cómo varía la acción bajo cierta transformación infinitesimal:

$$\begin{gather}\delta S = \int \delta(dt)L+\int dt\delta L\end{gather}$$

donde:

- $\delta(dt)=(\delta t)^\cdot dt$
- $\delta(p_\alpha\dot q^\alpha -H)=\delta p_\alpha \dot q^\alpha +p_\alpha \delta(\dot q^\alpha)-\delta H$
- $\delta(\dot q^\alpha)=(\delta q^\alpha)^\cdot-\dot q^\alpha(\delta t)^\cdot$ (véase Tema 1)
- $\delta H = \frac{\partial H}{\partial q^\alpha}\delta q^\alpha +\frac{\partial H}{\partial p_\alpha}\delta p_\alpha + \frac{\partial H}{\partial t}\delta t$. Si admitimos trayectorias clásicas, podemos usar las ecuaciones de Hamilton, de modo que esta variación se convierte SOBRE TRAYECTORIAS CLÁSICAS en: $\delta H = -p_\alpha \delta q^\alpha +q^\alpha \delta p_\alpha +\dot H\delta t$

La variación de la acción SOBRE TRAYECTORIAS CLÁSICAS queda entonces de la forma:

$$\begin{gather}\delta S = \int_{t_1}^{t_2}p_\alpha (\delta q^\alpha)^\cdot+\dot p_\alpha \delta q^\alpha -H(\delta t)^\cdot -\dot H\delta t\end{gather}$$

Pero el integrando se puede escribir como una derivada temporal total. Integrando:

$$\delta S = (p_\alpha \delta q^\alpha -H\delta t)|_{t_1}^{t_2}$$

De este modo, si el sistema es invariante bajo la transformación considerada, la acción variará bajo ella como mucho en la integral de la derivada temporal total de alguna función, de modo que: $inv. \Longrightarrow \delta S = \delta f|_{t_1}^{t_2}$. De este modo, pasando todo a la izquierda:

$$0 = (p_\alpha \delta q^\alpha -H\delta t-\delta f)|_{t_1}^{t_2}$$

Como esto se ha de cumplir para cualquier intervalo temporal, tenemos que la magnitud conservada asociada a esta transformación es:

$$\delta Q = p_\alpha \delta q^\alpha -H\delta t -\delta f_\blacksquare$$

## Invariante integral de Poincaré-Cartan

**Def.:** se define la "forma de Poincaré-Cartan" en el espacio de fases extendido como:

$$\boxed{\Theta:=p_\alpha dq^\alpha -Hdt=\theta-Hdt}$$

**Teorema:** las ecuaciones de Hamilton en el espacio de fases extendido se pueden formular en términos de la forma de Poincaré-Cartan:

$$\boxed{\iota_{\tilde \Delta_H}d\tilde \Theta=0}$$

**Teorema:** sea $\tilde \gamma$ una curva cerrada en el espacio de fases extendido. Entonces, la integral de la forma de Poincaré-Cartan a lo largo de dicha curva solamente es invariante bajo el flujo Hamiltoniano (evol. clásica):

$$\boxed{I_{\tilde \gamma}=\oint_{\tilde \gamma}\Theta}$$

Pues $I_{\tilde\gamma}=I_{\tilde \gamma'}\Longleftrightarrow \tilde \gamma'=\varphi^H_\sigma \gamma$ para algún valor del parámetro $\sigma$
