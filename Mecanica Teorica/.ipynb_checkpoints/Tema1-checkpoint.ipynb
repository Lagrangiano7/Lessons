# Mecánica Lagrangiana

**Objetivos:**

- Generalizar la Mecánica de Newton introduciendo variedades diferenciables ("geometrizar" la mecánica)
- Estudiar teoremas de conservación
- Introducir la Mecánica Relativista

# 1. Mecánica con variedades diferenciables

**Objetivo:** formalizar el planteamiento de la Mecánica Clásica

**Def.:** las "coordenadas generalizadas" de un sistema, denotadas $\{q\}$, son aquellas variables que determinan completamente la posición de nuestra partícula en cada instante.

**Def.:** el "espacio de configuración" $\mathcal Q$ es una variedad diferenciable de dimensión $n$ (número de grados de libertad del sistema). Cada punto de la variedad corresponde a una configuración de las coordenadas generalizadas.

**Def.:** las "velocidades generalizadas" son las derivadas temporales de las coordenadas generalizadas: $\{\dot q\}$.

**Def.:** el "espacio de fases en velocidades" $T\mathcal Q$ es el fibrado tangente de la variedad. Recuerda que, como buen fibrado tangente, un elemento de este será el par ordenado $(q, v)$ donde $q\in \mathcal Q$, $v\in T_q\mathcal Q$. Como también es variedad diferenciable, de dimensión $2n$, trabajaremos sobre ella porque nos permite especificar completamente el estado de movimiento de una partícula en cada instante de tiempo.

**OJO!:** la trayectoria física (la que vemos) de la partícula es una curva en el espacio de configuración $\gamma(t) : \mathbb R \to \mathcal Q$. Sin embargo, la trayectoria entendida como "el estado mecánico de la partícula en cada instante de tiempo" es una curva en el espacio de fases en velocidades, porque no solo especifica las posiciones sino también las velocidades en cada instante de tiempo: $\tilde \gamma(t):\mathbb R \to T\mathcal Q$

**Def.:** llamaremos "función dinámica" a cualquier función definida sobre la variedad:

$$f: T\mathcal Q \to \mathbb R$$

# 2. Principio de acción estacionaria

**Def.:** el "Lagrangiano" es una función dinámica y, en general, dependiente de un parámetro adicional (el tiempo), de modo que:

$$\mathcal L: T\mathcal Q\times \mathbb R \to \mathbb R$$

**Propiedad:** el Lagrangiano es una cantidad escalar. Esto es, tiene un valor dado sobre cada punto de la variedad, y es intrínseco en el sentido de que dicho valor no depende de las coordenadas que estemos empleando, siempre y cuando estas definan el mismo punto $p\in T\mathcal Q$

### La acción

**Def.:** la "acción" es un funcional definido sobre las funciones de la variedad tal que:

$$S:\mathcal F\to \mathbb R$$

**Nota:** un funcional es una función cuyo dominio son funciones. La acción lo es, porque no le pasamos variables sino funciones $q(t)$ o, si se quiere, Lagrangianos. Estos últimos sí que son funciones dinámicas porque se definen sobre las variables de la variedad

Dado un Lagrangiano, se define como:

$$\boxed{S[q]=\int_{t_0}^{t_1}\mathcal L(q, \dot q, t)dt}$$

**Propiedades:**

1. Tiene dimensiones de momento angular: $[S]=[E]\cdot [t]$
2. Es local en el tiempo (el integrando solo depende del instante integrado $t$ en cada paso de integración)
3. Es invariante bajo las transformaciones de simetría del sistema (no es requisito pero es muy recomendable para ver simetrías y conservaciones)
4. Para sistemas conservativos, el Lagrangiano es $\mathcal L = T-V$ y la acción no es solo extremal, también es mínima, para las trayectorias físicas

**Teorema:** "dados un punto inicial y otro final de una trayectoria, la trayectoria intermedia que seguirá el sistema será aquella que haga la acción extremal"

$$\boxed{(q(t), \dot q(t)) \text{ física }\Longleftrightarrow \delta S =0}$$

# 3. Ecuaciones de Euler-Lagrange

Las ecuaciones que dan las ecuaciones diferenciales cuya solución es la trayectoria física vienen dadas por las ecuaciones de Euler-Lagrange:

$$\frac{\partial \mathcal L}{\partial q^\alpha}-\frac{d}{dt}\frac{\partial \mathcal L}{\partial \dot q^\alpha}=0; \ \ \alpha=1, \ldots, n$$

**Dem.:**

- Supongamos que tenemos el Lagrangiano de un sistema definido sobre la variedad y sobre el tiempo: $\mathcal L(q, \dot q, t)$. Impongamos ahora que la acción sea estacionaria sobre trayectorias clásicas. Por el principio de mínima acción, estas son las trayectorias que hacen la acción extremal. Esto quiere decir que, si obtenemos la acción integrando una trayectoria infinitesimalmente alejada de la clásica (la real) en el espacio de fases en velocidades, obtendríamos la misma acción

$$\delta S = \delta \int_{t_0}^{t_1}\mathcal L(q, \dot q, t)dt$$

- Véase que, como la variación solo se efectúa, de acuerdo al teorema, en las variables de la variedad, el tiempo no se ve afectado y el símbolo integral y de variación conmutan. Por la regla de la cadena, tenemos:

$$\int_{t_0}^{t_1}\left[\frac{\partial \mathcal L}{\partial q^\alpha}\delta q^\alpha + \frac{\partial \mathcal L}{\partial \dot q^\alpha}\delta \dot q^\alpha\right]dt=0$$

Ahora hacemos:

$$\frac{\partial \mathcal L}{\partial \dot q^\alpha}\delta \dot q^\alpha = \frac{d}{dt}\left(\frac{\partial \mathcal L}{\partial \dot q^\alpha}\delta  q^\alpha\right)-\frac{d}{dt}\left(\frac{\partial\mathcal L}{\partial\dot q^\alpha}\right)\delta q^\alpha$$

Vemos que el primer término del RHS es una derivada temporal total, de modo que queda como un término de superficie en la acción, proporcional a la variación de la trayectoria en los puntos inicial y final. No obstante, como hemos dicho que dichos puntos son fijos, tenemos que $\delta q(t_0)=\delta q(t_1)\equiv 0$ y el término de superficie se anula. Entonces, quedamos con que:

$$\int_{t_0}^{t_1}\left[\frac{\partial \mathcal L}{\partial q^\alpha}-\frac{d}{dt}\frac{\partial \mathcal L}{\partial \dot q^\alpha}\right]\delta q^\alpha dt=0$$

Y como esta igualdad se debe cumplir para cualquier variación arbitraria $\delta q^\alpha$, tenemos que lo que se debe anular es el integrando, y quedamos finalmente con que las trayectorias físicas son aquellas que satisfacen las ecuaciones de Euler-Lagrange$_\blacksquare$

**Observación:**  el principio de acción estacionaria impone una condición global y sin embargo la consecuencia que se obtiene es local

**Nota:** hay un matiz sobre las ecuaciones de Euler-Lagrange: estrictamente, el Lagrangiano es una función definida sobre la variedad (el fibrado tangente), de modo que no tiene sentido pedir su derivada temporal total (ni de ninguna función que provenda de él) porque solo entiende de las variables de la variedad y, si acaso, del tiempo como un parámetro externo (explícitamente). Sin embargo, las ecuaciones de Euler-Lagrange tratan trayectorias físicas (las reales, las que suceden) sobre la variedad, de modo que al particularizar la expresión del Lagrangiano a la curva sobre la variedad que da la trayectoria: $(q(t), \dot q(t))$, entonces ya tenemos una función solo del tiempo y podemos hablar de derivadas temporales totales. De este modo, la forma más correcta de las ecuaciones de Euler-Lagrange llevan implícita la evaluación del Lagrangiano sobre la curva del espacio de fases en velocidades que da la trayectoria física:

$$\left.\frac{\partial \mathcal L}{\partial q^\alpha}\right |_{(q(t), \dot q(t))}-\left.\frac{d}{dt}\frac{\partial \mathcal L}{\partial \dot q^\alpha}\right |_{(q(t), \dot q(t))}=0; \ \ \alpha=1, \ldots, n$$

## 3.1 Las ecuaciones de Euler-Lagrange son covariantes

**Teorema:** las ecuaciones de Euler-Lagrange son covariantes. Esto es, tienen la misma forma en cualquier sistema de coordenadas

**Dem.:**

Esto se debe a que el Lagrangiano es un escalar, de modo que:

$$\mathcal L(q, \dot q, t) = \mathcal L'(q', \dot q', t)$$

Definamos entonces un "operador diferencial de Euler-Lagrange" sobre funciones de la variedad. Si las ecuaciones de Euler-Lagrange son realmente covariantes, entonces el Lagrangiano primado debe encontrarse en el kernel del operador.

$$(\mathcal O \mathcal L')_\alpha=\frac{\partial \mathcal L'}{\partial q'^\alpha}-\frac{d}{dt}\frac{\partial \mathcal L'}{\partial \dot q'^\alpha}$$

- Primer término:

$$\frac{\partial \mathcal L'}{\partial q'^\alpha} = \frac{\partial \mathcal L}{\partial q^\beta}\frac{\partial q^\beta}{\partial q'^\alpha}+\frac{\partial \mathcal L}{\partial \dot q^\beta}\frac{\partial \dot q^\beta}{\partial q'^\alpha}$$

**Importante:** se puede demostrar que $\frac{\partial \dot q^\beta}{\partial q'^\alpha}=\frac{d}{dt}\frac{\partial  q^\beta}{\partial q'^\alpha}$. Entonces, tenemos que el primer término es:

$$\frac{\partial \mathcal L'}{\partial q'^\alpha} =  \frac{\partial \mathcal L}{\partial q^\beta}\frac{\partial q^\beta}{\partial q'^\alpha} + \frac{\partial \mathcal L}{\partial \dot q^\beta}\frac{d}{dt}\frac{\partial  q^\beta}{\partial q'^\alpha}$$

- El segundo término es más fácil, porque las coordenadas generalizadas no dependen de las velocidades, de modo que:

$$\frac{\partial \mathcal L}{\partial \dot q'^\alpha} = \frac{\partial \mathcal L}{\partial \dot q^\beta}\frac{\partial \dot q^\beta}{\partial \dot q'^\alpha}$$

Sabiendo que: $q^\beta = q^\beta(q'^\alpha, t)\Longrightarrow \dot q^\beta = \frac{\partial q^\beta}{\partial q'^\alpha}\dot q'^\alpha+\frac{\partial q^\beta}{\partial t}$ y derivando a ambos lados por la velocidad generalizada en el sistema primado, tenemos que:

$$\frac{\partial \dot q^\beta}{\partial \dot q'^\alpha} = \frac{\partial  q^\beta}{\partial  q'^\alpha}$$

Ahora derivamos con respecto del tiempo y sale un término que se va con el del otro lado, de modo que al final queda:

$$(\mathcal O\mathcal L)_\alpha = \left[\frac{\partial \mathcal L}{\partial q^\beta}-\frac{d}{dt}\frac{\partial \mathcal L}{\partial \dot qP^\beta}\right]\frac{\partial q^\beta}{\partial q'^\alpha}$$

Pero la expresión entre corchetes es 0 porque son las ecuaciones de Euler-Lagrange del sistema original, antes de transformarlo. De este modo, tenemos que:

$$\mathcal O\mathcal L'\equiv 0$$

Y por ende las ecuaciones de Euler-Lagrange son covariantes$_\blacksquare$

## 3.2 Libertad de gauge de un Lagrangiano

- Un Lagrangiano (junto con las condiciones iniciales) determina de manera unívoca la dinámica de un sistema. Sin embargo, podemos tener varios Lagrangianos que induzcan la misma dinámica

- Dos Lagrangianos inducen la misma dinámica del sistema cuando:

1. Difieren en la derivada total de una función que no dependa de las velocidades: $\mathcal L_2 = \mathcal L_1 +\frac{d}{dt}\Phi(q, t)$
2. Chiripa (son estructuralmente distintos y no se relacionan mediante ninguna transformación gauge, pero sin embargo al introducir uno y otro en las ecuaciones E-L obtenemos lo mismo)

**Dem. (1):**

Sea la acción del primer Lagrangiano $S_1$. La del segundo será:

$$S_2 = \int_{t_0}^{t_1}\mathcal L_1dt + \int_{t_0}^{t_1}\frac{d}{dt}\Phi(q, t)dt = S_1+[\Phi(q(t_1), t_1)-\Phi(q_0, t_0)]$$

Vemos entonces que:

$$S_2=S_1+cte.$$

Así que las trayectorias que hagan extremal una de las acciones también hacen extremal la otra y viceversa:

$$\delta S_1=0\Longleftrightarrow \delta S_2=0_\blacksquare$$

**Idea:** podríamos definir una clase de equivalencia de todos los Lagrangianos posibles donde la relación de equivalencia sea "inducen la misma dinámica". De este modo, estudiando la clase de equivalencia de un Lagrnagiano dado podríamos conocer todas las transformaciones gauge que lo dejan invariante (aunque habría más elementos de la clase, porque podemos tener Lagrangianos que no se relacionen por transformaciones gauge que sin embargo induzcan la misma dinámica)

# 4. Función $h$ de Jacobi

**Def.:** dado un Lagrangiano, la función $h$ de Jacobi es otra función dinámica definida como:

$$\boxed{h(q, \dot q, t)=\frac{\partial \mathcal L}{\partial \dot q^\alpha}\dot q^\alpha-\mathcal L}$$

**Propiedad:**

$$\boxed{\frac{dh}{dt}=-\frac{\partial \mathcal L}{dt}}$$

**Propiedad:** siempre que el Lagrangiano sea una función dinámica estricta (no dependa del tiempo), entonces la función de Jacobi será una constante del movimiento

**<span style="color: red">Propiedad:</span>** la función de Jacobi coincide con la energía mecánica del sistema si (1) el término cinético va solo con términos cuadráticos en velocidades generalizadas y (2) el potencial no depende de las velocidades generalizadas

**Dem.:**

Sea una partícula libre. En coordenadas cartesianas, su Lagrangiano coincide con su energía cinética:

$$\mathcal L =\frac{1}{2}m\dot{\vec x}^2$$

Empleemos ahora coordenadas generalizadas. Tenemos entonces que:

$$\dot{\vec x} = \frac{\partial \vec x}{\partial q^\alpha}\dot q^\alpha+\frac{\partial\vec x}{\partial t}$$

Introduciendo en el Lagrangiano, tenemos que:

$$\mathcal L = T = T_2+T_1+T_0$$

donde:

- $T_2 = \frac{1}{2}m\frac{\partial \vec x}{\partial q^\alpha}\cdot \frac{\partial \vec x}{\partial q^\beta}\dot q^\alpha\dot q^\beta$
- $T_1=m\frac{\partial \vec x}{\partial q^\alpha}\cdot \frac{\partial \vec x}{\partial t}\dot q^\alpha$
- $T_0=\frac{1}{2}m\frac{\partial \vec x}{\partial t}\cdot \frac{\partial \vec x}{\partial t}$

donde el subínidce indica el grado que tiene cada término en velocidades generalizadas. De este modo, la energía cinética se puede descomponer en tres contribuciones:

$$T=T_2+T_1+T_0$$

**Teorema:** la contribución $T_i$ es una función homogénea en sentido de Euler de orden $i$ en las velocidades generalizadas.

Construyendo el Lagrangiano más general que podamos:

$$\mathcal L = T-V=T_2+T_1+T_0-V(q, \dot q, t)$$

Ahora construimos la función $h$ de Jacobi para compararla con $E=T+V$

$$h= \frac{\partial (T_2+T_1+T_0)}{\partial \dot q^\alpha}\dot q^\alpha-\mathcal L$$

$$h = 2T_2+T_1-\frac{\partial V}{\partial \dot q^\alpha}\dot q^\alpha-(T_2+T_1+T_0-V)$$

$$h=T_2-T_0+V-\frac{\partial V}{\partial \dot q^\alpha}\dot q^\alpha$$

Mientras que:

$$E:=T_2+T_1+T_0+V$$

Por ende, para que $h=E$, se debe cumplir:

1. $T_1=0$
2. $T_0=0$
3. $\frac{\partial V}{\partial\dot q^\alpha}=0_\blacksquare$

**Corolario:** si se cumplen estos requisitos y además el Lagrangiano no depende explícitamente del tiempo, entonces $E=cte.$

# 5. Coordenadas cíclicas y momentos canónicos conjugados

**Def.:** se define el momento canónico conjugado de una coordenada $q$ como:

$$P_q \equiv \frac{\partial \mathcal L}{\partial \dot q}$$

- Las ecuaciones de Euler-Lagrange nos dan un mecanismo muy natural para estudiar cantidades conservadas porque van con una derivada temporal total. De este modo:

**Teorema:** si una coordenada generalizada no aparece exlícitamente en el Lagrangiano, entonces se llama "cíclica" y su momento canónico conjugado es una constante del movimiento

**OJO!:** si encontramos un momento canónico conjugado que se conserve, es mejor no introducirlo en el Lagrangiano, porque al introducirlo en las ecuaciones de Euler-Lagrange te va a dar mal. Es mejor hacerlo todo hasta el final y solo sustituir cuando tengas las ecuaciones del movimiento planteadas. Con todo, si quieres introducirlo en el Lagrangiano, puedes pasarlo de término cinético a uno potencial (efectivo) cambiándole el signo a la contribución. Si quieres estudiar el potencial efectivo sin andar cambiando signos, también puedes plantear las ecuaciones del movimiento y luego usar:


$$\vec F = -\vec\nabla U_{ef}$$

**Nota:** conceptualmente, está mal sustituir cantidades conservadas en el Lagrangiano (entendido como función dinámica), porque estas solo se conservan sobre trayectorias clásicas, y nosotros lo estaríamos imponiendo como ligadura sobre todas las trayectorias (es decir, sobre el espacio de fases en velocidades), dando lugar a un error conceptual y a resultados asimismo erroneos. Lo correcto es introducir ligaduras con multiplicadores de Lagrange en las ecuaciones del movimiento o calcular las ecuaciones del movimiento, que son válidas por definición solo para trayectorias clásicas, y entonces ya sustituir la ligadura, que como también es válida solo para trayectorias clásicas, ya no habrá problema

# 6. Teorema de Noether

## Para transformaciones puntuales continuas

**Def.:** una "transformación puntual" es aquella transformación de las coordenadas generalizadas (por ejemplo, pasar de cartesianas a polares, aunque en general pueden depender del tiempo como parámetro externo)

$$\boxed{q^\alpha \to q'^\alpha(q, t)}$$

**Teorema:** si un Lagrangiano es invariante bajo una cierta transformación puntual, entonces se conserva la carga:


$$\boxed{\delta Q=\frac{\partial\mathcal L}{\partial \dot q^\alpha}\delta q^\alpha=P_\alpha\delta q^\alpha}$$

**Dem.:**

Al someter a un Lagrangiano a una transformación puntual (continua), tenemos que:

$$\delta \mathcal L = \frac{\partial \mathcal L}{\partial q^\alpha}\delta q^\alpha + \frac{\partial \mathcal L}{\partial \dot q^\alpha}\delta\dot q^\alpha$$

Reescribiendo:

$$\delta \mathcal L = \left[\frac{\partial \mathcal L}{\partial q^\alpha}-\frac{d}{dt}\frac{\partial \mathcal L}{\partial \dot q^\alpha}\right]\delta q^\alpha -\frac{d}{dt}\left(\frac{\partial \mathcal L}{\partial \dot q^\alpha}\delta q^\alpha\right)$$

Pero el corchete es 0 porque las trayectorias físicas cumplen las ecuaciones de Euler-Lagrange. Entonces, tenemos:

$$\delta\mathcal L=0 \Longleftrightarrow \frac{d}{dt}\left(\frac{\partial \mathcal L}{\partial \dot q^\alpha}\delta q^\alpha\right)=0_\blacksquare$$

**Nota:** fíjate que en general, para trayectorias off-shell, la variación del Lagrangiano es no trivial. Es decir, el Lagrangiano nuevo no induce la misma dinámica que el viejo. Sin embargo, como difieren en una derivada temporal total + un término que se anula on-shell, el Lagrangiano on-shell tras la variación sí que induce la misma dinámica. Por eso las cantidades conservadas lo son solo para trayectorias on-shell

**Generalización:** si un Lagrangiano es invariante bajo una cierta transformación puntual continua, entonces todos los Lagrangianos equivalentes tienen también una carga conservada tal que:

$$\mathcal L_2 = \mathcal L_1 +\frac{d}{dt}\Phi(q, t) \Longrightarrow \boxed{\delta Q_2=P_\alpha \delta q^\alpha -\delta \Phi=cte.}$$

## Para transformaciones más generales

Consideramos ahora aquellas transformaciones que dependan también del tiempo. Esto es, variamos el tiempo y variamos las coordenadas de la variedad. De este modo:

$$q \to q'(q, t) = q+\delta q(q, t); \ \ t\to t'(q, t)=t+\delta t(q, t)$$

Hasta ahora, como las transformaciones no dependían del tiempo  (eran puntuales, por hipótesis) podíamos evaluar las transformaciones en el Lagrangiano. Ahora, como estamos admitiendo que las transformaciones pueden asimismo alterar el tiempo, debemos considerar la invariancia en la acción. De este modo:

$$\delta S=\int_{t_0}^{t_1}(\delta \mathcal L dt+\mathcal L\delta(dt))$$

- Por un lado:

$$t'-t=\delta(t) \Longrightarrow \delta(dt)=dt'-dt=d(t'-t)=d(\delta t)=\frac{d}{dt}(\delta t)dt$$

- Por otro lado:

$$\delta \mathcal L = \frac{\partial \mathcal L}{\partial q^\alpha}\delta q^\alpha + \frac{\partial \mathcal L}{\partial \dot q^\alpha}\delta \dot q^\alpha+\frac{\partial \mathcal L}{\partial t}\delta t$$

Donde tenemos ahora ya la variación y la derivada temporal no conmutan en general, porque también estamos alterando el tiempo. De este modo, tenemos que $\delta \dot q^\alpha=\frac{d}{dt}(\delta q^\alpha)-\dot q^\alpha\frac{d}{dt}\delta t$

Introduciendo ambos términos en la integral de la acción y tras algunas manipulaciones, tenemos que:

$$\delta S = \int_{t_0}^{t_1}dt\left(E-L\right)(\delta q^\alpha-\dot q^\alpha\delta t) + \left.\left[(\mathcal L-P_\alpha\delta q^\alpha)\delta t+P_\alpha\delta q^\alpha\right]\right |_{t_0}^{t_1}$$

Por ende, la carga conservada es:

$$\boxed{\delta Q = \left[(\mathcal L-P_\alpha\dot q^\alpha)\delta t+P_\alpha\delta q^\alpha\right] - \delta \Phi}$$

De forma más compacta, y considerando N partículas (caso más general):

$$\boxed{\delta Q = \left[-h\delta t+P_\alpha\delta q^\alpha\right] - \delta \Phi}$$

**Observaciones:**

1. Si no permitimos las variaciones temporales, recuperamos la carga conservada para la invariancia del sistema bajo ese tipo de transformaciones
2. Metemos el término $\delta \Phi$ porque dos Lagrangianos son equivalentes si difieren en la derivada temporal total de cierta función $\Phi(q^\alpha, t)$, de modo que la acción podrá en general ser igual a dicha función evaluada en los bordes del tiempo y seguir siendo invariante

## Corolario: invariancia bajo reparametrizaciones temporales:

- Vamos a estudiar qué condiciones cumple el Lagrangiano del sistema si pedimos que la acción sea invariante bajo reparametrizaciones temporales. Esto es relevante porque en relatividad especial pediremos que la acción cumple esta propiedad

- Admitiendo que las coordenadas generalizadas no sufren ninguna transformación al reparametrizar el tiempo, encontramos que las condición que debe cumplir el Lagrangiano para que la acción permanezca invariable bajo reparametrizaciones temporales es:

$$h\equiv 0$$

Porque (1) la transformación es puramente temporal (no afecta a las coordenadas de la variedad) y (2) $\delta Q = h\delta t(t) = cte. \ \ \forall t \Longrightarrow h\equiv 0_\blacksquare$


**OJO!:** el concepto de energía depende del parámetro temporal que escojamos. Por ejemplo, la acción

$$S=C\int_{\sigma_0}^{\sigma_1}d\sigma \sqrt{ \overset \circ x^2 }$$

donde $\overset \circ x=\frac{dx}{d\sigma}$ y $\sigma$ es un parámetro temporal arbitrario, entonces $E=0$ porque la acción es invariante bajo reparametrizaciones temporales. No obstante, si escogemos un parámetro temporal que rompe esta invariancia, entonces ya tiene sentido hablar de energía. Por ejemplo, si tomamos $dt=C\frac{d\sigma}{m\overset\circ x^2}$ entonces:


$$S=\int_{t_0}^{t_1}dt\dot x^2$$

y ya tenemos $E=\frac{1}{2}m\dot x^2$. Es decir, como ya hemos escogido como parámetro temporal uno concreto (i.e: el tiempo, que en mecánica clásica es absoluto), ya se rompe esta invarincia bajo reparametrizaciones temporales y vuelve a tener sentido el concepto de energía, porque estamos en un inercial dado, donde la podemos medir

### Corolario: conservación de la energía

- Si en concreto el sistema es invariante bajo traslaciones temporales, tenemos que no varían las variables de la variedad y que $\delta t=cte.$, de modo que:

$$\mathcal L-P_\alpha\dot q^\alpha\equiv h=cte.$$

**Más fácil:** el lío de todo esto ha sido que las transformaciones temporales eran arbitrarias, y que además considerábamos transformaciones tanto sobre el tiempo como sobre las variables de la variedad. Sin embargo, si consideramos únicamente traslaciones temporales, dejando inalterada la variedad, entonces:


$$\mathcal L \to \mathcal L'(q, \dot q, t+\delta t); \ \ \delta t=cte.$$

Entonces, la variación del Lagrangiano será:

$$\delta L = \mathcal L'-\mathcal L = \frac{\partial\mathcal L}{\partial t}\delta t$$

Si el sistema es invariante bajo traslaciones temporales:

$$\delta \mathcal L = 0\Longleftrightarrow \frac{\partial\mathcal L}{\partial t}=0 \Longleftrightarrow \frac{dh}{dt}=0_\blacksquare$$

**Nota:** véase que ahora he trabajado con el Lagrangiano. Podría haber trabajado con la acción y me saldría lo mismo, porque el diferencial $dt\to dt'=d(t+\delta t)=dt$ y todo queda igual. Alternativamente, vemos que $\delta$ y $d/dt$ conmutan porque $\delta t=cte.$, de modo que el desarrollo anterior, más general, se reduce al nuestro de ahora para traslaciones temporales

**Interpretación:** la energía de un sistema se conserva si la dinámica de este no depende del instante inicial considerado (porque como $\delta t=cte.$, en el fondo al hacer la transformación temporal $t'=t+\delta t$ lo que estamos haciendo es fijar otro origen de tiempos)

# 7. Formulación intrínseca de la Mecánica Clásica

- Podemos formular la Mecánica Clásica como una teoría independiente de la carta escogida. Esto es, una formulación intrínseca

- La estructura es la misma: tenemos la variedad diferenciable $\mathcal TQ$ que es el espacio de fases en velocidades. Sobre cada punto de esta variedad tenemos un espacio tangente $T_qM$ donde viven los vectores, que son formas diferenciables del tipo:

$$X=X_1^\alpha \frac{\partial}{\partial q^\alpha}+X_2^\alpha \frac{\partial}{\partial\dot q^\alpha}$$

Asimismo, podemos definir un espacio cotangente $T_q^*M$ y en general un espacio tangente de tensores del tipo $(r, s)$

- La evolución temporal de una función dinámica vendrá dada por la derivada de Lie de dicha función sobre el flujo del campo vectorial $\Delta_L$, donde:

$$\Delta_L=\dot q^\alpha\frac{\partial}{\partial q^\alpha} + \ddot q^\alpha \frac{\partial}{\partial\dot q^\alpha}$$

De tal modo que:

$$\boxed{\frac{d}{dt}f(q^\alpha, \dot q^\alpha)=\mathfrak L_{\Delta_L}f}$$

**Nota:** si queremos que la función dependa explícitamente del tiempo, podemos hacerlo, pero el tiempo siempre será un parámetro externo, no una variable de nuestra variedad

$$\frac{df}{dt}(q, \dot q, t)=\Delta_L(f)+\frac{\partial f}{\partial t}$$

## Forma intrínseca de las ecuaciones de Euler-Lagrange

- Definamos el campo de 1-formas:

$$\Theta_L=\frac{\partial\mathcal L}{\partial \dot q^\alpha}dq^\alpha$$

Entonces, las ecuaciones de Euler-Lagrange se pueden escribir como la derivada de Lie de este campo con respecto al flujo del campo vectorial $\Delta_L$

$$\boxed{\mathfrak L_{\Delta_L}\Theta_L=d\mathcal L}$$

**Dem.:**

Calculemos primero el RHS, que es un campo de 1-formas:

$$d\mathcal L=\frac{\partial L}{\partial q^\alpha}dq^\alpha+\frac{\partial L}{\partial \dot q^\alpha}d\dot q^\alpha$$

Ahora calculemos el LHS por partes con cuidado:

Sabemos que una derivada de Lie actúa sobre cierto campo de 1-formas, en componentes, de la forma:

$$(\mathfrak L_{\Delta_L}\Theta_L)_\alpha = \Delta \Theta_\alpha + \Theta_\beta \partial_\alpha\Delta^\beta$$

Como $\Theta_\alpha$ solo tiene componentes no-nulos en los elementos de la base correspondientes a las coordenadas generalizadas $\{dq^\alpha\}$, podemos hacer este cálculo bastante fácilmente separando los elementos de la base en aquellos correspondientes a coords. generalizadas y aquellos correspondientes a velocidades generalizadas.

**Parte de coords. generalizadas:**

$$(\mathfrak L_\Delta\Theta)_{q^\alpha}=\Delta \Theta_{q^\alpha}+\Theta_\beta\frac{\partial \Delta^\beta}{\partial q^\alpha}$$

El primer término es $\Delta \Theta_{q^\alpha}=\dot P_\alpha$. En el segundo término, veamos que $\Theta_\beta=0$ a menos que los betas corran en coords. generalizadas, de modo que solo nos queda en la derivada que se contrae con este término $\Theta^{q^\beta}=\dot q^\beta$. Como velocidades y coords. generalizadas son variables distintas de la variedad, tenemos que: $\frac{\partial \Delta^{q^\beta}}{\partial q^\alpha}=\frac{\partial \dot q^\beta}{\partial q^\alpha}=0$ y por ende ya tenemos la primera mitad de componentes de la derivada de Lie:

$$(\mathfrak L_\Delta\Theta)_{q^\alpha}=\dot P_\alpha$$

Ahora vamos con la parte en velocidades generalizadas:

$$(\mathfrak L_\Delta \Theta)_{\dot q^\alpha}=\Delta \cdot 0 + \Theta_{q^\beta}\frac{\partial \Delta^{q^\beta}}{\partial \dot q^\alpha}$$

y como $\frac{\dot q^\beta}{\partial \dot q^\alpha}=\delta^\alpha_\beta$, tenemos que:

$$(\mathfrak L_\Delta \Theta)_{\dot q^\alpha} = P_\alpha$$

Comparemos ahora RHS con LHS:

$$\mathfrak L_\Delta \Theta = dL \Longleftrightarrow \dot P_\alpha dq^\alpha + P_\alpha d\dot q^\alpha=\frac{\partial \mathcal L}{\partial q^\alpha}dq^\alpha + \frac{\partial \mathcal L}{\partial\dot q^\alpha}d\dot q^\alpha$$

Donde, evidentemente, la igualdad ha de cumplirse término a término en los elementos de la base del fibrado cotangente. Entonces, tenemos que la igualdad en los elementos $\{d\dot q^\alpha\}$ es trivial, mientras que la igualdad en los elementos $\{d q^\alpha\}$ indica la igualdad:

$$\dot P_\alpha=\frac{\partial \mathcal L}{\partial q^\alpha}$$

Y desarrollando los momentos canónicos conjugados de las coordenadas $\{q^\alpha\}$ y pasándolo todo al mismo lado, tenemos:

$$\frac{d}{dt}\frac{\partial\mathcal L}{\partial \dot q^\alpha}-\frac{\partial\mathcal L}{\partial q^\alpha}=0_\blacksquare$$

**Meter Teorema de Noether en formulación intrínseca**

# 8. Relatividad Especial

- Quiero definir una acción para partículas relativistas (en general, para un sistema de $N$ partículas)
- Si tuviera una sola, podría definir la acción como:

$$S=\int_{\tau_0}^{\tau_1}d\tau \mathcal L(x^\mu(\tau), \dot x^\mu(\tau), \tau)$$

donde hemos escogido parametrizar la trayectoria con el tiempo propio de la partícula porque $\tau$ es Lorentz-invariante

- ¿Pero qué pasa si tenemos $N$ partículas? No tiene sentido decidir que $\tau$ sea el tiempo propio de una de ellas y describir el movimiento del resto de partículas desde este inercial (fíjate que quiero tener un solo parámetro temporal)
- **Solución:** pido que la acción sea invariante bajo reparametrizaciones temporales: de este modo da igual qué parámetro temporal use para integrar el Lagrangiano
- Con esto acabo la lista de propiedades que debe cumplir la acción relativista:

1. Lorentz-invariante (en clásica, como el tiempo es absoluto,  "Galileo-invariante" ya cubre cond. 2)
2. Invariante bajo reparametrizaciones temporales
3. En el límite $\beta<<1$ debo recuperar la acción clásica

**Nota:** el motivo por el que entiendo yo que realmente pedimos que la acción sea invariante bajo reparametrizaciones temporales es que, en relatividad especial, cada inercial tendrá su tiempo, y lógicamente un observador en dicho inercial parametrizará la trayectoria de una partícula con ese tiempo. Sin embargo, como el principio de relatividad dicta que las leyes de la Física han de ser las mismas en todo inercial, el parámetro con el que escojamos parametrizar la trayectoria de una partícula no puede afectar a esta. En el límite no-relativista, esto no es problema porque el tiempo es absoluto, de modo que todos los observadores escogen el mismo parámetro. 

## Lagrangiano de partícula libre (covariante)

- Invariancia Lorentz: la dependencia en velocidades ha de ser escalar (depende de módulo)
- Invariancia bajo reparamtrizaciones temporales: ha de ser lineal en velocidades para que la regla de la cadena del parámetro temporal se cancele al reparametrizarlo
- Conclusión: dependencia como $\sqrt{-\mathring{x}^\mu\mathring{x}_\mu}$


Concretamente, el Lagrangiano de una partícula libre es:

$$\mathcal L=-mc \sqrt{-\mathring{x}^\mu\mathring{x}_\mu}$$

**Observación:** la acción de este sistema es invariante bajo reparametrizaciones temporales, porque el parámetro temporal empleado $\sigma$ es arbitrario:

$$\boxed{S=-mc\int d\sigma \sqrt{-\mathring{x}^\mu\mathring{x}_\mu}} $$

Vemos que es invariante bajo reparametrizaciones temporales. Lo comprobamos explícitamente pasando del parámetro $\sigma$ a, por ejemplo, el tiempo propio de la partícula, $\tau$:

$$\tilde S=-mc\int_{\sigma_1}^{\sigma_2} d\sigma \sqrt{-\dot x^\mu \frac{d\tau}{d\sigma}\dot x_\mu \frac{d\tau}{d\sigma}}$$

Y sacando la regla de la cadena de la raíz:

$$\tilde S=-mc \int_{\tau_1}^{\tau_2} d\tau \sqrt{-\dot x^\mu \dot x_\mu}$$

Y queda igual en forma. Véase que esto es, si se quiere, en la variedad de Minkowski, antes de imponer que las velocidades han de ser físicas y por ende están sujetas a ligaduras adicionales, como que $\dot x^\mu \dot x_\mu=-c^2$. Estrictamente, esta ligadura habría que meterla con un multiplicador de Lagrange porque la acción solo considera trayectorias físicas, pero Garay dice que este acercamiento es inadecuado pero que propicia los resultados correctos para partículas libres y partículas acopladas a campos EM

**Observación:** escoger un parámetro temporal dado rompe la invariancia bajo reparametrizaciones temporales.

### Tiempo propio

- Escojamos por ejemplo como parámetro temporal el tiempo propio de la partícula: $\sigma=\tau$. Meterlo directamente en la acción es incorrecto, porque el que las cuadrivelocidades sean curvas de velocidad unidad es una ligadura, y como tal hay que introducirlo con un multiplicador de Lagrange. Sin embargo, para la partícula libre y para una sometida a un campo EM el resultado sale correcto. Lo mejor es sacar las ecuaciones de E-L con parámetro temporal arbitrario, y como entonces ya estamos on-shell por definición de E-L, entonces ya podemos escoger tiempo propio

$$\boxed{S=-mc^2\int_{\tau_0}^{\tau_1}d\tau}$$

**Observación:** en geometría diferencial, definimos las geodésicas como "aquellas curvas entre dos puntos de la variedad tales que su longitud de arco es extremal". Esto es, son la generalización de las rectas en la geometría euclídea, en el sentido de que buscamos una curva que nos dé el camino más corto entre dos puntos de entre todas las curvas que los conectan. Para ello, basta con minimizar:

$\delta l = 0 \Longrightarrow \delta\int ||\dot x(s)||ds \Longrightarrow \delta\int ds=0$, donde en el último paso hemos usado la propiedad de las curvas de velocidad unidad, que es una propiedad que surge cuando las parametrizamos con su parámetro natural: la longitud de arco. Asimismo, vemos que la longitud de arco de una trayectoria en el espacio-tiempo de Minkowski está estrechamente relacionada con el tiempo propio de la partícula:

$$d\tau = \frac{1}{c}\sqrt{-ds^2}$$

en la signatura $(-+++)$. Por tanto, estamos haciendo lo mismo!

**Observación:** en relatividad especial, la trayectoria que sigue una partícula libre es aquella que haga extremal su "longitud de arco", como en geometría diferencial. Es decir, **las partículas siguen geodésicas en la variedad de Minkowski**. Pedir que este continúe siendo el caso para una métrica arbitraria nos da la relatividad general.

**Razonamiento:** postulo: las partículas siguen geodésicas en la variedad. Una vez tengo eso, el estudio de la dinámica de una partícula en una variedad con métrica $g_{\mu\nu}$ se reduce al estudio de las geodésicas de dicha variedad.

**Observación:** también podíamos haber escrito esta acción como:

$$S=\int_{\tau_0}^{\tau_1}\frac{1}{2}m\dot x^2d\tau$$

Que formalmente recuerda a la acción de una partícula libre clásica, donde el Lagrangiano coincide con el término cinético. Efectivamente, como $\dot x^2=-c^2$, tenemos que esta acción se reduce a la expresión anterior

### Tiempo de cierto inercial $(t)$

- Vemos que este parámetro temporal tampoco es arbitrario sino que es bien conocido, porque es una variable de la variedad de Minkowski. Tenemos que:

$$x^\mu = (ct, \vec x); \ \ v^\mu = \frac{d}{dt}x^\mu = (c, \vec v)$$

- Como ya hemos demostrado que la acción como la hemos construido es invariante bajo reparametrizaciones temporales, tenemos que:

$$S=-mc\int_{t_0}^{t_1} dt\sqrt{-v^\mu v_\mu}$$

Introduciendo la expresión de la velocidad de laboratorio, tenemos finalmente que la acción expresada en la variable tiempo de laboratorio para una partícula libre es:

$$\boxed{S=-mc^2\int_{t_0}^{t_1}dt\sqrt{1-\frac{\vec v^2}{c^2}}}$$

- Alternativamente, podíamos haber sacado esta expresión recordando que $dt = d\tau \gamma$ y sustituyendo en la expresión de la acción expresada en términos del tiempo propio. Con todo, vemos claramente que estas acciones no son invariantes bajo reparametrizaciones temporales, porque incluso si podemos pasar de una a la otra, no conservan su forma como lo hacía la acción de parámetro temporal arbitrario. Sin embargo, en esa acción la energía era idénticamente 0 porque el tiempo no está bien definido, mientras que con estas parametrizaciones temporales la energía es no-nula porque ya se ha roto esa invariancia bajo reparametrizaciones temporales

- Concretamente, vemos que la acción con esta última acción recupera la acción clásica de una partícula libre

**Observación:** una vez hemos aprendido a expresar la acción de una partícula desde un inercial que no esté relacionado con ella, podemos calcular la acción de un sistema de $N$ partículas como siempre:

$$S=\displaystyle\sum S_n=\displaystyle\sum_n^N -m_nc^2\int_{t_0}^{t_1}dt\sqrt{1-\frac{\vec v_n^2}{c^2}}$$

# 9. Partículas cargadas acopladas a campos EM

**Objetivo:** estudiar la dinámica de una partícula relativista sometida a cierto campo EM. No estudiamos la dinámica del campo por la contribución de la carga de la partícula a este. Ponemos el campo de fondo y solo estudiamos cómo se mueve la partícula en su seno (si se quiere, admitimos que los campos generados por la partícula son despreciables en intensidad frente a los campos de fondo)

## Acción del sistema

- Además de las 3 propiedades que hemos dicho que debe cumplir la acción por ser relativista, debe además cumplir invariancia bajo la transformación gauge:

$$A_\mu \to A'_\mu = A_\mu + \partial_\mu f$$

Porque esta es una simetría de las ecuaciones de Maxwell, y por ende la acción debe reflejarla (si no me cambian los campos, no me cambia la Física, y si no me cambia la Física no me puede cambiar la acción, porque si no este cambio daría lugar a que otras trayectorias pasen a ser las físicas)

**Objetivo:** construir una magnitud (la acción) invariante Lorentz, invariante bajo esta transformación gauge

**Lo que tengo:** 

- $\mathring x^\mu d\sigma$
- $A_\mu$
- $\eta{\mu\nu}$
- $\epsilon^{\mu\nu\rho\sigma}$

**Las cantidades invariantes Lorentz que puedo hacer con ellas:**

- $A^2$
- $\mathring x^2(d\sigma)^2$
- $d\sigma \mathring x^\mu A_\mu$

Con esto, debo construir una cantidad que cumpla las 3 invariancias y la propiedad de recuperación del límite clásico que he pedido. Entonces, ha de estar construida con estas magnitudes invariantes (invariancia Lorentz) y además ser homogénea en las velocidades (invariancia bajo reparam. temporales). Hasta ahora, el razonamiento es el mismo que para la partícula libre, con la única diferencia de que ahora tenemos más cantidades invariantes disponibles con las que jugar. La expresión más general de la acción con estas 2 de 4 propiedades es:


$$S=\int d\sigma\sqrt{-\mathring x^2}F(A^2, \mathring x^\mu A_\mu/\sqrt{-\mathring x^2})$$

donde $F(x, y)$ función arbitraria. La invariancia gauge elimina todas las dependencias de la función (porque no son invariantes gauge) salvo dos:

$$\int d\sigma \sqrt{-\mathring x^2}, \ \int d\sigma\mathring x^\mu A_\mu$$

La primera es trivial porque no depende del potencial. La segunda también es invariante gauge porque al hacer la transformación $A_\mu\to A_\mu+\partial_\mu f$, nos queda la integral de un término que es derivada temporal total y por ende contribuye un término constante a la acción:

$$\int d\sigma \mathring x^\mu \partial_\mu f = \int d\sigma \frac{df}{d\sigma}$$

Por tanto, la acción será:

$$\boxed{S=\int_{\sigma_0}^{\sigma_1}d\sigma \left[-mc\sqrt{-\mathring x^2}+q\mathring x^\mu A_\mu\right]}$$

**Observación:** hemos introducido un factor de escala $q$ en el segundo término. Este se asocia con la carga de la partícula, y determina con qué intensidad (y signo) se acopla esta al campo EM

**Nota:** si tenemos $N$ partículas y admitimos que no interactúan entre sí a través de sus campos, entonces la acción es aditiva y cada una de ellas contribuye un término así a la acción. Todas las acciones parametrizadas con el parámetro temporal. Tampoco pasaría nada si elegimos un parámetro temporal distinto para cada partícula, porque como las partículas "no se ven", la suma acabaría dando lo mismo (si se quiere, cada una tiene asociado su propio espacio de fases)

**Recuperando la acción clásica:**

- Es interesante pararse un momento a sacar la contrapartida clásica de la acción de este sistema, porque de ella podemos sacar la versión no-relativista de estas ecuaciones. En vez de evaluar el límite clásico en cualquier parte posterior del desarrollo, considero más beneficioso evaluarlo aquí, porque podemos hacer un desarrollo paralelo (E-L, ecs. del movimiento, cantidades conservadas, etc.)
- Escogiendo el tiempo $t$ en un inercial dado y desarrollando las expresiones:

$$S=\int_{t_0}^{t_1} dt\left[-mc^2\sqrt{1-\frac{\dot{\vec {x}}^2}{c^2}}+q(c(-\phi/c) + \vec v\cdot \vec A )\right]$$

Ahora hacemos Taylor a primer orden en $v/c$. Esto es todo lo que necesitamos para recuperar el límite clásico (véase que $c$ desaparece de nuestras ecuaciones cuando hacemos esto). Entonces, tenemos:

$$\boxed{S_{clasico}=\int_{t_0}^{t_1}dt\left[\frac{1}{2}m\vec v^2+q(-\phi+\vec v\cdot \vec A)\right]}$$


**Recuperando la Ley de Lorentz**

Encontramos que el Lagrangiano de una partícula no-relativista acoplada a cierto campo EM es:

$$\mathcal L= \frac{1}{2}mx_i^2 -q\phi(x_\alpha, t) + q\dot x_iA_i(x_\alpha, t)$$

donde se ha admitido que los campos típicamente solo dependen de la posición y tal vez del tiempo, pero desde luego no de la velocidad de la partícula porque admitimos que son de fondo. Introduciendo en E-L (conviene trabajar por componentes):


$$\frac{\partial \mathcal L}{\partial x_j}-\frac{d}{dt}\left(\frac{\mathcal L}{\partial x_j}\right)=0$$

Basta con hacerlo para una de las tres componentes del espacio, $x_j$, porque la expresión en índices que obtenemos se puede pasar fácilmente a notación vectorial (y por tanto tridimensional) haciendo las identificaciones correspondientes:

$$-e\partial_j\phi +e\dot x^i\partial_j A_i -\frac{d}{dt}(m\dot x_j+eA_j)=0$$

Desarrollando:

$$m\ddot x_j=-e\partial_j\phi -e\partial_t A_j +e\dot x^i(\partial_j A_i-\partial_i A_j)$$

**Comentario:** $\partial_i A_j-\partial_j A_i=\epsilon_{ijk}B^k$

$$m\ddot x_j=-e\partial_j\phi -e\partial_t A_j +e\dot x^i(\partial_j A_i-\partial_i A_j)$$

Y haciendo las identificaciones correspondientes entre potenciales y campos eléctrico y magnético, concluimos lo que ya sabíamos: que una partícula clásica acoplada a un campo EM (sin interactuar con él) está sujeta a la fuerza de Lorentz (no relativista):

$$m\ddot{\vec x}=q\left(\vec E+\dot{\vec x}\times \vec A\right)$$

## Ecuaciones del movimiento

- Ya hemos hallado la acción del sistema como la cantidad más general que cumple las 4 propiedades que necesitábamos
- Introduzcamos el Lagrangiano que se lee de esta acción en las ecuaciones de Euler-Lagrange (recuerda, variables de la variedad son $x^\mu$, $\mathring x^\mu$) y tenemos, en términos del parámetro temporal arbitrario $\sigma$:

$$mc\frac{d}{d\sigma}\frac{\mathring x_\mu}{\sqrt{-\mathring x^2}}=q\mathring x^\nu F_{\mu\nu}$$

donde $F_{\mu\nu}=\partial_\mu A_\nu - \partial_\nu A_\mu$ es el tensor de Maxwell

- Si escogemos ahora como parámetro temporal concreto el tiempo propio del sistema, tenemos que las ecuaciones del movimiento son:

$$\boxed{m\ddot x_\mu = q\dot x^\nu F_{\mu \nu}}$$

- Si por otro lado escogemos el tiempo en algún inercial concreto, $t$, entonces tenemos:

$$\boxed{\begin{cases}\frac{d}{dt}(m\gamma\vec v)=q(\vec E+\vec v\times \vec  B) \\ \frac{d}{dt}(m\gamma c^2)=q\vec v\cdot \vec E\end{cases}}$$

**OJO!:** No son independientes, la segunda sale de multiplicar $\vec v$ en la primera

**Interpretación:** (1) la derivada del momento cinético relativista es igual a la fuerza de Lorentz. (2) Solo el campo eléctrico desarrolla trabajo sobre el sistema (i.e: le bombea energía), alterando la energía mecánica relativista de la partícula

- Estas ecuaciones también se pueden escribir en una sola, de la forma:

$$\frac{d}{dt}(m\gamma c^2+q\phi)=q(\partial_t\phi -\vec v\cdot \partial_t \vec A)$$

**Corolario:** para una partícula cargada acoplada a un campo EM, si el cuadri-potencial no depende explícitamente del tiempo $\partial_0 A^\mu=0$ (campos estáticos), entonces la energía del sistema, entendida como contribución mecánica + electrostática, se conserva

## Cantidades conservadas

- Ya hemos planteado la acción como magnitud fundamental de nuestra teoría, como la magnitud más general que cumple todas las propiedades que hemos pedido
- A partir de ella hemos leído el Lagrangiano y lo hemos introducido en las ecuaciones de Euler-Lagrange para sacar las ecuaciones del movimiento de una partícula cargada relativista sometida a un campo EM
- Estudiemos ahora la formulación en mecánica relativista del teorema de Noether

- Por el teorema de Noether sabemos que, si la acción es invariante bajo transformaciones arbitrarias en tiempo y coordenadas, entonces:

$$\boxed{\delta Q = p_\mu\delta  x^\mu- E\delta \sigma=cte. }$$

**Observación 1:** si la acción es invariante bajo reparametrizaciones temporales (que lo es, tal y como la hemos construido, antes de elegir un parámetro temporal concreto) entonces la energía del sistema es $E\equiv 0$

**Observación 2:** el cuadrimomento es una cantidad independiente de la parametrización temporal:

$$p_\mu =\frac{\partial \mathcal L^{(\sigma)}}{\partial \mathring x^\mu}$$

**Dem.:**

Escojamos otro parámetro temporal, por ejemplo $t$. Como el Lagrangiano es homogéneo en las velocidades, tenemos que:

$$\mathcal L \to \mathcal L\frac{dt}{d\sigma}$$

Asimismo, tenemos que $\mathring x^\mu \to \frac{dx^\mu}{dt}\frac{dt}{d\sigma}$

Poniendo esto en la expresión del cuadrimomento:

$$p_\mu = \frac{\partial \mathcal L\frac{dt}{d\sigma}}{\partial x'^\mu \frac{dt}{d\sigma}}=\frac{\partial \mathcal L^{(t)}}{\partial x'^\mu}_\blacksquare$$

### Cuadrimomento $p_\mu$

**Observación:** para una partícula acoplada a un campo EM, el cuadrimomento es:

$$\boxed{p_\mu = mc\frac{\mathring x^\mu}{\sqrt{-\mathring x^2}}+qA_\mu=m\dot x^\mu + qA_\mu}$$

- Por el teorema de Noether, si hacemos una traslación espacio-temporal al sistema de tal forma que la acción se mantenga invariante, entonces la proyección del **cuadrimomento** en la dirección de la traslación se conserva:

$$\delta Q = p_\mu n^\mu = cte.$$

**Observación:** si el vector director de la traslación espaciotemporal es de tipo espacio ($|\vec n|^2>(n^0)^2$) ("apunta más en las direcciones espaciales que en la temporal") entonces existe un inercial en el que $n^0=0$. En dicho inercial, lo que se conserve será el momento lineal (relativista) de la partícula. De aquí tenemos que, para una partícula cargada vista desde ese inercial, tendremos:

$$\hat n\cdot \vec p = \hat n\cdot (m\gamma\vec v+q\vec A)=cte.$$

**Fíjate:** el momento no es el cinético, porque el cuadrimomento se define a partir del Lagrangiano! Es un momento conjugado canónico, y como tal no tiene por qué coincidir con el momento lineal cinético! (y por ende la parte espacial del cuadrivector tampoco)

**Observación:** si el vector director de la traslación espaciotemporal es en cambio de tipo tiempo, entonces existirá algún inercial en el que lo que se conserve sea la energía:

$$cp^0 = E = m\gamma c^2 + q\phi=cte.$$

**¿Qué condición debe cumplir $S$ para ser invariante bajo traslaciones espacio-temporales?:**

Sometamos a la acción a cierta perturbación cte. en las coordenadas generalizadas, $\delta x^\nu=n^\nu$, de modo que:

$$\delta S= q\int d\sigma \mathring x^\mu \frac{\partial A_\mu}{\partial x^\nu}\delta x^\nu=0 \Longleftrightarrow \frac{\partial A_\mu}{\partial x^\nu}n^\nu=0$$

**Conclusión:** si el cuadripotencial es invariante bajo traslaciones en la dirección $n^\nu$, entonces la proyección del cuadrimomento sobre dicha dirección se conserva

### Cuadrimomento angular 

- Intuición: en mecánica clásica, el momento angular orbital se define como $\vec L=\vec x\times \vec p$, de modo que cabría esperar que ahora lo definamos de la forma $L^{\mu \nu}=x^\mu p^\nu - x^\nu p^\mu$. Esta intuición es correcta, pero veamos por qué.
- Sea la transformación de Lorentz infinitesimal:

$$\Lambda^\mu_\nu = \delta^\mu_\nu +\delta\omega^\mu_\nu$$

Entonces, la variación de las coordenadas será de la forma:

$$\delta x^\mu = x^\mu - \Lambda^\mu_\nu x^\nu = -\delta\omega^\mu_\nu x^\nu$$

Por el teorema de Noether, si la acción es invariante bajo transformaciones de este tipo, entonces la carga conservada correspondiente es:

$$\delta Q = p_\mu \delta x^\mu = -p_\mu \delta \omega^\mu_\nu x^\nu = -p^\mu x^\nu \delta\omega_{\mu\nu}$$

Donde se puede demostrar que $\delta\omega_{\mu\nu}$ es antisimétrico. Entonces, podemos separar esta expresión como:

$$\delta Q = \frac{1}{2}(x^\mu p^\nu-x^\nu p^\mu)\delta\omega_{\mu\nu}=L^{\mu\nu}\delta\omega_{\mu\nu}$$

**Observación:** las transformaciones de Lorentz $\Lambda^\mu_\nu$ pertenecen al grupo propio-ortocrono, de modo que no solo engloban rotaciones en el espacio sino también boosts de Lorentz. Se puede demostrar que podemos separar el tensor cuadrimomento angular (que tiene 6 componentes independientes) en dos contribuciones de trivectores que separen explícitamente la parte de rotaciones y la parte de boosts, de modo que:

$$\delta Q = \delta\vec \theta\cdot \vec J+\frac{\delta\vec v}{c}\vec K $$

donde $\vec J$ es el generador de las rotaciones (momento angular orbital) y $\vec K$ es el generador de los boosts de Lorentz

La relación entre los componentes de los trivectores y del tensor de cuadrimomento angular es:

$$\boxed{J^i=\frac{1}{2}\epsilon^{ijk}J_{jk}; \ \ \delta \theta^i = \frac{1}{2}\epsilon^{ijk}\delta\omega_{jk}}$$

$$\boxed{K_i = J_{i0}; \ \ \frac{\delta v^i}{c}=\delta\omega_{0i}}$$

**Conclusión:** si el sistema (la acción) es invariante bajo rotaciones en el espacio en torno a un eje dado, entonces se conserva el (tri)momento angular en la dirección del eje de rotación. Si el sistema es invariante bajo boosts de Lorentz en una dirección dada, entonces se conserva la proyección del generador de boosts de Lorentz en dicha dirección

**Nota:** sabemos que $\vec J$ es el momento angular orbital, ¿pero qué es $\vec K$? En mecánica no relativista, tenemos el generador de boosts galileanos ("vector de centro de masas de Newton-Wigner"), que es:

$$\vec G =m\vec x-t\vec p$$

Véase que es constante si no actúan fuerzas sobre el sistema. En este caso, el vector $\vec K$ no es más que la generalización relativista de esta magnitud

**Corolario:** para una carga acoplada a un campo EM, la acción es automáticamente invariante bajo boosts de Lorentz. Si además $A_\mu$ no depende de cierta coordenada angular, la componente de $\delta\vec\theta\cdot \vec J$ es asimismo una integral del movimiento

### Centro de inercia

- Sea un sistema de N partículas, de tal forma que en el instante $t$ se encuentren en las posiciones $\{x_n\}$. Entonces, el centro de inercia del sistema se define de forma análoga al centro de masas:

$$\boxed{X^i=\displaystyle\sum_n^N x^i\frac{p_0^n}{P_0}}$$

donde $p^0_n = E_n/c$ y $P^0=\displaystyle\sum p_n^0$

**Observación:** esto es análogo al CM en mecánica no relativista, solo que ahora el peso ya no es la masa sino la energía. En particular, si en un inercial dado todas las partículas se encuentran en reposo, entonces centro de masas y centro de inercia coinciden

**Observación:** esta expresión no es covariante, y por ende el CI será distinto en distintos inerciales

**Def.:** llamaremos "inercial del CI" al inercial propio del CI, es decir, al que cumpla $P^i=0$. 

**Conservaciones:**

Podemos escribir la posición del CI en términos de los generadores de boosts de Lorentz:

$$X^i=\frac{-K^i}{P^0}+x^0\frac{P^i}{P^0}$$

Si el sistema es invariante bajo boosts de Lorentz, entonces $K^i=cte.$ Si además es invariante bajo traslaciones temporales, entonces $P^0=cte.$ y $X^i$ solo depende del cuadrimomento total del sistema. Además, los generadores de boosts de Lorentz se pueden interpretar como las posiciones iniciales del centro de inercia:

$$Y^i=\frac{K^i}{P^0}=cte.$$

Si además el sistema es invariante bajo traslaciones espaciales, entonces $P^i=cte.$ y el CI se mueve con cuadrivelocidad constante $\frac{P^i}{P^0}$

## Invariantes de Casimir

- Hemos encontrado magnitudes que se conservan dadas las simetrías del sistema. Sin embargo, no son invariantes (cte. en todo inercial pero con distintos valores de la cte.)
- Hemos encontrado magnitudes invariantes en todos los inerciales, pero que sin embargo no son cantidades conservadas
- Podemos preguntarnos si existen cantidades conservadas de un sistema que además sean invariantes: valen lo mismo en todo inercial para todo tiempo

**Para el grupo de Poincaré, hay 2 (indeps.):**

1. $P^2$ (módulo del momento lineal total)
2. $W^2$ (módulo del vector de Pauli-Lyubarskii)

- Módulo del momento lineal es invariante bajo traslaciones y además por ser escalar también lo es bajo transformaciones de Lorentz:

$$p_\mu = \frac{\partial\mathcal L}{\partial \mathring x^\mu}$$

Porque si el Lagrangiano es invariante bajo traslaciones, entonces el momento lineal automáticamente también

Su módulo nos da una definición apropiada de la masa del sistema. Vayámonos al inercial más fácil para calcular cuánto vale: al del CI, porque $P^i=0$:

$$P^2 = -m^2c^2$$

**Interpretación:** podemos entender la masa en el contexto de la relatividad especial como la mínima energía que puede adoptar un sistema. En cualquier otro inercial, su energía será mayor. Véase que no cuenta cuando decimos que $E=0$ porque en ese caso no estamos en ningún inercial concreto, habida cuenta de que estamos dejando el parámetro temporal $\sigma$ arbitrario. Cuando colapsamos a un inercial concreto, entonces ahí ya $E\geq mc^2$

- Vector de Pauli Lyubarskii: $W^\mu = \frac{1}{2}\epsilon^{\mu\nu\rho\sigma}P_\nu J_{\rho\sigma}$

Hagamos el mismo truco para ver cuánto vale su módulo. En el inercial de CI, tenemos $P^i=0$, de modo que:

$$(W_{CI})^\mu=\frac{1}{2}\epsilon^{\mu0\rho\sigma}(P_{CI})_0(J_{CI})_{\rho\sigma}$$

De aquí leemos que:

$$(W_{CI})^0=0; \ \ (W_{CI})^i = \frac{1}{2}(P_{CI})^0\epsilon^{ijk}(J_{CI})_{jk}$$

Y por ende su módulo será:

$$\boxed{W^2=m^2c^2\vec S^2}$$

donde $S_{CI}^i=\frac{1}{2}\epsilon_{ijk}(J_{CI})^{jk}$, llamado "spin" del sistema, es el momento angular de este en cualquier inercial del CI. Es una magnitud intrínseca porque solo depende de la configuración de momentos y posiciones relativos de las partículas que componen el sistema.


**Propiedad:** $P\cdot W=0$ (ortogonalidad)

**Propiedad:** si además el sistema tiene $m=0$, entonces el módulo de $P^\mu$ y de $W^\mu$ y por tanto son proporcionales. La constante de proporcionalidad se llama "helicidad": $\boxed{W^\mu = -hP^\mu}$ con $\boxed{h=\frac{\vec P\cdot \vec S}{P^0}}$

**Observación:** si solo hay 1 partícula puntual clásica, entonces el spin es nulo (porque el momento angular en su centro de inercia es nulo). Sin embargo, en mecánica cuántica tenemos que las partíiculas pueden tener spin no-nulo incluso cuando sean puntuales. Entonces, el vector de Pauli-Lyubarskii ya no será nulo
