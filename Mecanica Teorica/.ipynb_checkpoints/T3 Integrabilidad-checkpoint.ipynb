# Teoría de Hamilton-Jacobi

## Formulación

**Idea:** podemos coger como coordenadas del espacio de fases las cantidades que queramos, siempre que lo mapeen. En particular, podemos coger como coordenadas las constantes de movimiento del sistema (i.e: las $2n$ variables cíclicas). Entonces, en este nuevo sistema coordenado, la dinámica sería trivial de integrar, porque todo sería constante. Es decir, tendríamos $H'(q', p',t)\equiv 0$. Para tener el problema resuelto en las coordendas originales, bastaría con deshacer la transformación. Por tanto, el problema consiste en encontrar aquella transformación canónica tal que el nuevo Hamiltoniano sea nulo

En particular, escogemos transformaciones de tipo I, y llamaremos $S(q, q', t)$ a la función generatriz. Como $H'=H+\partial_t F_1$, tendremos que:

$$\boxed{H(q, \partial_q S, t)+\partial_t S=0}$$

Esta es la ecuación central de la teoría, y es una ecuación para la generatriz de la transformación canónica, $S$. Véase que hemos escrito $p=\partial_q S$ como sabemos de la teoría de estas transformaciones. Una vez la encontremos, la transformación está completamente identificada y habremos resuelto el problema.

Hasta aquí llega el desarrollo de la idea, el resto de la teoría de H-J son formas en las que podemos encontrar soluciones aceptables de esta ecuación

**Notación:** como las nuevas coordenadas son todas constantes del movimiento, en el fondo son parámetros (e.g: condiciones iniciales). Por ende, emplearemos la notación:

$$q'^a\to \alpha^a$$
$$p'_b \to \beta_b$$

De este modo, la generatriz será $S=S(q, \alpha, t)$ y tendremos:

$$p_a = \frac{\partial S}{\partial q^a}$$
$$\beta_a = -\frac{\partial S}{\partial \alpha^a}$$

## Soluciones completas

**Observación:** no necesitamos hallar la solución general de esta ecuación, nos vale con encontrar una función generatriz que nos dé la transformación que queremos. Lo único, es imprescindible para nuestra formulación que la transformación sea invertible

**Def.:** diremos que una solución es "completa" si es invertible, es decir, si su Hessiana es no-degenerada:

$$\boxed{\det\frac{\partial^2S}{\partial q^a\partial \alpha^b}\neq 0}$$

### Función principal de Hamilton

**Def.:** la acción de un sistema es el funcional que toma un Lagrangiano y lo integra en función de un parámetro entre dos puntos del espacio de configuración en dos instantes dados, tal que se torna extremal cuando las trayectorias son clásicas.

**Def.:** la "función principal de Hamilton" $S_H$ es la acción on-shell (sobre trayectorias físicas). Es función de las condiciones iniciales y finales de los puntos que conecta, es decir: $S_H=S_H(q_0, q_1, t_0, t_1)$

**Teorema:** la función principal de Hamilton es una solución de la ec. de $H-J$ (pero no sirve de mucho, al menos en mecánica clásica, porque hallarla requiere haber resuelto la dinámica, que es precisamente lo que queremos conseguir)

**Ejemplo:** $S_H$ para la partícula libre en 1 dimensión

$$\mathcal L=\frac{\dot q^2}{2}$$

La trayectoria que conecta $q_0$ con $q_1$ en los instantes $t_0$ y $t_1$ es:

$$q(t)=q_0+\frac{q_1-q_0}{t_1-t_0}(t-t_0)$$

Notamos que, efectivamente, $q(t_0)=q_0$ y $q(t_1)=q_1$. Ahora hallamos el Lagrangiano on-shell e integramos:

$$\mathcal L_{O-S} =\frac{1}{2}\left(\frac{q_1-q_0}{t_1-t_0}\right)^2$$

Ahora lo integramos para hallar la función principal de Hamilton:

$$\boxed{S_H(q_0, q_1, t_0, t_1)=\frac{(q_1-q_0)^2}{2(t_1-t_0)}}$$

Ahora hallemos las variables restantes. Como $\delta S = (p\delta q-H\delta t)|_{t_0}^{t_1}$, tenemos que:

$$p_{0}=-\frac{\partial S}{\partial q_0}; p_1=\frac{\partial S}{\partial q_1}; H=-\frac{\partial S}{\partial t_1}$$

# Separación de variables

- Veamos ahora algunos casos típicos que nos permiten simplificar la resolución de la ec. de H-J

## Caso 1: $\partial_t H = 0$

- En este caso, la energía es conservada y podemos separar la dependencia temporal de la función de H-J del resto de su dependencia:

$$\boxed{S(q; \alpha, t)=W(q; \alpha)-Et}$$

Introduciendo esta expresión en la ec. de H-J, queda la ecuación de H-J independiente del tiempo:

$$\boxed{H(q, \partial_q W, t)=E}$$

## Caso 2: coordenada cíclica

- Exactamente la misma separación de variables aplica cuando existe una coordenada cíclica:

$$S(q, t)=S_0(q^i, \alpha, t)+\tilde \alpha_i \tilde q^i$$

donde las variables tildadas son las cíclicas. La ecuación que queda ahora es:

$$\boxed{\partial_t S_0+H(q, \partial_i S_0, \tilde \alpha_i, t)=0}$$

**Observación:** como la coordenada es cíclica, su momento canónico es una constante del movimiento, y por ende será una de las coordenadas que adoptemos tras la transformación. En particular, los momentos conservados pasan a ser nuevas coordenadas generalizadas

## Caso 3: variables separables

**Def.:** diremos que una variable $q^1$ es "separable" si ella y su momento conjugado siempre aparecen en la misma combinación en el Hamiltoniano. Refleja un posible cambio de variable para dar una coord. cíclica

La estrategia en estos casos es buscar soluciones de la forma:

$$S(q, t)=S_0(q^2, \ldots, q^n, t)+S_1(q^1)$$

Una solución de esta ecuación consiste en coger como parámetro la función que siempre sale repetida, $\alpha^q=f_1(q^1, \partial_1S_1)$, integrarla por cuadraturas y resolver la ec en 1 variable menos:

$$\partial_t S_0+H(q^2, \ldots, q^n, \partial_2 S_0, \ldots, \partial_n S_0, \alpha_1, t)=0$$

Si sucediera que $q^2$ es asimismo separable, entonces podríamos seguir el proceso de forma recursiva. Si todas las variables son separables, entonces una forma de la solución es:

$$\boxed{S(q, \alpha, t)=S_1(q^1; \alpha^1)+S_2(q^2; \alpha^1, \alpha^2)+\ldots+S_n(q^n; \alpha^1,\ldots, \alpha^n)+S_{n+1}(t; \alpha_1,\ldots,\alpha^n)}$$

donde $S_{n+1}=-\int dtH(t; \alpha)$

**Ejemplo:** $H=\frac{p^2}{2}+\frac{1}{2}g(q)(q'^2+p'^2)$

Observamos que (1) el Hamiltoniano es indep. del tiempo y (2) que $q'$ separable. Entonces, tenemos la separación:

$$S(q, q', t)\to W(q, q')-Et \to S_0(q)+S'(q')-Et$$

Podemos plantear la ec. de H-J independiente del tiempo y resolver, pero no hace falta matarse porque ya conocemos una solución para estos casos:

$$f_1(q', p')=\frac{1}{2}(q'^2+p'^2)\Longrightarrow \frac{1}{2}(q'^2+(S'_{q'})^2)=\alpha_1$$

# Teoremas de integrabilidad

## Teorema de Liouville

### Formulación sencilla

- Un sistema con $n$ grados de libertad que posee $n$ constantes del movimiento (1) independientes y (2) en involución es integrable por cuadraturas

### Formulación detallada

- Sea la variedad simpléctica $M$ asociada a un sistema de $n$ grados de libertad, y escojamos una carta de coordenadas canónicas $\xi$
- Sea el Hamiltoniano del sistema independiente del tiempo
- Sean $n$ constantes del movimiento $\{g_a\}$, entre las cuales obviamente se encuentra $H$

**Def.:** se define un "subconjunto de nivel" como aquella región del espacio de fases tal que los valores de las constantes del movimiento tienen un valor determinado. Esto es un subconjunto del espacio de fases, en general no será sub-variedad

$$U_g:=\{p\in M | g_a(\xi_p)=g_{a,0}\}\subset M$$

- **Premisas:**

1. $\{g_a\}$ son independientes (i.e: sus diferenciales son linealmente independientes o, equivalentemente, $dg_q\wedge \ldots \wedge dg_n\neq 0$)
2. $\{g_a\}$ Están en involución ($\{g_i, g_j\}=0$) --> en particular, como $H$ es una de esas constantes, esto implica que no dependen explícitamente del tiempo porque $\{g_i, H\}=\dot g_i-\partial_t g_i=0$

- **Tesis:**

1. $U_g$ es sub-variedad invariante bajo los flujos generados por las constantes del movimiento, y en particular bajo evolución temporal
2. Las ecuaciones del movimiento en las variables originales se pueden integrar por cuadraturas

## Teorema de Liouville-Arnold

**Premisas:**

1. Las del teorema de Liouville
2. $U_g$ es compacta y conexa

**Tesis:**

1. $U_g$ es difeomorfa a un toro n-dimensional
2. Existen variables de acción ángulo $(\phi^a, J_a)$ tales que:

$$\oint_{\gamma}d\phi=2\pi$$

donde la integral es a lo largo de una curva que recorre $U_g$ en una sola vuelta

**Nota:** pueden existir valores críticos de $\{g_a\}$ para los que $U_g$ deja de ser sub-variedad n-dimensional. Esto indica un cambio de régimen

### Corolario 

- Podemos hallar la generatriz de la transformación canónica de tipo 2 que nos lleva a las variables de acción-ángulo mediante:

$$\boxed{F_2(q,g)=\int_{\gamma_g(q)} p_a dq^a}$$

donde $\gamma_g(q)$ es una curva arbitraria pero contenida en $U_g$ tal que une un punto inicial $(q_0, g)$ con otro $(q, g)$

### Corolario

- En términos de estas variables acción-ángulo, las ecuaciones de Hamilton son de la forma:

$$\begin{cases}\dot \phi^a=\omega^a(J) \\ \dot J_a=0\end{cases}$$

Y por ende se pueden resolver inmediatamente:

$$\begin{cases}\phi^a(t)=\phi^a(0)+\omega^a t \\ J_a(t)=J_a(0)\end{cases}$$

**Nota:** esta construcción es muy útil porque nos permite calcular la frecuencia de un grado de libertad concreto sin tener que integrar la dinámica del problema completamente. Basta con calcular $J$ y después aplicar las ecuaciones de Hamilton para hacer $\dot \phi^\alpha \equiv \omega^\alpha=\partial_{J_\alpha}H$

# Variables de acción-ángulo

## Construcción en sistemas con 1 grado de libertad

- Sea un sistema con 1 grado de libertad e independiente del tiempo. Entonces, la energía se conserva y el sistema es integrable por cuadraturas
- Además, si las curvas de nivel $E=cte.$ son cerradas en una región del espacio de fases, podemos definir variables de acción-ángulo para él en esa región
- Sea un sistema sometido a un potencial $V(q)$, es decir, $H=\frac{p^2}{2m}+V(q)$, tal que $V$ acotado superiormente. Entonces,, para energías $E<V_{max}$, el sistema trazará curvas cerradas en el espacio de fases y podemos usar variables de acción-ángulo $(\phi, J)$
- Encontremos la generatriz de tipo 2 de la transformación canónica:

$$W(q, J)=\int_{q_0}^qpdq\Big|_{U_E}; \ \ q_0\in U_E$$

**OJO!:** no hemos especificado cuántas vueltas damos alrededor de la curva $U_E$, de modo que la generatriz es multivaluada. En particular, si hemos dado $k$ vueltas, tenemos que:


$$W=k\oint_{U_E}pdq+\int_{q_0}^qpdq\Big |_{U_E}$$

donde la segunda integral no llega a completar una vuelta alrededor de $U_E$

**Observación:** esta ambigüedad no afecta a $p=\partial_qW$ porque la primera integral no depende de $q$. Sin embargo, sí afecta a $\phi=\partial_JW$, de modo que $p$ está bien definido pero $\phi$ no lo está. Concretamente, $\phi$ está definida salvo por un número entero de veces el término $\partial_J\oint_{U_E}pdq$. 

**Condición:** imponemos ahora la condición de que $\oint_{U_E} d\phi=2\pi$ tras completar una sola vuelta (notación: $\oint_1$). De aquí, obtenemos la condición:

$$2\pi=\partial_J\oint_{U_E}pdq$$

Integrando esta expresión para $J$, tenemos que:

$$J=\frac{1}{2\pi}\oint_{1}pdq$$

**Observación:** este es precisamente el invariante integral de Poincaré, también conocida como invariante adiabático. Recordamos que tenía la propiedad de ser invariante bajo transformaciones canónicas, y en particular esto implica la evolución temporal, de modo que nos vale como variable de acción

Como ahora ya sabemos cuánto vale esta integral cerrada en términos de la variable $J$, podemos calcular la variable de ángulo fácilmente, y también la generatriz. En resumen, tenemos que:

$$\boxed{\begin{gather}J=\frac{1}{2\pi}\oint_1 pdq \\ \phi=2\pi k+\partial_J\int_{q_0}^q pdq\Big |_{U_E} \\ W(q, J)=2\pi k J +\int_{q_0}^q pdq\Big |_{U_E}\end{gather}}$$


**Interpretación:** Si entendemos $J(E)$ como el número que etiqueta la curva de nivel $U_E$ para un valor de la energía dado, entonces $\phi$ no es más que el ángulo que marca el punto de esa curva en el que nos encontramos (i.e: estamos usando polares, o más bien, elípticas). Entonces, esta ambigüedad es completamente natural

**Otra interpretación para J:** es precisamente el área encerrada por la curva de nivel $E=cte.$

## Resolución del caso unidimensional

- Hemos visto que el caso unidimensional siempre se puede reducir a cuadraturas. Resolvamos el problema más general unidimensional $H=\frac{p^2}{2m}+V(q)$ cuando $E<V_{max}$
- Fijada una curva de nivel, el momento sobre ella valdrá:

$$p=\pm \sqrt{2m[E-V(q)]}$$

Lógicamente, esta es la curva de nivel, parametrizada con una de las coordenadas. Lógicamente, el sistema tendrá dos puntos de retorno, que serán aquellos valores de $q$ tales que $E=V(q_*)$. Si se quiere, en el espacio de fases estos son los puntos en los que $p=0$. Hallemos la variable de acción del sistema:

$$J=\frac{1}{2\pi}\oint_{U_E}pdq=\frac{2}{2\pi}\int_{q_-(E)}^{q_+(E)}\sqrt{2m[E-V(q)]}dq$$

- Por otro lado, si tomamos por ejemplo 0 vueltas alrededor de $U_E$, tenemos que la generatriz de la transformación es:

$$W=sgn(p)\int_{q_-(E)}^q\sqrt{2m[E-V(q)]}dq$$

donde el signo aparece porque podemos hacer esta integral en la rama de arriba o en la de abajo (piensa en una circunferencia)

- Finalmente, la variable angular se puede determinar fácilmente ahora a partir de la generatriz:

$$\phi=\partial_J W = \partial_E W \frac{dE}{dJ}$$

Sin embargo, nosotros lo que tenemos es una expresión de $J$ en función de $E$, no al revés. Como la derivada es exacta, podemos emplear el teorema de la función implícita para invertir la expresión:

$$\phi = \partial_J W = \partial_E W \left(\frac{dJ}{dE}\right)^{-1}$$

Por ende, tenemos que:

$$\phi = \sigma\pi\frac{\int_{q_-(E)}^q [E-V(q)]^{-1/2}dq}{\int_{q_-}^{q_+}[E-V(q)]^{-1/2}dq}$$

## Caso general: $n$ grados de libertad

**Teorema:** en un sistema con $n$ grados de libertad que satisface las condiciones del teorema de Liouville-Arnold, las $n$ variables de acción son las correspondientes integrales de Poincaré:

$$J_a=\frac{1}{2\pi}\oint_{\gamma_a}p_b dq^b$$

La función generatriz de tipo 2 que nos da la transformación canónica $(q^a, p_a)\to(\phi^a, J_a)$ es:


$$W(q, J)=\int_{q_0}^qp_bdq^b\Big |_{U_g}$$

## Conexión con H-J:

**Teorema:** $W(q, J)$ es una solución completa de la ecuación de Hamilton-Jacobi independiente del tiempo

**Teorema:** si $W(q, g)$ es solución completa de H-J independiente del tiempo, entonces las $n$ constantes del movimiento están en involución. Si además el subconjunto de nivel es compacto, podemos construir variables de acción-ángulo

**Teorema:** si los flujos de $\{g_a\}$ son cerrados (en general se arrollan alrededor del n-toro) entonces las variables de acción ángulo son especialmente fáciles de construir. Las variables canónicas conjugadas $f^a=\partial_{g_a}W(q,g)$ son periódicas y por ende las variables de acción-ángulo son:

$$\boxed{\phi^a=2\pi\frac{f^a}{K^a}; \ \ J_a=\frac{1}{2\pi}g_a K^a}$$

donde $K^a=\oint_{U_g}df^a$
