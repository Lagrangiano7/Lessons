# 1. Introducción

## Definición

**Def.:** una variedad diferenciable $M$ es un espacio topológico de Haussdorf (dos puntos distintos tienen entornos disjuntos) y segundo-numerable (la topología tiene base numerable) junto con un Atlas maximal diferenciable

## Cartas

- Intuición: una variedad diferenciable es un espacio abstracto, pero podemos pensar en él como una superficie. Dado un punto de la variedad $p\in M$, deseamos poder etiquetarlo de manera única para diferenciarlo de los demás. Entonces, podemos coger una cierta región $U_M\subseteq M$ sobre la variedad y asociarle otra región $U_\mathbb R \subseteq R^n$ (donde $n$ es la dimensión de nuestra variedad). Asimismo, necesitamos una "regla" $\varphi:U_M\to U_\mathbb R$ que a cada punto dentro de la región de la variedad le asocie un punto (y solo uno) dentro de la región $U_\mathbb R$. Al par ($U_\mathbb R, \phi$) se le llama "carta" y nos permite asociarle coordenadas a un punto de la variedad

$$\text{Carta}:= \{U_\mathbb R, \varphi\}$$

tal que:

$$\varphi: \ U_M  \xrightarrow[p \ \longmapsto \ u_p]{} U_\mathbb R$$

De este modo, podemos entender:

$$p=\varphi^{-1}(u_p)$$

**Nota:** en general, podemos cubrir una misma región de la variedad con distintas cartas (son como parches en la superficie). Si por ejemplo tenemos las cartas $\alpha=(u^\alpha, \varphi_\alpha)$ y $\beta=(u^\beta, \varphi_\beta)$, entonces podemos "traducir" las coordenadas de un punto sobre la variedad de la carta $\beta$ a la $\alpha$ haciendo $U_\mathbb R^\beta \to U_M\to U_\mathbb R^\alpha$. Es decir, la composición:

$$\varphi_\alpha \ \circ \varphi^{-1}$$

Nos coge las coordenadas del punto en la carta $\beta$ y nos da las coordenadas de ese mismo punto en la carta $\alpha$

**Def.:** diremos que dos cartas son "compatibles" si podemos pasar de una a otra de forma suave. Esto permite hacer cambios de coordenadas usando la regla de la cadena (esencialmente, permite la idea del Jacobiano, que lleva implícita la idea de que el paso de unas coordenadas a otras es suave, i.e: diferenciable)

**Def.:** un "Atlas" es un conjunto de cartas suficientes para cubrir toda la variedad:

$$\mathcal A:=\{(u_i, \varphi_i)\}_i$$

**Def.:** un "Atlas maximal" es el conjunto de todas las cartas compatibles con un dado Atlas

# 2. Funciones sobre la variedad

- Intuición: como la variedad es un espacio abstracto, no sabemos definir funciones sobre ella, ni qué significa eso. Lo que sí sabemos es definir funciones sobre $\mathbb R^n$, y también sabemos llevar regiones de la variedad $M$ a regiones en $\mathbb R^n$, entonces ya tenemos el problema resuelto: definimos una función sobre la variedad pero luego pasamos sus puntos de la variedad a la región de $\mathbb R^n$ que nos dice la carta, y así estamos trabajando con funciones sobre los reales

**Def.:** se define una función $f$ sobre la variedad $M$ de la forma:

$$f: \ M\to \mathbb R$$

tal que:

$$f(p)=f(\varphi(p))$$

**Observación:** naturalmente, como la función está definida sobre la variedad, su valor **no debe** depender de la carta que estemos usando, porque es solo una forma de referirse a los puntos. La función, entendida como regla de asignación, coge un punto $p\in M$ y le asigna un valor real. Por ende, como siempre hablamos del mismo punto $p$, no debe importar qué carta escojamos para referirnos a él (i.e: el lenguaje no altera el significado)

## Derivación de funciones y concepto de vector tangente

- Como la variedad tiene en general dimensión $n$, tiene sentido hablar de derivada direccional
- Nos servimos de las curvas para derivar. Esto es, como sé definir la derivada de una curva en un punto (que se asocia con su vector tangente en ese punto), derivaré las funciones según la dirección de la curva en ese punto:

$$\frac{d}{dt}f(\gamma(t))|_{t=0}$$

**Observación:** véase que esto no depende de la curva $\gamma(t)$ seleccionada, siempre y cuando sean localmente iguales en ese punto (su primera derivada coincida, porque entonces tendrán el mismo vector tangente). Véase que, como lo hemos definido, tiene todo el sentido que hablemos de vectores tangentes:

$$\gamma: \mathbb R \to \mathbb U_M$$
$$f: U_M\to \mathbb R$$

Por ende, cuando escribimos $f(\gamma(t))$, tenemos una función $f \ \circ \ \gamma: \mathbb R \to \mathbb R$, y podemos entonces aplicar los conceptos de funciones de una variable.

Aplicando la regla de la cadena a $\frac{d}{dt}f(\gamma(t))$, tenemos:

$$\frac{d}{dt}f(\gamma(t))|_{t=0} = \left.\frac{\partial f}{\partial x^\alpha}\frac{dx^\alpha}{dt}\right|_{t=0}$$

Llamando $\left.X^\alpha(p)=\frac{dx^\alpha}{dt}\right|_{t=0}$ (**componentes contravariantes del vector**) (recordemos que $\gamma(t=0)=p)$, tenemos entonces un operador diferencial que nos da la derivada direccional de $f$ en la dirección del vector tangente de la curva $\gamma(t)$ en el punto $p$ de la variedad: hemos conseguido definir las derivadas direccionales en la variedad de forma clara

Vista la relación entre los vectores tangentes a la variedad en un punto $p$ y la derivación de funciones, definimos:

$$\boxed{X(p): C^\infty (p) \to \mathbb R \text{ que satisface (1) linealidad y (2) regla de Leibniz}}$$

**Def.:** el conjunto de todos los vectores sobre un punto $p$ recibe el nombre de "espacio tangente a la variedad en $p$", y se denota $T_pM$. Tiene estructura de espacio vectorial, y la "base natural" es:

$$\boxed{B_p:=\{\partial_\alpha |_p\}_{\alpha=1}^n}$$

Esto era de esperar, porque ya hemos visto cómo actúan los vectores $X(p)$ sobre las funciones de la variedad:

$$Xf = \frac{\partial f}{\partial x^\alpha}X^\alpha$$

De modo que parece natural definir:

$$X(p)=X^\alpha\partial_\alpha|_p$$

# 3. Fibrados

## Fibrado tangente

**Def.:** el "fibrado tangente" es la unión disjunta de todos los espacios tangentes a la variedad en distintos puntos de la misma:

$$TM := \displaystyle\coprod_{p\in M} T_qM$$

Esto es, un elemento del fibrado tangente será un par ordenado $(p, v)$ donde $p$ sea un punto de la variedad y $v$ sea un vector en el espacio tangente a la variedad en dicho punto, $v\in T_qM$. Especificar el punto de la variedad es necesario, porque de lo contrario no sabríamos a qué espacio vectorial pertenece el vector.

**Resultado:** si la variedad es de dimensión $n$, entonces el fibrado tangente, por ir a pares, tiene dimensión $2n$. Un ejemplo clásico es el espacio de fases en velocidades en mecánica Lagrangiana, donde tenemos la ordenación típica $\xi^\alpha=(q^1,\ldots,q^n,\dot q^1,\ldots, \dot q^n)$

**Importante:** el fibrado tangente tiene estructura de variedad diferenciable, esta vez de dimensión $2n$. Por ejemplo, en Mecánica Clásica, el Lagrangiano es una función definida sobre el fibrado tangente del espacio de configuración

**Nota:** los vectores tangentes a la variedad viven en el fibrado tangente. Antes estábamos fijándonos solo sobre puntos concretos $p\in M$, pero si extendemos la definición de "cualquier" a "todos" los puntos de la variedad, entonces tenemos un campo vectorial. Este vive en el fibrado tangente

## Fibrado cotangente

- Hemos visto que el espacio tangente a un punto $p\in M$ de la variedad tienen estructura de espacio vectorial (por eso los llamamos vectores). Entonces, también podemos definir el espacio dual de dicho campo vectorial. Este recibe el nombre de "espacio cotangente", y comprende todas las 1-formas definidas sobre el espacio tangente a la variedad en $p$:

$$\boxed{T_p^*M :=\{\omega: T_pM \xrightarrow[X(p)\longmapsto a]{} \mathbb R\}}$$

- La "base natural" del espacio cotangente es:

$$B_p^* :=\{dx^\alpha\}_{\alpha=1}^n$$

**Propiedad:** las 1-formas de la base dual, por definición de espacio dual, actúan sobre los vectores de la base del espacio vectorial para dar:

$$dx^\alpha|_p(\partial_\beta|_p)=\delta^\alpha_\beta$$

**Nota:** de forma completamente idéntica a cómo definíamos el fibrado tangente de la variedad, definimos el fibrado cotangente como la unión disjunta de todos sus espacios cotangentes:

$$T^* M := \displaystyle\coprod_{p\in M} T^*_qM$$

De nuevo, podemos pensar en un elemento del fibrado cotangente de la variedad como el par ordenado $(p, \omega)$ con $p\in M$ y $\omega \in T^*_pM$

**Proposición:** el diferencial exacto de una función de la variedad es una 1-forma:

$$df(p) = \left.\frac{\partial f}{\partial x^\alpha}\right|_pdx^\alpha|_p$$

Que vemos que está expandido en la base del espacio cotangente a la variedad en $p$, de modo que es un elemento de dicho espacio. Alternativamente, podemos ver cómo actúa $df$ sobre un elemento del espacio tangente en $p$:

$$df X = \frac{\partial f}{\partial x^\alpha}dx^\alpha X^\beta \partial_\beta$$

donde el elemento de la base dual y el de la base del espacio vectorial dan una delta, así que:

$$df X = X^\alpha \frac{\partial f}{\partial x^\alpha} \equiv Xf$$

**Observación:** los diferenciales exactos de funciones definidas sobre la variedad dan 1-formas cuya acción sobre los vectores del espacio tangente es actuarlos sobre la función diferenciada

**Observación:** en general, cuando tenemos una 1-forma actuando sobre un vector, podemos escribir esta actuación en componentes de la forma:

$$\omega(X)=\omega_\alpha X^\alpha$$

Este resultado no depende de la carta que estemos usando, porque tanto $\omega$ como $X$ son tensores

## Fibrado tensorial

- Vamos a generalizar lo que hemos encontrado hasta ahora en cuestión de fibrados
- **Proposición.:** el conjunto de todos los tensores $(r, s)$ definidos en $p$ tiene estructura de espacio vectorial, y se denota:

$$T_p^{(r, s)} := \{T^r_s(p)\}$$

- Observamos que esto es más general que los espacios tangente y cotangente. En particular:

$$T_pM \equiv T_p^{(1, 0)}$$
$$T_p^*M \equiv T_p^{(0,1)}$$

- La "base natural" de este espacio vectorial será:

$$B_p^{(r, s)}:=\{\partial_{\alpha_1}\otimes \ldots \otimes \partial_{\alpha_r}\otimes dx^{\beta_1}\otimes \ldots \otimes dx^{\beta_s}\}$$

- Un fibrado tensorial será entonces como viene siendo costumbre la unión disjunta de todos los espacios vectoriales de tensores de tipo $(r, s)$ sobre cada punto de la variedad:

$$T^{(r, s)}M := \displaystyle\coprod_{p\in M} T^{(r, s)}_p M$$

- Donde, como siempre, un elemento del fibrado tensorial será el par ordenado $(p, T^r_s)$ con $p\in M$, $T^r_s\in T^{(r, s)}_pM$

# 4. Curvas integrales de un campo vectorial

**Nota:** ahora que tenemos definido el fibrado tensorial, podemos definir tensores en cada punto de la variedad, de modo que podemos asignarle a cada punto de la varidad un tensor. Esta asignación, si se hace de forma suave, recibe el nombre de **campo tensorial**:

$$\phi: M \xrightarrow[p \ \longmapsto T^r_s(p)]{} T^{(r, s)}M$$

**Nota:** habitualmente, nos referimos al campo tensorial directamente por sus componentes: $T^r_s(p); \ p\in M$

**Def.:** las "curvas integrales" de un campo vectorial son aquellas curvas de la variedad a las que los vectores del campo vectorial son tangentes. Naturalmente, se construyen como:

$$\frac{d}{dt}x^\alpha(t)=V^\alpha [x^\mu(t)]$$

**Nota:** la condición inicial $x^\alpha(0)=x_0^\alpha$ me fija a qué curva integral me estoy refiriendo. Naturalmente, como el campo vectorial está evaluado sobre cada punto de la variedad, hay en general infinitas curvas integrales, y estas me describen el flujo de dicho campo (i.e: las trayectorias de partículas dentro de un cuerpo de agua en movimiento. Los vectores serían la velocidad de estas partículas. Si escojo un punto inicial, entonces ya me fija la trayectoria de una partícula concreta)

**Def.:** el "flujo" de un campo vectorial es el conjunto de todas las curvas integrales de este. Formalmente, es una asignación $\varphi: \mathbb R\times M \to M$ tal que:

$$\frac{d}{dt}\varphi(t, p)=V(\varphi(t, p))$$

**Importante:** si tenemos un campo vectorial definido sobre una variedad, por cada punto de la misma pasa una y solo una curva integral. Esto se debe al teorema de existencia y unicidad de EDOs de primer orden como es la de las curvas integrales. Decimos que estas forman una congruencia

**Propiedad:** las curvas integrales llenan la variedad

# 5. Generalización de la derivada de campos tensoriales

- Hay varias formas de derivar campos tensoriales en una variedad diferencial. Todos concuerdan para funciones definidas sobre la variedad

## Conexión y Derivada Covariante

- Sabemos que en cada punto $p\in M$ de la variedad tenemos un espacio tangente distinto, $T_pM$, de modo que a priori no sabemos comparar vectores del espacio tangente de un punto $p_1$ con los otros del espacio tangente sobre $p_2$
- Si estuviéramos en $\mathbb R^n$, podríamos hacer una traslación del origen de uno de los vectores hasta que coincida con el origen del otro, y por tanto ya podríamos compararlos. Sin embargo, ¿qué pasa con dos vectores definidos sobre distintos puntos en una esfera? ¿Cómo transporto uno para que compararlo con el otro si el origen de ambos vectores no puede salirse de la esfera?
- **Solución:** la "conexión".

- Esta nos dice cómo comparar (restar) tensores pertenecientes a puntos distintos de la variedad, y lo hace por definición. Esto se refleja en los símbolos de Christoffel de la variedad. Si pensamos en superficies, los símbolos de Christoffel nos dicen cómo se derivan los elementos de la base del espacio tangente (solo si estamos en la base natural, de lo contrario en general se llamarán "coeficientes de la conexión"):

$$\nabla_{e_j}e_i = \Gamma^k_{ij}e_k$$

**Interpretación:** los coeficientes de la conexión me dicen cómo varía cada vector de la base al derivarlo en la dirección de otro vector de la base. Naturalmente, el resultado es otro vector, que me da la derivada direccional.

- Básicamente, para saber cómo derivar un tensor, necesito saber cómo se deriva el objeto completo: no solo sus componentes sino también los elementos de la base sobre la que está definido, habida cuenta de que cambian en distintos puntos de la variedad (imagínate la base del espacio tangente de una superficie)

- Para que una formulación así conduzca a un tensor, lo único que hay que pedirle a los símbolos de Christoffel es que se transformen como:


$$\bar{\Gamma}_{k l}^{i} = \frac{\partial \bar{x}^{i}}{\partial x^{m}} \frac{\partial x^{n}}{\partial \bar{x}^{k}} \frac{\partial x^{p}}{\partial \bar{x}^{l}} \Gamma_{n p}^{m} + \frac{\partial^{2} x^{m}}{\partial \bar{x}^{k} \partial \bar{x}^{l}} \frac{\partial \bar{x}^{i}}{\partial x^{m}}$$

- Entonces, si deseamos derivar un vector, tenemos:

$$\nabla_{\mu} Y = \nabla_{\mu} (Y^\alpha \partial_\alpha) = \partial_\mu (Y^\alpha)\partial_\alpha + Y^\alpha\nabla_{\mu} (\partial_\alpha)$$

Sustituyendo la conexión (cómo se derivan los elementos de la base):

$$\nabla_\mu(\partial_\alpha)=\Gamma_{\mu\alpha}^\nu\partial_\nu$$

Tenemos entonces que:

$$(\nabla_\mu Y)^\alpha \equiv Y^\alpha_{;\mu}= \partial_\mu Y^\alpha + \Gamma^\alpha_{\mu\nu}Y^\nu$$

### Conexión de Levi-Civita

- Si sobre la variedad diferenciable hemos definido una métrica, entonces existe una sola derivada covariante compatible con dicha métrica y libre de torsión, que se conoce como la "conexión de Levi-Civita":

$$\Gamma^{k}_{ij} = \frac{1}{2} g^{kl} \left( \frac{\partial g_{jl}}{\partial x^{i}} + \frac{\partial g_{li}}{\partial x^{j}} - \frac{\partial g_{ij}}{\partial x^{l}} \right)$$

## Derivada de Lie

- Sea un campo vectorial $X$ definido sobre la variedad y sean sus líneas de campo (su flujo) $\Phi_X^t(p)$
- Supongamos que tenemos una función definida sobre la variedad, $f(p)$
- La derivada de Lie mide la variación del campo tensorial $f$ cuando la arrastramos a lo largo de las curvas integrales de $X$ una cantidad infinitesimal (i.e: variación pequeña en el parámetro de la curva). Concretamente:

$$\mathcal L_X f(p)= \left.\frac{d}{dt}\right|_{t=0}f(\Phi_X^t(p))=\displaystyle\lim_{t\to0}\frac{f(\Phi_X^t(p))-f(p)}{t}$$

- Como vemos de la definición con límites, no es más que la definición habitual de derivada, solo que como no sabemos en qué parámetro avanzarla, la avanzamos en la dirección de cierta curva. Esto se hace con curvas integrales de campos vectoriales, pero entiendo que se podría generalizar a cualquier curva
- Veamos ahora cómo se derivan algunos tensores en sentido de la derivada de Lie

### Función escalar

**En palabras:** la derivada de Lie sobre el campo vectorial $X$ de una función de la variedad es la actuación del campo vectorial sobre la función:

$$\boxed{\mathcal L_X f=Xf}$$

**Dem.:**

Partimos de la definición de la derivada de Lie de $f$ y aplicamos regla de la cadena:

$$\frac{d}{dt}f(\Phi_X^t(p))=df \dot \Phi_X^t(p)$$

Pero por definición de curva integral:

$$\dot \Phi_X^t(p)=X(\Phi_X^t(p))$$

Por tanto:

$$\frac{d}{dt}f(\Phi_X^t(p))=df \dot \Phi_X^t(p) = \left(\partial_\alpha f dx^\alpha\right)X^\beta\partial_\beta$$

donde todos los elementos de la base tanto del fibrado tangente como del cotangente están evaluadas sobre el punto adecuado de la curva integral. Recordando la actuación de la base dual $dx^\alpha \partial_\beta = \delta^\alpha_\beta$:

$$\mathcal L_x f(p)=\left.\frac{d}{dt}\right|_{t=0}f(\Phi_X^t(p)) = X^\alpha\partial_\alpha f=Xf$$

Evaluado sobre el punto de la curva integral.

### Campo vectorial

**En palabras:** la derivada de Lie de un campo vectorial $Y$ a lo largo de las curvas integrales de otro campo vectorial $X$ es el conmutador de ambos campos vectoriales, y este es a su vez otro campo vectorial (no es producto tensorial):


$$\boxed{\mathcal L_X Y = [X, Y]}$$

**Dem.:**

Nosotros lo que sabemos derivar son funciones sobre la variedad, y también sabemos que la derivada de Lie cumple la regla de Leibniz por construcción. Entonces, podemos derivar la función escalar que sale de actuar el campo vectorial $Y$ sobre alguna función $f$ y aplicando la regla de la cadena despejar la derivada de Lie del campo vectorial $Y$:

$$\mathcal L_X (Yf)=\mathcal L_X(Y)f + Y\mathcal L_X(f) $$

El primer término del RHS es el que queremos obtener, y los otros dos ya los conocemos:


$$XYf=\mathcal L_X (Y)f + YXf$$

Despejando:

$$\mathcal L_X(Y)f = (XY-YX)f$$

De modo que $\mathcal L_XY=[X, Y]$ en el sentido de operadores

**Nota:** en este sentido, la derivada de Lie dota al conjunto de todos los campos vectoriales suaves $\mathfrak X(M)$, que tiene asimismo estructura de espacio vectorial, de una operación de corchete de Lie que cumple:

1. Bilinealidad
2. Identidad de Jacobi
3. $[X, X]=0$

Por tanto, la derivada de Lie de campos vectoriales dota al conjunto $\mathfrak X(M)$ de estructura de "álgebra de Lie", que no es más que un espacio vectorial con un corchete de Lie definido

### Campo de 1-formas

- Como ya sabemos actuar la derivada de Lie sobre campos vectoriales, y también sabemos que las 1-formas comen campos vectoriales, vamos a hacer el mismo truco de siempre de la regla de la cadena, esta vez del campo de 1-formas comiéndose al campo vectorial:

$$\mathcal L_X(\sigma(Y)) = L_X(\sigma)Y+\sigma L_XY$$

El argumento de LHS es una función escalar, y el segundo término de RHS es un corchete de Lie. De este modo:

$$L_X(\sigma)Y = X\sigma Y-\sigma[X, Y]$$

En componentes:

$$\boxed{\left(L_X\sigma\right)_\alpha = \left(X^\beta\partial_\beta\sigma_\alpha + \sigma_\beta \partial_\alpha X^\beta\right)}$$

### Campos tensoriales de tipo $(r, s)$

- Ahora que hemos entendido de dónde salen las derivadas de Lie de algunos de los campos más recurrentes, veamos la expresión general de la derivada de Lie a lo largo de un campo vectorial $X$ de un campo tensorial de tipo $(r, s)$:

$$\boxed{(\mathcal L_XT)^{i_1\ldots i_r}_{j_1\ldots j_s} = \frac{\partial T^{\ldots}_\ldots}{\partial x^k}X^k - T^{k\ldots}_{\ldots}\frac{\partial X^{i_1}}{\partial x^k}- \ldots + T^{\ldots}_{k\ldots}\frac{\partial X^k}{\partial x^{j_1}} + \ldots}$$

**Observación:** como es de esperar, vemos que las expresiones que hemos obtenido hasta ahora se pueden obtener de esta expresión más general, introduciendo los campos tensoriales del tipo correspondiente.

# 6. Variedades de interés en Física

- Las variedades diferenciables son ubicuas en la formulación de teorías de la Física moderna, porque nos permiten trabajar con un espacio más útil que el euclídeo, en tanto que permiten introducir ligaduras de forma natural (i.e: partícula dentro de la esfera), nos permiten usar espacios con topología más complicada y la definición no-trivial (no nula) de aparataje adicional como el tensor métrico, que induce conceptos como el de curvatura. Asimismo, tenemos conceptos como el de geodésica, transporte paralelo, etc. que también pasan a ser no-triviales y por ende podemos usarlos para nuestro beneficio (por ejemplo, en Relatividad General la gravedad se toma como la distorsión del espacio-tiempo, que se ve reflejada a su vez en la métrica)

## Variedad Riemanniana

**Def.:** una "variedad Riemanniana" es una variedad diferencial (1) **real** en la que (2) cada espacio tangente $T_pM$ viene equipado con un **producto interno** que varía suavemente entre puntos cercanos de la variedad

- Un espacio euclídeo es un espacio vectorial sobre el cuerpo de los reales con dimensión finita y un producto interno definido. Pues las variedades Riemannianas son la generalización natural del espacio euclídeo

**OJO!:** el producto interno definido sobre cada espacio tangente $T_pM$ ha de ser definido positivo, no-degenerado, simétrico y venir representado por un tensor (tensor métrico) 0-2, y estas propiedades deben cumplirse en todos los $T_pM$

**Observación:** dentro de cada espacio tangente, el tensor métrico no es más que la matriz de Gram en dicho espacio vectorial

## Variedad pseudo-Riemanniana

**Def.:** una variedad pseudo-Riemanniana es una generalización de la variedad de Riemann. Todo es igual, y lo único que cambia es que pasamos de tener un producto escalar a tener un producto pseudoescalar: la signatura de la matriz de Gram en cada $T_pM$ ya no tiene por qué ser $(0,n)$ sino que puede ser cualquiera

## Variedad de Lorentz

**Def.:** una variedad de Lorentz es una variedad pseudo-Riemanniana en la que la signatura del producto interno en cada espacio tangente es, exactamente, $(1, n-1)$

**Nota:** dependiendo de la elección de signatura que hagamos, esto puede significar que hay 1 autovalor positivo y 3 negativos (QFT) o 1 negativo y 3 positivos (electrodinámica clásica, relatividad general)

## Variedad simpléctica

**Def.:** una variedad simpléctica es una variedad diferenciable en la que, sobre cada espacio tangente $T_pM$ a la variedad, se ha definido una 2-forma no degenerada:

$$\Omega_p: T_pM\times T_pM \to \mathbb R$$

Tal que:

- Antismétrica: $\Omega(X, Y)=-\Omega(Y, X)$ (bilinealidad y simetría son precisamente 2-forma)
- No-degenerada: $\Omega(X, Y)=0 \Longleftrightarrow X=0 \text{ o bien } Y=0$

**Teorema:** los espacios simplécticos solo pueden tener dimensión par

**Dem.:**

- Proposición: el determinante de una matriz antisimétrica es nulo si la dimensión de la matriz es impar:
- Dem.: $A^T=-A$ por ser antisimétrica, así que: $det(A)=det(A^T)=det(-A)=(-1)^n det(A)$. Si $n$ impar,$det(A)=0$. Si $n$ par, trivial y en general $det(A)$ podrá ser 0 o no.
- Pero si la forma bilineal tiene determinante no-nulo, entonces ya no puede ser no-degenerada. Por ende, ya no tenemos una forma bilineal antisimétrica. Entonces, solo podemos definir formas bilineales antisimétricas en dimensiones pares, y ese es el motivo por el que nuestra variedad simpléctica solo puede tener una dimensión par.
