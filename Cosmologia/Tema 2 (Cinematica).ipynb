# 1. Métrica de Robertson-Walker (RW)

## 1.1 Construcción

**Observación:** a escalas suficientemente grandes, sabemos que el Universo es (1) homogeneo e (2) isótropo. Esta es la **hipótesis cosmológica**, que ahora se ha confirmado

**Proposición:** la métrica a emplear debe ser la de Robertson-Walker, que es:

$$ds^2=dt^2-a^2(t)\left[\frac{dr^2}{1-kr^2}+r^2d\Omega^2\right]$$

**Caveat:** $1-kr^2>0$

**Def.:** decimos que un "observador comóvil" es aquel que se expande con el Universo (i.e: con el fluido cósmico), de modo que para dicho observador todo está estático. Por ende, $t$ ("tiempo cosmológico") es precisamente el tiempo propio de un observador comóvil

**Dem:**

Comencemos expandiendo una forma bilineal simétrica arbitraria:

$$ds^2=g_{00}dt^2 + 2g_{0i}dtx^i+g_{ij}dx^idx^j$$

**Observación:** por isotropía, no puede haber direcciones privilegiadas por la métrica, de modo que $g_{0i}=0$ para evitar que la dirección espacial $x^i$ sea privilegiada (i.e: aparezca explícitamente)

**Imposición:** hacemos $g_{00}=1$ porque queremos dar una descripción de un observador comóvil, así que queremos que el tiempo coordenado coincida con el propio de dicho observador

Para la parte espacial, se puede demostrar que la forma más general de una métrica isótropa es:

$$g_{ij}dx^idx^j=-a^2(t, r)[\lambda^2(r)dr^2+r^2d\Omega^2]$$

**Observación:** si deseamos que la métrica sea homogenea, el factor de escala no puede depender de la posición. Tampoco depende de ángulos porque ya hemos dicho que la métrica es isótropa. Entonces:

**Imposición:** el espacio-tiempo debe ser homogeneo, así que la curvatura espacial no puede depender de la posición

$$^3R=\frac{2}{a^2r^2}\frac{d}{dr}\left(r\left(1-\frac{1}{\lambda^2(r)}\right)\right)$$

Como no puede depender de $r$, esto fuerza una forma para $\lambda$:

$$r\left(1-\frac{1}{\lambda^2(r)}\right)=kr^3+C$$

Pero vemos que lógicamente $C=0$ para evitar una inconsitencia cuando $r\to 0$ (como estamos empleando coordenadas esféricas en la parte espacial, creo que el origen no está bien determinado). Entonces, tenemos:

$$\lambda^2(r)=\frac{1}{1-kr^2}$$

**Proposición:** el escalar de curvatura de este espacio-tiempo es:

$$R=-6\left(\frac{\ddot a}{a}+\left(\frac{\dot a}{a}\right)^2+\frac{k}{a^2}\right)$$

**Proposición:** la curvatura de las secciones espaciales viene completamente determinada por el parámetro $k$

$$\boxed{^3R = \frac{6k}{a^2} \begin{cases}k>0 \text{ \ (cerrado)} \\ k=0 \text{ \ (plano)}  \\ k<0 \text{ \ (abierto)}\end{cases}}$$

**Observación:** aunque la curvatura de las secciones espaciales sea una magnitud local (depende de las coordenadas aunque solo sea $t$), véase que su signo es una cantidad global: si el Universo es cerrado / plano / abierto, lo será para todo instante y todo lugar (al menos siempre que estemos en la misma carta)

## 1.2 Christoffels

**Proposición:** los símbolos de Christoffel de segunda especie no cambian si se toma una signatura u otra para la métrica (véase la fórmula, van siempre con 2 g's)

**Temporales:** 

$\Gamma^0_{ij}=-\frac{\dot a}{a}g_{ij}$

**Espaciales:** 

$\Gamma^i_{jk}=\frac{1}{2}g^{il}\left(\partial_k g_{lj}+\partial_j g_{lk}-\partial_l g_{jk}\right)$

$\Gamma^i_{0j}=\frac{\dot a}{a}\delta^i_j$

---

Más explícitamente:

$\boxed{t}$

$\Gamma^t_{rr}=\frac{a\dot a}{1-kr^2}$

$\Gamma^t_{\theta \theta}=a\dot ar^2$

$\Gamma^t_{\varphi\varphi}=a\dot r^2\sin^2\theta$

$\boxed{r}$

$\Gamma^r_{rt}=\Gamma^r_{tr}=\frac{\dot a}{a}$

$\Gamma^r_{rr}=-\frac{1}{2f(r)}\frac{df}{dr}$ // $f(r)=\frac{1}{1-kr^2}$

$\Gamma^r_{\theta\theta}=-rf(r)$

$\Gamma^r_{\varphi\varphi}=-rf(r)\sin^2\theta$

$\boxed{\theta}$

$\Gamma^\theta_{t\theta}=\Gamma^\theta_{\theta t}=\frac{\dot a}{a}$

$\Gamma^r_{r\theta}=\Gamma^r_{\theta r}=\frac{1}{r}$

$\Gamma^\theta_{\varphi\varphi}=-\sin\theta\cos\theta$

$\boxed{\varphi}$

$\Gamma^\varphi_{t\varphi}=\Gamma^\varphi_{\varphi t}=\frac{\dot a}{a}$

$\Gamma^\varphi_{r\varphi} =\Gamma^\varphi_{\varphi r} = \frac{1}{r}$

$\Gamma^\varphi_{\theta\varphi} = \Gamma^\varphi_{\varphi\theta}=\text{cotan \ }\theta $

## 1.4 Coordenadas comóviles

**Def.:** la distancia física entre dos puntos del espacio en un instante dado será:

$$dl_{P}:=\sqrt{-g_{ij}dx^i dx^j}$$

Véase que depende del tiempo debido al factor de escala. Como esperaríamos, las distancias escalan con la expansión del Universo

**Def.:** la distancia comóvil entre dos puntos del espacio es la que mide un observador comóvil, para el que no existe expansión porque se expande solidario al fluido. Entonces, este observador siempre medirá la misma distancia independientemente del tiempo, y será:

$$dl_C := -\sqrt{h_{ij}dx^idx^j}$$

con $h_{ij}$ la parte espacial de la métrica pero sin el factor de escala. Obviamente, ambas distancias están relacionadas por:

$$dl_P=a(t)dl_C$$

Deseamos ahora una forma más natural de tratar las distancias. En particular, desearíamos que la distancia entrara en la métrica directamente, sin factores dividiendo ($f(r)$). Entonces, definimos la coordenada comóvil como:

$$\boxed{d\chi :=\frac{dr}{\sqrt{1-kr^2}}}$$

**Interpretación:** la distancia comóvil mide distancias longitudinales, mientras que la radial mide distancias transversales. Podemos integrar la relación fácilmente para estudiar la relación entre ambas coordenadas:

$$r(\chi)=\begin{cases}\frac{1}{\sqrt k}\sin(\sqrt k \chi), \ \ \ \  \quad  k>0 \\ \chi, \ \ \ \ \ \ \  \quad \quad \quad \quad \quad  k=0 \\ \frac{1}{\sqrt {|k|}}\sinh(\sqrt{|k|} \chi), \ k<0\end{cases}$$

**Nota:** en el caso cerrado ($k>0$), es $\sqrt k \chi \in [0, \pi]$

En términos de esta nueva variable, la métrica queda de la forma:

$$\boxed{ds^2=dt^2-a^2(t)[d\chi^2+r_k^2(\chi)d\Omega^2]}$$

## 1.5 Tiempo conforme

**Def.:** el "tiempo conforme" se define como:

$$d\eta:=\frac{1}{a(t)}dt$$

con la condición $\eta(t=0)=0$

Con esta nueva coordenada temporal, la métrica queda:

$$ds^2=a^2(\eta)\left[1-d\chi^2+r_k^2(\chi)d\Omega^2\right]$$

**Observación:** en un Universo plano ($k=0$), la métrica de RW es conforme a la métrica de Minkowski. Por ende, tienen el mismo tensor de Weyl (nulo) y la estructura causal es la misma (conos de Minkowski)

**Nota:** para una señal luminosa siguiendo una trayectoria radial, tenemos que $d\eta^2-d\chi^2=0$, de modo que sobre esta geodésica la métrica es conforme a Minkowski. Por ende, tenemos que $\left|\frac{d\chi}{d\eta}\right|=1$

# 2. Efectos de la expansión en la cinemática

- La expansión del Universo hace que algunas cantidades que en Minkowski eran constantes en el tiempo para una partícula libre ya no lo sean

## 2.1 Dilatación temporal cosmológica

**Teorema:** la expansión del Universo hace que los intervalos temporales que comprende un evento aumenten en el tiempo de la forma:

$$\boxed{\frac{dt_0}{a(t_0)}=\frac{dt_1}{a(t_1)}}$$

**Interpretación:** supóngase que se emite una señal en una estrella en $r_1$ en el instante $t_1$, que llega al observador, situado en $r_0$, en el instante $t_0$. Durante el tiempo que ha tardado en llegar la señal, el Universo se habrá expandido, de modo que efectivamente la señal ha tenido que recorrer una distancia mayor a la que había entre la estrella y el observador cuando se emitió la señal. Por ende, la estrella emite un fotón en el instante $0$ y otro en el instante $1$, en la Tierra medirá que el primer fotón ha llegado en el instante $5$ y el segundo en el instante $8$. Por ende, la estrella ha emitido un pulso de duración de $1$ pero nosotros en la Tierra "medimos" que ha durado $3$ (entendiendo "medir" por la información que nos llega a nosotros)

**Dem:**

Sea la Tierra posicionada en $\chi_0$ y sea una estrella posicionada en $\chi_1$. La estrella emite una señal radialmente hacia la Tierra en un instante de tiempo conforme $\eta_1$, que llega a la Tierra en el instante $\eta_0$. Luego, la estrella emite otra señal en el instante $\eta_1+d\eta_1$ que llega a la Tierra en el instante $\eta_0+d\eta_0$. Como hablamos de luz, $ds^2=0$, y como la trayectoria se supone radial, $d\Omega^2=0$. Por ende, $d\eta^2-d\chi^2=0$ y tenemos una métrica que, sobre la trayectoria radial de la señal, es conforme a Minkowski.

Por ende, el delay con el que se nos llega la señal será igual al delay con el que se ha enviado: $d\eta_0=d\eta_1$. Sustituyendo la relación entre tiempo conforme y tiempo cosmológico:

$$\frac{dt_0}{a(t_0)}=\frac{dt_1}{a(t_1)}$$

donde $dt_0$ es el delay (en tiempo cosmológico) con el que llega la señal a la Tierra y $dt_1$ es el delay con el que se envió. Vemos que, si el Universo se ha expandido, entonces $dt_0>dt_1$: medimos que el pulso dura más de lo que duró al enviarse, porque el segundo fotón ha tenido que recorrer más distancia que el primero$_\blacksquare$

## 2.2 Redshift cosmológico

**Teorema:** las distancias físicas, y en particular las longitudes de onda, se corren al rojo debido a la expansión del Universo

$$\lambda_0=\frac{a(t_0)}{a(t_1)}\lambda_1$$

**Interpretación:** la expansión del Universo es esencialmente un factor de escala en la parte espacial de la métrica, de modo que podemos entender la expansión como una redefinición de cuánto vale la unidad de longitud para cada instante. Entonces, lo que en un instante era 1cm en otro instante serán 2, y el efecto que veremos nosotros será que la longitud de onda es mayor con el tiempo, dando lugar a un redshift.

**Def.:** se define el "redshift" ($z$) empíricamente como:

$$\frac{\lambda_{med}-\lambda_{teo}}{\lambda_{teo}}=1+z$$

donde "med" es medido y "teo" es el que se mide, por ejemplo, en el inercial propio de un átomo que emite esa linea

Entonces, de acuerdo a nuestra cinemática, tenemos que el redshift está relacionado con el factor de escala y, si resolvemos las ecuaciones de Einstein para obtener una expresión concreta de $a(t)$, tenemos una correspondencia $t\leftrightarrow z$ (por eso decimos que, a mayor redshift, más estamos observando el Universo temprano). Concretamente:

$$\boxed{1+z=\frac{a(t_0)}{a(t_1)}}$$

donde $t_0$ es cuando llega la señal luminosa y $t_1$ cuando se emitió

**Dem:**

Volvamos al caso del fotón en una trayectoria radial. Para trayectorias radiales, sabemos que las distancias vienen determinadas exclusivamente por $\chi$. En concreto, si medimos la longitud de onda (comóvil) de la señal al emitirse en la estrella: $\Delta\chi_1$ y la volvemos a medir cuando nos llegue a nosotros: $\Delta \chi_0$, veremos que coinciden. Pasando ahora a distancias físicas:

$$a(t_1)\lambda_1=a(t_0)\lambda_0 \Longrightarrow \lambda_0>\lambda_{1 \blacksquare}$$

## 2.3 Atenuación de energías / frecuencias

**Teorema:** la frecuencia de una señal se atenúa con un factor $\frac{1}{a(t)}$

**Interpretación:** parece que la energía "vale menos de sí misma" conforme el Universo se expande. Es decir, la energía de un sistema aislado no se conserva en un Universo en expansión (lógicamente, esto es compatible en excelente orden de aproximación con la conservación de la energía en procesos cotidianos porque el factor de escala tarda mucho en cambiar apreciablemente. Sin embargo, en la etapa inflacionaria, por ejemplo, entiendo que en efecto la conservación de la energía no se daba, si es que siquiera podemos aplicar RW al Universo tan temprano)

**Dem:**

Se me ocurre una demostración indirecta. No sé tratar directamente cómo se dilata el periodo, pero sí se calcular delays debidos a la expansión del Universo. Por ende, sea una estrella que emite dos señales idénticas de periodo $T$ separadas por un tiempo igual al periodo. Es decir, van desfasadas $\varphi=2\pi$. El desfase es invariante bajo la expansión trivialmente porque es una constante. Por ende, cuando lleguen a la Tierra, las señales seguirán teniendo el mismo desfase (seguirán separadas por un periodo), si bien su periodo habrá cambiado. Por ende, la separación temporal desde que nos llegue una hasta que nos llegue la otra será el nuevo periodo. Sabemos que el delay es el cociente entre factores de escala, de mood que el nuevo periodo es $T_0=\frac{a(t_0)}{a(t_1)}T_1$.

Como $2\pi = T\omega$, tenemos que $\omega_0=\frac{a(t_1)}{a(t_0)}\omega_1$ y por ende la onda efectivamente habrá sufrido redshift cosmológico. Además, obedeciendo a la cuantización de la energía de los fotones, tenemos que lo que le pase a la frecuencia del fotón le pasa a su energía, así que la energía, al menos de la luz, se atenúa también con un factor $1/a(t)_\blacksquare$

**Nota:** no sé si mi razonamiento es del todo correcto, porque $2\pi=T\omega$ implica que $c=1$, pero no sé si este es el caso a menos que estemos en coordenadas conformes

## 2.4 Atenuación del momento lineal

**Teorema:** tanto para partículas masivas como sin masa, el módulo de su momento lineal se atenúa como $|\vec p|\propto \frac{1}{a(t)}$

## 2.5 Frenado de cuerpos masivos

**Teorema:** un cuerpo masivo desplazándose libremente en un Universo en expansión se ve frenado:

$$\frac{1}{|\vec u|}\frac{d|\vec u|}{dt}=-\frac{\dot a}{a}$$

con $|\vec u|^2=-g_{ij}u^i u^j$ y $\vec u$ parte espacial de $u^\mu$

# 3. Midiendo distancias en un Universo en expansión

**Proposición:** en una geometría espacial euclídea y sin expansión, la luminosidad aparente de un cuerpo viene dada por:

$$l=\frac{L}{4\pi d^2}$$

donde $d$ es la distancia euclídea desde el punto de medición hasta el cuerpo que emite la radiación EM. Esta expresión también es válida en buena aproximación en un inercial local, porque recuperamos Minkowski (principio de equivalencia), aunque en la práctica aplica a regiones más grandes del espacio (dentro del radio de Hubble) porque la curvatura, si la hubiera, no es notable

Ahora queremos generalizar cómo medir distancias, y reinterpretarlo, porque cuando hay expansión y curvatura ya no es trivial. Los dos problemas que salen son (1) ¿cómo defino la distancia desde donde estoy hasta un objeto si la geometría no es euclídea?; (2) ¿cómo defino la distancia desde donde estoy hasta un objeto si la distancia se redefine a cada instante de tiempo?

## 3.1 Distancia por luminosidad aparente

- Generalizamos la expresión de $l$, ahora admitiendo que el Universo está en expansión y que la geometría es RW

- La intensidad aparente que mida yo (i.e: en mi posición y a tiempo $t_0$, que es cuando me llega la señal) será: $l=\frac{L_{yo}}{A_{yo}}$

Denominador: el área que cubra la onda cuando me llegue a mí será $A_{yo}=4\pi r_1^2a_0^2$

Numerador: intento ponerlo en términos de $L_{1}$, que es lo que sé medir (cuánto emite la fuente). Tengo entonces: $L_{yo}=\frac{dE_{yo}}{dt_0}$, donde $dE_{yo}=dE_1\frac{a(t_1)}{a_0}$. De este modo: $L_{yo}=\frac{dE_{yo}}{dt_1}\frac{dt_1}{dt_0}=\frac{dE_1}{dt_1}\left(\frac{a(t_1)}{a_0}\right)^2$. Sustituyendo todo y haciendo $a_0=1$:

$$\boxed{l=\frac{La^2(t_1)}{4\pi r_1^2}=\frac{L}{4\pi r_1^2(1+z)^2}}$$

De este  modo, definimos la "distancia por luminosidad" como $\boxed{d_L=r(1+z)}$, pero no es la distancia física como tal porque no va con $\chi$ sino con $r$, que mide distancias transversales, no longitudinales

## 3.2 Distancia por diámetro angular

En caso euclideo y sin expansión: $\theta=\frac{D}{d}$. En general, supongamos que el objeto tiene un cierto tamaño $D(t_1)$ en el instante de emisión. Como este es una longitud transversal, por la métrica de FRW tenemos que es $D(t_1)=ra(t_1)\Delta\theta$, pero también sabemos que la distancia por diámetro angular se define en analogía al caso euclídeo como: $d_A=\frac{D}{\theta}$. Comparando, tenemos que:

$$\boxed{d_A=ra(t_1)=\frac{r}{1+z}}$$

**Def.:** si conocemos el tamaño de un objeto en el Universo, lo llamamos **regla estándar** porque a partir de él podemos calcular su distancia por diámetro angular
