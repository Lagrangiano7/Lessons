# 1. Spinores de Weyl

**Proposición:** el álgebra de Lie del grupo de Lorentz propio ortocrono se puede expresar como:

$$\mathfrak{so}^+(1, 3)_C=\mathfrak{su}(2)_C\oplus \mathfrak{su}(2)_C$$

**Dem.:**

Podemos hacer combinaciones lineales COMPLEJAS de los generadores del grupo de Lorentz para llegar a que esta es la suma directa de dos ideales de dimensión 3 (como no podía ser de otra forma por el teorema de descomposición en ideales). Concretamente:

$$\begin{cases}[J_i^+, J_j^+]=i\varepsilon_{ijk}J_k^+ \\ [J_i^-, J_j^-]=i\varepsilon_{ijk}J_k^- \\ [J_i^+, J_j^-]=0\end{cases}$$

Estos ideales son, precisamente, $\mathfrak{su}(2)$. El único problema es que antes el álgebra de Lorentz era real, y ahora hemos complejificado para poder separarla. Esta complefijicación da muchos problemas después

Copiando el razonamiento del caso escalar, esperaríamos que los fermiones de S=1/2 sean excitaciones de campos de igual spin, esto es, irreps. $(1/2, 0)$ ó $(0,1/2)$ del grupo de Lorentz PO. Veremos que no, pero estas irreps. serán útiles para definir los spinores de Dirac

**Def.:** un "spinor de Weyl levógiro" es aquel campo que se transforma en la irrep. $(1/2, 0)$ del grupo de Lorentz propio ortocrono. El "dextrógiro" será el que se transforme en $(0,1/2)$

**Generadores:** en la irrep. $(1/2,0)$ tenemos entonces que $J_i^+=0, \ J_i^- = \frac{\sigma_i}{2} \Longrightarrow \begin{cases}J_i=\frac{\sigma_i}{2} \\ K_i=i\frac{\sigma_i}{2}\end{cases}$. Pues ya sabemos cómo son las transformaciones Lorentz, que van con la exponencial

$$\boxed{\Psi_{R/L}\to e^{\frac{1}{2}(i\theta_i\sigma_i\pm \beta_i\sigma_i)} \Psi_{R/L}}$$

**Proposición:** estas irreps. NO son unitarias (porque los generadores de los boosts van con exponencial real) $\Longrightarrow \Lambda_{R/L}^\dagger \Lambda_{R/L}\neq I$

# 2. Construyendo una teoría para fermiones

No hay tantas teorías que podamos construir solo con spinores de Weyl, y ninguna de ellas es Lorentz-invariante. Vamos a tener que empezar a construir cosas que sí lo sean. Concretamente, por ser teoría libre, quiero que mi Lagrangiano tenga: (1) término cinético y (2) término de masas

## 2.1 Término de masas

**Proposición:** $\boxed{\Lambda_L^\dagger \Lambda_R = \Lambda_R^\dagger \Lambda_L = I}$ (de inmediata comprobación)

Por ende, el único término escalar-Lorentz que podemos construir en los campos sin meter derivadas es: $\Psi_L^\dagger \Psi_R$ (o al revés). Como no tenemos motivo para preferir uno sobre el otro, metemos los dos por igual:

$$\mathcal L_m = -m(\Psi_L^\dagger \Psi_R+\Psi_R^\dagger \Psi_L)$$

## 2.2 Término cinético

- Sabemos que debe ir con una derivada, de modo que sea lo que sea que contraigamos con ella, debe ser un cuadrivector
- Tenemos que construir ahora una 4-tupla que se transforme como un cuadrivector. ¿Cómo es esto?:

$$\delta w_0=\beta_i w_i$$

$$\delta w_i = \beta_iw_0-\varepsilon_{ijk}\theta_j w_k$$

**Proposición:** $\Psi_R^\dagger \Psi_R$ se comporta como la parte temporal de un cuadrivector: $\Psi_R^\dagger \Psi_R\to \beta_i\Psi_R^\dagger \sigma_i\Psi_R$. Esto también nos dice directamente cuál es la parte espacial del cuadrivector. Habrá que comprobar que efectivamente se transforma como tal, y sí lo hace

**Observación:** $(\Psi_R^\dagger \Psi_R, \Psi_R^\dagger \sigma_i\Psi_R)$ se transforma como un cuadrivector, así que si lo contraigo con derivada me da un escalar que puedo usar para construir el término cinético

**Nota:** para mantener la notación covariante, definimos $\sigma^\mu=(I, \sigma_i)$ y $\bar\sigma^\mu=(I, -\sigma_i)$, así que ahora tenemos:

$$\mathcal L_K=i(\Psi_R^\dagger \sigma_\mu \partial^\mu\Psi_R)+i(\Psi_L^\dagger \sigma_\mu \partial^\mu\Psi_L)$$

**Nota:** podemos contraer más aún la notación y agrupar: $\gamma^\mu = \begin{pmatrix}0 & \sigma^\mu \\ \bar\sigma^\mu & 0\end{pmatrix}$ así como $\Psi=\begin{pmatrix}\Psi_L \\ \Psi_R\end{pmatrix}$ y tenemos:

$$\mathcal L = \bar \Psi(i\cancel \partial - m)\Psi$$

**Observación:** al construir así el spinor de Dirac, lo que estamos haciendo realmente es tomar la suma directa de las dos representaciones de Weyl, de modo que el spinor de Dirac se transforma en la representación $(1/2,0)\oplus(0,1/2)\equiv (1/2, 1/2)$ del grupo de Lorentz PO. Véase que el electrón sigue siendo una partícula elemental a pesar de que el campo que lo crea y destruye no se transforme en una irrep., porque lo que se tiene que transformar en irreps. de acuerdo a la clasificación de Wigner son los estados, no los campos que los crean / destruyen

**Proposición:** cada componente de $\Psi$ cumple por separado una ecuación de K-G. Esto es un resultado de la ec. de Dirac, y no son equivalentes (ec. de Dirac es más general que K-G por componentes, porque solo puedo hacer la derivación en un sentido):

$$(\square+m^2)\Psi=0$$

## 2.3 ¿Cómo se transforma el spinor de Dirac?

- Como el spinor de Dirac es la suma directa de los de Weyl, sabemos por teoría de representaciones que se transforma como: $\Psi\to \begin{pmatrix}\Lambda_L & 0 \\ 0& \Lambda_R\end{pmatrix}\Psi$

**Proposición:** también podemos definir el generador de las transformaciones del spinor de Dirac como: $S^{\mu\nu}=\frac{i}{4}[\gamma^\mu, \gamma^\nu]$, de modo que ahora tenemos $\Lambda_S = e^{i\theta_{\mu\nu}S^{\mu\nu}}$, donde:

$$S^{ij}=\frac{1}{2}\varepsilon_{ijk}J^k$$

$$S^{0i}=K^i$$

de modo que se recupera la forma habitual del grupo de Lorentz

**Propiedades de las matrices $\gamma$:**

1. $\{\gamma^\mu, \gamma^\nu\}=2\eta^{\mu\nu}$
2. $\Lambda_S^{-1}\gamma^\mu \Lambda_S= \Lambda^\mu_\nu \gamma^\nu$
3. $\Lambda_S(2\pi)=-I$

# 3. Estructuras algebraicas de los campos fermiónicos

Los spinores de Dirac son operadores de campo, y como tales, podemos verlos de 2 formas:

1. El conjunto de todas las COMPONENTES de $\Psi$'s y $\Psi^\dagger$'s forman un álgebra de operadores, cuya operación interna es la composición. Sobre ella, se pueden definir relaciones de (anti)conmutación, que son las que obedecen estos operadores

- $\{\Psi\}$ forma espacio vectorial
- $\{\bar\Psi\}$ es su espacio dual
- $\gamma^0$ se puede entender como una métrica, aunque no es euclídea (de hecho es Lorentziana), y como tal conecta $S\leftrightarrow S^*$

**Nota:** a priori, parecería que como $\Psi^\dagger$ y $\Psi$ ya se pueden contraer, no haría falta la métrica. Sin embargo, $\Psi^\dagger\Psi$ NO es Lorentz-invariante, de modo que no tendría sentido introducirlo en nuestras teorías. Sin embargo, $\Psi^\dagger\gamma^0\Psi$ sí es Lorentz-invariante. De este modo, tenemos que $\gamma^0(\Psi, \Psi):=\Psi^\dagger \gamma^0\Psi$ es un escalar de Lorentz

# 4. Invariancia gauge: QED

**Observación:** el Lagrangiano de Dirac tiene es invariante bajo un cambio de fase global: $U(1)$. Vamos a imponer ahora que este cambio de fase no se dé igual en todos los eventos del espacio tiempo, sino que sea distinto en cada uno. Esto es, la fase depende ahora del punto de la variedad

El Lagrangiano tal y como está ahora mismo no es invariante gauge, sino que cambia como: $\mathcal L\to \mathcal L +\bar\Psi\cancel\partial\alpha\Psi$

**Nota:** debemos modificar el operador del término cinético para que se trague este término extra si queremos forzar que nuestra teoría sea gauge-invariante. Por ende, debemos buscar un operador $D_\mu$ (derivada covariante) que podamos sustituir en el término cinético, que cumpla:

$$\mathcal L'=\mathcal L\Longrightarrow e^{i\alpha(x)}D'_\mu(e^{-i\alpha(x)})\Psi+D'_\mu\Psi=D_\mu\Psi$$

Probemos entonces con la forma más sencilla, que es $D_\mu = \partial_\mu + ieA_\mu$, donde $A_\mu$ es cualquier cuadrivector y $e$ es una constante. Si lo introducimos en el Lagrangiano e imponemos invariancia gauge, vemos que la derivada covariante debe transformarse de tal forma que cancele el factor $\alpha$ que nos sale de la derivada. Lógicamente, esta transformación debe venir del cuadrivector que hemos añadido a la derivada normal, así que este debe transformarse como: $A_\mu \to A_\mu-ie\partial_\mu\alpha$

Inmediatamente comprobamos que el cuadripotencial del campo EM es el candidato perfecto para este rol, de modo que hacemos la teoría invariante de la forma:

$$\mathcal L = \bar\Psi(i\cancel D-m)\Psi$$

**Intuición:** el juego que hacemos es el siguiente: cuando metemos un cambio de fase local, la derivada nos saca un factor extra que nos sobra si la teoría ha de ser invariante bajo estas transformaciones. Asimismo, como la Física debe ser independiente de las transformaciones gauge que hagamos en el cuadripotencial, cada vez que hagamos un cambio de fase local hacemos asimismo una transformación gauge en el cuadripotencial por el opuesto de esa fase. Entonces, la derivada se cancela exactamente y la teoría es invariante gauge

**Observación:** el hecho de introducir la derivada covariante mete en el Lagrangiano un término del tipo: $e\bar\Psi A_\mu \Psi$. Esto es precisamente un vértice a 3 campos. De este modo, **imponer invariancia gauge de una teoría sirviéndose de otro campo introduce una interacción entre las dos teorías**

## 4.1 QED

- Hemos identificado el campo que metemos en la derivada covariante como el cuadripotencial del campo EM. Sin embargo, no estamos teniendo en cuenta su dinámico
- Pero esta la conocemos de la teoría libre del campo EM. El término de interacción ya está metido también, así que se la podemos añadir y ya tenemos la teoría completa (full interacting picture) de la interacción electromagnética entre fermiones:

$$\mathcal L=\bar\Psi(i\cancel D-m)\Psi-\frac{1}{4}F^{\mu\nu}F_{\mu\nu}$$

# 5. Helicidad y quiralidad

**Def.:** la "quiralidad" es una propiedad de los spinores de Weyl: es sencillamente si son L ó R (más técnicamente, en qué irrep. del grupo de Lorentz PO se transforman)

Podemos construir un proyector ortogonal $P_{R/L}:=\frac{1\pm \gamma^5}{2}$ tal que proyecta sobre una de las dos componentes de quiralidad bien definida de un spinor de Dirac: $\Psi_{R/L}=P_{R/L}\Psi$, cosa que es trivial porque es suma directa de estos

Desacoplando la ec. de Dirac, queda:

$$\begin{pmatrix}-m & i\sigma^\mu \partial_\mu \\ i\bar \sigma^\mu\partial_\mu & -m\end{pmatrix}\begin{pmatrix}\Psi_L \\ \Psi_R\end{pmatrix}=0$$

**Observación:** el término de masa hace que la dinámica de sendas quiralidades se acople. En los neutrinos del SM, por ejemplo, como son massless, están desacopladas y solo observamos los Left

**Def.:** la "helicidad" es la proyección del spin de un campo sobre su momento lineal. Para S=1/2

$$\hat h:=\frac{\vec\sigma\cdot\vec p}{|\vec p|}$$

**Observación:** a diferencia de la quiralidad, que es una propiedad inherente al campo, la helicidad es un observable que dependerá del inercial en el que se mida, a menos que el campo de la teoría sea massless, como es el caso del fotón. De hecho, los spinores de Weyl NO son autoestados de la helicidad salvo que sean massless

**Propiedad:** en el límite quiral ó en el ultrarrelativista, helicidad y quiralidad se corresponden uno a uno y son, en este sentido, la misma cosa:

$$\begin{gather}\hat h\Psi_R=+\Psi_R \\ \hat h\Psi_L=-\Psi_L\end{gather}$$

**Def.:** decimos que una teoría es "quiral" si NO es simétrica bajo intercambio $L\leftrightarrow R$ / si los spinores de Weyl de distinta quiralidad pueden transformarse por separado de forma independiente / si no se mezclan quiralidades

**Observación:** QED y QCD NO son teorías quirales porque el término de masas mezcla spinores.

**¿Cuándo es una teoría de fermiones quiral?:**

- Una teoría de fermiones siempre va a tener la pinta: $\mathcal L = \bar \Psi(i\cancel D-m)\Psi$

- Para estudiar si es o no quiral, debemos estudiar si alguna de las dos partes de la teoría mezcla quiralidades

**Término de masa:** mezcla quiralidades

$m\bar \Psi \Psi=m(\Psi_L^\dagger \ \Psi_R^\dagger)\gamma^0 \begin{pmatrix}\Psi_L \\ \Psi_R\end{pmatrix} = m(\Psi_L^\dagger \Psi_R + \Psi_R^\dagger \Psi_L)$

**Observación:** por eso el SM no puede tener términos de masa introducidos explícitamente en la teoría, porque la interacción débil ha de ser quiral y los términos de masa rompen la quiralidad de la teoría

**Término cinético:** no mezcla quiralidades a menos que $D$ lleve una matriz que sí mezcle

$\bar \Psi \cancel D \Psi = (\Psi_R^\dagger, \Psi_L^\dagger) \begin{pmatrix}0 & \sigma^\mu \\ \bar \sigma^\mu & 0\end{pmatrix}D_\mu \begin{pmatrix}\Psi_L \\ \Psi_R\end{pmatrix} = \Psi_R^\dagger \sigma^\mu D_\mu \Psi_R+\Psi_L^\dagger \bar\sigma^\mu D_\mu \Psi_L$

Por ende, la derivada covariante puede tener la forma que se desee, que no va a mezclar quiralidades salvo que sea matricial y coincida que tenga una forma tal que sí mezcle

# 6. Soluciones de la ec. de Dirac libre (clásica)

Históricamente, Dirac introduce su ecuación para tratar de resolver el problema de la "distribución de probabilidad" que no es definida positiva --> lo consigue, pero sigue teniendo el problema de energías negativas

Sabemos que cada componente del spinor de Dirac cumple una ec. de Klein-Gordon, así que la solución a esto debe ser:

$$\Psi(x)=\int \frac{d^3\vec p}{(2\pi)^32E_p}u^s(p)e^{-ip\cdot x}$$

$$\chi(x)=\int \frac{d^3\vec p}{(2\pi)^32E_p}v^s(p)e^{+ip\cdot x}$$

Ambas soluciones tienen energía positiva, solo que una va con frecuencias positivas y la otra con negativas

Pero los spinores también deben cumplir la ec. de Dirac, que es más general, así que imponemos que la cumplan. En el inercial de centro de inercia, $p\cdot \sigma = mI$, así que, en representación de momentos:

$$\begin{pmatrix}-1 & 1 \\ 1 & -1\end{pmatrix}u^s(p)=0$$

$$\begin{pmatrix}-1 & -1 \\ -1 & -1\end{pmatrix}v^s(p)=0$$

Por ende:

$$u^s(p)=\begin{pmatrix}\xi^s \\ \xi^s\end{pmatrix}; \ \ v^s(p)=\begin{pmatrix}\eta^s \\ -\eta^s\end{pmatrix}$$

Con la elección más sencilla, me queda entonces que en el CM:

$$\boxed{\begin{gather}u_\uparrow(p)=\sqrt{m}\begin{pmatrix}1 \\ 0 \\ 1 \\ 0\end{pmatrix} \\ u_\downarrow(p)=\sqrt{m}\begin{pmatrix}0 \\ 1 \\ 0 \\ 1\end{pmatrix} \\ v_\uparrow(p)=\sqrt{m}\begin{pmatrix}-1 \\ 0 \\ 1 \\ 0\end{pmatrix} \\ v_\downarrow(p)=\sqrt{m}\begin{pmatrix}0 \\ 1 \\ 0 \\ -1\end{pmatrix}\end{gather}}$$

Y qué pasa en un inercial arbitrario? (i.e: haciendo un boost a esta solución)

$$\boxed{\begin{gather}u^s(p)=\sqrt{m}\begin{pmatrix}\sqrt{p\cdot \sigma}\xi^s \\ \sqrt{p\cdot \bar\sigma}\xi^s\end{pmatrix} \\ v^s(p)=\sqrt{m}\begin{pmatrix}\sqrt{p\cdot \sigma}\eta^s \\ -\sqrt{p\cdot \bar\sigma}\eta^s\end{pmatrix}\end{gather}}$$

donde podemos tomar $\xi^s, \eta^s$ como antes

**Nota:** en general, como las soluciones involucran tomar la raíz cuadrada de una matriz, no queda más remedio que irse al CM y luego volverse al inercial original. Sin embargo, como $\sigma_z$ es diagonal, si el boost del CM es en dir. z, nos sigue quedando algo diagonal en la raíz y no ace falta hacer esto

**Normalización:**

$$\boxed{\begin{gather}\bar u^s u^{s'}=2m\delta_{ss'} \\ \bar v^s v^{s'}=-2m\delta_{ss'} \\ \bar u^s v^{s'}=\bar v^s u^{s'}=0\end{gather}}$$

**Propiedad:**

$$\boxed{\begin{gather}\displaystyle\sum_{s=1}^2u^s\bar u^s=\cancel p + m \\ \displaystyle\sum_{s=1}^2v^s\bar v^s=\cancel p - m\end{gather}}$$

# 7. Cuantización del campo fermiónico

Ya tenemos las soluciones de la ec. de Dirac así que ahora cuantizamos normal:

$$\Psi(x)=\displaystyle\sum_{s=1}^2\int \frac{d^3\vec p}{(2\pi)^3\sqrt{2E_p}}[a_p^su^s(p)e^{-ip\cdot x}+(b_p^s)^\dagger v^s(p)e^{ip\cdot x}]$$

$$\bar \Psi(x)=\displaystyle\sum_{s=1}^2\int \frac{d^3\vec p}{(2\pi)^3\sqrt{2E_p}}[(a_p^s)^\dagger \bar u^s(p)e^{ip\cdot x}+b_p^s\bar v^s(p)e^{-ip\cdot x}]$$

**Proposición:** los operadores de creación / destrucción ANTIconmutan:

$$\boxed{\{a_p^s, (b_{p'}^{s'})^\dagger\}=2E_p(2\pi)^3\delta^3(\vec p-\vec p \ ')\delta_{ss'}}$$

**MUCHÍSIMO OJO!:** las relaciones de anticonmutación de los campos vienen a raíz de los operadores de creación y destrucción. Los $u,v$ y sus barrados que vayamos a usar luego en el cálculo de las amplitudes son coeficientes normales, no obedecen ninguna relación de anticonmutación porque NO SON operadores

# 8. Propagador fermiónico libre

**Nota:** como los campos ahora anticonmutan, debemos poner un signo "-" en el propagador de Feynman:

$$\boxed{(S_F)_\alpha\beta:=\langle 0|T(\Psi_\alpha(y)\bar\Psi_\beta(x))= \Theta(x^0>y^0)\langle 0|\Psi_\alpha(y)\bar \Psi_\beta(x)|0\rangle - \Theta(y^0>x^0)\langle 0|\bar \Psi_\beta(x)\Psi_\alpha(y)|0\rangle}$$

**Observación:** a veces se prescinde de los índices espinoriales al escribir este propagador, pero entonces fíjate que al intercambiar los campos por ordenación temporal cuando toque, tendremos un escalar Lorentz o una matriz dependiendo de en qué orden vayan. La interpretación correcta es como está aquí, por índices del espacio de Dirac

**Proposición:** el propagador de Feynman de la teoría fermiónica libre es:

$$\boxed{S_F(x-y)=-(i\cancel\partial-m)\int\frac{d^4p}{(2\pi)^4} \frac{ie^{ip\cdot(x-y)}}{p^2-m^2+i\varepsilon}}$$

# 9. Operadores discretos en QFT

Así como podemos hacer transformaciones continuas y de gauge a los campos, también podemos someterlos a transformaciones discretas. Observo que habitualmente estas suelen ser reps. de $C_2$ aunque cada una tiene su propia forma de actuar


## ¿Cómo actúan los operadores discretos sobre campos?

- La mayoría de operadores discretos son del tipo $\mathcal O^2=I$, de modo que con representaciones de un grupo $C_2=\{e, \mathcal O\}$
- Por otro lado, sabemos que los campos viven en un álgebra de operadores y, a otro nivel, en un espacio vectorial
- Si queremos actuar este operador discreto sobre los campos, recordemos que la representación debe ser rep. sobre un álgebra, de modo que debe cumplir:

1. $\rho(\mathcal O)(\alpha\Psi_1+\beta\Psi_2)=\alpha_1\rho(\mathcal O)\Psi_1 + \beta\rho(\mathcal O)\Psi_2$
2. $\rho(\mathcal O)(\bar\Psi_1\Psi_2)=\rho(\mathcal O)(\Psi_1)\rho(\mathcal O)(\Psi_2)$
3. $\rho(\mathcal O)(\{\Psi_1, \Psi_2\})=\{\rho(\mathcal O)\Psi_1, \rho(\mathcal O)\Psi_2\}$

## Conjugación de carga

**Def.:** se define el operador "conjugación de carga" $C$ en spinores como:

$$\boxed{C\Psi\equiv \Psi_c=-i\gamma_2\Psi^*}$$

**Propiedades:**

1. $C^2=I$
2. Transforma partículas a antipartículas (e invierte el spin)
3. El condensado es invariante bajo conjugación de carga: $C(\bar\Psi\Psi)=\bar\Psi_c\Psi_c=\bar\Psi\Psi$
4. $C(\bar\Psi\gamma^\mu\Psi)=-\bar\Psi\gamma^\mu\Psi$

**Observación adicional:** el condensado es una cantidad real

**Dem:**

1. $C(C\Psi)=-i\gamma_2(C\Psi)^*=-i\gamma_2(-i\gamma_2\Psi^*)^*=-i\gamma_2i\gamma_2^*\Psi=\Psi$
2. Hay 2 formas de llegar a este resultado: (1) viendo que $\Psi_c$ se acopla al campo EM como la antipartícula (i.e: con carga de signo opuesto); (2) viendo cómo actúa sobre las soluciones de la ec. de Dirac libre

(1)

- Sea un leptón acoplado a la interacción EM. Entonces, la ec. de Dirac para el campo correspondiente es: $[i(\cancel \partial +i e\cancel A)-m]\Psi=0$

- Ahora tratemos de poner la ec. en términos de $\Psi_c$ para ver qué ecuación obedece

- Paso 1: conjugamos complejo toda la ec: $(-i\gamma^{*\mu}\partial_\mu - e\gamma^{*\mu}A_\mu-m)\Psi^*=0$

- Paso 2: $\Psi_c=-i\gamma_2\Psi^* \Longrightarrow \Psi^*=-i\gamma_2\Psi_c$ --> sustituimos

- Paso 3: finalmente actuamos con $\gamma_2$ por la izquierda, porque $\gamma_2\gamma^{*\mu}\gamma_2=\gamma^\mu$

- Resultado:


$$[i(\cancel\partial-ie\cancel A)-m]\Psi_c=0$$

Como el campo $\Psi_c$ se acopla a la interacción EM con carga opuesta, es razonable sospechar que puede estar cambiando partículas $\leftrightarrow$ antipartículas

(2)

Nos vamos al CM y estudiamos cómo actúa sobre las soluciones de la ec. de Dirac libre. Esto me parece más convincente porque (1) se ve el efecto explícitamente y (2) no depende de ninguna interacción, porque el razonamiento anterior podría en principio funcionar para QED pero tal vez no para QCD y habría que ir probando

Las transformaciones (en CM por simplicidad) son:

$$\boxed{\begin{gather}u_\uparrow \leftrightarrow v_\downarrow \\ u_\downarrow \leftrightarrow v_\uparrow \end{gather}}$$

3. Vemos cómo se transforma el condensado:

- Primero estudiemos cómo se transforma el spinor barrado: $\bar\Psi_c=\bar\Psi_c^\dagger\gamma^0=[C\Psi]^\dagger \gamma^0=[-i\gamma_2\Psi^*]^\dagger \gamma^0=+i\Psi^t\gamma_2^\dagger\gamma^0$
- Por ende: $\bar\Psi_c=i\Psi^t\gamma_2\gamma^0$
- Ahora desarrollemos $\bar\Psi_c\Psi_c=i\Psi^t\gamma_2\gamma^0(-i\gamma_2)\Psi^*$
- Tenemos las propiedades: $\gamma_2\gamma^0\gamma_2=\gamma^0$ y $(\gamma^0)^t=\gamma^0$, por lo que la expresión queda como:
- $\bar\Psi_c\Psi_c=\Psi^t\gamma^0\Psi^*$
- Expresándolo como un traspuesto: $=(\Psi^\dagger \gamma^0\Psi)^t$ y recordando que es un número, por lo que $c^t=c$, hemos acabado

Condensado es real:

- Si es real, su parte imaginaria es 0, y por ende $Im[\bar\Psi\Psi]\propto\bar\Psi\Psi-(\bar\Psi\Psi)^*=0$
- Desarrollemos: $=\Psi^\dagger \gamma^0\Psi-\Psi^t\gamma^0\Psi^*$
- Haciendo el mismo argumento de antes, como el segundo sumando es un número, debe ser igual a su traspuesta, así que la expresamos así, y queda justo el primer sumando --> hemos terminado, sí da 0 $_\blacksquare$

**Proposición:** la teoría fermiónica libre es invariante bajo conjugación de carga (como cabría esperar, porque no hay nada que distinga partículas de antipartículas)


**Proposición:** cuando introducimos acoplo a un campo EM, esta invariancia no tiene por qué darse, porque tenemos un término de interacción que va como: $\bar\Psi\cancel A\Psi$, y sabemos que $C(\bar\Psi\gamma^\mu\Psi)=-\bar\Psi\gamma^\mu\Psi$. Todo depende de cómo se transforme el campo compensador

**Solución:** si queremos imponer que nuestra teoría con interacción sea invariante bajo conjugación de carga, tal y como se observa empíricamente, entonces debemos pedir que el cuadripotencial se transforme como:

$$\boxed{C(A_\mu)=-A_\mu}$$

**IMPORTANTE:** mucho más fácil es trabajar con la conjugación de carga como sigue (además es independiente de la representación (Dirac / Weyl / otra) que estemos usando):

**Def.:** $\boxed{\Psi\xrightarrow{C}C\bar\Psi^T}$

**Propiedades:**

1. $C^\dagger = C^{-1}$
2. $\boxed{C^{-1}\gamma^\mu C^=-(\gamma^\mu)^T}$
3. $C\gamma^5C^{-1}=(\gamma^5)^T$ (inmediatamente)
4. $\bar \Psi \xrightarrow{C}\Psi^TC^{-1}$

**Ejemplo:** determinar cómo se transforma el bilineal $\bar \Psi\gamma^\mu \Psi$ bajo conjugación de carga:

$\bar\Psi\gamma^\mu \Psi\to \Psi^T(C^{-1}\gamma^\mu C)\bar\Psi^T= -\Psi^T(\gamma^\mu)^T\bar\Psi^T=-\bar\Psi\gamma^\mu \Psi$ --> impar bajo conjugación de carga (esto es lógico: esta corriente está asociada a la electromagnética, y como las antipartículas tienen carga opuesta que las partículas ordinarias, tanto la densidad de carga como la densidad de corriente van a cambiar en un signo)

## Paridad

**Def.:** el operador de "paridad" es aquel que invierte la dependencia espacial de los campos

### Sobre campos escalares

En campos escalares: $P\phi(t, \vec x)=\phi(t, -\vec x)$

Si estos campos tienen paridad espacial bien definida, entonces serán autoestados del operador paridad: $P\phi(t, \vec x)=\pm \phi(t, \vec x)$

**Def.:** campos escalares con paridad par: **escalares**; escalares con paridad impar: **pseudoescalares**

**Observación:** $P^2=I$ de nuevo, de modo que tenemos otra vez un grupo cíclico de dos elementos: $C_2=\{e, P\}$

### Sobre spinores

**Proposición:** el operador de paridad invierte la quiralidad del spinor de Weyl:

$$\boxed{\begin{gather}P\Psi_{L/R}=\Psi_{R/L}\end{gather}}$$

Por ende, sobre spinores de Dirac actuará intercambiando los dos spinores de Weyl de la suma directa. Este es precisamente el efecto de $\gamma^0$, así que:

$$\boxed{P\Psi=\gamma^0\Psi}$$

**Dem:**

El argumento es como sigue:

1. Yo sé cómo se transforma Lorentz un spinor de Weyl
2. Yo sé cómo se transforman los generadores del grupo de Lorentz bajo paridad
3. Aplico paridad al spinor transformado Lorentz, y si veo que este nuevo campo se transforma con quiralidad opuesta, he acabado

- $\Psi_L\to (1+\theta_i J_i-\beta_i\sigma_i)\Psi_L$
- Escojo rep. del grupo $C_2$ tal que $P=\begin{pmatrix}1 & & & \\ & -1 & & \\ & & -1 & \\ & & & -1\end{pmatrix}$, y veo que $[P, J_i]=0$ y que $\{P, K_i\}=0$

- Dicho de otra forma: $PJ_iP^{-1}=J_i$ y $PK_iP^{-1}=-K_i$

- Ahora aplico paridad al spinor de Weyl:

$$P\Psi_L'=(1+\theta_iPJ_i-\beta_i PK_i)(P^{-1}P)\Psi_R$$

Desarrollando la expresión, me queda entonces que:

$$P\Psi_L'=(1+\theta_iJ_i+\beta_i K_i)(P\Psi_R)$$

Conclusión: $P\Psi_L$ se transforma precisamente como $\Psi_R$, así que debe ser $P\Psi_L=\Psi_R$. Un argumento completamente equivalente da el mismo resultado para la otra quiralidad

**Propiedades:**

1. El condensado es par: $P(\bar\Psi\Psi)=\bar\Psi\Psi$
2. Los cuadrivectores "copian" a las coordenadas, de modo que si $Px^\mu=(t, -\vec x)$, entonces esperaríamos que los cuadrivectores se transformen así bajo paridad. Si esto es cierto, el cuadrivector es **vectorial**, y si es al revés, diremos que es **axial**
3. La corriente vectorial $V^\mu=\bar\Psi\gamma^\mu\Psi$ es vectorial
4. La corriente axial $A^\mu=\bar\Psi\gamma^\mu\gamma^5\Psi$ es axial
5. El Lagrangiano de la teoría fermiónica libre es invariante bajo paridad
6. El lagrangiano de QED es invariante bajo paridad si y solo si el cuadripotencial es vectorial bajo paridad (imponemos que lo sea, porque observamos que QED sí parece ser inv. bajo paridad)

**Dem:**

1. Condensado es par

$P(\bar\Psi)=[P\Psi]^\dagger\gamma^0 = \Psi^\dagger (\gamma^0)^2=\Psi^\dagger$

Por ende, $P(\bar\Psi\Psi)=\bar\Psi\Psi$

3. Definamos $V^\mu=\bar\Psi \gamma^\mu\Psi$. Si su componente temporal es invariante bajo paridad y la espacial cambia de signo, hemos terminado

Recordemos que $P(\bar\Psi)=\Psi^\dagger$

$j^0=\bar\Psi\gamma^0\Psi=\Psi^\dagger\Psi$: $P(j^0)=\Psi^\dagger \gamma^0(\gamma^0\Psi)=\Psi^\dagger\Psi=j^0$

$j^i=\bar\Psi\gamma^i\Psi$: $P(j^i)=\Psi^\dagger\gamma^i\gamma^0\Psi=-\Psi^\dagger\gamma^0\gamma^i\Psi=-j^i$

donde he usado que $\{\gamma^\mu, \gamma^\nu\}=2\eta^{\mu\nu}$, así que $\gamma^0$ anticonmuta con las gammas espaciales

4. Misma lógica que antes: $A^\mu=i\bar\Psi\gamma^\mu\gamma^0\gamma^1\gamma^2\gamma^3\Psi$

$A^0=i\bar\Psi\gamma^1\gamma^2\gamma^3\Psi$ --> $P(A^0)=i\Psi^\dagger \gamma^1\gamma^2\gamma^3\gamma^0\Psi$

Como la gamma temporal y las espaciales anticonmutan, y tenemos que pasar la $\gamma^0$ 3 veces hacia la izquierda, esto nos saca un factor $(-1)^3=-1$, así que $P(A^0)=-A^0$

$A^i=i\bar\Psi\gamma^i\gamma^0\gamma^1\gamma^2\gamma^3\Psi$ --> $P(A^i)=i\Psi^\dagger \gamma^i\gamma^0\gamma^1\gamma^2\gamma^3\gamma^0\Psi$

Al pasar $\gamma^0$ hacia la izquierda, las 3 primeras gammas espaciales me meten un signo "-" cada una --> $(-1)^3$, luego consigo misma conmuta trivialmente y después nos queda pasar por una última espacial, así que $(-1)^4=1$. Por ende, $P(A^i)=A^i$ y la corriente es en efecto axial

5. El término de masas es invariante por (1). El término cinético es invariante porque la paridad también actúa sobre la derivada, cambiando el signo de la parte espacial. Como la conmutación de $\gamma^0$ con las espaciales mete un signo "-" y consigo misma conmuta, ya está:

$\mathcal L_K=i\bar\Psi\cancel\partial_\mu \Psi \mapsto i\Psi^\dagger\gamma^\mu( -\delta_{\mu, (xyz)})\partial_\mu\gamma^0\Psi$

donde la delta es notación abreviada para el producto de 3 deltas (i.e: las componentes espaciales introducen un signo "-"). Véase que no podemos simplemente ver que hay 3 coordenadas que cambian de signo e introducir un signo "-" global porque $(-1)^3=-1$, porque estamos siguiendo el convenio de sumación de Einstein. Es decir, esto quiere decir que los sumandos de gamma*derivada cambian de signo en la suma, no sale ningún signo "-" global

Estos signos "-" de las coordenadas espaciales de la derivada se cancelan con los signos "-" que salen de conmutar $\gamma^0$ con las gamma's espaciales que acompañan a cada término de esa derivada, de modo que finalmente tenemos: $P(\mathcal L_K)=\mathcal L_K$, así que la teoría fermiónica libre es en efecto invariante bajo paridad, como cabría esperar

6. Imponemos que QED sea inv. bajo paridad y veamos qué condición nos fija eso en el único "grado de libertad" que tenemos, que es el cuadripotencial (i.e: lo único que aún no hemos visto cómo se transforma bajo paridad). Lo único que cambia entre la teoría libre y la teoría interactuante (en acoplo mínimo) es que tenemos un término adicional de interacción, que va como $\bar\Psi\cancel A \Psi$ (el de la derivada covariante, vamos)

$P(\bar\Psi\gamma^\mu A_\mu)=\Psi^\dagger \gamma^\mu P(A_\mu)\gamma^0\Psi=\Psi^\dagger[\gamma^0P(A_0)\gamma^0 + \gamma^iP(A_i)\gamma^0]\Psi$

Fíjate: si imponemos que el transformado sea igual al original, entonces necesitamos que la primera parte no introduzca ningún signo y que la segunda sí lo haga, para contrarrestar el signo que va a meter el conmutar la $\gamma^0$ con las espaciales. Por ende, pidiendo $P(A_0)=A_0, P(A_i)=-A_i$, el término de interacción es un escalar:

$=\Psi^\dagger [A_0+\gamma^0\gamma^iA_i]\Psi$. Ahora metemos dos $\gamma^0$ delante del primer sumando (recuerda que $(\gamma^0)^2=I$), y tenemos entonces que: $=\bar\Psi\cancel A\Psi_\blacksquare$
