# 1. Formulación

**Def.:** QED es la teoría de la electrodinámica cuántica, que es una teoría obtenida a partir de coger el Lagrangiano libre de Dirac e imponer invariancia bajo la transformación $U(1)$ local. Esto nos fuerza a cambiar el término cinético para que sea invariante, introduciendo una derivada covariante. Esta derivada covariante lleva en las tripas el cuadripotencial del campo EM, que es una conexión del fibrado principal de $U(1)$. El tensor EM, $F=dA$, es el Riemann de dicho fibrado, de modo que solamente existe interacción EM si el fibrado tiene curvatura



# 2. Reglas de Feynman

Al pedir invariancia $U(1)$ local, introducimos la derivada covariante. Si la expandimos, esta añade un término adicional $\mathcal L_{int}=e\bar\Psi\cancel A\Psi$ al Lagrangiano fermiónico libre, de modo que introduce la famosa interacción de vértice a 3 campos

Tenemos entonces 3 elementos: líneas externas, líneas internas y vértices. Cada uno introduce un ingrediente y todos juntos forman la amplitud invariante del proceso que representa el diagrama. En este curso, nos quedaremos a nivel árbol

**Recuerda:** los diagramas de Feynman se leen de atrás hacia delante

1. Líneas externas:

- PO saliente: $\bar u^s$ / PO entrante: $u^s$
- AP saliente: $\bar v^s$ / AP entrante: $v^s$
- Fotón saliente: $(\epsilon_\mu^i)^*$ / Fotón entrante: $\epsilon_\mu^i$

2. Vértice: $-ie\gamma^\mu$
3. Líneas internas:

- PO: $\frac{i(\cancel p+m)}{p^2-m^2+i\varepsilon}=\frac{i}{\cancel p-m+i\varepsilon}$
- Fotón: $-i\frac{\eta_{\mu\nu}}{p^2+i\varepsilon}$

**Nota:** siempre dibujaremos los diagramas con las líneas internas siendo partículas ordinarias, nunca antipartículas. Esto es por convenio, pero si quisiéramos usar antipartículas como líneas internas, tampoco pasaría nada, solo que el propagador iría con $\cancel p-m$

**IMPORTANTE:** cuando tengamos diagramas en varios canales, **cada uno llevará el signo de la paridad de la permutación del diagrama**. Es decir, si tenemos un diagrama del tipo $[\bar u_3\gamma^\mu \bar u_1][\bar u_4\gamma_\mu u_2]$ entonces llevará un signo "-" porque $3142\to 1342\to 1324 \to 1234$ y sale un "-" (**NOTA:** parece que Jacobo usa el convenio contrario. De todos modos, esta determinación del signo dedice el signo del potencial efectivo a bajas energías así que conviene preguntar)

# 3. Identidad de Ward-Takahashi en QED

- Las identidades de Ward-Takahashi son la generalización a QFT del teorema de Noether
- En teoría clásica de campos, si la teoría tiene una simetría continua parametrizada con $n$ parámetros, entonces existen $n$ corrientes conservadas tales que $\partial_\mu J^\mu_n=0$
- En QFT ahora las corrientes son operadores, así que lo más cercano que podemos pedir es que el valor esperado de vacío de la derivada sea nulo: $\langle \partial_\mu J^\mu_n\rangle = 0$
- A veces, al cuantizar la teoría, algunas corrientes que en la contraparte clásica eran conservadas ahora ya no son homogeneas en la identidad de Ward-Takahashi. A esto se le llama **anomalía** (e.g: la corriente axial singlete de la simetría de isospin)
- En QED, la teoría es invariante bajo $U(1)$ global (y no es anómala), de modo que tenemos: $\langle \partial_\mu j^\mu\rangle = 0$
- **Observación crucial:** esta corriente es precisamente la que sale en los diagramas (los diagramas a tree level son el acoplo de dos corrientes conservadas). De este modo, lo que digamos aplicará no solamente a la invariancia $U(1)$ global sino también a la local. Por eso la identidad de Ward en QED está asociada a la invariancia gauge de las amplitudes calculadas

## 3.1 Invariancia gauge de la amplitud

- Sabemos que el cuadripotencial se puede transformar gauge como: $A_\mu\to A_\mu+\partial_\mu\alpha(x)$ sin cambiar la física. Equivalentemente, en espacio de momentos, si el fotón tiene momento $k^\mu$ entonces la transformación gauge es: $\epsilon_\mu\to \epsilon_\mu+\lambda p_\mu$
- **Observación:** la amplitud ha de ser invariante bajo la transformación gauge, y como los fotones reales se acoplan como: $\epsilon_\mu(\vec k)\mathcal M^{\mu}$, entonces si hacemos una transformación gauge esto debería seguir cumpliéndose:

$$(\epsilon_\mu(\vec k)+\lambda k_\mu)\mathcal M^{\mu}=\mathcal M + \lambda k_\mu M^{\mu}$$

Por ende, debemos imponer que:

$$\boxed{k_\mu M^{\mu}=0}$$

Esta es la llamada **identidad de Ward**

**OJO!:** perturbativamente, esta identidad se cumple orden a orden, no diagrama a diagrama (i.e: hay que sumar todos los diagramas a un orden para que se cumpla)

## 3.2 Los modos no físicos del fotón no intervienen en los observables despolarizados

- Cuando trabajamos con probabilidades despolarizadas, siempre hacemos la suma sobre las polarizaciones del fotón, que nos da: $\displaystyle\sum_{\{pol\}}^4\epsilon_\mu^i\epsilon^{j*}_\nu=-\eta_{\mu\nu}$. Sin embargo, solemos escribir solamente la suma sobre las polarizaciones físicas $\epsilon^1, \epsilon^2$
- Hay otras dos no-físicas: la temporal y la longitudinal (recuerda que como el cuadripotencial es un cuadrivector, hay que usar una base de 4 vectores para expandirlo, pero como el campo EM solo tiene dos polarizaciones, dos de estos vectores no serán modos físicos)
- En general, la relación es:

$$\displaystyle\sum_{i=\{pol\}}^4 \epsilon_\mu^i=-\eta_{\mu\nu}+\frac{k_\mu n_\nu+n_\mu k_\nu}{k\cdot n}$$

De este modo, por la identidad de Ward, el último término se cancela y tenemos la prescripción que solemos usar

# 4. QED con varios tipos de leptones

Hasta ahora, hemos considerado implícitamente que nuestra teoría solo tiene: un tipo de fermión, su antipartícula, y otra partícula asociada a la interacción (en el sentido de que aparece cuando imponemos inv. gauge, en sí la interacción también se puede dar mediante la propagación de un fermión virtual, véase Compton scattering)

Sin embargo, sabemos que en la Naturaleza hay más de un fermión elemental de S=1/2 que se acopla a la interacción EM. Concretamente, tenemos los 6 quarks y las 3 familias de leptones, en total 9 partículas elementales. ¿Cómo se construye una QED que comprenda todas estas simultáneamente? --> Tengo que introducir sabor a la teoría

**Observación:** la interacción EM no cambia el sabor. Por ende, todos nuestros vértices deben ser: entra partícula de sabor A --> sale (anti)partícula de sabor A + fotón. Es decir, tenemos que "copiar" QED para cada familia que estemos considerando

Si consideramos solamente leptones (no metemos quarks, pero todo sería idéntico), entonces tenemos que sumar 3 copias de nuestro Lagrangiano de QED

**Def.:** para ahorrar notación, definimos el "vector de sabor": $\boxed{\Psi:=\begin{pmatrix}\Psi_e \\ \Psi_\mu \\ \Psi_\tau\end{pmatrix}}$

De modo que tenemos:

$$\boxed{\mathcal L_{all \ QED} = \bar\Psi(i\cancel D-M)\Psi-\frac{1}{4}F^2}$$

con $M=\begin{pmatrix}m_e & & \\ & m_\mu & \\ & & m_\tau\end{pmatrix}$ **"matriz de masas"**

## 4.1 El espacio de sabor

Aunque parezca una forma de ahorrar notación, debemos conferirle cierta estructura a esta nueva forma de agrupar los campos de distintos sabores. Lo hago a continuación:

**Def.:** sea una teoría con campos de $N$ sabores, donde por "sabor" se entiende "tipo de partícula distinta". Incluso los quarks pertenecientes a la misma generación tienen distintos sabores (up, down, etc.). Entonces, se define un solo campo que contiene todos los anteriores, de la forma: $\Psi=\begin{pmatrix}\Psi_1 \\ \vdots \\ \Psi_n\end{pmatrix}$ donde cada subíndice es un sabor

**Proposición:** esta tupla de sabores es un elemento vector de un especio vectorial, que llamaremos "espacio de sabor"

**Dem:** cada campo (componente) por separado ya era parte de un espacio vectorial, por el principio de superposición. Si definimos suma y producto por escalar habituales para esta tupla, tenemos que cada componente se suma / multiplica por separado, de modo que efectivamente el espacio de sabor es $V_{sabor}=\displaystyle\bigoplus_{i=1}^N V_i$, donde $V_i$ es el espacio vectorial al que pertenece el campo de sabor i-ésimo. Véase que este resultado era evidente porque es la suma directa de espacios vectoriales la que permite precisamente expresar elementos vectores del espacio suma como una tupla de todos los demás vectores, escogiendo la basde adaptada

**Proposición:** trivialmente, los operadores asociados a otros espacios (Lorentz, etc.) no tienen nada que ver con el espacio de sabor, de modo que llevan una matriz identidad de sabor: $\cancel DI_3$. Esto es por cómo hemos construido el espacio de sabor como una forma de ahorrar notación de suma

**Proposición:** si se quiere, el espacio de sabor es euclídeo, con producto escalar usual. Esto es porque cuando escribimos $\bar \Psi(i\cancel D+M)\Psi$, realmente estamos usando la matriz identidad como un bilineal, de modo que podemos verla como la matriz de Gram del espacio de sabor

## 4.2 El origen de los números electrónico, muónico, tauónico y leptónico

**Observación:** sabemos que QED (con un solo sabor) tiene una simetría gauge $U(1)$. Sin embargo, como ahora tenemos 3 copias de QED, podemos hacer un cambio de fase global distinto para cada sabor, y cada contribución a la teoría será invariante por separado. Por ende, tenemos ahora 3 simetrías gauge $U(1)$, y con ellas una carga conservada por sabor (relacionada con la carga eléctrica). Esta carga es precisamente el número asociado a la especie leptónica que postularon los primeros Físicos de Partículas como números conservados para explicar por qué algunas reacciones parecían estar prohibida. Ahora sabemos que $U(1)\to Q$ (eléctrica) por ende, conservación de número de un leptón es conservación de carg

**Observación:** si $L_e, L_\mu, L_\tau$ son conservados por separado, es evidente que su suma también será conservada. Este es el **número leptónico**, que precede cronológicamente a los números leptónicos específicos. Obviamente, hace referencia a la carga eléctrica global del sistema, que también se debe conservar

# 5. Ejemplo de proceso completo: $\mu^- e^- \to \mu^- e^-$

Vamos a calcular:

1. $\mathcal M$
2. $\bar{|\mathcal M|^2}$ (i.e: probabilidad despolarizada)
3. Poner esta probabilidad en términos de invariantes de Mandelstam
4. Estudio de los límites no-relativista y ultrarrelativista



## 5.1 Cálculo de $\mathcal M$

Trivialmente, este proceso solo tiene canal $t$ porque los canales $s$ y $u$ requerirían de vértices que no son aceptables en QED. De este modo, solo hay canal t (i.e: intercambio de un fotón, y tenemos):

$$\mathcal M=\frac{e^2}{t}[\bar u_3\gamma^\mu u_1][\bar u_4\gamma_\mu u_2]$$

## 5.2 Cálculo $\bar{|\mathcal M|^2}$

- Queremos calcular esta probabilidad para luego poder enchufarla en la expresión de sección eficaz diferencial, porque la mayoría de detectores no discriminan estados de polarización (i.e: spin)
- Recuerda: hay 4 soluciones a la ec. de Dirac libre, que corresponden a $u^s, v^s$, donde $s$ denota el spin de la partícula a la que están asociados. Se cumplen también las relaciones:

$$\displaystyle\sum_s u^s(\vec p)\bar u^s(\vec p)=\cancel p+m$$

$$\displaystyle\sum_s v^s(\vec p)\bar v^s(\vec p)=\cancel p-m$$

- En un experimento despolarizado, no vamos a saber qué polarización llevan las partícula entrantes, así que habrá que promediar: $\frac{1}{(2s_1+1)(2s_2+1)}\displaystyle\sum_{s_1, s_2}$. Tampoco sabremos qué polarización tendrán las partículas salientes, ni nos interesa, así que sumamos todas: $\displaystyle\sum_{s_3,s_4}$. Por ende:

$$\bar{|\mathcal M|^2}:=\frac{1}{(2s_1+1)(2s_2+1)}\displaystyle\sum_{s_1, s_2, s_3, s_4}|\mathcal M|^2$$

Es conveniente expresar $|\mathcal M|^2=\mathcal M\mathcal M^\dagger$ (la trasposición en este caso es trivial), de modo que tenemos:

$\bar{|\mathcal M|^2}=\frac{1}{4}\displaystyle\sum_{s_1, s_2, s_3, s_4}\frac{e^4}{t^2}[\bar u_3\gamma^\mu u_1][\bar u_4\gamma_\mu u_2][\bar u_1\gamma^\nu u_3][\bar u_2\gamma_\nu u_4]$, donde es inmediato comprobar que $[\bar u_i\gamma^\mu \bar u_j]^\dagger = \bar u_j\gamma^\mu u_i$. Podemos reorganizar esto como un producto de las parejas 1-3 y 2-4:

$\bar{|\mathcal M|^2}=\frac{1}{4}\left(\displaystyle\sum_{s_1, s_3}[\bar u_3\gamma^\mu u_1][\bar u_1\gamma^\nu u_3]\right) \left(\displaystyle\sum_{s_2, s_4}[\bar u_4\gamma_\mu u_2][\bar u_2\gamma_\nu u_4]\right)$

Pasando a notación de índices en el espacio de Dirac, podemos reorganizar las expresiones dentro de cada paréntesis para agruparlas 1-1, 2-2, 3-3, 4-4 y poder usar la propiedad de arriba. Es decir:

$A=\displaystyle\sum_{s_1, s_3}[\bar u_3\gamma^\mu u_1][\bar u_1\gamma^\nu u_3]=\displaystyle\sum_{s_1, s_3}[\bar u_{3,a}(\gamma^\mu)_{ab} u_{1,b}][\bar u_{1,c}(\gamma^\nu)_{cd} u_{3,d}]$

$A=\displaystyle\sum_{s_1, s_3}[u_{3,d}\bar u_{3,a}(\gamma^\mu)_{ab} u_{1,b}][\bar u_{1,c}(\gamma^\nu)_{cd}] = \displaystyle\sum_{s_3}u_{3,d}\bar u_{3,a} \displaystyle\sum_{s_3}u_{1,b}\bar u_{1,c}(\gamma^\mu)_{ab}(\gamma^\nu)_{cd}=(\cancel k_3+m_\mu)_{da}(\cancel k_1+m_\mu)_{bc}(\gamma^\mu)_{ab}(\gamma^\nu)_{cd}$

Recuerda: $(AB)_{ab}=A_{ac}B_{cb}$, así que:

$A=[(\cancel k_3+m_\mu)\gamma^\mu]_{db}[(\cancel k_1+m_\mu)\gamma^\nu]_{bd}=[(\cancel k_3+m_\mu)\gamma^\mu]_{db}[(\cancel k_1+m_\mu)\gamma^\nu]_{dd}\equiv Tr[(\cancel k_3+m_\mu)\gamma^\mu(\cancel k_1+m_\mu)\gamma^\nu]$

Por tanto quedamos con la siguiente probabilidad despolarizada:

$$\bar{|\mathcal M|^2}=\frac{e^4}{4t^2}Tr[(\cancel k_3+m_\mu)\gamma^\mu(\cancel k_1+m_\mu)\gamma^\nu]Tr[(\cancel k_4+m_e)\gamma_\mu(\cancel k_2+m_e)\gamma_\nu]$$

**Teoremas de traza:**

1. $Tr[\text{número impar de $\gamma$'s}]=0$
2. $Tr[\gamma^\mu\gamma^\nu]=4\eta^{\mu\nu}$
3. $Tr[\mu\nu\rho\sigma]=4(\eta_{\mu\nu}\eta_{\rho\sigma}-\eta_{\mu\rho}\eta_{\nu\sigma}+\eta_{\mu\sigma}\eta_{\nu\rho})$
4. $Tr[\gamma^\mu \gamma^5]=0$ (número impar de $\gamma$'s)

Aplicando 1, 2 y 3 llegamos a que:

$$\bar{|\mathcal M|^2}=\frac{8e^4}{t^2}[(12)(34)+(14)(23)-(13)m_e^2-(24)m_\mu^2+2m_e^2m_\mu^2]$$

donde observo que la notación: $(ij)\equiv k_i\cdot k_j$ es muy conveniente para los cálculos intermedios

## 5.3 Poner en términos de invariantes de Mandelstam

- Ya hemos visto en temas anteriores que en scattering $2\to 2$ podemos usar las variables $s, t$
- La forma más fácil de proceder es coger los invariantes de Mandelstam por sus definiciones y despejar el producto que nos interese. Por ejemplo, nos queda: $(12)=\frac{1}{2}(s-m_2^2-m_\mu^2)$. Haciendo esto y simplificando tenemos que:

$$\boxed{\bar{|\mathcal M|^2}=\frac{2e^4}{t^2}\left[(s-m_e^2-m_\mu^2)^2+(u-m_e^2-m_\mu^2)^2+2(m_e^2+m_\mu^2)t\right]}$$

## 5.4 Límite ultrarrelativista

- Como $E=\sqrt s >>m_e, m_\mu$, y como $t, u$ serán la variable $s$ en sus respectivos canales cruzados, tenemos que todas las variables de Mandelstam son mucho más grandes que las masas y por ende:

$$\bar{|\mathcal M|^2}\approx 2e^4\frac{s^2+u^2}{t^2}$$

## 5.5 Límite no-relativista

- En el límite no-relativista, la energía debe ser comparable a la masa, es decir, $E_{CM}=s_{th}=(m_e+m_\mu)^2$

- Por su lado, $t_{th}=0$ y $u=(m_e-m_\mu)^2$. Sustituyendo:

$$\bar{|\mathcal M|^2}\approx \frac{16e^4}{t^2}m_e^2m_\mu^2$$


**Observación:** esta divergencia infrarroja no es una patología de QED, la teoría sigue siendo válida en el límite no-relativista, prueba de ello es que se recupera el mismo resultado que en la teoría de scattering cuántico no-relativista por un potencial Coulombiano. Si te pones a mirarlo es una puta movida, las palabras clave son teorema de Bloch-Nordsieck y que los estados asintóticos no están bien definidos para un potencial Coulombiano porque este no decae suficientemente rápido. Las soluciones son meter fotones blandos o admitir que el estado asintótico está rodeado por una nube de fotones

## 5.6 Recuperando potencial de Coulomb

- En el límite no-relativista, en aproximación de Born, tenemos que la amplitud de scattering está directamente relacionada con el potencial (en espacio de momentos) mediante:

$$\tilde V(\vec q) = -\frac{\mathcal M}{4m_\mu m_e}$$

No podemos sacar el signo de la amplitud en el límite no-relativista a partir de los resultados que ya tenemos porque hemos estado trabajando con la amplitud al cuadrado. Sin embargo, sabemos que en el límite no-relativista:

1. El momento transferido es puramente espacial: $q^\mu = p_1^\mu-p_3^\mu=(0,\vec q)$
2. Los spinores se reducen a sus componentes grandes
3. La corriente fermiónica temporal domina sobre las espaciales

En última instancia, la amplitud del proceso no es más que la contracción Lorentz de dos corrientes electromagnéticas (la del muón y la del electrón), a saber:

$$i\mathcal M = [\bar u_3ie\gamma^\mu u_1]\left(\frac{-i\eta_{\mu\nu}}{t}\right)[\bar u_4ie\gamma^\nu u_2]=\frac{-i}{t}j_{1-3}^\mu j_{2-4,\mu}$$

Por 1, tenemos que: $t=(p_1-p_3)^2=-|\vec q|^2$, así que:

$i\mathcal M \approx \frac{i}{|\vec q|^2}j_{1-3}^0j_{2-4,0}$

donde $j_{1-3}^0=\bar u_3\gamma^0u_1$. Por la identidad de Gordon: $\bar u(p_3)\gamma^\mu u(p_1)=\frac{1}{2m_\mu}\bar u(p_3)\left[(p_1+p_3)^\mu +i\sigma^{\mu\nu}(p_3-p_1)_\nu\right]u(p_1)$, tenemos que $\bar u(p_3)\gamma^\mu u(p_1)=\frac{1}{2m_\mu}\bar u(p_3)\left[2m_\mu -\frac{1}{2}[\gamma^0, \gamma^0(m_\mu-m_\mu)] \right]u(p_1)=\bar u(p_3)u(p_1)$. Pero como $p_3^\mu=(m_\mu, \vec 0)=p_1^\mu$, tenemos que esta igualdad es: $\bar u_3(p_3)u_1(p_3)=2m_\mu \delta_{s_1s_3}$. De este modo, introduciendo en la amplitud:

$\mathcal M \approx \frac{e^2}{-|\vec q|^2}(2m_\mu)(2m_e)$, así que finalmente a bajas energías tenemos que la amplitud es:

$$\mathcal M\approx -\frac{4e^2}{|\vec q|^2}m_em_\mu$$

De acuerdo a la aproximación de Born, la amplitud se relaciona con el potencial de scattering en esopacio de momentos como:

$$\tilde V(\vec q)=-\frac{\mathcal M}{4m_em_\mu}$$

Así que ya podemos leer el potencial de nuestra interacción:

$$\tilde V(\vec q)=\frac{e^2}{|\vec q|^2}$$

Para hallar el potencial en representación de posiciones hacemos una transformada de Fourier:

$$V(r)=\int\frac{d^3\vec q}{(2\pi)^3}\tilde V(\vec q)e^{-i\vec q\cdot \vec r}=\frac{e^2}{4\pi r}=\frac{\alpha_{EM}}{r}$$

Y por ende el potencial efectivo a bajas energías es Coulombiano y repulsivo. Es el mismo que ya conocíamos (nota: el scattering de dos partículas se puede reducir al caso del scattering efectivo de una sola partícula, por eso nos sale un potencial como si la carga estuviera en reposo en vez de dos cargas acercándose una a otra, donde habría campos magnéticos.
