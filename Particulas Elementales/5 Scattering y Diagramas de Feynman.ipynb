# 1. Teoría de scattering y experimento

En un experimento de scattering, tengo una serie de partículas que me entran, interactúan, y salen otras --> las que salen me dan información sobre cómo han interactuado las que entraban

La conexión entre estos dos estados se hace a través de la matriz de scattering: $\mathcal M:=\langle f|S|i\rangle$. Por ende, si se quiere, $S$ avanza el estado de entrada hasta el de salida, y nosotros comprobamos cuál es la amplitud de probabilidad de que dicho estado coincida con uno $|f\rangle$ que nos interesa

La matriz $S$ se puede separar en (1) el caso donde no hay interacción y (2) el caso donde sí la hay. Es evidente que el caso donde sí la hay debe estar muy ligado a la amplitud de probabilidad de que el evento tenga lugar

$$S=I+iT$$

Si consideramos un proceso de interacción $2\to n$, tenemos entonces que el elemento de matriz es:

$$\langle p_1',\ldots,p_n'|S-I|p_1,p_2\rangle = i(2\pi)^4\delta^4(p_{i}-p_{f})\mathcal M$$

## 1.1 Sección eficaz

Lo que se mide realmente en un experimento son la sección eficaz diferencial $\frac{d\sigma}{d\Omega}$ y la total $\sigma$

**Gedankenexperiment:** haciendo estadística frecuentista, podemos imaginarnos un experimento donde lanzamos una partícula contra un objetivo y estudiamos si se dispersa o no. Al repetir el experimento un número infinito de veces, sabremos que la probabilidad de que una partícula se disperse al lanzarla contra el objetivo es: $P=\displaystyle\lim_{N\to\infty}\frac{N_{disp}}{N}$

**Def.:** supongamos que tenemos una partícula entrante, modelada por su flujo $F$, que van a dar contra un objetivo. Supongamos que la interacción con dicho objetivo dura un cierto intervalo de tiempo $\Delta t_{int}$. Si tuviésemos garantizado que la partículas se va a dispersar, entonces esperaríamos observar un flujo saliente igual al entrante. Asimismo, como la partícula saldría al acabar la interacción (i.e: sale al de $\Delta t_{int}$), entonces podríamos asignarle al objetivo un cierto área efectiva: $\sigma_1=\frac{1}{F\times \Delta t_{int}}$. Como no tenemos garantizado que la partícula vaya a dispersar, habrá que multiplicar esta sección eficaz por la probabilidad de que haya dispersión que ya hemos calculado antes: $\sigma=\sigma_1 P_{disp}$. Esto es la sección eficaz

**Def.:** se define la **luminosidad** $L(t)$ de un cierto experimento como:

$$N_{disp}=\int d\sigma\int dt L(t)$$

Es una propiedad del aparataje y se calibra con un experimento de sección eficaz conocido

**Fíjate:** medir la sección eficaz de un proceso es entonces equivalente a medir la probabilidad de que este ocurra. Esto me conecta directamente experimento y teoría, así que puedo comprobar las predicciones de mis teorías a través de esta magnitud

# 2. Fijando la cinemática del scattering

**Intuición:** hemos dicho que medir $\sigma$ equivale a medir la probabilidad de un proceso. Es de esperar entonces que $\sigma$ se relacione de alguna manera con la amplitud de probabilidad, pero esta relación no puede ser directa porque $\mathcal M$ no es medible. Esto ya lo sabíamos, porque como nos tiene que quedar proporcional a la probabilidad de que ocurra el proceso, es de esperar que obtengamos una relación del tipo: $\sigma\propto |\mathcal M|^2$

**Proposición:** para un proceso $2\to n$, tenemos:

$$d\sigma=\frac{1}{(2E_1)(2E_2)}\frac{1}{|\vec u_1-\vec u_2|}(2\pi)^4\delta^4(p_i-p_f)|\mathcal M|^2d\Pi_f$$

con $d\Pi_f=\displaystyle\prod_{i=3}^n\frac{d^3\vec p_i}{(2\pi)^32E_j}$ (**espacio de fases final**)

**Nota:** esta expresión es Lorentz-invariante, y de hecho el denominador se puede escribir como: $(2E_1)(2E_2)|\vec u_1-\vec u_2|=4\sqrt{(p_1\cdot p_2)^2-m_1^2m_2^2}$

**Def.:** equivalentemente, se suele definir una medida $d$LIPS$=(2\pi)^4\delta^4(p_i-p_f)d\Pi_f$ que se llama "Lorentz-Invariant Phase Space"

**Útil:** el LIPS para un estado final de 2 partículas es: $\boxed{d^2LIPS=\frac{1}{(4\pi)^2}\frac{|\vec p_3|}{E_i}d\Omega}$

En términos de esta medida, el ritmo de transición es:

$$\boxed{dW_{i\to f}=\displaystyle\prod_{i=1}^m\frac{1}{2E_i V}V|\mathcal M|^2 d^nLIPS}$$

**Proposición:** para un proceso de scattering $2\to 2$, esta expresión se reduce a (en inercial CM):

$$\boxed{\frac{d\sigma}{d\Omega}=\frac{1}{64\pi^2s}\frac{|\vec p_3|}{|\vec p_1|}|\mathcal M|^2}$$

**Dem:**

Centrémonos en el caso donde hay dispersión, de modo que $\langle f|S|i\rangle = i(2\pi)^4\delta^4(p_i-p_f)\mathcal M$

Primera nota: hasta ahora, hemos trabajado con estados normalizados de tal forma que $\langle a|b\rangle$ sea invariante Lorentz. Sin embargo, como $S$ es unitaria, necesariamente debemos normalizarlos a 1.

Problema: hay una delta en la normalización

Solución: supongo que el proceso tiene lugar en una región acotada del espacio-tiempo que, formalmente, puedo hacer tender a infinito. Impongo condiciones de contorno periódicas en mi caja. Esto es un artefacto, pero no importará siempre y cuando todo se cancele al final y me mantenga consistente con su uso.

En el espacio infinito, la normalización es: $\langle\vec p|\vec k\rangle = (2\pi)^32E_p\delta^3(\vec p-\vec k)$. En la caja, la normalización será: $\langle\vec p|\vec k\rangle = 2E_pV\delta_{\vec p\vec k}$. Esto es porque al pasar entre sumatorio e integral, tenemos que hacer: $\sum\to \frac{V}{(2\pi)^3}\int d^3\vec p$, esto es, es la densidad de estados. Véase Física Estadística.

Por ende, para que los estados dentro de nuestra caja estén normalizados a 1, tenemos que dividirlos por la raíz de ese factor, de modo que usaremos esta nueva normalización con notación normal: $|\vec p\rangle = \frac{1}{\sqrt{2E_p}V}|\vec p\rangle_{antes}$

Entonces, tenemos que nuestra probabilidad es: $P_{i\to f}=|\langle f|S|i\rangle|^2=(2\pi)^8\delta^4(p_i-p_f)\delta^4(p_i-p_f)|\mathcal M|^2\displaystyle\prod_{i=1}^m\frac{1}{2E_i V}\displaystyle\prod_{j=1}^n\frac{1}{2E_j V}$, donde $m$ es el número de partículas inicial y $n$ el número de partículas final. Véase que estos factores vienen precisamente de haber cambiado la normalización de los estados cuando tomamos elemento de matriz de $S$

Recordemos ahora de teoría de distribuciones que si tenemos dos deltas iguales, entonces una de ellas se puede poner como una delta de 0. Esto es: $(\delta^4(p_i-p_f))^2=\delta^4(0)\delta^4(p_i-p_f)$. A su vez, tenemos que: $\delta^4(0)=\frac{1}{(2\pi)^4}\int dx^4e^{ix\cdot 0}=\frac{1}{(2\pi)^4}V\cdot T$, donde $V$ es el volumen del espacio que estamos considerando y $T$ el "volumen del tiempo" que estamos considerando. Esto es, estamos considerando solamente una región del espacio de Minkowski de volumen $V\cdot T$. Tenemos entonces que:

$$P_{i\to f}=(2\pi)^4\delta^4(p_i-p_f)VT|\mathcal M|^2\displaystyle\prod_{i=1}^m\frac{1}{2E_i V}\displaystyle\prod_{j=1}^n\frac{1}{2E_j V}$$

**Observación:** no tiene mucho sentido hablar de la probabilidad de ir de $i\to f$, sino que más bien tiene sentido hablar del ritmo de transición, que es lo que vemos en nuestro experimento. Para ello, podemos definir dicho ritmo como la probabilidad dividida por el tiempo que estamos considerando, de modo que eso ya nos quita uno de los reguladores. Tenemos entonces:

$$W_{i\to f}=(2\pi)^4\delta^4(p_i-p_f)V|\mathcal M|^2\displaystyle\prod_{i=1}^m\frac{1}{2E_i V}\displaystyle\prod_{j=1}^n\frac{1}{2E_j V}$$

**Observación:** como en scattering estamos trabajando con estados asintóticos, realmente no tiene sentido hablar del ritmo de transición de un estado a otro, sino de un estado a un intervalo de estados tal que el momento de la partícula final j-ésima esté entre $\vec k_j$ y $\vec k_j+d\vec k_j$. Por ende, debemos multiplicar por la densidad de estados final, que hemos dicho que es: $d\rho=\displaystyle\prod_{j=1}^n\frac{V}{(2\pi)^3}d^3\vec p_j$. Por ende, tenemos que el diferencial de ritmo de transición de un estado inicial a un intervalo de estados finales es:

$$\boxed{dW_{i\to f}=\displaystyle\prod_{i=1}^m\frac{1}{2E_i V}V(2\pi)^4\delta^4(p_i-p_f)|\mathcal M|^2\displaystyle\prod_{j=1}^n\frac{d^3\vec p_j}{(2\pi)^32E_j}}$$

Paso 2: sección eficaz

La sección eficaz se puede entender como el ritmo de transición entre el flujo de partículas entrantes: $d\sigma=\frac{dW_{i\to f}}{F}$

A su vez, el flujo de partículas es la densidad de partículas entrantes por la velocidad neta del haz. Solo consideramos el caso de scattering de 2 partículas, y recordemos que este (Taylor) se puede reducir al caso de scattering de 1 partícula contra un  objetivo fijo, de modo que la densidad es entonces $\frac{1}{V}$. Tenemos entonces que $F=nu=\frac{1}{V|\vec u_1-\vec u_2|}$. Ojo: esto no introduce una singularidad si las partículas tienen igual velocidad, porque recordemos que en CM están enfrentadas, de modo que en todo caso se suman

Introduciendo la definición del flujo de partículas entrantes, tenemos entonces que:

$$d\sigma=\frac{1}{(2E_1)(2E_2)}\frac{1}{|\vec u_1-\vec u_2|}(2\pi)^4\delta^4(p_i-p_f)|\mathcal M|^2d\Pi_f$$

**Paso 1:** denominador

Sabemos que el momento relativista de una partícula es: $\vec p=\gamma m\vec v$, y también sabemos que su energía total (si es libre, y los estados asintóticos lo son) es $E=\gamma m$. Por ende, para una partícula relativista, tenemos $\vec u=\frac{\vec p}{E}$. Ahora sustituimos en la expresión del flujo:

$4E_1E_2|\vec u_1-\vec u_2|=4E_1E_2|\frac{\vec p_1}{E_1}-\frac{\vec p_2}{E_2}|$, pero como estamos en CM, tenemos que $\vec p_2=-\vec p_1$, así que

$=4E_1E_2|\vec p_1|\frac{E_1+E_2}{E_1E_2}=4|\vec p_1|\sqrt{s}$

# 3. Fijando la cinemática de las desintegraciones

Ahora fijamos la cinemática de un proceso de desintegración general: $1\to n$

Como ya hemos calculado el ritmo de transición, es fácil, porque el ritmo de desintegración es precisamente el ritmo de transición de un estado de 1 partícula a un estado final de $n$ partículas, de modo que:

$$d\Gamma_{1\to n}=\frac{1}{2M}(2\pi)^4|\mathcal M|^2d\Pi_f$$

Particularizando para la desintegración a 2 partículas, tenemos que:

$$\boxed{d\Gamma=\frac{1}{32\pi^2M^2}|\mathcal M|^2|\vec p_3|d\Omega}$$

**Dem:**

$d\Gamma:=dW_{1\to n}=\frac{1}{2MV}V(2\pi)^4|\mathcal M|^2d\Pi_f$

Si particularizamos al caso de 2 partículas finales, enchufamos la expresión de $d^2LIPS$ y obtenemos:

$d\Gamma=\frac{1}{2MV}V|\mathcal M|^2\frac{1}{(4\pi)^2}\frac{|\vec p_3|}{M}d\Omega$ y obtenemos inmediatamente el resultado final $_\blacksquare$

Ahora queremos construir el aparataje teórico que subyace al cálculo de la dinámica. Para ello, vamos a hacer lo siguiente:

1. Expresamos $\langle f|S|i\rangle$ en términos de $(\square_1+m_1^2)\ldots(\square_n+m_n^2)\langle \phi(x_1)\ldots\phi(x_n)\rangle$ porque las funciones a $n$ puntos son algo más tangible e incluso se pueden calcular computacionalmente (lattice). Esto es fórmula LSZ

3. Aparte: como el propagador de Feynman es la función de Green de la teoría libre en la que estemos, actuando juntos darán una delta. Puedo pensarlo al revés: una delta puede expresarse así, de modo que para calcular función a $m$ puntos, ya tengo la estructura: propagador*operador actuando sobre función a $m$ puntos --> puedo aplicar Schiwnger-Dyson (LSZ me ha servido para llegar a la función a $m$ puntos en primer lugar)

# 4. Fórmula LSZ

**Teorema:** sea el scattering de $2\to (n-2)$. Entonces, podemos relacionar la amplitud de scattering con una función a $n$ puntos como sigue:

$$\boxed{\langle p_3\ldots p_n|S|p_1p_2\rangle = \left[i\int d^4x_1e^{-ip_1\cdot x_1}(\square_1+m_1^2)\right]\times \ldots \times \left[i\int d^4x_ne^{ip_n\cdot x_1}(\square_n+m_n^2)\right]\langle \Omega|T\left\{\phi(x_1)\ldots\phi(x_n)\right\}|\Omega\rangle}$$

donde $|\Omega\rangle$ es el estado de vacío de la teoría CON interacción, que es diferente a $|0\rangle$ (vacío sin interacción)

**Nota:** la diferencia entre estos dos estados de vacío es que, cuando permitimos interacción, el vacío se polariza (QED), aparecen condensados (QCD), etc. Por ende, el vacío se vuelve mucho más rico y ya no es "aquel estado tal que $a_p|0\rangle=0$). Sin embargo, podemos expresar todo en elementos de matriz del vacío libre (llamado así porque es el vacío con interacción de la teoría libre, que no tiene interacción) como (**teorema de Gell-Mann-Low**):

$$\langle \Omega|T\{\phi(x_1)\ldots\phi(x_n)\}|\Omega\rangle =\frac{\langle 0|T\{\phi_I(x_1)\ldots\phi_I(x_n)\exp(i\int d^4x\mathcal L_{int})\}|0\rangle}{\langle 0|T\{\exp(i\int d^4x\mathcal L_{int})\}|0\rangle}$$

donde $\phi_I$ son los campos en imagen de interacción


**IMPORTANTE:** esta fórmula nos dice que **la interacción ocurre necesariamente off-shell**, porque de lo contrario, si uno solo de los campos está on-shell, la amplitud se anula idénticamente porque cumple la ec. de Klein-Gordon

**Dem:**

Paso 1: expreso $\langle f |S|i\rangle$ constryendo los estados $|i\rangle, |f\rangle$ con operadores de construcción

Paso 2: reescribo en términos de restas de operadores construcción ó creación aprovechando que tengo ordenación temporal

Paso 3: encuentro relación entre esta resta y la transformada de Fourier del operador KG actuando sobre un campo --> sustituyo --> hecho

**Paso 1:**

Vamos a crear los estados asintóticos:

$$\begin{gather}|i\rangle=|p_1p_2\rangle=\sqrt{2E_1}\sqrt{2E_2}a_{p_1}^\dagger(-\infty) a_{p_2}^\dagger (-\infty)|\Omega\rangle \\ |f\rangle=|p_3\ldots p_n\rangle=\sqrt{2E_3}\ldots\sqrt{2E_n}a_{p_3}^\dagger(\infty)\ldots a_{p_n}^\dagger (\infty)|\Omega\rangle\end{gather}$$

donde los estados asintóticos son libres, pero siguen siendo excitaciones del vacío físico porque la teoría tiene interacción

Por ende, me queda:

$$\langle p_3\ldots p_n|S|p_1p_2\rangle = 2^{n/2}\sqrt{E_1\ldots E_n}\langle \Omega| a_{p_3}(\infty)\ldots a_{p_n}(\infty)a^\dagger_{p_1}(-\infty)a^\dagger_{p_2}(-\infty)|\Omega\rangle$$

donde $S$ ha desaparecido porque ya está codificada en los operadores de creación y destrucción, ya que la amplitud de que suceda un proceso es precisamente la amplitud de que observemos un estado asintótica $|f\rangle$ en $t\to\infty$ habiendo empezado con un estado asintótico $|i\rangle$ en $t\to-\infty$

**Paso 2:**

No es raro que los operadores creación / destrucción dependan del tiempo. En la teoría libre, lo hacen trivialmente, a través de una fase, por lo que usualmente se absorbe en la exponencial y esta se expresa como $ip\cdot x$ en lugar de $i\vec p\cdot \vec x$. Sin embargo, en una teoría con interacción, la cosa ya no es tan sencilla y como estamos en imagen de Heisenberg, en general los operadores creación / destrucción dependen del tiempo. Sin embargo, como nuestros estados asintóticos son libres, podemos seguir escribiendo estos campos con la expresión habitual

**Observación:** el elemento de matriz que tenemos ya está bien ordenado temporalmente sin que se lo hayamos pedido explícitamente (si se quiere, no podía ser de otra manera, porque hemos respetado los estados asintóticos de entrada y salida). Entonces, nada cambia si ponemos un operador de ordenación temporal. Además, si hacemos esto, nada cambia si en vez de poner $a_p(\infty)$ para una de las partículas salientes ponemos $a_p(\infty)-a_p(-\infty)$, porque la ordenación temporal mandará este término a la derecha, donde aniquilará al vacío en interacción y todo ese término contribuirá con $0$. El razonamiento es análogo para las partículas entrantes. Por tanto, podemos escribir el elemento de matriz como:

$$\langle p_3\ldots p_n|S|p_1p_2\rangle = 2^{n/2}\sqrt{E_1\ldots E_n}\langle \Omega|T\left\{ (a_{p_3}(\infty)-a_{p_3}(-\infty))\ldots  (a_{p_n}(\infty)-a_{p_n}(-\infty))(a^\dagger_{p_1}(-\infty)-a^\dagger_{p_1}(\infty))(a^\dagger_{p_2}(-\infty)-a^\dagger_{p_2}(\infty))\right\}|\Omega\rangle$$

**Paso 3:**

Vamos a intentar encontrar una forma de expresar esta resta de operadores en términos de lo que sabemos manejar bien, que son los campos

**Nota:** demostración de Jacobo está mal, refiero resultado directamente:

$$FT[(\square+m^2)\phi(x)]=\sqrt{2E_p}[a_p(\infty)-a_p(-\infty)]$$

Por ende, hacemos la sustitución de RHS por LHS en la expresión de arriba y ya hemos terminado $_\blacksquare$

# 5. Fórmula de Schwinger-Dyson

**Teorema:** en una teoría con cierto término de interacción $\mathcal L_{int}(\phi)$ que solo depende del propio campo, la fórmula de Schwinger-Dyson nos dice exactamente cuánto vale $(\square+m^2)\langle\ldots\rangle$:

$$\boxed{(\square_x+m^2)\langle \phi(x)\phi(x_1)\ldots\phi(x_n)\rangle= \langle \frac{\delta\mathcal L_{int}(\phi(x))}{\delta\phi}\phi(x_1)\ldots\phi(x_n)\rangle-i\displaystyle\sum_j\delta^4(x-x_j)\langle \phi(x_1)\ldots\phi(x_{j-1})\phi(x_{j+1})\ldots\phi(x_n))\rangle}$$

# 6. Diagramas de Feynman (posiciones)

Ahora ya podemos calcular funciones a $n$ puntos, tanto en las teorías libres como en las que tienen interacción



## 6.1 Función a 2 puntos en teoría libre (sin masa)

Queremos calcular $\langle\phi(x_1)\phi(x_2)\rangle$. Este es precisamente el propagador de Feynman de $1\to 2$, así que esperaríamos obtener $D_F(x_1, x_2)$

- Metemos una delta: $\langle\phi(x_1)\phi(x_2)\rangle = \int d^4 \delta^4(x-x_1)\langle \phi(x)\phi(x_2)\rangle$
- Recordamos que el propagador de Feynman es la función de Green de la teoría que toque, en este caso de K-G, así que $\square_x D_F(x,x_1)=-i\delta^4(x-x_1)$
- Sustituimos:  $\langle\phi(x_1)\phi(x_2)\rangle = i\int \square_x D_F(x, x_1)\langle \phi(x)\phi(x_2)\rangle$
- Integrando por partes 2 veces (y desechando los términos de superficie) podemos hacer que el d'Alembertiano actúe sobre la función a 2 puntos, efectivamente conmutando $\square_x$ con el propagador
- $\langle\phi(x_1)\phi(x_2)\rangle = i\int D_F(x, x_1)\square_x\langle \phi(x)\phi(x_2)\rangle$
- Ahora esto ya lo sabemos calcular con Schwinger-Dyson:
- $\square_x\langle \phi(x)\phi(x_2)\rangle=-i\delta^4(x-x_1)\langle \Omega|\Omega\rangle = -i\delta^4(x-x_1)$ (el vacío está por definición normalizado a 1)
- $\langle \phi(x_1)\phi(x_2)rangle = i(-i)\int d^4 x D_F(x, x_1)\delta^4(x-x_2)=D_F(x_2, x_1)$

Y ya hemos terminado. Recordemos que, como el propagador de Feynman ya lleva ordenación temporal en las tripas, $D_F(x_1, x_2)=D_F(x_2, x_1)$

Pictóricamente, podríamos haber representado el evento $x_1$ y el evento $x_2$ conectados por una linea, que es el propagador:

$\langle \phi(x_1)\phi(x_2)\rangle = x_1 --- x_2$

## 6.2 LSZ "amputa" las patas externas de los diagramas de Feynman

Cuando calculamos diagramas de Feynman, vemos que las únicas líneas que meten propagadores son las internas, las externas no. ¿Por qué es esto?

Supongamos que tenemos un cierto proceso, y que queremos calcular su amplitud de probabilidad --> calculamos el elemento de matriz de $S$ --> por LSZ, esto queda en términos de una función a $n$ puntos --> la expandimos en teoría de perturbaciones hasta el orden deseado, expresando los propagadores como las transformadas de Fourier en espacio de posiciones de los propagadores en momentos --> integramos en las variables espaciales intermedias que nos introduce el cálculo perturbativo en posiciones --> los propagadores exteriores, al integrar, sueltan una delta --> se van y solo quedan los propagadores de las líneas internas

A esto nos referimos con que LSZ "amputa" las patas externas: a que no aparecen los propagadores de estas líneas. Sin embargo, véase que esto no es solo por virtud de LSZ, sino que también se debe a cómo hemos hecho el desarrollo perturbativo en la constante de acoplo
