# 1. Primer modelo de interacción débil

**Observación:** se observa que otra interacción (la débil) debe existir porque ninguna de las otras explica procesos como la desintegración beta

Fermi postula una teoría:

$$\mathcal L = -\frac{G_F}{\sqrt 2}(\bar p\gamma^\mu n)(\bar e\gamma_\mu \nu_e)+h.c$$

**Problema:** no puede ser una teoría de interacción fundamental porque no involucra partículas fundamentales. En su lugar, probemos con sus constituyentes:

$$\mathcal L = -\frac{G_F}{\sqrt 2}(\bar u\gamma^\mu d)(\bar e\gamma_\mu \nu_e)+h.c$$

**Problema:** experimento de Wu demuestra que la interacción débil viola paridad, y esta teoría es par (i.e: invariante) bajo paridad. La simetría de paridad debe estar explícitamente rota

**Solución:** meto una corriente axial para que desentone con la paridad de la corriente vectorial, porque $(V-A)(V-A)$ no tiene paridad bien definida

$$\mathcal L = -\frac{G_F}{\sqrt 2}(\bar u\gamma^\mu(1-\gamma^5) d)(\bar e\gamma_\mu(1-\gamma^5) \nu_e)+h.c$$

Este es el llamado "modelo V-A"

**Nota:** no puedo meter términos de masa habituales $\bar\Psi m\Psi$ porque conectan partes left y right, y este modelo es decididamente quiral (porque los neutrinos Right se supone que no existen, y se puede demostrar que si los neutrinos solo se acoplan left, entonces todo debe acoplarse solo left para preservar la invariancia Lorentz)

**Observación experimental:** solo se producen neutrinos de quiralidad Left --> mi modelo ya predecía esto, porque fíjate que esencialmente tenemos un proyector de quiralidad Left!

$$\mathcal L = -4\frac{G_F}{\sqrt 2}(\bar u\gamma^\mu d_L)(\bar e\gamma_\mu \nu^e_L)+h.c$$

**Proposición:** si una de las componentes de la corriente solo tiene quiralidad Left, a ella solo se acoplará la quiralidad Left del otro spinor, i.e: $\bar \Psi\gamma^\mu\Psi_L=\bar \Psi\gamma^\mu\Psi_L$

$$\mathcal L = -4\frac{G_F}{\sqrt 2}(\bar u_L\gamma^\mu d_L)(\bar e_L\gamma_\mu \nu^e_L)+h.c$$

**Observación:** se observa el proceso $\mu^-\to e^-\bar\nu_e\nu_\mu$ así que ahora tengo que dar cuenta también de los muones, así como del acoplo de todos con todos (i.e: cada corriente es una línea final --> principio en el diagrama de Feynman, de modo que ups por ejemplo solo pueden pasar a ser downs, interaccionando por ejemplo con un electrón que solo puede dar un neutrino). 

$$\mathcal L = -4\frac{G_F}{\sqrt 2}[(\bar u_L\gamma^\alpha d_L)+(\bar e_L\gamma_\alpha \nu^e_L)+(\bar\mu_L\gamma^\alpha \nu^\mu_L)][h.c]_\alpha$$

**Idea:** la estructura siempre parece ser la misma, la interacción débil me pasa de una partícula a otra (i.e: $u\to d, e\to \nu_e$, etc.) así que me surge la idea de organizarlos en un doblete de $SU(2)$

$$Q_L=\begin{pmatrix}u_L \\ d_L\end{pmatrix}; \quad L_e=\begin{pmatrix}\nu^e_L \\ e_L\end{pmatrix}; L_\mu=\begin{pmatrix}\nu^\mu_L \\ \mu_L\end{pmatrix}$$

Construyendo los habituales operadores escalera de $SU(2)$, me queda:

$$\mathcal L = -4\frac{G_F}{\sqrt 2}\displaystyle\sum_{a=1}^2[\bar Q_L\gamma^\alpha \frac{\tau^a}{2} Q_L)+(\bar L_e\gamma_\alpha \frac{\tau^a}{2} L_e)+(\bar L_\mu\gamma^\alpha \frac{\tau^a}{2}L_\mu)][h.c]_{a\alpha}$$

**Observación:** hay simetría $SU(2)_L$ porque esto es como el módulo de un vector, cada corriente se transforma pero al contraerla con su adjunta todo queda invariante

**Observación:** qué raro que no salga $\tau^3$ para completar los generadores de $SU(2)$. Lo cierto es que sí sale, pero daría lugar a una corriente neutra (porque $\tau^\pm$ sí portan carga pero $\tau^0$ no), y costó 20 años que se dieran cuenta de que existía dicha corriente

**Problema:** esta teoría funciona bien para algunos procesos pero para otros sigue divergiendo a altas energías

# 2. Construyendo teoría de interacción electrodébil (gauge)

**Qué nos ha enseñado la fenomenología que hemos hecho?:**

1. Se debe violar paridad
2. Solo se acoplan partes left
3. Tiene que haber invariancia $SU(2)$ global al organizar las partículas en dobletes de sabor
4. Tiene que haber invariancia $U(1)$ global (Glashow vio que esta adición era la mínima necesaria para reproducir parte electromagnética, permitir corrientes neutras y mantener renormalizabilidad de la teoría gauge)

**Nota:** en la teoría que vamos a construir a continuación verás que aparecen también partes right. Estas solo se acoplan a la $U(1)$ así que no se está rompiendo el postulado 2

## 2.1 Acondicionando teoría libre

Partamos del lagrangiano libre para todas las partículas susceptibles a interactuar débilmente (menos el Higgs, que viene luego), ya escrito en términos de dobletes para las partes left, ya teniendo en cuenta las 3 familias:

$$\mathcal L_{free}=\displaystyle\sum[i\bar Q_i\cancel\partial Q_i+i\bar L_i\cancel\partial L_i + i\bar u_{i,R}\cancel\partial u_{i,R}+i\bar d_{i,R}\cancel\partial d_{i,R}+i\bar l_{i,R}\cancel\partial l_{i,R}+i\bar e_{i,R}\cancel\partial e_{i,R}]$$

donde $u_i$ son los 3 quarks de carga positiva (parte de arriba de las familias) y $d_i$ son los de carga negativa. No meto término de masas porque me mezcla quiralidades y me rompe $SU(2)_L$, tendré que buscar una forma de generar masas más adelante (mecanismo de Higgs)

**Observación:** la teoría libre así escrita (i.e: con dobletes left y sin masas) es inv. bajo $U(1)$ y bajo $SU(2)_L$

**Observación:** no estamos acoplando la parte right del neutrino, podríamos hacerlo pero para qué si no se observa (i.e: no se acopla a la dinámica --> no tiene término cinético --> solo puede tener término de masas pero como tiene masa 0, fuera)

**Def.:** $SU(2)_L$ se llama **isospin debil** en analogía al isospin de QCD, pero la diferencia es que el isospin débil es una simetría exacta porque el límite quiral es ahora una imposición

**Def.:** dados los números cuánticos de tercera componente de isospin y de hipercarga, podemos construir la carga eléctrica de una partícula con la relación de Gell-Mann-Nishijima:

$$Q=I_3+\frac{Y}{2}$$

## 2.2 Promocionando simetrías a locales

Promocionemos ahora $SU(2)_L\times U(1)_Y$ de simetrías globales a locales y veamos lo que pasa

**Observación:** la derivada covariante para las componentes left y right es diferente: ambas llevan la parte de hipercarga, pero la left lleva además la parte de $SU(2)_L$ local:

$$\boxed{D^R_\mu = \partial_\mu-ig'\frac{Y}{2}B_\mu}$$

$$\boxed{D^L_\mu = \partial_\mu-ig'\frac{Y}{2}B_\mu-igW_\mu^a\frac{\tau^a}{2}}$$

De este modo, la teoría completa (aún sin masa) es:

$$\mathcal L=\displaystyle\sum[i\bar Q_i\cancel D Q_i+i\bar L_i\cancel D L_i + i\bar u_{i,R}\cancel D u_{i,R}+i\bar d_{i,R}\cancel D d_{i,R}+i\bar l_{i,R}\cancel D l_{i,R}+i\bar e_{i,R}\cancel D e_{i,R}]-\frac{1}{4}B_{\mu\nu}B^{\mu\nu}-\frac{1}{4}W_{\mu\nu}^a W^{\mu\nu}_a$$

Sé que los bosones gauge deben tener masa. Esto no es simplemente porque la interacción sea de corto alcance, porque entonces en QCD los gluones también deberían tenerla y no la tienen, sino que el corto alcance de la teoría se debe a que la no-abelianidad del grupo produce que los gluones tengan carga de color, permitiendo las interacciones entre ellos y por ende el corto alcance de la interacción. Todo esto también se podría decir de los bosones débiles porque $SU(2)$ también es no-abeliano, pero lo que veremos pasa aquí es que el Higgs adopta su valor esperado de vacío introduciendo una escala en el sistema y no permitiendo que se llegue a una fase confinada. Si no considerammos el Higgs, el modelo no describe la Naturaleza pero tampoco hay garantía de que $SU(2)$ confine. El confinamiento se estudia exclusivamente mirando el running de la constante de acoplo

# 3. Dotando de masa a los bosones: mecanismo de Higgs

## 3.1 Mecanismo de Higgs con campo escalar

**Descripción:** el mecanismo de Higgs es aquel en el que los bosones de una simetría gauge adquieren masa al acoplarse al valor esperado de vacío del campo de la teoría

Partamos del modelo sigma lineal para un campo escalar complejo, e impongamos ahora una invariancia gauge $U(1)$:

$$\mathcal L=D_\mu\phi(D^\mu \phi)^*+m^2\phi\phi^*-\frac{\lambda}{4!}(\phi\phi^*)^2-\frac{1}{4}F^2$$

El valor mínimo del potencial sigue siendo el mismo, de modo que volvemos a tener infinitos posibles estados de vacío, todos igual de válidos:

$$\langle\Omega_\theta|\phi|\Omega_\theta\rangle = \sqrt{\frac{2m^2}{\lambda}}e^{i\theta}$$

Volvemos a coger el vacío como $\theta=0$ por comodidad, y al hacer esto ya hemos roto espontáneamente la simetría $U(1)$, que ahora es local! Hemos roto espontáneamente una simetría gauge

Parametricemos ahora el campo como siempre:

$$\phi(x)=\frac{1}{\sqrt 2}(v+\sigma(x))e^{i\pi(x)/F}$$

Introducimos en el Lagrangiano y desacoplamos la teoría. Cuando el campo $\phi$ adopte su valor esperado en el vacío (i.e: cuando se rompa la simetría gauge espontáneamente), tendremos:

$$\mathcal L_G = -\frac{1}{4}F^2+\frac{1}{2}\partial_\mu\pi \partial^\mu\pi +\frac{1}{2}e^2v^2A^2+evA_\mu\partial^\mu\pi$$

**Observación:** el término con $A^2$ se puede interpretar como un término de masas para $A_\mu$, con $m_A=ev$ (tiene el signo correcto, véase el Lagrangiano de Proca)

**Observación:** al romper espontáneamente la simetría gauge, hemos introducido un acoplo entre el campo gauge y el bosón de Goldstone resultante de romper una simetría, pero como tenemos libertad de gauge por construcción de la teoría, podemos tomar $A_\mu'=A_\mu+\frac{1}{e}\partial_\mu\\alpha$, y como el Goldstone se transforma como $\pi\to \pi-F\alpha(x)$, podemos escoger $\alpha=\frac{\pi(x)}{F}$ para eliminar completamente el Goldstone. Este gauge se llama **gauge unitario** y decimos que la simetría gauge se ha "comido" el Goldstone. Por ende, tenemos que, en este gauge:

$$\mathcal L_G=-\frac{1}{4}F^2+\frac{1}{2}m_A^2A^2$$

## 3.2 Ruptura espontánea de la simetría electrodébil

Vamos a intentar usar un modelo sigma lineal para dotar de masa a los bosones débiles con la ruptura espontánea de la simetría gauge de la teoría. Construyamos el modelo sigma no-lineal con simetría $SU(2)\times U(1)$:

$$\mathcal L =\partial_\mu H^\dagger \partial^\mu H +m^2H^\dagger H-\frac{\lambda}{4!}(H^\dagger H)^2$$

donde $H=\begin{pmatrix}H_+ \\ H_0\end{pmatrix}\equiv \begin{pmatrix}H_1+iH_2 \\ H_3+iH_4\end{pmatrix}$

**Asigno:** $Y_H=+1$, de modo que la carga de la primera componente (por Gell-Mann-Nishijima) es +1 y la de la segunda es 0

**Observación:** el mínimo del potencial es $H_1^2+\ldots+H_4^2=v^2$. Si quiero romper la simetría (global o gauge, en sí da igual, el mecanismo de ruptura es el mismo), entonces me quedo por ejemplo con el estado de vacío que me dé el valor esperado de vacío $H_1=H_2=H_4=0, H_3=v$ (esto es porque esta componente es neutra, de lo contrario el vacío estaría eléctricamente cargado. Esto es matemáticamente posible pero no es lo que se observa en la Naturaleza). Por ende:

$$\langle \Omega|H|\Omega\rangle=\frac{1}{\sqrt 2}\begin{pmatrix}0 \\ v\end{pmatrix}$$

**Observación:** si hacemos tranasformación $SU(2)$ general, el valor esperado de vacío no es invariante. Si hacemos $U(1)$, tampoco lo es. Sin embargo, tenemos que $\sigma_3+Y$ sí aniquilan el vacío o, equivalentemente, $e^{i\alpha_3\sigma_3/2}e^{iY/2}\begin{pmatrix}0 \\ v\end{pmatrix}=\begin{pmatrix}0 \\ v\end{pmatrix}$ porque $Y=1$ para este campo, por definición. De este modo, observamos que sí hay algo que aniquila el vacío:

$$Q=\sigma_3+\frac{Y}{2}$$

De este modo, la simetría no está totalmente rota, sino que tenemos un patrón de ruptura:

$$SU(2)\times U(1)_Y\to U(1)_Q$$

**Observación:** este es precisamente el patrón de ruptura que necesitamos en nuestra teoría, porque la carga eléctrica no se rompe espontáneamente. De lo contrario, los fotones también adquirirían masa

Promocionemos ahora la simetría global a local. El patrón de ruptura es el mismo porque solo depende de la simetría. Parametricemos ahora el campo $H$ como siempre, como su valor esperado de vacío + unas fluctuaciones radiales y otras transversales:

$$H(x)=\frac{1}{\sqrt 2}\begin{pmatrix}0 \\ v+h(x)\end{pmatrix}e^{i\frac{\pi_a(x)\tau^a}{2}}$$

del acoplo de los bosones gauge $W_\mu^a , B_\mu$ al valor esperado de vacío del campo $H$ saldrán sus masas

**Observación:** el acoplo es de la forma: $\frac{g^2v^2}{8}[W_\mu^1 W^\mu_1+W_\mu^2 W^\mu_2+(\frac{g'}{g}B_\mu- W_\mu^3)^2]$. Esta forma de escribir el acoplo no es muy conveniente porque no se ve claramente qué bosones están adquiriendo masa. Esto apunta a que no estamos trabajando en la base adecuada para estudiar la asignación de masas. Para diagonalizar dicha parte, rotamos:

$$Z_\mu = \cos\theta_W W_\mu^3-\sin\theta_WB_\mu$$
$$A_\mu = \sin\theta_W W_\mu^3+\cos\theta_WB_\mu$$

donde $\tan\theta_W=\frac{g'}{g}$. Si además definimos $W_\mu^\pm = W_\mu 1 \pm iW_\mu^2$, entonces ya tenemos:

$$\boxed{m_W=\frac{1}{2}gv}$$

$$\boxed{m_Z=\frac{1}{2}\sqrt{g^2+(g')^2}=\frac{m_W}{\cos\theta_W}}$$

**Observación:** como la teoría gauge es no-abeliana, existe interacción entre los campos gauge. Además, por la ruptura espontánea de la simetría, hay acoplo de los campos gauge al campo de Higgs $h(x)$. Concretamente, los acoplos son:

**A 3 campos:**

- $\gamma W^+ W^-$
- $ZW^+W^-$
- $hW^+W^-$
- $hZZ$
- $hhh$

**A 4 campos:**

- $W^+W^-W^+W^-$
- $\gamma \gamma W^+W^-$
- $ZZW^+W^-$
- $\gamma Z W^+W^-$
- $hhW^+W^-$
- $hhZZ$
- $hhhh$

**Observación:** se han medido los acoplos y efectivamente un desarrollo perturbativo en diagramas de Feynman tiene sentido. Concretamente:

$$e\approx 0.303; \ \sin^2\theta_W\approx 0.223; \ g=\frac{e}{\sin\theta_W}\approx 0.64; \ g'\approx \frac{e}{\cos\theta_W}\approx 0.34$$

**Observación:** esta teoría ya no nos da los problemas que teníamos antes de divergencia UV, ahora todo bien

# 4. Sector fermiónico

Ya sé cómo generar las masas de los bosones gauge, pero **cómo genero la masa de los fermiones sin meter término de masas?**

**Solución:** acoplo con el doblete $H$ de la forma:

$$\mathcal L_{mass}=-y\bar LHe_R+h.c$$

Tras darse el SSB, tenemos entonces que $\mathcal L_{mass}\xrightarrow{SSB}-\frac{y}{\sqrt 2}v\bar ee$, que es un término de masas de Dirac

Esto funciona bien porque los neutrinos no tienen masa, de modo que por construcción este término solo da masa a la partícula correspondiente a la segunda componente del doblete de isospin (en el caso de $L_i$, a $e^-, \mu^-, \tau^-$). Esto funciona entonces también para los quarks d, s, b pero no para u, c, t. Para estos, usamos el acoplo:

$$\tilde{\mathcal L}_{mass}=-y\bar Q\tilde Hu_R+h.c.$$

con $\tilde H = i\tau_2H=\begin{pmatrix}v+h(x) \\ 0\end{pmatrix}$

**Problema:** este acoplo no permite que se mezclen las diferentes familias. Esto no tendría por qué ser así y de hecho sabemos que no es así porque observamos decays de kaones en piones (i.e: $s\to u/d$). De este modo, en general el acoplo de Yukawa no puede ser a través de una cte. $y$ sino de una matriz que mezcle familias:

$$\mathcal L_{mass}=-Y_{ij}^d\bar Q_iHd_R^j + h.c.$$

donde $Y_{ij}^{u/d}$ recibe el nombre de **matriz de Yukawa**

Nos interesa expresar el Lagrangiano en la base en la que las masas sean diagonales, porque los estados de masa bien definida son los que se propagan. Nada nos garantiza que la matriz de Yukawa sea diagonalizable, pero sí que podemos "diagonalizarla" (en el sentido más laxo de asignarle una matriz diagonal) mediante una transformación biunitaria: $Y_d = U_d M_d K_d^\dagger$ e ídem para $Y_u$. Entonces, el término de masas (tras SSB) queda para el sector de quarks como:

$$\mathcal L^Q_{mass}=-\frac{v}{\sqrt 2}(\bar d_LU_dM_dK_d^\dagger + \bar u_LU_uM_uK_u^\dagger u_R)$$

donde en esta notación $d_L=\begin{pmatrix}d_L \\ s_L \\ b_L\end{pmatrix}$ y equivalentemente para $u_L$. Puedo redefinir mis multipletes de familia de tal forma que absorban las transformaciones a ambos lados de las matrices diagonales, y aasí obtengo un Lagrangiano diagonal en las masas al precio de haber mezclado los sabores de las familias:

$$\mathcal L^Q_{mass}=-m^d_j \bar d_L^j d_R^j - m_u^j\bar u_L^j u_R^j$$

donde $m^{d/u}_j=\frac{v}{\sqrt 2}(M_{d/u})_{jj}$

**Problema:** la parte cinética me mezcla sabores al pasar a base de masas

**En definitiva:** si seguimos manipulando el Lagrangiano, conseguimos que este quede de la forma:

$$\begin{gather}\mathcal L_{mass}^Q=\bar u_j(i\cancel\partial-m_u^j)u^j+\bar d_j(i\cancel\partial-m^d_j)d_j \\ + e[\frac{2}{3}(\bar u^i\gamma^\mu u^i)-\frac{1}{3}\bar d_i\gamma^\mu d_i]A_\mu \\ + \frac{e}{\sin\theta_W}J^Z_\mu Z^\mu \\ + \frac{e}{\sqrt 2\cos\theta_W}(j_+^\mu W^\mu_++j_-^\mu W^\mu_-)\end{gather}$$

donde $j_+^\mu = \bar u_L^i\gamma^\mu V_{ij}d_L^j$, $j_-^\mu = \bar d_L^i\gamma^\mu V_{ij}u_L^j$ y donde

$$V=\begin{pmatrix}V_{ud} & V_{us} & V_{ub} \\ V_{cd} & V_{cs} & V_{cb} \\ V_{td} & V_{ts} & V_{tb}\end{pmatrix}$$

es la matriz **CKM**

**Observación:** esta matriz es unitaria, y como tal tiene $9$ parámetros libres. Por ende, se puede entender como 3 rotaciones + 6 fases, pero 5 de estas 6 (la que queda es la de carga, que no me puedo quitar) se pueden reabsorber, de modo que puedo parametrizarla como:

$$V=\begin{pmatrix}1 & 0 & 0 \\ 0 & \cos\theta_{23} & \sin\theta_{23} \\ 0 & -\sin\theta_{23} & \cos\theta_{23}\end{pmatrix}\begin{pmatrix}\cos\theta_{13} & 0 & 0 \\ 0 & 1 & 0 \\ -\sin\theta_{13}e^{i\delta} &0& \cos\theta_{13}\end{pmatrix}\begin{pmatrix}\cos\theta_{12}& \sin\theta_{12} & 0 \\ -\sin\theta_{12} & \cos\theta_{23} & 0 \\ 0 & 0 & 1\end{pmatrix}$$

donde se han medido:

$$\theta_{12}\approx 13.02^\circ; \ \theta_{23}\approx-2.36^\circ; \ \theta_{13}\approx 0.20^\circ; \ \delta\approx 64^\circ$$

**Nota:** como en toda matriz unitaria, las columnas (las filas también) sumadas módulo al cuadrado tienen que dar 1. Como la matriz CKM es unitaria por construcción, si la suma diera menor que 1 esto indicaría que nos faltan familias, pero dentro de error la suma sí que da 1
