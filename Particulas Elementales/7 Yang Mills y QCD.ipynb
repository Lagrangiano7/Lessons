# 1. Teorías de Yang-Mills: $SU(N)$

**Nota:** vamos a construir una teoría gauge (en fermiones) con simetría local $SU(N)$. Esto se puede hacer de dos maneras:

1. Al igual que en QED, "a lo bestia": impongo que L tenga invariancia gauge --> me obliga a cambiar derivada --> añado campo compensador a la derivada (la suma de ambos es la derivada covariante), que se transforma como el campo, de modo que la transformación se cancela con la que viene de $\bar \Psi$ y ya tenemos teoría invariante
2. Pidiendo que la diferencia entre dos campos sea consistente con las transformaciones gauge: si la teoría va a ser invariante, entonces debe ser: $\Psi(x)-\Psi(y)=fase\cdot (\Psi'(x)-\Psi'(y))$. Esto obliga a definir un objeto que nos dé la diferencia así, porque de normal no sale (claro, porque la teoría aún no es gauge-invariante) --> "línea de Wilson" --> como ahora sé restar de forma consistente, también sé derivar --> derivada covariante

**Nota:** ahora tenemos un espacio no-trivial adicional, que es el "espacio de N-color". Cuando escribamos $\Psi$ será realmente: $\Psi=\begin{pmatrix}\Psi_1 \\ \vdots \\ \Psi_N\end{pmatrix}$

## 1.1 Generalizando QED a grupos gauge no abelianos: campo compensador

**Proposición:** sea una teoría de fermiones con invariancia gauge $SU(N)$. Entonces, la derivada covariante toma la forma:

$$\boxed{D_\mu = \partial_\mu + igA_\mu}$$

donde $\boxed{A_\mu = A_\mu^aT^a}$ es el **campo compensador** y $T^a$ son los generadores de $SU(N)$

**Proposición:** bajo una transformación gauge, el campo compensador se transforma como en QED + constantes de estructura que QED no tenía por ser abeliano:

$$\boxed{A_\mu\to (A_\mu)'=A_\mu - \frac{1}{g}\partial_\mu \theta^a(x)T^a - i\theta^a(x)[A_\mu, T_a]}$$

**Nota:** lo pongo con "+" porque a mí me sale con "+" en la dem. En "invitation to QFT" va con "+" también (pág. 58, ec. 4.41)

Equivalentemente (en la base habitual del álgebra de Lie de $SU(N)$):

$$\boxed{A_\mu^a\to (A_\mu^a)'=A_\mu^a - \frac{1}{g}\partial_\mu \theta^a(x) - f^{abc}\theta_b(x)A_{\mu c}}$$

con la convención de signo $D_\mu=\partial_\mu+igA_\mu$, en muchos libros se encuentra con el signo contrario. Cambia el signo de la constante de acoplo para obtener esas expresiones

**IMPORTANTE:** las teorías de Yang-Mills siempre dan lugar a mediadores que son (1) bosónicos, de spin 1 (porque campo compensador es cuadrivector); (2) sin masa (porque no aparece término de masa en el Lagrangiano tras imponer invariancia gauge)

**Dem.:**

Partamos de la teoría libre: $\mathcal L = \bar\Psi(i\cancel \partial-m)\Psi$. Vemos que esta no es invariante gauge bajo ningún grupo gauge, porque según metamos una dependencia de punto en el campo, automáticamente la derivada va a sacar un término extra que antes no había, por regla de Leibniz. Entonces, es preciso modificar el operador del término cinético añadiendo un campo que se transforme de la forma contraria al término extra que sale, de tal forma que ambos términos "residuales" de la transformación se cancelen y el Lagrangiano se mantenga invariante (siempre pedimos $\delta\mathcal L =0$, aunque yo creo que esto es demasiado restrictivo, en general bastaría con $\delta\mathcal L = \partial_\mu f^\mu$ --> no, cambia topología, esto es avanzado)

**Observación:** el término de masas no tiene el problema que comentamos porque, como las transformaciones consideradas son unitarias, no se transforma: $\bar \Psi m \Psi = m\Psi^\dagger \gamma^0 \Psi \to m(U\Psi)^\dagger\gamma^0(U\Psi)=m\Psi^\dagger \gamma^0(U^\dagger U)\Psi=\bar \Psi m \Psi$ (véase que $\gamma^0$ vive en el espacio de Dirac, de modo que tiene una identidad en el espacio de "N-color" que ahora estamos considerando. Introduzcamos ahora la derivada covariante: $\mathcal L = \bar\Psi(i\cancel D-m)\Psi$. Inspirados por el término de masa, pedimos que $\cancel D\Psi$ se transforme igual que $\Psi$ para que se cancele la transformación con la que viene de $\bar \Psi$.

¿Cómo se transforma $\Psi$?: $\delta\Psi=i\theta_a(x)T_a\Psi$

¿Cómo se transforma $D_\mu \Psi$?: $\delta(D_\mu \Psi)=\delta([\partial_\mu -igA_\mu]\Psi)=\delta(\partial_\mu \Psi)-ig\delta(A_\mu)\Psi-igA_\mu \delta \Psi$

Ahora impongamos que la derivada covariante se transforme como el campo: $\delta(\cancel D\Psi)=\theta_aT_a(\cancel D\Psi)$ e igualemos, despejando $\delta A_\mu$ y obteniendo el resultado

$$ig\delta A_\mu = i\partial_\mu \theta_a T_a+g\theta_a[A_\mu, T_a]$$

Expresemos el conmutador en términos de las constantes de estructura:

$(A_\mu)' = A_\mu + \frac{1}{g}\partial_\mu \theta_c T_c -i \theta_aA_\mu^b[T^b, T^a]$, donde $[T^a, T^b]=if^{abc}T^c$. Sustituyendo:

$(A_\mu)' = A_\mu + \frac{1}{g}\partial_\mu \theta_c T_c -i \theta_aA_\mu^bif^{bac}T_c$

Expandiendo todo en la base del álgebra $T^c$ **tal que** las constantes de estructura quedan totalmente antisimétricas (esto ya nos fija la base, pero es la habitual)

$(A_\mu^c T_c)'=A_\mu^c T_c + \frac{1}{g}\partial_\mu \theta^c T_c + \theta_aA_\mu^bf^{bac}T_c$

De este modo, obtenemos cómo se transforma cada componente del campo compensador bajo una transformación gauge:

$(A_\mu^c)'=A_\mu^c + \frac{1}{g}\partial_\mu \theta^c + \theta_aA_\mu^bf^{bac}$

Cambiemos $a\leftrightarrow c$ para dejarlo como en el enunciado

$(A_\mu^a)'=A_\mu^a + \frac{1}{g}\partial_\mu \theta^a + \theta_cA_\mu^bf^{bca}$

De nuevo, para dejarlo como en el enunciado, donde $\theta$ y $A_\mu$ tienen los subíndices cambiados, hacemos $b\leftrightarrow c$:

$(A_\mu^a)'=A_\mu^a + \frac{1}{g}\partial_\mu \theta^a + \theta_bA_\mu^cf^{cba}$

Y finalmente usando que en esta base las constantes de estructura son totalmente antisimétricas:

$(A_\mu^a)'=A_\mu^a + \frac{1}{g}\partial_\mu \theta^a - \theta_bA_\mu^cf^{abc}_\blacksquare$

## 1.2 Línea de Wilson

En el anterior apartado, hemos procedido "a lo bruto", imponiendo por nuestros cuernos que el Lagrangiano sea invariante bajo la transformación gauge deseada. Ahora, pidamos algo más elegante: que la diferencia entre dos campos sea coherente con la transformación gauge (i.e: si mi teoría es invariante gauge, es de esperar que una diferencia de campos pueda mapearse a otra consistentemente):

$$|\Psi(y)-\Psi(x)|$$

debería mantenerse invariante gauge. Sin embargo, notamos que no lo hace: $|\Psi(y)-\Psi(x)|\to |U(y)\Psi(y)-U(x)\Psi(x)|$. Por ende, debemos definir un objeto matemático que nos permita definir una diferencia de campos consistente con la transformación gauge: la línea de Wilson

**Def.:** se define la "línea de Wilson" como aquella forma que se transforma como: $\boxed{W(x,y)\to U(x)W(x,y)U^{-1}(y)}$

Esto es, si los campos son vectores en la transformación gauge porque se transforman como $\Psi(x)\to U(x)\Psi(x)$, entonces la línea de Wilson no es más que una forma bilineal porque se transforma como esperaríamos (i.e: como las matrices en Álgebra al cambiar de base) 

**Propiedad (yo):** el objeto $\bar\Psi(y)W(x,y)\Psi(x)$ es gauge-invariante. Según chati, se puede interpretar como "quark en $x$ conectado a antiquark en $y$ a través de un campo gauge". Se usa en Lattice, y también como potencial

**Observación:** la línea de Wilson no es más que el concepto de transporte paralelo para los campos de nuestra variedad: en geometría diferencial, tienes una variedad --> vectores tangentes en cada punto --> si quieres comparar vectores de distintos puntos, tienes que definir una forma de transportar vectores de un punto a otro --> esto se hace con transporte paralelo (paralelo porque no quieres que rote con respecto a la conexión, de forma que llegue "como estaba", aunque solo llegará intacto en espacio euclídeo). Pues ahora en QFT, en cada punto del espacio-tiempo tenemos vectores (los operadores de campo evaluados en ese punto) --> para comparar los campos en dos puntos distintos, hay que transportar uno a la posición del otro --> como la curvatura del fibrado tangente no es 0, hay que definir un transporte paralelo

**Nota:** fíjate que el transporte paralelo no es más que la generalización de las traslaciones en el espacio euclídeo. En este, cuando queremos comparar dos vectores, sencillamente trasladamos uno hasta donde está el otro, de forma que sus orígenes coincidan, y ya los comparamos --> cuando la curvatura es no-nula, no podemos hacer eso, y por eso definimos el transporte paralelo como generalización

**Nota:** componer dos líneas de Wilson (que tocan el mismo punto) da otra línea de Wilson, pero en general estas no forman un grupo porque si no tocan el mismo punto lo que obtenemos ya no es una línea de Wilson porque no se transforma como tal. Si componemos varias líneas de Wilson para dar lugar a un loop, tenemos que el objeto resultante SÍ es otra línea de Wilson, y además se transforma como $W(x,x)$, pero esto no nos permite identificar el loop con 1 porque solo se transforman igual, pero no valen lo mismo (i.e: parece que las transformaciones no determinan completamente las líneas de Wilson)

**Nota:** otra cantidad gauge invariante es la **plaqueta** de Wilson: $Tr W(C)$, porque la traza, siendo cíclica, permite poner las matrices de transformación gauge juntas y cancelarlas

## 1.3 Tensor de fuerza

Ahora que tenemos una derivada covariante en el fibrado principal, podemos construir el Riemann de dicho fibrado como se hace habitualmente en geometría diferencial: estudiando la no-conmutatividad de la derivada covariante

**Proposición:** $[D_\mu, D_\nu]=-ieF_{\mu\nu}$

donde $\boxed{F_{\mu\nu} = \partial_\mu A_\nu-\partial_\nu A_\mu -ig[A_\mu, A_\nu]}$

equivalentemente, por componentes: 

$\boxed{F_{\mu\nu}^a = \partial_\mu A_\nu^a-\partial_\nu A_\mu^a +gf^{abc}A_\mu^bA_\nu^c}$

**Propiedad:** $F_{\mu\nu}$ sigue siendo antisimétrico en sus índices Lorentz

**Transformación gauge:** $F_{\mu\nu}^a\to F_{\mu\nu}^a-f^{abc}\theta_b(x) F_{\mu\nu}^c$

**Def.:** el Lagrangiano que codifica la dinámica del campo gauge es entonces:

$$\boxed{\mathcal L_{YM}=-\frac{1}{4}F_{\mu\nu}^aF^{\mu\nu}_a}$$

**Observación:** el Lagrangiano $\mathcal L_\theta=\theta\varepsilon^{\mu\nu\alpha\beta}F_{\mu\nu}^a F_{\alpha\beta}^a$ es igual de válido, y se transforma igual bajo transformaciones gauge y de Lorentz. Se puede expresar como una derivada total, de modo que clásicamente no importaría en la dinámica de los campos. Sin embargo, en QFT sí que importa porque en la formulación de integral de caminos se ve que pesa distintos sectores topológicos del campo gauge. $F^*F$ es CP impar, de modo que $CP\mathcal L_\theta = -\mathcal L_\theta$ y la teoría ya no es CP invariante, a menos que $\theta=0$. En QCD, esta teoría predice que el neutrón tiene momento dipolar eléctrico $d\propto \theta$, pero experimentalmente se mide que el neutrón no tiene tal momento, de modo que $\theta\leq 10^{-10}$. Actualmente no se sabe por qué esta teoría está tan suprimida. Este es el llamado **strong CP problem**

## 1.4 Puntualización sobre la derivada covariante

**Nota 1:** la derivada covariante es una derivación, en el sentido que cumple la regla de Leibniz. Podría parecer que el campo compensador la rompe, pero no. Veamos por qué:

- En cada evento del espacio-tiempo, tenemos un grupo de Lie $SU(N)(x)$ con su correspondiente álgebra de Lie, $\mathfrak g_{SU(N)}(x)$
- $A_\mu(x)=A_\mu^a(x)T^a\in \mathfrak g_{SU(N)}(x)$ es elemento vector del álgebra de Lie gauge en ese evento
- Aunque escribamos $A_\mu$ "tal cual" en nuestras expresiones, realmente es un elemento de un álgebra de Lie, y como tal no podemos usarlo directamente, sino que tenemos que usar una representación para actuarlo sobre operadores de campo
- De este modo, si tenemos que aplicarlo sobre dos campos que se están multiplicando, en el espacio gauge tenemos: $A_\mu(\phi_1\phi_2)\equiv \rho(A_\mu)(\phi_1\otimes \phi_2)$, de modo que $\rho=\rho_1\otimes \rho_2$, y por teoría de representaciones de álgebras de Lie: $\rho_1\otimes \rho_2=\rho_1\otimes I_2+I_1\otimes \rho_2$. Es de aquí de donde sale que el campo compensador actúa conforme a una regla de Leibniz sobre los operadores de campo, y por ende $D_\mu$ sí es una derivación

**Nota 2:** el campo compensador es una conexión afín. Recordemos que en geometría diferencial, teníamos que la derivada covariante de un cuadrivector era: $D_\nu v^\mu \equiv v^\mu_{;\nu}=v_{,\nu}-\Gamma^\mu_{\alpha\nu}v^\alpha$. Ahora, tenemos la estructura: $D_\mu\Psi=\partial_\mu \Psi-igA_\mu\Psi$. En este sentido, el campo compensador es una conexión afín en el fibrado principal

**Nota 3:** la línea de Wilson es el concepto de transporte paralelo. La motivación para definir la línea de Wilson era que queríamos tener una forma consistente de comparar campos (i.e: restarlos) evaluados en cualesquiera dos eventos del espacio-tiempo. Al igual que en geometría diferencial queríamos una forma de comparar vectores del fibrado tangente en dos puntos distintos de la variedad, ahora estamos haciendo exactamente lo mismo, porque cada evento tiene un campo vectorial de operadores de campo

**Nota 4:** la línea de Wilson se puede interpretar como "la fase acumulada por un campo al moverlo de un evento a otro": $\Psi(x)=W(x,y)\Psi(y)$ donde $W(x,y)=\mathcal Pe^{ig\int A_\mu dx^\mu}$ donde $\mathcal P$ es "path-ordering" y es necesario porque como en general el grupo gauge ya no es abeliano, hay que mantener el orden al integrar. Véase que en QED esto es sencillamente la fase de Berry, que justifica el efecto de Ahronov Bohm, y se interpreta como que el campo EM no solo introduce interacción sino también origina un cambio de fase en los campos que se propagan en su seno

**Nota 5:** si cerramos la línea de Wilson obtenemos un **loop de Wilson**. Intuitivamente, cuantifica "cuánto ha rotado el campo en el espacio de color al recorrer un loop". Naturalmente, esto da de nuevo el tensor de fuerza: $W(C)=\mathcal Pe^{ig\oint_CA_{\mu}dx^\mu}=1+igF_{\mu\nu}+\ldots$

## Sobre las corrientes conservadas

- Cuando nuestra teoría tiene una simetría global, tenemos una corriente Noether conservada: $\partial_\mu j^\mu=0$
- Si pedimos que la teoría tenga una simetría gauge promoviendo esta simetría global a una local, entonces lo que antes era simetría global va a ser ahora un caso particular de la transformación gauge en el que las funciones de la transformación son constantes, de modo que no son dos simetrías distintas. Entonces, ya no es $\partial_\mu j^\mu=0$ a menos que la teoría sea abeliana
- En esencia, lo que ha pasado es que hemos pasado de no tener curvatura en el fibrado principal a sí tenerla, de modo que ahora la corriente es COVARIANTEMENTE conservada: $D_\mu j^\mu=0$
- Analogía útil: en espacio-tiempo plano (e.g: Minkowski) tenemos que el tensor de energía-momento del contenido del sistema es conservado de la forma $T^{\mu\nu}_{,\nu}=0$. Sin embargo, cuando metemos curvatura porque hay gravedad, se tiene por el principio de equivalencia que $T^{\mu\nu}_{;\nu}=0$.
- Es que esto es exactamente lo mismo: existe un principio de equivalencia análogo al de GR, no por nada sino porque el argumento matemático es el mismo: cogiendo coordenadas normales de Riemann, puedo anular la conexión afín en un punto, de modo que recupero SR en el caso de GR anulando el efecto del campo gravitatorio en cada punto (i.e: me dejo caer con el observador en caída libre y recupero SR). En el caso de las simetrías gauge, puedo escoger el gauge de tal forma que el campo gauge se me anule en un punto. En ambos casos, la curvatura sigue estando ahí, igual que pasaba con el Riemann. Puedo hacerlo, porque el potencial no es mensurable, lo mensurable son las fuerzas de marea, y en ambos casos eso no se anula incluso al anular la conexión
- Este "anular la conexión afín en cada punto" y luego transportar paralelamente es precisamente lo que subyace a la prescripción $\partial_\mu \to D_\mu$, y por eso aplica tanto a GR como a teorías gauge

**Observación:** el campo compensador actúa siempre en una representación del álgebra de Lie del grupo gauge (en cada punto, por el carácter local del grupo), de modo que puede actuar diferente dependiendo del objeto sobre el que actúe. Por ejemplo, en QED, tenemos que $\partial_\mu j^\mu=0$ y $D_\mu j^\mu=0$, no $D_\mu j^\mu=0+ieA_\mu j^\mu\neq 0$. Esto se debe a que $j^\mu=\bar \Psi \gamma^\mu \Psi$ es gauge-invariante, de modo que es un "escalar", no tiene índices a los que la conexión ($A$) pueda acoplarse, de modo que no actúa sobre ella y por ende $A_\mu j^\mu=0$

## 2.1 Motivando el color

**Observación:** se observa en el zoo de partículas que existe $\Delta^{++}(u\uparrow u\uparrow u\uparrow)$. En una descripción no-relativista, principio de exclusión de Pauli prohibe esta configuración. Equivalentemente, la función de onda de este sistema es par bajo trasposición de cualquiera de las 3 partículas, porque tenemos: $\Psi_{\Delta^{++}}=\Psi_{esp}\otimes \Psi_{sabor}\otimes \Psi_{spin}$ (nota: la parte espacial as par también porque tiene paridad +)

**Solución:** debe haber una propiedad que permita que la función de onda sea antisimétrica --> color: $\Psi_{\Delta^{++}}=\Psi_{esp}\otimes \varepsilon_{abc}|u\uparrow^au\uparrow^b u \uparrow^c \rangle$.

**OJO!:** los quarks siguen siendo indistinguibles (si no, la antisimetría bajoo intercambio no tendría sentido). El color, tal y como se ha introducido por ahora, es un grado de libertad interno, como el spin

**Observación:** el color NO se observa en la Naturaleza, de modo que debe estar confinado en el interior de los hadrones: **todos los hadrones son singletes de color**

¿Cómo se integra el color como grado de libertad en nuestra QFT?: como el color es un grado de libertad interno pero parece que QCD trata igual a todos los colores, postulamos que nuestra teoría tiene una simetría global $SU(3)$. Queremos que haya dinámica, y para ello promocionamos esta simetría a una local. Esto no es arbitrario, pero tampoco encuentro que esté demasiado bien justificado. Tal vez una justificación posible sea que, como la teoría es indiferente frente al color que tengan los quarks, esta indiferencia debe darse en cada punto del espacio-tiempo

La cuestión es que tenemos entonces una teoría de Yang-Mills con $N=3$, de modo que:

$$\mathcal L_{QCD}=\bar\Psi(i\cancel D-M)\Psi-\frac{1}{4}\mathcal G_{\mu\nu}^a \mathcal G^{\mu\nu}_a$$

**Nota:** esta forma de escribir el Lagrangiano es incluso más compacta que en la sección anterior, porque estamos sumando a todos los sabores de quark, de modo que $\Psi=\begin{pmatrix}q_u \\ q_d \\ q_s \\ \vdots \\ q_t\end{pmatrix}$, a su vez $q_u=\begin{pmatrix}q_{u,R} \\ q_{u,B} \\ q_{u,G}\end{pmatrix}$ y cada uno de estos es un spinor de Dirac. De este modo:

$$\Psi=sabor\otimes color \otimes Dirac \otimes Lorentz$$

Hasta aquí todo ha sido puramente clásico. Ahora cuanticemos y veamos las reglas de Feynman que salen de esta teoría

## 2.2.1 Vértices de QCD

Expandamos el Lagrangiano de QCD:

$\mathcal L = \bar \Psi(i(\cancel\partial-ig\cancel A)-M)\Psi-\frac{1}{4}G^2 = \bar \Psi(i\cancel\partial-M)\Psi +g\bar\Psi\cancel A\Psi-\frac{1}{4}G^2$

$\mathcal L = \mathcal L_{0, ferm}+\mathcal L_{0,gauge}+\mathcal L_{int}$

# 3. Identidades de Slavnov-Taylor

En QED, teníamos la identidad de Ward: $k_\mu M^\mu=0$, que venía de que la amplitud a un orden dado es gauge-invariante. En QCD, tenemos algo análogo, que son las identidades de Slavnov-Taylor. Estas dotan a la amplitud de una forma:

$$\boxed{iM^{\mu\nu}_{ab}=i\delta_{ab}(p^\mu p^\nu-p^2\eta^{\mu\nu})M(p^2)}$$

Como corolario, se sigue cumpliendo una identidad de tipo Ward en QCD, pero que ahora es una instancia particular de unas identidades más generales, que son las de Slavnov-Taylor:

$$\boxed{k\cdot M=0}$$

# 3. Potencial de quarks

En electrodinámica, históricamente primer se cuantiza el espacio de fases de la partícula y se mantiene el campo EM como clásico (i.e: el potencial es el de Coulomb). Después ya se cuantiza el campo electromagnético y obtenemos QED. Sin embargo, como la interacción fuerte no está a la escala mesoscópica sino que a la escala de energía de la Física Clásica queda confinada dentro de los hadrones, no era posible que se hubiera detectado y como tal no hay un potencial clásico previo al modelo puramente de QFT --> ahora el procedimiento es al revés, nos preguntamos cómo será el potencial semiclásico que sienten los quarks debido a sus interaciones fuertes. Este es:

$$\boxed{V_S(r)=A\frac{\alpha_S(r)}{r}+\sigma r}$$

donde:

- $\alpha_S(r)$: constante de acoplamiento de la interacción fuerte, renormalizada a la escala que toque para la separación dada
- $\sigma$: "string tension". Por argumentos de lattice, vemos que el Wilson loop va como $e^{-\sigma S}$ donde $S$ es la superficie del lattice que comprende el loop, y de este comportamiento se concluye que la teoría es confinante. Si no lo fuera, como es el caso de QED, iría con el perímetro en vez del área

## 3.1 Confinamiento

**Observación:** para $r$ grandes, el potencial se vuelve confinante. Sin embargo, creo recordar que este potencial solo es válido en la teoría pura gauge, porque en el momento en el que permitimos la producción de pares de quarks, llegará un momento en el que el tubo de flujo se romperá, dando lugar a un par $q\bar q$ y por ende dejando 2 mesones donde antes solo había uno

**Nota:** el confinamiento parece estar relacionado con el hecho de que QCD se vuelve no-perturbativa a bajas energías. Sin embargo, la demostración analítica de que QCD en efecto es confinante aún es desconocida.

**Nota para mí:** cuando estudiamos los átomos con el método de potencial de Fermi, vemos que incluso para el Helio, que tiene solo 2 electrones, la aproximación de gas de electrones ya funciona muy bien. Tal vez algo parecido se podría aplicar para mesones y bariones

## 3.2 Libertad asintótica

**Observación:** para $r$ pequeños, el string tension apenas contribuye y la teoría tiende asintóticamente a un potencial Coulombiano. Esto no es coincidencia: las cargas fuertes, al igual que las electromagnéticas, son isotrópicas en su potencial. Sin embargo, a energías grandes / distanias pequeñas, la interacción fuerte se torna perturbativa

**Observación:** si miramos los vértices, vemos que si $E\to \infty \Longrightarrow g_S\to 0$, entonces solo nos queda un vértice, que es el de $q\bar qg$, y por ende QCD se comporta asintóticamente como QED

**Nota:** se llama libertad asintótica porque cuanto más se acerquen los quarks, más débil es la interacción entre estos

# 3.3 ¿Las cargas de color se atraen o se repelen?

En QED: las cargas opuestas se atraen, las iguales se repelen

En QCD: todo es repulsivo salvo si tenemos color x anticolor

**Observación:** si tenemos un $q$ y $\bar q$ interactuando, como cada uno está en una irrep. de color, tenemos: $3\otimes\bar 3=8\oplus 1$, donde el octete es repulsivo y el singlete de color es atractivo. **Por eso todos los mesones están en singlete de color**. Un argumento equivalente debe aplicar para los bariones

La "demostración" se puede ver a nivel de árbol (no hace falta más porque estamos con régimen no-relativista) estudiando $q\bar q\to q\bar q$. Si consideramos la interacción electromagnética y tomamos el límite no-relativista, encontramos que $\mathcal M_t=-4m_e^2V$ donde $V$ es el potencial. En el caso de $e^-e^-\to e^-e^-$, tenemos que $\mathcal M_t=-\frac{4m_e^2e^2}{t}V$, de modo que $V=\frac{e^2}{q^2}$, de modo que la interacción entre dos electrones es repulsiva (si pasamos a espacio de momentos, recuperamos Coulomb). Si ahora consideramos el proceso $u\bar d\to u \bar d$, tenemos canal "s" y canal "t" como en el caso de que interaccionaran en QED, pero ahora hay una consideración de color.

Supongamos que tenemos $u^1\bar d^2\to u^j\bar d^c$. Entonces, tenemos:

Canal t: $\mathcal M_t=\frac{(ig_S)^2}{t}[\bar u^j\gamma^\mu u^1][\bar v^c\gamma_\mu v^2](-\frac{1}{6}\delta_{j1}\delta_{2c})$. Por ende, $V=\frac{1}{6}\frac{g_S^2}{q^2}\delta_{j1}\delta_{2c}$ y la interacción: (1) solo se da si cada quark conserva su color; (2) en caso de darse, es repulsiva

**Nota sobre potenciales:** el canal "s" de una aniquilación nunca contribuye a un potencial porque describe estados intermedios. Con todo, en este ejemplo no se calcula el diagrama en canal $s$ porque el vértice iría como $u\bar d g$ y esto no puede ser porque QCD no mezcla sabores, al igual que QED (i.e: es diagonal en sabores)

# 4. Renormalización

**Observación:** cuando calculo la polarización del vacío en QCD (i.e: corrección del propagador del gluón a 1 loop), tengo 4 diagramas que me contribuyen:

1. Par quark-antiquark
2. Loop de gluones
3. Tadpole
4. Loop de ghosts

Centrémonos por ejemplo en el primero: este irá como $\mathcal M_1 \sim \int\frac{d^4k}{(2\pi)^4}\frac{i}{(k-p)^2-m^2+i\varepsilon}\frac{i}{k^2-m^2+i\varepsilon}Tr[\gamma^\mu(\cancel k-\cancel p+m)\gamma^\nu(\cancel k+m)]$

**Observación:** la integral diverge porque cuando $k\to\infty$ tenemos que va como $\frac{1}{k^4}$, y la medida de integración es $d^4k \sim k^3\cdots$, de modo que queda una divergencia logarítmica

**¿Qué se hace? (Regularización dimensional):** (hay otras formas de proceder, como integral de Feynman)

- Hago integral en $d=4-\varepsilon$ dimensiones de modo que ya no diverge (tengo que meter un factor con dimensiones, $\mu^\varepsilon$, para corregir las dimensiones de mi medida) --> ahora me queda algo del tipo "$\frac{1}{\varepsilon}+\log\left(\frac{\tilde \mu^2}{-p^2}\right)+$cosas", donde $\tilde\mu = 4\pi e^{-\gamma_E}\mu^2$

Véase que el resultado sigue siendo divergente cuando tomamos el límite físico $\varepsilon\to 0$

**¿Cómo quito la divergencia?: Renormalización**

1. Redefino la constante de acoplo como: $\alpha_0:=\mu^{\varepsilon/2}Z_\alpha \alpha(\mu)$, donde:
    - $\alpha_0$: "bare coupling constant"
    - $Z_\alpha=1+\frac{\alpha}{3\pi}\frac{1}{\varepsilon}+\ldots$, que también es divergente
    - $\alpha(\mu)$ se llama **running coupling constant** y es FINITA, el infinito que cancela la divergencia va en $Z_\alpha$
2. La divergencia de $Z_\alpha$ cancela la de la amplitud y nos queda algo finito pero que ahora depende de la escala de renormalización

**Teorema:** el resultado Físico no depende de la forma en la que hemos renormalizado. Esto es: $\frac{d\alpha_0}{d\mu}=0$

Esto nos dice cómo depende la running coupling constant de $\mu$, dando lugar a la ecuación del grupo de renormalización:

$$\mu\frac{d\alpha}{d\mu}=\beta(\alpha)$$

QED (1er orden): $\boxed{\beta_{QED}(\alpha)=\frac{2}{3\pi}\alpha^2}$

QCD (1er orden): $\boxed{\beta_S(\alpha)=-\frac{\alpha_S^2}{2\pi}\left(\frac{11}{3}N_C -\frac{2}{3}N_f\right)}$. Con todos los sabores y colores, tenemos: $\beta_S(\alpha)=-\frac{7}{2\pi}\alpha^2=-\beta_0\alpha^2$

**Observación:** QED es perturbativo a bajas energías, porque $\frac{d\alpha_{QED}}{d\mu}>0$. Conforme vayamos a más altas energías, dejará de ser perturbativo, pero la escala a la que esto sucede aún ni la podemos alcanzar en el CERN

**Observación:** QCD (con 3 colores y cualquier número de sabores) es perturbativo a altas energías porque $\frac{d\alpha_S}{d\mu}<0$

**Nota para mí:** la única forma de invertir este comportamiento y que QCD sea perturbativo a bajas energías es si ponemos 1 solo color y los 6 sabores, que no creo que tenga relevancia física porque necesitamos los 3 colores

**Def.:** se define la "escala hadrónica" (previamente fijando la escala de renormmalización) como la escala a la que la constante de acoplo (running) de QCD diverge, de modo que este se torna no perturbativo. Para $\mu=m_Z$, es $\boxed{\Lambda_\chi(m_Z)\approx 300MeV}$

Esto se obtiene resolviendo la ec. del grupo de renormalización de QCD a primer orden: $\alpha_S(\mu)=\frac{1}{\beta_0\log\left(\frac{\mu}{\Lambda_\chi}\right)} \longrightarrow \Lambda_\chi(\mu)=\mu e^{-\alpha_S(\mu)/\beta_0}$
