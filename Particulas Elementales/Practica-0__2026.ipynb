### _SymPy_

Módulo para cálculo simbólico.

[Sympy User Guide](http://docs.sympy.org/latest/index.html "Sympy Tutorial")

```python
# Construcción de expresiones con lenguaje simbólico
# Ojo: mostrar expresiones largas y complicadas en SymPy puede ser lento para el programa

from sympy import * # importar todas las funciones de SymPy
init_printing() # para que las expresiones salgan más bonitas

x, y, z, t, nu, a, l = symbols('x y z t nu alpha lamda') # permite empleo de letras griegas para los símbolos

expr = sin( a ** 2 - l ) + nu ** 3

expr
```

```python
# Definir funciones y ecuaciones

f, g, h = symbols( 'f g h', cls=Function )

Eq( 2 * x ** 2 + 3, -5 )
```

```python
f1, g1, p1 = symbols('f1 g1 p1', cls=Function)

f1 = t ** 2
g1 = t ** 4


p = Piecewise((f1, t <0), (g1, True))
p.subs(t, 2), p.subs(t, -2)
```

```python
# Derivadas

diff( exp(x ** 2), x, 3 )
```

```python
expr = exp( x * sin(y) )
expr
```

```python
dexpr = Derivative( expr, x, 3, y, 7 )
dexpr
```

```python
dexpr.doit()
```

```python
# Integrales

integrate( cos(x) ** 2, x )
```

```python
integrate( exp( -x ** 2 - y ** 2 ), (x, -oo, oo), (y, -oo, oo) ) # infinito = oo
```

```python
integ = Integral( x ** y * exp( -x ), (x, 0, oo) )
integ
```

```python
integ.doit()
```

```python
# Límites

limit( sin(x) / x, x, 0 )
```

```python
expr = Limit( 1 / x, x, 0, '-' )
expr
```

```python
expr.doit()
```

```python
# Series

expr = exp( x * cos(x) )
expr.series( x, pi, 3 )
```

```python
# Resolución de ecuaciones

expr = Eq( x ** 2 - 2 * x - 1, 0 )
solveset( expr, x, domain=S.Reals )
```

```python
roots( x ** 2 + 2 * x + 1, x ) # para ver la multiplicidad de las raíces
```

```python
# Ecuaciones diferenciales

f(x).diff(x)
```

```python
diffeq = Eq( f(x).diff(x, 2) - 2 * f(x).diff(x) + f(x), sin(x) )
diffeq
```

```python
dsolve( diffeq, f(x) )
```

```python
# Simplificación

def my_function(x, A=1, B=0):
    """
    Optional arguments:
        - A: polynomial part (default: A=1)
        - B: trigonometric part (default: B=0)
    """
    m = 2  
    
    return A * x * ( x - 1 ) ** 2 * ( x + 3 ) + B * ( cos( m * x ) + sin( x ) * cos( x ) + sin( x ) ** m )

expr = my_function(x, 1, 1)
expr
```

```python
expr.simplify()
```

```python
# Sustitución de variables

expr.subs(x, 3)
```

```python
expr.subs(x, 3).evalf()
```

```python
# Expandir y factorizar

expr = my_function(x)
expr

expr.expand()
```

```python
expr.collect(x)
```

```python
expr.factor()
```

```python
# Matrices en SymPy

m_rot = Matrix([[cos(x), sin(x)], [- sin(x), cos(x)]])
m_rot, simplify(m_rot.inv())
```

### _Matplotlib_

Módulo para visualización de datos.

[Matplotlib Tutorials](https://matplotlib.org/tutorials/index.html)

```python
import numpy as np 
import matplotlib as mpl 
import matplotlib.pyplot as plt

%matplotlib inline
# necesario incluir esta línea para que nos muestre las gráficas en el notebook
```

```python
# Ejemplo

x = np.linspace( 0, 2 * np.pi, 50) # generamos 50 valores en el eje x

y1 = np.sin( x ) # y evaluamos dos funciones en dichos valores
y2 = np.cos( x )
```

```python
# Representación en subplots:
fig, (ax1, ax2) = plt.subplots(nrows= 1, ncols = 2, figsize=(10,5)) # creamos matriz de plots con cierto tamaño
ax1.plot(x, y1)
ax2.plot(x, y2); # ; para que no muestre línea de texto
```

```python
# O bien pintar ambas en un solo gráfico
fig, ax = plt.subplots(nrows = 1, figsize=(10,5))
ax.plot(x, y1)
ax.plot(x, y2);
```

```python
# Opciones de plots:

fig, ax = plt.subplots(1, figsize=(10,5)) # crear figura
ax.grid(True) # malla activada

ax.plot(x, y1, 'ro', label='seno') # 'ro': puntos rojos, 'b--': línea rayada azul (consultar todas las posibilidades)
ax.plot(x, y2, 'b--', label='coseno') # label para etiquetar cada curva

x_min, x_max, y_min, y_max = ax.axis('tight') # límites de los ejes ajustados a los datos
ax.axis([-0.5, 6.5, y1.min() - 0.5, y1.max() + 0.5]) # cambiamos los límites de los ejes
ax.axes.xaxis.set_label_text("X Axis", fontdict={"size":15}) # etiquetas de los ejes
ax.axes.yaxis.set_label_text("Y Axis", fontdict={"size":15})


ax.title.set_text("Gráficos con Matplotlib") # título de la figura
ax.legend(loc=3); # leyenda
```

```python
# Otros tipos de gráficos

datos = np.random.standard_normal((500, 2))
variable1 = datos[:,0]
variable2 = datos[:,1]
variable3 = np.random.randint(0,3, len(datos))

fig, ax1 = plt.subplots(nrows = 1, figsize=(10,5))
sc = ax1.scatter(variable1, variable2, c = variable3, marker = 'o') # gráfico tipo scatter o de dispersión

ax1.set_title('Diagrama de dispersion con barra de colores')
ax1.set_xlabel('Variable 1')
ax1.set_ylabel('Variable 2')

plt.colorbar(sc);
```

```python
fig, ax1 = plt.subplots(nrows = 1, figsize=(10,5))

ax1.hist(datos, label=['columna0', 'columna 1'], bins = 25) # gráfico tipo histograma

ax1.set_xlabel('Valores')
ax1.set_ylabel('Frecuencia')

ax1.set_title('Histograma')
ax1.legend(loc=0);
```

```python
# Para plots en 3D:

from mpl_toolkits.mplot3d import Axes3D  # noqa: F401 unused import

fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(111, projection='3d')

r = np.linspace(0, 1.25, 50)
p = np.linspace(0, 2 * np.pi, 50) # coordenadas polares
R, P = np.meshgrid(r, p) # crear el grid de coordenadas de los ejes
Z = (( R ** 2 - 1 ) ** 2 )
X, Y = R*np.cos(P), R*np.sin(P) # coordenadas cartesianas

ax.plot_surface(X, Y, Z, cmap=plt.cm.YlGnBu_r)

ax.set_zlim(0, 1)
ax.set_xlabel(r'$\phi_\mathrm{real}$', fontsize=20) # texto LaTeX en las etiquetas
ax.set_ylabel(r'$\phi_\mathrm{im}$', fontsize=20)
ax.set_zlabel(r'$V(\phi)$', fontsize=20)

plt.show();
```
