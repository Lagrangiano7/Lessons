**Objetivos:**

- Entender cómo se construye el Lagrangiano de teoría de campo vectorial libre y por qué son bosones
- Cuantizar la teoría

# 1. Campo vectorial masivo

La clasificación de Wigner clasifica las partículas elementales (las irreps. del grupo de Poincaré) por su spin y masa. Como tenemos ahora S=1 (vectorial), tenemos $2S+1=3$ grados de libertad, pero no hay irrep. del grupo de Lorentz con tal dimensión --> tendremos que usar una de dimensión 4 y tendremos un grado de libertad espurio



## 1.1 Construyendo teoría

- Tenemos que construir teoría que tenga 3 grados de libertad (i.e: que permita una ligadura)
- Tras prueba y error, parametrizamos un Lagrangiano del tipo: $\mathcal L = \frac{a}{2}A_\mu\square A^\mu + \frac{b}{2}A_\mu\partial^\mu\partial_\nu A^\nu+\frac{1}{2}m^2A^2$
- Cojo Euler-Lagrange y aplico $\partial_\mu$, de modo que me queda: $[(a+b)\square+m^2]\partial_\mu A^\mu=0$, así que si escojo $a=-b$, tengo $\partial_\mu A^\mu=0$
- Esta ecuación no es dinámica, es una ligadura, y por ende me restringe un grado de libertad --> ya tengo los 3 que quería
- **Proposición:** el Lagrangiano de un campo vectorial masivo libre es:

$$\boxed{\mathcal L = -\frac{1}{4}F_{\mu\nu}F^{\mu\nu}+\frac{1}{2}m^2A^2}$$

- **Observación:** el término de masa rompe la invariancia gauge típica de $A_\mu$

**Proposición:** con esta teoría, la densidad de energía del campo es definida positiva: $T^{00}>0$, de modo que se cumple la condición de energía débil de RG

## 1.2 Ecuaciones del movimiento

- Por E.L., tenemos entonces que este campo cumple Klein-Gordon en cada componente, con la ligadura adicional que hemos encontrado (esto es fundamental porque, si no, tendríamos 4 grados de libertad)

$$\boxed{\begin{gather}(\square + m^2)A_\mu=0; \\ \partial_\mu A^\mu=0 \end{gather}}$$

## 1.3 Cuantización

**Observación:** el campo vectorial masivo obedece una ecuación de Klein-Gordon en cada componente (+ ligadura) --> cada componente deberá ser como un campo escalar (real, porque si no el conteo de grados de libertad es el doble), y eso ya sé cómo se cuantiza

$$A_\mu(x)=\displaystyle\sum_{i=1}^3\int\frac{d^3\vec p}{(2\pi)^3\sqrt{2E_p}}[\epsilon_\mu^ia_i(p)e^{-ip\cdot x}+\epsilon^{i*}_\mu a_i^\dagger(p)e^{ip\cdot x}]$$

**Interpretación:** $a_i^\dagger (p)$ crea una partícula con trimomento $\vec p$ y con polarización $\epsilon^i_\mu$

# 2. Campo vectorial sin masa

- Ahora queremos estudiar el caso de una partícula de S=1 que no tiene masa (por ejemplo, fotón)



## 2.1 Construyendo teoría

- Por consistencia con el caso masivo, podemos tomar $m\to 0$ y tenemos la teoría del campo vectorial libre sin masa:

$$\mathcal L = -\frac{1}{4}F^2$$

## 2.2 Libertad de gauge

**Observación:** esta teoría permanece inalterada si hacemos la transformación: $A_\mu\to A_\mu+\partial_\mu \alpha$. Esto se llama "libertad de gauge" y veremos en la teoría completa de QED que está asociada a la invariancia gauge $U(1)$

## 2.3 Conteo de grados de libertad: 2

**Teorema:** el campo vectorial libre sin masa tiene 2 grados de libertad, no 3 como el caso masivo

**Dem:**

- Separo ecs. de E-L en $\mu=0$ y $\mu=i$--> $-\partial_i^2A_0-\partial_t \partial_i A_i=0$ + parte de "i"
- Cojo gauge de Coulomb: $\partial_j A_j=0$ (-1 grado de libertad)
- Ahora tengo que la ec. de $\mu=0$ es: $\partial_i^2A_0=0$
- Siempre que $\partial_i^2\alpha=0$ en mis transformaciones gauge, estas dejan todo inv. porque $\partial_i A_i \to \partial_i A_i+\partial_i^2\alpha$
- Si derivo otra vez, me sale la primera ec. del mov., así que se cumple trivialmente --> es otra ligadura, o, si se quiere, puedo hacer incluso $A_0=0$ (-1 grado de libertad)
- Por ende, tengo 2 grados de libertad --> esto explica que el fotón tenga 2 polarizaciones y no 3 si tiene spin 1

## 2.4 Cuantización

- Igual que en caso masivo, pero ahora solo hay 2 polarizaciones físicas (nota: los fotones virtuales tendrán 3 polarizaciones porque no están en su capa de masas, pero da igual porque la polarización de un fotón virtual no entra en su propagador)

$$A_\mu(x)=\displaystyle\sum_{i=1}^2\int\frac{d^3\vec p}{(2\pi)^3\sqrt{2E_p}}[\epsilon_\mu^ia_i(p)e^{-ip\cdot x}+\epsilon^{i*}_\mu a_i^\dagger(p)e^{ip\cdot x}]$$

## 2.4 Propagador del fotón

**Proposición:** el propagador del fotón es: $\Pi_{\mu\nu}=\frac{1}{p^2}[\eta_{\mu\nu}-(1-\xi)\frac{p_\mu p_\nu}{p^2}]$

Normalmente, se coge determinación de Feynman-t'Hooft ($\xi=1$), de modo que queda el propagador que siempre se ve:

$$\boxed{i\Pi_{\mu\nu}=-i\frac{\eta_{\mu\nu}}{p^2+i\varepsilon}}$$

**Dem:**

- Quiero calcular $\Pi_{\mu\nu}(x-y):=\langle 0|T(A_\mu(x)A_\nu(y))|0\rangle=i\int\frac{d^4p}{(2\pi)^4}\Pi_{\mu\nu}(p)$
- Truco: en vez de hacer la integral a lo bestia, resuelvo el caso donde existen corrientes, porque ahí la función de Green sale convolucionada directamente, de la forma: $A_\mu=\Pi_{\mu\nu}J^\nu$
- Ecuaciones con corrientes: $\partial_\mu F^{\mu\nu}=J^\nu$
- Paso a momentos: $(-p^2\eta^{\mu\nu}+p^\mu p^\nu)A_\mu=J^\nu$
- Parece que ya casi lo tengo, porque solo me queda invertir el operador que actúa sobre $A$, pero no se puede porque su determinante es 0 (fíjate que es simétrico)
- Físicamente, no se puede porque diferentes elecciones de gauge me dan diferentes propagadores
- Solución: como sé que la Física es independiente del gauge que elija (porque mi teoría tiene esa invariancia, incluso en presencia de corrientes), elijo el gauge que más me guste (Lorenz) y lo meto en la teoría como una ligadura, con su correspondiente multiplicador de Lagrange
- $\mathcal L = -\frac{1}{4}F^2-J_\mu A^\mu - \frac{1}{2\xi}(\partial_\mu A^\mu)^2$
- Ahora, las ecuaciones del movimiento en representación de momentos me quedan de la forma:
- $(-p^2\eta^{\mu\nu}+(1-\frac{1}{\xi})p^\mu p^\nu)A_\mu=J^\nu$
- Ahora sí puedo invertir porque el determinante ya no es 0, y me queda:
- $\Pi_{\mu\nu}=\frac{1}{p^2}[\eta_{\mu\nu}-(1-\xi)\frac{p_\mu p_\nu}{p^2}]$

Y ya hemos acabado. Véase que puedo seleccionar la determinación que quiera escogiendo $\xi$, porque la he metido con la ligadura del gauge, pero la teoría es independiente del gauge
