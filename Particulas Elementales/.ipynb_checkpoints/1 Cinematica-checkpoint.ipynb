# Tema 1: Cinemática relativista

**Objetivo:** fijar la cinemática de los procesos más comunes

1. Scattering $2\to 2$
2. Desintegraciones a 2 cuerpos
3. Desintegraciones a 3 cuerpos

## 1. Scattering $2\to 2$

### Grados de libertad: 2

**Proposición:** un proceso de scattering 2 a 2 tiene 2 grados de libertad cinemáticos, que tomamos como $s, t$ (Mandelstam)

**Dem:**

- Consideremos lo que sucede en un proceso 2->2 a nivel de cuadrimomentos. Tenemos: $p_A^\mu=(E_A,\vec p_A)$ y $p_B^\mu=(E_B, -\vec p_A)$
- Por ende, tenemos a priori 3 grados de libertad: las 3 componentes del momento cinético de $A$, por ejemplo
- Sin embargo: $p_A^2=m_A^2$, de modo que tenemos una ligadura --> 2 grados de libertad. También podíamos tomar la energía del sistema en el centro de masas y el ángulo de scattering, que es equivalente a tomar $s, t$ $_\blacksquare$

### Determinación de energías (CM)

**Proposición:** la energía de cada partícula (ya sea reactivo o producto) solo depende de $s$ de la siguiente forma:

$$\boxed{\begin{gather}E_A = \frac{s+m_A^2-m_B^2}{2\sqrt s} \\ E_B = \frac{s+m_B^2-m_A^2}{2\sqrt s} \\ E_C = \frac{s+m_C^2-m_D^2}{2\sqrt s} \\ E_D = \frac{s+m_D^2-m_C^2}{2\sqrt s}\end{gather}}$$

**Dem:**

Hacemos una y las otras son iguales. Por ejemplo, para la partícula $A$ del proceso $A+B\to C+D$, tenemos:

$s=(E_A+E_B)^2=E_A^2+E_B^2+2E_AE_B$

También es útil: $E_B^2=E_A^2+m_B^2-m_A^2$, y ahora sustituimos dejando todo en potencias de $E_A$:

$s=2E_A^2+2E_AE_B++m_B^2-m_A^2$

Sacando factor común de $E_A$, tenemos: $E_A(E_A+E_B)=E_A\sqrt{s}$, así que llegamos a la expresión que buscábamos demostrar $_\blacksquare$

**Proposición:** el módulo del momento cinético viene fijado por una función de Källén de $s$:

$$\boxed{|\vec p_A|^2=\frac{\lambda(s, m_A^2, m_B^2)}{4s}}$$

donde $\lambda(x, y, z)=(x+y-z)^2-4xy$

**Nota:** véase que las energías y magnitudes equivalentes no dependen de $t$, lógicamente, porque la energía de una partícula en el vacío no puede depender del ángulo con el que salga dispersada

**Proposición:** el ángulo de scattering viene completamente determinado por $s, t$ (aunque yo creo que se podría sacar más fácil directamente de $t$ que usando la expresión a continuación):

$$\boxed{\cos\theta=\frac{s(t-u)+(m_C^2-m_D^2)(m_A^2-m_B^2)}{\sqrt{\lambda(s, m_A^2, m_B^2)\lambda(s, m_C^2, m_D^2)}}}$$

### Determinación de energías (LAB)

**Nota:** en el sistema de referencia de laboratorio (que en general es cualquier inercial que no sea el de CM), es de esperar que las energías de las partículas individuales sí dependan del ángulo con el que estas salen dispersadas, porque el CM ya no está en reposo con respecto de nuestro inercial

**Nota:** ahora habrá 2 ángulos de scattering a priori. Se puede demostrar que solo uno de ellos es independinte, de modo que nos quedaremos con el formado por $\hat{AC}$ como siempre

**Nota:** supongamos por simplicidad que el inercial de LAB es aquel en el que $B$ está en reposo. Entonces, en este inercial:

$$\boxed{\begin{gather}E_A = \frac{s-m_A^2-m_B^2}{2m_B} \\ E_B = m_B \\ E_C = \frac{m_B^2+m_C^2-u}{2m_B} \\ E_D = \frac{m_B^2+m_D^2-t}{2m_B}\end{gather}}$$

Además, los módulos de los momentos van como funciones de Callen de $t$:

$$\boxed{\begin{gather}|\vec p_C|^2=\frac{\sqrt{\lambda(t, m_B^2, m_C^2)}}{2m_B} \\ |\vec p_D|^2=\frac{\sqrt{\lambda(t, m_B^2, m_D^2)}}{2m_B}\end{gather}}$$

Finalmente, el ángulo de scattering de A con C es:

$$\boxed{\cos\theta=\frac{(u-m_C^2)(t-m_D^2)+m_B^2(2m_C^2+2m_D^2-s-m_A^2)}{\sqrt{\lambda(t, m_B^2, m_D^2)\lambda(u, m_B^2, m_C^2)}}}$$

## 2. Desintegraciones a 2 cuerpos

### Grados de libertad: 0

**Proposición:** las desintegraciones a 2 cuerpos vienen completamente fijadas por la cinemática

**Dem:**

El estado inicial contribuye con 0 grados de libertad, lógicamente. El estado final tiene, a priori, 3 grados de libertad (las 3 componentes del momento cinético de una de las partículas hijas). Impongamos las ligaduras y veamos que tenemos tantas como grados de libertad a priori:

1. $p_B^2=m_B^2\longrightarrow -1$
2. $p_C^2=m_C^2\longrightarrow -1$
3. $E_B+E_C=m_A \longrightarrow -1$

Por ende, hay 0 grados de libertad en el sistema $_\blacksquare$

**Nota:** en scattering 2->2, al situarnos en el inercial de CM, ya estamos escogiendo un eje privilegiado en el espacio (el eje en el que se mueven las partículas), de modo que el ángulo con el que salen las partículas de salida tiene significado. En este caso, sin embargo, no hay forma con significado físico de asignar un eje, y esta asignación es por ende arbitraria, así que el ángulo con el que salgan los productos no puede contarse como grado de libertad

## 3. Desintegraciones a $n$ cuerpos

### Grados de libertad: $3n-7$

### Constrains e igualdades cinemáticas útiles

- Sea un proceso de desintegración a $n$ cuerpos: $A\to 1+\ldots+n$
- **Def.:** sea $t_k=(p_A-p_k)^2$
- **Constraint:**

$$\boxed{\left(\displaystyle\sum_{j\neq k}^n m_j\right)^2 \leq t_k \leq (m_A-m_k)^2}$$

- **Relación: suma de t's**

$$\boxed{\displaystyle\sum_{k=1}^n t_k=(n-2)m_A^2+\displaystyle\sum_{k=1}^n m_k^2}$$

- **Def.:** sea $s_{ij}=(p_i+p_j)^2$

- **Relación: suma de s's**

$$\boxed{\displaystyle\sum_{i<j}^ns_{ij}=(n-2)\displaystyle\sum_{i=1}^n m_i^2+m_A^2}$$

**Dem:**

$\sum_{i<j} s_{ij}^n=\sum_{i<j}^n (m_i^2+m_j^2+2p_i\cdot p_j)$

- Primera parte: hacemos la suma directamente, y vemos que cada término de masa aparece $n-1$ veces. Entonces, es igual a $(n-1)\displaystyle\sum_{i=1}^n m_i^2$

- Segunda parte: recurrimos astutamente a la expansión del cuadrado de un sumatorio, concretamente del de cuadrimomentos, porque involucra justamente el término de interés:

$\left(\sum_{i=1}^np_i\right)^2\equiv m_A^2=\sum_{i=1}^nm_i^2+2\sum_{i<j}p_i\cdot p_j$

Sustituyendo las dos relaciones que hemos encontrado en la expresión de partida, llegamos directamente a la fórmula que queríamos demostrar $_\blacksquare$

### Determinación de energías y ángulos

$$\boxed{E_i = \frac{m_A^2+m_i^2-t_i}{2m_A}}$$

$$\boxed{|\vec p_i|=\frac{\sqrt{\lambda(m_A^2, m_i^2, t_i)}}{2m_A}}$$

$$\boxed{\cos\theta_{jk}=\frac{(m_j^2-t_j)(m_k^2-t_k)+m_A^2(2(m_j^2+m_k^2)-t_i-m_i^2)}{\sqrt{\lambda(t_j, m_A^2, m_j^2)\lambda(t_k, m_A^2, m_k^2)}}}$$

## 4. Desintegraciones a 3 cuerpos

**Proposición:** por conservación de cuadrimomento, es inmediato que las s's y las t's están relaciondas uno a uno:

$$\boxed{\begin{gather}t_1=s_{23} \\ t_2 = s_{13} \\ t_3 = s_{12}\end{gather}}$$

### Diagramas de Dalitz

**Proposición:** sea una desintegración a 3 cuerpos $A\to 1+2+3$. Este sistema tiene 2 grados de libertad, que podemos tomar como $s_{12}, s_{23}$. El tercero vendrá fijado por la regla de suma de s's: $s_{12}+s_{13}+s_{23}=m_1^2+m_2^2+m_3^2+m_A^2$

**Proposición:** aunque las variables son independientes, una acota a la otra, de modo que una desintegración no puede ocupar cualquier par de valores en el plano $(s_{12}, s_{23})$, sino que quedan en una especie de triángulo conocido como diagrama de Dalitz

**Proposición:** $s_{23}\Big |^M_m = \frac{1}{4s_{13}} \{(m_A^2+m_3^2-m_1^2-m_2^2)^2 - [\lambda^{1/2}(s_{13}, m_A^2, m_2^2) \mp \lambda^{1/2}(s_{13}, m_1^2, m_3^2)]\}$

**Uso 1 de los diagramas de Dalitz: desintegraciones a 3 cuerpos**

El diagrama de Dalitz en el fondo nos da un mapa experimental del módulo de la amplitud del proceso de desintegración: $|\mathcal M(s_{13}, s_{23})|$

Cada punto es un evento, así que la estructura del diagrama nos dice lo siguiente:

1. Vacío: no hay decay
2. Uniforme: decay dominado por el espacio de fases (i.e: no hay estructura cinemática especial)
3. Deformado: el proceso viene en parte fijado por la cinemática
4. Bandas: resonancias

**Uso 2: detección de resonancias**

La mayoría de decays no son a 3 cuerpos, y los que sí lo parecen, suelen ser A --> resonancia + partícula --> 3 cuerpos

**Proposición:** las resonancias aparecen como bandas en el diagrama de Dalitz (horizontales, verticales o con pendiente $-1$ dependiendo de los ejes que hayamos cogido)

**Dem:**

Consideremos el diagrama de Dalitz con ejes $(s_{13}, s_{23})$

Sea un proceso de decay con resonancia tal que $A\to R+2\to (1+3)_R+2$

Tenemos que $s_{13}\equiv (p_1+p_3)^2 \equiv p_R^2=m_R^2 = cte.$ Por ende, pase lo que pase en los demás canales (ejes), tendremos una banda vertical en $s_13=m_R^2$. Es una banda y no una recta porque las resonancias no tienen masa bien definida, sino que esta va con cierta anchura (supongo que son tan cortas que $\Delta E \delta t$ hace que se ensanche mucho la banda de energía, y con ella la de masa)

Si equivalentemente tenemos $A\to R+1\to (2+3)_R+1$, entonces tendremos una banda horizontal con $s_{23}=m_R^2$

Finalmente, si tenemos $A\to R+3\to (1+2)_R+3$, entonces es $s_{12}=m_R^2$, pero esta no coincide con ninguno de nuestros ejes. Sin embargo, recordando la condición on-shell para decays a 3 partículas, tenemos: $s_{12}+s_{13}+s_{23}=m_A^2+m_1^2+m_2^2+m_3^2$. De este modo, tenemos una banda en el diagrama de Dalitz tal que: $s_{23}=-s_{13}+\text{ masas}-m_R^2$
