# 1. Ruptura espontánea de la simetría

**Proposición:** sea una teoría con cierta simetría global de dimensión $N$. Entonces, existen $N$ operadores de carga conservados $\dot Q_a=0$, y el álgebra de operadores de estas cargas es isomorfo al álgebra de Lie del grupo de simetría

$$\boxed{\mathfrak g_Q\cong \mathfrak g_G}$$

De hecho, el álgebra de operadores de carga no es sino una representación del álgebra (abstracta) del grupo de Lie.

**Proposición:** estos operadores de carga son generadores de la simetría, en el sentido de que:

$$[Q_a(t), \phi(t, \vec x)]=-i\frac{\delta \phi}{\delta\alpha_a}$$

**Proposición:** los observables se transforman bajo una simetría global como: $\phi\to U\phi U^{-1}$, si bien luego en la representación concreta del grupo en la que estemos esta transformación puede ser lineal, por eso escribimos $\phi\to U\phi$. Un ejemplo típico es el de las rotaciones: $R x R^{-1}=R_{ij}x_i$

**Ruptura espontánea de la simetría:**

- Sea un estado de vacío en interacción $|\Omega\rangle$. Su energía no tiene por qué ser nula, sino que en general: $H|\Omega\rangle = E_0|\Omega\rangle$
- Sea una carga conservada $Q$, de modo que $[Q(t), H(t)]=0$
- Supóngase que esta carga NO aniquila el vacío: $Q|\Omega\rangle \neq 0$
- Entonces, existe un estado $Q|\Omega\rangle$ con la misma energía que $|\Omega\rangle$, porque $H(Q|\Omega\rangle)=E_0(Q|\Omega\rangle)$ por la conmutatividad del Hamiltoniano y la carga
- **Def.:** un generador de simetría que no deja el vacío invariante se dice **roto**

Veámoslo de otra manera: sea una teoría que tiene cierta simetría continua y cierto potencial $V(\phi)$. Entonces, se define el estado de vacío como aquel para el que el valor esperado del campo es mínimo (i.e: el mínimo del potencial):

$$\langle \Omega|\phi_i|\Omega\rangle=v_i$$

Tomemos ahora el conmutador de una de las cargas con uno de los campos:

$$\langle \Omega|[Q_a, \phi_i]|\Omega\rangle \equiv (T^a)_{ij}v_j$$

donde $T^a$ es el generador asociado a $Q_a$. Si el operador de carga aniquila el vacío, entonces se tiene que $T^a_{ij}v_j=0$, y por eso se dice coloquialmente que **el generador de una simetría no-rota aniquila el vacío**. Sin embargo, estrictamente los generadores de un grupo de simetría no son operadores en un espacio de Hilbert así que no actúan sobre el vacío, que es un ket en un espacio de Hilbert (o de Fock, da igual). En su lugar, deben hacerlo las cargas. Como es más práctico y además equivalente, muchas veces diremos que una simetría está rota si el generador actuando sobre el valor esperado de vacío no lo aniquila o, lo que es lo mismo, si la transformación uniparamétrica asociada al generador considerado sobre el valor esperado de vacío no lo deja invariante

**Proposición:** dado un generador de simetría roto, estas tres afirmaciones son equivalentes:

1. Su operador de carga asociado no aniquila el **estado** de vacío: $Q_a|\Omega\rangle\neq 0$
2. El generador no aniquila el **valor esperado** de vacío: $T^a_{ji}v_j\neq 0$
3. La transformación uniparamétrica asociada a este generador no deja el **valor esperado** de vacío invariante: $(e^{iT_a\alpha_a})_{ji}v_j\neq v_j$ (sin suma en los índices $a$)

**Interpretación:** sobre todo el tercer punto lo deja claro: la ruptura espontánea de la simetría se da cuando una teoría tiene una cierta simetría pero el valor esperado de vacío no la tiene

**Teorema de Goldstone:** por cada generador de una simetría global espontáneamente roto, aparece un bosón de masa nula con los mismos números cuánticos que el generador

Nota: ahora ya no nos sorprende hablar de que un generador tenga números cuánticos: como generador de un álgebra abstracta no los tiene, pero su representación, que es el operador de carga, sí.

# 2. Simetrías de isospin y quiral de QCD

## Simetría de isospin

Consideremos QCD con solo dos sabores (por ejemplo, que estamos a tan baja energía que los demás sabores no se acoplan a la dinámica). Entonces, tenemos:

$$\mathcal L=\bar u(i\cancel D-m_u)u+\bar d(i\cancel D-m_d)d-\frac{1}{4}G^2$$

En buena aproximación, tenemos que $m_u\approx m_d$, así que:

$$\mathcal L^{iso}=\bar u(i\cancel D-m)u+\bar d(i\cancel D-m)d-\frac{1}{4}G^2$$

Y podemos comprimir la notación poniendo ambos quarks en un doblete de sabor: $\Psi=\begin{pmatrix}u \\ d\end{pmatrix}$

De modo que ahora:

$$\mathcal L^{iso}= \bar \Psi(i\cancel D-m_u)\Psi-\frac{1}{4}G^2$$

donde el paréntesis es obviamente la identidad en el espacio de sabor para que quede lo mismo que antes, no hemos hecho nada aún, pero esta notación es sumamente útil porque hace explícita la simetría que pueda tener la teoría debido a que las masas son iguales

**Observación:** la teoría es simétrica bajo la transformación global $SU(2)$. Esto es el isospin, y de aquí viene la convención de que el up tiene isospin 1/2 y el down -1/2

## Simetría quiral

Realmente, la masa de los quarks es tan pequeña comparada con incluso el mesón más ligero (el pion) que podemos admitir que son massless en buena aproximación. Es un resultado conocido de la teoría de fermiones de Dirac que al tomar masas nulas la dinámica de las quiralidades se desacopla. Del Lagrangiano original se ve entonces que:

$$\mathcal L^{quiral}=\bar u_L(i\cancel D)u_L+u_R(i\cancel D)u_R+\bar d_L(i\cancel D)d_L+\bar d_R(i\cancel D)d_R-\frac{1}{4}G^2$$

Nota: acuérdate de que $u_{L/R}$ en esta notación no es el spinor de Weyl literalmente, sino el spinor de Dirac con solo el spinor de Weyl y la parte del otro spinor a 0, es decir, es la proyección: $u_{L/R}=P_{L/R} \ u$

Podemos volver a poner los spinores en dobletes, ahora con quiralidad definida:

$$\mathcal L^{quad}= \bar \Psi_L(i\cancel D)\Psi_L+\bar \Psi_R(i\cancel D)\Psi_R-\frac{1}{4}G^2$$

**Observación:** este Lagrangiano tiene una simetría $SU(2)_L\times SU(2)_R$. Es decir, podemos hacer las mismas rotaciones que en el límite de isospin pero para cada quiralidad por separado

**Proposición:** se pueden tomar combinaciones lineales de las corrientes conservadas de tal forma que tengan paridad bien definida. Tendremos entonces corientes vectoriales y axiales, y sus correspondientes cargas. Véase que la carga vectorial será par bajo paridad y la axial, impar. Por ende, esperamos que haya estados de paridad distinta de igual energía (i.e: degenerados en masa), porque si el estado $|\Psi\rangle$ es par bajo paridad, entonces $Q_A|\Psi\rangle$ será impar pero de igual energía

**Observación:** esto no es lo que se observa, sino que los dos estados más ligeros de paridades opuestas del espectro hadrónico se encuentran separados por un 100% de la masa del más ligero. Este efecto no puede ser introducido por el hecho de que la simetría quiral es aproximada, porque es demasiado grande. Lo que está pasando es que la simetría quiral está espontáneamente rota, y $Q_A|\Psi\rangle$ no tiene ni que ser un estado de partícula bien definido

$$SU(2)_L\times SU(2)_R\to SU(2)_{L+R}$$

**Problema:** idealmente, quiero estudiar esta ruptura en QCD, pero no sé nada sobre su vacío

**Solución:** formulo la teoría "de juguete" (resulta que no lo es tanto) que presente la misma simetría quiral con el mismo patrón de ruptura espontánea

# 2. Modelo $\sigma$ lineal escalar

Consideremos la teoría de un campo escalar complejo:

$$\mathcal L_\sigma = \partial_\mu \phi \partial^\mu \phi^* +m^2\phi\phi^*-\frac{\lambda}{4!}(\phi\phi^*)^2; \ \ m,\lambda>0$$

**Observación:** el término de masas lleva el signo cambiado con respecto a cómo suele entrar, porque $m$ no es ahora la masa de los campos sino un parámetro del potencial. La masa de los campos será la derivada segunda del potencial con respecto a estos, como en Física del Estado Sólido

**Observación:** el potencial es: $V(\phi)=-m^2\phi\phi^*+\frac{\lambda}{4!}(\phi\phi^*)^2$

**Observación:** la teoría tiene una simetría $U(1)$ global

**OJO!:** la masa de la partícula depende del potencial, como sucede a los electrones dentro de un cristal, de modo que tenemos:

$$(M_{eff})_{ij}=\frac{\partial^2V}{\partial \phi_i\phi_j}$$

donde $\phi = \phi_1+i\phi_2$

![Mexican Hat Potential](Mexican_Hat_Potential.jpg)

De este modo, en $\phi=0$ tenemos masa imaginaria (la config. es inestable, es máximo de potencial) y en el mínimo tenemos $m_{eff}=\sqrt 2m$

**Observación:** el potencial presenta infinitos mínimos (i.e: infinitas configuraciones de mínima energía), y por ende podemos definir infinitos estados de vacío: $|\Omega_\theta\rangle$ tq. $\langle \Omega_\theta|\phi|\Omega_\theta\rangle = \sqrt{\frac{2m^2}{\lambda}}e^{i\theta}$

Puedo elegir uno, el que sea, así que por simplicidad defino el vacío como aquel en el que $\theta=0$

**Observación:** el valor esperado de vacío del campo es: $\langle \Omega|\phi|\Omega\rangle=v=\sqrt{\frac{2m^2}{\lambda}}$

**CRUCIAL:** al escoger un vacío concreto, ya estoy rompiendo la simetría $U(1)$, porque la teoría la sigue teniendo pero el valor esperado de vacío del campo ya no (i.e: he escogido una fase, si transformo $U(1)$ me estaré yendo a otros estados que ya no son mi vacío)

Es habitual que el valor esperado del vacío de un campo sea 0, así que parametricemos el campo de la forma:

$$\phi(x)=\frac{1}{\sqrt 2}(v+\sigma(x))e^{i\pi(x)/F}$$

donde obviamente la config. de vacío corresponderá ahora a $\pi=0, \sigma=0$. Literalmente estamos parametrizando el campo en una especie de coordenadas polares porque así queda explícita la simetría del potencial, del cual es la variable

De este modo, queda un Lagrangiano de la forma:

$$\mathcal L_\sigma=\frac{1}{2}\partial_\mu \sigma \partial^\mu\sigma-V(\sigma)+\frac{1}{2}\partial_\mu \pi\partial^\mu \pi+\mathcal L_{int}(\sigma, \pi)$$

**Observación:** la fase del campo es ahora otro campo de masa 0. Este es precisamente el bosón de Goldstone proveniente de la ruptura espontánea de la simetría $U(1)$ global. El campo $\sigma$ tiene masa porque cuesta energía moverse radialmente en el potencial. El campo $\pi$ no la tiene porque como el potencial tiene simetría acimutal, moverse transversalmente no gasta ni da energía


**Haciendo la teoría model-independent:** si ahora desacoplamos todo lo dependiente de la teoría, es decir, $m, \lambda\to\infty$ con $\frac{m^2}{\lambda}<\infty$ entonces queda:

$$\mathcal L=\frac{1}{2}\partial_\mu\pi\partial^\mu\pi$$

Esto se llama **nonlinear $\sigma$ model** y es el límite a bajas energías del linear sigma model. No tenemos suficiente energía para excitar los modos radiales, de modo que solo observamos los bosones de Goldstone. Igual que en $\chi PT$

# 2. Linear $\sigma$ model para ruptura espontánea de la simetría quiral

Hemos visto que el linear sigma model nos da ruptura espontánea de una simetría siguiendo el patrón: tengo potencial con simetría, así que tengo infinitos mínimos (por ende vacíos), todos igual de válidos --> escojo uno (aquí ya estoy rompiendo la simetría) --> me queda un valor esperado de vacío del campo que no es invariante bajo las transformaciones de simetría --> he roto espontáneamente la simetría de la teoría al escoger a mano un vacío

Ahora tenemos una simetría $SU(2)_L\times SU(2)_R$ que se rompe en $SU(2)_{L+R}$, y queremos construir un sigma model para estudiar su ruptura espontánea. Para ello, necesitamos primero un objeto que se transforme bajo este grupo, y esto puede ser una matriz: 

$$\Sigma\to L\Sigma R^\dagger$$

$$\Sigma^\dagger = R\Sigma^\dagger L^\dagger$$

donde $L=e^{i\theta^a_L\tau^a}, R=e^{i\theta^a_R\tau^a}$. Construyamos entonces el modelo sigma lineal correspondiente:

$$\mathcal L=Tr[\partial_\mu \Sigma \partial^\mu \Sigma^\dagger] + m^2Tr[\Sigma\Sigma^\dagger]-\frac{\lambda}{4!}Tr[(\Sigma\Sigma^\dagger)^2]$$

Comprobamos inmediatamente que $SU(2)_L\times SU(2)_R$ es una simetría de la teoría porque la traza es cíclica

Ahora tenemos un potencial análogo al del sigma model para campo escalar:

$$V(\Sigma)=-m^2Tr[\Sigma\Sigma^\dagger]+\frac{\lambda}{4!}Tr[(\Sigma\Sigma^\dagger)^2]$$

donde el mínimo es: $\Sigma_{min}=\sqrt{\frac{2m^2}{\lambda}}\begin{pmatrix}1 & 0 \\ 0 & 1\end{pmatrix}e^{i\theta_a\tau_a}$

Igual que antes, escogemos el vacío como aquel en el que $\{\theta_a\}=0$, tenemos que el estado de vacío es tal que el valor esperado de nuestro campo es:

$$ \langle \Sigma\rangle = v\begin{pmatrix}1 & 0 \\ 0 & 1\end{pmatrix}$$

**Observación:** el valor esperado de vacío no es ya invariante bajo la simetría de la teoría: $L\langle \Sigma\rangle R^\dagger = vLR^\dagger \neq \langle \Sigma\rangle$ A MENOS QUE $L=R$, en cuyo caso sí porque el grupo es unitario. Por ende, el valor esperado de vacío es invariante bajo un subgrupo del grupo de simetría del Lagrangiano, y tenemos por ende que la simetría se ha roto espontáneamente como:

$$SU(2)_L\times SU(2)_R\to SU(2)_{L+R}$$

que es precisamente el patrón de ruptura que buscábamos

Como hemos roto 3 generadores, es de esperar que obtengamos tres bosones de Goldstone. Para ver esto explícitamente, parametricemos el campo como antes:

$$\Sigma(x)=\frac{1}{\sqrt 2}(v+\sigma(x))e^{2i\pi^a \tau_a/F}$$

Si transformamos $\Sigma$, vemos que los $\theta^a$ se transforman "raro", pero si hacemos $L=R$ (la simetría que aún se sostiene para el valor esperado de vacío del campo), **encontramos que cómo se transformen los Goldstones es un resultado de aplicar la transformación de los generadores rotos al campo de los Goldstones**. Si por ejemplo fuera $U(1)$, tenemos que $e^{-i\alpha} e^{i\pi(x)/F}=e^{i(\pi-\alpha F)/F}$, así que el Goldstone se transforma como $\pi\to \pi(x)-F\alpha$

Si desacoplamos la teoría, tenemos que:

$$\mathcal L=\frac{F^2}{4}Tr[\partial_\mu U\partial^\mu U^\dagger]$$

donde $U=e^{2i\pi_a\tau_a/F}$ (esto es $\chi PT$ "pura" (i.e: sin masa), a primer orden en los momentos). Si además queremos meter masa porque en QCD los quarks la tienen, podemos meter:

$$\mathcal L=\frac{F^2}{4}Tr[\partial_\mu U\partial^\mu U^\dagger]+\frac{F^2}{4}Tr[2B_0M(U+U^\dagger)]$$

A menos que pidamos que la matriz de masas $M$ se transforme como un espurión, este término de masas rompe explícitamente la simetría quiral, pero si las masas son pequeñas en comparación a la masa de los Goldstones, como es el caso en QCD, entonces la simetría quiral se sostiene en buena aproximación y todo lo que hemos dicho vale, de nuevo, en buena aproximación
