# 2. Cuantización Canónica del Campo Escalar

## 1. ¿Por qué usamos campos en QFT?

**Teorema de Wigner:** las partículas elementales son irreps. del grupo de Poincaré, etiquetadas por su masa y spin (entendemos una partícula como el ket $|p, s\rangle$)

**Necesidad:** queremos que estas irreps. sean unitarias para que las amplitudes de probabilidad no dependan del inercial, ya que de lo contrario la probabilidad de observar un mismo proceso desde dos inerciales sería distinta $\to$ las leyes de la Física no serían iguales en todos los sistemas de referencia inerciales $\to$ rompemos principio de relatividad de Einstein y con ello la teoría ya no es relativista

**Problema:** no existen irreps. unitarias del grupo de Poincaré de dimensión finita. Tampoco pasa nada, porque nuestros estados de partículas ya eran infinitos al venir etiquetados por $\{|p, s\rangle\}$

**Solución:** no me queda más remedio que usar irrep. unitaria infinita. **La embebo en un campo** (en el sentido de que en vez de coger todos los $\{|p, s\rangle\}$, puedo recuperar estos estados a través de operadores de campo actuando sobre el vacío)

**Ventaja:** los campos me permiten hacer que mi teoría sea local en el espacio tiempo

**Nota:** en los 60s se intentó formular una teoría relativista solo con partículas, sin introducir campos, imponiendo relatividad, unitariedad, analiticidad y causalidad (S matrix programme)

### Normalización invariante

Por ortogonalidad, sabemos que $\langle \vec k|\vec p\rangle \propto \delta^3(\vec k-\vec p)$

Sin embargo, esa delta no es invariante Poincaré, y queremos que lo sea por lo dicho anteriormente. De hecho: $\delta^3(\vec k-\vec p)\to \delta^3(\vec k \ '-\vec p\ ')\frac{E'}{E}$. Entonces ya está: si pasamos la energía al otro lado, tenemos que la cantidad E*delta sí es invariante Poincaré, y por ello podemos escoger una normalización invariante, de la forma:

$$\boxed{\langle \vec k |\vec p\rangle = 2E_p(2\pi)^3\delta^3(\vec k-\vec p)}$$

## 2. Cuantización canónica

**Nota:** aunque se suele separar entre "primera" y "segunda" cuantización, realmente son el mismo procedimiento, solo que uno se aplica a la mecánica clásica de la partícula y la otra a la mecánica clásica del campo

**Pasos para cuantización canónica:**

1. Tengo un sistema clásico con unos ciertos grados de libertad
2. Promuevo el estado del sistema a un cierto elemento vector de un espacio de Hilbert (o Fock)
3. Promuevo los observables a operadores actuando sobre dichos vectores
4. Impongo relaciones de (anti)conmutación entre estos operadores --> ahora forman un álgebra



**Aplicación: primera cuantización**

1. Cojo sistema mecánico clásico (en general de varias partículas, aunque no tiene ni por qué involucrar partículas), cuyo estado viene determinado por la tupla $(q_1, \ldots, q_N, p_1, \ldots, p_N)$
2. Promuevo el estado a un elemento vector de un espacio de Hilbert
3. Promuevo los observables ($x, \vec p, \vec J,\ldots$) a operadores que actúan sobre los elementos de este espacio de Hilbert
4. Impongo relación de conmutación $[x,p]=i\hbar$

**Aplicación: segunda cuantización**

1. Cojo campo clásico: $\phi(x)=\int\frac{d^3\vec p}{\sqrt{2E_p}}(a_p e^{ip\cdot x}+a_p^\dagger e^{-ip\cdot x})$
2. Estados del sistema vienen dados por kets en espacio de Fock: $|p, s\rangle$
3. Promuevo campos a operadores de campo: $\phi(x)\to \hat \phi(x)=\int\frac{d^3\vec p}{\sqrt{2E_p}}(\hat a_p e^{ip\cdot x}+\hat a_p^\dagger(x)e^{-ip\cdot x})$
4. Impongo relaciones de conmutación (a tiempos iguales) entre campo y momento canónico o, equivalentemente, de sus modos de Fourier: $[\hat \phi(\vec x, t), \hat\Pi(\vec x \ ', t)]=i\delta^{(3)}(\vec x-\vec x \ ') \Longleftrightarrow [\hat a_p, \hat a_{p'}^\dagger]=(2\pi)^3\delta^{(3)}(\vec p-\vec p \ ')$

Si se quiere, al cuantizar un campo, tenemos infinitos osciladores armónicos, con sus correspondientes operadores de creación y destrucción, de tal forma que:

$$\boxed{|\vec p\rangle = \sqrt{2E_p}a_p^\dagger|0\rangle}$$

**Nota:** este factor es necesario para que la creación de partículas sea consistente con la normalización que hemos elegido (hágase $\langle \vec k|\vec p\rangle$)

### Espectro de la teoría

El Lagrangiano de la teoría escalar libre es: $\mathcal L = \frac{1}{2}(\partial \phi)^2 -\frac{1}{2}m^2\phi^2$

La densidad Lagrangiana asociada a esta teoría es: $\mathcal H(x) = \frac{1}{2}[\Pi^2+(\vec\nabla \phi)^2+m^2\phi^2]$

Si integramos en el espacio físico, tenemos el Hamiltoniano asociado a la teoría: $H=\int d^3\vec d\mathcal H = \frac{1}{2}\int\frac{d^3\vec p}{(2\pi)^3 E_p}E_p[a_p^\dagger a_p + a_p^\dagger a_p]=\int\frac{d^3\vec p}{(2\pi)^3}E_p[a_p^\dagger a_p + \frac{1}{2}[a_p, a_p^\dagger]]$

Este Hamiltoniano se corresponde con el de un MAS para cada modo de Fourier del campo escalar, y como tal vemos que el conmutador da precisamente que la energía del vacío es la suma del estado fundamental de cada oscilador, y como tenemos infinitos, diverge

**Solución:** restamos energía del vacío o tomamos ordenación normal. En ausencia de gravedad, ambas dan un resultado equivalente, donde, como esperaríamos, las únicas entidades que contribuyen a la energía son las partículas (i.e: las excitaciones provocadas por los campos) y el vacío tiene por construcción energía 0. Este fix se puede entender con una analogía: tenemos un recipiente con infinita agua y una bola de hierro en el fondo. Queremos determinar la masa de la bola pesando el recipiente. Lógicamente, al pesarlo naïvely, obtendremos que la masa de todo ello es infinito, así que si queremos solo la masa de la bola, hay que "tarar la balanza" sin bola, de modo que quitamos la masa del agua. Esto está también muy relacionado con el tratamiento de infinitos en renormalización, donde también se restan cantidades infinitas de este modo para obtener información física

**Nota:** cuando hay gravedad, el espacio-tiempo ya no es Minkowski y la energía del vacío no debería poder eliminarse así, de modo que queda la cuestión de si tenemos energía infinita, de si estamos definiendo las cosas bien, de si estamos usando el formalismo correcto y de si hay Física más allá

### Los campos escalares son bosónicos

**Proposición:** el campo escalar da lugar a partículas bosónicas de spin 0

**Dem:**

Que son de spin 0 es inmediato porque el campo, al ser escalar, ya tiene spin 0. Las excitaciones del campo serán bosónicas si el estado del sistema no cambia al permutar dos partículas: $|k_1 k_2\rangle = |k_2 k_1\rangle$?

$|k_1k_2\rangle = a_{k_1}^\dagger a_{k_2}^\dagger |0\rangle$

$|k_2k_1\rangle = a_{k_2}^\dagger a_{k_1}^\dagger |0\rangle$

Pero como el conmutador de dos operadores creación es 0, podemos intercambiar su orden, así que $|k_2k_1\rangle = a_{k_2}^\dagger a_{k_1}^\dagger |0\rangle=a_{k_1}^\dagger a_{k_2}^\dagger |0\rangle=|k_1k_2\rangle_\blacksquare$

**Observación:** lo que en QM era un principio, vemos que en QFT es un resultado natural de la teoría

## 3. Microcausalidad y propagadores

**Def.:** un "propagador" es la amplitud de probabilidad de que los modos del campo se propaguen de un evento del espacio tiempo a otro

**Proposición:** para dos eventos $x, y$ tales que $x^0>y^0$, el propagador del campo escalar libre es: $D(x-y)=\int\frac{d^3\vec p}{(2\pi)^32E_p}e^{-ip\cdot(x-y)}$. Lógicamente, si $x^0<y^0$, esto debe anularse, porque no puede haber propagación hacia el pasado

**Proposición:** para dos eventos simultaneos pero separados espacialmente, tenemos que $D(x-y)\xrightarrow[r\to\infty]{}e^{-mr}$, de modo que la propagación está suprimida exponencialmente al alejarnos del cono de Minkowski, pero la amplitud de probabilidad de que se dé la propagación sigue sin ser 0: hay un efecto de tunelaje fuera del cono de Minkowski

**Observación:** el propagador se filtra por el cono de Minkowski. Consideremos la propagación puramente temporal hacia el futuro, tal que $y=(0,\vec 0), \ x=(t, \vec 0)$ con $t>0$. Entonces, es inmediato que: $D(x-y)\xrightarrow[t\to\infty]{}e^{-imt}$. No pasa nada porque el propagador no es una cantidad observable, así que no podemos pedirle que respete relatividad especial

**Criterio de microcausalidad:** dos acciones separadas por un intervalo spacelike (i.e: sobre eventos a lados distintos del cono de Minkowski) están causalmente desconectadas. Entonces, debería dar igual el orden en el que actuemos los operadores de campo. Encontramos que este es un resultado de nuestra teoría:

$$\langle 0 |[\phi(x), \phi(y)]|0\rangle = D(x-y)-D(y-x)$$

que es 0 para $(x-y)^2<0$ (spacelike)

**Def.:** deseamos que la propagación de los modos entre dos eventos siempre se haga en el sentido temporal creciente, de modo que definimos el propagador de Feynman entre los eventos $x, y$ como:

$$\boxed{D_F(x-y)=\langle 0|T(\phi(x)\phi(y))|0\rangle=D(x-y)=\int \frac{d^4p}{(2\pi)^4}\frac{i}{p^2-m^2+i\varepsilon}e^{-ip\cdot(x-y)}}$$

donde $T$ es el operador de ordenación temporal: $T(\phi(x)\phi(y))=\Theta(x^0-y^0)\phi(x)\phi(y)+\Theta(y^0-x^0)\phi(y)\phi(x)$

**Observación:** véase que en el propagador de Feynman no salen las Theta's explícitamente. Esto se debe a que hace un uso astuto del teorema del residuo con la prescripción de Feynman en el denominador para seleccionar solamente la parte que corresponda a la ordenación temporal

### ¿Por qué conmutan los operadores de creación / destrucción?

Si hubiéramos impuesto anticonmutación en lugar de conmutación, tendríamos entonces que el propagador queda de la forma:

$$D(x-y)=i\int \frac{d^4\vec p}{(2\pi)^4}\frac{p^0}{E_p}\frac{e^{ip\cdot(x-y)}}{p^2-m^2+i\varepsilon}$$

que obviamente no es Lorentz-invariante sino que se transforma, pero recuerda que este propagador es esencialmente la función a 2 puntos, que es el valor esperado de dos campos escalares y por ende debería ser asimismo un escalar

**Conclusión:** por consistencia, los modos de creación / destrucción el campo escalar conmutan

**Nota: este criterio de consistencia es equivalente a pedir que se cumpla microcausalidad o a pedir que el espectro de la teoría esté inferiormente acotado, y de hecho los 3 son equivalentes**

# 4. Campo escalar complejo

Ahora tenemos dos grados de libertad (en cada evento): tomamos $\phi, \phi^*$ como grados de libertad independientes


## 4.1 Cuantización

Todo igual, la única diferencia es que como el campo ya no es real (i.e: autoadjunto), ahora tenemos distintos operadores de creación y destrucción: damos lugar a la distinción entre partícula y antipartícula

$$\boxed{\begin{gather}\phi(x)=\int\frac{d^3\vec p}{(2\pi)^3\sqrt{2E_p}}(a_p e^{-ip\cdot x}+b_p^\dagger e^{ip\cdot x}) \\ \phi^*(x)=\int\frac{d^3\vec p}{(2\pi)^3\sqrt{2E_p}}(b_p e^{-ip\cdot x}+a_p^\dagger e^{ip\cdot x})\end{gather}}$$

Las relaciones de conmutación son las habituales:

$$[\hat a_p, \hat a_{p'}^\dagger]=(2\pi)^3\delta^{(3)}(\vec p-\vec p \ ')$$

$$[\hat b_p, \hat b_{p'}^\dagger]=(2\pi)^3\delta^{(3)}(\vec p-\vec p \ ')$$

Dode: $a$ asociado a partículas ordinarias y $b$ a antipartículas

## 4.2 Simetría $U(1)$ global

**Proposición:** la teoría libre del campo escalar complejo tiene una simetría de fase global

Corriente Noether conservada: $j^\mu=i[\partial_\mu\phi^*\phi - \phi^*\partial_\mu \phi]$

Carga conservada: $Q=\int d^3\vec p[a_p^\dagger a_p-b_p^\dagger b_p]$ (carga eléctrica, porque va con los operadores de densidad de partículas - antipartículas)
